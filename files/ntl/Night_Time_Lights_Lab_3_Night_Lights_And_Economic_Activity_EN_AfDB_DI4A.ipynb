## 01 · Introduction: why night-time lights measure the economy

*What night-time lights measure — and what they do not*

> 🎯 **Objective** — Understand what night-time lights (NTL) measure, why they say something about economic
> activity, and where what they can tell us ends.

**The intuition.** Producing, transporting, trading and consuming all require energy, and part of that energy
escapes at night as light that is visible from space. The more active an area, the brighter it shines. Satellite
sensors measure this light everywhere, every night, with the same method: it is a **homogeneous, spatially
detailed and timely** indicator, where national accounts are annual, published late and rarely regional.

**What the literature says.**
- *Henderson, Storeygard & Weil (2012, AER)* — growth in lights tracks GDP growth; the elasticity of GDP with
  respect to lights is about 0.3 (section 7 estimates the reverse regression, lights on GDP, whose
  elasticity is of the order of the inverse, around 3 — compare like with like); lights improve growth estimates where statistics are weak.
- *Chen & Nordhaus (2011, PNAS)* — lights add information mainly for countries with fragile statistics
  and for subnational regions.
- *Elvidge et al. (2017, 2021)* — monthly and annual VIIRS products (VNL V2): stray-light correction,
  background and ephemeral-light masks.
- *Román et al. (2018)* — NASA Black Marble: BRDF and lunar correction, suited to daily series.
- *Li et al. (2020, Scientific Data)* — harmonised DMSP–VIIRS series, 1992–2021.
- *Gibson, Olivia & Boe-Gibson (2020-2021)* — limits of lights in rural and sparsely populated areas.

**What they do not say.** Lights measure *radiance* at around 01:30, not value added.
Agriculture, daytime trade, much of the services sector and the rural informal economy are barely visible.
Section 11 and the dashboard's "Method & limitations" page detail these biases.

**Two generations of sensors.**

| Sensor | Period | Resolution | Measurement | Use here |
|---|---|---|---|---|
| DMSP-OLS | 1992–2013 | ~1 km | digital number 0–63, saturates in cities | long series (harmonised version), diagnostics |
| VIIRS DNB | 2012 → | ~500 m | calibrated radiance (nW·cm⁻²·sr⁻¹) | all main analyses |
| Black Marble VNP46A2 | 2012 → | ~500 m, **daily** | BRDF/moon-corrected radiance | detailed shock analysis |

> 🔗 **Where this notebook sits in the series.** Parts 1 and 2 download and dissect NASA Black Marble files (`VNP46A1`–`A4`) on your machine. This notebook changes source on purpose: it reads the **VIIRS VNL V2** monthly and annual composites of the Earth Observation Group, plus Black Marble `VNP46A2` for daily shocks, **directly in Google Earth Engine**. The reason is scale: a national analysis by region and by month over ten years would mean downloading thousands of tiles, whereas Earth Engine computes the regional sums on its servers and returns only the tables. The files collected in Parts 1 and 2 are therefore not needed here; what carries over is the method — quality flags, masks, fill values, the order of the operations.

```python
# --- 2.1 Environment: project code + missing packages (Colab / Kaggle / local) ----------
# The notebook is SELF-CONTAINED: if the src/ folder is not next to it (e.g. the .ipynb was
# opened alone in Colab), the project code embedded below is unpacked automatically.
import importlib.util, subprocess, sys, os, io, base64, zipfile
from pathlib import Path

IN_COLAB = importlib.util.find_spec("google.colab") is not None
# A cloned repository (src/ next to the notebook) is used as is. Otherwise the embedded code is ALWAYS
# unpacked again: an old copy left in /content (a Colab session keeps its files after "Restart
# session") must never hide the notebook's version.
ROOT = next((p for p in (Path.cwd(), Path.cwd().parent) if (p / "src" / "pipeline.py").exists()), None)
if ROOT is None:
    ROOT = Path("/content/ntl-economic-analysis") if IN_COLAB else Path.cwd() / "ntl-economic-analysis"
    ROOT.mkdir(parents=True, exist_ok=True)
    with zipfile.ZipFile(io.BytesIO(base64.b64decode(PROJECT_BUNDLE))) as z:
        z.extractall(ROOT)   # overwrites src/ and dashboard/; outputs/ (cache) is kept
    print(f"Project code (src/, dashboard/) unpacked into {ROOT}")

NEEDED = {"ee": "earthengine-api", "geemap": "geemap", "geopandas": "geopandas", "statsmodels": "statsmodels",
          "plotly": "plotly", "folium": "folium", "pycountry": "pycountry", "ruptures": "ruptures",
          "dotenv": "python-dotenv"}
missing = [pkg for mod, pkg in NEEDED.items() if importlib.util.find_spec(mod) is None]
if missing:
    print("Installing:", ", ".join(missing))
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", *missing], check=False)
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
importlib.invalidate_caches()
print("Project root:", ROOT)

# Colab: maps (geemap / ipyleaflet) are "third-party" widgets. Colab's custom widget manager is enabled
# here; otherwise Colab opens the "Code snippets › Third-party Jupyter widgets" panel.
if IN_COLAB:
    from google.colab import output
    output.enable_custom_widget_manager()
```

```python
# --- 2.2 Imports, configuration, logging ---------------------------------------------------
import logging, warnings, json, datetime as dt
import numpy as np, pandas as pd
import plotly.graph_objects as go, plotly.express as px, plotly.io as pio

from src.config import NTLConfig, DATASETS, package_versions
from src import cache, gee, viz, pipeline as P, indicators as ind, timeseries as ts
from src import messages as M, export as X
from src.boundaries import get_boundaries, attach_parents, area_km2

warnings.filterwarnings("ignore", category=FutureWarning)
logging.basicConfig(level=logging.INFO, format="%(message)s")
pd.set_option("display.float_format", lambda v: f"{v:,.3f}")
pio.templates.default = "plotly_white"

def show_md(text):
    display(Markdown(text))

fnum, fpct = M._f, M._pct      # Number formatting: fnum(0.79) -> "0.79" ; fpct(0.031) -> "+3.1%"

cfg = NTLConfig(iso3=ISO3, start_year=START_YEAR, end_year=END_YEAR, admin_levels=tuple(ADMIN_LEVELS),
                gee_project=GEE_PROJECT, custom_events=CUSTOM_EVENTS, daily_windows=DAILY_WINDOWS,
                events_csv=EVENTS_CSV, output_dir=str(ROOT / "outputs"), dashboard_dir=str(ROOT / "dashboard"))

# Basemap for geemap maps. External tile servers (OpenStreetMap, CARTO) refuse requests coming from
# Colab ("Access blocked", "API key required"). The basemap is therefore drawn by Earth Engine
# itself — the same server as the data, which always works in Colab: dark sea, dark-grey land,
# national borders (USDOS LSIB).
BASEMAP_NAME = "Earth Engine basemap (borders)"
ESRI_DARK = "https://server.arcgisonline.com/ArcGIS/rest/services/Canvas/World_Dark_Gray_Base/MapServer/tile/{z}/{y}/{x}"
ESRI_ATTR = "Tiles © Esri — Esri, HERE, Garmin, © OpenStreetMap contributors"

def ee_basemap():
    """Earth Engine image used as a basemap (no key, no third-party server)."""
    import ee
    lsib = ee.FeatureCollection("USDOS/LSIB_SIMPLE/2017")
    sea = ee.Image.constant(0).visualize(min=0, max=1, palette=["0b0f14"])
    land = ee.Image().byte().paint(lsib, 1).visualize(min=0, max=1, palette=["1f2732"])
    borders = ee.Image().byte().paint(lsib, 1, 1).visualize(min=0, max=1, palette=["6b7280"])
    return ee.ImageCollection([sea, land, borders]).mosaic()

def _no_colab_secret_lookup():
    """geemap queries the Colab secret GOOGLE_MAPS_API_KEY for every new map. Outside the UI
    ("Run all", busy session) that request times out: TimeoutException. Only the environment
    variable is read instead — no Google Maps key is needed here."""
    import os
    import geemap.coreutils as gcu
    gcu.get_google_maps_api_key = lambda key="GOOGLE_MAPS_API_KEY": os.environ.get(key)

def make_map(center, zoom=6):
    """geemap map with no external basemap: the default OSM layer is removed and replaced by ee_basemap()."""
    import geemap
    _no_colab_secret_lookup()
    m = geemap.Map(center=center, zoom=zoom)
    for lyr in list(m.layers):
        url = str(getattr(lyr, "url", ""))
        if url.startswith("http") and "earthengine" not in url:
            m.remove_layer(lyr)
    m.addLayer(ee_basemap(), {}, BASEMAP_NAME)
    return m
```

```python
# --- 2.3 Earth Engine: only if a required acquisition step is not cached yet -----------
cfg.resolve()   # validates ISO3 (clear error if unknown); END_YEAR=None -> provisional, refined below
REQUIRED = ["boundaries_ADM0", "boundaries_ADM1", "monthly", "annual"]
NEED_GEE = FORCE_DOWNLOAD or not all(cache.exists(cfg.cache_dir, s) for s in REQUIRED)
GEE_OK = False
if not SYNTHETIC_TEST:
    try:
        gee.init_gee(cfg.gee_project, interactive=NEED_GEE)
        GEE_OK = True
    except gee.GEEError as e:
        if NEED_GEE:
            raise
        print("Earth Engine not initialised — analysis from the cache (interactive GEE maps disabled).")
NEED_GEE = NEED_GEE or (GEE_OK and FORCE_DOWNLOAD)
if GEE_OK and END_YEAR is None:
    cfg.end_year = gee.last_complete_year()
cfg.resolve()
print(f"Country: {cfg.country['name']} ({cfg.iso3}) · period {cfg.start_year}–{cfg.end_year} · "
      f"levels {', '.join(cfg.admin_levels)} · cache {cfg.cache_dir}")
```

## 03 · Data acquisition and preprocessing

*From thousands of satellite images to clean, quality-controlled tables*

> 🎯 **Objective** — Turn thousands of satellite images into clean tables: one row per administrative
> unit and per month (or year), after removing everything that is not human activity.
>
> 📚 **Key concepts**
> - **Sum of lights (SoL)**: sum of the pixel radiances over an area. It is an *index*, not an energy. Pixels are
>   15 arc-seconds wide, so their area shrinks with latitude: fine inside a country, but compare countries far
>   apart in latitude with care.
> - **Cloud-free coverage (`cf_cvg`)**: number of usable observations of a pixel in the month. A pixel
>   that was never observed is not *dark*, it is *missing*.
> - **Gas flares**: gas flaring at oil sites shines as brightly as a city without any local economic
>   activity; it must be masked.
> - **Blooming / saturation**: light "spills over" around cities; DMSP saturates (value 63) in city centres.
>
> 🛠️ **Method (and why)**
>
> | Step | Rule | Rationale |
> |---|---|---|
> | Clouds | pixel valid if `cf_cvg` ≥ `cf_cvg_min` (2); a cloudy pixel takes its own annual radiance; unit-month kept if ≥ 80% of pixels are valid | do not mistake a cloudy month for a recession, nor assume the hidden pixels look like the average |
> | Negatives | set to 0 | calibration noise over dark areas |
> | Outliers | absolute cap + spikes > 5 × the pixel's annual median replaced by the median | fires, lightning, reflections |
> | Background | pixels unlit in the VNL V2 annual composite → 0 (if that year is not published yet, pixels clearly lit in the month are kept) | EOG already masks background and ephemeral lights there; a newly electrified area must not vanish |
> | Gas flares | pixels ≥ 50 nW, < 5 inhabitants, not built-up (union over years) + 2 km halo | removes flared gas and its spill-over |
> | DMSP | share of saturated pixels measured; DN > 7 threshold on the harmonised series | recommendation of Li et al. (2020) |
>
> 🔎 **Key takeaway** — The quality-control table (3.4) shows the share of excluded months, the share of light
> removed as gas flares and DMSP saturation. If any of them is high, the affected results are less reliable.

```python
# --- 3.1 Administrative boundaries (geoBoundaries CGAZ, FAO GAUL as fallback) -------------
if GEE_OK:   # server-side collections are needed for zonal statistics and maps
    GDFS, FCS, BMETA = get_boundaries(cfg)
    for lv, g in GDFS.items():
        cache.save(cfg.cache_dir, f"boundaries_{lv}", g)
    cache.save(cfg.cache_dir, "boundary_meta", BMETA)
else:
    GDFS = {lv: cache.load(cfg.cache_dir, f"boundaries_{lv}") for lv in ("ADM0", *cfg.admin_levels)}
    GDFS = attach_parents({k: v for k, v in GDFS.items() if v is not None})
    BMETA, FCS = cache.load(cfg.cache_dir, "boundary_meta") or {}, None
NAMES = P.unit_names(GDFS)
for lv, g in GDFS.items():
    print(f"{lv}: {len(g):>4} units — source {BMETA.get(lv, '?')}")
```

```python
# --- 3.2 Earth Engine context: projection, annual composites, gas-flare mask --------------
from src import preprocessing as PP
CTX = None
EMPTY_FC = {"type": "FeatureCollection", "features": []}
# Version of the computations: changing it invalidates the cache (here: cloudy pixels imputed, background
# mask of the latest year, unmasked threshold sensitivity — methodological review of 24 Oct 2026)
CODE_VERSION = "2026-10-24-review"
_rp = cache.load(cfg.cache_dir, "run_params")
OUTDATED_CODE = _rp is not None and _rp.get("code_version") != CODE_VERSION
if OUTDATED_CODE:
    print("⚠️ Cache produced by an earlier version of the code: everything is recomputed.")
if GEE_OK:   # server-side collections are needed for zonal statistics and maps
    from src import acquisition as A
    CTX = A.GEEContext(cfg, FCS)
    print("VNL V2 annual composites:", {y: c.split('/')[-1] for y, c in CTX.annual_years.items()})
    if cfg.flare_mode != "off":
        # candidates detected once, vectorised, cached; the mask is then a cheap painted vector
        POLYS = cache.cached(cfg.cache_dir, "flare_polygons", lambda: CTX.build_flare_mask() or EMPTY_FC,
                             force=FORCE_DOWNLOAD or OUTDATED_CODE)
        CTX.flare = PP.flare_mask_from_vectors(cfg, POLYS, CTX.proj)
POLYS = cache.load(cfg.cache_dir, "flare_polygons") or EMPTY_FC
FLARES = PP.flare_candidates_table(POLYS)
print(f"Gas-flare candidates: {0 if FLARES is None else len(FLARES)} (list exported for visual checking)")
```

```python
# --- 3.3 Zonal statistics (each step cached; optional steps degrade gracefully) ----------
# The cache is stored per country (outputs/<ISO3>/cache) but not per period or threshold, so the
# parameters' "signature" is kept to detect a stale cache (e.g. START_YEAR changed).
PARAMS = {k: str(getattr(cfg, k)) for k in (
    "start_year", "end_year", "admin_levels", "boundary_source", "pop_source", "cf_cvg_min", "min_valid_frac",
    "noise_floor", "radiance_cap", "spike_factor", "flare_mode", "flare_rad_min", "flare_pop_max", "flare_buffer_m")}
PARAMS["code_version"] = CODE_VERSION
OLD = cache.load(cfg.cache_dir, "run_params")
STALE = OLD is not None and OLD != PARAMS
if STALE:
    changed = sorted(k for k in PARAMS if OLD.get(k) != PARAMS[k])
    if CTX is None:
        raise RuntimeError(f"The cache was built with different parameters ({', '.join(changed)}) and Earth Engine "
                           "is not available to recompute it. Restore the parameters or initialise Earth Engine.")
    print(f"⚠️ Parameters changed since the last run ({', '.join(changed)}): statistics are recomputed.")
REFRESH = (FORCE_DOWNLOAD or STALE) and CTX is not None

def step(name, fn, required=True):
    if CTX is None and not cache.exists(cfg.cache_dir, name):
        if required:
            raise RuntimeError(f"'{name}' is not in the cache and Earth Engine is unavailable.")
        return None
    return cache.cached(cfg.cache_dir, name, fn, force=REFRESH, required=required)

LV = ("ADM0", *cfg.admin_levels)
if CTX is not None:
    from src import acquisition as A
MONTHLY = step("monthly", lambda: A.monthly_zonal(CTX, LV))
ANNUAL = step("annual", lambda: A.annual_zonal(CTX, LV))
LONG = step("long", lambda: A.harmonized_zonal(CTX, ("ADM0", "ADM1")), required=False)
DMSP = step("dmsp", lambda: A.dmsp_zonal(CTX, ("ADM0",)), required=False)
DAILY = step("daily", lambda: A.daily_zonal(CTX, A.default_daily_windows(cfg)), required=False)
NO2 = step("no2", lambda: A.no2_zonal(CTX, LV), required=False)
RASTERS = step("rasters_index", lambda: A.export_pngs(CTX, cfg.years, cfg.cache_dir / "rasters"), required=False)
if CTX is not None:
    cache.save(cfg.cache_dir, "run_params", PARAMS)

# World Bank: an empty response (network down, API unavailable) must not be cached, otherwise
# section 7 would stay empty even once the network is back.
from src.worldbank import fetch_wdi
def _wdi():
    d = fetch_wdi(cfg.iso3)
    if "gdp_const_usd" not in d or d["gdp_const_usd"].notna().sum() == 0:
        raise RuntimeError("no GDP received from the World Bank API (network?)")
    return d
_old_wdi = cache.load(cfg.cache_dir, "wdi")
_bad_wdi = _old_wdi is not None and ("gdp_const_usd" not in _old_wdi or _old_wdi["gdp_const_usd"].notna().sum() == 0)
WDI = cache.cached(cfg.cache_dir, "wdi", _wdi, force=FORCE_DOWNLOAD or _bad_wdi, required=False)
WDI = WDI if WDI is not None else pd.DataFrame({"year": []})
if "date" in MONTHLY:
    MONTHLY["date"] = pd.to_datetime(MONTHLY["date"])
print(f"Monthly: {len(MONTHLY):,} rows · Annual: {len(ANNUAL):,} rows · Long series: "
      f"{0 if LONG is None else LONG['year'].nunique()} years · Daily: {0 if DAILY is None else len(DAILY):,} rows · "
      f"GDP: {WDI.get('gdp_const_usd', pd.Series(dtype=float)).notna().sum()} years")

# Consistency check: GHSL population of the country vs World Bank (a gap > 30 % signals a problem)
_p = ANNUAL.loc[ANNUAL["level"] == "ADM0"].groupby("year")["pop"].sum()
if "population" in WDI and len(_p):
    _w = WDI.set_index("year")["population"].dropna()
    _y = max(set(_p.index) & set(_w.index), default=None)
    if _y is not None:
        _r = _p[_y] / _w[_y]
        print(f"Population check {_y}: GHSL {_p[_y]/1e6:,.2f} M vs World Bank {_w[_y]/1e6:,.2f} M (ratio {_r:.2f})",
              "✔" if 0.7 <= _r <= 1.3 else "⚠️ abnormal gap: check the population source")
```

```python
# --- 3.4 Quality control ---------------------------------------------------------------------
nid = P.national_id(GDFS)
nat_m = MONTHLY[MONTHLY["unit_id"] == nid]
nat_a = ind.add_derived(ANNUAL[ANNUAL["unit_id"] == nid]).sort_values("year")
qc = {
    "Months observed (national)": f"{nat_m['sol_adj'].notna().sum()} / {len(nat_m)}",
    "Mean cloud-free coverage": f"{nat_m['valid_frac'].mean():.0%}",
    "Unit-months excluded (coverage < 80%)": f"{MONTHLY['sol_adj'].isna().mean():.1%}",
    "Light removed as gas flares (last year)": f"{nat_a['flare_share'].iloc[-1]:.1%}" if "flare_share" in nat_a else "n/a",
    "Annual products used": ", ".join(sorted(ANNUAL["product"].astype(str).str.split("/").str[-1].unique())),
    "Population (source, year)": f"{cfg.pop_source.upper()} — {ANNUAL['pop_source_year'].iloc[-1]}",
}
if "imputed_share" in nat_m:
    qc["Light imputed for cloudy pixels (mean)"] = f"{nat_m['imputed_share'].mean():.1%}"
if "annual_ref_year" in MONTHLY:
    _ref = MONTHLY.assign(y=MONTHLY["date"].dt.year).groupby("y")["annual_ref_year"].first()
    _late = [f"{y} (mask {int(r)})" for y, r in _ref.items() if int(r) != int(y)]
    qc["Years masked with another year's composite"] = ", ".join(_late) if _late else "none"
if DMSP is not None and len(DMSP):
    qc["DMSP saturation (pixels at 63, mean)"] = f"{DMSP.loc[DMSP['unit_id'] == nid, 'sat_share'].mean():.2%}"
display(pd.Series(qc, name="Quality control").to_frame())

# Annual VNL series: continuity at the V2.1 -> V2.2 switch (different processing)
_na = ANNUAL[ANNUAL["unit_id"] == nid].sort_values("year")
_v22 = _na["product"].astype(str).str.contains("V22")
if _v22.any() and (~_v22).any():
    _s = _na.set_index("year")["sol"]; _y = int(_na.loc[_v22, "year"].min())
    if {_y - 2, _y - 1, _y, _y + 1} <= set(_s.index):
        print(f"VNL V2.1 → V2.2 switch in {_y}: growth {_y-2}→{_y-1}: {fpct(_s[_y-1]/_s[_y-2]-1)} · switch "
              f"{_y-1}→{_y}: {fpct(_s[_y]/_s[_y-1]-1)} · {_y}→{_y+1}: {fpct(_s[_y+1]/_s[_y]-1)}  → the GDP regressions "
              "of section 7 include a version dummy.")

# Harmonised long series: continuity at the DMSP -> VIIRS junction (2013 -> 2014)
if LONG is not None and len(LONG):
    ln = LONG[LONG["unit_id"] == nid].sort_values("year").set_index("year")["dn_sum"]
    if {2012, 2013, 2014, 2015} <= set(ln.index):
        g_before = ln[2013] / ln[2012] - 1
        g_junction = ln[2014] / ln[2013] - 1
        g_after = ln[2015] / ln[2014] - 1
        print(f"Harmonised series — growth 2012→13: {fpct(g_before)} · junction 2013→14: {fpct(g_junction)} · "
              f"2014→15: {fpct(g_after)}  → a jump at the junction very different from neighbouring years signals "
              "a residual sensor break.")
```

## 04 · Exploration: maps and descriptive statistics

*Look at the data before computing*

> 🎯 **Objective** — *Look* at the data before computing: where the light is, how it has changed, which
> regions dominate.
>
> 📚 **Key concepts** — Radiance has a highly skewed distribution (a few very bright urban pixels,
> many faint pixels): it is mapped on a **logarithmic scale**. **Light per capita**
> distinguishes an active region from one that is merely populated.
>
> 🛠️ **Method** — Interactive map (light in key years, change map, regional statistics on hover), before/after map, regional choropleth,
> descriptive statistics by ADM1.
>
> 🔎 **Key takeaway** — Spot artefacts by eye (residual gas flares, very bright isolated sites,
> cloudy areas) and the geography of activity: capital, corridors, ports, mines.

```python
# --- 4.1 Descriptive statistics by ADM1 (last year) -----------------------------------------
Y1 = int(ANNUAL["year"].max()); Y0 = int(ANNUAL["year"].min())
a1 = ind.add_derived(ANNUAL[(ANNUAL["level"] == "ADM1") & (ANNUAL["year"] == Y1)]).copy()
a1["name"] = a1["unit_id"].map(NAMES)
a1["sol_share"] = a1["sol"] / a1["sol"].sum()
a1["pop_share"] = a1["pop"] / a1["pop"].sum()
desc = (a1[["name", "sol", "sol_share", "pop", "pop_share", "sol_pc", "lit_area_share", "dark_share"]]
        .sort_values("sol", ascending=False).set_index("name"))
display(desc.style.format({"sol": "{:,.0f}", "sol_share": "{:.1%}", "pop": "{:,.0f}", "pop_share": "{:.1%}",
                           "sol_pc": "{:.3f}", "lit_area_share": "{:.1%}", "dark_share": "{:.0%}"})
        .bar(subset=["sol_share"], color="#9ec5f4"))
```

```python
# --- 4.2 Interactive map: light year by year, over the whole period ---------------------
# "Year" slider + play button: every year of the period. Each year's tiles are prepared once by
# Earth Engine; changing the year only swaps the tile address (instant).
NTL_MAP_VIS = {"min": -0.3, "max": 1.9,   # log10 radiance: 0.5 nW (threshold) → ~80 nW (urban cores)
               "palette": ["2b0b57", "5c1a8a", "9b2a8c", "d6456c", "f9804a", "fcc35a", "fcfdbf"]}
CHANGE_VIS = {"min": -1, "max": 1, "palette": ["d7263d", "f28f6b", "f6d6c8", "cfe3f6", "5fa8e8", "1f78d1"]}

def light_change(y0, y1, min_change=0.1):
    """Δlog10(radiance) between two years; masked if |Δ| < min_change (±26%) or if the pixel is
    dark in both years. +1 = light ×10, −1 = light ÷10."""
    floor = cfg.noise_floor
    a = A.annual_image(CTX, y0).unmask(0).max(floor)
    b = A.annual_image(CTX, y1).unmask(0).max(floor)
    d = b.log10().subtract(a.log10())
    return d.updateMask(a.gt(floor).Or(b.gt(floor)).And(d.abs().gte(min_change))).clip(CTX.region)

def region_layer(gdfs, a1, annual):
    """ADM1 + readable statistics (shown on hover)."""
    y0, y1 = int(annual["year"].min()), int(annual["year"].max())
    s0 = annual[(annual["level"] == "ADM1") & (annual["year"] == y0)].set_index("unit_id")["sol"]
    g = gdfs["ADM1"][["unit_id", "geometry"]].merge(
        a1[["unit_id", "name", "sol", "sol_share", "sol_pc", "dark_share"]], on="unit_id", how="left")
    chg = g["unit_id"].map(s0)
    f = lambda v, fmt: "n/a" if v != v else format(v, fmt)
    g = g.assign(**{
        "Region": g["name"].fillna(g["unit_id"]),
        f"Sum of lights {y1}": [f(v, ",.0f") for v in g["sol"]],
        "National share": [f(100 * v, ".1f") + "%" for v in g["sol_share"]],
        "Light per capita": [f(v, ".3f") for v in g["sol_pc"]],
        f"Change {y0}→{y1}": [f(100 * (v / c - 1), "+.0f") + "%" if c and c == c else "n/a"
                              for v, c in zip(g["sol"], chg)],
        "Population in the dark": [f(100 * v, ".0f") + "%" for v in g["dark_share"]],
    })
    return g.drop(columns=["unit_id", "name", "sol", "sol_share", "sol_pc", "dark_share"])

def ee_tiles(img, vis):
    """URL of the Earth Engine tiles of an image (one request per image, then cached by the browser)."""
    return img.getMapId(vis)["tile_fetcher"].url_format

w, s, e, n = GDFS["ADM0"].total_bounds
if CTX is not None:
    import ipyleaflet, ipywidgets as W
    YRS = [y for y in cfg.years if y in CTX.annual_years]
    YA, YB = YRS[0], YRS[-1]
    print("Preparing Earth Engine tiles: {n} years ({a}–{b})…".format(n=len(YRS), a=YA, b=YB))
    NTL_URLS = {y: ee_tiles(A.annual_image(CTX, y).log10(), NTL_MAP_VIS) for y in YRS}

    m = make_map(center=[(s + n) / 2, (w + e) / 2], zoom=6)
    ntl = ipyleaflet.TileLayer(url=NTL_URLS[YB], name="Night-time light (slider year)", max_zoom=24,
                               attribution="VIIRS VNL V2 (EOG) · Google Earth Engine")
    m.add(ntl)
    m.addLayer(light_change(YA, YB), CHANGE_VIS, "Change {a}→{b} (blue = increase, red = decrease)".format(a=YA, b=YB), shown=False)
    m.add_gdf(region_layer(GDFS, a1, ANNUAL), layer_name="ADM1 regions (hover = statistics)",
              style={"color": "#E5E7EB", "weight": 0.8, "fillOpacity": 0.01},
              hover_style={"color": "#F5C242", "weight": 2.5, "fillOpacity": 0.05}, zoom_to_layer=False,
              style_callback=lambda feat: {"color": "#E5E7EB", "fillColor": "#000000"})

    # IntSlider (whole years) linked to the Play button: no decimal values from the browser
    year = W.IntSlider(min=YA, max=YB, step=1, value=YB, description="Year", continuous_update=True,
                       layout=W.Layout(width="330px"))
    play = W.Play(min=YA, max=YB, step=1, value=YB, interval=1600, description="Play",
                  layout=W.Layout(width="90px"))
    W.jslink((play, "value"), (year, "value"))
    nearest = lambda y: min(YRS, key=lambda v: abs(v - int(y)))   # missing year -> closest available
    year.observe(lambda ch: setattr(ntl, "url", NTL_URLS[nearest(ch["new"])]), names="value")
    m.add(ipyleaflet.WidgetControl(widget=W.HBox([play, year]), position="topright"))
    m.add_colorbar(NTL_MAP_VIS, label="Radiance, log10(nW·cm⁻²·sr⁻¹)", position="bottomleft")
    m.add_legend(title="Change {a}→{b}".format(a=YA, b=YB), position="bottomright", legend_dict={"Strong increase (≥ ×3)": "1f78d1", "Increase (+26% to ×3)": "5fa8e8", "Decrease (−21% to ÷3)": "f28f6b", "Strong decrease (≤ ÷3)": "d7263d"})
    m.fit_bounds([[s, w], [n, e]])
    print("Slider at the top right: pick a year or click ▶ to animate {a}–{b}. Layers: top-right icon. Hover over a region to see its statistics.".format(a=YA, b=YB))
    display(m)
else:
    # Offline fallback: the static PNG overlays exported for the dashboard, one layer per year
    import folium
    R = RASTERS or {}
    w, s, e, n = R.get("bounds", GDFS["ADM0"].total_bounds)
    m = folium.Map(location=[(s + n) / 2, (w + e) / 2], zoom_start=6, tiles=ESRI_DARK, attr=ESRI_ATTR)
    for i, (y, f) in enumerate(sorted(R.get("ntl", {}).items())):
        folium.raster_layers.ImageOverlay(str(cfg.cache_dir / "rasters" / f), bounds=[[s, w], [n, e]], name=y,
                                          opacity=0.9, show=(i == len(R.get("ntl", {})) - 1),
                                          mercator_project=False).add_to(m)
    folium.LayerControl(collapsed=False).add_to(m)
    display(m)
```

```python
# --- 4.3 Compare two years: synchronised side-by-side maps ------------------------------
# Two linked maps (shared pan and zoom). Pick any year for each side.
if CTX is not None:
    import ipyleaflet, ipywidgets as W

    def side(y0, label):
        mp = make_map(center=[(s + n) / 2, (w + e) / 2], zoom=6)
        mp.layout.height = "520px"
        lyr = ipyleaflet.TileLayer(url=NTL_URLS[y0], name=str(y0), max_zoom=24, attribution="VIIRS · Google Earth Engine")
        mp.add(lyr)
        mp.add_gdf(GDFS["ADM1"][["geometry"]], layer_name="ADM1", zoom_to_layer=False,
                   style={"color": "#E5E7EB", "weight": 0.6, "fillOpacity": 0},
                   hover_style={"weight": 0.6}, style_callback=lambda feat: {"color": "#E5E7EB", "fillColor": "#000000"})
        dd = W.Dropdown(options=YRS, value=y0, description=label, layout=W.Layout(width="190px"))
        dd.observe(lambda ch: setattr(lyr, "url", NTL_URLS[ch["new"]]), names="value")
        mp.add(ipyleaflet.WidgetControl(widget=dd, position="topright"))
        mp.fit_bounds([[s, w], [n, e]])
        return mp

    mL, mR = side(YA, "Left"), side(YB, "Right")
    W.jslink((mL, "center"), (mR, "center"))
    W.jslink((mL, "zoom"), (mR, "zoom"))
    mL.layout.width = mR.layout.width = "50%"
    print("Pan or zoom one map and the other follows. Change each map's year with its menu (top right).")
    display(W.HBox([mL, mR], layout=W.Layout(width="100%")))
else:
    print("Interactive comparison: available with Earth Engine (and in the dashboard, Map page).")
```

## 05 · National and regional time dynamics

*Trend, seasonality and regional series*

> 🎯 **Objective** — Describe how light activity evolves and separate the trend from seasonal variation.
>
> 📚 **Key concepts**
> - **Seasonality**: light varies over the year for reasons unrelated to the economy (snow, vegetation, length of
>   nights, festivities). Comparing December with November is therefore meaningless.
> - **STL** (*Seasonal-Trend decomposition using Loess*): splits the series into trend + season + remainder. It is
>   applied to the logarithm, because seasonality is proportional to the level.
> - **Year-on-year change**: comparing a month with the same month of the previous year neutralises the season.
>
> 🛠️ **Method** — Robust STL (period 12) on monthly log(SoL); robust annual trend (Theil–Sen) on the
> annual composites; series by region (ADM1); harmonised long DMSP–VIIRS series (Li et al., 2020).
>
> 🔎 **Key takeaway** — The *seasonally adjusted* curve is the one to read. A brief rise in the observed series
> may be nothing more than a seasonal effect.

```python
DYN = P.dynamics(MONTHLY, ANNUAL, GDFS, cfg)
viz.national_series(DYN["stl"], None, None, DYN["anomalies"], cfg.country["name"]).show()
```

```python
# Year-on-year growth (same month) and ADM1 annual series
yoy = DYN["yoy"].dropna()
fig = go.Figure(go.Bar(x=yoy.index, y=100 * yoy, marker_color=np.where(yoy >= 0, viz.SERIES[0], viz.SERIES[7]),
                       hovertemplate="%{x|%b %Y}<br>%{y:.1f}%<extra></extra>"))
viz.style(fig, "Year-on-year change in the sum of lights (same month, previous year)", viz.SRC_VIIRS,
          "12-month change (%)", None, legend=False).show()
a = ANNUAL[ANNUAL["level"] == "ADM1"]
viz.regional_lines(a, "sol", NAMES, "Annual sum of lights by region (ADM1)",
                   "Sum of radiances (nW·cm⁻²·sr⁻¹)", source=viz.SRC_VIIRS + " (VNL V2 annual)").show()
```

```python
# Long-run harmonised series (DMSP-like units) — read as a trend only
if LONG is not None and len(LONG):
    ln = LONG[LONG["unit_id"] == P.national_id(GDFS)].sort_values("year")
    fig = go.Figure()
    for src, g in ln.groupby("source"):
        fig.add_scatter(x=g["year"], y=g["dn_sum"], mode="lines+markers", name=src)
    fig.add_vline(x=2013.5, line=dict(color=viz.MUTED, dash="dot"))
    viz.style(fig, "Harmonised long DMSP–VIIRS series (DMSP units)",
              "Li et al. (2020), Harmonized DMSP-VIIRS, via GEE community catalog", "Sum of DN (> 7)", "Year").show()
```

## 06 · Regional growth and convergence

*Who grows, who stagnates, who declines?*

> 🎯 **Objective** — Identify regions that grow, stagnate or decline, and see whether activity is concentrating or
> spreading out.
>
> 📚 **Key concepts**
> - **Theil–Sen slope**: median of the slopes between all pairs of years; insensitive to an outlying year.
> - **Mann–Kendall test**: is the trend significant or due to chance?
> - **β-convergence**: if initially dimly lit regions grow faster, they are catching up.
> - **σ-convergence**: does the dispersion across regions decrease over time?
> - **Herfindahl index (HHI)**: concentration of light across units (1 = everything in a single unit).
>
> 🛠️ **Method** — Robust growth per unit; class "Growth" or "Decline" if |slope| ≥ 1%/yr and p < 0.10 after
> Benjamini–Hochberg correction (hundreds of units are tested at once), otherwise "Stagnation"; β-convergence
> regression with robust errors, initial level averaged over the first three years and growth measured on the
> following years (a single noisy starting year would manufacture convergence); NO₂ (Sentinel-5P) as a complementary
> indicator of industrial activity.
>
> 🔎 **Key takeaway** — A class is assigned only if the trend is statistically established: a lot of
> "stagnation" often means "not enough information", not "standing still".

```python
GROWTH = P.regional_growth(ANNUAL, GDFS, cfg, NO2)
for lv, tr in GROWTH["trends"].items():
    print(lv, GROWTH["class_counts"][lv])
lvl = P.finest_level(GDFS)
viz.growth_ranking(GROWTH["trends"][lvl], NAMES, f"Annual growth of light by {lvl} unit, {Y0}–{Y1}").show()
```

```python
# Choropleth of growth classes
gg = GDFS[lvl].merge(GROWTH["trends"][lvl][["unit_id", "class", "growth"]], on="unit_id")
gg["growth_pct"] = (100 * gg["growth"]).round(1)
viz.choropleth(gg, "class", f"Light dynamics {Y0}–{Y1} ({lvl})", "Class",
               categorical=viz.CLASS_COLORS, tooltip_cols=["class", "growth_pct"])
```

```python
# Concentration and convergence
conc = GROWTH["concentration"]["ADM1"]
display(conc.assign(top1_name=conc["top1_name"].map(NAMES)).set_index("year")
        .style.format({"top1_share": "{:.1%}", "top10pct_share": "{:.1%}", "hhi": "{:.3f}"}))
b = GROWTH["beta_convergence"]; bd = GROWTH["beta_data"]
fig = px.scatter(bd.assign(name=bd["unit_id"].map(NAMES), log_init=np.log(bd["init"]), growth_pct=100 * bd["g"]),
                 x="log_init", y="growth_pct", hover_name="name", trendline="ols",
                 color_discrete_sequence=[viz.SERIES[0]])
viz.style(fig, f"β-convergence of light per capita ({b['level']}): β = {b['beta']:.3f} "
          f"[{b['ci'][0]:.3f}, {b['ci'][1]:.3f}]", viz.SRC_VIIRS + "; GHSL/WorldPop population",
          "Growth of light per capita (%/yr)", f"log light per capita {b.get('init_years', b['first_year'])}").show()
sig = GROWTH["sigma_convergence"]
print("σ-convergence (std. dev. of log light per capita):",
      " → ".join(f"{r.year}: {r.sigma_log:.2f}" for r in sig.iloc[[0, len(sig) // 2, -1]].itertuples()))
if "no2_corr" in GROWTH:
    c = GROWTH["no2_corr"]
    print(f"NO₂ vs light ({c['year']}, {c['n']} units): Spearman ρ = {c['rho']:.2f} (p = {c['p_value']:.3f})")
```

```python
cc = GROWTH["class_counts"][lvl]; lo, hi = b["ci"]
conv = ("convergence" if hi < 0 else "divergence" if lo > 0 else "no significant convergence or divergence")
top = NAMES.get(conc["top1_name"].iloc[-1], conc["top1_name"].iloc[-1])
show_md(f'''**🔎 Key takeaway** — Out of {sum(cc.values())} {lvl} units: **{cc.get("Growth", 0)}** in significant
growth, **{cc.get("Stagnation", 0)}** stagnating, **{cc.get("Decline", 0)}** declining. **{top}** accounts for
**{fpct(conc["top1_share"].iloc[-1], 0, False)}** of national light ({fpct(conc["top1_share"].iloc[0], 0, False)} in {Y0}).
Across units, light per capita shows: **{conv}** (β = {fnum(b["beta"], 3)}, CI [{fnum(lo, 3)}, {fnum(hi, 3)}], n = {b["n"]}) — confidence
**{"medium" if (hi < 0 or lo > 0) else "low"}**.
*Economic reading*: rising concentration signals a polarisation of activity (agglomeration around the
capital or ports); convergence indicates electrification and activities spreading towards peripheral
regions.''')
```

## 07 · NTL–GDP relationship and activity estimation

*Elasticity, nowcasting and experimental regional GDP*

> 🎯 **Objective** — Check that lights track official GDP, measure the **elasticity**, then use this
> relationship to estimate activity where statistics are missing (recent year, regions).
>
> 📚 **Key concepts**
> - **Elasticity** β in log(NTL) = α + β·log(GDP): +1% GDP ↔ +β% light. This is the reverse of the regression of
>   Henderson et al. (GDP on lights, ≈ 0.3): their 0.3 corresponds here to a β of the order of 3.
> - **Regression in levels vs in growth rates**: two series that both rise over time look correlated
>   even without any link (spurious regression). Estimation in growth rates is more demanding.
> - **HAC standard errors (Newey–West)**: correct for autocorrelation in the residuals of annual series.
> - **Diagnostics**: Durbin–Watson (autocorrelation, ≈ 2 if absent), Breusch–Pagan (heteroskedasticity),
>   Jarque–Bera (normality of residuals).
> - **Pseudo-real-time prediction**: each year is predicted by a model estimated on the *earlier* years only, as a
>   forecaster would have done at the time, and compared with two naive forecasts (last year's growth, mean past
>   growth). Lights add information only if they beat both.
> - **Bootstrap interval**: a second 95% interval for β, computed by resampling the years, shown next to the HAC one.
>
> 🛠️ **Method** — Real GDP (constant 2015 USD, World Bank API). Three estimations: VIIRS levels, VIIRS growth,
> levels on the harmonised long series (with a sensor dummy); a VNL version dummy when the annual series mixes V2.1 and V2.2. Estimate for the not-yet-published year via
> Δlog(GDP) = a + γ·Δlog(NTL). Experimental regional GDP: national GDP allocated pro rata to light.
>
> 🔎 **Key takeaway** — With 10 to 30 annual data points, an elasticity remains fragile: read the confidence
> interval, not just the point estimate. GDP estimates derived from lights are **experimental**.

```python
GDP = P.ntl_gdp(ANNUAL, MONTHLY, WDI, GDFS, LONG)
if "error" in GDP:
    print("⚠️", GDP["error"], "— section 7 limited.")
else:
    rows = []
    for k, lab in (("levels", "Levels (VIIRS)"), ("growth", "Growth (VIIRS)"), ("long", "Levels, long series")):
        e = GDP.get(k)
        if e and np.isfinite(e.get("beta", np.nan)):
            rows.append({"Specification": lab, "β": e["beta"], "CI low": e["ci"][0], "CI high": e["ci"][1],
                         "boot. CI low": e.get("ci_boot", (np.nan, np.nan))[0],
                         "boot. CI high": e.get("ci_boot", (np.nan, np.nan))[1],
                         "p": e["p_value"], "R²": e["r2"], "n": e["n"], **{k2: v for k2, v in e["diagnostics"].items()},
                         "Confidence": e["confidence"]})
    display(pd.DataFrame(rows).set_index("Specification").round(3))
    if GDP.get("vnl_versions_mixed"):
        print("The annual series mixes VNL V2.1 and V2.2: the regressions include a version dummy.")
    viz.ntl_gdp_scatter(GDP["levels"], cfg.country["name"]).show()
```

```python
# Residuals and pseudo-real-time tracking
if "levels" in GDP:
    r = GDP["levels"]["data"]
    fig = go.Figure(go.Scatter(x=r["year"], y=r["resid"], mode="lines+markers", line_color=viz.SERIES[0]))
    fig.add_hline(y=0, line=dict(color=viz.AXIS))
    viz.style(fig, "Residuals of the levels model (a visible pattern = unstable relationship)",
              viz.SRC_VIIRS + "; " + viz.SRC_WB, "Residual (log)", "Year", legend=False).show()
    tr = GDP["tracking"]
    fig = go.Figure()
    fig.add_scatter(x=tr["year"], y=100 * tr["dl_gdp"], name="Official GDP", mode="lines+markers", line_color=viz.INK2)
    fig.add_scatter(x=tr["year"], y=100 * tr["dl_gdp_loo"], name="Estimated from lights (pseudo-real time)",
                    mode="lines+markers", line=dict(color=viz.SERIES[0], dash="dot"))
    if "dl_gdp_naive_last" in tr:
        fig.add_scatter(x=tr["year"], y=100 * tr["dl_gdp_naive_last"], name="Naive forecast (last year's growth)",
                        mode="lines", line=dict(color=viz.MUTED, dash="dash", width=1.4))
    nc = GDP["nowcast"].get("estimates")
    if nc is not None and len(nc):
        fig.add_scatter(x=nc["year"], y=100 * nc["growth_est"], name="Estimate (unpublished year)", mode="markers",
                        marker=dict(color=viz.SERIES[1], size=12, symbol="diamond"))
    viz.style(fig, "Real GDP growth: official and estimated from lights",
              viz.SRC_WB + "; monthly VIIRS aggregated by year", "Annual growth (%)", "Year").show()
    if nc is not None and len(nc):
        display(nc.assign(**{c: nc[c] / 1e9 for c in ("gdp_est", "lo95", "hi95", "lo80", "hi80")})
                .rename(columns={"gdp_est": "Estimated GDP (USD bn, 2015)"}).round(2))
```

```python
# Experimental regional GDP (ADM1, last year with official national GDP)
if "regional_gdp" in GDP and len(GDP["regional_gdp"]):
    rg = GDP["regional_gdp"]; yl = rg["year"].max()
    rg05 = GDP["regional_gdp_g05"].set_index(["unit_id", "year"])["gdp_est"]
    t = rg[rg["year"] == yl].assign(name=lambda d: d["unit_id"].map(NAMES))
    t["gdp_est_gamma05"] = [rg05.get((u, yl), np.nan) for u in t["unit_id"]]
    display(t[["name", "share", "gdp_est", "gdp_est_gamma05", "gdp_pc_est"]].sort_values("gdp_est", ascending=False)
            .set_index("name").style.format({"share": "{:.1%}", "gdp_est": "{:,.0f}", "gdp_est_gamma05": "{:,.0f}",
                                             "gdp_pc_est": "{:,.0f}"}))
    print("γ = 1: allocation pro rata to light; γ = 0.5: sensitivity giving more weight to dimly lit "
          "regions. The gap between the two columns measures the uncertainty due to this assumption.")
```

```python
if "levels" in GDP:
    L, G = GDP["levels"], GDP["growth"]
    show_md(f'''**🔎 Key takeaway** — Elasticity in levels: **β = {fnum(L["beta"])}** (95% CI {fnum(L["ci"][0])} to {fnum(L["ci"][1])},
R² = {fnum(L["r2"])}, n = {L["n"]}) — confidence **{L["confidence"]}** ({L["confidence_reason"]}). In growth rates:
β = {fnum(G.get("beta", np.nan))} (CI {fnum(G.get("ci", (np.nan, np.nan))[0])} to {fnum(G.get("ci", (np.nan, np.nan))[1])}).
In pseudo-real time, growth estimated from lights has a correlation of **{fnum(GDP.get("tracking_corr", np.nan))}** with
official growth (mean absolute error: {fnum(GDP.get("tracking_mae_pp", np.nan), 1)} percentage points of growth, against
{fnum(min(GDP.get("naive_last_mae_pp", np.nan), GDP.get("naive_mean_mae_pp", np.nan)), 1)} for the best naive forecast, over
{GDP.get("tracking_n", 0)} years) — lights {"**beat**" if GDP.get("tracking_relative_mae", np.nan) < 1 else "**do not beat**"} the naive forecast.
*Economic reading*: β < 1 means light grows more slowly than GDP (shift to services, energy
efficiency, LEDs); a β that is unstable across specifications argues against using lights to estimate the level of GDP,
and for using them only as a signal of direction and spatial distribution.''')
```

## 08 · Shock detection and analysis

*Pandemic, conflict, disasters: troughs and recovery*

> 🎯 **Objective** — Identify the shocks visible in the lights (pandemic, conflict, disasters, energy
> crises, electrification) and measure the speed of recovery.
>
> 📚 **Key concepts**
> - **Anomaly**: a month whose STL remainder is extreme (|robust z| ≥ 3.5 and deviation ≥ 5%).
> - **Structural break**: a lasting change in level or slope; detected by piecewise linear segmentation
>   with a BIC-type penalty (at least 12 months on each side).
> - **Trough**: maximum deviation from the average level of the 12 months before the shock; **recovery**: return to that level.
>
> 🛠️ **Method** — Event catalogue (COVID-19 + `CUSTOM_EVENTS` + optional ACLED/EM-DAT export); impact on
> the national and regional seasonally adjusted series; daily Black Marble zoom (high quality, cloud- and snow-free,
> 7-day rolling median, compared with the same days of the previous year so that the season cancels out).
> A trough "stands out" when it is deeper than 95% of the troughs measured the same way on every window before the
> event (placebo test). To use national COVID-19 dates, add an event with `"id": "covid19"` to `CUSTOM_EVENTS`.
>
> 🔎 **Key takeaway** — A shock is retained only if it exceeds ordinary variability. A break without a
> known event must be documented before it is interpreted.

```python
SHOCKS = P.shocks(DYN, cfg, DAILY, GDFS)
viz.national_series(DYN["stl"], SHOCKS["events"], DYN["breaks"], DYN["anomalies"], cfg.country["name"]).show()
imp = SHOCKS["impacts"]
if len(imp):
    cols = ["label", "type", "start", "end", "trough_date", "trough_pct", "gap_vs_trend_pct", "months_to_recovery",
            "recovered", "distinguishable_from_noise"]
    display(imp[[c for c in cols if c in imp]].round({"trough_pct": 1, "gap_vs_trend_pct": 1}))
display(DYN["breaks"].round({c: 2 for c in DYN["breaks"].columns if c != "date"}))
if len(SHOCKS["unexplained_breaks"]):
    print("⚠️ Breaks with no catalogued event within ±6 months:",
          ", ".join(pd.to_datetime(SHOCKS["unexplained_breaks"]["date"]).dt.strftime("%m/%Y")))
```

```python
# Regional impact (heat table) and daily Black Marble index
reg = SHOCKS["regional"]
if len(reg):
    piv = reg.assign(name=reg["unit_id"].map(NAMES)).pivot_table(index="name", columns="event", values="trough_pct")
    display(piv.round(1).style.background_gradient(cmap="RdBu", vmin=-30, vmax=30).format("{:+.1f}%"))
for w, idx in SHOCKS["daily"].items():
    ev = SHOCKS["events"][SHOCKS["events"]["id"] == w]
    lab = ev["label"].iloc[0] if len(ev) else w
    viz.daily_index_fig(idx, lab, ev["start"].iloc[0] if len(ev) else None, ev["end"].iloc[0] if len(ev) else None).show()
```

```python
if len(imp) and imp["computable"].any():
    r = imp[imp["computable"]].sort_values("trough_pct").iloc[0]
    rec = f"a return to the pre-shock level after **{int(r['months_to_recovery'])} months**" if r["recovered"] else \
          "**no return** to the pre-shock level by the end of the data"
    show_md(f'''**🔎 Key takeaway** — The deepest shock is **{r["label"]}**: a trough of **{fpct(r["trough_pct"] / 100)}**
in {pd.Timestamp(r["trough_date"]):%m/%Y}, {rec}. This trough {"stands out" if r["distinguishable_from_noise"] else "does not stand out"}
from ordinary variability (confidence **{"medium" if r["distinguishable_from_noise"] else "low"}**).
Breaks without a catalogued explanation: {len(SHOCKS["unexplained_breaks"])}.
*Economic reading*: light captures stoppages of night-time activity well (lockdowns, outages, population
displacement); it underestimates shocks that hit dimly lit activities (agriculture, daytime services).''')
else:
    show_md("**🔎 Key takeaway** — No catalogued event overlaps the period; add the country's events "
            "to `CUSTOM_EVENTS` (section 2) and read the breaks detected above.")
```

## 09 · Spatial inequality and access to electricity

*Gini, Theil and populations “in the dark”*

> 🎯 **Objective** — Measure the spatial concentration of light per capita and locate the populations "in the
> dark".
>
> 📚 **Key concepts**
> - **Gini** (0 = equality, 1 = all light in a single unit) and **Lorenz curve** (cumulative share of
>   light against cumulative share of population).
> - **Theil T**: decomposes into inequality *between* regions and *within* regions — useful to know at which
>   level to act.
> - **Population in the dark**: inhabitants of populated pixels (≥ 5 inhabitants) whose radiance stays below the detection
>   threshold (0.5 nW). It is a *proxy* for lack of access to electricity, sensitive to the chosen threshold.
> - **Urban/rural gap**: light per capita in urban areas (GHSL degree of urbanisation) / in rural areas.
>
> 🛠️ **Method** — Population-weighted indices on the finest units; bootstrap interval (300
> draws) for the change in Gini; sensitivity of the population in the dark to 3
> thresholds, computed on the unmasked annual radiance (EOG's background mask would make them identical), gas-flare
> halos excluded; comparison with the World
> Bank access rate (EG.ELC.ACCS.ZS).
>
> 🔎 **Key takeaway** — The population-in-the-dark figure is published with its **range** (depending on the threshold),
> never to the nearest person.

```python
INEQ = P.inequality(ANNUAL, GDFS, cfg, WDI)
display(INEQ["series"].set_index("year").round(3))
viz.lorenz_curves(INEQ["lorenz"], INEQ["gini_years"], f"Lorenz curves of light per capita ({INEQ['level']})").show()
print(f"Change in Gini {Y0}→{Y1}: {INEQ['gini_change']:+.3f} (95% bootstrap CI: "
      f"{INEQ['gini_change_ci'][0]:+.3f} to {INEQ['gini_change_ci'][1]:+.3f})")
```

```python
# Population "in the dark": national trend, sensitivity, World Bank comparison, units
nat = INEQ["national"]; sens = INEQ["dark_sensitivity"]
fig = go.Figure()
for i, (t, g) in enumerate(sens.groupby("threshold")):
    fig.add_scatter(x=g["year"], y=g["pop_dark"] / 1e6, mode="lines+markers", name=f"threshold {t:g} nW",
                    line=dict(color=viz.SEQ_ORANGE[2 + 2 * i], width=2))
viz.style(fig, "Population living in unlit inhabited areas, by radiance threshold",
          viz.SRC_VIIRS + "; GHSL/WorldPop", "Millions of people", "Year").show()
if "access_comparison" in INEQ and len(INEQ["access_comparison"]):
    ac = INEQ["access_comparison"]
    fig = go.Figure()
    fig.add_scatter(x=ac["year"], y=ac["ntl_access_pct"], name="Inhabited population in lit areas (lights)",
                    mode="lines+markers", line_color=viz.SERIES[1])
    fig.add_scatter(x=ac["year"], y=ac["elec_access_pct"], name="Access to electricity (World Bank)",
                    mode="lines+markers", line=dict(color=viz.INK2, dash="dot"))
    viz.style(fig, "Two measures of access to electricity", viz.SRC_VIIRS + "; " + viz.SRC_WB + " (EG.ELC.ACCS.ZS)",
              "% of population", "Year").show()
dl = INEQ.get(f"dark_{P.finest_level(GDFS)}")
if dl is not None:
    viz.dark_population_bars(dl, NAMES, f"Where do people in the dark live? ({P.finest_level(GDFS)}, {Y1})").show()
```

```python
# Map of inhabited but unlit areas
if CTX is not None:
    w, s, e, n = GDFS["ADM0"].total_bounds
    dm = make_map(center=[(s + n) / 2, (w + e) / 2], zoom=6)
    dm.addLayer(A.dark_inhabited_image(CTX, Y1), {"min": 1, "max": 2, "palette": ["f5c242", "d7263d"]},
                f"Inhabited areas: lit (yellow) / in the dark (red), {Y1}")
    dm.add_gdf(GDFS["ADM1"][["geometry"]].assign(Region=GDFS["ADM1"]["unit_id"].map(NAMES)), layer_name="ADM1 regions",
               style={"color": "#E5E7EB", "weight": 0.8, "fillOpacity": 0.01},
               hover_style={"color": "#F5C242", "weight": 2.5, "fillOpacity": 0.05}, zoom_to_layer=False,
               style_callback=lambda feat: {"color": "#E5E7EB", "fillColor": "#000000"})
    dm.add_legend(title=f"Inhabited areas, {Y1}", position="bottomright",
                  legend_dict={"lit": "f5c242", f"in the dark (< {cfg.noise_floor:g} nW)": "d7263d"})
    dm.fit_bounds([[s, w], [n, e]])
    display(dm)
elif RASTERS and RASTERS.get("dark"):
    import folium
    w, s, e, n = RASTERS["bounds"]; f = list(RASTERS["dark"].values())[0]
    dm = folium.Map(location=[(s + n) / 2, (w + e) / 2], zoom_start=6, tiles=ESRI_DARK, attr=ESRI_ATTR)
    folium.raster_layers.ImageOverlay(str(cfg.cache_dir / "rasters" / f), bounds=[[s, w], [n, e]], opacity=0.9,
                                      mercator_project=False).add_to(dm)
    display(dm)
ur = nat[["year", "sol_pc_urban", "sol_pc_rural", "urban_rural_ratio"]].set_index("year")
display(ur.round(3))
```

```python
r1 = nat.iloc[-1]; lo, hi = INEQ["gini_change_ci"]
s1 = sens[sens["year"] == sens["year"].max()]
gap = INEQ.get("access_gap_pp")
show_md(f'''**🔎 Key takeaway** — Gini of light per capita: **{fnum(INEQ["series"]["gini"].iloc[-1])}** in {Y1}
({"significant increase" if lo > 0 else "significant decrease" if hi < 0 else "no significant change"} since {Y0}).
{"The *between*-region component accounts for **" + fpct(INEQ["series"]["theil_between_share"].iloc[-1], 0, False) + "** of the Theil index." if np.isfinite(INEQ["series"]["theil_between_share"].iloc[-1]) else ""}
About **{fnum(r1["pop_dark"]/1e6)} M** people live in inhabited areas with no detectable light
(range depending on the threshold: {fnum(min(s1["pop_dark"].min(), r1["pop_dark"])/1e6)} to {fnum(max(s1["pop_dark"].max(), r1["pop_dark"])/1e6)} M) — confidence **medium**.
{"Average gap with the World Bank access rate: **" + format(gap, "+.0f") + " points**." if gap is not None else ""}
Urban light per capita is **{fnum(r1["urban_rural_ratio"], 1)} times** rural light per capita.
*Economic reading*: the map of areas in the dark helps target electrification programmes; a persistent gap with
household surveys signals either low-consumption connections (invisible from space) or areas not covered by
the surveys.''')
```

## 10 · Industrial zones: where does the country produce?

*Lights, non-residential buildings, NO₂ and OpenStreetMap validation*

> 🎯 **Objective** — Locate the country's industrial, logistics and energy zones, measure their weight in light
> activity and follow their dynamics since the start of the period.
>
> 💡 **Why it is useful**
> - Industry is the GDP component **best captured** by night lights: sites lit all night, shift work, logistics —
>   unlike agriculture or daytime services.
> - Few countries have an **up-to-date, geolocated register** of industrial zones. Satellites provide a homogeneous map,
>   comparable across countries and updated every year.
> - The light of a zone is a **leading indicator** of its activity (opening, expansion, slowdown), to be compared with
>   the industrial production index and business surveys.
> - Uses: targeting surveys, planning electricity and roads, measuring the local impact of a shock.
>
> 📚 **Key concepts**
> - **Non-residential built-up** (GHSL BUILT-S R2023A, band `built_surface_nres`): built surface used mainly for
>   non-residential purposes (factories, warehouses, retail, facilities), estimated at 10 m by the JRC.
> - **Light per resident**: a very bright but sparsely inhabited area is probably an activity zone.
> - **Tropospheric NO₂** (Sentinel-5P TROPOMI): tracer of combustion (industry, power plants, traffic).
> - **Industrial score**: weighted mean of four signals normalised to 0–1. The weights and thresholds below are
>   expert choices, not estimates: change them in `IND.PARAMS` and check that the main zones survive before any use.
> - **Precision / recall**: share of detected zones confirmed by OpenStreetMap / share of OSM zones found.
>
> 🛠️ **Method** (VIIRS grid ~460 m, entirely on Earth Engine)
>
> | Signal | Data | Normalisation | Weight |
> |---|---|---|---|
> | Light intensity | VIIRS VNL V2, median of the last 3 years | log10(radiance)/2 | 30% |
> | Non-residential built-up | GHSL BUILT-S | non-residential share / 0.6 | 35% |
> | Few residents | GHSL GHS-POP | 1 − log10(1 + people/km²)/log10(5000) | 15% |
> | NO₂ emissions | Sentinel-5P, annual mean | (NO₂ − median)/(p98 − median) | 20% |
>
> A pixel is a **candidate** when its radiance exceeds 5 nW, at least 1% of its area is built-up and its score exceeds
> 0.55 — or when it belongs to a gas flare detected in section 3 (*energy / extraction* class). Neighbouring pixels are
> grouped into **zones**; each zone gets its annual light (Theil–Sen trend), its share of national light, its
> non-residential share, its population and its location (ADM2). The list is then compared with the
> `landuse=industrial` areas of **OpenStreetMap** (Overpass API).
>
> ⚠️ **Limitations** — "Non-residential" buildings also include retail, ports, airports and public facilities; NO₂ is
> measured at ~5 km and also follows traffic; OpenStreetMap completeness varies across countries. Zones are
> **candidates**, to be checked against national registers and high-resolution imagery (Sentinel-2 views in 10.3).
>
> 🔎 **Key takeaway** — We publish *candidate zones* with their score and validation, never an official list of
> industrial zones.

```python
# --- 10.1 Candidate industrial zones (Earth Engine, result cached) ---------------------------
from src import industry as IND
ICOL = IND.CLASS_COLOR
IND_RES = IND.run(CTX, cfg, GDFS, NAMES, ANNUAL, cache, lang="en", force=REFRESH)
ZONES, IND_SUM = IND_RES["zones"], IND_RES["summary"]
if ZONES is None or not len(ZONES):
    print("⚠️ No candidate zone (Earth Engine unavailable and nothing cached, or no pixel passes the thresholds).")
else:
    v = IND_RES["validation"]
    print(f"{IND_SUM['n']} candidate zones: {IND_SUM['n_major']} major hubs, {IND_SUM['n_energy']} energy sites · "
          f"{M._num(IND_SUM['area_km2'])} km² · {M._pct(IND_SUM['share_national'], 0, False)} of national light · "
          f"growth {M._pct(IND_SUM['growth_total'])}/yr")
    print("OpenStreetMap validation:", (f"{M._pct(v['precision'], 0, False)} of zones contain an OSM industrial area "
          f"within 1 km ({v['osm_areas']} OSM areas in the country, recall {M._pct(v['recall'], 0, False)})")
          if v else "not available (Overpass API unreachable) — run again later.")
    t = ZONES.head(15)
    display(pd.DataFrame({
        "Zone": t["zone_id"], "Location": t["name"], "Type": t["class_label"],
        "Area (km²)": [M._num(x, 1) for x in t["area_km2"]],
        "Share of national light": [M._pct(x, 2, False) for x in t["share_national"]],
        "Growth/yr": [M._pct(x) for x in t["growth"]], "Dynamics": t["dynamics_label"],
        "Non-resid. built": [M._pct(x, 0, False) for x in t["nres_share"]],
        "People/km²": [M._num(x) for x in t["pop_density"]],
        "OSM": ["✔" if o > 0 else "–" for o in t["osm_n"].fillna(0)]}).set_index("Zone"))
```

```python
# --- 10.2 Map: industrial zones over night-time lights -----------------
if ZONES is not None and len(ZONES):
    w, s, e, n = GDFS["ADM0"].total_bounds
    labels = IND.CLASS_LABEL["en"]
    zl = ZONES[["geometry", "class"]].assign(**{
        "Zone": ZONES["zone_id"] + " · " + ZONES["name"], "Type": ZONES["class_label"],
        "Area": [M._num(x, 1) + " km²" for x in ZONES["area_km2"]],
        "Share of national light": [M._pct(x, 2, False) for x in ZONES["share_national"]],
        "Growth": [M._pct(x) + "/yr" for x in ZONES["growth"]],
        "Non-residential built-up": [M._pct(x, 0, False) for x in ZONES["nres_share"]],
        "Confirmed by OSM": ["✔" if o > 0 else "–" for o in ZONES["osm_n"].fillna(0)]})
    mx = float(ZONES["sol"].max())
    if CTX is not None:
        import ipyleaflet, ipywidgets as W
        print("Preparing the industrial zones map…")
        IMG, CAND, INFO = IND.indicator_layers(CTX)
        im = make_map(center=[(s + n) / 2, (w + e) / 2], zoom=6)
        im.addLayer(A.annual_image(CTX, Y1).log10(), NTL_MAP_VIS, "Night-time lights {y}".format(y=Y1), True, 0.55)
        im.addLayer(IMG.select("nres_share").updateMask(IMG.select("built").gte(0.02)),
                    {"min": 0, "max": 1, "palette": ["1b1035", "5b2a86", "b5179e", "f72585"]}, "Non-residential share of built-up (GHSL)", False)
        im.addLayer(IMG.select("s_no2").updateMask(IMG.select("s_no2").gt(0.05)),
                    {"min": 0, "max": 1, "palette": ["2c7bb6", "abd9e9", "fdae61", "d7191c"]}, "NO₂ index (Sentinel-5P)", False, 0.6)
        im.addLayer(IMG.select("score").updateMask(IMG.select("rad").gte(2)),
                    {"min": 0.3, "max": 0.9, "palette": ["313695", "74add1", "fee090", "f46d43", "a50026"]}, "Industrial score (lit pixels)", False)
        im.add_gdf(zl, layer_name="Candidate industrial zones (hover = profile)", zoom_to_layer=False,
                   style={"weight": 1.6, "fillOpacity": 0.45}, hover_style={"weight": 3, "color": "#FFFFFF", "fillOpacity": 0.75},
                   style_callback=lambda f: {"color": ICOL[f["properties"]["class"]], "fillColor": ICOL[f["properties"]["class"]]})
        marks = []
        for _, r in ZONES.head(80).iterrows():
            mk = ipyleaflet.CircleMarker(location=(float(r["lat"]), float(r["lon"])), radius=int(4 + 22 * (r["sol"] / mx) ** 0.5),
                                         color="#FFFFFF", weight=1, fill_color=ICOL[r["class"]], fill_opacity=0.85)
            mk.popup = W.HTML(f"<b>{r['zone_id']} · {r['name']}</b><br>{r['class_label']}<br>"
                              f"Share of national light : {M._pct(r['share_national'], 2, False)}<br>Growth : {M._pct(r['growth'])}/yr")
            marks.append(mk)
        im.add(ipyleaflet.LayerGroup(layers=marks, name="Circles ∝ light emitted (click = profile)"))
        osm = cache.load(cfg.cache_dir, "osm_industrial")
        if osm is not None and len(osm):
            im.add(ipyleaflet.GeoJSON(name="OpenStreetMap industrial areas (validation)", data={"type": "FeatureCollection", "features": [
                {"type": "Feature", "properties": {}, "geometry": {"type": "Point", "coordinates": [float(a), float(b)]}}
                for a, b in zip(osm["lon"], osm["lat"])]},
                point_style={"radius": 2.5, "color": "#22D3EE", "fillColor": "#22D3EE", "fillOpacity": 0.9, "weight": 0}))
        im.add_legend(title="Candidate industrial zones", position="bottomright",
                      legend_dict={**{labels[k]: v.lstrip("#") for k, v in ICOL.items()}, "OSM industrial area": "22D3EE"})
        im.fit_bounds([[s, w], [n, e]])
        print("Layers (icon at the top right): non-residential built-up, NO₂, industrial score. Hover a zone for its profile, click a circle for the summary.")
        display(im)
    else:   # offline: folium on a key-free Esri dark basemap
        import folium
        fm = folium.Map(location=[(s + n) / 2, (w + e) / 2], zoom_start=6, tiles=ESRI_DARK, attr=ESRI_ATTR)
        folium.GeoJson(zl.to_json(), name="Candidate industrial zones (hover = profile)",
                       style_function=lambda f: {"color": ICOL[f["properties"]["class"]], "fillColor": ICOL[f["properties"]["class"]],
                                                 "weight": 1.5, "fillOpacity": 0.5},
                       tooltip=folium.GeoJsonTooltip(fields=[c for c in zl.columns if c not in ("geometry", "class")])).add_to(fm)
        for _, r in ZONES.head(80).iterrows():
            folium.CircleMarker((r["lat"], r["lon"]), radius=4 + 22 * (r["sol"] / mx) ** 0.5, color="#FFFFFF", weight=1,
                                fill=True, fill_color=ICOL[r["class"]], fill_opacity=0.85,
                                tooltip=f"{r['zone_id']} · {r['name']}").add_to(fm)
        folium.LayerControl().add_to(fm)
        display(fm)
```

```python
# --- 10.3 Ranking, industrial signature and close-up views (Sentinel-2) ------------------------
if ZONES is not None and len(ZONES):
    lab_cls = IND.CLASS_LABEL["en"]
    top = ZONES.head(15).iloc[::-1]
    ylab = top["name"] + " · " + top["zone_id"]
    fig = go.Figure([go.Bar(y=ylab[top["class"] == c], x=100 * top.loc[top["class"] == c, "share_national"], orientation="h",
                            name=lab_cls[c], marker_color=col) for c, col in ICOL.items() if (top["class"] == c).any()])
    fig.update_layout(title="The 15 zones emitting the most light", barmode="overlay", height=560,
                      xaxis_title="Share of national light (%)", yaxis={"categoryorder": "array", "categoryarray": list(ylab)},
                      legend={"orientation": "h", "y": 1.06})
    fig.show()
    z = ZONES[ZONES["pop"] > 0].copy()
    z["Non-residential share of built-up (%)"] = 100 * z["nres_share"]
    z["Light per resident"] = z["sol"] / z["pop"]
    fig = px.scatter(z, x="Non-residential share of built-up (%)", y="Light per resident", size="sol", color="class_label",
                     hover_name="name", log_y=True, size_max=38, color_discrete_map={lab_cls[k]: v for k, v in ICOL.items()},
                     labels={"class_label": "Type"},
                     title="Industrial signature: further right and higher = more industrial profile (size = light)")
    fig.update_layout(height=500); fig.show()
    if CTX is not None:
        import ipywidgets as W
        cards = []
        for _, r in ZONES.head(3).iterrows():
            png = IND.zone_closeup(CTX, r.geometry, Y1)
            if png:
                cards.append(W.VBox([W.Image(value=png, format="png", layout=W.Layout(width="330px")),
                                     W.HTML(f"<b>{r['zone_id']} · {r['name']}</b><br><span style='color:#555'>{r['class_label']} · "
                                            f"{M._pct(r['share_national'], 1, False)} of national light</span>")],
                                    layout=W.Layout(margin="0 12px 0 0")))
        if cards:
            show_md(f"**Close-up views of the top three zones** — Sentinel-2 (ESA Copernicus), {Y1}, zone outline in yellow.")
            display(W.HBox(cards))
```

```python
if ZONES is not None and len(ZONES):
    v = IND_RES["validation"]; z0 = ZONES.iloc[0]
    up, down = IND_SUM["n_up"], IND_SUM["n_down"]
    conf = "🟠 medium" if v and v["precision"] >= 0.5 else "🔴 low"
    show_md(f"""**🔎 Key takeaway** — {IND_SUM['n']} candidate zones, including {IND_SUM['n_major']} major hubs, emit
**{M._pct(IND_SUM['share_national'], 0, False)}** of national light over {M._num(IND_SUM['area_km2'])} km².
The largest is **{z0['name']}** ({M._pct(z0['share_national'], 1, False)} of the country's light, {M._pct(z0['growth'])}/yr).
Since {Y0}, {up} zones have been expanding (> +5%/yr) and {down} declining; the light of all zones changes by
{M._pct(IND_SUM['growth_total'])} per year. OpenStreetMap validation: {(M._pct(v['precision'], 0, False) + " of zones confirmed") if v else "not available"}.

*Confidence {conf}* — *candidate* zones: check the largest ones on imagery (10.3) and in the national register of
industrial zones before any statistical use. Results are included in the dashboard ("Industrial zones" tab).""")
```

## 12 · Exporting results for the dashboard

*Static files for a site that runs without Earth Engine*

> 🎯 **Objective** — Write **static** files (simplified GeoJSON, CSV, JSON, PNG) so that the site works without
> Earth Engine: GEE tiles expire after a few hours, a PNG does not.
>
> 🛠️ **Method** — `src/export.py` writes everything under `dashboard/data/<ISO3>/` and updates `dashboard/data/countries.json`
> (several countries can coexist on the same site). The site is **bilingual**: the French version is written to
> `dashboard/fr/` (EN / FR button at the top right); the country flag and a local copy of Plotly are downloaded. The manifest keeps the parameters, versions and availability of
> the datasets: this is what guarantees reproducibility.

```python
MANIFEST = {
    "config": {k: v for k, v in cfg.to_dict().items() if k not in ("country",)},
    "versions": package_versions(),
    "availability": AVAIL.to_dict(orient="records") if AVAIL is not None else [],
    "boundaries": BMETA, "pop_source": cfg.pop_source.upper(),
    "datasets": {k: v["id"] for k, v in DATASETS.items()},
    "references": [
        "Henderson, J. V., Storeygard, A., Weil, D. N. (2012). Measuring Economic Growth from Outer Space. AER 102(2).",
        "Chen, X., Nordhaus, W. D. (2011). Using luminosity data as a proxy for economic statistics. PNAS 108(21).",
        "Elvidge, C. D. et al. (2021). Annual time series of global VIIRS nighttime lights. Remote Sensing 13(5).",
        "Román, M. O. et al. (2018). NASA's Black Marble nighttime lights product suite. RSE 210.",
        "Li, X. et al. (2020). A harmonized global nighttime light dataset 1992–2018. Scientific Data 7.",
        "Gibson, J. et al. (2021). Which night lights data should we use in economics, and where? JDE 149.",
        "Cleveland, R. B. et al. (1990). STL: A seasonal-trend decomposition procedure based on Loess. JOS 6(1).",
    ] + [v["citation"] for v in DATASETS.values() if "citation" in v],
}
OUT = X.export_dashboard(cfg, GDFS, R, ANNUAL, MONTHLY, NAMES, MESSAGES, RECS, M.LIMITS, MANIFEST,
                         RASTERS, cfg.cache_dir / "rasters", synthetic=SYNTHETIC_TEST)
if FLARES is not None:
    FLARES.to_csv(OUT / "flare_candidates.csv", index=False)
(cfg.out / "manifest.json").write_text(json.dumps(X._clean(MANIFEST), ensure_ascii=False, indent=1), encoding="utf-8")
# French version of the site (dashboard/fr/): same figures, texts generated in French
from src import i18n
OUT_FR = i18n.export_other_language("en", cfg, GDFS, R, ANNUAL, MONTHLY, NAMES, MANIFEST, RASTERS,
                                    cfg.cache_dir / "rasters", ROOT, synthetic=SYNTHETIC_TEST)
print("Exported files:", ", ".join(sorted(p.name for p in OUT.iterdir())))
print("French version:", "dashboard/fr/" if OUT_FR else "not generated (src_fr modules missing)")
print("Local preview: python -m http.server -d dashboard 8000  →  http://localhost:8000")
```

## 13 · Publishing the dashboard on GitHub Pages

*One call, a secure token*

> 🎯 **Objective** — Publish the site in a single call: `publish_dashboard(ISO3, REPO_NAME, GITHUB_USER)`.
>
> 🔐 **Token security — read before running**
> 1. Create a **fine-grained personal access token** (GitHub → *Settings* → *Developer settings* → *Fine-grained tokens*),
>    with a **short expiry** (7 to 30 days), restricted to the dashboard repository only.
> 2. Minimum permissions: **Contents** (read/write), **Pages** (read/write); **Administration**
>    only to create the repository; **Workflows** only if `PUSH_SOURCE = True`.
>    *Tip*: a fine-grained token can only target an existing repository — first create the empty public repository on GitHub,
>    or use an "All repositories" token with Administration for the time it takes to create it.
> 3. Storage: **Colab** → 🔑 *Secrets* icon, name `GITHUB_TOKEN`, with notebook access granted; **local** → `.env` file
>    (`GITHUB_TOKEN=...`), already listed in `.gitignore`. Otherwise, the token is requested through masked input.
> 4. The token never appears in the code, outputs, logs, Git history or URLs: publishing
>    goes through the REST API (`Authorization` header), never through a `https://token@github.com` URL, and every error
>    message is redacted.
> 5. **Before sharing this notebook, clear the output of the publishing cell** (*Edit → Clear output*), then
>    **revoke** the token if it is no longer needed.
>
> 🛠️ **What the function does** — checks the token (`GET /user`) → creates the public repository if it does not exist → pushes the site
> to the `gh-pages` branch in a single timestamped commit (country, period, date) → enables Pages → waits for the build to finish →
> displays the URL `https://<user>.github.io/<repository>/`. Re-running updates the site without duplicating the repository.

```python
from src.publish import publish_dashboard, PublishError

if not PUBLISH:
    print("Publishing disabled (PUBLISH = False). Set PUBLISH to True in section 2 to publish.")
elif SYNTHETIC_TEST:
    print("Synthetic test dataset: public publishing refused.")
else:
    try:
        # repository landing page: illustrated README (real satellite images of the country), licence, citation
        from src import readme
        def LANDING(url, owner):
            print("Preparing the repository landing page (Sentinel-2 / VIIRS images)…")
            files = readme.build(cfg.dashboard_dir, "en", url, REPO_NAME, owner, ctx=CTX, out_dir=cfg.out / "landing")
            meta = json.loads((Path(cfg.dashboard_dir) / "data" / cfg.iso3 / "meta.json").read_text(encoding="utf-8"))
            return files, readme.repo_metadata("en", meta, url)
        URL = publish_dashboard(cfg.iso3, REPO_NAME, GITHUB_USER, site_dir=cfg.dashboard_dir,
                                period=(Y0, Y1), project_dir=ROOT if PUSH_SOURCE else None, landing=LANDING)
    except PublishError as e:
        print("❌ Publishing failed:", e)
    else:
        print("English version:", URL, "· French version:", URL.rstrip("/") + "/fr/")
```
