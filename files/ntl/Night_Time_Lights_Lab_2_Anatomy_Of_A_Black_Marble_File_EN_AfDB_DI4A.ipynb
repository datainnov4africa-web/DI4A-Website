## What this notebook is for

Night-time lights are one of the very few data sources available to a national statistical office that are
**free, global, daily and politically neutral**. But between the file NASA publishes and a publishable
indicator lies a step that almost nobody explains properly: **understanding what is inside the file**.

A Black Marble file is not an image. It is a **binder** holding some twenty tables of integers, each with
its own meaning, its own scale and its own "empty" value. Reading that binder wrongly — forgetting a scale
factor, mistaking a fill value for a zero, ignoring a quality flag — produces numbers that *look* right and
are **wrong**.

This notebook takes the file apart piece by piece. By the end, you will be able to answer, with evidence,
eight questions:

## The storyline: one question, asked twenty times

A training notebook that strings together eighteen independent demonstrations is forgotten within a week. This one follows **a single question**, from start to finish:

<div style="background:#E8F5EF;border:1px solid #00704A44;border-left:4px solid #00704A;border-radius:8px;
 padding:13px 17px;margin:10px 0;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
 <div style="color:#00704A;font-weight:700;font-size:13.5px;">&#128273;&nbsp; The question this notebook answers</div>
 <div style="color:#231F20;font-size:13.5px;margin-top:5px;line-height:1.6;"><b style="font-size:15px;">What is the sum of lights of the chosen territory — and how can you be sure that figure is right?</b><br><br>We will answer it for the first time in step 09, very fast and very badly. Then each step will correct one mistake, and the figure will move: sometimes by a few percent, sometimes by a factor of ten. Step 19 lays every answer side by side. <b>None of the mistakes corrected along the way produces an error message</b> — which is exactly what makes them expensive.</div>
</div>
The country being followed is set by the `COUNTRY` parameter in step 02 (**Cote d'Ivoire** by default). Change it, and the whole notebook follows.

## Mini-glossary — worth keeping at hand

<table style="width:100%;border-collapse:collapse;border:none;"><tr><td style="width:25.0000%;vertical-align:top;padding:5px;">
 <div style="background:#F4F7F5;border:1px solid #D5DED9;border-top:3px solid #0E7C86;border-radius:10px;
  padding:13px 15px;height:100%;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
  <div style="color:#00704A;font-weight:700;font-size:14px;margin-bottom:5px;">Radiance</div>
  <div style="color:#231F20;font-size:13px;line-height:1.55;">The light intensity measured by the sensor, in <b>nW·cm⁻²·sr⁻¹</b> (nanowatts per square centimetre per steradian). It is the underlying physical quantity.</div>
 </div></td><td style="width:25.0000%;vertical-align:top;padding:5px;">
 <div style="background:#F4F7F5;border:1px solid #D5DED9;border-top:3px solid #0E7C86;border-radius:10px;
  padding:13px 15px;height:100%;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
  <div style="color:#00704A;font-weight:700;font-size:14px;margin-bottom:5px;">Pixel</div>
  <div style="color:#231F20;font-size:13px;line-height:1.55;">One cell of the image. Here &asymp;&nbsp;<b>500&nbsp;m</b> across at the equator (15 arc-seconds).</div>
 </div></td><td style="width:25.0000%;vertical-align:top;padding:5px;">
 <div style="background:#F4F7F5;border:1px solid #D5DED9;border-top:3px solid #0E7C86;border-radius:10px;
  padding:13px 15px;height:100%;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
  <div style="color:#00704A;font-weight:700;font-size:14px;margin-bottom:5px;">Tile</div>
  <div style="color:#231F20;font-size:13px;line-height:1.55;">A <b>10°&nbsp;×&nbsp;10°</b> square cut out of the globe, identified as <code>hHHvVV</code>. A country spans 1 to 6 tiles.</div>
 </div></td><td style="width:25.0000%;vertical-align:top;padding:5px;">
 <div style="background:#F4F7F5;border:1px solid #D5DED9;border-top:3px solid #0E7C86;border-radius:10px;
  padding:13px 15px;height:100%;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
  <div style="color:#00704A;font-weight:700;font-size:14px;margin-bottom:5px;">HDF5 / <code>.h5</code></div>
  <div style="color:#231F20;font-size:13px;line-height:1.55;">A scientific file format that stores several tables <i>and</i> their descriptions in a single file, like a binder with dividers.</div>
 </div></td></tr><tr><td style="width:25.0000%;vertical-align:top;padding:5px;">
 <div style="background:#F4F7F5;border:1px solid #D5DED9;border-top:3px solid #0E7C86;border-radius:10px;
  padding:13px 15px;height:100%;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
  <div style="color:#00704A;font-weight:700;font-size:14px;margin-bottom:5px;">Dataset</div>
  <div style="color:#231F20;font-size:13px;line-height:1.55;">One table of numbers inside the binder. A Black Marble file holds around twenty of them.</div>
 </div></td><td style="width:25.0000%;vertical-align:top;padding:5px;">
 <div style="background:#F4F7F5;border:1px solid #D5DED9;border-top:3px solid #0E7C86;border-radius:10px;
  padding:13px 15px;height:100%;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
  <div style="color:#00704A;font-weight:700;font-size:14px;margin-bottom:5px;">Attribute</div>
  <div style="color:#231F20;font-size:13px;line-height:1.55;">A piece of metadata attached to a dataset: unit, scale factor, fill value, description…</div>
 </div></td><td style="width:25.0000%;vertical-align:top;padding:5px;">
 <div style="background:#F4F7F5;border:1px solid #D5DED9;border-top:3px solid #0E7C86;border-radius:10px;
  padding:13px 15px;height:100%;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
  <div style="color:#00704A;font-weight:700;font-size:14px;margin-bottom:5px;"><code>_FillValue</code></div>
  <div style="color:#231F20;font-size:13px;line-height:1.55;">The conventional value meaning «&nbsp;<b>no data here</b>&nbsp;». It must be converted to <code>NaN</code>, never treated as a zero.</div>
 </div></td><td style="width:25.0000%;vertical-align:top;padding:5px;">
 <div style="background:#F4F7F5;border:1px solid #D5DED9;border-top:3px solid #0E7C86;border-radius:10px;
  padding:13px 15px;height:100%;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
  <div style="color:#00704A;font-weight:700;font-size:14px;margin-bottom:5px;">Quality flag</div>
  <div style="color:#231F20;font-size:13px;line-height:1.55;">A small integer code attached to each pixel telling you how far it can be trusted.</div>
 </div></td></tr><tr><td style="width:25.0000%;vertical-align:top;padding:5px;">
 <div style="background:#F4F7F5;border:1px solid #D5DED9;border-top:3px solid #0E7C86;border-radius:10px;
  padding:13px 15px;height:100%;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
  <div style="color:#00704A;font-weight:700;font-size:14px;margin-bottom:5px;">BRDF</div>
  <div style="color:#231F20;font-size:13px;line-height:1.55;"><i>Bidirectional Reflectance Distribution Function</i> — the correction that makes measurements comparable despite view angle and moonlight.</div>
 </div></td><td style="width:25.0000%;vertical-align:top;padding:5px;">
 <div style="background:#F4F7F5;border:1px solid #D5DED9;border-top:3px solid #0E7C86;border-radius:10px;
  padding:13px 15px;height:100%;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
  <div style="color:#00704A;font-weight:700;font-size:14px;margin-bottom:5px;">EPSG:4326</div>
  <div style="color:#231F20;font-size:13px;line-height:1.55;">The coordinate system in degrees (longitude / latitude), the one used by GPS.</div>
 </div></td><td style="width:25.0000%;vertical-align:top;padding:5px;">
 <div style="background:#F4F7F5;border:1px solid #D5DED9;border-top:3px solid #0E7C86;border-radius:10px;
  padding:13px 15px;height:100%;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
  <div style="color:#00704A;font-weight:700;font-size:14px;margin-bottom:5px;">Composite</div>
  <div style="color:#231F20;font-size:13px;line-height:1.55;">An image obtained by combining several nights to remove clouds and noise (monthly&nbsp;=&nbsp;A3, annual&nbsp;=&nbsp;A4).</div>

 <div style="color:#231F20;font-size:14px;margin-top:4px;">Knowing where we are running, and installing the strict minimum</div>

**What this step does.** The notebook first identifies the execution environment
(Colab, Kaggle, local Jupyter, VS&nbsp;Code), then checks the libraries.

The distinction that matters is between the **required core** and the **optional extras**:

- The **core** — `numpy`, `pandas`, `matplotlib`, `h5py` — is enough to run **the whole** of this
  notebook. All four are already installed on Colab and Kaggle.
- The **extras** — `geopandas`, `folium`, `rasterio`, `requests` — add convenience (administrative
  boundaries, interactive map, GeoTIFF export, downloading). If one is missing or refuses to install, the
  notebook says so and **carries on**: no step ever blocks.

````python
# =============================================================================
#  STEP 01 — ENVIRONMENT AND DEPENDENCIES
# =============================================================================
import importlib.util
import os
import platform
import subprocess
import sys
from pathlib import Path

# --- Setting: set to False to forbid any automatic installation ---------------
AUTO_INSTALL = True


def detect_environment() -> str:
    """Identify the platform the notebook is running on."""
    if "google.colab" in sys.modules:
        return "Google Colab"
    if os.environ.get("KAGGLE_KERNEL_RUN_TYPE") or Path("/kaggle/input").exists():
        return "Kaggle"
    if os.environ.get("VSCODE_PID") or os.environ.get("VSCODE_CWD"):
        return "VS Code"
    if "ipykernel" in sys.modules:
        return "Local Jupyter"
    return "Python (outside a notebook)"


ENVIRONMENT = detect_environment()

# Import name  →  pip package name
CORE = {"numpy": "numpy", "pandas": "pandas", "matplotlib": "matplotlib", "h5py": "h5py"}
EXTRAS = {"requests": "requests", "shapely": "shapely",
          "geopandas": "geopandas", "folium": "folium", "rasterio": "rasterio"}


def pip_install(packages) -> bool:
    """Install a list of packages with pip. Returns True if everything went well."""
    try:
        subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", *packages],
                              stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
        return True
    except Exception:
        return False


def check_packages(table, label_text, required):
    """Check that the modules are present, try to install the missing ones."""
    missing = [pip for imp, pip in table.items() if importlib.util.find_spec(imp) is None]
    if missing and AUTO_INSTALL:
        print(f"  installing {', '.join(missing)} ...")
        pip_install(missing)
        missing = [pip for imp, pip in table.items() if importlib.util.find_spec(imp) is None]

    available = {}
    for imp in table:
        is_present = importlib.util.find_spec(imp) is not None
        available[imp] = is_present
        if is_present:
            try:
                version = getattr(__import__(imp), "__version__", "?")
            except Exception:
                version = "?"
            print(f"  [ok]      {imp:<12} {version}")
        else:
            print(f"  [{'MISSING' if required else 'absent '}]  {imp:<12} ({label_text})")
    return available


print("=" * 74)
print(f"  ENVIRONMENT       : {ENVIRONMENT}")
print(f"  PYTHON            : {platform.python_version()}  ({platform.system()} {platform.machine()})")
print(f"  WORKING DIRECTORY : {Path.cwd()}")
print("=" * 74)

print("\nRequired core")
core_state = check_packages(CORE, "essential", required=True)

print("\nOptional extras")
AVAILABLE = check_packages(EXTRAS, "convenience", required=False)

if not all(core_state.values()):
    raise SystemExit(
        "\nA core library is missing. Install it by hand, then re-run:\n"
        "    pip install numpy pandas matplotlib h5py")

print("\n" + "=" * 74)
print("  Core complete — the notebook can run from end to end.")
if not all(AVAILABLE.values()):
    missing_extras = [k for k, v in AVAILABLE.items() if not v]
    print(f"  Missing extras: {', '.join(missing_extras)} — the steps concerned will simply")
    print("  be flagged and worked around, without interrupting the notebook.")
print("=" * 74)
````

**What this step does.** A single cell gathers every choice you will have to make.
Nowhere else does the notebook contain one: you edit this cell, you hit *Run all*, and all
20 steps, every table and every map adapt to the country you picked.

Four decisions really matter:

1. **`COUNTRY`** — an ISO 3166-1 alpha-3 code (`SEN`, `CIV`, `MAR`, `KEN`, `TZA`, `COD`…). It
   determines which tiles to download, which border is used for clipping, which cities appear on
   the maps, and even what the demonstration dataset contains. **It is the one indispensable
   setting.**
2. **`EARTHDATA_TOKEN_VALUE`** — your NASA Earthdata token, if you want to work on the **real**
   data. Without it, the notebook falls back on a structurally faithful demonstration dataset.
3. **`QA_MAX`** and **`MASK_WATER`** — the two methodological trade-offs that will change your
   numbers, and that you will have to justify in your documentation.
4. **`DOWNSAMPLE`** — your memory budget. `1` = full resolution, `4` = sixteen times fewer
   pixels, useful on a modest machine.

 <div style="color:#231F20;font-size:13.5px;margin-top:5px;line-height:1.6;">Black Marble data is free, but the archive requires authentication. <b>1.</b> Create an account on <code>urs.earthdata.nasa.gov</code> (free, no affiliation required). <b>2.</b> Log in, open the <i>Generate Token</i> tab, click <i>Generate Token</i>. <b>3.</b> Copy the long string you get — it is valid for 60 days.<br><br>The notebook looks for that token in <b>four</b> places, in this order: the <code>EARTHDATA_TOKEN_VALUE</code> variable below, the <code>EARTHDATA_TOKEN</code> environment variable, the file <code>~/.earthdata_token</code>, and then — if <code>ASK_FOR_TOKEN = True</code> — a masked prompt on screen.</div>

 <div style="color:#231F20;font-size:13.5px;margin-top:5px;line-height:1.6;">It is worth a password: anyone who gets hold of it acts on the NASA archive in your name. <b>Do not paste it into the cell</b> if you intend to share this notebook, push it to GitHub, or project it in a room.  <div style="color:#231F20;font-size:13.5px;margin-top:5px;line-height:1.6;">The notebook embeds the borders of 88 countries across Africa (including Cabo Verde, São Tomé and Príncipe and the Seychelles), the Mediterranean rim and the Middle East, simplified from <b>Natural Earth</b> (public domain, 1:50 M scale), together with 873 reference cities and their populations.
````python
# =============================================================================
#  STEP 02 — SETTINGS   ◄◄◄  THE ONLY CELL TO EDIT
# =============================================================================

# -----------------------------------------------------------------------------
#  1. WHICH COUNTRY ARE YOU FOLLOWING?
# -----------------------------------------------------------------------------
#     An ISO 3166-1 alpha-3 code. The whole notebook adapts to it: border,
#     cities, tiles to download, clipping, maps and indicators.
#     Examples: "CIV" Cote d'Ivoire · "SEN" Senegal · "SDN" Sudan
#               "NGA" Nigeria · "MAR" Morocco · "KEN" Kenya · "ZAF" South Africa
#               "COD" DR Congo · "EGY" Egypt · "TZA" Tanzania · "ETH" Ethiopia
#     88 countries are available; an unknown code stops the notebook and lists them.
# -----------------------------------------------------------------------------
COUNTRY = "CIV"

# -----------------------------------------------------------------------------
#  2. WHERE IS YOUR DATA?
# -----------------------------------------------------------------------------
#     None  -> automatic detection (working directory, /kaggle/input, mounted Drive...)
#     Else  -> explicit path to the folder holding your .h5 files
#
#     Windows  : DATA_DIR = r"C:\Users\moi\blackmarble_data"
#     macOS    : DATA_DIR = "/Users/moi/blackmarble_data"
#     Linux    : DATA_DIR = "/home/moi/blackmarble_data"
#     Colab    : DATA_DIR = "/content/drive/MyDrive/blackmarble_data"
#     Kaggle   : DATA_DIR = "/kaggle/input/mon-dataset-ntl"
# -----------------------------------------------------------------------------
DATA_DIR = None

# -----------------------------------------------------------------------------
#  3. YOUR EARTHDATA TOKEN  (to download the real data)
# -----------------------------------------------------------------------------
#     Black Marble products are free, but the archive requires a NASA
#     Earthdata account and a personal token:
#         1. create an account at  https://urs.earthdata.nasa.gov
#         2. open the "Generate Token" tab, then copy the long string you get
#         3. paste it below, OR better: see the note that follows step 02.
#
#     Left empty, the notebook still works: it falls back on its demonstration
#     set. A token is never shared and never committed.
# -----------------------------------------------------------------------------
EARTHDATA_TOKEN_VALUE = ""      # or leave empty and use EARTHDATA_TOKEN / ~/.earthdata_token
ASK_FOR_TOKEN = True            # prompt for the token if none is found elsewhere
DOWNLOAD = True                 # False = never reach out to the network
WANTED_YEARS = (2021, 2022, 2023)

# -----------------------------------------------------------------------------
#  4. WHICH PRODUCT SHOULD WE EXPLORE FIRST?
#     "VNP46A4" annual  ·  "VNP46A3" monthly  ·  "VNP46A2" daily
# -----------------------------------------------------------------------------
MAIN_PRODUCT = "VNP46A4"

# -----------------------------------------------------------------------------
#  5. QUALITY FILTERING
#     QA_MAX : "auto" (recommended) keeps only the codes that the file's own
#              legend labels as good / high quality. For the monthly and annual
#              composites that is code 0 (1 = poor quality, 3 nights or fewer;
#              2 = gap-filled from history). For the daily product it depends on
#              the legend written in the file (read at step 11).
#              An integer forces the threshold: keep pixels whose flag is <= QA_MAX.
#     MASK_WATER : set ocean and sea pixels to NaN (codes read from the file's legend)
# -----------------------------------------------------------------------------
QA_MAX = "auto"
MASK_WATER = True

# -----------------------------------------------------------------------------
#  6. DISPLAY
#     DOWNSAMPLE : 1 = full resolution ; 4 = mean of 4x4 blocks
#                  (affects display only, never the computations)
# -----------------------------------------------------------------------------
DOWNSAMPLE = 1
FIG_DPI = 115

# -----------------------------------------------------------------------------
#  7. DEMONSTRATION MODE
#     ALLOW_DEMO : if no real data is found or downloaded, build .h5 files
#                  that are structurally faithful, so that the notebook can run
#                  from end to end.
#     DEMO_PX    : side, in pixels, of the MAIN demonstration tile — the one
#                  holding your largest city. Real NASA tiles are 2400 x 2400;
#                  1200 is enough to keep the zoom on an urban area readable
#                  while keeping the notebook fast. Drop to 600 on a modest
#                  machine.
# -----------------------------------------------------------------------------
ALLOW_DEMO = True
DEMO_PX = 1200

# =============================================================================
print("Settings recorded.")
print(f"  COUNTRY                 = {COUNTRY}")
print(f"  DATA_DIR             = {DATA_DIR if DATA_DIR else 'automatic detection'}")
print(f"  MAIN_PRODUCT    = {MAIN_PRODUCT}")
print(f"  QA_MAX               = {QA_MAX}   (\"auto\" = the codes the file labels as good quality)")
print(f"  MASK_WATER          = {MASK_WATER}")
print(f"  DOWNLOAD          = {DOWNLOAD}"
      f"   (token given in the cell: {'yes' if EARTHDATA_TOKEN_VALUE else 'no'})")
````

 padding:13px 17px;margin:10px 0;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
 <div style="color:#00704A;font-weight:700;font-size:13.5px;">&#128161;&nbsp; Why <code>QA_MAX = "auto"</code>?</div>
 <div style="color:#231F20;font-size:13.5px;margin-top:5px;line-height:1.6;">The meaning of a quality code <b>depends on the product</b>. In the monthly and annual composites, code <code>0</code> is good quality, code <code>1</code> is <i>poor quality</i> (3 nights or fewer behind the pixel) and code <code>2</code> is a value gap-filled from history. In the daily product the coding is different, and it has changed between collections. Rather than hard-coding one table, <code>"auto"</code> reads the legend written in the file and keeps only the codes it labels as good quality — code <code>0</code> for the composites. Forcing <code>QA_MAX&nbsp;=&nbsp;1</code> on a composite gives fuller coverage, at the price of pixels resting on almost nothing. Whatever you choose, <b>document it</b> in your metadata.</div>

**What this step does.** Four cells, in this order:

1. **The toolbox** — the constants of the Black Marble grid, the functions that know how to find a group
   inside an HDF5 file and read an attribute cleanly, and the visual identity of the figures.
2. **The safety net** — a generator of demonstration `.h5` files, triggered only if no real data is found.
3. **The inventory** — the notebook walks the data folder and catalogues everything it found there. That
   catalogue drives the whole rest of the notebook.
4. **A first look** — a night map of your area, with no filtering at all, to see what we are talking
   about before taking it apart.

<div style="background:#E8F5EF;border:1px solid #00704A44;border-left:4px solid #00704A;border-radius:8px;
 padding:13px 17px;margin:10px 0;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
 <div style="color:#00704A;font-weight:700;font-size:13.5px;">&#128273;&nbsp; The naming convention used here</div>
 <div style="color:#231F20;font-size:13.5px;margin-top:5px;line-height:1.6;">The functions in this notebook carry explicit, spelled-out names (<code>read_attribute</code>, <code>find_data_group</code>, <code>decode_filename</code>…) rather than abbreviations. This is deliberately more verbose: the code has to be readable aloud during a training session, and picked up afterwards by a colleague who was not in the room.</div>
</div>

````python
# =============================================================================
#  STEP 03 a — TOOLBOX
# =============================================================================
import json
import re
import warnings

import h5py
import matplotlib as mpl
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from matplotlib.colors import LinearSegmentedColormap, ListedColormap, LogNorm

warnings.filterwarnings("ignore")

# -----------------------------------------------------------------------------
#  1. VISUAL IDENTITY  (palette taken from the Bank's style guide)
# -----------------------------------------------------------------------------
COLOURS = dict(
    green="#00A86A", dark_green="#00704A", forest="#00553A", gold="#F5C242",
    ochre="#D49A00", teal="#0E7C86", terra="#C4621D", brick="#B83B2E",
    ink="#231F20", slate="#5E6964", mist="#F4F7F5", mint="#E8F5EF", sage="#D5DED9",
)

# "Night seen from orbit" palette: black → green → gold → cream
CMAP_NIGHT = LinearSegmentedColormap.from_list(
    "afdb_night", ["#04120C", "#00553A", "#00704A", "#00A86A", "#F5C242", "#FFF6DF"])
# Light sequential palette, for maps on a white background
CMAP_GREEN = LinearSegmentedColormap.from_list(
    "afdb_green", ["#E8F5EF", "#9ED9C0", "#57BD92", "#10A06A", "#00704A", "#00553A"])

_fonts = {f.name for f in mpl.font_manager.fontManager.ttflist}
FONT = "Calibri" if "Calibri" in _fonts else "DejaVu Sans"

mpl.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "figure.dpi": FIG_DPI, "font.family": FONT, "font.size": 10.5,
    "axes.titlesize": 12.5, "axes.titleweight": "bold", "axes.titlecolor": COLOURS["forest"],
    "axes.labelcolor": COLOURS["slate"], "axes.edgecolor": COLOURS["sage"],
    "xtick.color": COLOURS["slate"], "ytick.color": COLOURS["slate"],
    "axes.grid": False, "axes.spines.top": False, "axes.spines.right": False,
    "legend.frameon": False, "figure.autolayout": False,
})


def figure_title(fig, title, subtitle="", source=""):
    """Apply the standard institutional framing to a figure."""
    fig.text(0.012, 0.985, title, ha="left", va="top", fontsize=13.5,
             fontweight="bold", color=COLOURS["forest"])
    if subtitle:
        fig.text(0.012, 0.938, subtitle, ha="left", va="top", fontsize=10,
                 color=COLOURS["slate"])
    if source:
        fig.text(0.012, 0.004, source, ha="left", va="bottom", fontsize=7.2,
                 style="italic", color=COLOURS["slate"])


# -----------------------------------------------------------------------------
#  2. THE BLACK MARBLE GRID
# -----------------------------------------------------------------------------
#  The globe is cut into 10° × 10° tiles, indexed h00..h35 (west → east)
#  and v00..v17 (north → south). Each tile holds 2400 × 2400 pixels, i.e. a
#  resolution of 15 arc-seconds (~ 463 m at the equator).
# -----------------------------------------------------------------------------
TILE_DEGREES = 10.0
TILE_PIXELS = 2400
RESOLUTION_ARCSEC = 15.0

# The name of the HDF5 group depends on the product
GROUP_BY_PRODUCT = {
    "VNP46A1": "HDFEOS/GRIDS/VNP_Grid_DNB/Data Fields",
    "VNP46A2": "HDFEOS/GRIDS/VNP_Grid_DNB/Data Fields",
    "VNP46A3": "HDFEOS/GRIDS/VIIRS_Grid_DNB_2d/Data Fields",
    "VNP46A4": "HDFEOS/GRIDS/VIIRS_Grid_DNB_2d/Data Fields",
}
FAMILY = {"VNP46A1": "daily (raw)", "VNP46A2": "daily",
          "VNP46A3": "monthly", "VNP46A4": "annual"}


def tile_numbers(tile: str):
    """'h17v08' → (17, 8)"""
    m = re.search(r"h(\d{2})v(\d{2})", tile)
    if not m:
        raise ValueError(f"Unreadable tile identifier: {tile!r}")
    return int(m.group(1)), int(m.group(2))


def tile_extent(tile: str):
    """'h17v08' → (west, east, south, north) in decimal degrees."""
    h, v = tile_numbers(tile)
    west = -180.0 + h * TILE_DEGREES
    north = 90.0 - v * TILE_DEGREES
    return (west, west + TILE_DEGREES, north - TILE_DEGREES, north)


def tiles_covering(west, east, south, north):
    """List the tiles needed to cover a geographic extent."""
    h0, h1 = int((west + 180) // 10), int((east + 180 - 1e-9) // 10)   # 1e-9: no empty tile when an edge falls on 10°
    v0, v1 = int((90 - north) // 10), int((90 - south - 1e-9) // 10)
    return [f"h{h:02d}v{v:02d}" for v in range(v0, v1 + 1) for h in range(h0, h1 + 1)]


# -----------------------------------------------------------------------------
#  3. READING AN HDF5 FILE WITHOUT GETTING IT WRONG
# -----------------------------------------------------------------------------
def find_data_group(file_: h5py.File) -> str:
    """Find the 'Data Fields' group, whatever the product.

    We never hard-code the path: we walk the tree and keep the first group
    whose name ends with 'Data Fields'.
    """
    found = []
    file_.visititems(
        lambda name, obj: found.append(name)
        if isinstance(obj, h5py.Group) and name.endswith("Data Fields") else None)
    if not found:
        raise KeyError("No 'Data Fields' group: this file is not a VNP46 product.")
    return found[0]


def read_attribute(obj, name, default=None):
    """Read an HDF5 attribute and convert it to a plain Python type.

    HDF5 attributes come back in very different shapes: bytes, a one-element
    numpy array, a multi-element array... This function brings everything back
    to a string, a number, or a list.
    """
    if name not in obj.attrs:
        return default
    value = obj.attrs[name]
    if isinstance(value, (bytes, bytearray, np.bytes_)):
        return value.decode("utf-8", "replace").strip()
    if isinstance(value, str):
        return value.strip()
    flat = np.ravel(value)
    if flat.size == 0:
        return default
    if flat.size == 1:
        element = flat[0]
        if isinstance(element, (bytes, bytearray, np.bytes_)):
            return element.decode("utf-8", "replace").strip()
        return element.item()
    return flat.tolist()


def all_attributes(obj) -> dict:
    """Return every attribute of an HDF5 object, properly converted."""
    return {name: read_attribute(obj, name) for name in obj.attrs}


# Order of preference for the main radiance layer
RADIANCE_LAYERS = [
    "DNB_BRDF-Corrected_NTL",              # A2: the night's own retrieval, the one the quality flag describes
    "Gap_Filled_DNB_BRDF-Corrected_NTL",   # A2: gap-filled version (older values fill the holes)
    "AllAngle_Composite_Snow_Free",        # A3 / A4: the usual reference
    "NearNadir_Composite_Snow_Free",       # A3 / A4: near-nadir view
    "OffNadir_Composite_Snow_Free",
    "AllAngle_Composite_Snow_Covered",
]


def pick_radiance_layer(group: h5py.Group) -> str:
    """Pick the main radiance layer present in the file."""
    present = set(group.keys())
    for name in RADIANCE_LAYERS:
        if name in present:
            return name
    # Fallback: any 2-D layer whose unit is a radiance
    for name, obj in group.items():
        if isinstance(obj, h5py.Dataset) and obj.ndim == 2:
            if "nW" in str(read_attribute(obj, "units", "")):
                return name
    raise KeyError("No identifiable radiance layer in this file.")


# -----------------------------------------------------------------------------
#  4. DECODING A FILENAME
# -----------------------------------------------------------------------------
NAME_PATTERN = re.compile(
    r"^(?P<product>VNP46A[1-4])"              # product
    r"\.A(?P<year>\d{4})(?P<day>\d{3})"       # A + year + Julian day
    r"\.(?P<tile>h\d{2}v\d{2})"               # tile
    r"\.(?P<version>\d{3})"                   # collection version
    r"\.(?P<timestamp>[^.]+)"                 # production timestamp
    r"\.h5$", re.IGNORECASE)


def decode_filename(name: str):
    """Break a VNP46 filename down into its constituent parts."""
    m = NAME_PATTERN.match(str(name))
    if not m:
        return None
    d = m.groupdict()
    d["year"] = int(d["year"])
    d["day"] = int(d["day"])
    d["date"] = pd.Timestamp(d["year"], 1, 1) + pd.Timedelta(days=d["day"] - 1)
    d["family"] = FAMILY.get(d["product"].upper(), "?")
    d["extent"] = tile_extent(d["tile"])
    return d


print(f"Toolbox loaded  ·  figure font: {FONT}")
print(f"Black Marble grid : tiles of {TILE_DEGREES:.0f}° × {TILE_DEGREES:.0f}°, "
      f"{TILE_PIXELS} x {TILE_PIXELS} pixels, {RESOLUTION_ARCSEC:.0f} arc-seconds")
````

<div style="background:#E8F5EF;border:1px solid #00704A44;border-left:4px solid #00704A;border-radius:8px;
 padding:13px 17px;margin:10px 0;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
 <div style="color:#00704A;font-weight:700;font-size:13.5px;">&#128161;&nbsp; The two functions worth remembering from this cell</div>
 <div style="color:#231F20;font-size:13.5px;margin-top:5px;line-height:1.6;"><code>find_data_group(f)</code> and <code>read_attribute(obj, name)</code>. They are what makes your code robust: the group path changes between the daily product and the composites, and HDF5 attributes arrive sometimes as bytes, sometimes as a one-element array. Hard-coding those two things is the most frequent cause of scripts that «&nbsp;worked last year&nbsp;» and break on another product.</div>
</div>

in plain text. Two lines per country, in a compact format you can read with the naked eye.

Why embed data rather than download it? Because a training notebook has to run where the Wi-Fi
does not. The encoding chosen — one point per `lon,lat`, rings separated by `|` — holds the whole

#### What the chosen country implies

The next cell decodes that data and derives from it **everything the notebook needs in order to be
generic**:

| Function | What it gives you |
|---|---|
| `country_mask(LON, LAT)` | for each pixel, does it fall inside the country? |
| `land_mask(LON, LAT)` | land or sea, all countries together — used by the demonstration generator |
| `country_tiles()` | the Black Marble tiles to download in order to cover the country |
| `country_cities()` | the landmark cities, from the most to the least populous |
| `draw_country_outline(ax)` | the border, laid over any map |
| `log_storyline(...)` | the storyline counter, which will track the sum of lights |

The point-in-polygon test is done by `matplotlib.path.Path`, which you already have installed.
**No GIS library is required** — no `geopandas`, no `rasterio`, no `GDAL`: that is exactly what
lets this notebook start in thirty seconds on Colab as well as on Kaggle.

````python
# =============================================================================
#  STEP 03 a-bis (2/2) — GEOGRAPHY OF THE CHOSEN COUNTRY, AND THE STORYLINE
# =============================================================================
from matplotlib.path import Path as _Path


# --- 1. Decoding the embedded data -------------------------------------------
def _decode_rings(text):
    anneaux = []
    for morceau in text.split("|"):
        points = [p.split(",") for p in morceau.split(" ") if p]
        anneaux.append(np.array([(float(x), float(y)) for x, y in points]))
    return anneaux


def _decode_cities(text):
    cities = []
    for morceau in text.split(";"):
        if not morceau:
            continue
        name_, reste = morceau.split("@")
        lon, lat, pop = reste.split(",")
        cities.append((name_, float(lon), float(lat), int(pop)))
    return cities


COUNTRY_NAMES = dict(NAMES_SRC)
COUNTRY_OUTLINES = {code: _decode_rings(t) for code, t in BORDERS_SRC.items()}
COUNTRY_CITIES = {code: _decode_cities(t) for code, t in CITIES_SRC.items()}

# Bounding box (west, east, south, north) of each country, from its outline.
COUNTRY_EXTENTS = {}
for _code, _anneaux in COUNTRY_OUTLINES.items():
    _xs = np.concatenate([a[:, 0] for a in _anneaux])
    _ys = np.concatenate([a[:, 1] for a in _anneaux])
    COUNTRY_EXTENTS[_code] = (float(_xs.min()), float(_xs.max()),
                            float(_ys.min()), float(_ys.max()))

# matplotlib paths, built once, so that the point-in-polygon tests stay fast
# even on a 2400 × 2400 pixel grid.
_PATHS = {c: [_Path(a) for a in anneaux] for c, anneaux in COUNTRY_OUTLINES.items()}


# A few ISO codes that Natural Earth spells differently.
ALIAS_PAYS = {"SSD": "SDS", "PSE": "PSX",
              # territories folded into a country (see the conventions in the data cell)
              "ESH": "MAR", "SAH": "MAR", "SOL": "SOM", "SOMALILAND": "SOM", "CYN": "CYP"}


def check_country(code):
    """Stop the notebook at once, and legibly, if the country code is unknown."""
    code = str(code).strip().upper()
    code = ALIAS_PAYS.get(code, code)
    if code in COUNTRY_OUTLINES:
        return code
    catalogue = sorted((NAMES_SRC.get(c, c), c) for c in COUNTRY_OUTLINES)
    rows, tampon = [], []
    for name_, c in catalogue:
        tampon.append(f"{c} {name_[:20]}")
        if len(tampon) == 3:
            rows.append("   " + "   ".join(f"{x:<26}" for x in tampon))
            tampon = []
    if tampon:
        rows.append("   " + "   ".join(f"{x:<26}" for x in tampon))
    raise SystemExit(
        f"\nUnknown country code: '{code}'. Use an ISO 3166-1 alpha-3 code.\n"
        f"The {len(COUNTRY_OUTLINES)} countries available:\n" + "\n".join(rows)
        + "\n\nFix COUNTRY in step 02, then re-run from there.")


COUNTRY = check_country(COUNTRY)
COUNTRY_NAME = COUNTRY_NAMES.get(COUNTRY, COUNTRY)
COUNTRY_EXTENT = COUNTRY_EXTENTS[COUNTRY]


# --- 2. Geographic membership ------------------------------------------------
def country_mask(LON, LAT, code=None):
    """Boolean mask: True for the pixels that fall inside the country.

    A point-in-polygon test, which matplotlib performs with no GIS library at
    all — and that is what keeps this notebook portable.
    """
    code = COUNTRY if code is None else code
    LON = np.asarray(LON)
    points = np.column_stack([LON.ravel(), np.asarray(LAT).ravel()])
    dedans = np.zeros(points.shape[0], dtype=bool)
    for path_ in _PATHS.get(code, []):
        dedans |= path_.contains_points(points)
    return dedans.reshape(LON.shape)


def land_mask(LON, LAT):
    """Boolean mask: True over dry land, all countries taken together.

    The demonstration generator uses it to place oceans and coastlines in the
    right spot, whichever country has been chosen.
    """
    LON = np.asarray(LON)
    o, e = float(LON.min()), float(LON.max())
    s, n = float(np.asarray(LAT).min()), float(np.asarray(LAT).max())
    points = np.column_stack([LON.ravel(), np.asarray(LAT).ravel()])
    land = np.zeros(points.shape[0], dtype=bool)
    for code, (po, pe, ps, pn) in COUNTRY_EXTENTS.items():
        if pe < o or po > e or pn < s or ps > n:      # country outside the frame
            continue
        for path_ in _PATHS[code]:
            land |= path_.contains_points(points)
    return land.reshape(LON.shape)


def coordinate_grid(extent, forme):
    """LON / LAT grids (pixel centres) for a given extent and array shape."""
    west, east, south, north = extent
    rows, columns = forme
    lon = west + (np.arange(columns) + 0.5) * (east - west) / columns
    lat = north - (np.arange(rows) + 0.5) * (north - south) / rows
    return np.meshgrid(lon, lat)


def draw_country_outline(ax, code=None, colour="white", lw=1.5, alpha=0.95, label=None):
    """Lay the national border over a map that has already been drawn."""
    code = COUNTRY if code is None else code
    for i, anneau in enumerate(COUNTRY_OUTLINES.get(code, [])):
        ax.plot(anneau[:, 0], anneau[:, 1], color=colour, lw=lw, alpha=alpha,
                zorder=6, solid_joinstyle="round",
                label=label if i == 0 else None)


def city_weight(population):
    """Population -> relative light weight, compressed as in reality.

    A city ten times more populous does not shine ten times brighter: the 0.45
    exponent reproduces that compression. Used both by the demonstration
    generator and for the size of the circles on the maps.
    """
    return float(np.clip((population / 3.0e6) ** 0.45, 0.07, 1.25))


def country_cities(code=None, number=8):
    """The country's main cities, most populous first."""
    code = COUNTRY if code is None else code
    return COUNTRY_CITIES.get(code, [])[:number]


def cities_within(extent, number=40):
    """Every reference city that falls inside a given extent."""
    west, east, south, north = extent
    retenues = []
    for liste in COUNTRY_CITIES.values():
        for name_, lon, lat, pop in liste:
            if west <= lon <= east and south <= lat <= north:
                retenues.append((name_, lon, lat, pop))
    return sorted(retenues, key=lambda v: -v[3])[:number]


def country_tiles(code=None):
    """The Black Marble tiles needed to cover the country."""
    code = COUNTRY if code is None else code
    return tiles_covering(*COUNTRY_EXTENTS[code])


COUNTRY_TILES = country_tiles()


# --- 3. The storyline: one question, followed from end to end ----------------
STORYLINE = []


def log_storyline(step, label_, total_, pixels, perimeter, comment=""):
    STORYLINE.append({"step": step, "treatment": label_,
                      "sum of lights": float(total_), "pixels kept": int(pixels),
                      "perimeter": perimeter, "comment": comment})
    print(("   [STORYLINE] step %02d · %-34s SoL = %15.0f"
           % (step, label_, total_)).replace(",", " "))


def storyline_table():
    if not STORYLINE:
        return pd.DataFrame()
    t = pd.DataFrame(STORYLINE)
    changes = [""]
    for i in range(1, len(t)):
        before, after = t["sum of lights"].iloc[i - 1], t["sum of lights"].iloc[i]
        changes.append("%+.1f %%" % (100 * (after - before) / before) if before else "—")
    t["change"] = changes
    t["sum of lights"] = t["sum of lights"].map(lambda v: f"{v:,.0f}".replace(",", " "))
    t["pixels kept"] = t["pixels kept"].map(lambda v: f"{v:,}".replace(",", " "))
    return t.set_index("step")


# --- 4. What the chosen country implies --------------------------------------
_villes = country_cities(number=5)
print("=" * 74)
print(f"  COUNTRY FOLLOWED  : {COUNTRY} — {COUNTRY_NAME}")
print(f"  Bounding box      : longitude [{COUNTRY_EXTENT[0]:+.2f}° , {COUNTRY_EXTENT[1]:+.2f}°]"
      f"   latitude [{COUNTRY_EXTENT[2]:+.2f}° , {COUNTRY_EXTENT[3]:+.2f}°]")
print(f"  Tiles to cover    : {', '.join(COUNTRY_TILES)}   ({len(COUNTRY_TILES)} tile(s))")
print(f"  Landmark cities   : {', '.join(v[0] for v in _villes) if _villes else 'none'}")
print(f"  Storyline question: what is the sum of lights of {COUNTRY_NAME}?")
print("=" * 74)
````

#### One single way of drawing a night map

Night-lights maps are hard to read, and almost always badly made. Three mistakes come back again
and again: a **linear scale** (99 % of the pixels are crushed into the first percent of the
palette, and the map looks black); a colour bar graduated in **`log1p`**, which nobody knows how
to read; and **holes indistinguishable from black**, so that you cannot see that the data is
missing.

The next cell settles all three, once and for all. Every map in the notebook goes through
`night_map()` and `colour_bar()`, so every map shares:

- a **logarithmic scale** with a floor at 0.15 nW·cm⁻²·sr⁻¹ — the black of a rural sky;
- a colour bar graduated in **real physical units**, not in logarithms;
- a **single light grey** for "no data", everywhere in the notebook;
- **legends placed below the map**, which never hide the data;
- **landmark cities** circled and named, with a halo so they stay readable on a dark background.

````python
# =============================================================================
#  STEP 03 a-ter — ONE SINGLE WAY OF DRAWING A NIGHT MAP
# =============================================================================
#  Every map in the notebook goes through these few functions. As a result:
#  one colour scale, one convention for holes, a colour bar graduated in REAL
#  physical units — not in log1p, which nobody can read back — and a scale bar
#  in kilometres on every map.
# -----------------------------------------------------------------------------
import matplotlib.patheffects as patheffects

RADIANCE_FLOOR = 0.15        # nW·cm⁻²·sr⁻¹: the black of a rural sky
NO_DATA_GREY = "#A9B5B0"    # the colour, everywhere, of "no data"


def halo(colour):
    """A legibility outline: dark behind light text, and the other way round.

    Without it a white label vanishes on a light map and a black label vanishes
    on a night map.
    """
    claire = str(colour).lower() in ("white", "#fff", "#ffffff")
    fond = "#0B1A14CC" if claire else "#FFFFFFCC"
    return [patheffects.withStroke(linewidth=2.4, foreground=fond)]


def night_map(ax, radiance, extent, vmin=RADIANCE_FLOOR, vmax=None,
                   interpolation="nearest"):
    """Draw a radiance on a logarithmic scale, with the holes in grey.

    The logarithmic scale is not an affectation: on a linear one, 99 % of the
    pixels are crushed into the first percent of the palette and the map looks
    uniformly black. (Step 15 demonstrates it.)
    """
    if DOWNSAMPLE and DOWNSAMPLE > 1:          # DISPLAY ONLY: block mean of the image,
        f_ = int(DOWNSAMPLE)                    # the data and every figure computed stay untouched
        h_, w_ = (radiance.shape[0] // f_) * f_, (radiance.shape[1] // f_) * f_
        with warnings.catch_warnings():
            warnings.simplefilter("ignore", category=RuntimeWarning)
            radiance = np.nanmean(radiance[:h_, :w_].reshape(h_ // f_, f_, w_ // f_, f_), axis=(1, 3))
    fini = radiance[np.isfinite(radiance) & (radiance > 0)]
    if vmax is None:
        vmax = float(np.nanpercentile(fini, 99.7)) if fini.size else 10 * vmin
    vmax = max(float(vmax), vmin * 20)
    visible = np.where(np.isfinite(radiance), np.clip(radiance, vmin, None), np.nan)
    image = ax.imshow(visible, cmap=CMAP_NIGHT, norm=LogNorm(vmin=vmin, vmax=vmax),
                      extent=extent, origin="upper", aspect="auto",
                      interpolation=interpolation)
    ax.set_facecolor(NO_DATA_GREY)
    return image


def colour_bar(fig, image, ax, label="radiance  (nW·cm⁻²·sr⁻¹)", fraction=0.046):
    """The colour bar: always in the same place, always labelled."""
    barre = fig.colorbar(image, ax=ax, fraction=fraction, pad=0.025)
    barre.set_label(label, fontsize=8.5, color=COLOURS["slate"])
    barre.ax.tick_params(labelsize=7.5)
    barre.outline.set_edgecolor(COLOURS["sage"])
    return barre


def scale_bar(ax, colour="white"):
    """A scale bar in kilometres, at the bottom right of the map.

    Without it a map in degrees says nothing about the real size of what you
    are looking at — and that is precisely what participants have to take away:
    a pixel is 500 m, a tile is 1,100 km.
    """
    o, e = ax.get_xlim()
    s, n = ax.get_ylim()
    km_per_degree = 111.32 * max(np.cos(np.deg2rad((s + n) / 2)), 0.15)
    width_km = abs(e - o) * km_per_degree
    length_km = next((c for c in (1000, 500, 200, 100, 50, 20, 10, 5, 2)
                        if c <= 0.34 * width_km), 1)
    length_deg = length_km / km_per_degree
    x1 = e - 0.055 * (e - o)
    x0 = x1 - length_deg
    y = s + 0.055 * (n - s)
    row_, = ax.plot([x0, x1], [y, y], color=colour, lw=3.0,
                     solid_capstyle="butt", zorder=9)
    row_.set_path_effects(halo(colour))
    text = ax.text((x0 + x1) / 2, y + 0.022 * (n - s), f"{length_km} km",
                    ha="center", va="bottom", fontsize=8, color=colour,
                    fontweight="bold", zorder=9)
    text.set_path_effects(halo(colour))


def legend_below_axes(ax, entries, columns=4, title_=None):
    """A legend placed BELOW the map: it never hides the data."""
    handles = [plt.Rectangle((0, 0), 1, 1, fc=colour, ec="none") for _, colour in entries]
    labels = [name_ for name_, _ in entries]
    return ax.legend(handles, labels, loc="upper center", bbox_to_anchor=(0.5, -0.14),
                     ncol=columns, fontsize=8, frameon=False, title=title_,
                     title_fontsize=8.5, handlelength=1.2, handleheight=1.2,
                     columnspacing=1.3, borderpad=0.2)


def label_cities(ax, cities, colour="white", size=8.0, contour=True):
    """Place the landmark cities on a map, with a halo to stay legible.

    The label flips to the left of the dot when the dot is near the right-hand
    edge: otherwise the name of the country's largest city gets clipped.
    """
    o, e = ax.get_xlim()
    bas, haut = ax.get_ylim()
    offset = 0.015 * abs(e - o)
    # Anti-overlap: two cities too close together on screen get a single
    # label — otherwise the names pile up and nothing stays readable.
    min_gap = 0.055 * abs(e - o)
    placed = []
    for city_name, city_lon, city_lat, population in cities:
        if not (o <= city_lon <= e and bas <= city_lat <= haut):
            continue
        if any(abs(city_lon - x) < min_gap and abs(city_lat - y) < min_gap * 0.55
               for x, y in placed):
            continue
        placed.append((city_lon, city_lat))
        ax.scatter([city_lon], [city_lat], s=12 + 46 * city_weight(population),
                   facecolor="none", edgecolor=colour, lw=1.1, zorder=7)
        to_the_left = (city_lon - o) > 0.78 * (e - o)
        text = ax.text(city_lon + (-offset if to_the_left else offset),
                        city_lat + offset * 0.6, city_name,
                        ha="right" if to_the_left else "left",
                        fontsize=size, color=colour, zorder=8)
        if contour:
            text.set_path_effects(halo(colour))


def categorical_map(ax, codes, extent, legend, couleurs, title_=None):
    """Map of a layer of integer codes (quality, land/water, cloud).

    Returns the (label, colour) list of the classes actually present, for the
    legend — we never show a class that is absent from the file.
    """
    presents = sorted(int(k) for k in np.unique(codes))
    indexed = np.zeros(codes.shape, dtype="int16")
    for i, code in enumerate(presents):
        indexed[codes == code] = i
    ax.imshow(indexed, cmap=ListedColormap([couleurs.get(k, COLOURS["sage"])
                                           for k in presents]),
              extent=extent, origin="upper", aspect="auto", interpolation="nearest",
              vmin=-0.5, vmax=len(presents) - 0.5)
    ax.set_facecolor(NO_DATA_GREY)
    if title_:
        ax.set_title(title_, pad=8, fontsize=10.5)
    return [(f"{k} · {str(legend.get(k, 'undocumented'))[:28]}",
             couleurs.get(k, COLOURS["sage"])) for k in presents]


print("Map conventions loaded: logarithmic scale, grey = no data,")
print("colour bar in nW·cm⁻²·sr⁻¹, scale bar in kilometres.")
````

````python
# =============================================================================
#  STEP 03 b — SAFETY NET: DEMONSTRATION FILE GENERATOR
#  (defined now, called only if no real data could be obtained)
# =============================================================================

# --- Layer specification, modelled on collection 2 (version 002) ------------
# Composites A3/A4: float32 radiances with fill -999.9, uint16 night counts.
# Daily A2: uint16 radiances x 0.1 with fill 65535. Always read the attributes.
SPEC_COMPOSITE = {   # products VNP46A3 (monthly) and VNP46A4 (annual)
    "AllAngle_Composite_Snow_Free":
        dict(type="float32", scale=1.0, fill=-999.9, unite="nW_cm-2_sr-1",
             title_="All angle composite, snow-free, BRDF-corrected NTL"),
    "AllAngle_Composite_Snow_Free_Num":
        dict(type="uint16", scale=1.0, fill=65535, unite="Number of Observations",
             title_="Number of observations, all-angle snow-free composite"),
    "AllAngle_Composite_Snow_Free_Std":
        dict(type="float32", scale=1.0, fill=-999.9, unite="nW_cm-2_sr-1",
             title_="Standard deviation, all-angle snow-free composite"),
    "AllAngle_Composite_Snow_Free_Quality":
        dict(type="uint8", scale=1.0, fill=255, unite="Quality Flag",
             title_="Quality flag, all-angle snow-free composite"),
    "NearNadir_Composite_Snow_Free":
        dict(type="float32", scale=1.0, fill=-999.9, unite="nW_cm-2_sr-1",
             title_="Near-nadir composite, snow-free, BRDF-corrected NTL"),
    "NearNadir_Composite_Snow_Free_Num":
        dict(type="uint16", scale=1.0, fill=65535, unite="Number of Observations",
             title_="Number of observations, near-nadir snow-free composite"),
    "OffNadir_Composite_Snow_Free":
        dict(type="float32", scale=1.0, fill=-999.9, unite="nW_cm-2_sr-1",
             title_="Off-nadir composite, snow-free, BRDF-corrected NTL"),
    "Land_Water_Mask":
        dict(type="uint8", scale=1.0, fill=255, unite="none",
             title_="Land/Water background mask"),
    "DNB_Platform":
        dict(type="uint8", scale=1.0, fill=255, unite="none",
             title_="Source satellite platform"),
}

SPEC_DAILY = {   # product VNP46A2
    "DNB_BRDF-Corrected_NTL":
        dict(type="uint16", scale=0.1, fill=65535, unite="nW_cm-2_sr-1",
             title_="BRDF-corrected nighttime lights radiance"),
    "Gap_Filled_DNB_BRDF-Corrected_NTL":
        dict(type="uint16", scale=0.1, fill=65535, unite="nW_cm-2_sr-1",
             title_="Gap-filled BRDF-corrected nighttime lights radiance"),
    "DNB_Lunar_Irradiance":
        dict(type="uint16", scale=0.1, fill=65535, unite="nW_cm-2_sr-1",
             title_="Lunar irradiance at observation time"),
    "Latest_High_Quality_Retrieval":
        dict(type="uint8", scale=1.0, fill=255, unite="Days",
             title_="Days since latest high-quality retrieval"),
    "Mandatory_Quality_Flag":
        dict(type="uint8", scale=1.0, fill=255, unite="none",
             title_="Mandatory quality flag"),
    "QF_Cloud_Mask":
        dict(type="uint16", scale=1.0, fill=65535, unite="none",
             title_="VIIRS cloud mask quality flags (bit field)"),
    "Snow_Flag":
        dict(type="uint8", scale=1.0, fill=255, unite="none",
             title_="Snow / no-snow flag"),
}

# --- Official flag descriptions, taken from the NASA user guides ------------
DESC_QUALITY_COMPOSITE = "0 = Good quality 1 = Poor quality 2 = Gap filled 255 = Fill value"
DESC_QUALITY_DAILY = ("0 = High-quality, persistent nighttime lights "
                           "1 = High-quality, ephemeral nighttime lights "
                           "2 = Poor-quality, outlier corrected by VIIRS 255 = No retrieval")
DESC_LAND_WATER = ("0 = Shallow ocean 1 = Land 2 = Coastline 3 = Shallow inland water "
                  "4 = Ephemeral water 5 = Deep inland water 6 = Moderate or continental ocean 7 = Deep ocean")


def _neighbourhood(masque, rayon_px):
    """True if at least one true pixel lies in a square of side 2·radius+1.

    Computed with an integral image: the cost does not depend on the radius,
    which is what lets us draw a continental shelf more than a degree wide
    without slowing everything down.
    """
    if rayon_px <= 0:
        return masque.copy()
    integral = np.cumsum(np.cumsum(masque.astype("int32"), axis=0), axis=1)
    integral = np.pad(integral, ((1, 0), (1, 0)))
    h, w = masque.shape
    l0 = np.clip(np.arange(h) - rayon_px, 0, h)
    l1 = np.clip(np.arange(h) + rayon_px + 1, 0, h)
    c0 = np.clip(np.arange(w) - rayon_px, 0, w)
    c1 = np.clip(np.arange(w) + rayon_px + 1, 0, w)
    total_ = (integral[np.ix_(l1, c1)] - integral[np.ix_(l0, c1)]
             - integral[np.ix_(l1, c0)] + integral[np.ix_(l0, c0)])
    return total_ > 0


_TILE_GEOGRAPHY = {}


def tile_geography(tile, side):
    """Land, sea, coastline and cities of a tile — computed once only.

    Everything comes from the real borders embedded in step 03 a-bis, so the
    demonstration set looks like the chosen country, whichever country it is.
    """
    key = (tile, side)
    if key in _TILE_GEOGRAPHY:
        return _TILE_GEOGRAPHY[key]

    extent = tile_extent(tile)
    LON, LAT = coordinate_grid(extent, (side, side))
    land = land_mask(LON, LAT)
    resolution = TILE_DEGREES / side

    def radius(degres):
        return max(int(round(degres / resolution)), 1)

    coastline = land & _neighbourhood(~land, radius(0.10))         # land at the water's edge
    shelf = (~land) & _neighbourhood(land, radius(1.10))        # shallow sea
    open_sea = (~land) & ~shelf                                # deep ocean

    land_water = np.ones(LON.shape, dtype="uint8")
    land_water[shelf] = 0
    land_water[open_sea] = 7
    land_water[coastline] = 2

    # Moonglint and coastal traffic: the sea is not black, and that is exactly
    # what distorts coastal indicators when water is left unmasked.
    # The decay away from the coast must be CONTINUOUS: discrete steps would
    # produce perfectly sharp rings, which no real image ever shows.
    steps_ = np.linspace(0.04, 1.15, 11)
    proximity = np.zeros(LON.shape, dtype="float32")
    for degres in steps_:
        proximity += (_neighbourhood(land, radius(degres)) & (~land)).astype("float32")
    proximity /= len(steps_)                       # 1 at the shore, 0 offshore
    glint = (0.20 * proximity ** 2.4).astype("float32")

    cities = cities_within(extent, number=25)
    _TILE_GEOGRAPHY[key] = dict(LON=LON, LAT=LAT, land=land, land_water=land_water,
                                  shelf=shelf, glint=glint, cities=cities,
                                  resolution=resolution)
    return _TILE_GEOGRAPHY[key]


def _radiance_field(tile, side, seed=0, intensity=1.0, noise=0.05):
    """Build a plausible radiance field for this tile."""
    geo = tile_geography(tile, side)
    LON, LAT, land = geo["LON"], geo["LAT"], geo["land"]
    rng = np.random.default_rng(seed + tile_numbers(tile)[0] * 101 + tile_numbers(tile)[1])
    # The rural background must be CONTINUOUS from one tile to the next, otherwise
    # the step 14 mosaic shows rectangles of brightness: its phases are therefore
    # drawn from a generator that does NOT depend on the tile.
    rng_fond = np.random.default_rng(seed)

    # The rural background: a SMOOTH variation (relief, density of hamlets)
    # plus a fine grain. A purely random background would look like noise and
    # would hide the urban structure we actually care about.
    _lisse = _smooth_field(LON, LAT, rng_fond, modes=5, scale=(0.8, 3.0))
    _lisse = (_lisse - _lisse.mean()) / max(float(_lisse.std()), 1e-6)
    _grain = rng.normal(0.0, 1.0, LON.shape)
    # LOG-NORMAL distribution: that is the shape rural radiances really have,
    # and it is what the histogram of step 15 has to reveal.
    _fond = np.exp(-2.35 + 0.55 * _lisse + 0.42 * _grain)
    field = np.where(land, _fond, 0.0).astype("float32")

    # --- the cities -----------------------------------------------------------
    #  An urban area is not a round blob. We build it in three layers: a core,
    #  sub-centres, outgoing axes — then modulate the whole thing with a fine
    #  texture, so that the zoom of step 14 shows built-up fabric rather than a
    #  smooth gradient.
    urban = np.zeros(LON.shape, dtype="float32")
    for city_name, clon, clat, population in geo["cities"]:
        weight = city_weight(population)
        # sigma = radius of the dense core. 0.031° ~ 3.4 km for a capital:
        # the order of magnitude of a genuinely saturated city centre.
        sigma = 0.010 + 0.022 * weight
        rng_ville = np.random.default_rng(int(abs(clon) * 1000 + abs(clat) * 37))
        lobe = np.zeros(LON.shape, dtype="float32")
        for amplitude, radius in [(190.0, 1.0), (24.0, 3.0), (4.5, 8.0)]:
            lobe += amplitude * weight * intensity * np.exp(
                -(((LON - clon) / (sigma * radius)) ** 2
                  + ((LAT - clat) / (sigma * radius)) ** 2) / 2)
        for _k in range(9):                                  # districts, sub-centres
            angle = rng_ville.uniform(0, 2 * np.pi)
            dist = rng_ville.uniform(1.0, 4.0) * sigma
            lobe_radius = sigma * rng_ville.uniform(0.25, 0.50)
            lobe += (55.0 * weight * intensity * rng_ville.uniform(0.3, 1.0)) * np.exp(
                -(((LON - clon - dist * np.cos(angle)) / lobe_radius) ** 2
                  + ((LAT - clat - dist * np.sin(angle)) / lobe_radius) ** 2) / 2)
        for _k in range(4):                                  # outgoing road axes
            angle = rng_ville.uniform(0, np.pi)
            le_long = (LON - clon) * np.cos(angle) + (LAT - clat) * np.sin(angle)
            en_travers = -(LON - clon) * np.sin(angle) + (LAT - clat) * np.cos(angle)
            lobe += 16.0 * weight * intensity \
                * np.exp(-(en_travers / (sigma * 0.18)) ** 2 / 2) \
                * np.exp(-(le_long / (sigma * 4.0)) ** 2 / 2)
        urban += lobe

    # Texture of the built-up area: without it a city stays a perfectly smooth
    # gradient, whereas a real urban area is grainy at 500 m resolution.
    if geo["cities"]:
        texture = _smooth_field(LON, LAT, rng, modes=8, scale=(40.0, 130.0))
        urban = urban * (1.0 + 0.30 * texture / max(float(np.abs(texture).max()), 1e-6))

    field = field + urban

    if noise:
        field *= (1 + rng.normal(0, noise, LON.shape)).clip(0.4, 2.0)

    # The sea: moonglint, plus a few ships or light-attracting fisheries.
    if geo["shelf"].any():
        rng_mer = np.random.default_rng(seed + 23)
        sea = geo["glint"].copy()
        indices = np.flatnonzero(geo["shelf"].ravel())
        for _ in range(min(6, indices.size)):
            k = int(rng_mer.choice(indices))
            plon, plat = LON.ravel()[k], LAT.ravel()[k]
            sea = sea + 7.0 * np.exp(-(((LON - plon) / 0.045) ** 2
                                       + ((LAT - plat) / 0.045) ** 2) / 2)
        sea = sea * np.exp(rng_mer.normal(0.0, 0.33, LON.shape)).clip(0.3, 3.0)
        # We ADD the urban lights to the sea rather than overwrite them: a port
        # capital spills over the water, and a simplified coastline would
        # otherwise classify half the city as "ocean".
        field = np.where(geo["shelf"], sea + 0.4 * urban, field)

    return field.clip(0, 6000).astype("float32"), LON, LAT


def _encode(values_, spec):
    """Real value -> stored value, honouring the type, scale factor and fill value."""
    if spec["type"].startswith("float"):            # Collection 2 composites: decimals stored as such
        return np.where(np.isfinite(values_), values_, spec["fill"]).astype(spec["type"])
    brut = np.round(np.nan_to_num(values_) / spec["scale"])
    return np.clip(brut, 0, spec["fill"] - 1).astype(spec["type"])


def _as_bytes(text):
    """Encode a string as UTF-8 bytes for an HDF5 attribute."""
    return np.frombuffer(str(text).encode("utf-8"), dtype="S1").tobytes()


def _write_layer(groupe, name_, tableau, spec, extra=None):
    """Write a dataset with the full set of NASA-style attributes."""
    jeu = groupe.create_dataset(name_, data=tableau, compression="gzip", compression_opts=4)
    jeu.attrs["long_name"] = _as_bytes(spec["title_"])
    jeu.attrs["units"] = _as_bytes(spec["unite"])
    jeu.attrs["scale_factor"] = np.array([spec["scale"]], dtype="float64")
    jeu.attrs["add_offset"] = np.array([0.0], dtype="float64")
    jeu.attrs["_FillValue"] = np.array([spec["fill"]], dtype=tableau.dtype)
    jeu.attrs["valid_min"] = np.array([0], dtype=tableau.dtype)
    jeu.attrs["valid_max"] = np.array([1.0e5 if spec["type"].startswith("float") else spec["fill"] - 1],
                                    dtype=tableau.dtype)
    for key, val in (extra or {}).items():
        jeu.attrs[key] = _as_bytes(val) if isinstance(val, str) else val


def _struct_metadata(tile, side, grille):
    """Reproduce the StructMetadata.0 block specific to the HDF-EOS5 format."""
    west, east, south, north = tile_extent(tile)
    return (f"GROUP=SwathStructure\nEND_GROUP=SwathStructure\n"
            f"GROUP=GridStructure\n\tGROUP=GRID_1\n"
            f"\t\tGridName=\"{grille}\"\n\t\tXDim={side}\n\t\tYDim={side}\n"
            f"\t\tUpperLeftPointMtrs=({west * 1e6:.6f},{north * 1e6:.6f})\n"
            f"\t\tLowerRightMtrs=({east * 1e6:.6f},{south * 1e6:.6f})\n"
            f"\t\tProjection=HE5_GCTP_GEO\n\t\tGridOrigin=HE5_HDFE_GD_UL\n"
            f"\tEND_GROUP=GRID_1\nEND_GROUP=GridStructure\nEND\n")


def _smooth_field(LON, LAT, rng, modes=6, scale=(0.4, 2.6)):
    """A smooth random field: used for quality defects and clouds, which come
    in patches and never as salt-and-pepper noise."""
    field = np.zeros(LON.shape)
    for _k in range(modes):
        fx, fy = rng.uniform(scale[0], scale[1], 2)
        px, py = rng.uniform(0, 2 * np.pi, 2)
        field += np.sin(fx * LON + px) * np.sin(fy * LAT + py)
    return field


def make_demo_file(path_, product, tile, year, day, side=480,
                           seed=0, intensity=1.0):
    """Write a structurally faithful demonstration .h5 file."""
    path_ = Path(path_)
    path_.parent.mkdir(parents=True, exist_ok=True)
    journalier = product == "VNP46A2"
    grille = "VNP_Grid_DNB" if journalier else "VIIRS_Grid_DNB_2d"
    spec_produit = SPEC_DAILY if journalier else SPEC_COMPOSITE

    radiance, LON, LAT = _radiance_field(tile, side, seed, intensity,
                                         noise=0.25 if journalier else 0.05)
    geo = tile_geography(tile, side)
    land_water = geo["land_water"]
    rng = np.random.default_rng(seed + 7)

    with h5py.File(path_, "w") as f:
        west, east, south, north = tile_extent(tile)
        h, v = tile_numbers(tile)
        for key, val in [("ShortName", product), ("PlatformShortName", "Suomi-NPP"),
                         ("InstrumentShortName", "VIIRS"), ("ProcessingLevel", "Level 3"),
                         ("ProductionTime", f"{year}-06-01 00:00:00.000"),
                         ("HorizontalTileNumber", f"{h:02d}"),
                         ("VerticalTileNumber", f"{v:02d}"),
                         ("DEMO_SYNTHETIC", "TRUE - faithful structure, simulated values")]:
            f.attrs[key] = _as_bytes(val)
        for key, val in [("WestBoundingCoord", west), ("EastBoundingCoord", east),
                         ("SouthBoundingCoord", south), ("NorthBoundingCoord", north)]:
            f.attrs[key] = np.array([val], dtype="float64")

        f.create_group("HDFEOS INFORMATION").create_dataset(
            "StructMetadata.0", data=_as_bytes(_struct_metadata(tile, side, grille)))

        groupe = f.create_group(f"HDFEOS/GRIDS/{grille}/Data Fields")
        for name_, spec in spec_produit.items():
            if name_ == "Land_Water_Mask":
                _write_layer(groupe, name_, land_water, spec, {"description": DESC_LAND_WATER})
            elif name_ == "DNB_Platform":
                _write_layer(groupe, name_, np.zeros(radiance.shape, "uint8"), spec,
                               {"description": "0 = Suomi-NPP 1 = NOAA-20 2 = NOAA-21 3 = NOAA-20 and NOAA-21"})
            elif name_.endswith("_Quality") or name_ == "Mandatory_Quality_Flag":
                field_q = _smooth_field(LON, LAT, rng, modes=9, scale=(0.5, 4.0))
                field_q += 0.40 * _smooth_field(LON, LAT, rng, modes=12, scale=(3.0, 9.0))
                part1, part2 = (0.22, 0.10) if journalier else (0.07, 0.025)
                q = np.zeros(radiance.shape, "uint8")
                q[field_q >= np.quantile(field_q, 1 - part1)] = 1
                q[field_q >= np.quantile(field_q, 1 - part2)] = 2
                q[land_water == 7] = 255                 # offshore: no retrieval
                _write_layer(groupe, name_, q, spec, {
                    "description": DESC_QUALITY_DAILY if journalier else DESC_QUALITY_COMPOSITE})
            elif name_.endswith("_Num"):
                _write_layer(groupe, name_,
                               np.clip(rng.poisson(230 if product == "VNP46A4" else 14, radiance.shape),
                                       0, 65534).astype("uint16"), spec)
            elif name_.endswith("_Std"):
                _write_layer(groupe, name_, _encode(radiance * 0.18, spec), spec)
            elif name_ == "Snow_Flag":
                _write_layer(groupe, name_, np.zeros(radiance.shape, "uint8"), spec,
                               {"description": "0 = No snow 1 = Snow"})
            elif name_ == "QF_Cloud_Mask":
                field_n = _smooth_field(LON, LAT, rng, modes=6, scale=(0.3, 1.9))
                field_n += 0.35 * _smooth_field(LON, LAT, rng, modes=10, scale=(2.2, 6.0))
                detection = np.zeros(radiance.shape, dtype="uint16")
                detection[field_n >= np.quantile(field_n, 0.55)] = 1
                detection[field_n >= np.quantile(field_n, 0.76)] = 2
                detection[field_n >= np.quantile(field_n, 0.89)] = 3
                qf = np.full(radiance.shape, 0b10, dtype="uint16")
                qf |= (detection << 6).astype("uint16")
                _write_layer(groupe, name_, qf, spec,
                               {"description": "VIIRS cloud mask, bit field"})
            elif name_ == "DNB_Lunar_Irradiance":
                _write_layer(groupe, name_, _encode(np.full(radiance.shape, 12.5), spec), spec)
            elif name_ == "Latest_High_Quality_Retrieval":
                _write_layer(groupe, name_,
                               np.clip(rng.poisson(2, radiance.shape), 0, 254).astype("uint8"), spec)
            elif "NearNadir" in name_:
                _write_layer(groupe, name_, _encode(radiance * 0.97, spec), spec)
            elif "OffNadir" in name_:
                _write_layer(groupe, name_, _encode(radiance * 1.04, spec), spec)
            elif "Gap_Filled" in name_:
                _write_layer(groupe, name_, _encode(radiance * 1.01, spec), spec)
            else:
                # Main radiance layer: the open sea gets the fill value
                encode = _encode(radiance, spec)
                encode[land_water == 7] = spec["fill"]
                _write_layer(groupe, name_, encode, spec)

        if not journalier:
            resolution = TILE_DEGREES / side
            axes = {"lon": (west + (np.arange(side) + 0.5) * resolution,
                            "Longitude", "degrees_east"),
                    "lat": (north - (np.arange(side) + 0.5) * resolution,
                            "Latitude", "degrees_north")}
            for axis_name, (values_, titre_axe, unite) in axes.items():
                jeu = groupe.create_dataset(axis_name, data=values_.astype("float64"))
                jeu.attrs["long_name"] = _as_bytes(titre_axe)
                jeu.attrs["units"] = _as_bytes(unite)
                jeu.attrs["_FillValue"] = np.array([-999.9], dtype="float64")
    return path_


def make_demo_set(folder, tiles=None, years=(2021, 2022, 2023)):
    """Build a complete set for THE CHOSEN COUNTRY: annual, monthly, daily."""
    folder = Path(folder)
    tiles = list(tiles or COUNTRY_TILES)

    # The "main" tile is the one holding the country's largest city.
    principales = country_cities(number=1)
    main_tile = tiles[0]
    if principales:
        _, city_lon, city_lat, _ = principales[0]
        for t in tiles:
            o, e, s, n = tile_extent(t)
            if o <= city_lon <= e and s <= city_lat <= n:
                main_tile = t
                break

    # A very open_sea country may need a dozen tiles: building them all would keep
    # the room waiting for nothing. We keep the tiles closest to the main urban
    # area, and we say so.
    TILE_CAP = 9
    dropped = []
    if len(tiles) > TILE_CAP:
        def _distance(t):
            o, e, s, n = tile_extent(t)
            if not principales:
                return 0.0
            _, city_lon, city_lat, _ = principales[0]
            return (((o + e) / 2 - city_lon) ** 2 + ((s + n) / 2 - city_lat) ** 2) ** 0.5
        ordered = sorted(tiles, key=_distance)
        tiles, dropped = sorted(ordered[:TILE_CAP]), sorted(ordered[TILE_CAP:])
        if main_tile not in tiles:
            main_tile = tiles[0]

    # Every tile must share the SAME pixel size, otherwise the mosaic of step
    # 14 will not fit back together. So we scale the side to the tile count.
    divisor = {1: 1, 2: 1.35}.get(len(tiles), 1.8 if len(tiles) <= 4 else 2.6)
    side = max(280, int(DEMO_PX / divisor))

    # The most recent year is built on EVERY tile: it is the one used for the
    # mosaic and the clipping of step 14. The earlier years only exist to show
    # that a series exists, so the main tile is enough.
    latest = max(years)
    jobs = [("VNP46A4", tile, latest, 1, f"annual/{latest}", 1.0)
               for tile in tiles]
    for year in sorted(a for a in years if a != latest):
        jobs.append(("VNP46A4", main_tile, year, 1, f"annual/{year}",
                        1 - 0.07 * (latest - year)))
    for day in (1, 91, 182, 274):
        jobs.append(("VNP46A3", main_tile, latest, day,
                        f"monthly/{latest}", 1.0))
    for day in (12, 200):
        jobs.append(("VNP46A2", main_tile, latest, day,
                        f"daily/{latest}", 1.0))

    print(f"  country {COUNTRY} · tiles {', '.join(tiles)}"
          f" · {side} × {side} pixels per tile"
          f" (main tile: {main_tile})")
    if dropped:
        print(f"  tiles dropped to keep the run short: {', '.join(dropped)}")
    print(f"  {len(jobs)} files to build ...")
    written = []
    for i, (product, tile, year, day, subfolder, intensity) in enumerate(jobs):
        name_ = f"{product}.A{year}{day:03d}.{tile}.002.DEMO{i:09d}.h5"
        written.append(make_demo_file(folder / subfolder / name_, product, tile,
                                             year, day, side=side, seed=year + day,
                                             intensity=intensity))
    return written


print("Demonstration generator ready (it is only called as a last resort).")
````

unlikely. As soon as one rung returns something, it stops.

````python
# =============================================================================
#  STEP 03 c — GETTING THE DATA: YOUR FILES, NASA, OR THE DEMONSTRATION
# =============================================================================

NOMS_DOSSIERS_USUELS = ["blackmarble_data", "blackmarble", "black_marble", "BlackMarble",
                        "ntl_data", "NTL", "ntl", "data", "data_", "Collecte",
                        "Collecte/blackmarble_data", "VNP46"]

LAADS_ARCHIVE = "https://ladsweb.modaps.eosdis.nasa.gov"
ARCHIVE_ID = 5200          # the archive that holds the whole VNP46 suite


# --- 1. The token -------------------------------------------------------------
def earthdata_token():
    """Look for the token in four places, from the most explicit to the most discreet.

    A token is a personal secret: the good practice is NOT to put it in the
    notebook, but in an environment variable or in a file in your home folder.
    That way the notebook stays shareable exactly as it is.
    """
    if EARTHDATA_TOKEN_VALUE.strip():
        return EARTHDATA_TOKEN_VALUE.strip(), "the cell in step 02"
    depuis_env = os.environ.get("EARTHDATA_TOKEN", "").strip()
    if depuis_env:
        return depuis_env, "the EARTHDATA_TOKEN environment variable"
    for loader in ("colab", "kaggle"):                 # notebook secret stores
        try:
            if loader == "colab":
                from google.colab import userdata          # type: ignore
                value = (userdata.get("EARTHDATA_TOKEN") or "").strip()
            else:
                from kaggle_secrets import UserSecretsClient   # type: ignore
                value = (UserSecretsClient().get_secret("EARTHDATA_TOKEN") or "").strip()
            if value:
                return value, ("Google Colab Secrets" if loader == "colab" else "Kaggle Secrets")
        except Exception:
            pass
    file_ = Path.home() / ".earthdata_token"
    try:
        if file_.exists() and file_.read_text(encoding="utf-8").strip():
            return file_.read_text(encoding="utf-8").strip(), f"the file {file_}"
    except OSError:
        pass
    if ASK_FOR_TOKEN:
        try:
            import getpass
            print("No token found in the cell, in the environment or in ~/.earthdata_token.")
            typed = getpass.getpass(
                "Earthdata token — paste it, or leave empty for demonstration mode: ").strip()
            if typed:
                return typed, "typed in at the keyboard"
            print("  (no token entered: the notebook will switch to demonstration mode)")
        except Exception:
            # Non-interactive run (nbconvert, papermill, scheduled job): nobody is
            # at the keyboard, so we carry on without a token.
            print("  (non-interactive run: the token cannot be asked for)")
    return "", "no source"


TOKEN, TOKEN_SOURCE = earthdata_token()


# --- 2. The download ----------------------------------------------------------
def list_granules(product, year, day, token, timeout=40):
    """List the files available for one product, one year and one Julian day.

    "requests" is imported here rather than at the top of the notebook: it is
    only needed for downloading, and its absence must never stop the notebook.
    """
    import requests
    url = (f"{LAADS_ARCHIVE}/api/v2/content/details/allData/{ARCHIVE_ID}/{product}"
           f"/{year}/{day:03d}?fields=all&formats=json")
    response = requests.get(url, headers={"Authorization": f"Bearer {token}"}, timeout=timeout)
    response.raise_for_status()
    return response.json().get("content", []) or []


def _readable_h5(path_):
    """True if the file really opens as HDF5 (a truncated download does not)."""
    try:
        with h5py.File(path_, "r"):
            return True
    except Exception:
        return False


def download_vnp46(product, year, day, tiles, folder, token, timeout=300):
    """Download the granules matching the requested tiles. Returns the list written."""
    import requests
    folder = Path(folder)
    folder.mkdir(parents=True, exist_ok=True)
    written = []
    for element in list_granules(product, year, day, token):
        name_ = element.get("name", "")
        if not name_.endswith(".h5") or not any(f".{t}." in name_ for t in tiles):
            continue
        cible = folder / name_
        if cible.exists() and _readable_h5(cible):
            written.append(cible)
            continue
        lien = element.get("downloadsLink") or (
            f"/archive/allData/{ARCHIVE_ID}/{product}/{year}/{day:03d}/{name_}")
        url = lien if lien.startswith("http") else LAADS_ARCHIVE + lien
        with requests.get(url, headers={"Authorization": f"Bearer {token}"},
                          stream=True, timeout=timeout) as flux:
            flux.raise_for_status()
            expected = int(flux.headers.get("content-length", 0))
            partial = cible.with_suffix(".h5.part")          # never leave a truncated .h5 behind
            with open(partial, "wb") as sortie:
                for bloc in flux.iter_content(chunk_size=1 << 20):
                    sortie.write(bloc)
        if (expected and partial.stat().st_size < expected) or not _readable_h5(partial):
            partial.unlink(missing_ok=True)
            raise IOError(f"incomplete or unreadable download: {name_}")
        partial.replace(cible)
        print(f"      {name_}  ({cible.stat().st_size / 1e6:.1f} MB)")
        written.append(cible)
    return written


def collect_for_country(folder, token):
    """Download enough to run the notebook: 3 annual, 1 monthly, 1 daily."""
    commandes = [("VNP46A4", a, 1) for a in WANTED_YEARS]
    commandes += [("VNP46A3", max(WANTED_YEARS), 1),
                  ("VNP46A2", max(WANTED_YEARS), 12)]
    obtenus = []
    for product, year, day in commandes:
        print(f"   {product}  {year}  day {day:03d} ...")
        obtenus += download_vnp46(product, year, day, COUNTRY_TILES, folder, token)
    return obtenus


# --- 3. Locating files already present ----------------------------------------
def _contient_vnp46(folder, limite=4000):
    try:
        for i, file_ in enumerate(Path(folder).rglob("*.h5")):
            if file_.name.upper().startswith("VNP46"):
                return True
            if i > limite:
                break
    except (PermissionError, OSError):
        pass
    return False


def locate_data(chemin_explicite=None):
    if chemin_explicite:
        d = Path(chemin_explicite).expanduser()
        if d.exists():
            return d, ("explicit path (DATA_DIR)" if _contient_vnp46(d)
                       else "explicit path (DATA_DIR) — no VNP46 detected")
        print(f"  ! DATA_DIR points at a folder that does not exist: {d}")
    # Files collected by notebook 1 ("Collect"): <NTL_HOME or NTL_LOCAL>/h5_cache/<COUNTRY>
    roots = [Path(os.environ["NTL_HOME"])] if os.environ.get("NTL_HOME") else []
    roots += [Path("/content/NTL_LOCAL"), Path("/kaggle/working/NTL_LOCAL"),
              Path.home() / "NTL_LOCAL", Path.cwd() / "NTL_LOCAL"]
    for root_ in roots:
        d = root_ / "h5_cache" / COUNTRY
        if d.exists() and _contient_vnp46(d):
            return d, "files collected by notebook 1 (NTL_LOCAL/h5_cache)"
    racine_kaggle = Path("/kaggle/input")
    if racine_kaggle.exists():
        for d in sorted(racine_kaggle.iterdir()):
            if d.is_dir() and _contient_vnp46(d):
                return d, "Kaggle dataset"
    for d in [Path("/content/drive/MyDrive"), Path("/content")]:
        if d.exists() and _contient_vnp46(d):
            return d, "Google Colab"
    ici = Path.cwd()
    for base in [ici, *list(ici.parents)[:4]]:
        for name_ in NOMS_DOSSIERS_USUELS:
            d = base / name_
            if d.exists() and _contient_vnp46(d):
                return d, "usual local folder"
    if _contient_vnp46(ici):
        return ici, "working directory"
    return None, "no data found"


# --- 4. The ladder of priority, applied ---------------------------------------
print("Looking for the data ...")
DATA_FOLDER, ORIGINE = locate_data(DATA_DIR)
DEMO_MODE = False

if DATA_FOLDER is None and DOWNLOAD:
    print(f"\n  No local file. Earthdata token: {TOKEN_SOURCE}.")
    if not TOKEN:
        print("  → Without a token there is no download. To obtain one:")
        print("    https://urs.earthdata.nasa.gov  then the Generate Token tab.")
    elif not AVAILABLE.get("requests"):
        print("  → The requests library is missing: no download possible.")
    else:
        cible = Path.cwd() / "blackmarble_data"
        print(f"  Downloading from LAADS DAAC into {cible} ...")
        try:
            obtenus = collect_for_country(cible, TOKEN)
            if obtenus:
                DATA_FOLDER, ORIGINE = cible, "downloaded from LAADS DAAC"
                print(f"  {len(obtenus)} file(s) obtained.")
            else:
                print("  No granule matches the requested tiles.")
        except Exception as error:
            print(f"  Download interrupted: {type(error).__name__} — {error}")
            print("  The most frequent causes: an expired token, no connection,")
            print("  or access to the product not yet accepted in your Earthdata profile.")

if DATA_FOLDER is None:
    if ALLOW_DEMO:
        DATA_FOLDER = Path.cwd() / f"ntl_demo_{COUNTRY}"
        if not _contient_vnp46(DATA_FOLDER):
            print(f"\n  Building the demonstration set for {COUNTRY_NAME} ...")
            make_demo_set(DATA_FOLDER, COUNTRY_TILES, WANTED_YEARS)
        ORIGINE, DEMO_MODE = f"demonstration set built for {COUNTRY}", True
    else:
        raise SystemExit("No data found and ALLOW_DEMO = False. "
                         "Set DATA_DIR in step 02, or supply a token.")


# --- 5. Inventory -------------------------------------------------------------
def dresser_inventaire(folder):
    """Walk the folder and catalogue every usable VNP46 file."""
    rows = []
    for file_ in sorted(Path(folder).rglob("*.h5")):
        info = decode_filename(file_.name)
        if info is None:
            continue
        rows.append({
            "product": info["product"].upper(), "family": info["family"],
            "year": info["year"], "day": info["day"], "date": info["date"].date(),
            "tile": info["tile"], "size_MB": round(file_.stat().st_size / 1e6, 2),
            "path": str(file_),
        })
    columns = ["product", "family", "year", "day", "date", "tile", "size_MB", "path"]
    if not rows:
        return pd.DataFrame(columns=columns)
    return (pd.DataFrame(rows).sort_values(["product", "year", "day", "tile"])
            .reset_index(drop=True))


INVENTORY = dresser_inventaire(DATA_FOLDER)

if INVENTORY.empty:
    raise SystemExit(
        f".h5 files exist under {DATA_FOLDER} but none follows the VNP46 naming\n"
        f"convention. Check that they were not renamed when downloaded.")

missing_tiles = [t for t in COUNTRY_TILES if t not in set(INVENTORY["tile"])]


def sample_file(product=None):
    """A representative file: if possible the one that contains the capital."""
    subset = INVENTORY[INVENTORY["product"] == product] if product else INVENTORY
    if subset.empty:
        for fallback in ("VNP46A4", "VNP46A3", "VNP46A2", "VNP46A1"):
            subset = INVENTORY[INVENTORY["product"] == fallback]
            if not subset.empty:
                break
    if subset.empty:
        subset = INVENTORY
    # The MOST RECENT year first: otherwise the storyline would silently compare
    # two different vintages.
    subset = subset.sort_values(["year", "day"], ascending=False)
    main_cities = country_cities(number=1)
    if main_cities:
        _, city_lon, city_lat, _ = main_cities[0]
        for _, row_ in subset.iterrows():
            o, e, s, n = tile_extent(row_["tile"])
            if o <= city_lon <= e and s <= city_lat <= n:
                return Path(row_["path"])
    return Path(subset.iloc[0]["path"])


SAMPLE_FILE = sample_file(MAIN_PRODUCT)

# A demonstration set left behind in the working directory would be found by the
# automatic detection and taken for real data. The files built here carry the
# global attribute DEMO_SYNTHETIC: reading it is enough.
with h5py.File(SAMPLE_FILE, "r") as _f_controle:
    if read_attribute(_f_controle, "DEMO_SYNTHETIC", "") and not DEMO_MODE:
        DEMO_MODE = True
        ORIGINE += " (demonstration set, recognised by its DEMO_SYNTHETIC attribute)"

# --- 6. Report ----------------------------------------------------------------
banner = (f"  DEMONSTRATION MODE — {COUNTRY_NAME} — simulated values, faithful structure  "
           if DEMO_MODE else f"  REAL DATA — {COUNTRY_NAME}  ")
print("\n" + "=" * 78)
print(banner.center(78, "*" if DEMO_MODE else "="))
print("=" * 78)
print(f"  Origin            : {ORIGINE}")
print(f"  Folder            : {DATA_FOLDER}")
print(f"  VNP46 files       : {len(INVENTORY)}   ({INVENTORY['size_MB'].sum():.1f} MB)")
print(f"  Products present  : {', '.join(sorted(INVENTORY['product'].unique()))}")
print(f"  Tiles present     : {', '.join(sorted(INVENTORY['tile'].unique()))}")
print(f"  Country tiles     : {', '.join(COUNTRY_TILES)}"
      + (f"   ⚠ missing: {', '.join(missing_tiles)}" if missing_tiles else "   (all present)"))
print(f"  Years covered     : {INVENTORY['year'].min()} – {INVENTORY['year'].max()}")
print(f"  Sample file       : {SAMPLE_FILE.name}")
print("=" * 78)

summary = (INVENTORY.groupby(["product", "family"])
            .agg(files=("path", "size"),
                 years=("year", lambda s: f"{s.min()}–{s.max()}"),
                 tiles=("tile", lambda s: s.nunique()),
                 volume_MB=("size_MB", "sum"))
            .reset_index())
display(summary)
display(INVENTORY.drop(columns=["path"]).head(12))
````

#### First look: what does this data actually look like?

Before taking the file apart, let us look at it. The cell below does the bare minimum — open, decode, display — with no filtering at all. This is deliberately **the naive read**, the one everybody performs on day one. It already produces a convincing image, and that is precisely the problem: nothing on this map flags the errors hiding inside it.

````python
# =============================================================================
#  STEP 03 d — FIRST LOOK: YOUR COUNTRY, SEEN AT NIGHT
# =============================================================================
preview_info = decode_filename(SAMPLE_FILE.name)

with h5py.File(SAMPLE_FILE, "r") as f:
    groupe_apercu = f[find_data_group(f)]
    preview_layer = pick_radiance_layer(groupe_apercu)
    preview_dataset = groupe_apercu[preview_layer]
    ech_a = read_attribute(preview_dataset, "scale_factor", 1.0)
    dec_a = read_attribute(preview_dataset, "add_offset", 0.0)
    vide_a = read_attribute(preview_dataset, "_FillValue", None)
    raw_preview = preview_dataset[:].astype("float32")

preview_radiance = np.where(np.isclose(raw_preview, float(vide_a)) if vide_a is not None else False, np.nan, raw_preview * ech_a + dec_a)
preview_extent = tile_extent(preview_info["tile"])
LON_APERCU, LAT_APERCU = coordinate_grid(preview_extent, preview_radiance.shape)
cities = country_cities(number=7)
main_city = cities[0] if cities else None

fig, (axA, axB) = plt.subplots(1, 2, figsize=(13.0, 6.1),
                               gridspec_kw={"width_ratios": [1.1, 1]})
fig.subplots_adjust(top=0.765, bottom=0.105, left=0.055, right=0.945, wspace=0.2)

# ---- A: the whole tile, with the border laid over it ------------------------
imA = night_map(axA, preview_radiance, preview_extent)
axA.set_xlim(preview_extent[0], preview_extent[1])
axA.set_ylim(preview_extent[2], preview_extent[3])
draw_country_outline(axA, colour=COLOURS["gold"], lw=1.9)
label_cities(axA, [v for v in cities
                       if preview_extent[0] <= v[1] <= preview_extent[1]
                       and preview_extent[2] <= v[2] <= preview_extent[3]], size=8.4)
scale_bar(axA)
axA.set_xlabel("longitude (°)")
axA.set_ylabel("latitude (°)")
axA.set_title(f"Tile {preview_info['tile']} — 10° × 10°, i.e. ≈ 1,100 km across",
              pad=8, fontsize=11)

# ---- B: zoom on the main urban area -----------------------------------------
if main_city is not None:
    nom_v, clon, clat, _pop = main_city
    margin = 0.9
    window = ((np.abs(LON_APERCU - clon) <= margin) & (np.abs(LAT_APERCU - clat) <= margin)
               & np.isfinite(preview_radiance))
    zoom_ceiling = (float(np.nanpercentile(preview_radiance[window], 99.8))
                    if window.any() else None)
    imB = night_map(axB, preview_radiance, preview_extent, vmax=zoom_ceiling)
    axB.set_xlim(clon - margin, clon + margin)
    axB.set_ylim(clat - margin, clat + margin)
    draw_country_outline(axB, colour=COLOURS["gold"], lw=1.9)
    label_cities(axB, [v for v in cities_within(
        (clon - margin, clon + margin, clat - margin, clat + margin), number=8)], size=9.5)
    scale_bar(axB)
    axB.set_title(f"{nom_v} and its region — {int(2 * margin * 111):d} km across",
                  pad=8, fontsize=11)
    axA.add_patch(plt.Rectangle((clon - margin, clat - margin), 2 * margin, 2 * margin,
                                facecolor="none", edgecolor=COLOURS["gold"], lw=1.3,
                                ls=(0, (4, 3)), zorder=7))
    colour_bar(fig, imB, axB)
else:
    axB.axis("off")
    colour_bar(fig, imA, axA)
axB.set_xlabel("longitude (°)")

figure_title(fig, f"{COUNTRY_NAME} seen at night — and what we are about to learn to read in it",
             "Gold outline: the national border. White circles: the landmark cities.\n"
             "The colour scale is logarithmic, because a city centre is a thousand times brighter than the countryside.",
             f"Source: {SAMPLE_FILE.name}   ·   layer {preview_layer}   ·   borders from Natural Earth"
             + ("   ·   DEMONSTRATION MODE — simulated values, faithful structure" if DEMO_MODE else ""))
plt.show()

print(f"Layer displayed  : {preview_layer}")
print(f"Non-empty pixels : {np.isfinite(preview_radiance).sum():,} out of {preview_radiance.size:,}"
      .replace(",", " "))
print("No filtering has been applied yet: this is the raw image, decoded, nothing more.")
````

**The starting point.** The sensor that produces this data is called **VIIRS-DNB**
(*Visible Infrared Imaging Radiometer Suite — Day/Night Band*). It flies on the **Suomi-NPP** (since 2011)
and **NOAA-20** (since 2017) satellites, which pass over every point on the globe at around
**1:30 a.m. local time**. The VNP46 products come from Suomi-NPP; their NOAA-20 equivalents are called VJ146. In Collection 2, the composites record in their `DNB_Platform` layer which satellite supplied each pixel — worth checking before reading a break in a series.

A raw night-time measurement is, however, unusable as it stands: it depends on the **moon**, the **clouds**,
the **view angle**, the **aerosols**, the **snow**. NASA's *Black Marble* project applies a chain of
corrections — including the **BRDF** correction, which neutralises the effect of view angle and lunar
illumination — and publishes four products, differing in their **degree of aggregation over time**.

````python
# =============================================================================
#  STEP 04 — THE BLACK MARBLE PRODUCT FAMILY
# =============================================================================
PRODUCT_CATALOGUE = pd.DataFrame([
    dict(product="VNP46A1", cadence="Daily (raw)",
         content="Top-of-atmosphere radiance, uncorrected, + observation geometry",
         layers="~ 26", brdf_corrected="No",
         use="Sensor diagnostics, research. Rarely used in official statistics."),
    dict(product="VNP46A2", cadence="Daily",
         content="BRDF-corrected, moonlight-adjusted radiance, one night",
         layers="~ 7", brdf_corrected="Yes",
         use="A dated event: power cut, conflict, disaster, cyclone."),
    dict(product="VNP46A3", cadence="Monthly",
         content="One-month composite, clouds and outliers removed",
         layers="~ 17", brdf_corrected="Yes",
         use="Seasonality, within-year monitoring, monthly series."),
    dict(product="VNP46A4", cadence="Annual",
         content="One-year composite, the most stable of the four",
         layers="~ 17", brdf_corrected="Yes",
         use="Multi-year trend, territorial indicator, publication reference."),
])

# Characteristics shared by the whole family
SHARED = pd.DataFrame([
    ("Sensor", "VIIRS Day/Night Band on Suomi-NPP (VNP46; NOAA-20 equivalent = VJ146)"),
    ("Overpass time", "~ 01:30, local time"),
    ("Spatial resolution", "15 arc-seconds ~ 463 m at the equator (often rounded to 500 m)"),
    ("Tile size", "2400 × 2400 pixels = 10° × 10°"),
    ("Coordinate system", "Geographic, WGS 84 - EPSG:4326"),
    ("Radiance unit", "nW·cm⁻²·sr⁻¹ (nanowatt per cm2 per steradian)"),
    ("Temporal coverage", "From 19 January 2012 onwards"),
    ("Current collection", "Collection 2 - version 002"),
    ("Format", "HDF-EOS5 (.h5)"),
    ("Access", "LAADS DAAC / Earthdata - free account, access token"),
    ("Licence", "NASA public data, free re-use with citation"),
], columns=["Characteristic", "Value"])

print("THE FOUR PRODUCTS")
display(PRODUCT_CATALOGUE.set_index("product"))
print("\nWHAT THE WHOLE FAMILY SHARES")
display(SHARED.set_index("Characteristic"))
````

````python
# --- Diagram: the same signal, four levels of temporal aggregation ----------
fig, ax = plt.subplots(figsize=(11.2, 4.3))
fig.subplots_adjust(top=0.72, bottom=0.17, left=0.11, right=0.985)

rng_fig = np.random.default_rng(3)
nights = np.arange(365)
# Simulated daily signal: baseline + season + clouds (random dips)
base = 100 + 9 * np.sin(2 * np.pi * (nights - 60) / 365)
clouds = rng_fig.random(365) < 0.35
daily_signal = base * (1 - 0.55 * clouds * rng_fig.random(365)) + rng_fig.normal(0, 4, 365)

ax.plot(nights, daily_signal, lw=0.8, color=COLOURS["sage"], zorder=1,
        label="VNP46A2 - a single night (noisy: clouds, moon)")
monthly = pd.Series(daily_signal).groupby(nights // 30).mean()
centres = np.array(monthly.index) * 30 + 15
ax.step(centres, monthly.values, where="mid", lw=2.2, color=COLOURS["teal"], zorder=2,
        label="VNP46A3 - monthly composite")
ax.axhline(daily_signal.mean(), lw=2.6, color=COLOURS["gold"], zorder=3,
           label="VNP46A4 - annual composite")
ax.plot(nights, base, lw=1.4, ls=(0, (4, 3)), color=COLOURS["dark_green"], zorder=2,
        label="Underlying real signal")

ax.set_xlim(0, 364)
ax.set_ylim(0, 135)
ax.set_xlabel("Day of year")
ax.set_ylabel("Radiance (arbitrary units)")
ax.set_xticks([0, 90, 181, 273, 364])
ax.set_xticklabels(["1 Jan", "1 Apr", "1 Jul", "1 Oct", "31 Dec"])
ax.legend(loc="lower left", fontsize=8.6, ncol=2)

figure_title(fig,
             "The same place, seen through three different products",
             "The more you aggregate over time, the more stable the signal - and the less precisely dated it is.\n"
             "The product does not change the reality measured: it changes the trade-off between temporal precision and robustness.",
             "Teaching illustration built on a simulated signal - not a measurement.")
plt.show()
````

**The idea.** A name like `VNP46A4.A2023001.h17v08.002.2024015113045.h5` looks like a barcode.
It is one: each of its six segments carries information you can use **without opening the file**. That is
what makes it possible to inventory several thousand files in a few seconds — exactly what step 03 did.

The part that surprises people most is the third segment: the date is not in calendar form but as a
**Julian day** (*day-of-year*), a number from 001 to 366. `A2023001` therefore reads "year 2023, day 1",
that is, 1 January 2023.

````python
# =============================================================================
#  STEP 05 — DECODING THE FILENAME
# =============================================================================
sample_name = SAMPLE_FILE.name
info = decode_filename(sample_name)
west, east, south, north = info["extent"]

# What the Julian day means depends on the product
if info["product"] == "VNP46A4":
    date_meaning = f"the whole of {info['year']} (day 001 marks the start of the period)"
elif info["product"] == "VNP46A3":
    date_meaning = f"the month of {info['date'].strftime('%B %Y')} (the day marks the 1st of the month)"
else:
    date_meaning = f"the night of {info['date'].strftime('%d/%m/%Y')}"

reading = pd.DataFrame([
    ("1", "Product", info["product"],
     f"{info['family'].capitalize()} - VNP46 = Black Marble suite, A4 = annual, A3 = monthly, A2 = daily"),
    ("2", "Year", f"A{info['year']}",
     f"The 'A' signals a date; the year is {info['year']}"),
    ("3", "Julian day", f"{info['day']:03d}",
     f"{info['date'].strftime('%d/%m/%Y')} - {date_meaning}"),
    ("4", "Tile", info["tile"],
     f"longitude [{west:+.0f}° , {east:+.0f}°]  ·  latitude [{south:+.0f}° , {north:+.0f}°]"),
    ("5", "Version", info["version"],
     "002 = Collection 2. Version 001 is obsolete: never mix the two."),
    ("6", "Timestamp", info["timestamp"],
     ("NASA production date and time - useful for spotting a reprocessing"
      if info["timestamp"].isdigit() else
      f"Production segment '{info['timestamp']}'")),
], columns=["#", "Segment", "Value", "Interpretation"])

print(f"FILE ANALYSED: {sample_name}\n")
display(reading.set_index("#"))

# --- Check: how many files does the inventory decode without error? ----------
total_h5 = len(list(Path(DATA_FOLDER).rglob("*.h5")))
print(f"\nOut of {total_h5} .h5 file(s) present, {len(INVENTORY)} were decoded "
      f"by reading the name alone ({100 * len(INVENTORY) / max(total_h5, 1):.0f} %).")
````

````python
# --- Annotated diagram of the filename ---------------------------------------
segments = [(info["product"], "PRODUCT", COLOURS["green"]),
            (f"A{info['year']}{info['day']:03d}", "DATE (year + Julian day)", COLOURS["gold"]),
            (info["tile"], "TILE  (10° × 10°)", COLOURS["teal"]),
            (info["version"], "VERSION", COLOURS["terra"]),
            (info["timestamp"], "PRODUCTION", COLOURS["slate"]),
            ("h5", "FORMAT", COLOURS["dark_green"])]

widths = np.array([max(len(v), 6) for v, _, _ in segments], dtype=float)
widths = widths / widths.sum()
edges = np.concatenate([[0], np.cumsum(widths)])

fig, ax = plt.subplots(figsize=(11.2, 2.9))
fig.subplots_adjust(top=0.66, bottom=0.04, left=0.02, right=0.98)
ax.axis("off")
ax.set_xlim(0, 1)
ax.set_ylim(0, 1)

for i, (value, label_, colour) in enumerate(segments):
    x0, x1 = edges[i], edges[i + 1]
    ax.add_patch(plt.Rectangle((x0 + 0.004, 0.52), x1 - x0 - 0.008, 0.26,
                               facecolor=colour, edgecolor="none", zorder=2))
    ax.text((x0 + x1) / 2, 0.65, value, ha="center", va="center", color="white",
            fontsize=11, fontweight="bold", family="monospace", zorder=3)
    ax.plot([(x0 + x1) / 2, (x0 + x1) / 2], [0.50, 0.40], lw=1, color=colour, zorder=1)
    ax.text((x0 + x1) / 2, 0.355, label_, ha="center", va="top", color=COLOURS["ink"],
            fontsize=8.2, fontweight="bold", wrap=True)

ax.text(0.5, 0.90, ".".join(s[0] for s in segments), ha="center", va="center",
        family="monospace", fontsize=10.5, color=COLOURS["slate"])
figure_title(fig, "Anatomy of a VNP46 filename",
             "Six segments separated by dots - each usable without opening the file.")
plt.show()
````

 <div style="color:#231F20;font-size:14px;margin-top:4px;">Why your country fits into one to six squares</div>
</div>

**The problem the grid solves.** A global image at 500 m resolution would weigh terabytes. NASA
therefore cuts the globe into **10° × 10° squares**: 36 columns (`h00` to `h35`, west to east) by 18 rows
(`v00` to `v17`, north to south), i.e. 648 tiles. Each tile is 2400 × 2400 pixels.

Working out the upper-left corner of a tile takes two lines, and that is all you need to georeference a
file:

```
west_longitude = -180 + 10 × h
north_latitude =  +90 - 10 × v
```

The practical consequence: **a country almost always straddles several tiles**, and you will have to
assemble (*mosaic*) them before any national analysis. The cell below computes the tiles needed for a few
African countries.

````python
# =============================================================================
#  STEP 06 — THE TILE GRID
# =============================================================================
#  The bounding boxes come from the Natural Earth borders embedded in step 03,
#  so they are consistent with the clipping that step 14 will perform.
# -----------------------------------------------------------------------------
_reperes = ["CIV", "SEN", "NGA", "MAR", "KEN", "ZAF", "EGY", "COD", "TZA", "SDN"]
_a_montrer = [COUNTRY] + [c for c in _reperes if c != COUNTRY and c in COUNTRY_EXTENTS][:6]

rows = []
for code in _a_montrer:
    o, e, s, n = COUNTRY_EXTENTS[code]
    tiles = tiles_covering(o, e, s, n)
    rows.append({
        "code": code,
        "country": COUNTRY_NAMES.get(code, code) + ("   ◄ your country" if code == COUNTRY else ""),
        "tiles": " ".join(tiles),
        "count": len(tiles),
        "raw pixels": f"{len(tiles) * TILE_PIXELS ** 2 / 1e6:.1f} M",
        "annual weight ≈": f"{len(tiles) * 25:.0f} MB",
    })
display(pd.DataFrame(rows).set_index("code"))

print(f"\nYour country — {COUNTRY_NAME} ({COUNTRY})")
print(f"  Bounding box   : longitude [{COUNTRY_EXTENT[0]:+.2f}° , {COUNTRY_EXTENT[1]:+.2f}°]"
      f"   latitude [{COUNTRY_EXTENT[2]:+.2f}° , {COUNTRY_EXTENT[3]:+.2f}°]")
print(f"  Tiles to cover : {', '.join(COUNTRY_TILES)}")

o, e, s, n = tile_extent(info["tile"])
resolution_deg = TILE_DEGREES / TILE_PIXELS
print(f"\nTile of the sample file: {info['tile']}")
print(f"  upper-left corner   : longitude {o:+.4f}°  ·  latitude {n:+.4f}°")
print(f"  lower-right corner  : longitude {e:+.4f}°  ·  latitude {s:+.4f}°")
print(f"  size of one pixel   : {resolution_deg:.8f}° = {resolution_deg * 3600:.1f} arc-seconds")
print(f"                        ≈ {resolution_deg * 111320:.0f} m in longitude at the equator")
print(f"  pixels in the tile  : {TILE_PIXELS} × {TILE_PIXELS} = "
      f"{TILE_PIXELS ** 2:,} pixels".replace(",", " "))
````

````python
# --- Map: the Black Marble grid, and the country inside it -------------------
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13.2, 6.6),
                               gridspec_kw={"width_ratios": [1.12, 1]})
fig.subplots_adjust(top=0.775, bottom=0.155, left=0.055, right=0.985, wspace=0.18)

tiles_present = set(INVENTORY["tile"].unique())
country_tile_set = set(COUNTRY_TILES)
TITRE_TUILES = (f"{COUNTRY_NAME} and the tile that covers it" if len(country_tile_set) == 1
                else f"{COUNTRY_NAME} and the {len(country_tile_set)} tiles that cover it")

# ---- Panel 1: Africa, every border, and the grid ----------------------------
for code, anneaux in COUNTRY_OUTLINES.items():
    o, e, s, n = COUNTRY_EXTENTS[code]
    if e < -22 or o > 56 or n < -38 or s > 40:
        continue
    for anneau in anneaux:
        ax1.fill(anneau[:, 0], anneau[:, 1], facecolor=COLOURS["mint"],
                 edgecolor=COLOURS["sage"], lw=0.5, zorder=1)
for anneau in COUNTRY_OUTLINES[COUNTRY]:
    ax1.fill(anneau[:, 0], anneau[:, 1], facecolor=COLOURS["green"], alpha=0.85,
             edgecolor=COLOURS["forest"], lw=1.3, zorder=3)

for h in range(int((-25 + 180) // 10), int((58 + 180) // 10) + 1):
    for v in range(int((90 - 42) // 10), int((90 + 40) // 10) + 1):
        tile = f"h{h:02d}v{v:02d}"
        o, e, s, n = tile_extent(tile)
        if e < -22 or o > 56 or n < -38 or s > 40:
            continue
        couvre = tile in country_tile_set
        ax1.add_patch(plt.Rectangle((o, s), 10, 10, facecolor="none",
                                    edgecolor=COLOURS["gold"] if couvre else "#C7D2CD",
                                    lw=2.2 if couvre else 0.7, zorder=5 if couvre else 2))
        if couvre:
            ax1.text(o + 0.4, n - 0.5, tile, ha="left", va="top", fontsize=8,
                     fontweight="bold", color=COLOURS["ochre"], zorder=6)

ax1.set_xlim(-22, 56)
ax1.set_ylim(-38, 40)
ax1.set_aspect("equal")
ax1.set_xlabel("longitude (°)")
ax1.set_ylabel("latitude (°)")
ax1.set_title("The global grid, restricted to Africa", pad=8, fontsize=11)
legend_below_axes(ax1, [(COUNTRY_NAME, COLOURS["green"]),
                        ("tiles to cover", COLOURS["gold"]),
                        ("10° × 10° grid", "#C7D2CD")], columns=3)

# ---- Panel 2: the country and its tiles -------------------------------------
o_p, e_p, s_p, n_p = COUNTRY_EXTENT
margin = max(1.0, 0.16 * max(e_p - o_p, n_p - s_p))
# The limits are set BEFORE drawing: each tile label must fall inside the
# visible window, not at the tile corner, which is often off-frame.
ax2.set_xlim(o_p - margin, e_p + margin)
ax2.set_ylim(s_p - margin, n_p + margin)
for tile in sorted(country_tile_set):
    o, e, s, n = tile_extent(tile)
    presente = tile in tiles_present
    ax2.add_patch(plt.Rectangle((o, s), 10, 10,
                                facecolor=COLOURS["gold"] if presente else "#F3F6F4",
                                edgecolor=COLOURS["ochre"], lw=1.6,
                                alpha=0.26 if presente else 1.0, zorder=1))
    x_etiquette = max(o, o_p - margin) + 0.03 * (e_p - o_p + 2 * margin)
    y_etiquette = max(s, s_p - margin) + 0.03 * (n_p - s_p + 2 * margin)
    ax2.text(x_etiquette, y_etiquette, tile, ha="left", va="bottom", fontsize=8.5,
             fontweight="bold", color=COLOURS["ochre"], zorder=6,
             bbox=dict(boxstyle="round,pad=0.25", fc="white", ec=COLOURS["ochre"], lw=0.8))

for code, anneaux in COUNTRY_OUTLINES.items():
    o, e, s, n = COUNTRY_EXTENTS[code]
    if e < o_p - margin - 8 or o > e_p + margin + 8 or n < s_p - margin - 8 or s > n_p + margin + 8:
        continue
    for anneau in anneaux:
        ax2.fill(anneau[:, 0], anneau[:, 1],
                 facecolor=COLOURS["green"] if code == COUNTRY else COLOURS["mint"],
                 alpha=0.85 if code == COUNTRY else 0.6,
                 edgecolor=COLOURS["forest"] if code == COUNTRY else COLOURS["sage"],
                 lw=1.4 if code == COUNTRY else 0.6, zorder=3 if code == COUNTRY else 2)

ax2.set_aspect("equal")
label_cities(ax2, country_cities(number=7), colour=COLOURS["ink"], size=8.6)
ax2.set_xlabel("longitude (°)")
ax2.set_title(TITRE_TUILES, pad=8, fontsize=11)
legend_below_axes(ax2, [("tile present in your data", COLOURS["gold"]),
                        ("tile missing", "#F3F6F4")], columns=2)

figure_title(fig, "A tile is not a country",
             "The globe is cut into 36 × 18 = 648 squares of 10° × 10°, each of 2400 × 2400 pixels.\n"
             "Borders, for their part, ignore that grid: you will have to mosaic, then clip.",
             "Grid: NASA Black Marble (VNP46, collection 2). Borders: Natural Earth 1:50 M, public domain.")
plt.show()
````

**The mental image.** Open a `.h5` file thinking of a **binder with dividers**, not of a
photograph. Three notions are enough:

- a **group** is a divider — it holds no data, only other groups or datasets; it is the equivalent of a
  folder;
- a **dataset** is a sheet of numbers — a table, here almost always two-dimensional, of integer type;
- an **attribute** is a note stuck onto a group or a sheet: unit, scale factor, fill value, description of
  the codes.

In a Black Marble file the useful path is always of the form
`HDFEOS / GRIDS / <grid name> / Data Fields`. The grid name, however, **changes with the product**:
`VIIRS_Grid_DNB_2d` for the A3 and A4 composites, `VNP_Grid_DNB` for the A2 daily product. That is exactly
why the `find_data_group()` function searches for the group instead of assuming it.

````python
# =============================================================================
#  STEP 07 — THE FILE TREE
# =============================================================================
print(f"FILE : {SAMPLE_FILE.name}")
print(f"SIZE : {SAMPLE_FILE.stat().st_size / 1e6:.2f} MB")
print("=" * 78)

counters = {"groups": 0, "datasets": 0, "bytes": 0}


def show_node(name, obj):
    """Print one node of the tree with proportional indentation."""
    depth = name.count("/")
    leaf = name.split("/")[-1]
    margin = "   " * depth
    if isinstance(obj, h5py.Group):
        counters["groups"] += 1
        n_attrs = len(obj.attrs)
        suffix = f"   [{n_attrs} attribute(s)]" if n_attrs else ""
        print(f"{margin}|- > {leaf}/{suffix}")
    else:
        counters["datasets"] += 1
        counters["bytes"] += obj.size * obj.dtype.itemsize
        shape = "x".join(str(d) for d in obj.shape) if obj.shape else "scalar"
        print(f"{margin}|- . {leaf:<42} {shape:>12}  {str(obj.dtype):<8} "
              f"{len(obj.attrs):>2} attr.")


with h5py.File(SAMPLE_FILE, "r") as f:
    # 1. The global attributes of the file
    globals_ = all_attributes(f)
    if globals_:
        print("\nGLOBAL ATTRIBUTES OF THE FILE")
        for key, value in list(globals_.items())[:14]:
            text = str(value)
            print(f"   {key:<26} = {text[:46]}{'...' if len(text) > 46 else ''}")

    # 2. The full tree
    print("\nTREE")
    print("   >  group (divider)      .  dataset (sheet of numbers)\n")
    f.visititems(show_node)

    group_path = find_data_group(f)

print("\n" + "=" * 78)
print(f"  {counters['groups']} group(s)  ·  {counters['datasets']} dataset(s)  ·  "
      f"{counters['bytes'] / 1e6:.1f} MB once decompressed in memory")
print(f"  Useful group detected automatically:\n     {group_path}")
print("=" * 78)
````

style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span></div>
 <div style="color:#231F20;font-size:14px;margin-top:4px;">Around twenty tables, five roles</div>
</div>

**What to understand.** A file does not hold *one* radiance image but **several versions** of
that image, plus auxiliary layers. Telling them apart is the key to choosing the right one. The five
families:

<table style="width:100%;border-collapse:collapse;border:none;"><tr><td style="width:33.3333%;vertical-align:top;padding:5px;">
 <div style="background:#F4F7F5;border:1px solid #D5DED9;border-top:3px solid #00A86A;border-radius:10px;
  padding:13px 15px;height:100%;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
  <div style="color:#00704A;font-weight:700;font-size:14px;margin-bottom:5px;">Radiance</div>
  <div style="color:#231F20;font-size:13px;line-height:1.55;">The measurement itself, in nW·cm⁻²·sr⁻¹. Several variants depending on view angle (<code>AllAngle</code>, <code>NearNadir</code>, <code>OffNadir</code>) and on snow cover (<code>Snow_Free</code>, <code>Snow_Covered</code>). <b>In Africa we always take <code>AllAngle_Composite_Snow_Free</code>.</b></div>
 </div></td><td style="width:33.3333%;vertical-align:top;padding:5px;">
 <div style="background:#F4F7F5;border:1px solid #D5DED9;border-top:3px solid #00A86A;border-radius:10px;
  padding:13px 15px;height:100%;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
  <div style="color:#00704A;font-weight:700;font-size:14px;margin-bottom:5px;">Uncertainty</div>
  <div style="color:#231F20;font-size:13px;line-height:1.55;"><code>…_Num</code> gives the <b>number of nights</b> actually used to build the composite; <code>…_Std</code> gives the <b>standard deviation</b> across those nights. A pixel computed from 2 nights with a huge standard deviation is not worth a pixel computed from 25 stable ones.</div>
 </div></td><td style="width:33.3333%;vertical-align:top;padding:5px;">
 <div style="background:#F4F7F5;border:1px solid #D5DED9;border-top:3px solid #00A86A;border-radius:10px;
  padding:13px 15px;height:100%;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
  <div style="color:#00704A;font-weight:700;font-size:14px;margin-bottom:5px;">Quality</div>
  <div style="color:#231F20;font-size:13px;line-height:1.55;"><code>…_Quality</code> (composites) or <code>Mandatory_Quality_Flag</code> (daily): one code per pixel telling you whether to trust it. Detailed in step 11.</div>
 </div></td></tr><tr><td style="width:33.3333%;vertical-align:top;padding:5px;">
 <div style="background:#F4F7F5;border:1px solid #D5DED9;border-top:3px solid #00A86A;border-radius:10px;
  padding:13px 15px;height:100%;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
  <div style="color:#00704A;font-weight:700;font-size:14px;margin-bottom:5px;">Context</div>
  <div style="color:#231F20;font-size:13px;line-height:1.55;"><code>Land_Water_Mask</code> distinguishes land, coast, inland water and ocean. <code>DNB_Platform</code> says which satellite supplied the measurement — useful for spotting a break in a series.</div>
 </div></td><td style="width:33.3333%;vertical-align:top;padding:5px;">
 <div style="background:#F4F7F5;border:1px solid #D5DED9;border-top:3px solid #00A86A;border-radius:10px;
  padding:13px 15px;height:100%;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
  <div style="color:#00704A;font-weight:700;font-size:14px;margin-bottom:5px;">Coordinates</div>
  <div style="color:#231F20;font-size:13px;line-height:1.55;"><code>lat</code> and <code>lon</code>: the explicit coordinates of the rows and columns, present in the composites. They let you <b>check</b> the georeferencing you computed.</div>
 </div></td></tr></table>

````python
# =============================================================================
#  STEP 08 — THE LAYER CATALOGUE
# =============================================================================

def classify_layer(name: str) -> str:
    """Translate the technical name of a layer into an understandable role."""
    if name in ("lat", "lon", "latitude", "longitude"):
        return "5 · Coordinates"
    if name == "Land_Water_Mask":
        return "4 · Context - land / water"
    if name == "DNB_Platform":
        return "4 · Context - satellite"
    if name.endswith("_Quality") or "Quality_Flag" in name:
        return "3 · Quality"
    if "Cloud" in name or name == "Snow_Flag":
        return "3 · Quality - cloud / snow"
    if name.endswith("_Num"):
        return "2 · Uncertainty - no. of observations"
    if name.endswith("_Std"):
        return "2 · Uncertainty - standard deviation"
    if "Lunar" in name:
        return "4 · Context - lunar illumination"
    if "Retrieval" in name:
        return "3 · Quality - freshness"
    if "Composite" in name or "NTL" in name:
        return "1 · RADIANCE"
    return "6 · Other"


rows = []
with h5py.File(SAMPLE_FILE, "r") as f:
    group = f[find_data_group(f)]
    main_layer = pick_radiance_layer(group)
    for name, obj in group.items():
        if not isinstance(obj, h5py.Dataset):
            continue
        rows.append({
            "layer": name,
            "role": classify_layer(name),
            "shape": "x".join(str(d) for d in obj.shape) if obj.shape else "scalar",
            "type": str(obj.dtype),
            "units": read_attribute(obj, "units", ""),
            "scale": read_attribute(obj, "scale_factor", ""),
            "offset": read_attribute(obj, "add_offset", read_attribute(obj, "offset", "")),
            "_FillValue": read_attribute(obj, "_FillValue", ""),
            "MB": round(obj.size * obj.dtype.itemsize / 1e6, 2),
        })

LAYER_CATALOGUE = pd.DataFrame(rows).sort_values(["role", "layer"]).reset_index(drop=True)

print(f"{len(LAYER_CATALOGUE)} layer(s) in {SAMPLE_FILE.name}")
print(f"Radiance layer selected automatically: {main_layer}\n")
display(LAYER_CATALOGUE)

print("\nMEMORY WEIGHT BY ROLE")
display(LAYER_CATALOGUE.groupby("role").agg(layers=("layer", "size"), MB=("MB", "sum")))
````

 padding:13px 17px;margin:10px 0;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
 <div style="color:#D49A00;font-weight:700;font-size:13.5px;">&#9888;&nbsp; Pitfall no. 1: believing <code>_Num</code> and <code>_Std</code> are incidental</div>
 <div style="color:#231F20;font-size:13.5px;margin-top:5px;line-height:1.6;">They are not. In a frequently cloudy area, a monthly composite may rest on <b>two or three nights only</b>. The radiance value looks normal, but its uncertainty is enormous. Get into the habit of mapping <code>…_Num</code> next to the radiance: it is the best way to spot the regions where your indicator rests on almost nothing. It is also what will let you, when the time comes, justify a limitation in your methodological note.</div>
</div>

### 09 · Decoding the numbers

<div style="border-left:6px solid #F5C242;background:#E8F5EF;border-radius:0 12px 12px 0;
 padding:14px 22px;margin-top:22px;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
 <div style="color:#D49A00;font-size:11px;font-weight:700;letter-spacing:2.5px;text-transform:uppercase;">
  STEP 09 · 20 &nbsp;&nbsp;<span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#F5C242;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span></div>
 <div style="color:#231F20;font-size:14px;margin-top:4px;">Three attributes that separate the true from the false</div>
</div>

**The problem.** The number stored in the file is not always the radiance itself. The daily products
store **integers** to save space; the monthly and annual composites of Collection 2 store decimals but
mark empty pixels with a special value. In both cases the real radiance is obtained by a conversion, and the parameters of that conversion are written **in the file
itself**, as attributes:

$$\text{radiance} = \text{stored value} \times \texttt{scale\_factor} + \texttt{add\_offset}$$

with one exception: a particular value — the `_FillValue` — does not mean "zero" but
**"no data here"**. It must be converted to `NaN` **before** any operation.

Concretely, for the daily layer `DNB_BRDF-Corrected_NTL`: `scale_factor = 0.1`, `add_offset = 0`,
`_FillValue = 65535`. A stored value of `1247` is therefore **124.7 nW·cm⁻²·sr⁻¹**, and a stored value of
`65535` is worth **nothing at all**. For the Collection 2 composite `AllAngle_Composite_Snow_Free`:
`scale_factor = 1`, `_FillValue = −999.9` — the stored value is the radiance, except for `−999.9`.
These figures are examples: the cell below reads them from your file.

<div style="background:#FBEDEA;border:1px solid #B83B2E44;border-left:4px solid #B83B2E;border-radius:8px;
 padding:13px 17px;margin:10px 0;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
 <div style="color:#B83B2E;font-weight:700;font-size:13.5px;">&#9940;&nbsp; The mistake that silently invalidates a whole analysis</div>
 <div style="color:#231F20;font-size:13.5px;margin-top:5px;line-height:1.6;">Treating the <code>_FillValue</code> as a number. In a daily file (<code>_FillValue = 65535</code>, <code>scale_factor = 0.1</code>) every empty pixel becomes <b>6,553.5 nW·cm⁻²·sr⁻¹</b> — more than the centre of Lagos — and the sum of lights is inflated. In a Collection 2 composite (<code>_FillValue = −999.9</code>) every empty pixel subtracts 999.9 and the sum can even turn negative. Either way, the figure is wrong by orders of magnitude without a single error message appearing. The resulting map still looks fine. <b>It is the first thing to check</b> when a figure looks too good.</div>
</div>

````python
# =============================================================================
#  STEP 09 — DECODING THE NUMBERS: scale_factor, add_offset, _FillValue
# =============================================================================
with h5py.File(SAMPLE_FILE, "r") as f:
    group = f[find_data_group(f)]
    ds = group[main_layer]

    attributes = all_attributes(ds)
    scale = read_attribute(ds, "scale_factor", 1.0)
    offset = read_attribute(ds, "add_offset", read_attribute(ds, "offset", 0.0))
    fill = read_attribute(ds, "_FillValue", None)

    # A small square sample, taken where there is some light
    side = ds.shape[0]
    preview = ds[::max(side // 300, 1), ::max(side // 300, 1)].astype("float64")
    valid_mask = preview != fill if fill is not None else np.ones_like(preview, bool)
    if valid_mask.any():
        i_max, j_max = np.unravel_index(np.argmax(np.where(valid_mask, preview, -1)),
                                        preview.shape)
        stride = max(side // 300, 1)
        r0 = int(np.clip(i_max * stride - 2, 0, side - 5))
        c0 = int(np.clip(j_max * stride - 2, 0, side - 5))
    else:
        r0 = c0 = 0
    raw_block = ds[r0:r0 + 5, c0:c0 + 5].astype("float64")

print(f"LAYER: {main_layer}\n")
print("ATTRIBUTES READ FROM THE FILE (never hard-coded)")
for key, value in attributes.items():
    text = str(value)
    print(f"   {key:<16} = {text[:62]}{'...' if len(text) > 62 else ''}")

print(f"\nTHE DECODING EQUATION")
print(f"   radiance = stored_value × {scale} + {offset}")
print(f"   then     : any value equal to {fill} becomes NaN\n")

# --- Applied to a 5 × 5 block -------------------------------------------------
decoded_block = np.where(np.isclose(raw_block, float(fill)), np.nan, raw_block * scale + offset)

print(f"5 × 5 BLOCK TAKEN AT POSITION (row {r0}, column {c0})\n")
print("   stored values (raw integers)")
print(pd.DataFrame(raw_block.astype("int64")).to_string(
    header=[f"c{c0 + k}" for k in range(5)], index=[f"r{r0 + k}" for k in range(5)]))
print(f"\n   decoded radiances (nW·cm⁻²·sr⁻¹)")
print(pd.DataFrame(np.round(decoded_block, 2)).to_string(
    header=[f"c{c0 + k}" for k in range(5)], index=[f"r{r0 + k}" for k in range(5)]))

# --- Demonstrating the danger -------------------------------------------------
raw_array = None
with h5py.File(SAMPLE_FILE, "r") as f:
    raw_array = f[find_data_group(f)][main_layer][:].astype("float64")

wrong = (raw_array * scale + offset)                    # without handling the _FillValue
right = np.where(np.isclose(raw_array, float(fill)), np.nan, wrong)      # handled correctly
n_empty = int(np.sum(np.isclose(raw_array, float(fill))))

print("\n" + "=" * 78)
print("  WHAT FORGETTING THE _FillValue COSTS")
print("=" * 78)
print(f"  Fill pixels            : {n_empty:,} out of {raw_array.size:,} "
      f"({100 * n_empty / raw_array.size:.1f} %)".replace(",", " "))
print(f"  Sum of lights - handled correctly : {np.nansum(right):>18,.0f}".replace(",", " "))
print(f"  Sum of lights - _FillValue ignored: {np.nansum(wrong):>18,.0f}".replace(",", " "))
if np.nansum(wrong) < 0:
    print("  The sum even turns NEGATIVE: impossible for light, and yet no error message.")
elif np.nansum(right) > 0:
    print(f"  Error factor                      : "
          f"x {np.nansum(wrong) / np.nansum(right):.1f}")
print("=" * 78)

# --- The storyline starts here ------------------------------------------------
log_storyline(9, "no precaution at all", np.nansum(wrong), int(np.isfinite(wrong).sum()),
              f"tile {info['tile']}", "the _FillValue counted as a measurement")
log_storyline(9, "_FillValue set to NaN", np.nansum(right), int(np.isfinite(right).sum()),
              f"tile {info['tile']}", "first defensible value")
````

````python
# --- Figure: from the stored integer to the radiance --------------------------
fig, axes = plt.subplots(1, 3, figsize=(12.4, 4.3))
fig.subplots_adjust(top=0.70, bottom=0.13, left=0.04, right=0.98, wspace=0.22)

for ax, data, title_, cmap, fmt in [
        (axes[0], raw_block, "1 · What is stored\n(integers)", CMAP_GREEN, "{:.0f}"),
        (axes[1], decoded_block, "2 · After decoding\n(× scale + offset)", CMAP_GREEN, "{:.1f}"),
        (axes[2], np.isnan(decoded_block).astype(float),
         "3 · Where the holes are\n(_FillValue → NaN)",
         ListedColormap([COLOURS["mint"], COLOURS["brick"]]), None)]:
    finite = data[np.isfinite(data)]
    vmin, vmax = (float(finite.min()), float(finite.max())) if finite.size else (0.0, 1.0)
    ax.imshow(np.nan_to_num(data, nan=vmin), cmap=cmap, vmin=vmin, vmax=max(vmax, vmin + 1e-9))
    for i in range(data.shape[0]):
        for j in range(data.shape[1]):
            v = data[i, j]
            if fmt is None:
                text = "empty" if np.isnan(decoded_block[i, j]) else "ok"
                colour = "white" if np.isnan(decoded_block[i, j]) else COLOURS["dark_green"]
            elif np.isnan(v):
                text, colour = "NaN", COLOURS["brick"]
            else:
                text = fmt.format(v)
                colour = "white" if (v - vmin) > 0.6 * (vmax - vmin + 1e-9) else COLOURS["ink"]
            ax.text(j, i, text, ha="center", va="center", fontsize=8.2,
                    fontweight="bold", color=colour)
    ax.set_title(title_, fontsize=10.5, pad=7)
    ax.set_xticks(range(5)); ax.set_xticklabels([f"c{c0 + k}" for k in range(5)], fontsize=7.5)
    ax.set_yticks(range(5)); ax.set_yticklabels([f"r{r0 + k}" for k in range(5)], fontsize=7.5)
    ax.tick_params(length=0)

figure_title(fig, "The conversion, pixel by pixel",
             f"layer '{main_layer}'   ·   "
             f"radiance = stored × {scale} + {offset}   ·   _FillValue = {fill} → NaN",
             f"Source: {SAMPLE_FILE.name}" + ("   ·   DEMONSTRATION MODE" if DEMO_MODE else ""))
plt.show()
````

<div style="background:#F4F7F5;border:1px solid #00A86A44;border-left:4px solid #00A86A;border-radius:8px;
 padding:13px 17px;margin:10px 0;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
 <div style="color:#00A86A;font-weight:700;font-size:13.5px;">&#128200;&nbsp; How to read these three panels</div>
 <div style="color:#231F20;font-size:13.5px;margin-top:5px;line-height:1.6;">The first shows the integers as they sleep inside the file; the second the same cells after the equation has been applied; the third marks in red the cells that hold <b>no data at all</b>. If the third panel is entirely green, your block is full — the usual case deep inside a continent. On a coastal block, the red cells are the ocean.</div>
</div>

<div style="background:#F2F7FA;border:1px solid #0E7C8644;border-left:4px solid #0E7C86;border-radius:8px;padding:14px 18px;margin:14px 0;font-family:'Segoe UI',Calibri,sans-serif;"><div style="color:#0E7C86;font-weight:700;font-size:13.5px;">&#9999;&nbsp; Your turn — decode a pixel by hand</div><div style="color:#231F20;font-size:13.5px;margin-top:6px;line-height:1.6;">A pixel of the daily <code>DNB_BRDF-Corrected_NTL</code> layer (<code>scale_factor = 0.1</code>) holds the stored value <b>3,042</b>. Another holds <b>65,535</b>.<br><b>(a)</b> What radiance is the first one? &nbsp; <b>(b)</b> And the second? &nbsp; <b>(c)</b> If you naively sum a hundred pixels of the second kind, by how much do you inflate your sum of lights?</div></div>

<details><summary><i>Show the answer</i></summary>

**(a)** `3042 × 0.1 + 0 = 304.2 nW·cm⁻²·sr⁻¹` — a high value, typical of a city centre.

**(b)** Nothing. `65535` is the `_FillValue`: it means "no data here", and must become `NaN`. It is not a zero, and still less a measurement.

**(c)** `100 × 65535 × 0.1 = 655,350`. For comparison, that is the order of magnitude of the sum of lights of an entire urban area — produced here by a hundred **empty** pixels. In a Collection 2 composite the same mistake works the other way: a hundred fill values of `−999.9` *remove* 99,990 from the sum.

</details>

<div style="border-left:4px solid #F5C242;background:#FFFBF0;border-radius:0 8px 8px 0;
 padding:9px 17px;margin:14px 0 2px 0;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
 <span style="color:#D49A00;font-size:10.5px;font-weight:700;letter-spacing:2px;text-transform:uppercase;">Worth remembering</span>
 <span style="color:#231F20;font-size:13.5px;margin-left:12px;"><code>_FillValue</code> is not a zero. Mistaking one for the other corrupts everything, silently.</span>

### 10 · Where is each pixel?

<div style="border-left:6px solid #00A86A;background:#E8F5EF;border-radius:0 12px 12px 0;
 padding:14px 22px;margin-top:22px;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
 <div style="color:#D49A00;font-size:11px;font-weight:700;letter-spacing:2.5px;text-transform:uppercase;">
  STEP 10 · 20 &nbsp;&nbsp;<span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span></div>
 <div style="color:#231F20;font-size:14px;margin-top:4px;">Going from (row, column) to (longitude, latitude) without a GIS library</div>
</div>

**What you need to know.** Black Marble products sit on a **regular geographic grid**
(EPSG:4326): no complicated projection, no resampling. Pixel (0, 0) is the **north-west corner** of the
tile, and the size of a pixel is constant in degrees:

```
resolution = 10° / 2400 = 0.00416667°  (i.e. 15 arc-seconds)

longitude(column) = west  + (column + 0.5) × resolution
latitude(row)     = north - (row    + 0.5) × resolution
```

The `+ 0.5` places the coordinate at the **centre** of the pixel rather than at its corner — a discreet
convention, but forgetting it shifts your whole grid by half a pixel, about 230 metres.

The A3 and A4 composites also carry explicit `lat` and `lon` datasets. The cell below computes the
coordinates, then **checks them** against the stored values: a verification any serious notebook should do
once.

````python
# =============================================================================
#  STEP 10 — GEOREFERENCING: FROM PIXEL TO COORDINATES
# =============================================================================
sample_tile = info["tile"]
west, east, south, north = tile_extent(sample_tile)
n_rows, n_cols = raw_array.shape
resolution = TILE_DEGREES / n_rows


def pixel_to_lonlat(row, column):
    """Centre of pixel (row, column) in decimal degrees."""
    return (west + (column + 0.5) * resolution,
            north - (row + 0.5) * resolution)


def lonlat_to_pixel(longitude, latitude):
    """The inverse operation: which cell holds this point?"""
    column = int((longitude - west) / resolution)
    row = int((north - latitude) / resolution)
    if not (0 <= row < n_rows and 0 <= column < n_cols):
        return None
    return row, column


# Full axes (useful for display and export)
LONGITUDES = west + (np.arange(n_cols) + 0.5) * resolution
LATITUDES = north - (np.arange(n_rows) + 0.5) * resolution

# Extent in the form imshow expects: (left, right, bottom, top)
DISPLAY_EXTENT = (west, east, south, north)

# Affine transform, in GDAL / rasterio form
AFFINE_TRANSFORM = (resolution, 0.0, west, 0.0, -resolution, north)

print(f"TILE {sample_tile}   ·   {n_rows} × {n_cols} pixels\n")
print(f"  North-west corner : longitude {west:+10.5f}°   latitude {north:+9.5f}°")
print(f"  South-east corner : longitude {east:+10.5f}°   latitude {south:+9.5f}°")
print(f"  Resolution        : {resolution:.8f}°  =  {resolution * 3600:.2f} arc-seconds")
print(f"  Transform         : (a={resolution:.8f}, b=0, c={west:.5f}, "
      f"d=0, e={-resolution:.8f}, f={north:.5f})")

print("\nA FEW CHECK PIXELS")
checks = [(0, 0, "north-west corner"),
          (0, n_cols - 1, "north-east corner"),
          (n_rows - 1, 0, "south-west corner"),
          (n_rows // 2, n_cols // 2, "centre of the tile")]
check_table = pd.DataFrame(
    [{"row": r_, "column": c_, "landmark": name_,
      "longitude": round(pixel_to_lonlat(r_, c_)[0], 5),
      "latitude": round(pixel_to_lonlat(r_, c_)[1], 5)}
     for r_, c_, name_ in checks])
display(check_table.set_index("landmark"))

# --- Check against the lat / lon datasets of the file -------------------------
with h5py.File(SAMPLE_FILE, "r") as f:
    group = f[find_data_group(f)]
    has_coordinates = "lat" in group and "lon" in group
    if has_coordinates:
        lat_file = group["lat"][:]
        lon_file = group["lon"][:]

if has_coordinates:
    lat_gap = float(np.max(np.abs(lat_file - LATITUDES)))
    lon_gap = float(np.max(np.abs(lon_file - LONGITUDES)))
    print(f"\nCHECK against the 'lat' and 'lon' datasets of the file")
    print(f"  maximum gap in latitude  : {lat_gap:.3e}°  "
          f"({lat_gap * 111320:.4f} m)")
    print(f"  maximum gap in longitude : {lon_gap:.3e}°  "
          f"({lon_gap * 111320:.4f} m)")
    verdict = "the two methods agree" if max(lat_gap, lon_gap) < 1e-6 \
              else "SIGNIFICANT GAP - check the pixel-centring convention"
    print(f"  → {verdict}")
else:
    print("\n  This product carries no 'lat' / 'lon' datasets: "
          "the calculation above is the only source of georeferencing.")

# --- Locating a city in the grid ----------------------------------------------
tile_cities = cities_within(DISPLAY_EXTENT, number=10)
if tile_cities:
    print("\nLOCATING A FEW CITIES IN THE GRID")
    city_rows = []
    for city_name, vlon, vlat, _population in tile_cities:
        position = lonlat_to_pixel(vlon, vlat)
        if position is None:
            continue
        r_v, c_v = position
        value = raw_array[r_v, c_v]
        city_radiance = (np.nan if np.isclose(value, float(fill)) else value * scale + offset)
        city_rows.append({"city": city_name, "longitude": vlon, "latitude": vlat,
                          "row": r_v, "column": c_v,
                          "radiance": round(city_radiance, 1)})
    display(pd.DataFrame(city_rows).set_index("city"))
````

````python
# --- Figure: the same file in two reference frames ---------------------------
radiance_full = np.where(np.isclose(raw_array, float(fill)), np.nan,
                             raw_array * scale + offset)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12.8, 5.9))
fig.subplots_adjust(top=0.775, bottom=0.125, left=0.065, right=0.945, wspace=0.26)

# ---- Panel 1: the array frame (row, column) ---------------------------------
ceiling = float(np.nanpercentile(radiance_full[np.isfinite(radiance_full)
                                                  & (radiance_full > 0)], 99.7))
visible = np.where(np.isfinite(radiance_full),
                   np.clip(radiance_full, RADIANCE_FLOOR, None), np.nan)
ax1.imshow(visible, cmap=CMAP_NIGHT,
           norm=LogNorm(vmin=RADIANCE_FLOOR, vmax=max(ceiling, RADIANCE_FLOOR * 20)),
           origin="upper", aspect="auto", interpolation="nearest")
ax1.set_facecolor(NO_DATA_GREY)
ax1.set_xlabel("column (array index)")
ax1.set_ylabel("row (array index)")
ax1.set_title("Array frame: (row, column)", pad=8, fontsize=11)
for x, y, text, va, ha in [(0, 0, " (0, 0)  row 0 = the NORTH", "top", "left"),
                            (0, n_rows - 1, f" ({n_rows - 1}, 0)  last row = the SOUTH",
                             "bottom", "left")]:
    ax1.annotate(text, xy=(x, y), xytext=(x + 0.04 * n_cols,
                                           y + (0.06 if va == "top" else -0.06) * n_rows),
                 fontsize=8.5, color="white", fontweight="bold", va=va, ha=ha,
                 bbox=dict(boxstyle="round,pad=0.32", fc=COLOURS["brick"], ec="none"),
                 arrowprops=dict(arrowstyle="->", color=COLOURS["brick"], lw=1.3))

# ---- Panel 2: the geographic frame ------------------------------------------
im2 = night_map(ax2, radiance_full, DISPLAY_EXTENT)
ax2.set_xlim(DISPLAY_EXTENT[0], DISPLAY_EXTENT[1])
ax2.set_ylim(DISPLAY_EXTENT[2], DISPLAY_EXTENT[3])
draw_country_outline(ax2, colour=COLOURS["gold"], lw=1.6)
label_cities(ax2, cities_within(DISPLAY_EXTENT, number=7), size=8.4)
scale_bar(ax2)
ax2.set_xlabel("longitude (°)")
ax2.set_ylabel("latitude (°)")
ax2.set_title("Geographic frame: (longitude, latitude)", pad=8, fontsize=11)
colour_bar(fig, im2, ax2)

figure_title(fig, "The same image, two reference frames",
             "Row 0 of the array is the NORTHERNMOST: latitude decreases as the row index grows.\n"
             "Forgetting that minus sign is where almost every upside-down map comes from.",
             f"Source : {SAMPLE_FILE.name}   ·   layer {main_layer}"
             + ("   ·   DEMONSTRATION MODE — simulated values" if DEMO_MODE else ""))
plt.show()
````

### 11 · Quality flags

<div style="border-left:6px solid #00A86A;background:#E8F5EF;border-radius:0 12px 12px 0;

 <div style="color:#D49A00;font-size:11px;font-weight:700;letter-spacing:2.5px;text-transform:uppercase;">
  STEP 11 · 20 &nbsp;&nbsp;<span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span  <div style="color:#231F20;font-size:14px;margin-top:4px;">Knowing which pixels to trust</div>
</div>

**The principle.** NASA attaches to every pixel a small integer code summarising how much
confidence it deserves. Its meaning **depends on the product**, and good practice is to read the coding
from the `description` attribute in the file rather than assume it.

For the composites (A3, A4) — layer `…_Quality`:

| code | meaning | should you keep it? |
|:---:|---|---|
| **0** | good quality | yes, always |
| **1** | poor quality | depends on the use |
| **2** | gap-filled value | avoid for a published indicator |
| **255** | fill value | never |

For the daily product (A2) — layer `Mandatory_Quality_Flag`:

| code | meaning | comment |
|:---:|---|---|
| **0** | high quality, **persistent** lights | the heart of the signal: cities, infrastructure |
| **1** | high quality, **ephemeral** lights | fires, gas flares, fishing, seasonal activity |
| **2** | poor quality, outlier corrected | handle with care |
| **255** | no retrieval | thick cloud, moon too bright, data absent |

````python
# =============================================================================
#  STEP 11 — QUALITY FLAGS
# =============================================================================

# Fallback codings, in case the file carries no 'description' attribute
FALLBACK_CODES = {
    "composite": {0: "Good quality", 1: "Poor quality",
                  2: "Gap-filled value", 255: "Fill value"},
    "daily": {0: "High quality - persistent lights",
              1: "High quality - ephemeral lights",
              2: "Poor quality - outlier corrected",
              255: "No retrieval"},
    "land_water": {0: "Shallow ocean", 1: "Land", 2: "Coastline",
                   3: "Shallow inland water", 4: "Ephemeral water",
                   5: "Deep inland water", 6: "Moderate or continental ocean", 7: "Deep ocean"},
}


def parse_description(text: str) -> dict:
    """Extract a {code: label} dictionary from a 'description' attribute.

    NASA descriptions have the form:
        "0 = Good quality 1 = Poor quality 2 = Gap filled 255 = Fill value"
    The look-ahead (?=...) lets us cut just before the next code.
    """
    if not text:
        return {}
    pattern = r"(\d+)\s*=\s*([^\n\t0-9][^\n\t]*?)(?=\s+\d+\s*=|$)"
    return {int(code): label.strip() for code, label in re.findall(pattern, str(text))}


def resolve_qa_max(legend, qa_max=None):
    """Highest quality code kept. "auto": the run of codes, from 0, that the
    legend labels as good / high quality (and not poor / low)."""
    qa_max = QA_MAX if qa_max is None else qa_max
    if str(qa_max).strip().lower() != "auto":
        return int(qa_max)
    good = {k for k, v in legend.items()
            if re.search(r"good|high|bonne|haute", str(v), re.I)
            and not re.search(r"poor|low|médiocre|mauvaise|faible", str(v), re.I)}
    k = 0
    while k + 1 in good:
        k += 1
    return k


def water_codes_from_legend(legend):
    """Land_Water_Mask codes to drop: those whose label names an ocean or the sea.
    Read from the file, because the coding is not the same in every collection."""
    codes = sorted(k for k, v in legend.items()
                   if k != 255 and re.search(r"ocean|\bsea\b|océan|\bmer\b", str(v), re.I))
    return codes or [0, 7]


def find_flag_layers(path_) -> list:
    """List the flag layers present, in a useful reading order."""
    found = []
    with h5py.File(path_, "r") as f:
        group = f[find_data_group(f)]
        for name, obj in group.items():
            if not (isinstance(obj, h5py.Dataset) and obj.ndim == 2):
                continue
            if (name.endswith("_Quality") or "Quality_Flag" in name
                    or name in ("Land_Water_Mask", "QF_Cloud_Mask", "Snow_Flag", "DNB_Platform")):
                found.append(name)
    priority = {"Quality": 0, "Land_Water_Mask": 1, "QF_Cloud_Mask": 2}
    return sorted(found, key=lambda n: (0 if "Quality" in n else priority.get(n, 3), n))


flag_layers = find_flag_layers(SAMPLE_FILE)
print(f"Flag layers detected: {', '.join(flag_layers)}\n")

# --- Analysing the main quality layer ----------------------------------------
quality_name = next((n for n in flag_layers if "Quality" in n), None)

if quality_name is None:
    print("No quality layer in this file.")
    quality_flag, quality_legend = None, {}
    QA_EFFECTIVE = resolve_qa_max({}, 0 if str(QA_MAX).lower() == "auto" else QA_MAX)
else:
    with h5py.File(SAMPLE_FILE, "r") as f:
        obj = f[find_data_group(f)][quality_name]
        quality_flag = obj[:]
        description = read_attribute(obj, "description", "") or read_attribute(obj, "long_name", "")

    quality_legend = parse_description(description)
    if not quality_legend:
        family = "daily" if info["product"] == "VNP46A2" else "composite"
        quality_legend = FALLBACK_CODES[family]
        print("  ('description' attribute missing - fallback coding applied)")

    QA_EFFECTIVE = resolve_qa_max(quality_legend)
    print(f"LAYER: {quality_name}")
    print(f"description: {str(description)[:110]}")
    print(f"threshold applied: codes <= {QA_EFFECTIVE}"
          + ("  (QA_MAX = \"auto\": read from the legend above)" if str(QA_MAX).lower() == "auto" else "") + "\n")

    codes, counts = np.unique(quality_flag, return_counts=True)
    distribution = pd.DataFrame({
        "code": codes,
        "meaning": [quality_legend.get(int(k), "undocumented code") for k in codes],
        "pixels": counts,
        "share_%": np.round(100 * counts / quality_flag.size, 2),
        "kept (threshold %d)" % QA_EFFECTIVE: ["yes" if int(k) <= QA_EFFECTIVE else "no" for k in codes],
    })
    display(distribution.set_index("code"))

    kept = int(np.sum(quality_flag <= QA_EFFECTIVE))
    print(f"\nWith the threshold (codes <= {QA_EFFECTIVE}): {kept:,} pixels kept out of "
          f"{quality_flag.size:,} ({100 * kept / quality_flag.size:.1f} %)"
          .replace(",", " "))

# --- The storyline moves on ---------------------------------------------------
if quality_flag is not None:
    radiance_quality = np.where(quality_flag <= QA_EFFECTIVE, radiance_full, np.nan)
    log_storyline(11, f"+ quality filter <= {QA_EFFECTIVE}", np.nansum(radiance_quality),
                  int(np.isfinite(radiance_quality).sum()), f"tile {info['tile']}",
                  "doubtful pixels leave the computation")
````

````python
# --- Figure: the quality flags, and the price of the threshold ---------------
if quality_flag is not None:
    palette_q = {0: COLOURS["green"], 1: COLOURS["gold"], 2: COLOURS["terra"],
                 3: COLOURS["teal"], 255: COLOURS["sage"]}

    fig, axes = plt.subplots(1, 3, figsize=(13.4, 5.5),
                             gridspec_kw={"width_ratios": [1, 1, 0.78]})
    fig.subplots_adjust(top=0.765, bottom=0.205, left=0.055, right=0.985, wspace=0.24)

    # 1. the map of the classes
    entries = categorical_map(axes[0], quality_flag, DISPLAY_EXTENT,
                                 quality_legend, palette_q,
                                 "Quality class, pixel by pixel")
    draw_country_outline(axes[0], colour=COLOURS["ink"], lw=1.2)
    scale_bar(axes[0], colour=COLOURS["ink"])
    axes[0].set_xlabel("longitude (°)")
    axes[0].set_ylabel("latitude (°)")
    legend_below_axes(axes[0], entries, columns=2)

    # 2. what the threshold keeps
    kept = (quality_flag <= QA_EFFECTIVE).astype("int16")
    entrees2 = categorical_map(
        axes[1], kept, DISPLAY_EXTENT,
        {0: "dropped by the threshold", 1: "kept"},
        {0: COLOURS["brick"], 1: COLOURS["green"]},
        f"What the threshold keeps (codes <= {QA_EFFECTIVE})")
    draw_country_outline(axes[1], colour=COLOURS["ink"], lw=1.2)
    axes[1].set_xlabel("longitude (°)")
    legend_below_axes(axes[1], entrees2, columns=2)

    # 3. the trade-off curve
    thresholds = [0, 1, 2, 3]
    shares = [100 * np.mean(quality_flag <= s) for s in thresholds]
    bars = axes[2].bar([str(s) for s in thresholds], shares, width=0.6, edgecolor="none",
                         color=[COLOURS["gold"] if s == QA_EFFECTIVE else COLOURS["green"]
                                for s in thresholds])
    for barre, part in zip(bars, shares):
        axes[2].text(barre.get_x() + barre.get_width() / 2, part + 2.0, f"{part:.1f} %",
                     ha="center", fontsize=9, fontweight="bold", color=COLOURS["forest"])
    axes[2].set_ylim(0, 112)
    axes[2].set_xlabel("quality threshold")
    axes[2].set_ylabel("pixels kept (%)")
    axes[2].set_title("The threshold is a trade-off", pad=8, fontsize=10.5)
    axes[2].spines["left"].set_visible(False)
    axes[2].tick_params(left=False)
    axes[2].set_xticks(range(len(thresholds)))
    axes[2].set_xticklabels([f"{s}\n◄ your setting" if s == QA_EFFECTIVE else str(s)
                             for s in thresholds], fontsize=9)
    for label in axes[2].get_xticklabels():
        if "◄" in label.get_text():
            label.set_color(COLOURS["ochre"])
            label.set_fontweight("bold")

    figure_title(fig, "Quality flags: which pixels can you trust?",
                 f"Layer '{quality_name}'. The lower the threshold, the sounder the analysis — and the less complete the coverage.\n"
                 "Look for patterns on the left-hand map: bands betray the orbits, patches betray the clouds.",
                 f"Source : {SAMPLE_FILE.name}"
                 + ("   ·   DEMONSTRATION MODE — simulated distribution" if DEMO_MODE else ""))
    plt.show()
````

 <div style="color:#00A86A;font-weight:700;font-size:13.5px;">&#128200;&nbsp; How to read these three panels</div>

A reader must be able to reproduce your figure exactly. It is also what will protect you the day somebody gets a different result from the same files.</div>

<div style="background:#F2F7FA;border:1px solid #0E7C8644;border-left:4px solid #0E7C86;border-radius:8px;padding:14px 18px;margin:14px 0;font-family:'Segoe UI',Calibri,sans-serif;"><div style="color:#0E7C86;font-weight:700;font-size:13.5px;">&#9999;&nbsp; Your turn — pick a threshold, and defend it</div><div style="color:#231F20;font-size:13.5px;margin-top:6px;line-height:1.6;">Go back to step 02, change <code>QA_MAX</code> from <code>"auto"</code> to <code>1</code>, then re-run steps 02 to 12.<br><b>(a)</b> How many pixels do you gain? &nbsp; <b>(b)</b> On how many nights do those pixels rest? &nbsp; <b>(c)</b> For a 2014-2024 series meant to measure a <i>trend</i>, which of the two thresholds would you defend before a validation committee?</div></div>

<details><summary><i>Show the answer</i></summary>

**(a)** The third panel of the step 11 figure answers directly: it is the gap between bar `0` and bar `1`.

**(b)** On an annual or monthly composite, code `1` means *poor quality*: three nights or fewer behind the pixel. Map the `…_Num` layer (step 16) and you will find them in the cloudiest areas. Their value can be right or wrong by a wide margin — a single cloudy or moonlit night is enough to move it.

**(c)** `QA_MAX = "auto"` (code `0` only) for a trend. Pixels resting on three nights add year-to-year noise unrelated to the phenomenon measured. What matters most is to **document the choice**, and to show both series when they diverge.

</details>

### 12 · The masks: land, water and clouds

<div style="border-left:6px solid #00A86A;background:#E8F5EF;border-radius:0 12px 12px 0;
 padding:14px 22px;margin-top:22px;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
 <div style="color:#D49A00;font-size:11px;font-weight:700;letter-spacing:2.5px;text-transform:uppercase;">
  STEP 12 · 20 &nbsp;&nbsp;<span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#00A86A;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span></div>
 <div style="color:#231F20;font-size:14px;margin-top:4px;">Including bitwise reading, which frightens people for no good reason</div>
</div>

**Two masks, two very different logics.**

`Land_Water_Mask` is simple: one code from 0 to 7 per pixel, each code naming a surface type. It is what
lets you exclude the ocean — indispensable, because the sea is not black: moonglint, ships and fisheries
produce a signal there that would artificially inflate any coastal sum of lights.

`QF_Cloud_Mask` (daily product A2) is different: **a single 16-bit integer encodes about ten pieces of
information**, each occupying one or more bits. Reading that number as an ordinary integer makes no sense;
you have to **extract slices of bits** from it. The technique, called *bit masking*, fits on one line:

```python
field_value = (integer >> first_bit_position) & ((1 << number_of_bits) - 1)
```

You shift the integer to the right to bring the field you want all the way right, then erase everything
above it. The main fields of the VIIRS `QF_Cloud_Mask`:

<table style="width:100%;border-collapse:collapse;border:none;"><tr><td style="width:33.3333%;vertical-align:top;padding:5px;">
 <div style="background:#F4F7F5;border:1px solid #D5DED9;border-top:3px solid #0E7C86;border-radius:10px;
  padding:13px 15px;height:100%;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
  <div style="color:#00704A;font-weight:700;font-size:14px;margin-bottom:5px;">Bit 0 — Day / night</div>
  <div style="color:#231F20;font-size:13px;line-height:1.55;">0 = night, 1 = day. A pixel acquired in daylight means nothing for night-time lights.</div>
 </div></td><td style="width:33.3333%;vertical-align:top;padding:5px;">
 <div style="background:#F4F7F5;border:1px solid #D5DED9;border-top:3px solid #0E7C86;border-radius:10px;
  padding:13px 15px;height:100%;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
  <div style="color:#00704A;font-weight:700;font-size:14px;margin-bottom:5px;">Bits 1-3 — Land/water background</div>
  <div style="color:#231F20;font-size:13px;line-height:1.55;">A second description of the surface, finer than <code>Land_Water_Mask</code>.</div>
 </div></td><td style="width:33.3333%;vertical-align:top;padding:5px;">
 <div style="background:#F4F7F5;border:1px solid #D5DED9;border-top:3px solid #0E7C86;border-radius:10px;
  padding:13px 15px;height:100%;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
  <div style="color:#00704A;font-weight:700;font-size:14px;margin-bottom:5px;">Bits 4-5 — Cloud mask quality</div>
  <div style="color:#231F20;font-size:13px;line-height:1.55;">From 00 (poor) to 11 (high). If it is poor, the cloud detection that follows is worth little.</div>
 </div></td></tr><tr><td style="width:33.3333%;vertical-align:top;padding:5px;">
 <div style="background:#F4F7F5;border:1px solid #D5DED9;border-top:3px solid #0E7C86;border-radius:10px;
  padding:13px 15px;height:100%;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
  <div style="color:#00704A;font-weight:700;font-size:14px;margin-bottom:5px;">Bits 6-7 — Cloud detection</div>
  <div style="color:#231F20;font-size:13px;line-height:1.55;">00 = confident clear, 01 = probably clear, 10 = probably cloudy, 11 = confident cloudy.</div>
 </div></td><td style="width:33.3333%;vertical-align:top;padding:5px;">
 <div style="background:#F4F7F5;border:1px solid #D5DED9;border-top:3px solid #0E7C86;border-radius:10px;
  padding:13px 15px;height:100%;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
  <div style="color:#00704A;font-weight:700;font-size:14px;margin-bottom:5px;">Bit 8 — Shadow</div>
  <div style="color:#231F20;font-size:13px;line-height:1.55;">Cloud shadow detected over the pixel.</div>
 </div></td><td style="width:33.3333%;vertical-align:top;padding:5px;">
 <div style="background:#F4F7F5;border:1px solid #D5DED9;border-top:3px solid #0E7C86;border-radius:10px;
  padding:13px 15px;height:100%;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
  <div style="color:#00704A;font-weight:700;font-size:14px;margin-bottom:5px;">Bit 9 — Cirrus</div>
  <div style="color:#231F20;font-size:13px;line-height:1.55;">Thin high-altitude cloud — often invisible to the eye on the image.</div>
 </div></td></tr></table>

````python
# =============================================================================
#  STEP 12 — THE MASKS: LAND / WATER, THEN CLOUDS BIT BY BIT
# =============================================================================

# --- 1. The land / water mask -------------------------------------------------
with h5py.File(SAMPLE_FILE, "r") as f:
    group = f[find_data_group(f)]
    has_land_water = "Land_Water_Mask" in group
    if has_land_water:
        obj = group["Land_Water_Mask"]
        land_water_mask = obj[:]
        land_water_desc = read_attribute(obj, "description", "")

if has_land_water:
    land_water_legend = parse_description(land_water_desc) or FALLBACK_CODES["land_water"]
    codes, counts = np.unique(land_water_mask, return_counts=True)
    print("LAND / WATER MASK  -  Land_Water_Mask\n")
    display(pd.DataFrame({
        "code": codes,
        "surface type": [land_water_legend.get(int(k), "undocumented") for k in codes],
        "pixels": counts,
        "share_%": np.round(100 * counts / land_water_mask.size, 2),
    }).set_index("code"))

    # Codes treated as water to be excluded: read from the legend (ocean, sea)
    WATER_CODES = water_codes_from_legend(land_water_legend)
    water_share = 100 * np.mean(np.isin(land_water_mask, WATER_CODES))
    print(f"\nWith MASK_WATER = {MASK_WATER}, codes {WATER_CODES} would be dropped "
          f"({water_share:.1f} % of the tile).")
else:
    land_water_mask = None
    print("This product holds no Land_Water_Mask layer.")


# --- 2. The cloud mask, read bit by bit --------------------------------------
CLOUD_FIELDS = {
    "Day / night":            (0, 1, {0: "Night", 1: "Day"}),
    "Land / water background": (1, 3, {0: "Land and desert", 1: "Land, no desert",
                                       2: "Inland water", 3: "Sea water",
                                       5: "Coastal zone"}),
    "Cloud mask quality":     (4, 2, {0: "Poor", 1: "Low",
                                      2: "Medium", 3: "High"}),
    "Cloud detection":        (6, 2, {0: "Confident clear", 1: "Probably clear",
                                      2: "Probably cloudy", 3: "Confident cloudy"}),
    "Cloud shadow":           (8, 1, {0: "No", 1: "Yes"}),
    "Cirrus":                 (9, 1, {0: "No", 1: "Yes"}),
}


def extract_bits(integer, first_bit, number_of_bits):
    """Extract a bit field from an integer (or an array of integers)."""
    return (np.asarray(integer) >> first_bit) & ((1 << number_of_bits) - 1)


# We look for a daily file, the only one that carries QF_Cloud_Mask
daily_files = INVENTORY[INVENTORY["product"] == "VNP46A2"]
if daily_files.empty:
    print("\nNo VNP46A2 file in the inventory: "
          "demonstrating the bit decoding on a made-up integer.\n")
    sample_integer = np.uint16(0b0000_0010_1000_0010)
    cloud_array = None
else:
    daily_path = Path(daily_files.iloc[-1]["path"])
    with h5py.File(daily_path, "r") as f:
        daily_group = f[find_data_group(f)]
        cloud_array = daily_group["QF_Cloud_Mask"][:] if "QF_Cloud_Mask" in daily_group else None
    print(f"\nCLOUD MASK  -  QF_Cloud_Mask   ({daily_path.name})\n")
    sample_integer = (np.uint16(cloud_array.flat[0]) if cloud_array is not None
                      else np.uint16(0b0000_0010_1000_0010))

# Decoding one integer, field by field
print(f"Example: the integer {int(sample_integer)} is, in binary, "
      f"{int(sample_integer):016b}\n")
bit_rows = []
for field_name, (first, width, meanings) in CLOUD_FIELDS.items():
    value = int(extract_bits(sample_integer, first, width))
    position = (f"bit {first}" if width == 1
                else f"bits {first}-{first + width - 1}")
    bit_rows.append({"field": field_name, "position": position,
                     "bits": format(value, f"0{width}b"),
                     "value": value,
                     "meaning": meanings.get(value, "undocumented code")})
display(pd.DataFrame(bit_rows).set_index("field"))

# Statistics over the whole tile
if cloud_array is not None:
    detection = extract_bits(cloud_array, 6, 2)
    detection_meanings = CLOUD_FIELDS["Cloud detection"][2]
    codes, counts = np.unique(detection, return_counts=True)
    print("\nDISTRIBUTION OF CLOUD DETECTION OVER THE WHOLE TILE")
    display(pd.DataFrame({
        "code": codes,
        "meaning": [detection_meanings.get(int(k), "?") for k in codes],
        "pixels": counts,
        "share_%": np.round(100 * counts / detection.size, 2),
    }).set_index("code"))
    cloudy_share = 100 * np.mean(detection >= 2)
    print(f"\nNight analysed: {cloudy_share:.1f} % of the tile is probably or "
          f"confidently cloudy.")
    if cloudy_share > 40:
        print("→ Beyond 40 %, a conclusion drawn from this single night would be unwise.")

# --- The storyline moves on ---------------------------------------------------
radiance_after_masks = radiance_full.copy()
if land_water_mask is not None and MASK_WATER:
    radiance_after_masks[np.isin(land_water_mask, WATER_CODES)] = np.nan
if quality_flag is not None:
    radiance_after_masks[quality_flag > QA_EFFECTIVE] = np.nan
log_storyline(12, "+ water mask", np.nansum(radiance_after_masks),
              int(np.isfinite(radiance_after_masks).sum()), f"tile {info['tile']}",
              "the ocean no longer counts as dark territory")
````

````python
# --- Figure: the masks, and what is left once they are applied ---------------
panels = []
if land_water_mask is not None:
    panels.append(("land_water", land_water_mask))
if "cloud_array" in dir() and cloud_array is not None:
    panels.append(("nuage", extract_bits(cloud_array, 6, 2)))

if panels:
    fig, axes = plt.subplots(1, len(panels) + 1,
                             figsize=(4.55 * (len(panels) + 1), 5.5))
    fig.subplots_adjust(top=0.765, bottom=0.205, left=0.055, right=0.95, wspace=0.22)
    axes = np.atleast_1d(axes)

    for ax, (kind, data_) in zip(axes, panels):
        if kind == "land_water":
            code_colours = {0: "#2C7DA0", 1: "#BFE0CE", 2: COLOURS["gold"],
                              3: "#4EA8DE", 4: "#89C2D9", 5: "#1B3B6F",
                              6: COLOURS["green"], 7: "#0A2342"}
            entries = categorical_map(ax, data_, DISPLAY_EXTENT, land_water_legend,
                                         code_colours, "Land_Water_Mask — surface type")
        else:
            code_colours = {0: COLOURS["mint"], 1: "#BFE3D0",
                              2: COLOURS["gold"], 3: COLOURS["brick"]}
            entries = categorical_map(ax, data_, DISPLAY_EXTENT,
                                         CLOUD_FIELDS["Cloud detection"][2],
                                         code_colours,
                                         "QF_Cloud_Mask, bits 6-7 — cloud detection")
        draw_country_outline(ax, colour=COLOURS["ink"], lw=1.2)
        ax.set_xlabel("longitude (°)")
        legend_below_axes(ax, entries, columns=2)
    axes[0].set_ylabel("latitude (°)")
    scale_bar(axes[0], colour=COLOURS["ink"])

    # Last panel: the cumulative effect of the masks on the radiance
    ax = axes[-1]
    masked_radiance = radiance_full.copy()
    if land_water_mask is not None and MASK_WATER:
        masked_radiance[np.isin(land_water_mask, WATER_CODES)] = np.nan
    if quality_flag is not None:
        masked_radiance[quality_flag > QA_EFFECTIVE] = np.nan
    image = night_map(ax, masked_radiance, DISPLAY_EXTENT)
    ax.set_xlim(DISPLAY_EXTENT[0], DISPLAY_EXTENT[1])
    ax.set_ylim(DISPLAY_EXTENT[2], DISPLAY_EXTENT[3])
    draw_country_outline(ax, colour=COLOURS["gold"], lw=1.6)
    scale_bar(ax)
    remaining = 100 * np.mean(np.isfinite(masked_radiance))
    ax.set_title(f"Radiance after masking — {remaining:.0f} % of pixels", pad=8, fontsize=10.5)
    ax.set_xlabel("longitude (°)")
    legend_below_axes(ax, [("pixels dropped (grey)", NO_DATA_GREY)], columns=1)
    colour_bar(fig, image, ax)

    figure_title(fig, "The masks, and what is left once they are applied",
                 "In grey: the pixels dropped. They are the ones that must never enter a sum or a mean.\n"
                 "On a coastal tile the ocean can be 30 to 50 % of the pixels — and it is not black.",
                 f"Source : {SAMPLE_FILE.name}"
                 + ("   ·   DEMONSTRATION MODE" if DEMO_MODE else ""))
    plt.show()
````

 <div style="color:#00704A;font-weight:700;font-size:13.5px;">&#128161;&nbsp; Why masking water really does change the figures</div>
 <div style="color:#231F20;font-size:13.5px;margin-top:5px;line-height:1.6;">On a coastal tile, the ocean can account for 30 to 50 % of the pixels. ### 13 · The reading function

<div style="border-left:6px solid #F5C242;background:#E8F5EF;border-radius:0 12px 12px 0;
 padding:14px 22px;margin-top:22px;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
 <div style="color:#D49A00;font-size:11px;font-weight:700;letter-spacing:2.5px;text-transform:uppercase;">
  STEP 13 · 20 &nbsp;&nbsp;<span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#F5C242;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span></div>
 <div style="color:#231F20;font-size:14px;margin-top:4px;">Steps 07 to 12 fit into a single function</div>
</div>

**What this step does.** Everything we took apart separately is put back together here into a
single function, `read_ntl()`. It applies the six operations **in the right order**, and that order is not
negotiable:

1. open the file and **find** the data group (never assume it);
2. **pick** the radiance layer;
3. read the attributes and **decode**: `value × scale_factor + add_offset`;
4. replace the **`_FillValue` with `NaN`** — before any computation;
5. drop the pixels whose **quality flag** exceeds the threshold;
6. drop the **water** pixels, then compute the georeferencing.

Swapping steps 3 and 4 — decoding before neutralising the `_FillValue` — gives you exactly the disaster
quantified in step 09. The function returns a dictionary holding the radiance, the extent, the affine
transform and a **processing log**: the number of pixels lost at each filter.

<div style="background:#E8F5EF;border:1px solid #00704A44;border-left:4px solid #00704A;border-radius:8px;
 padding:13px 17px;margin:10px 0;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
 <div style="color:#00704A;font-weight:700;font-size:13.5px;">&#128273;&nbsp; Why a processing log?</div>
 <div style="color:#231F20;font-size:13.5px;margin-top:5px;line-height:1.6;">Because an indicator without traceability cannot be published. The <code>log</code> field returned by <code>read_ntl()</code> tells you how many pixels were dropped, and by which filter. Copied into your metadata, it answers in advance the question the first attentive reviewer will ask: «&nbsp;what exactly was this figure computed on?&nbsp;»</div>
</div>

````python
# =============================================================================
#  STEP 13 — THE CLEAN READING FUNCTION
# =============================================================================
WATER_CODES_TO_DROP = [0, 7]      # fallback only (old MODIS-style coding); the file's legend wins


def _downsample(array, factor):
    """Mean of factor × factor blocks, ignoring NaNs."""
    if factor is None or factor <= 1:
        return array
    h, w = array.shape
    h2, w2 = (h // factor) * factor, (w // factor) * factor
    cut = array[:h2, :w2].reshape(h2 // factor, factor, w2 // factor, factor)
    return np.nanmean(cut, axis=(1, 3))


def read_ntl(path_, layer=None, qa_max=None, mask_water=None,
             downsample=1, verbose=False):
    """Read a VNP46 file and return a clean, georeferenced radiance.

    Parameters
    ----------
    path_ : str or Path
        The .h5 file to read.
    layer : str, optional
        Name of the radiance layer. Chosen automatically by default.
    qa_max : int, optional
        Highest quality flag accepted. Defaults to the value of QA_MAX.
    mask_water : bool, optional
        Drop ocean pixels. Defaults to the value of MASK_WATER.
    downsample : int
        Kept for compatibility and ignored: the data keeps its full resolution.
        DOWNSAMPLE only changes how the maps are drawn (night_map).
    verbose : bool
        Print the processing log as it goes.

    Returns
    -------
    dict with keys: radiance, extent, transform, longitudes, latitudes,
    layer, metadata, log.
    """
    path_ = Path(path_)
    qa_max = QA_MAX if qa_max is None else qa_max
    mask_water = MASK_WATER if mask_water is None else mask_water

    name_info = decode_filename(path_.name)
    if name_info is None:
        raise ValueError(f"Filename does not follow the VNP46 convention: {path_.name}")

    log = []

    with h5py.File(path_, "r") as f:
        # --- 1. find the group, never assume it ------------------------------
        group = f[find_data_group(f)]

        # --- 2. pick the radiance layer --------------------------------------
        layer_name = layer or pick_radiance_layer(group)
        ds = group[layer_name]

        # --- 3. read the attributes FROM the file ----------------------------
        factor = read_attribute(ds, "scale_factor", 1.0)
        offset = read_attribute(ds, "add_offset", read_attribute(ds, "offset", 0.0))
        empty = read_attribute(ds, "_FillValue", None)
        units_ = read_attribute(ds, "units", "")

        raw = ds[:].astype("float32")
        total = raw.size
        log.append(("pixels in the file", total, 100.0))

        # --- 4. neutralise the _FillValue BEFORE decoding --------------------
        invalid = ~np.isfinite(raw)
        if empty is not None:
            invalid |= np.isclose(raw, float(empty))
        radiance = raw * factor + offset
        radiance[invalid] = np.nan
        radiance[radiance < 0] = np.nan          # a negative radiance does not exist
        lost = int(invalid.sum())
        log.append(("dropped: _FillValue", -lost, -100.0 * lost / total))

        # --- 5. filter on the quality flag -----------------------------------
        flag_name = next((n for n in (f"{layer_name}_Quality", "Mandatory_Quality_Flag")
                          if n in group), None)
        if flag_name is not None and qa_max is not None:
            flag = group[flag_name][:]
            legend_q = parse_description(read_attribute(group[flag_name], "description", "")) or \
                FALLBACK_CODES["daily" if name_info["product"].upper() in ("VNP46A1", "VNP46A2") else "composite"]
            qa_max = resolve_qa_max(legend_q, qa_max)          # "auto" -> read from the legend
            to_drop = (flag > qa_max) & np.isfinite(radiance)
            radiance[to_drop] = np.nan
            log.append((f"dropped: quality > {qa_max}", -int(to_drop.sum()),
                        -100.0 * int(to_drop.sum()) / total))

        # --- 6. mask the water ------------------------------------------------
        if mask_water and "Land_Water_Mask" in group:
            land_water = group["Land_Water_Mask"][:]
            water_codes = water_codes_from_legend(parse_description(
                read_attribute(group["Land_Water_Mask"], "description", ""))) if parse_description(
                read_attribute(group["Land_Water_Mask"], "description", "")) else WATER_CODES_TO_DROP
            to_drop = np.isin(land_water, water_codes) & np.isfinite(radiance)
            radiance[to_drop] = np.nan
            log.append(("dropped: water", -int(to_drop.sum()),
                        -100.0 * int(to_drop.sum()) / total))

        metadata = {"product": name_info["product"], "tile": name_info["tile"],
                    "date": name_info["date"].date(), "layer": layer_name,
                    "scale_factor": factor, "add_offset": offset,
                    "_FillValue": empty, "units": units_,
                    "quality_flag": flag_name, "qa_max": qa_max,
                    "water_masked": bool(mask_water), "file": path_.name}

    # --- 7. georeferencing -----------------------------------------------------
    w, e, s, n = tile_extent(name_info["tile"])
    # No downsampling here: every sum and statistic uses the full resolution.
    rows_, cols_ = radiance.shape
    res_lon, res_lat = (e - w) / cols_, (n - s) / rows_

    log.append(("usable pixels", int(np.isfinite(radiance).sum()),
                100.0 * float(np.isfinite(radiance).mean())))

    if verbose:
        print(f"{path_.name}")
        for label_, value, share in log:
            print(f"   {label_:<28} {value:>12,}   {share:>7.2f} %".replace(",", " "))

    return {"radiance": radiance,
            "extent": (w, e, s, n),
            "transform": (res_lon, 0.0, w, 0.0, -res_lat, n),
            "longitudes": w + (np.arange(cols_) + 0.5) * res_lon,
            "latitudes": n - (np.arange(rows_) + 0.5) * res_lat,
            "layer": metadata["layer"],
            "metadata": metadata,
            "log": pd.DataFrame(log, columns=["step", "pixels", "share_%"])}


# --- Putting it to work -------------------------------------------------------
RESULT = read_ntl(SAMPLE_FILE, downsample=DOWNSAMPLE)

print("PROCESSING LOG")
display(RESULT["log"].set_index("step"))

print("\nOUTPUT METADATA - copy this into your documentation")
display(pd.Series(RESULT["metadata"], name="value").to_frame())

radiance = RESULT["radiance"]
valid = radiance[np.isfinite(radiance)]
print(f"\nRadiance: {valid.size:,} usable pixels".replace(",", " "))
print(f"   minimum {valid.min():.3f}   median {np.median(valid):.3f}   "
      f"maximum {valid.max():.1f}   (nW·cm⁻²·sr⁻¹)")
print(f"   sum of lights (SoL): {valid.sum():,.0f}".replace(",", " "))

# --- The storyline: the function must reproduce the step 12 figure ------------
log_storyline(13, "via read_ntl() (check)", valid.sum(), valid.size,
              f"tile {RESULT['metadata']['tile']}",
              "same result as step 12: the function is faithful")
````

 padding:13px 17px;margin:10px 0;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
 <div style="color:#00A86A;font-weight:700;font-size:13.5px;">&#128200;&nbsp; How to read the log</div>
 <div style="color:#231F20;font-size:13.5px;margin-top:5px;line-height:1.6;">Every negative line is a deliberate loss. Add them up: if you drop more than half the tile, that is not necessarily abnormal (a coastal tile is half ocean), but you must be able to explain it. A sharp loss on the «&nbsp;quality&nbsp;» line usually signals a very cloudy period — that is the moment to look at the <code>…_Num</code> layer.</div>

### 14 · From the square to the country

<div style="border-left:6px solid #F5C242;background:#E8F5EF;border-radius:0 12px 12px 0;
 padding:14px 22px;margin-top:22px;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
 <div style="color:#D49A00;font-size:11px;font-weight:700;letter-spacing:2.5px;text-transform:uppercase;">
  STEP 14 · 20 &nbsp;&nbsp;<span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#F5C242;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span></div>
 <div style="color:#231F20;font-size:14px;margin-top:4px;">Mosaic the tiles, then clip on a border</div>
</div>

**The missing link.** So far we have worked on *one tile*. But a tile is not a territory: it is an arbitrary 10° × 10° square that ignores borders. No publishable indicator can come out of it as it stands.

Two operations separate the tile from the indicator, and this step performs both:

1. **Mosaic** — paste the tiles covering the country back together into a common grid. Because they share the same resolution and the same projection, the pasting is a simple index calculation: no resampling, hence no distortion of the values.
2. **Clip** — drop every pixel outside the border, with `country_mask()`.

The order matters: mosaic first, clip second. The other way round would force you to handle pixels straddling two tiles along the border.

````python
# =============================================================================
#  STEP 14 — FROM THE SQUARE TO THE COUNTRY: MOSAIC, THEN CLIP
# =============================================================================

# --- 1. Which tiles do we assemble? ------------------------------------------
batch = INVENTORY[INVENTORY["product"] == MAIN_PRODUCT]
if batch.empty:
    batch = INVENTORY
target_year = int(batch["year"].max())
batch = batch[batch["year"] == target_year]
batch_tiles = [t for t in COUNTRY_TILES if t in set(batch["tile"])] or sorted(batch["tile"].unique())

print(f"MOSAIC  ·  product {MAIN_PRODUCT}  ·  year {target_year}  ·  {COUNTRY_NAME}")
print(f"  tiles of the country : {', '.join(COUNTRY_TILES)}")
print(f"  tiles available      : {', '.join(batch_tiles)}")
if set(COUNTRY_TILES) - set(batch_tiles):
    print(f"  ⚠ missing            : {', '.join(sorted(set(COUNTRY_TILES) - set(batch_tiles)))}")
    print("    (the mosaic will be incomplete: the final figure will be understated)")
print()

# --- 2. Read every tile with the SAME function and the SAME settings ---------
pieces = {}
for t in batch_tiles:
    path_t = Path(batch[batch["tile"] == t].iloc[-1]["path"])
    pieces[t] = read_ntl(path_t, downsample=DOWNSAMPLE)
    print(f"  {t} : {pieces[t]['radiance'].shape[0]} × "
          f"{pieces[t]['radiance'].shape[1]} pixels")

# --- 3. Paste the pieces into one common grid --------------------------------
o_g = min(pieces[t]["extent"][0] for t in batch_tiles)
e_g = max(pieces[t]["extent"][1] for t in batch_tiles)
s_g = min(pieces[t]["extent"][2] for t in batch_tiles)
n_g = max(pieces[t]["extent"][3] for t in batch_tiles)

first = pieces[batch_tiles[0]]
mosaic_resolution = (first["extent"][1] - first["extent"][0]) / first["radiance"].shape[1]

MOSAIC = np.full((int(round((n_g - s_g) / mosaic_resolution)),
                    int(round((e_g - o_g) / mosaic_resolution))), np.nan, dtype="float32")
for t, res_t in pieces.items():
    o_t, _, _, n_t = res_t["extent"]
    col0 = int(round((o_t - o_g) / mosaic_resolution))
    lig0 = int(round((n_g - n_t) / mosaic_resolution))
    h_t, w_t = res_t["radiance"].shape
    MOSAIC[lig0:lig0 + h_t, col0:col0 + w_t] = res_t["radiance"]

MOSAIC_EXTENT = (o_g, e_g, s_g, n_g)
LON_M, LAT_M = coordinate_grid(MOSAIC_EXTENT, MOSAIC.shape)

print(f"\n  Mosaic obtained : {MOSAIC.shape[0]} × {MOSAIC.shape[1]} pixels")
print(f"  Extent          : longitude [{o_g:+.0f}° , {e_g:+.0f}°]  ·  "
      f"latitude [{s_g:+.0f}° , {n_g:+.0f}°]")

# --- 4. Clip on the national border ------------------------------------------
INSIDE_COUNTRY = country_mask(LON_M, LAT_M)
COUNTRY_RADIANCE = np.where(INSIDE_COUNTRY, MOSAIC, np.nan)

valid_mosaic = MOSAIC[np.isfinite(MOSAIC)]
valid_country = COUNTRY_RADIANCE[np.isfinite(COUNTRY_RADIANCE)]

# Pixel area: on a grid in degrees it shrinks with the cosine of latitude (-13 % at 30°).
# The plain sum is fine inside one country; to compare countries far apart in latitude,
# use the area-weighted sum (radiance x km²).
PIXEL_KM2 = (mosaic_resolution * 111.32) ** 2 * np.cos(np.deg2rad(LAT_M))
sol_area_country = float(np.nansum(COUNTRY_RADIANCE * PIXEL_KM2))

display(pd.DataFrame([
    {"perimeter": "raw mosaic (the whole tiles)",
     "usable pixels": f"{valid_mosaic.size:,}".replace(",", " "),
     "sum of lights": f"{valid_mosaic.sum():,.0f}".replace(",", " "),
     "median radiance": round(float(np.median(valid_mosaic)), 3)},
    {"perimeter": f"national territory — {COUNTRY_NAME}",
     "usable pixels": f"{valid_country.size:,}".replace(",", " "),
     "sum of lights": f"{valid_country.sum():,.0f}".replace(",", " "),
     "median radiance": round(float(np.median(valid_country)), 3) if valid_country.size else np.nan,
     "area-weighted SoL (radiance x km²)": f"{sol_area_country:,.0f}".replace(",", " ")},
]).set_index("perimeter"))

share_kept = 100 * valid_country.size / max(valid_mosaic.size, 1)
print(f"\nThe territory of {COUNTRY_NAME} accounts for only {share_kept:.1f} % of the mosaic's")
print("usable pixels. Everything else belongs to neighbouring countries or to the sea:")
print("including it in a national indicator would make no sense at all.\n")

# --- 5. The storyline moves on ------------------------------------------------
tile_perimeter = ("the whole tile" if len(batch_tiles) == 1
                    else f"{len(batch_tiles)} whole tiles")
log_storyline(14, "mosaic of the tiles", valid_mosaic.sum(), valid_mosaic.size,
                tile_perimeter,
                "same figure as step 13 when a single tile is enough")
log_storyline(14, "clipped on the border", valid_country.sum(), valid_country.size,
                f"{COUNTRY} territory", "the figure becomes a national indicator")
````

````python
# --- Figure: three scales, from the mosaic to the district -------------------
country_city_list = country_cities(number=9)
main_city = country_city_list[0] if country_city_list else None
o_p, e_p, s_p, n_p = COUNTRY_EXTENT
country_margin = max(0.6, 0.08 * max(e_p - o_p, n_p - s_p))

fig, axes = plt.subplots(1, 3, figsize=(13.6, 5.9))
fig.subplots_adjust(top=0.765, bottom=0.135, left=0.05, right=0.945, wspace=0.26)

# (1) the full mosaic, with the border laid over it
im1 = night_map(axes[0], MOSAIC, MOSAIC_EXTENT)
axes[0].set_xlim(MOSAIC_EXTENT[0], MOSAIC_EXTENT[1])
axes[0].set_ylim(MOSAIC_EXTENT[2], MOSAIC_EXTENT[3])
draw_country_outline(axes[0], colour=COLOURS["gold"], lw=1.7)
axes[0].add_patch(plt.Rectangle((o_p - country_margin, s_p - country_margin),
                                e_p - o_p + 2 * country_margin, n_p - s_p + 2 * country_margin,
                                facecolor="none", edgecolor="white", lw=1.0,
                                ls=(0, (4, 3)), alpha=0.85, zorder=7))
scale_bar(axes[0])
axes[0].set_title(f"1 · The mosaic — {' + '.join(batch_tiles)}", pad=8, fontsize=10.5)
axes[0].set_ylabel("latitude (°)")

# (2) the country alone, after clipping
im2 = night_map(axes[1], COUNTRY_RADIANCE, MOSAIC_EXTENT)
axes[1].set_xlim(o_p - country_margin, e_p + country_margin)
axes[1].set_ylim(s_p - country_margin, n_p + country_margin)
draw_country_outline(axes[1], colour=COLOURS["gold"], lw=1.7)
label_cities(axes[1], country_city_list, size=8.2)
scale_bar(axes[1])
axes[1].set_title(f"2 · {COUNTRY_NAME} alone — after clipping", pad=8, fontsize=10.5)

# (3) the main urban area
if main_city is not None:
    nom_v, clon, clat, _pop = main_city
    city_margin = 0.55
    window = ((np.abs(LON_M - clon) <= city_margin) & (np.abs(LAT_M - clat) <= city_margin)
               & np.isfinite(COUNTRY_RADIANCE))
    city_ceiling = (float(np.nanpercentile(COUNTRY_RADIANCE[window], 99.9))
                 if window.any() else None)
    im3 = night_map(axes[2], COUNTRY_RADIANCE, MOSAIC_EXTENT, vmax=city_ceiling,
                         interpolation="bilinear")
    axes[2].set_xlim(clon - city_margin, clon + city_margin)
    axes[2].set_ylim(clat - city_margin, clat + city_margin)
    draw_country_outline(axes[2], colour=COLOURS["gold"], lw=1.7)
    label_cities(axes[2], cities_within(
        (clon - city_margin, clon + city_margin, clat - city_margin, clat + city_margin), number=5), size=9.5)
    scale_bar(axes[2])
    axes[2].set_title(f"3 · {nom_v} — {int(2 * city_margin * 111):d} km across",
                      pad=8, fontsize=10.5)
    axes[1].add_patch(plt.Rectangle((clon - city_margin, clat - city_margin), 2 * city_margin, 2 * city_margin,
                                    facecolor="none", edgecolor="white", lw=1.1,
                                    ls=(0, (4, 3)), zorder=8))
    colour_bar(fig, im3, axes[2])
else:
    axes[2].axis("off")
    colour_bar(fig, im2, axes[1])

for ax in axes[:2]:
    ax.set_xlabel("longitude (°)")
axes[2].set_xlabel("longitude (°)")

figure_title(fig, f"Three scales, one single dataset — {COUNTRY_NAME}",
             "The mosaic assembles the tiles; the national mask drops everything that is not the country;\n"
             "the zoom shows that at 500 m resolution an urban area reads district by district.",
             f"Product {MAIN_PRODUCT}, year {target_year}. Borders: Natural Earth 1:50 M, public domain."
             + ("   ·   DEMONSTRATION MODE — simulated values" if DEMO_MODE else ""))
plt.show()
````

 padding:13px 17px;margin:10px 0;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
 <div style="color:#00A86A;font-weight:700;font-size:13.5px;">&#128200;&nbsp; How to read these three maps</div>
 <div style="color:#231F20;font-size:13.5px;margin-top:5px;line-height:1.6;"><b>1 · the mosaic</b> — the tiles assembled edge to edge, with no visible seam. <b>2 · the country alone</b> — the same dataset, clipped to the border: everything belonging to the neighbours or to the sea has turned grey. <b>3 · the city</b> — roughly 120 km across, which is the window an urban planner works in.<br><br>The first two maps share the <b>same scale</b>, so the comparison is honest. The third has a scale of its own, because a neighbourhood does not have the dynamic range of a country — and that is precisely the kind of detail that belongs in a methodological note.</div>
</div>

<div style="background:#E8F5EF;border:1px solid #00704A44;border-left:4px solid #00704A;border-radius:8px;

 <div style="color:#00704A;font-weight:700;font-size:13.5px;">&#128273;&nbsp; Why this step changes the nature of the figure</div>
 <div style="color:#231F20;font-size:13.5px;margin-top:5px;line-height:1.6;">Before it, the sum of lights described <i>a square</i>. After it, it describes <i>a country</i> — an entity somebody is accountable for, for which a GDP, a population and a census exist. <b>This is the precise moment a satellite measurement becomes a potential official statistic</b>, and that is why it has to be documented as one: which border, which version, which reference date.</div>
</div>

 padding:9px 17px;margin:14px 0 2px 0;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
 <span style="color:#D49A00;font-size:10.5px;font-weight:700;letter-spacing:2px;text-transform:uppercase;">Worth remembering</span>
 <span style="color:#231F20;font-size:13.5px;margin-left:12px;">Mosaic then clip: this is the operation that turns a satellite image into a territorial statistic.</span>

### 15 · The distribution of radiances

<div style="border-left:6px solid #00A86A;background:#E8F5EF;border-radius:0 12px 12px 0;
 padding:14px 22px;margin-top:22px;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
 <div style="color:#D49A00;font-size:11px;font-weight:700;letter-spacing:2.5px;text-transform:uppercase;">
  STEP 15 · 20 &nbsp;&nbsp;<span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#00A86A;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span></div>
 <div style="color:#231F20;font-size:14px;margin-top:4px;">The most important statistical property, and the most ignored</div>
</div>

**What you absolutely must understand.** Night-time radiance is not normally distributed. It is
**extremely skewed**: the vast majority of pixels are very dark, and a handful of urban pixels concentrate
most of the total light. It is not unusual for **1 % of the pixels to carry more than half the sum of
lights**.

Three practical consequences, which decide the quality of your analysis:

- **Mapping on a linear scale shows nothing**: the map is black, with three white dots. You have to switch
  to a **logarithmic scale** — which is why every map in this notebook uses a logarithmic colour scale, graduated in real units.
- **The mean is a poor summary.** It is dominated by a few pixels. For a comparison between regions, the
  **sum** (SoL) and the **median** say two different things, both useful.
- **A single outlying pixel can move a national indicator.** A gas flare, a fire, a ship: always map your
  extreme values before publishing.

````python
# =============================================================================
#  STEP 15 — THE DISTRIBUTION OF RADIANCES
# =============================================================================
valid = radiance[np.isfinite(radiance)]
positive = valid[valid > 0]

# --- Descriptive statistics ---------------------------------------------------
quantiles = [1, 5, 25, 50, 75, 90, 95, 99, 99.9]
summary = pd.DataFrame({
    "statistic": (["usable pixels", "mean", "standard deviation", "minimum"]
                  + [f"percentile {q}" for q in quantiles] + ["maximum", "sum (SoL)"]),
    "value": ([f"{valid.size:,}".replace(",", " "),
               f"{valid.mean():.3f}", f"{valid.std():.3f}", f"{valid.min():.3f}"]
              + [f"{np.percentile(valid, q):.3f}" for q in quantiles]
              + [f"{valid.max():.1f}", f"{valid.sum():,.0f}".replace(",", " ")]),
})
display(summary.set_index("statistic"))

# --- Concentration: what share of light for what share of pixels? -------------
sorted_values = np.sort(valid)[::-1]
cumulative = np.cumsum(sorted_values) / sorted_values.sum()
concentration = pd.DataFrame({
    "share of the brightest pixels": ["0.1 %", "1 %", "5 %", "10 %", "25 %", "50 %"],
    "share of total light": [
        f"{100 * cumulative[max(int(p * sorted_values.size / 100) - 1, 0)]:.1f} %"
        for p in (0.1, 1, 5, 10, 25, 50)],
})
print("\nCONCENTRATION OF LIGHT")
display(concentration.set_index("share of the brightest pixels"))

half_threshold = 100 * (np.searchsorted(cumulative, 0.5) + 1) / sorted_values.size
print(f"\nThe brightest {half_threshold:.2f} % of pixels are enough to reach "
      f"half of the sum of lights.")
````

````python
# --- Figure: four views of the same distribution -----------------------------
fig = plt.figure(figsize=(12.8, 7.8))
grille = fig.add_gridspec(2, 2, left=0.075, right=0.945, top=0.80, bottom=0.085,
                          hspace=0.46, wspace=0.26)
axA = fig.add_subplot(grille[0, 0])
axB = fig.add_subplot(grille[0, 1])
axC = fig.add_subplot(grille[1, 0])
axD = fig.add_subplot(grille[1, 1])

# (a) on a linear scale: everything is crushed to the left
axA.hist(valid, bins=90, color=COLOURS["sage"], edgecolor="none")
axA.set_yscale("log")
axA.set_title("(a) Linear scale: everything is crushed to the left", pad=7, fontsize=10.5)
axA.set_xlabel("radiance (nW·cm⁻²·sr⁻¹)")
axA.set_ylabel("number of pixels (log)")
axA.text(0.97, 0.9, "the map looks black,\nand the mean means nothing",
         transform=axA.transAxes, ha="right", va="top", fontsize=8.2,
         color=COLOURS["brick"], style="italic")

# (b) on a logarithmic scale: the shape appears
if positive.size:
    bornes = np.logspace(np.log10(max(positive.min(), 1e-3)), np.log10(positive.max()), 48)
    axB.hist(positive, bins=bornes, color=COLOURS["dark_green"], edgecolor="none")
    axB.set_xscale("log")
    axB.set_yscale("log")
    bas, haut = axB.get_ylim()
    for rank, (centile, colour, label) in enumerate(
            [(50, COLOURS["gold"], "median"), (99, COLOURS["brick"], "99th percentile")]):
        v = float(np.percentile(positive, centile))
        axB.axvline(v, color=colour, lw=1.8, ls="--", zorder=4)
        axB.text(v * 1.15, haut ** (0.92 - 0.14 * rank) * bas ** (0.08 + 0.14 * rank),
                 f"{label}\n{v:.2f}", color=colour, fontsize=8.4,
                 fontweight="bold", va="top", ha="left")
axB.set_title("(b) Logarithmic scale: the shape appears", pad=7, fontsize=10.5)
if float(RESULT["metadata"]["scale_factor"]) != 1.0:    # integers x 0.1: quantised values
    axB.text(0.5, 0.04, "the isolated bars on the left are the file's quantisation:\n"
                        "radiance is stored there in steps of 0.1 nW·cm⁻²·sr⁻¹",
             transform=axB.transAxes, ha="center", va="bottom", fontsize=7.6,
             color=COLOURS["slate"], style="italic")
axB.set_xlabel("radiance (log scale)")
axB.set_ylabel("number of pixels (log)")

# (c) the concentration curve
pixel_share = 100 * np.arange(1, sorted_values.size + 1) / sorted_values.size
pas = max(sorted_values.size // 4000, 1)
axC.plot(pixel_share[::pas], 100 * cumulative[::pas], lw=2.4, color=COLOURS["teal"], zorder=4)
axC.plot([0, 100], [0, 100], lw=1.1, ls=":", color=COLOURS["slate"], zorder=3)
axC.fill_between(pixel_share[::pas], 100 * cumulative[::pas], pixel_share[::pas],
                 color=COLOURS["teal"], alpha=0.14, zorder=2)
axC.axhline(50, color=COLOURS["gold"], lw=1.2, ls="--", zorder=3)
axC.scatter([half_threshold], [50], s=48, color=COLOURS["gold"], zorder=6)
axC.annotate(f"{half_threshold:.1f} % of the pixels\ncarry 50 % of the light",
             xy=(half_threshold, 50), xytext=(26, 26), fontsize=8.6,
             color=COLOURS["ochre"], fontweight="bold",
             arrowprops=dict(arrowstyle="->", color=COLOURS["ochre"], lw=1.3))
axC.set_xlim(0, 100)
axC.set_ylim(0, 102)
axC.set_xlabel("share of pixels, brightest to darkest (%)")
axC.set_ylabel("cumulative share of light (%)")
axC.set_title("(c) A handful of pixels does everything", pad=7, fontsize=10.5)
axC.text(72, 56, "perfectly even\ndistribution (theoretical)", fontsize=7.6,
         color=COLOURS["slate"], style="italic", ha="center")

# (d) the map, with the same convention as everywhere else
imD = night_map(axD, radiance, RESULT["extent"])
axD.set_xlim(RESULT["extent"][0], RESULT["extent"][1])
axD.set_ylim(RESULT["extent"][2], RESULT["extent"][3])
draw_country_outline(axD, colour=COLOURS["gold"], lw=1.5)
scale_bar(axD)
axD.set_title("(d) The matching map, on a log scale", pad=7, fontsize=10.5)
axD.set_xlabel("longitude (°)")
axD.set_ylabel("latitude (°)")
colour_bar(fig, imD, axD, fraction=0.05)

figure_title(fig, "Night-time radiance is a highly skewed quantity",
             "Ignoring that skew means producing black maps, misleading means,\n"
             "and indicators that a single outlying pixel can tip over.",
             f"Source : {RESULT['metadata']['file']}   ·   layer {RESULT['layer']}"
             + ("   ·   DEMONSTRATION MODE — simulated values" if DEMO_MODE else ""))
plt.show()
````

### 16 · Comparing the layers with one another

<div style="border-left:6px solid #00A86A;background:#E8F5EF;border-radius:0 12px 12px 0;
 padding:14px 22px;margin-top:22px;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
 <div style="color:#D49A00;font-size:11px;font-weight:700;letter-spacing:2.5px;text-transform:uppercase;">
  STEP 16 · 20 &nbsp;&nbsp;<span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#00A86A;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span></div>
 <div style="color:#231F20;font-size:14px;margin-top:4px;">Why the file holds several versions of the same image</div>
</div>

**What this comparison reveals.** A composite offers several views of the same scene:
`AllAngle` (all viewing geometries pooled), `NearNadir` (only the near-vertical overpasses), `OffNadir`
(the oblique ones). They do not give exactly the same figure.

On top of that come the two uncertainty layers, the ones most often forgotten: `…_Num` says **over how many
nights** the composite was computed, and `…_Std` **how much those nights diverged**. A pixel with high
radiance and a high standard deviation, computed over three nights, does not carry the same evidential
weight as a stable pixel computed over twenty-five — and yet the two look alike on a radiance map.

````python
# =============================================================================
#  STEP 16 — COMPARING THE LAYERS WITH ONE ANOTHER
# =============================================================================
with h5py.File(SAMPLE_FILE, "r") as f:
    group = f[find_data_group(f)]
    available = {name: obj for name, obj in group.items()
                 if isinstance(obj, h5py.Dataset) and obj.ndim == 2}

    # (1) the radiance variants
    variants = [n for n in available
                if ("Composite" in n or "NTL" in n)
                and not n.endswith(("_Num", "_Std", "_Quality"))]

    rows = []
    arrays = {}
    for name in variants:
        obj = group[name]
        fac = read_attribute(obj, "scale_factor", 1.0)
        off = read_attribute(obj, "add_offset", 0.0)
        nod = read_attribute(obj, "_FillValue", None)
        values = obj[:].astype("float32")
        values = np.where(np.isclose(values, float(nod)) if nod is not None else False,
                          np.nan, values * fac + off)
        arrays[name] = values
        ok = values[np.isfinite(values)]
        rows.append({"layer": name, "valid_pixels": ok.size,
                     "mean": round(float(ok.mean()), 3) if ok.size else np.nan,
                     "median": round(float(np.median(ok)), 3) if ok.size else np.nan,
                     "maximum": round(float(ok.max()), 1) if ok.size else np.nan,
                     "sum (SoL)": round(float(ok.sum()), 0) if ok.size else np.nan})

    # (2) the uncertainty layers
    num_name = next((n for n in available if n.endswith("_Num")), None)
    std_name = next((n for n in available if n.endswith("_Std")), None)
    num_array = group[num_name][:].astype("float32") if num_name else None
    if num_array is not None:                    # the fill value is read, never assumed
        _num_fill = read_attribute(group[num_name], "_FillValue", None)   # 255 (uint8) or 65535 (uint16)
        if _num_fill is not None:
            num_array[np.isclose(num_array, float(_num_fill))] = np.nan
    if std_name is not None:
        std_obj = group[std_name]
        f_std = read_attribute(std_obj, "scale_factor", 1.0)
        v_std = read_attribute(std_obj, "_FillValue", None)
        std_array = std_obj[:].astype("float32")
        std_array = np.where(np.isclose(std_array, float(v_std)) if v_std is not None else False,
                             np.nan, std_array * f_std)
    else:
        std_array = None

print("THE RADIANCE VARIANTS PRESENT IN THE FILE\n")
comparison = pd.DataFrame(rows).sort_values("sum (SoL)", ascending=False)
display(comparison.set_index("layer"))

# Relative gap between the reference layer and the others
reference = RESULT["layer"]
if reference in arrays and len(arrays) > 1:
    base = np.nansum(arrays[reference])
    print(f"\nGAP IN SUM OF LIGHTS RELATIVE TO '{reference}'")
    for name, values in arrays.items():
        if name == reference:
            continue
        gap = 100 * (np.nansum(values) - base) / base if base else np.nan
        print(f"   {name:<44} {gap:+7.2f} %")

if num_name:
    finite = num_array[np.isfinite(num_array)]
    print(f"\nNUMBER OF OBSERVATIONS  ({num_name})")
    print(f"   median {np.median(finite):.0f} nights   ·   "
          f"minimum {finite.min():.0f}   ·   maximum {finite.max():.0f}")
    low = 100 * np.mean(finite <= 5)
    print(f"   {low:.1f} % of pixels rest on 5 nights or fewer")
````

````python
# --- Figure: radiance, number of observations, standard deviation ------------
panels = [("Decoded radiance", np.where(np.isfinite(radiance), np.log1p(radiance), np.nan),
           CMAP_NIGHT, "log(1 + nW·cm⁻²·sr⁻¹)", True)]
if num_array is not None:
    num_display = num_array
    panels.append((f"Number of nights used\n({num_name})", num_display,
                   CMAP_GREEN, "nights", False))
if std_array is not None:
    panels.append((f"Standard deviation across those nights\n({std_name})",
                   np.where(np.isfinite(std_array), np.log1p(std_array), np.nan),
                   "magma", "log(1 + std. dev.)", False))

fig, axes = plt.subplots(1, len(panels), figsize=(4.55 * len(panels), 4.9))
fig.subplots_adjust(top=0.72, bottom=0.165, left=0.055, right=0.97, wspace=0.26)
axes = np.atleast_1d(axes)

for ax, (title_, data, cmap, label_, is_log) in zip(axes, panels):
    finite = data[np.isfinite(data)]
    vmin = float(np.nanpercentile(finite, 1)) if finite.size else 0.0
    vmax = float(np.nanpercentile(finite, 99.5)) if finite.size else 1.0
    im = ax.imshow(data, cmap=cmap, vmin=vmin, vmax=max(vmax, vmin + 1e-9),
                   extent=RESULT["extent"], origin="upper", aspect="auto",
                   interpolation="nearest")
    ax.set_facecolor("#DCE4E0")
    ax.set_title(title_, pad=7, fontsize=10.2)
    ax.set_xlabel("longitude (°)")
    bar = fig.colorbar(im, ax=ax, fraction=0.045, pad=0.02)
    bar.set_label(label_, fontsize=7.6)
    bar.ax.tick_params(labelsize=7)
axes[0].set_ylabel("latitude (°)")

figure_title(fig, "A radiance on its own does not say whether you can trust it",
             "The two right-hand panels are the ones people forget: they say over how many nights each pixel\n"
             "was computed, and how much those nights diverged. They are what justifies your limitations.",
             f"Source: {RESULT['metadata']['file']}"
             + ("   ·   DEMONSTRATION MODE" if DEMO_MODE else ""))
plt.show()
````

 <div style="color:#00704A;font-weight:700;font-size:13.5px;">&#128161;&nbsp; Which variant should you choose?</div>

### 17 · Comparing the products with one another

<div style="border-left:6px solid #00A86A;background:#E8F5EF;border-radius:0 12px 12px 0;
 padding:14px 22px;margin-top:22px;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
 <div style="color:#D49A00;font-size:11px;font-weight:700;letter-spacing:2.5px;text-transform:uppercase;">
  STEP 17 · 20 &nbsp;&nbsp;<span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span  <div style="color:#231F20;font-size:14px;margin-top:4px;">A2, A3 and A4 side by side — differences of structure, not just of period</div>
</div>

**The last structural point.** The three products do not differ only in their reference period.
They do not share the same group path, the same layer names, the same quality layer, or the same coding of
that quality. A script written for A4 that receives an A2 does not always crash: sometimes it runs, and
produces wrong numbers.

The cell below opens one file of each available product and builds the comparison.

````python
# =============================================================================
#  STEP 17 — COMPARING THE PRODUCTS WITH ONE ANOTHER
# =============================================================================
rows = []
previews = []

for product in ["VNP46A4", "VNP46A3", "VNP46A2", "VNP46A1"]:
    subset = INVENTORY[INVENTORY["product"] == product]
    if subset.empty:
        continue
    path_ = Path(subset.iloc[-1]["path"])
    with h5py.File(path_, "r") as f:
        group_path = find_data_group(f)
        group = f[group_path]
        datasets = {n: o for n, o in group.items() if isinstance(o, h5py.Dataset)}
        layer = pick_radiance_layer(group)
        ds = group[layer]
        flag_name = next((n for n in (f"{layer}_Quality", "Mandatory_Quality_Flag")
                          if n in group), "-")
        rows.append({
            "product": product,
            "family": FAMILY[product],
            "HDF5 group": group_path.replace("HDFEOS/GRIDS/", ".../").replace("/Data Fields", ""),
            "layers": len(datasets),
            "radiance layer": layer,
            "type": str(ds.dtype),
            "scale_factor": read_attribute(ds, "scale_factor", ""),
            "_FillValue": read_attribute(ds, "_FillValue", ""),
            "quality layer": flag_name,
            "shape": "x".join(str(d) for d in ds.shape),
        })
    try:
        previews.append((product, read_ntl(path_, downsample=DOWNSAMPLE)))
    except Exception as error:
        print(f"  {product}: could not be read ({error})")

COMPARISON = pd.DataFrame(rows).set_index("product")
print("STRUCTURAL COMPARISON OF THE PRODUCTS PRESENT IN YOUR DATA\n")
display(COMPARISON.T)

if len(previews) > 1:
    print("\nWHAT EACH PRODUCT GIVES OVER THE SAME AREA\n")
    summary = []
    for product, res in previews:
        v = res["radiance"][np.isfinite(res["radiance"])]
        summary.append({"product": product, "date": res["metadata"]["date"],
                        "tile": res["metadata"]["tile"],
                        "usable pixels": f"{v.size:,}".replace(",", " "),
                        "coverage_%": round(100 * v.size / res["radiance"].size, 1),
                        "median": round(float(np.median(v)), 3) if v.size else np.nan,
                        "SoL": f"{v.sum():,.0f}".replace(",", " ") if v.size else "-"})
    display(pd.DataFrame(summary).set_index("product"))
````

````python
# --- Figure: the same area seen through each product -------------------------
if previews:
    fig, axes = plt.subplots(1, len(previews), figsize=(4.5 * len(previews), 4.9))
    fig.subplots_adjust(top=0.72, bottom=0.165, left=0.055, right=0.98, wspace=0.22)
    axes = np.atleast_1d(axes)

    shared_ceiling = max(
        float(np.nanpercentile(np.log1p(r["radiance"][np.isfinite(r["radiance"])]), 99.5))
        for _, r in previews if np.isfinite(r["radiance"]).any())

    for ax, (product, res) in zip(axes, previews):
        data = res["radiance"]
        image = np.where(np.isfinite(data), np.log1p(data), np.nan)
        ax.imshow(image, cmap=CMAP_NIGHT, vmin=0, vmax=shared_ceiling,
                  extent=res["extent"], origin="upper", aspect="auto",
                  interpolation="nearest")
        ax.set_facecolor("#DCE4E0")
        coverage = 100 * float(np.isfinite(data).mean())
        ax.set_title(f"{product} - {FAMILY[product]}\n{res['metadata']['date']}   "
                     f"·   {coverage:.0f} % usable pixels",
                     pad=7, fontsize=10)
        ax.set_xlabel("longitude (°)")
    axes[0].set_ylabel("latitude (°)")

    figure_title(fig, "The same place, depending on the product chosen",
                 "Identical colour scale for all three. The daily product shows more holes (clouds, moon);\n"
                 "the annual one is the most complete - at the cost of any information about the date.",
                 "All maps in log1p, same bounds."
                 + ("   ·   DEMONSTRATION MODE - simulated values" if DEMO_MODE else ""))
    plt.show()
````

### 18 · Exporting

<div style="border-left:6px solid #00A86A;background:#E8F5EF;border-radius:0 12px 12px 0;

 <div style="color:#D49A00;font-size:11px;font-weight:700;letter-spacing:2.5px;text-transform:uppercase;">
  STEP 18 · 20 &nbsp;&nbsp;<span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span></div>
 <div style="color:#231F20;font-size:14px;margin-top:4px;">Leaving the notebook for a GIS, a spreadsheet or a repository</div>
</div>

**Why export.** The notebook is a place to explore, not a place to store. The product of your
work has to be openable by a colleague in QGIS, attachable to a report, or committable to a GitHub
repository. Three formats cover most needs:

- **GeoTIFF** — the universal GIS format. It carries the georeferencing: QGIS and ArcGIS place it in the
  right spot with no settings at all. This is the format to prefer for any sharing.
- **CSV** — one pixel per row, with its longitude, latitude and radiance. Heavy, but understandable by
  everyone and joinable to administrative data in a spreadsheet.
- **NPZ** — NumPy's compressed archive: quick to read back in Python, ideal as an intermediate cache
  between two processing steps.

If `rasterio` is not installed, the cell still produces a `.tif` file accompanied by a **world file**
(`.tfw`) and a `.prj`: most GIS packages will georeference it correctly.

````python
# =============================================================================
#  STEP 18 — EXPORTING
# =============================================================================
OUTPUT_DIR = Path.cwd() / "ntl_outputs"
OUTPUT_DIR.mkdir(exist_ok=True)

meta = RESULT["metadata"]
base = f"{meta['product']}_{meta['tile']}_{meta['date']}"
if DEMO_MODE:
    base = "DEMO_" + base
exported = []

# --- 1. GeoTIFF ---------------------------------------------------------------
tif_path = OUTPUT_DIR / f"{base}_radiance.tif"
res_lon, _, west_e, _, res_lat_neg, north_e = RESULT["transform"]

if AVAILABLE.get("rasterio"):
    import rasterio
    from rasterio.transform import Affine

    transform = Affine(res_lon, 0.0, west_e, 0.0, res_lat_neg, north_e)
    with rasterio.open(
            tif_path, "w", driver="GTiff",
            height=radiance.shape[0], width=radiance.shape[1], count=1,
            dtype="float32", crs="EPSG:4326", transform=transform,
            nodata=np.nan, compress="deflate", tiled=True) as out:
        out.write(radiance.astype("float32"), 1)
        out.update_tags(**{k: str(v) for k, v in meta.items()})
    exported.append(("Georeferenced GeoTIFF (rasterio)", tif_path))
else:
    # Fallback without rasterio: plain TIFF + world file + projection
    try:
        from PIL import Image
        Image.fromarray(radiance.astype("float32")).save(tif_path)
        (OUTPUT_DIR / f"{base}_radiance.tfw").write_text(
            f"{res_lon:.12f}\n0.0\n0.0\n{res_lat_neg:.12f}\n"
            f"{west_e + res_lon / 2:.12f}\n{north_e + res_lat_neg / 2:.12f}\n")
        (OUTPUT_DIR / f"{base}_radiance.prj").write_text(
            'GEOGCS["WGS 84",DATUM["WGS_1984",SPHEROID["WGS 84",6378137,298.257223563]],'
            'PRIMEM["Greenwich",0],UNIT["degree",0.0174532925199433]]')
        exported.append(("TIFF + .tfw world file + .prj (no rasterio)", tif_path))
    except Exception as error:
        print(f"  TIFF export failed: {error}")

# --- 2. CSV: one pixel per row ------------------------------------------------
stride = max(radiance.shape[0] // 220, 1)          # thinned out to stay readable
thin_radiance = radiance[::stride, ::stride]
LON, LAT = np.meshgrid(RESULT["longitudes"][::stride], RESULT["latitudes"][::stride])
pixel_table = pd.DataFrame({
    "longitude": np.round(LON.ravel(), 6),
    "latitude": np.round(LAT.ravel(), 6),
    "radiance": np.round(thin_radiance.ravel(), 4),
}).dropna(subset=["radiance"])

csv_path = OUTPUT_DIR / f"{base}_pixels.csv"
pixel_table.to_csv(csv_path, index=False, encoding="utf-8")
exported.append((f"CSV - {len(pixel_table):,} pixels (1 in {stride})".replace(",", " "), csv_path))

# --- 3. NumPy archive ---------------------------------------------------------
npz_path = OUTPUT_DIR / f"{base}_radiance.npz"
np.savez_compressed(npz_path, radiance=radiance.astype("float32"),
                    longitudes=RESULT["longitudes"], latitudes=RESULT["latitudes"],
                    extent=np.array(RESULT["extent"]))
exported.append(("Compressed NumPy archive", npz_path))

# --- 4. Human-readable metadata -----------------------------------------------
json_path = OUTPUT_DIR / f"{base}_metadata.json"
documentation = {
    "source": "NASA Black Marble - VNP46 suite, collection 2 (version 002)",
    "warning": ("SIMULATED VALUES - demonstration set, do not use to produce "
                "any figure" if DEMO_MODE else "Real NASA data"),
    "processing": {k: str(v) for k, v in meta.items()},
    "extent_WGS84": {"west": RESULT["extent"][0], "east": RESULT["extent"][1],
                     "south": RESULT["extent"][2], "north": RESULT["extent"][3]},
    "coordinate_system": "EPSG:4326 (WGS 84)",
    "affine_transform": list(RESULT["transform"]),
    "dimensions": {"rows": int(radiance.shape[0]), "columns": int(radiance.shape[1])},
    "downsample": DOWNSAMPLE,
    "processing_log": RESULT["log"].to_dict(orient="records"),
    "citation": ("Roman, M. O. et al. (2018). NASA's Black Marble nighttime lights product "
                 "suite. Remote Sensing of Environment, 210, 113-143."),
}
json_path.write_text(json.dumps(documentation, indent=2, ensure_ascii=False,
                                default=str), encoding="utf-8")
exported.append(("JSON metadata", json_path))

# --- Report -------------------------------------------------------------------
print("FILES WRITTEN")
print("=" * 78)
for label_, path_ in exported:
    print(f"  {label_}")
    print(f"     {path_}   ({path_.stat().st_size / 1e3:.0f} kB)")
print("=" * 78)
print(f"\nOutput folder: {OUTPUT_DIR}")
if not AVAILABLE.get("rasterio"):
    print("\n  (rasterio missing: the GeoTIFF was produced with a .tfw world file.")
    print("   For a fully compliant GeoTIFF:  pip install rasterio)")

print("\nPREVIEW OF THE CSV")
display(pixel_table.head(8))
````

**The moment of truth.** Since step 09, every correction has been recorded in `STORYLINE` with the value of the indicator at that instant. Here they all are together.

Read this table the way an **auditor** would: each of these lines is a figure an analyst could have published in good faith, believing the work was done. Only one is defensible. The others do not differ by rounding — they differ by factors.

````python
# =============================================================================
#  STEP 19 — THE STORYLINE: WHAT EACH CORRECTION COST
# =============================================================================
if not STORYLINE:
    print("No checkpoint recorded: run steps 09 to 14 first.")
else:
    print(f"THE SAME QUESTION, ASKED {len(STORYLINE)} TIMES — {COUNTRY_NAME}\n")
    display(storyline_table())

    values_ = [e["sum of lights"] for e in STORYLINE]
    labels_ = [f"{e['step']:02d} · {e['treatment']}" for e in STORYLINE]
    perimeters = [e["perimeter"] for e in STORYLINE]

    palette_p = [COLOURS["brick"], COLOURS["green"], COLOURS["teal"], COLOURS["gold"],
                 COLOURS["terra"]]
    colour_by_perimeter = {}
    for p in perimeters:
        colour_by_perimeter.setdefault(
            p, palette_p[len(colour_by_perimeter) % len(palette_p)])

    hauteur = 0.46 * len(values_) + 2.5
    fig, ax = plt.subplots(figsize=(12.0, hauteur))
    fig.subplots_adjust(top=1 - 1.18 / hauteur, bottom=1.02 / hauteur,
                        left=0.295, right=0.955)

    y = np.arange(len(values_))[::-1]
    ax.barh(y, values_, height=0.62, edgecolor="none",
            color=[colour_by_perimeter[p] for p in perimeters])
    reference = values_[-1]
    for yi, v in zip(y, values_):
        facteur = v / reference if reference else np.nan
        suffixe = f"   (× {facteur:,.1f})".replace(",", " ") if facteur >= 1.05 else ""
        ax.text(v * 1.06, yi, f"{v:,.0f}".replace(",", " ") + suffixe, va="center",
                ha="left" if v >= 0 else "right", fontsize=8.8, color=COLOURS["ink"], fontweight="bold")

    ax.set_yticks(y)
    ax.set_yticklabels(labels_, fontsize=9)
    if min(values_) > 0:
        ax.set_xscale("log")
        ax.set_xlim(max(min(values_) * 0.45, 1), max(values_) * 6)
    else:   # a negative sum (fill value -999.9 counted as a measurement): symmetric log scale
        ax.set_xscale("symlog", linthresh=max(1.0, abs(reference)))
        ax.set_xlim(min(values_) * 6, max(values_) * 6)
    ax.set_xlabel("sum of lights, logarithmic scale (nW·cm⁻²·sr⁻¹)")
    ax.spines["left"].set_visible(False)
    ax.tick_params(left=False)
    ax.legend(handles=[plt.Rectangle((0, 0), 1, 1, fc=c, ec="none", label=p)
                       for p, c in colour_by_perimeter.items()],
              loc="lower right" if min(values_) > 0 else "lower left", fontsize=8.2, frameon=True, framealpha=0.95,
              facecolor="white", edgecolor=COLOURS["sage"], title="perimeter",
              title_fontsize=8.5)

    figure_title(fig, f"The same territory, {len(set(np.round(values_, 3)))} different figures",
                 "Every bar is a defensible answer to the same question — and only one of them is right.\n"
                 "The gap between the first and the last is not a margin of error: it is an order of magnitude.",
                 "Checkpoints recorded at steps 09 to 14 of the notebook."
                 + ("   ·   DEMONSTRATION MODE — simulated values" if DEMO_MODE else ""))
    plt.show()

    naive, final = values_[0], values_[-1]
    print("\n" + "=" * 78)
    print(f"  First answer, with no precaution at all : {naive:>18,.0f}".replace(",", " "))
    print(f"  Final answer, every correction applied  : {final:>18,.0f}".replace(",", " "))
    if final and naive > 0:
        print(f"  Ratio between the two                   : × {naive / final:,.1f}"
              .replace(",", " "))
    print("=" * 78)
    print("  None of these steps is optional, and none of them produces an error")
    print("  message. That is exactly what makes them dangerous.")
````

 padding:13px 17px;margin:10px 0;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
 <div style="color:#D49A00;font-weight:700;font-size:13.5px;">&#9888;&nbsp; What this figure should change in your practice</div>
 <div style="color:#231F20;font-size:13.5px;margin-top:5px;line-height:1.6;">An indicator derived from satellite data is not reproducible without its processing chain. Two teams starting from the <b>same file</b> can publish figures an order of magnitude apart, without either having made a visible mistake. The only protection is the processing log and the metadata of step 18: layer used, quality threshold, decision on water, reference border. <b>Publish the chain, not just the number.</b></div>

 <div style="color:#B83B2E;font-weight:700;font-size:13.5px;">&#9940;&nbsp; The six artefacts to know about before concluding anything</div>
 <div style="color:#231F20;font-size:13.5px;margin-top:5px;line-height:1.6;"><b>Urban blooming</b> — light spills beyond the built-up area and inflates «&nbsp;lit&nbsp;» surfaces. &nbsp;<b>Saturation</b> — the most intense urban cores plateau, and growth there becomes invisible. &nbsp;<b>Gas flares</b> — an oil site can dominate the radiance of a whole region with no relation to its economic activity. &nbsp;<b>Seasonality</b> — clouds and bush fires create cycles that are not economic. &nbsp;<b>Sensor break</b> — the arrival of NOAA-20 alongside Suomi-NPP can introduce a step change in a long series. &nbsp;<b>Noise in dimly lit rural areas</b> — near the detection threshold the signal-to-noise ratio collapses: which is precisely where the populations the indicator claims to describe actually live.</div>
