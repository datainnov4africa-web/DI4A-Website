## À quoi sert ce notebook

Les lumières nocturnes sont l'une des rares sources de données **gratuites, mondiales, quotidiennes et
politiquement neutres** dont dispose un institut national de statistique. Mais entre le fichier que la NASA
met en ligne et un indicateur publiable, il y a une étape que presque personne n'explique correctement :
**comprendre ce qu'il y a dans le fichier**.

Un fichier Black Marble n'est pas une image. C'est un **classeur** contenant une vingtaine de tableaux de
nombres entiers, chacun avec sa propre signification, sa propre échelle et sa propre valeur « vide ».
Lire ce classeur de travers — oublier un facteur d'échelle, confondre une valeur de remplissage avec un
zéro, ignorer un drapeau qualité — produit des chiffres qui *semblent* corrects et qui sont **faux**.

Ce notebook démonte le fichier pièce par pièce. À la fin, vous saurez répondre, preuve à l'appui, à
huit questions :

## Le fil rouge : une seule question, posée vingt fois

Un notebook de formation qui enchaîne dix-huit démonstrations indépendantes s’oublie en une semaine. Celui-ci suit **une question unique**, du début à la fin :

<div style="background:#E8F5EF;border:1px solid #00704A44;border-left:4px solid #00704A;border-radius:8px;
 padding:13px 17px;margin:10px 0;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
 <div style="color:#00704A;font-weight:700;font-size:13.5px;">&#128273;&nbsp; La question à laquelle ce notebook répond</div>
 <div style="color:#231F20;font-size:13.5px;margin-top:5px;line-height:1.6;"><b style="font-size:15px;">Quelle est la somme des lumières du territoire choisi — et comment être sûr que ce chiffre est juste ?</b><br><br>Nous y répondrons une première fois à l’étape 09, très vite et très mal. Puis chaque étape corrigera une erreur, et le chiffre bougera : parfois de quelques pourcents, parfois d’un facteur dix. L’étape 19 met toutes ces réponses côte à côte. <b>Aucune des erreurs corrigées en chemin ne produit de message d’erreur</b> — c’est exactement ce qui les rend coûteuses.</div>
</div>
Le pays suivi est défini par le paramètre `PAYS` à l’étape 02 (**Côte d’Ivoire** par défaut). Changez-le, et tout le notebook suit.

Le notebook a été conçu pour tourner **à l’identique** dans les quatre environnements ci-dessus. Il
n’utilise aucune bibliothèque exotique, aucun chemin absolu et aucun service en ligne obligatoire.

<div style="background:#E8F5EF;border:1px solid #00704A44;border-left:4px solid #00704A;border-radius:8px;
 padding:13px 17px;margin:10px 0;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
 <div style="color:#00704A;font-weight:700;font-size:13.5px;">&#128273;&nbsp; Le principe qui rend ce notebook portable</div>
 <div style="color:#231F20;font-size:13.5px;margin-top:5px;line-height:1.6;">Au démarrage, le notebook <b>cherche vos vraies données</b> (dossier local, <code>/kaggle/input</code>, Google&nbsp;Drive monté…). S’il n’en trouve aucune, il <b>fabrique lui-même de petits fichiers <code>.h5</code> de démonstration</b> dont la structure interne est rigoureusement conforme à celle des vrais produits NASA : mêmes chemins de groupes, mêmes noms de couches, mêmes types entiers, mêmes attributs, mêmes valeurs de remplissage et mêmes drapeaux qualité. <b>Tout ce que le notebook enseigne sur la structure reste donc exact</b> — seules les valeurs de radiance sont simulées. Un bandeau vous rappellera en permanence dans quel mode vous êtes.</div>
</div>

<div style="background:#FDF4E0;border:1px solid #D49A0044;border-left:4px solid #D49A00;border-radius:8px;

 <div style="color:#D49A00;font-weight:700;font-size:13.5px;">&#9888;&nbsp; Ce que le mode démonstration ne remplace pas</div>
 <div style="color:#231F20;font-size:13.5px;margin-top:5px;line-height:1.6;">Les valeurs simulées ne doivent <b>jamais</b> servir à produire un chiffre, une carte publiée ou un indicateur. Elles servent uniquement à faire fonctionner la mécanique pédagogique. Dès que vous disposez de vos fichiers NASA, renseignez <code>DATA_DIR</code> à l’étape 02 et relancez : l’intégralité du notebook s’exécute alors sur vos données réelles, sans une seule autre modification.</div>

 <div style="color:#231F20;font-size:14px;margin-top:4px;">Savoir où l’on tourne, et installer le strict nécessaire</div>

**Ce que fait cette étape.** Le notebook identifie d’abord l’environnement d’exécution
(Colab, Kaggle, Jupyter local, VS&nbsp;Code), puis vérifie les bibliothèques.

La distinction importante est celle entre **noyau requis** et **extensions optionnelles** :

- Le **noyau** — `numpy`, `pandas`, `matplotlib`, `h5py` — suffit à faire tourner **la totalité** de ce
  notebook. Ces quatre bibliothèques sont déjà présentes sur Colab et Kaggle.
- Les **extensions** — `geopandas`, `folium`, `rasterio`, `requests` — ajoutent du confort (frontières
  administratives, carte interactive, export GeoTIFF, téléchargement). Si l’une manque ou refuse de
  s’installer, le notebook le signale et **continue** : aucune étape ne se bloque.

````python
# =============================================================================
#  ÉTAPE 01 — ENVIRONNEMENT ET DÉPENDANCES
# =============================================================================
import importlib.util
import os
import platform
import subprocess
import sys
from pathlib import Path

# --- Réglage : mettre à False pour interdire toute installation automatique ---
INSTALLER_AUTOMATIQUEMENT = True


def detecter_environnement() -> str:
    """Identifie la plateforme d'exécution du notebook."""
    if "google.colab" in sys.modules:
        return "Google Colab"
    if os.environ.get("KAGGLE_KERNEL_RUN_TYPE") or Path("/kaggle/input").exists():
        return "Kaggle"
    if os.environ.get("VSCODE_PID") or os.environ.get("VSCODE_CWD"):
        return "VS Code"
    if "ipykernel" in sys.modules:
        return "Jupyter local"
    return "Python (hors notebook)"


ENVIRONNEMENT = detecter_environnement()

# Nom d'import  ->  nom du paquet pip
NOYAU = {"numpy": "numpy", "pandas": "pandas", "matplotlib": "matplotlib", "h5py": "h5py"}
EXTENSIONS = {"requests": "requests", "shapely": "shapely",
              "geopandas": "geopandas", "folium": "folium", "rasterio": "rasterio"}


def installer(paquets) -> bool:
    """Installe une liste de paquets via pip. Renvoie True si tout s'est bien passé."""
    try:
        subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", *paquets],
                              stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
        return True
    except Exception:
        return False


def verifier(table, etiquette, obligatoire):
    """Vérifie la présence des modules, tente l'installation des manquants."""
    manquants = [pip for imp, pip in table.items() if importlib.util.find_spec(imp) is None]
    if manquants and INSTALLER_AUTOMATIQUEMENT:
        print(f"  installation de {', '.join(manquants)} ...")
        installer(manquants)
        manquants = [pip for imp, pip in table.items() if importlib.util.find_spec(imp) is None]

    disponibles = {}
    for imp in table:
        present = importlib.util.find_spec(imp) is not None
        disponibles[imp] = present
        if present:
            try:
                version = getattr(__import__(imp), "__version__", "?")
            except Exception:
                version = "?"
            print(f"  [ok]      {imp:<12} {version}")
        else:
            print(f"  [{'MANQUE' if obligatoire else 'absent'}]  {imp:<12} ({etiquette})")
    return disponibles


print("=" * 74)
print(f"  ENVIRONNEMENT   : {ENVIRONNEMENT}")
print(f"  PYTHON          : {platform.python_version()}  ({platform.system()} {platform.machine()})")
print(f"  DOSSIER COURANT : {Path.cwd()}")
print("=" * 74)

print("\nNoyau requis")
etat_noyau = verifier(NOYAU, "indispensable", obligatoire=True)

print("\nExtensions optionnelles")
DISPO = verifier(EXTENSIONS, "confort", obligatoire=False)

if not all(etat_noyau.values()):
    raise SystemExit(
        "\nIl manque une bibliothèque du noyau. Installez-la à la main puis relancez :\n"
        "    pip install numpy pandas matplotlib h5py")

print("\n" + "=" * 74)
print("  Noyau complet — le notebook peut s'exécuter intégralement.")
if not all(DISPO.values()):
    absentes = [k for k, v in DISPO.items() if not v]
    print(f"  Extensions absentes : {', '.join(absentes)} — les étapes concernées seront")
    print("  simplement signalées puis contournées, sans interrompre le notebook.")
print("=" * 74)
````

````python
# =============================================================================
#  ÉTAPE 02 — PARAMÈTRES   ◄◄◄  LA SEULE CELLULE À MODIFIER
# =============================================================================

# -----------------------------------------------------------------------------
#  1. QUEL PAYS SUIVEZ-VOUS ?
# -----------------------------------------------------------------------------
#     Un code ISO 3166-1 alpha-3. Le notebook entier s'y adapte : frontière,
#     villes, tuiles à télécharger, découpage, cartes et indicateurs.
#     Exemples : "CIV" Côte d'Ivoire · "SEN" Sénégal · "SDN" Soudan
#                "NGA" Nigéria · "MAR" Maroc · "KEN" Kenya · "ZAF" Afrique du Sud
#                "COD" RD Congo · "EGY" Égypte · "TZA" Tanzanie · "ETH" Éthiopie
#     88 pays sont disponibles ; un code inconnu arrête le notebook avec la liste.
# -----------------------------------------------------------------------------
PAYS = "SEN"

# -----------------------------------------------------------------------------
#  2. OÙ SONT VOS DONNÉES ?
# -----------------------------------------------------------------------------
#     None  -> détection automatique (dossier courant, /kaggle/input, Drive monté...)
#     Sinon -> chemin explicite vers le dossier contenant vos fichiers .h5
#
#     Windows  : DATA_DIR = r"C:\Users\moi\blackmarble_data"
#     macOS    : DATA_DIR = "/Users/moi/blackmarble_data"
#     Linux    : DATA_DIR = "/home/moi/blackmarble_data"
#     Colab    : DATA_DIR = "/content/drive/MyDrive/blackmarble_data"
#     Kaggle   : DATA_DIR = "/kaggle/input/mon-dataset-ntl"
# -----------------------------------------------------------------------------
DATA_DIR = None

# -----------------------------------------------------------------------------
#  3. VOTRE JETON EARTHDATA  (pour télécharger les vraies données)
# -----------------------------------------------------------------------------
#     Les produits Black Marble sont gratuits, mais l'accès demande un compte
#     NASA Earthdata et un jeton personnel :
#         1. créez un compte sur  https://urs.earthdata.nasa.gov
#         2. onglet « Generate Token », puis copiez la longue chaîne obtenue
#         3. collez-la ci-dessous, OU mieux : voir la note qui suit l'étape 02.
#
#     Laissé vide, le notebook fonctionne quand même : il bascule sur son jeu
#     de démonstration. Un jeton ne se partage pas et ne se commite jamais.
# -----------------------------------------------------------------------------
JETON_EARTHDATA = ""            # ou laissez vide et utilisez EARTHDATA_TOKEN / ~/.earthdata_token
DEMANDER_LE_JETON = True        # demander le jeton au clavier si aucun n'est trouvé ailleurs
TELECHARGER = True              # False = ne jamais sortir sur le réseau
ANNEES_VOULUES = (2021, 2022, 2023)

# -----------------------------------------------------------------------------
#  4. QUEL PRODUIT EXPLORER EN PRIORITÉ ?
#     "VNP46A4" annuel  ·  "VNP46A3" mensuel  ·  "VNP46A2" journalier
# -----------------------------------------------------------------------------
PRODUIT_PRINCIPAL = "VNP46A4"

# -----------------------------------------------------------------------------
#  5. FILTRAGE QUALITÉ
#     QA_MAX : on garde les pixels dont le drapeau qualité est <= à cette valeur
#              0 = uniquement la meilleure qualité   (le plus strict)
#              1 = bonne qualité + lumières éphémères (recommandé)
#              2 = on accepte aussi les pixels comblés / de qualité médiocre
#     MASQUER_EAU : mettre à NaN les pixels d'océan et d'eau profonde
# -----------------------------------------------------------------------------
QA_MAX = 1
MASQUER_EAU = True

# -----------------------------------------------------------------------------
#  6. AFFICHAGE
#     SOUS_ECHANTILLONNAGE : 1 = pleine résolution ; 4 = moyenne des blocs 4x4
#                            (n'affecte que l'affichage, jamais les calculs)
# -----------------------------------------------------------------------------
SOUS_ECHANTILLONNAGE = 1
FIG_DPI = 115

# -----------------------------------------------------------------------------
#  7. MODE DÉMONSTRATION
#     AUTORISER_DEMO : si aucune donnée réelle n'est trouvée ni téléchargée,
#                      fabriquer des fichiers .h5 structurellement conformes
#                      pour pouvoir exécuter le notebook de bout en bout.
#     DEMO_PX        : côté, en pixels, de la tuile de démonstration PRINCIPALE
#                      — celle qui contient votre plus grande ville. Les vraies
#                      tuiles NASA font 2400 x 2400 ; 1200 suffit pour que le
#                      zoom sur une agglomération reste lisible, tout en gardant
#                      le notebook rapide. Descendez à 600 sur machine modeste.
# -----------------------------------------------------------------------------
AUTORISER_DEMO = True
DEMO_PX = 1200

# =============================================================================
print("Paramètres enregistrés.")
print(f"  PAYS                 = {PAYS}")
print(f"  DATA_DIR             = {DATA_DIR if DATA_DIR else 'détection automatique'}")
print(f"  PRODUIT_PRINCIPAL    = {PRODUIT_PRINCIPAL}")
print(f"  QA_MAX               = {QA_MAX}   (garder les pixels de drapeau <= {QA_MAX})")
print(f"  MASQUER_EAU          = {MASQUER_EAU}")
print(f"  TELECHARGER          = {TELECHARGER}"
      f"   (jeton fourni dans la cellule : {'oui' if JETON_EARTHDATA else 'non'})")
````

 <div style="color:#00704A;font-weight:700;font-size:13.5px;">&#128161;&nbsp; Pourquoi <code>QA_MAX = 1</code> et non 0 ?</div>
 <div style="color:#231F20;font-size:13.5px;margin-top:5px;line-height:1.6;">Le drapeau <code>1</code> correspond aux <i>lumières éphémères de haute qualité</i> : feux de brousse, torchères, activités saisonnières. **Ce que fait cette étape.** Quatre cellules, dans cet ordre :

1. **La boîte à outils** — les constantes de la grille Black Marble, les fonctions qui savent retrouver
   un groupe dans un fichier HDF5 et lire proprement un attribut, et la charte graphique des figures.
2. **Le filet de sécurité** — un générateur de fichiers `.h5` de démonstration, activé seulement si aucune
   donnée réelle n’est trouvée.
3. **L’inventaire** — le notebook parcourt le dossier de données et dresse le catalogue de tout ce qu’il y
   a trouvé. Ce catalogue pilote tout le reste du notebook.
4. **Un premier regard** — une carte nocturne de votre zone, sans aucun filtrage, pour voir de quoi
   l’on parle avant de le démonter.

<div style="background:#E8F5EF;border:1px solid #00704A44;border-left:4px solid #00704A;border-radius:8px;
 padding:13px 17px;margin:10px 0;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
 <div style="color:#00704A;font-weight:700;font-size:13.5px;">&#128273;&nbsp; La convention de nommage en français</div>
 <div style="color:#231F20;font-size:13.5px;margin-top:5px;line-height:1.6;">Les fonctions de ce notebook portent des noms français explicites (<code>lire_attribut</code>, <code>trouver_groupe_donnees</code>, <code>decoder_nom_fichier</code>…) plutôt que des abréviations. C’est volontairement plus verbeux : le code doit pouvoir être lu à voix haute pendant une séance de formation, et repris par un collègue qui n’était pas dans la salle.</div>

````python
# =============================================================================
#  ÉTAPE 03 a — BOÎTE À OUTILS
# =============================================================================
import json
import re
import warnings

import h5py
import matplotlib as mpl
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from matplotlib.colors import LinearSegmentedColormap, ListedColormap, LogNorm

warnings.filterwarnings("ignore")

# -----------------------------------------------------------------------------
#  1. CHARTE GRAPHIQUE  (palette du guide de style de la Banque)
# -----------------------------------------------------------------------------
COULEURS = dict(
    vert="#00A86A", vert_fonce="#00704A", foret="#00553A", or_="#F5C242",
    ocre="#D49A00", sarcelle="#0E7C86", terre="#C4621D", brique="#B83B2E",
    encre="#231F20", ardoise="#5E6964", brume="#F4F7F5", menthe="#E8F5EF", sauge="#D5DED9",
)

# Palette « vue de nuit depuis l'orbite » : noir -> vert -> or -> crème
CMAP_NUIT = LinearSegmentedColormap.from_list(
    "bad_nuit", ["#04120C", "#00553A", "#00704A", "#00A86A", "#F5C242", "#FFF6DF"])
# Palette séquentielle claire, pour les cartes sur fond blanc
CMAP_VERT = LinearSegmentedColormap.from_list(
    "bad_vert", ["#E8F5EF", "#9ED9C0", "#57BD92", "#10A06A", "#00704A", "#00553A"])

_polices = {f.name for f in mpl.font_manager.fontManager.ttflist}
POLICE = "Calibri" if "Calibri" in _polices else "DejaVu Sans"

mpl.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "figure.dpi": FIG_DPI, "font.family": POLICE, "font.size": 10.5,
    "axes.titlesize": 12.5, "axes.titleweight": "bold", "axes.titlecolor": COULEURS["foret"],
    "axes.labelcolor": COULEURS["ardoise"], "axes.edgecolor": COULEURS["sauge"],
    "xtick.color": COULEURS["ardoise"], "ytick.color": COULEURS["ardoise"],
    "axes.grid": False, "axes.spines.top": False, "axes.spines.right": False,
    "legend.frameon": False, "figure.autolayout": False,
})


def titre_figure(fig, titre, sous_titre="", source=""):
    """Applique l'habillage institutionnel standard à une figure."""
    fig.text(0.012, 0.985, titre, ha="left", va="top", fontsize=13.5,
             fontweight="bold", color=COULEURS["foret"])
    if sous_titre:
        fig.text(0.012, 0.938, sous_titre, ha="left", va="top", fontsize=10,
                 color=COULEURS["ardoise"])
    if source:
        fig.text(0.012, 0.004, source, ha="left", va="bottom", fontsize=7.2,
                 style="italic", color=COULEURS["ardoise"])


# -----------------------------------------------------------------------------
#  2. LA GRILLE BLACK MARBLE
# -----------------------------------------------------------------------------
#  Le globe est découpé en tuiles de 10° x 10°, indexées h00..h35 (ouest -> est)
#  et v00..v17 (nord -> sud). Chaque tuile contient 2400 x 2400 pixels, soit une
#  résolution de 15 secondes d'arc (~ 463 m à l'équateur).
# -----------------------------------------------------------------------------
TUILE_DEGRES = 10.0
TUILE_PIXELS = 2400
RESOLUTION_ARCSEC = 15.0

# Le nom du groupe HDF5 dépend du produit
GROUPE_PAR_PRODUIT = {
    "VNP46A1": "HDFEOS/GRIDS/VNP_Grid_DNB/Data Fields",
    "VNP46A2": "HDFEOS/GRIDS/VNP_Grid_DNB/Data Fields",
    "VNP46A3": "HDFEOS/GRIDS/VIIRS_Grid_DNB_2d/Data Fields",
    "VNP46A4": "HDFEOS/GRIDS/VIIRS_Grid_DNB_2d/Data Fields",
}
FAMILLE = {"VNP46A1": "journalier (brut)", "VNP46A2": "journalier",
           "VNP46A3": "mensuel", "VNP46A4": "annuel"}


def numeros_tuile(tuile: str):
    """'h17v08' -> (17, 8)"""
    m = re.search(r"h(\d{2})v(\d{2})", tuile)
    if not m:
        raise ValueError(f"Identifiant de tuile illisible : {tuile!r}")
    return int(m.group(1)), int(m.group(2))


def emprise_tuile(tuile: str):
    """'h17v08' -> (ouest, est, sud, nord) en degrés décimaux."""
    h, v = numeros_tuile(tuile)
    ouest = -180.0 + h * TUILE_DEGRES
    nord = 90.0 - v * TUILE_DEGRES
    return (ouest, ouest + TUILE_DEGRES, nord - TUILE_DEGRES, nord)


def tuiles_couvrant(ouest, est, sud, nord):
    """Liste les tuiles nécessaires pour couvrir une emprise géographique."""
    h0, h1 = int((ouest + 180) // 10), int((est + 180) // 10)
    v0, v1 = int((90 - nord) // 10), int((90 - sud) // 10)
    return [f"h{h:02d}v{v:02d}" for v in range(v0, v1 + 1) for h in range(h0, h1 + 1)]


# -----------------------------------------------------------------------------
#  3. LIRE UN FICHIER HDF5 SANS SE TROMPER
# -----------------------------------------------------------------------------
def trouver_groupe_donnees(fichier: h5py.File) -> str:
    """Retrouve le groupe 'Data Fields', quel que soit le produit.

    On ne code jamais le chemin en dur : on parcourt l'arborescence et on
    retient le premier groupe dont le nom se termine par 'Data Fields'.
    """
    trouves = []
    fichier.visititems(
        lambda nom, obj: trouves.append(nom)
        if isinstance(obj, h5py.Group) and nom.endswith("Data Fields") else None)
    if not trouves:
        raise KeyError("Aucun groupe 'Data Fields' : ce fichier n'est pas un produit VNP46.")
    return trouves[0]


def lire_attribut(objet, nom, defaut=None):
    """Lit un attribut HDF5 et le convertit en type Python simple.

    Les attributs HDF5 arrivent sous des formes très variées : octets, tableau
    numpy à un élément, tableau à plusieurs éléments... Cette fonction ramène
    tout à une chaîne, un nombre, ou une liste.
    """
    if nom not in objet.attrs:
        return defaut
    valeur = objet.attrs[nom]
    if isinstance(valeur, (bytes, bytearray, np.bytes_)):
        return valeur.decode("utf-8", "replace").strip()
    if isinstance(valeur, str):
        return valeur.strip()
    plat = np.ravel(valeur)
    if plat.size == 0:
        return defaut
    if plat.size == 1:
        element = plat[0]
        if isinstance(element, (bytes, bytearray, np.bytes_)):
            return element.decode("utf-8", "replace").strip()
        return element.item()
    return plat.tolist()


def tous_les_attributs(objet) -> dict:
    """Renvoie tous les attributs d'un objet HDF5, proprement convertis."""
    return {nom: lire_attribut(objet, nom) for nom in objet.attrs}


# Ordre de préférence pour la couche de radiance principale
COUCHES_RADIANCE = [
    "Gap_Filled_DNB_BRDF-Corrected_NTL",   # A2 : version comblée
    "DNB_BRDF-Corrected_NTL",              # A2 : version brute corrigée BRDF
    "AllAngle_Composite_Snow_Free",        # A3 / A4 : la référence usuelle
    "NearNadir_Composite_Snow_Free",       # A3 / A4 : vue proche du nadir
    "OffNadir_Composite_Snow_Free",
    "AllAngle_Composite_Snow_Covered",
]


def choisir_couche_radiance(groupe: h5py.Group) -> str:
    """Choisit la couche de radiance principale présente dans le fichier."""
    present = set(groupe.keys())
    for nom in COUCHES_RADIANCE:
        if nom in present:
            return nom
    # Repli : n'importe quelle couche 2D dont l'unité est une radiance
    for nom, obj in groupe.items():
        if isinstance(obj, h5py.Dataset) and obj.ndim == 2:
            if "nW" in str(lire_attribut(obj, "units", "")):
                return nom
    raise KeyError("Aucune couche de radiance identifiable dans ce fichier.")


# -----------------------------------------------------------------------------
#  4. DÉCODER UN NOM DE FICHIER
# -----------------------------------------------------------------------------
MOTIF_NOM = re.compile(
    r"^(?P<produit>VNP46A[1-4])"      # produit
    r"\.A(?P<annee>\d{4})(?P<jour>\d{3})"   # A + année + jour julien
    r"\.(?P<tuile>h\d{2}v\d{2})"      # tuile
    r"\.(?P<version>\d{3})"           # version de la collection
    r"\.(?P<horodatage>[^.]+)"        # horodatage de production
    r"\.h5$", re.IGNORECASE)


def decoder_nom_fichier(nom: str):
    """Décompose un nom de fichier VNP46 en ses éléments constitutifs."""
    m = MOTIF_NOM.match(str(nom))
    if not m:
        return None
    d = m.groupdict()
    d["annee"] = int(d["annee"])
    d["jour"] = int(d["jour"])
    d["date"] = pd.Timestamp(d["annee"], 1, 1) + pd.Timedelta(days=d["jour"] - 1)
    d["famille"] = FAMILLE.get(d["produit"].upper(), "?")
    d["emprise"] = emprise_tuile(d["tuile"])
    return d


print(f"Boîte à outils chargée  ·  police des figures : {POLICE}")
print(f"Grille Black Marble     : tuiles de {TUILE_DEGRES:.0f}° × {TUILE_DEGRES:.0f}°, "
      f"{TUILE_PIXELS} × {TUILE_PIXELS} pixels, {RESOLUTION_ARCSEC:.0f} secondes d'arc")
````

<div style="background:#E8F5EF;border:1px solid #00704A44;border-left:4px solid #00704A;border-radius:8px;
 padding:13px 17px;margin:10px 0;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
 <div style="color:#00704A;font-weight:700;font-size:13.5px;">&#128161;&nbsp; Les deux fonctions à retenir de cette cellule</div>
 <div style="color:#231F20;font-size:13.5px;margin-top:5px;line-height:1.6;"><code>trouver_groupe_donnees(f)</code> et <code>lire_attribut(objet, nom)</code>. Ce sont elles qui rendent votre code robuste : le chemin du groupe change entre le journalier et les composites, et les attributs HDF5 arrivent tantôt en octets, tantôt en tableau à un élément. Coder ces deux choses en dur est la cause la plus fréquente de scripts qui « marchaient l’an dernier » et cassent sur un autre produit.</div>
</div>

#### Les données de référence, embarquées dans le notebook

La cellule suivante ne contient aucune logique : c'est un **dictionnaire de frontières et de
villes**, écrit en clair. Une ligne par pays, dans un format compact lisible à l'œil nu.

Pourquoi embarquer des données plutôt que les télécharger ? Parce qu'un notebook de formation doit
tourner là où le Wi-Fi ne tourne pas. L'encodage retenu — un point par `lon,lat`, les anneaux

#### Ce que le pays choisi implique

La cellule suivante décode ces données et en tire **tout ce dont le notebook a besoin pour être
générique** :

| Fonction | Ce qu'elle donne |
|---|---|
| `masque_pays(LON, LAT)` | pour chaque pixel, est-il à l'intérieur du pays ? |
| `masque_terre(LON, LAT)` | terre ou mer, tous pays confondus — sert au générateur de démonstration |
| `tuiles_du_pays()` | les tuiles Black Marble à télécharger pour couvrir le pays |
| `villes_du_pays()` | les villes repères, de la plus peuplée à la moins peuplée |
| `tracer_contour_pays(ax)` | la frontière, posée sur n'importe quelle carte |
| `noter_fil_rouge(...)` | le compteur du fil rouge, qui suivra la somme des lumières |

Le test point-dans-polygone est fait par `matplotlib.path.Path`, que vous avez déjà installé.
**Aucune bibliothèque SIG n'est nécessaire** — ni `geopandas`, ni `rasterio`, ni `GDAL` : c'est
exactement ce qui permet à ce notebook de démarrer en trente secondes sur Colab comme sur Kaggle.

````python
# =============================================================================
#  ÉTAPE 03 a-bis (2/2) — GÉOGRAPHIE DU PAYS CHOISI, ET FIL ROUGE
# =============================================================================
from matplotlib.path import Path as _Chemin


# --- 1. Décodage des données embarquées --------------------------------------
def _decoder_anneaux(texte):
    anneaux = []
    for morceau in texte.split("|"):
        points = [p.split(",") for p in morceau.split(" ") if p]
        anneaux.append(np.array([(float(x), float(y)) for x, y in points]))
    return anneaux


def _decoder_villes(texte):
    villes = []
    for morceau in texte.split(";"):
        if not morceau:
            continue
        nom, reste = morceau.split("@")
        lon, lat, pop = reste.split(",")
        villes.append((nom, float(lon), float(lat), int(pop)))
    return villes


NOMS_PAYS = dict(NOMS_SRC)
CONTOURS_PAYS = {code: _decoder_anneaux(t) for code, t in FRONTIERES_SRC.items()}
VILLES_PAYS = {code: _decoder_villes(t) for code, t in VILLES_SRC.items()}

# Emprise (ouest, est, sud, nord) de chaque pays, déduite de son contour.
EMPRISES_PAYS = {}
for _code, _anneaux in CONTOURS_PAYS.items():
    _xs = np.concatenate([a[:, 0] for a in _anneaux])
    _ys = np.concatenate([a[:, 1] for a in _anneaux])
    EMPRISES_PAYS[_code] = (float(_xs.min()), float(_xs.max()),
                            float(_ys.min()), float(_ys.max()))

# Chemins matplotlib (construits une fois) et boîtes englobantes, pour que les
# tests d'appartenance restent rapides même sur 2400 × 2400 pixels.
_CHEMINS = {c: [_Chemin(a) for a in anneaux] for c, anneaux in CONTOURS_PAYS.items()}


# Quelques codes ISO que Natural Earth écrit autrement.
ALIAS_PAYS = {"SSD": "SDS", "PSE": "PSX", "ESH": "SAH", "SOMALILAND": "SOL"}


def verifier_pays(code):
    """Arrête le notebook tout de suite, et lisiblement, si le code est inconnu."""
    code = str(code).strip().upper()
    code = ALIAS_PAYS.get(code, code)
    if code in CONTOURS_PAYS:
        return code
    catalogue = sorted((NOMS_SRC.get(c, c), c) for c in CONTOURS_PAYS)
    lignes, tampon = [], []
    for nom, c in catalogue:
        tampon.append(f"{c} {nom[:20]}")
        if len(tampon) == 3:
            lignes.append("   " + "   ".join(f"{x:<26}" for x in tampon))
            tampon = []
    if tampon:
        lignes.append("   " + "   ".join(f"{x:<26}" for x in tampon))
    raise SystemExit(
        f"\nCode pays inconnu : « {code} ». Utilisez un code ISO 3166-1 alpha-3.\n"
        f"Les {len(CONTOURS_PAYS)} pays disponibles :\n" + "\n".join(lignes)
        + "\n\nCorrigez PAYS à l'étape 02, puis réexécutez à partir de là.")


PAYS = verifier_pays(PAYS)
NOM_PAYS = NOMS_PAYS.get(PAYS, PAYS)
EMPRISE_PAYS = EMPRISES_PAYS[PAYS]


# --- 2. Appartenance géographique --------------------------------------------
def masque_pays(LON, LAT, code=None):
    """Masque booléen : True pour les pixels situés à l'intérieur du pays.

    Un test point-dans-polygone, que matplotlib sait faire sans aucune
    bibliothèque SIG — c'est ce qui garde le notebook portable.
    """
    code = PAYS if code is None else code
    LON = np.asarray(LON)
    points = np.column_stack([LON.ravel(), np.asarray(LAT).ravel()])
    dedans = np.zeros(points.shape[0], dtype=bool)
    for chemin in _CHEMINS.get(code, []):
        dedans |= chemin.contains_points(points)
    return dedans.reshape(LON.shape)


def masque_terre(LON, LAT):
    """Masque booléen : True sur la terre ferme (tous pays confondus).

    Sert au générateur de démonstration à placer océans et littoraux au bon
    endroit, quel que soit le pays choisi.
    """
    LON = np.asarray(LON)
    o, e = float(LON.min()), float(LON.max())
    s, n = float(np.asarray(LAT).min()), float(np.asarray(LAT).max())
    points = np.column_stack([LON.ravel(), np.asarray(LAT).ravel()])
    terre = np.zeros(points.shape[0], dtype=bool)
    for code, (po, pe, ps, pn) in EMPRISES_PAYS.items():
        if pe < o or po > e or pn < s or ps > n:      # pays hors du cadre
            continue
        for chemin in _CHEMINS[code]:
            terre |= chemin.contains_points(points)
    return terre.reshape(LON.shape)


def grille_coordonnees(emprise, forme):
    """Grilles LON / LAT (centres de pixels) pour une emprise et une forme données."""
    ouest, est, sud, nord = emprise
    lignes, colonnes = forme
    lon = ouest + (np.arange(colonnes) + 0.5) * (est - ouest) / colonnes
    lat = nord - (np.arange(lignes) + 0.5) * (nord - sud) / lignes
    return np.meshgrid(lon, lat)


def tracer_contour_pays(ax, code=None, couleur="white", lw=1.5, alpha=0.95, etiquette=None):
    """Superpose la frontière nationale sur une carte déjà tracée."""
    code = PAYS if code is None else code
    for i, anneau in enumerate(CONTOURS_PAYS.get(code, [])):
        ax.plot(anneau[:, 0], anneau[:, 1], color=couleur, lw=lw, alpha=alpha,
                zorder=6, solid_joinstyle="round",
                label=etiquette if i == 0 else None)


def poids_de_la_ville(population):
    """Population -> poids lumineux relatif, compressé comme dans la réalité.

    Une ville dix fois plus peuplée n'éclaire pas dix fois plus : l'exposant 0,45
    reproduit cette compression. Sert à la fois au générateur de démonstration et
    à la taille des cercles sur les cartes.
    """
    return float(np.clip((population / 3.0e6) ** 0.45, 0.07, 1.25))


def villes_du_pays(code=None, nombre=8):
    """Les principales villes du pays, de la plus peuplée à la moins peuplée."""
    code = PAYS if code is None else code
    return VILLES_PAYS.get(code, [])[:nombre]


def villes_dans(emprise, nombre=40):
    """Toutes les villes de référence situées dans une emprise donnée."""
    ouest, est, sud, nord = emprise
    retenues = []
    for liste in VILLES_PAYS.values():
        for nom, lon, lat, pop in liste:
            if ouest <= lon <= est and sud <= lat <= nord:
                retenues.append((nom, lon, lat, pop))
    return sorted(retenues, key=lambda v: -v[3])[:nombre]


def tuiles_du_pays(code=None):
    """Les tuiles Black Marble nécessaires pour couvrir le pays."""
    code = PAYS if code is None else code
    return tuiles_couvrant(*EMPRISES_PAYS[code])


TUILES_PAYS = tuiles_du_pays()


# --- 3. Le fil rouge : une seule question, suivie d'un bout à l'autre ---------
FIL_ROUGE = []


def noter_fil_rouge(etape, libelle, somme, pixels, perimetre, commentaire=""):
    FIL_ROUGE.append({"étape": etape, "traitement": libelle,
                      "somme des lumières": float(somme), "pixels retenus": int(pixels),
                      "périmètre": perimetre, "commentaire": commentaire})
    print(("   [FIL ROUGE] étape %02d · %-34s SoL = %15.0f"
           % (etape, libelle, somme)).replace(",", " "))


def tableau_fil_rouge():
    if not FIL_ROUGE:
        return pd.DataFrame()
    t = pd.DataFrame(FIL_ROUGE)
    variations = [""]
    for i in range(1, len(t)):
        avant, apres = t["somme des lumières"].iloc[i - 1], t["somme des lumières"].iloc[i]
        variations.append("%+.1f %%" % (100 * (apres - avant) / avant) if avant else "—")
    t["variation"] = variations
    t["somme des lumières"] = t["somme des lumières"].map(lambda v: f"{v:,.0f}".replace(",", " "))
    t["pixels retenus"] = t["pixels retenus"].map(lambda v: f"{v:,}".replace(",", " "))
    return t.set_index("étape")


# --- 4. Ce que le pays choisi implique ---------------------------------------
_villes = villes_du_pays(nombre=5)
print("=" * 74)
print(f"  PAYS SUIVI        : {PAYS} — {NOM_PAYS}")
print(f"  Emprise           : longitude [{EMPRISE_PAYS[0]:+.2f}° , {EMPRISE_PAYS[1]:+.2f}°]"
      f"   latitude [{EMPRISE_PAYS[2]:+.2f}° , {EMPRISE_PAYS[3]:+.2f}°]")
print(f"  Tuiles à couvrir  : {', '.join(TUILES_PAYS)}   ({len(TUILES_PAYS)} tuile(s))")
print(f"  Villes repères    : {', '.join(v[0] for v in _villes) if _villes else 'aucune'}")
print(f"  Question du fil rouge : quelle est la somme des lumières de {NOM_PAYS} ?")
print("=" * 74)
````

 padding:9px 17px;margin:14px 0 2px 0;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
 <span style="color:#D49A00;font-size:10.5px;font-weight:700;letter-spacing:2px;text-transform:uppercase;">À retenir</span>
 <span style="color:#231F20;font-size:13.5px;margin-left:12px;">Changez les trois lettres de <code>PAYS</code> et le notebook entier change de pays : tuiles, frontière, villes, cartes et chiffres.</span>

#### Une seule façon de dessiner une carte nocturne

Les cartes de lumières nocturnes sont difficiles à lire, et presque toujours mal faites. Trois
erreurs reviennent sans cesse : une **échelle linéaire** (99 % des pixels s'écrasent dans le
premier pour-cent de la palette, et la carte paraît noire) ; une **barre de couleur graduée en
`log1p`**, que personne ne sait relire ; et des **trous indiscernables du noir**, de sorte qu'on
ne voit pas que la donnée manque.

La cellule suivante règle les trois une fois pour toutes. Toutes les cartes du notebook passent
par `carte_nocturne()` et `barre_de_couleur()`, donc toutes partagent :

- une **échelle logarithmique** avec un plancher à 0,15 nW·cm⁻²·sr⁻¹ — le noir du ciel rural ;
- une barre de couleur graduée en **vraies unités physiques**, pas en logarithme ;
- un **gris clair unique** pour « pas de donnée », partout dans le notebook ;
- des **légendes posées sous la carte**, qui ne masquent jamais les données ;
- des **villes repères** entourées et nommées, avec un halo pour rester lisibles sur fond noir.

````python
# =============================================================================
#  ÉTAPE 03 a-ter — UNE SEULE FAÇON DE DESSINER UNE CARTE NOCTURNE
# =============================================================================
#  Toutes les cartes du notebook passent par ces quelques fonctions. Conséquence :
#  une seule échelle de couleurs, une seule convention pour les trous, une barre
#  de couleur graduée en VRAIES unités physiques — et non en log1p, que personne
#  ne sait relire — et une barre d'échelle en kilomètres sur chaque carte.
# -----------------------------------------------------------------------------
import matplotlib.patheffects as patheffects

PLANCHER_RADIANCE = 0.15        # nW·cm⁻²·sr⁻¹ : le noir du ciel rural
GRIS_SANS_DONNEE = "#A9B5B0"    # la couleur, partout, de « pas de donnée »


def halo(couleur):
    """Contour de lisibilité : sombre derrière un texte clair, et l'inverse.

    Sans cela, une étiquette blanche disparaît sur une carte claire et une
    étiquette noire disparaît sur une carte nocturne.
    """
    claire = str(couleur).lower() in ("white", "#fff", "#ffffff")
    fond = "#0B1A14CC" if claire else "#FFFFFFCC"
    return [patheffects.withStroke(linewidth=2.4, foreground=fond)]


def carte_nocturne(ax, radiance, emprise, vmin=PLANCHER_RADIANCE, vmax=None,
                   interpolation="nearest"):
    """Trace une radiance en échelle logarithmique, les trous en gris.

    L'échelle logarithmique n'est pas une coquetterie : en linéaire, 99 % des
    pixels sont écrasés dans le premier pour-cent de la palette, et la carte
    paraît uniformément noire. (C'est démontré à l'étape 15.)
    """
    fini = radiance[np.isfinite(radiance) & (radiance > 0)]
    if vmax is None:
        vmax = float(np.nanpercentile(fini, 99.7)) if fini.size else 10 * vmin
    vmax = max(float(vmax), vmin * 20)
    visible = np.where(np.isfinite(radiance), np.clip(radiance, vmin, None), np.nan)
    image = ax.imshow(visible, cmap=CMAP_NUIT, norm=LogNorm(vmin=vmin, vmax=vmax),
                      extent=emprise, origin="upper", aspect="auto",
                      interpolation=interpolation)
    ax.set_facecolor(GRIS_SANS_DONNEE)
    return image


def barre_de_couleur(fig, image, ax, etiquette="radiance  (nW·cm⁻²·sr⁻¹)", fraction=0.046):
    """La barre de couleur : toujours au même endroit, toujours étiquetée."""
    barre = fig.colorbar(image, ax=ax, fraction=fraction, pad=0.025)
    barre.set_label(etiquette, fontsize=8.5, color=COULEURS["ardoise"])
    barre.ax.tick_params(labelsize=7.5)
    barre.outline.set_edgecolor(COULEURS["sauge"])
    return barre


def barre_echelle(ax, couleur="white"):
    """Barre d'échelle en kilomètres, en bas à droite de la carte.

    Sans elle, une carte en degrés ne dit rien de la taille réelle de ce qu'on
    regarde — et c'est précisément ce que les participants doivent acquérir :
    un pixel fait 500 m, une tuile 1 100 km.
    """
    o, e = ax.get_xlim()
    s, n = ax.get_ylim()
    km_par_degre = 111.32 * max(np.cos(np.deg2rad((s + n) / 2)), 0.15)
    largeur_km = abs(e - o) * km_par_degre
    longueur_km = next((c for c in (1000, 500, 200, 100, 50, 20, 10, 5, 2)
                        if c <= 0.34 * largeur_km), 1)
    longueur_deg = longueur_km / km_par_degre
    x1 = e - 0.055 * (e - o)
    x0 = x1 - longueur_deg
    y = s + 0.055 * (n - s)
    ligne, = ax.plot([x0, x1], [y, y], color=couleur, lw=3.0,
                     solid_capstyle="butt", zorder=9)
    ligne.set_path_effects(halo(couleur))
    texte = ax.text((x0 + x1) / 2, y + 0.022 * (n - s), f"{longueur_km} km",
                    ha="center", va="bottom", fontsize=8, color=couleur,
                    fontweight="bold", zorder=9)
    texte.set_path_effects(halo(couleur))


def legende_sous_laxe(ax, entrees, colonnes=4, titre=None):
    """Une légende posée SOUS la carte : elle ne masque jamais les données."""
    poignees = [plt.Rectangle((0, 0), 1, 1, fc=couleur, ec="none") for _, couleur in entrees]
    etiquettes = [nom for nom, _ in entrees]
    return ax.legend(poignees, etiquettes, loc="upper center", bbox_to_anchor=(0.5, -0.14),
                     ncol=colonnes, fontsize=8, frameon=False, title=titre,
                     title_fontsize=8.5, handlelength=1.2, handleheight=1.2,
                     columnspacing=1.3, borderpad=0.2)


def etiqueter_villes(ax, villes, couleur="white", taille=8.0, contour=True):
    """Pose les villes repères sur une carte, avec un halo pour rester lisible.

    L'étiquette bascule à gauche du point quand celui-ci est près du bord droit :
    sinon le nom de la plus grande ville du pays se retrouve coupé.
    """
    o, e = ax.get_xlim()
    bas, haut = ax.get_ylim()
    decalage = 0.015 * abs(e - o)
    # Anti-chevauchement : deux villes trop proches à l'écran ne reçoivent
    # qu'une étiquette — sinon les noms se superposent et plus rien n'est lisible.
    ecart_min = 0.055 * abs(e - o)
    posees = []
    for nom_ville, vlon, vlat, population in villes:
        if not (o <= vlon <= e and bas <= vlat <= haut):
            continue
        if any(abs(vlon - x) < ecart_min and abs(vlat - y) < ecart_min * 0.55
               for x, y in posees):
            continue
        posees.append((vlon, vlat))
        ax.scatter([vlon], [vlat], s=12 + 46 * poids_de_la_ville(population),
                   facecolor="none", edgecolor=couleur, lw=1.1, zorder=7)
        a_gauche = (vlon - o) > 0.78 * (e - o)
        texte = ax.text(vlon + (-decalage if a_gauche else decalage),
                        vlat + decalage * 0.6, nom_ville,
                        ha="right" if a_gauche else "left",
                        fontsize=taille, color=couleur, zorder=8)
        if contour:
            texte.set_path_effects(halo(couleur))


def carte_categorielle(ax, codes, emprise, legende, couleurs, titre=None):
    """Carte d'une couche de codes entiers (qualité, terre/eau, nuage).

    Renvoie la liste (libellé, couleur) des classes réellement présentes, pour
    la légende — on ne montre jamais une classe absente du fichier.
    """
    presents = sorted(int(k) for k in np.unique(codes))
    indexe = np.zeros(codes.shape, dtype="int16")
    for i, code in enumerate(presents):
        indexe[codes == code] = i
    ax.imshow(indexe, cmap=ListedColormap([couleurs.get(k, COULEURS["sauge"])
                                           for k in presents]),
              extent=emprise, origin="upper", aspect="auto", interpolation="nearest",
              vmin=-0.5, vmax=len(presents) - 0.5)
    ax.set_facecolor(GRIS_SANS_DONNEE)
    if titre:
        ax.set_title(titre, pad=8, fontsize=10.5)
    return [(f"{k} · {str(legende.get(k, 'non documenté'))[:28]}",
             couleurs.get(k, COULEURS["sauge"])) for k in presents]


print("Conventions de carte chargées : échelle logarithmique, gris = pas de donnée,")
print("barre de couleur en nW·cm⁻²·sr⁻¹, barre d'échelle en kilomètres.")
````

 <span style="color:#D49A00;font-size:10.5px;font-weight:700;letter-spacing:2px;text-transform:uppercase;">À retenir</span>
 <span style="color:#231F20;font-size:13.5px;margin-left:12px;">Une carte de lumières nocturnes sans échelle logarithmique et sans barre de couleur en unités physiques n’est pas une carte : c’est une tache.</span>
</div>

n’est trouvée**. Ce qu’il produit n’est pas un fichier « bidon » : c’est un fichier HDF5 véritable,

  <div style="color:#231F20;font-size:13px;line-height:1.55;">Distribution plausible des classes de qualité, masque terre/eau, masque nuage encodé bit à bit.</div>
 </div></td></tr></table>

````python
# =============================================================================
#  ÉTAPE 03 b — FILET DE SÉCURITÉ : GÉNÉRATEUR DE FICHIERS DE DÉMONSTRATION
#  (défini maintenant, appelé seulement si aucune donnée réelle n'est obtenue)
# =============================================================================

# --- Spécification des couches, calquée sur la collection 2 (version 002) ----
SPEC_COMPOSITE = {   # produits VNP46A3 (mensuel) et VNP46A4 (annuel)
    "AllAngle_Composite_Snow_Free":
        dict(type="uint16", echelle=0.1, remplissage=65535, unite="nW_cm-2_sr-1",
             titre="All angle composite, snow-free, BRDF-corrected NTL"),
    "AllAngle_Composite_Snow_Free_Num":
        dict(type="uint8", echelle=1.0, remplissage=255, unite="Number of Observations",
             titre="Number of observations, all-angle snow-free composite"),
    "AllAngle_Composite_Snow_Free_Std":
        dict(type="uint16", echelle=0.1, remplissage=65535, unite="nW_cm-2_sr-1",
             titre="Standard deviation, all-angle snow-free composite"),
    "AllAngle_Composite_Snow_Free_Quality":
        dict(type="uint8", echelle=1.0, remplissage=255, unite="Quality Flag",
             titre="Quality flag, all-angle snow-free composite"),
    "NearNadir_Composite_Snow_Free":
        dict(type="uint16", echelle=0.1, remplissage=65535, unite="nW_cm-2_sr-1",
             titre="Near-nadir composite, snow-free, BRDF-corrected NTL"),
    "NearNadir_Composite_Snow_Free_Num":
        dict(type="uint8", echelle=1.0, remplissage=255, unite="Number of Observations",
             titre="Number of observations, near-nadir snow-free composite"),
    "OffNadir_Composite_Snow_Free":
        dict(type="uint16", echelle=0.1, remplissage=65535, unite="nW_cm-2_sr-1",
             titre="Off-nadir composite, snow-free, BRDF-corrected NTL"),
    "Land_Water_Mask":
        dict(type="uint8", echelle=1.0, remplissage=255, unite="none",
             titre="Land/Water background mask"),
    "DNB_Platform":
        dict(type="uint8", echelle=1.0, remplissage=255, unite="none",
             titre="Source satellite platform"),
}

SPEC_JOURNALIER = {   # produit VNP46A2
    "DNB_BRDF-Corrected_NTL":
        dict(type="uint16", echelle=0.1, remplissage=65535, unite="nW_cm-2_sr-1",
             titre="BRDF-corrected nighttime lights radiance"),
    "Gap_Filled_DNB_BRDF-Corrected_NTL":
        dict(type="uint16", echelle=0.1, remplissage=65535, unite="nW_cm-2_sr-1",
             titre="Gap-filled BRDF-corrected nighttime lights radiance"),
    "DNB_Lunar_Irradiance":
        dict(type="uint16", echelle=0.1, remplissage=65535, unite="nW_cm-2_sr-1",
             titre="Lunar irradiance at observation time"),
    "Latest_High_Quality_Retrieval":
        dict(type="uint8", echelle=1.0, remplissage=255, unite="Days",
             titre="Days since latest high-quality retrieval"),
    "Mandatory_Quality_Flag":
        dict(type="uint8", echelle=1.0, remplissage=255, unite="none",
             titre="Mandatory quality flag"),
    "QF_Cloud_Mask":
        dict(type="uint16", echelle=1.0, remplissage=65535, unite="none",
             titre="VIIRS cloud mask quality flags (bit field)"),
    "Snow_Flag":
        dict(type="uint8", echelle=1.0, remplissage=255, unite="none",
             titre="Snow / no-snow flag"),
}

# --- Descriptions officielles des drapeaux (reprises des guides NASA) --------
DESC_QUALITE_COMPOSITE = "0 = Good quality 1 = Poor quality 2 = Gap filled 255 = Fill value"
DESC_QUALITE_JOURNALIER = ("0 = High-quality, persistent nighttime lights "
                           "1 = High-quality, ephemeral nighttime lights "
                           "2 = Poor-quality, outlier corrected by VIIRS 255 = No retrieval")
DESC_TERRE_EAU = ("0 = Shallow ocean 1 = Land 2 = Coastline 3 = Shallow inland water "
                  "4 = Ephemeral water 5 = Deep inland water 6 = Continental 7 = Deep ocean")


def _voisinage(masque, rayon_px):
    """True si au moins un pixel vrai dans un carré de côté 2·rayon+1.

    Calculé par image intégrale : le coût ne dépend pas du rayon, ce qui permet
    de dessiner un plateau continental de plus d'un degré sans ralentir.
    """
    if rayon_px <= 0:
        return masque.copy()
    integrale = np.cumsum(np.cumsum(masque.astype("int32"), axis=0), axis=1)
    integrale = np.pad(integrale, ((1, 0), (1, 0)))
    h, w = masque.shape
    l0 = np.clip(np.arange(h) - rayon_px, 0, h)
    l1 = np.clip(np.arange(h) + rayon_px + 1, 0, h)
    c0 = np.clip(np.arange(w) - rayon_px, 0, w)
    c1 = np.clip(np.arange(w) + rayon_px + 1, 0, w)
    somme = (integrale[np.ix_(l1, c1)] - integrale[np.ix_(l0, c1)]
             - integrale[np.ix_(l1, c0)] + integrale[np.ix_(l0, c0)])
    return somme > 0


_GEOGRAPHIE_TUILE = {}


def geographie_tuile(tuile, cote):
    """Terre, mer, littoral et villes d'une tuile — calculés une seule fois.

    Tout vient des frontières réelles embarquées à l'étape 03 a-bis : le jeu de
    démonstration ressemble donc au pays choisi, quel que soit ce pays.
    """
    cle = (tuile, cote)
    if cle in _GEOGRAPHIE_TUILE:
        return _GEOGRAPHIE_TUILE[cle]

    emprise = emprise_tuile(tuile)
    LON, LAT = grille_coordonnees(emprise, (cote, cote))
    terre = masque_terre(LON, LAT)
    resolution = TUILE_DEGRES / cote

    def rayon(degres):
        return max(int(round(degres / resolution)), 1)

    littoral = terre & _voisinage(~terre, rayon(0.10))         # terre au bord de l'eau
    plateau = (~terre) & _voisinage(terre, rayon(1.10))        # mer peu profonde
    large = (~terre) & ~plateau                                # océan profond

    terre_eau = np.ones(LON.shape, dtype="uint8")
    terre_eau[plateau] = 0
    terre_eau[large] = 7
    terre_eau[littoral] = 2

    # Reflet lunaire et trafic côtier : la mer n'est pas noire, et c'est ce qui
    # fausse les indicateurs côtiers quand on oublie de masquer l'eau.
    # La décroissance depuis la côte doit être CONTINUE : des paliers donneraient
    # des anneaux parfaitement nets, que l'on ne voit dans aucune image réelle.
    paliers = np.linspace(0.04, 1.15, 11)
    proximite = np.zeros(LON.shape, dtype="float32")
    for degres in paliers:
        proximite += (_voisinage(terre, rayon(degres)) & (~terre)).astype("float32")
    proximite /= len(paliers)                       # 1 au bord de l'eau, 0 au large
    reflet = (0.20 * proximite ** 2.4).astype("float32")

    villes = villes_dans(emprise, nombre=25)
    _GEOGRAPHIE_TUILE[cle] = dict(LON=LON, LAT=LAT, terre=terre, terre_eau=terre_eau,
                                  plateau=plateau, reflet=reflet, villes=villes,
                                  resolution=resolution)
    return _GEOGRAPHIE_TUILE[cle]


def _champ_radiance(tuile, cote, graine=0, intensite=1.0, bruit=0.05):
    """Fabrique un champ de radiance plausible pour cette tuile."""
    geo = geographie_tuile(tuile, cote)
    LON, LAT, terre = geo["LON"], geo["LAT"], geo["terre"]
    rng = np.random.default_rng(graine + numeros_tuile(tuile)[0] * 101 + numeros_tuile(tuile)[1])
    # Le fond rural doit être CONTINU d'une tuile à l'autre, sinon la mosaïque de
    # l'étape 14 montre des rectangles de luminosité : on tire donc ses phases
    # d'un générateur qui ne dépend PAS de la tuile.
    rng_fond = np.random.default_rng(graine)

    # Le fond rural : une variation LISSE (relief, densité de hameaux) plus un
    # grain fin. Un fond purement aléatoire donnerait une image de bruit, et
    # masquerait complètement la structure urbaine qui nous intéresse.
    _lisse = _champ_lisse(LON, LAT, rng_fond, modes=5, echelle=(0.8, 3.0))
    _lisse = (_lisse - _lisse.mean()) / max(float(_lisse.std()), 1e-6)
    _grain = rng.normal(0.0, 1.0, LON.shape)
    # Distribution LOG-NORMALE : c'est la forme qu'ont réellement les radiances
    # rurales, et c'est elle que l'histogramme de l'étape 15 doit faire voir.
    _fond = np.exp(-2.35 + 0.55 * _lisse + 0.42 * _grain)
    champ = np.where(terre, _fond, 0.0).astype("float32")

    # --- les villes -----------------------------------------------------------
    #  Une agglomération n'est pas une tache ronde. On la construit en trois
    #  couches : un coeur, des sous-centres, des axes sortants — puis on module
    #  le tout par une texture fine, pour que le zoom de l'étape 14 montre du
    #  bâti et non un dégradé.
    urbain = np.zeros(LON.shape, dtype="float32")
    for nom_ville, clon, clat, population in geo["villes"]:
        poids = poids_de_la_ville(population)
        # sigma = rayon du coeur dense. 0,031° ≈ 3,4 km pour une capitale :
        # c'est l'ordre de grandeur d'un centre-ville réellement saturé.
        sigma = 0.010 + 0.022 * poids
        rng_ville = np.random.default_rng(int(abs(clon) * 1000 + abs(clat) * 37))
        lobe = np.zeros(LON.shape, dtype="float32")
        for amplitude, rayon in [(190.0, 1.0), (24.0, 3.0), (4.5, 8.0)]:
            lobe += amplitude * poids * intensite * np.exp(
                -(((LON - clon) / (sigma * rayon)) ** 2
                  + ((LAT - clat) / (sigma * rayon)) ** 2) / 2)
        for _k in range(9):                                  # quartiers, sous-centres
            angle = rng_ville.uniform(0, 2 * np.pi)
            dist = rng_ville.uniform(1.0, 4.0) * sigma
            rayon_lobe = sigma * rng_ville.uniform(0.25, 0.50)
            lobe += (55.0 * poids * intensite * rng_ville.uniform(0.3, 1.0)) * np.exp(
                -(((LON - clon - dist * np.cos(angle)) / rayon_lobe) ** 2
                  + ((LAT - clat - dist * np.sin(angle)) / rayon_lobe) ** 2) / 2)
        for _k in range(4):                                  # axes routiers sortants
            angle = rng_ville.uniform(0, np.pi)
            le_long = (LON - clon) * np.cos(angle) + (LAT - clat) * np.sin(angle)
            en_travers = -(LON - clon) * np.sin(angle) + (LAT - clat) * np.cos(angle)
            lobe += 16.0 * poids * intensite \
                * np.exp(-(en_travers / (sigma * 0.18)) ** 2 / 2) \
                * np.exp(-(le_long / (sigma * 4.0)) ** 2 / 2)
        urbain += lobe

    # Texture du bâti : sans elle, une ville reste un dégradé parfaitement lisse,
    # alors qu'une vraie agglomération est granuleuse à 500 m de résolution.
    if geo["villes"]:
        texture = _champ_lisse(LON, LAT, rng, modes=8, echelle=(40.0, 130.0))
        urbain = urbain * (1.0 + 0.30 * texture / max(float(np.abs(texture).max()), 1e-6))

    champ = champ + urbain

    if bruit:
        champ *= (1 + rng.normal(0, bruit, LON.shape)).clip(0.4, 2.0)

    # La mer : reflet lunaire, et quelques navires ou pêcheries à lumière.
    if geo["plateau"].any():
        rng_mer = np.random.default_rng(graine + 23)
        mer = geo["reflet"].copy()
        indices = np.flatnonzero(geo["plateau"].ravel())
        for _ in range(min(6, indices.size)):
            k = int(rng_mer.choice(indices))
            plon, plat = LON.ravel()[k], LAT.ravel()[k]
            mer = mer + 7.0 * np.exp(-(((LON - plon) / 0.045) ** 2
                                       + ((LAT - plat) / 0.045) ** 2) / 2)
        mer = mer * np.exp(rng_mer.normal(0.0, 0.33, LON.shape)).clip(0.3, 3.0)
        # On AJOUTE les lumières urbaines à la mer au lieu de les écraser : une
        # capitale portuaire déborde sur l'eau, et un trait de côte simplifié
        # classerait sinon la moitié de la ville en « océan ».
        champ = np.where(geo["plateau"], mer + 0.4 * urbain, champ)

    return champ.clip(0, 6000).astype("float32"), LON, LAT


def _encoder(valeurs, spec):
    """Réel -> entier stocké, en respectant le facteur d'échelle."""
    brut = np.round(np.nan_to_num(valeurs) / spec["echelle"])
    return np.clip(brut, 0, spec["remplissage"] - 1).astype(spec["type"])


def _octets(texte):
    """Encode une chaîne en octets UTF-8 pour un attribut HDF5."""
    return np.frombuffer(str(texte).encode("utf-8"), dtype="S1").tobytes()


def _ecrire_couche(groupe, nom, tableau, spec, extra=None):
    """Écrit un jeu de données avec la panoplie complète d'attributs NASA."""
    jeu = groupe.create_dataset(nom, data=tableau, compression="gzip", compression_opts=4)
    jeu.attrs["long_name"] = _octets(spec["titre"])
    jeu.attrs["units"] = _octets(spec["unite"])
    jeu.attrs["scale_factor"] = np.array([spec["echelle"]], dtype="float64")
    jeu.attrs["add_offset"] = np.array([0.0], dtype="float64")
    jeu.attrs["_FillValue"] = np.array([spec["remplissage"]], dtype=tableau.dtype)
    jeu.attrs["valid_min"] = np.array([0], dtype=tableau.dtype)
    jeu.attrs["valid_max"] = np.array([spec["remplissage"] - 1], dtype=tableau.dtype)
    for cle, val in (extra or {}).items():
        jeu.attrs[cle] = _octets(val) if isinstance(val, str) else val


def _struct_metadata(tuile, cote, grille):
    """Reproduit le bloc StructMetadata.0 propre au format HDF-EOS5."""
    ouest, est, sud, nord = emprise_tuile(tuile)
    return (f"GROUP=SwathStructure\nEND_GROUP=SwathStructure\n"
            f"GROUP=GridStructure\n\tGROUP=GRID_1\n"
            f"\t\tGridName=\"{grille}\"\n\t\tXDim={cote}\n\t\tYDim={cote}\n"
            f"\t\tUpperLeftPointMtrs=({ouest * 1e6:.6f},{nord * 1e6:.6f})\n"
            f"\t\tLowerRightMtrs=({est * 1e6:.6f},{sud * 1e6:.6f})\n"
            f"\t\tProjection=HE5_GCTP_GEO\n\t\tGridOrigin=HE5_HDFE_GD_UL\n"
            f"\tEND_GROUP=GRID_1\nEND_GROUP=GridStructure\nEND\n")


def _champ_lisse(LON, LAT, rng, modes=6, echelle=(0.4, 2.6)):
    """Champ aléatoire lisse : sert aux défauts de qualité et aux nuages, qui
    forment des taches et jamais du bruit poivre-et-sel."""
    champ = np.zeros(LON.shape)
    for _k in range(modes):
        fx, fy = rng.uniform(echelle[0], echelle[1], 2)
        px, py = rng.uniform(0, 2 * np.pi, 2)
        champ += np.sin(fx * LON + px) * np.sin(fy * LAT + py)
    return champ


def fabriquer_fichier_demo(chemin, produit, tuile, annee, jour, cote=480,
                           graine=0, intensite=1.0):
    """Écrit un fichier .h5 de démonstration structurellement conforme."""
    chemin = Path(chemin)
    chemin.parent.mkdir(parents=True, exist_ok=True)
    journalier = produit == "VNP46A2"
    grille = "VNP_Grid_DNB" if journalier else "VIIRS_Grid_DNB_2d"
    spec_produit = SPEC_JOURNALIER if journalier else SPEC_COMPOSITE

    radiance, LON, LAT = _champ_radiance(tuile, cote, graine, intensite,
                                         bruit=0.25 if journalier else 0.05)
    geo = geographie_tuile(tuile, cote)
    terre_eau = geo["terre_eau"]
    rng = np.random.default_rng(graine + 7)

    with h5py.File(chemin, "w") as f:
        ouest, est, sud, nord = emprise_tuile(tuile)
        h, v = numeros_tuile(tuile)
        for cle, val in [("ShortName", produit), ("PlatformShortName", "Suomi-NPP"),
                         ("InstrumentShortName", "VIIRS"), ("ProcessingLevel", "Level 3"),
                         ("ProductionTime", f"{annee}-06-01 00:00:00.000"),
                         ("HorizontalTileNumber", f"{h:02d}"),
                         ("VerticalTileNumber", f"{v:02d}"),
                         ("DEMO_SYNTHETIC", "TRUE - structure conforme, valeurs simulees")]:
            f.attrs[cle] = _octets(val)
        for cle, val in [("WestBoundingCoord", ouest), ("EastBoundingCoord", est),
                         ("SouthBoundingCoord", sud), ("NorthBoundingCoord", nord)]:
            f.attrs[cle] = np.array([val], dtype="float64")

        f.create_group("HDFEOS INFORMATION").create_dataset(
            "StructMetadata.0", data=_octets(_struct_metadata(tuile, cote, grille)))

        groupe = f.create_group(f"HDFEOS/GRIDS/{grille}/Data Fields")
        for nom, spec in spec_produit.items():
            if nom == "Land_Water_Mask":
                _ecrire_couche(groupe, nom, terre_eau, spec, {"description": DESC_TERRE_EAU})
            elif nom == "DNB_Platform":
                _ecrire_couche(groupe, nom, np.zeros(radiance.shape, "uint8"), spec,
                               {"description": "0 = Suomi-NPP 1 = NOAA-20"})
            elif nom.endswith("_Quality") or nom == "Mandatory_Quality_Flag":
                champ_q = _champ_lisse(LON, LAT, rng, modes=9, echelle=(0.5, 4.0))
                champ_q += 0.40 * _champ_lisse(LON, LAT, rng, modes=12, echelle=(3.0, 9.0))
                part1, part2 = (0.22, 0.10) if journalier else (0.07, 0.025)
                q = np.zeros(radiance.shape, "uint8")
                q[champ_q >= np.quantile(champ_q, 1 - part1)] = 1
                q[champ_q >= np.quantile(champ_q, 1 - part2)] = 2
                q[terre_eau == 7] = 255                 # au large : aucune restitution
                _ecrire_couche(groupe, nom, q, spec, {
                    "description": DESC_QUALITE_JOURNALIER if journalier else DESC_QUALITE_COMPOSITE})
            elif nom.endswith("_Num"):
                _ecrire_couche(groupe, nom,
                               np.clip(rng.poisson(14, radiance.shape), 0, 254).astype("uint8"), spec)
            elif nom.endswith("_Std"):
                _ecrire_couche(groupe, nom, _encoder(radiance * 0.18, spec), spec)
            elif nom == "Snow_Flag":
                _ecrire_couche(groupe, nom, np.zeros(radiance.shape, "uint8"), spec,
                               {"description": "0 = No snow 1 = Snow"})
            elif nom == "QF_Cloud_Mask":
                champ_n = _champ_lisse(LON, LAT, rng, modes=6, echelle=(0.3, 1.9))
                champ_n += 0.35 * _champ_lisse(LON, LAT, rng, modes=10, echelle=(2.2, 6.0))
                detection = np.zeros(radiance.shape, dtype="uint16")
                detection[champ_n >= np.quantile(champ_n, 0.55)] = 1
                detection[champ_n >= np.quantile(champ_n, 0.76)] = 2
                detection[champ_n >= np.quantile(champ_n, 0.89)] = 3
                qf = np.full(radiance.shape, 0b10, dtype="uint16")
                qf |= (detection << 6).astype("uint16")
                _ecrire_couche(groupe, nom, qf, spec,
                               {"description": "VIIRS cloud mask, bit field"})
            elif nom == "DNB_Lunar_Irradiance":
                _ecrire_couche(groupe, nom, _encoder(np.full(radiance.shape, 12.5), spec), spec)
            elif nom == "Latest_High_Quality_Retrieval":
                _ecrire_couche(groupe, nom,
                               np.clip(rng.poisson(2, radiance.shape), 0, 254).astype("uint8"), spec)
            elif "NearNadir" in nom:
                _ecrire_couche(groupe, nom, _encoder(radiance * 0.97, spec), spec)
            elif "OffNadir" in nom:
                _ecrire_couche(groupe, nom, _encoder(radiance * 1.04, spec), spec)
            elif "Gap_Filled" in nom:
                _ecrire_couche(groupe, nom, _encoder(radiance * 1.01, spec), spec)
            else:
                # Couche de radiance principale : le large reçoit la valeur de remplissage
                encode = _encoder(radiance, spec)
                encode[terre_eau == 7] = spec["remplissage"]
                _ecrire_couche(groupe, nom, encode, spec)

        if not journalier:
            resolution = TUILE_DEGRES / cote
            axes = {"lon": (ouest + (np.arange(cote) + 0.5) * resolution,
                            "Longitude", "degrees_east"),
                    "lat": (nord - (np.arange(cote) + 0.5) * resolution,
                            "Latitude", "degrees_north")}
            for nom_axe, (valeurs, titre_axe, unite) in axes.items():
                jeu = groupe.create_dataset(nom_axe, data=valeurs.astype("float64"))
                jeu.attrs["long_name"] = _octets(titre_axe)
                jeu.attrs["units"] = _octets(unite)
                jeu.attrs["_FillValue"] = np.array([-999.9], dtype="float64")
    return chemin


def fabriquer_jeu_demo(dossier, tuiles=None, annees=(2021, 2022, 2023)):
    """Fabrique un jeu complet pour LE PAYS CHOISI : annuel, mensuel, journalier."""
    dossier = Path(dossier)
    tuiles = list(tuiles or TUILES_PAYS)

    # La tuile « principale » est celle qui contient la plus grande ville du pays.
    principales = villes_du_pays(nombre=1)
    tuile_principale = tuiles[0]
    if principales:
        _, vlon, vlat, _ = principales[0]
        for t in tuiles:
            o, e, s, n = emprise_tuile(t)
            if o <= vlon <= e and s <= vlat <= n:
                tuile_principale = t
                break

    # Un très grand pays peut demander une douzaine de tuiles : fabriquer tout
    # ferait attendre la salle pour rien. On garde les tuiles les plus proches de
    # la principale agglomération, et on le dit.
    PLAFOND_TUILES = 9
    ecartees = []
    if len(tuiles) > PLAFOND_TUILES:
        def _distance(t):
            o, e, s, n = emprise_tuile(t)
            if not principales:
                return 0.0
            _, vlon, vlat, _ = principales[0]
            return (((o + e) / 2 - vlon) ** 2 + ((s + n) / 2 - vlat) ** 2) ** 0.5
        ordonnees = sorted(tuiles, key=_distance)
        tuiles, ecartees = sorted(ordonnees[:PLAFOND_TUILES]), sorted(ordonnees[PLAFOND_TUILES:])
        if tuile_principale not in tuiles:
            tuile_principale = tuiles[0]

    # Toutes les tuiles doivent partager la MÊME finesse, sinon la mosaïque de
    # l'étape 14 ne se recolle pas. On ajuste donc le côté au nombre de tuiles.
    diviseur = {1: 1, 2: 1.35}.get(len(tuiles), 1.8 if len(tuiles) <= 4 else 2.6)
    cote = max(280, int(DEMO_PX / diviseur))

    # L'année la plus récente est fabriquée sur TOUTES les tuiles : c'est elle qui
    # sert à la mosaïque et au découpage de l'étape 14. Les années précédentes ne
    # servent qu'à montrer qu'une série existe : la tuile principale suffit.
    derniere = max(annees)
    travaux = [("VNP46A4", tuile, derniere, 1, f"annuel/{derniere}", 1.0)
               for tuile in tuiles]
    for annee in sorted(a for a in annees if a != derniere):
        travaux.append(("VNP46A4", tuile_principale, annee, 1, f"annuel/{annee}",
                        1 - 0.07 * (derniere - annee)))
    for jour in (1, 91, 182, 274):
        travaux.append(("VNP46A3", tuile_principale, derniere, jour,
                        f"mensuel/{derniere}", 1.0))
    for jour in (12, 200):
        travaux.append(("VNP46A2", tuile_principale, derniere, jour,
                        f"journalier/{derniere}", 1.0))

    print(f"  pays {PAYS} · tuiles {', '.join(tuiles)}"
          f" · {cote} × {cote} pixels par tuile"
          f" (tuile principale : {tuile_principale})")
    if ecartees:
        print(f"  tuiles écartées pour tenir le temps d'exécution : {', '.join(ecartees)}")
    print(f"  {len(travaux)} fichiers à fabriquer ...")
    ecrits = []
    for i, (produit, tuile, annee, jour, sous_dossier, intensite) in enumerate(travaux):
        nom = f"{produit}.A{annee}{jour:03d}.{tuile}.002.DEMO{i:09d}.h5"
        ecrits.append(fabriquer_fichier_demo(dossier / sous_dossier / nom, produit, tuile,
                                             annee, jour, cote=cote, graine=annee + jour,
                                             intensite=intensite))
    return ecrits


print("Générateur de démonstration prêt (il ne sera appelé qu'en dernier recours).")
````

#### Localiser les données, puis dresser l’inventaire

Le notebook cherche vos fichiers en suivant une **échelle de priorité** : du plus explicite au plus
improbable. Dès qu’un niveau donne un résultat, il s’arrête.

````python
# =============================================================================
#  ÉTAPE 03 c — OBTENIR LES DONNÉES : VOS FICHIERS, LA NASA, OU LA DÉMONSTRATION
# =============================================================================

NOMS_DOSSIERS_USUELS = ["blackmarble_data", "blackmarble", "black_marble", "BlackMarble",
                        "ntl_data", "NTL", "ntl", "data", "donnees", "Collecte",
                        "Collecte/blackmarble_data", "VNP46"]

ARCHIVE_LAADS = "https://ladsweb.modaps.eosdis.nasa.gov"
ARCHIVE_ID = 5200          # l'archive qui contient toute la suite VNP46


# --- 1. Le jeton ---------------------------------------------------------------
def jeton_earthdata():
    """Cherche le jeton dans quatre endroits, du plus explicite au plus discret.

    Un jeton est un secret personnel : la bonne pratique est de NE PAS le mettre
    dans le notebook, mais dans une variable d'environnement ou un fichier de
    votre dossier personnel. Ainsi le notebook reste partageable tel quel.
    """
    if JETON_EARTHDATA.strip():
        return JETON_EARTHDATA.strip(), "la cellule de l'étape 02"
    depuis_env = os.environ.get("EARTHDATA_TOKEN", "").strip()
    if depuis_env:
        return depuis_env, "la variable d'environnement EARTHDATA_TOKEN"
    fichier = Path.home() / ".earthdata_token"
    try:
        if fichier.exists() and fichier.read_text(encoding="utf-8").strip():
            return fichier.read_text(encoding="utf-8").strip(), f"le fichier {fichier}"
    except OSError:
        pass
    if DEMANDER_LE_JETON:
        try:
            import getpass
            print("Aucun jeton trouvé dans la cellule, l'environnement ni ~/.earthdata_token.")
            saisi = getpass.getpass(
                "Jeton Earthdata — collez-le, ou laissez vide pour le mode démonstration : ").strip()
            if saisi:
                return saisi, "une saisie au clavier"
            print("  (aucun jeton saisi : le notebook passera en mode démonstration)")
        except Exception:
            # Exécution non interactive (nbconvert, papermill, tâche planifiée) :
            # il n'y a personne au clavier, on continue sans jeton.
            print("  (exécution non interactive : impossible de demander le jeton)")
    return "", "aucune source"


JETON, ORIGINE_JETON = jeton_earthdata()


# --- 2. Le téléchargement ------------------------------------------------------
def lister_granules(produit, annee, jour, jeton, delai=40):
    """Liste les fichiers disponibles pour un produit, une année et un jour julien.

    « requests » est importé ici, et non en tête de notebook : il ne sert qu'au
    téléchargement, et son absence ne doit jamais empêcher le notebook de tourner.
    """
    import requests
    url = (f"{ARCHIVE_LAADS}/api/v2/content/details/allData/{ARCHIVE_ID}/{produit}"
           f"/{annee}/{jour:03d}?fields=all&formats=json")
    reponse = requests.get(url, headers={"Authorization": f"Bearer {jeton}"}, timeout=delai)
    reponse.raise_for_status()
    return reponse.json().get("content", []) or []


def telecharger_vnp46(produit, annee, jour, tuiles, dossier, jeton, delai=300):
    """Télécharge les granules correspondant aux tuiles demandées. Renvoie la liste écrite."""
    import requests
    dossier = Path(dossier)
    dossier.mkdir(parents=True, exist_ok=True)
    ecrits = []
    for element in lister_granules(produit, annee, jour, jeton):
        nom = element.get("name", "")
        if not nom.endswith(".h5") or not any(f".{t}." in nom for t in tuiles):
            continue
        cible = dossier / nom
        if cible.exists() and cible.stat().st_size > 0:
            ecrits.append(cible)
            continue
        lien = element.get("downloadsLink") or (
            f"/archive/allData/{ARCHIVE_ID}/{produit}/{annee}/{jour:03d}/{nom}")
        url = lien if lien.startswith("http") else ARCHIVE_LAADS + lien
        with requests.get(url, headers={"Authorization": f"Bearer {jeton}"},
                          stream=True, timeout=delai) as flux:
            flux.raise_for_status()
            with open(cible, "wb") as sortie:
                for bloc in flux.iter_content(chunk_size=1 << 20):
                    sortie.write(bloc)
        print(f"      {nom}  ({cible.stat().st_size / 1e6:.1f} Mo)")
        ecrits.append(cible)
    return ecrits


def collecter_pour_le_pays(dossier, jeton):
    """Télécharge de quoi faire tourner le notebook : 3 annuels, 1 mensuel, 1 journalier."""
    commandes = [("VNP46A4", a, 1) for a in ANNEES_VOULUES]
    commandes += [("VNP46A3", max(ANNEES_VOULUES), 1),
                  ("VNP46A2", max(ANNEES_VOULUES), 12)]
    obtenus = []
    for produit, annee, jour in commandes:
        print(f"   {produit}  {annee}  jour {jour:03d} ...")
        obtenus += telecharger_vnp46(produit, annee, jour, TUILES_PAYS, dossier, jeton)
    return obtenus


# --- 3. Localiser des fichiers déjà présents ----------------------------------
def _contient_vnp46(dossier, limite=4000):
    try:
        for i, fichier in enumerate(Path(dossier).rglob("*.h5")):
            if fichier.name.upper().startswith("VNP46"):
                return True
            if i > limite:
                break
    except (PermissionError, OSError):
        pass
    return False


def localiser_donnees(chemin_explicite=None):
    if chemin_explicite:
        d = Path(chemin_explicite).expanduser()
        if d.exists():
            return d, ("chemin explicite (DATA_DIR)" if _contient_vnp46(d)
                       else "chemin explicite (DATA_DIR) — aucun VNP46 détecté")
        print(f"  ! DATA_DIR pointe vers un dossier inexistant : {d}")
    racine_kaggle = Path("/kaggle/input")
    if racine_kaggle.exists():
        for d in sorted(racine_kaggle.iterdir()):
            if d.is_dir() and _contient_vnp46(d):
                return d, "jeu de données Kaggle"
    for d in [Path("/content/drive/MyDrive"), Path("/content")]:
        if d.exists() and _contient_vnp46(d):
            return d, "Google Colab"
    ici = Path.cwd()
    for base in [ici, *list(ici.parents)[:4]]:
        for nom in NOMS_DOSSIERS_USUELS:
            d = base / nom
            if d.exists() and _contient_vnp46(d):
                return d, "dossier local usuel"
    if _contient_vnp46(ici):
        return ici, "dossier courant"
    return None, "aucune donnée trouvée"


# --- 4. L'échelle de priorité, appliquée --------------------------------------
print("Recherche des données ...")
DOSSIER_DONNEES, ORIGINE = localiser_donnees(DATA_DIR)
MODE_DEMO = False

if DOSSIER_DONNEES is None and TELECHARGER:
    print(f"\n  Aucun fichier local. Jeton Earthdata : {ORIGINE_JETON}.")
    if not JETON:
        print("  → Sans jeton, pas de téléchargement. Pour en obtenir un :")
        print("    https://urs.earthdata.nasa.gov  puis l'onglet « Generate Token ».")
    elif not DISPO.get("requests"):
        print("  → La bibliothèque « requests » est absente : téléchargement impossible.")
    else:
        cible = Path.cwd() / "blackmarble_data"
        print(f"  Téléchargement depuis LAADS DAAC vers {cible} ...")
        try:
            obtenus = collecter_pour_le_pays(cible, JETON)
            if obtenus:
                DOSSIER_DONNEES, ORIGINE = cible, "téléchargé depuis LAADS DAAC"
                print(f"  {len(obtenus)} fichier(s) obtenus.")
            else:
                print("  Aucun granule ne correspond aux tuiles demandées.")
        except Exception as erreur:
            print(f"  Téléchargement interrompu : {type(erreur).__name__} — {erreur}")
            print("  Causes les plus fréquentes : jeton expiré, pas de connexion,")
            print("  ou accès au produit non encore accepté dans votre profil Earthdata.")

if DOSSIER_DONNEES is None:
    if AUTORISER_DEMO:
        DOSSIER_DONNEES = Path.cwd() / f"ntl_demo_{PAYS}"
        if not _contient_vnp46(DOSSIER_DONNEES):
            print(f"\n  Fabrication du jeu de démonstration pour {NOM_PAYS} ...")
            fabriquer_jeu_demo(DOSSIER_DONNEES, TUILES_PAYS, ANNEES_VOULUES)
        ORIGINE, MODE_DEMO = f"jeu de démonstration fabriqué pour {PAYS}", True
    else:
        raise SystemExit("Aucune donnée trouvée et AUTORISER_DEMO = False. "
                         "Renseignez DATA_DIR à l'étape 02, ou fournissez un jeton.")


# --- 5. Inventaire -------------------------------------------------------------
def dresser_inventaire(dossier):
    """Parcourt le dossier et catalogue tous les fichiers VNP46 exploitables."""
    lignes = []
    for fichier in sorted(Path(dossier).rglob("*.h5")):
        infos = decoder_nom_fichier(fichier.name)
        if infos is None:
            continue
        lignes.append({
            "produit": infos["produit"].upper(), "famille": infos["famille"],
            "annee": infos["annee"], "jour": infos["jour"], "date": infos["date"].date(),
            "tuile": infos["tuile"], "taille_Mo": round(fichier.stat().st_size / 1e6, 2),
            "chemin": str(fichier),
        })
    colonnes = ["produit", "famille", "annee", "jour", "date", "tuile", "taille_Mo", "chemin"]
    if not lignes:
        return pd.DataFrame(columns=colonnes)
    return (pd.DataFrame(lignes).sort_values(["produit", "annee", "jour", "tuile"])
            .reset_index(drop=True))


INVENTAIRE = dresser_inventaire(DOSSIER_DONNEES)

if INVENTAIRE.empty:
    raise SystemExit(
        f"Des fichiers .h5 existent sous {DOSSIER_DONNEES} mais aucun ne suit la convention\n"
        f"de nommage VNP46. Vérifiez qu'ils n'ont pas été renommés au téléchargement.")

tuiles_manquantes = [t for t in TUILES_PAYS if t not in set(INVENTAIRE["tuile"])]


def fichier_exemple(produit=None):
    """Un fichier représentatif : si possible celui qui contient la capitale."""
    sous = INVENTAIRE[INVENTAIRE["produit"] == produit] if produit else INVENTAIRE
    if sous.empty:
        for repli in ("VNP46A4", "VNP46A3", "VNP46A2", "VNP46A1"):
            sous = INVENTAIRE[INVENTAIRE["produit"] == repli]
            if not sous.empty:
                break
    if sous.empty:
        sous = INVENTAIRE
    # On prend l'année la PLUS RÉCENTE d'abord : sans cela, le fil rouge
    # comparerait sans le dire deux millésimes différents.
    sous = sous.sort_values(["annee", "jour"], ascending=False)
    principales = villes_du_pays(nombre=1)
    if principales:
        _, vlon, vlat, _ = principales[0]
        for _, ligne in sous.iterrows():
            o, e, s, n = emprise_tuile(ligne["tuile"])
            if o <= vlon <= e and s <= vlat <= n:
                return Path(ligne["chemin"])
    return Path(sous.iloc[0]["chemin"])


FICHIER_EXEMPLE = fichier_exemple(PRODUIT_PRINCIPAL)

# Un jeu de démonstration oublié dans le dossier de travail serait retrouvé par
# la détection automatique et pris pour de vraies données. Les fichiers fabriqués
# ici portent l'attribut global DEMO_SYNTHETIC : il suffit de le lire.
with h5py.File(FICHIER_EXEMPLE, "r") as _f_controle:
    if lire_attribut(_f_controle, "DEMO_SYNTHETIC", "") and not MODE_DEMO:
        MODE_DEMO = True
        ORIGINE += " (jeu de démonstration, reconnu à son attribut DEMO_SYNTHETIC)"

# --- 6. Restitution ------------------------------------------------------------
bandeau = (f"  MODE DÉMONSTRATION — {NOM_PAYS} — valeurs simulées, structure conforme  "
           if MODE_DEMO else f"  DONNÉES RÉELLES — {NOM_PAYS}  ")
print("\n" + "=" * 78)
print(bandeau.center(78, "*" if MODE_DEMO else "="))
print("=" * 78)
print(f"  Origine           : {ORIGINE}")
print(f"  Dossier           : {DOSSIER_DONNEES}")
print(f"  Fichiers VNP46    : {len(INVENTAIRE)}   ({INVENTAIRE['taille_Mo'].sum():.1f} Mo)")
print(f"  Produits présents : {', '.join(sorted(INVENTAIRE['produit'].unique()))}")
print(f"  Tuiles présentes  : {', '.join(sorted(INVENTAIRE['tuile'].unique()))}")
print(f"  Tuiles du pays    : {', '.join(TUILES_PAYS)}"
      + (f"   ⚠ manquantes : {', '.join(tuiles_manquantes)}" if tuiles_manquantes else "   (toutes présentes)"))
print(f"  Années couvertes  : {INVENTAIRE['annee'].min()} – {INVENTAIRE['annee'].max()}")
print(f"  Fichier d'exemple : {FICHIER_EXEMPLE.name}")
print("=" * 78)

synthese = (INVENTAIRE.groupby(["produit", "famille"])
            .agg(fichiers=("chemin", "size"),
                 annees=("annee", lambda s: f"{s.min()}–{s.max()}"),
                 tuiles=("tuile", lambda s: s.nunique()),
                 volume_Mo=("taille_Mo", "sum"))
            .reset_index())
display(synthese)
display(INVENTAIRE.drop(columns=["chemin"]).head(12))
````

<div style="background:#FDF4E0;border:1px solid #D49A0044;border-left:4px solid #D49A00;border-radius:8px;
 padding:13px 17px;margin:10px 0;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
 <div style="color:#D49A00;font-weight:700;font-size:13.5px;">&#9888;&nbsp; Si le bandeau affiche « MODE DÉMONSTRATION »</div>
 <div style="color:#231F20;font-size:13.5px;margin-top:5px;line-height:1.6;">Le notebook fonctionne, mais sur des valeurs simulées. Pour basculer sur vos données : placez vos fichiers <code>VNP46A*.h5</code> dans un dossier, renseignez <code>DATA_DIR</code> à l’étape 02, puis réexécutez à partir de là. Rien d’autre n’est à changer.</div>
</div>

#### Premier regard : à quoi ressemblent ces données ?

Avant de démonter le fichier, regardons-le. La cellule ci-dessous fait le minimum — ouvrir, décoder, afficher — sans aucun filtrage. C’est volontairement **la lecture naïve**, celle que tout le monde fait le premier jour. Elle produit déjà une image convaincante, et c’est bien le problème : rien, sur cette carte, ne signale les erreurs qui s’y cachent.

````python
# =============================================================================
#  ÉTAPE 03 d — PREMIER REGARD : VOTRE PAYS, VU DE NUIT
# =============================================================================
infos_apercu = decoder_nom_fichier(FICHIER_EXEMPLE.name)

with h5py.File(FICHIER_EXEMPLE, "r") as f:
    groupe_apercu = f[trouver_groupe_donnees(f)]
    couche_apercu = choisir_couche_radiance(groupe_apercu)
    jeu_apercu = groupe_apercu[couche_apercu]
    ech_a = lire_attribut(jeu_apercu, "scale_factor", 1.0)
    dec_a = lire_attribut(jeu_apercu, "add_offset", 0.0)
    vide_a = lire_attribut(jeu_apercu, "_FillValue", None)
    brut_apercu = jeu_apercu[:].astype("float32")

radiance_apercu = np.where(brut_apercu == vide_a, np.nan, brut_apercu * ech_a + dec_a)
emprise_apercu = emprise_tuile(infos_apercu["tuile"])
LON_APERCU, LAT_APERCU = grille_coordonnees(emprise_apercu, radiance_apercu.shape)
villes = villes_du_pays(nombre=7)
ville_phare = villes[0] if villes else None

fig, (axA, axB) = plt.subplots(1, 2, figsize=(13.0, 6.1),
                               gridspec_kw={"width_ratios": [1.1, 1]})
fig.subplots_adjust(top=0.765, bottom=0.105, left=0.055, right=0.945, wspace=0.2)

# ---- A : la tuile entière, et la frontière posée dessus ---------------------
imA = carte_nocturne(axA, radiance_apercu, emprise_apercu)
axA.set_xlim(emprise_apercu[0], emprise_apercu[1])
axA.set_ylim(emprise_apercu[2], emprise_apercu[3])
tracer_contour_pays(axA, couleur=COULEURS["or_"], lw=1.9)
etiqueter_villes(axA, [v for v in villes
                       if emprise_apercu[0] <= v[1] <= emprise_apercu[1]
                       and emprise_apercu[2] <= v[2] <= emprise_apercu[3]], taille=8.4)
barre_echelle(axA)
axA.set_xlabel("longitude (°)")
axA.set_ylabel("latitude (°)")
axA.set_title(f"La tuile {infos_apercu['tuile']} — 10° × 10°, soit ≈ 1 100 km de côté",
              pad=8, fontsize=11)

# ---- B : zoom sur la principale agglomération ------------------------------
if ville_phare is not None:
    nom_v, clon, clat, _pop = ville_phare
    marge = 0.9
    fenetre = ((np.abs(LON_APERCU - clon) <= marge) & (np.abs(LAT_APERCU - clat) <= marge)
               & np.isfinite(radiance_apercu))
    plafond_zoom = (float(np.nanpercentile(radiance_apercu[fenetre], 99.8))
                    if fenetre.any() else None)
    imB = carte_nocturne(axB, radiance_apercu, emprise_apercu, vmax=plafond_zoom)
    axB.set_xlim(clon - marge, clon + marge)
    axB.set_ylim(clat - marge, clat + marge)
    tracer_contour_pays(axB, couleur=COULEURS["or_"], lw=1.9)
    etiqueter_villes(axB, [v for v in villes_dans(
        (clon - marge, clon + marge, clat - marge, clat + marge), nombre=8)], taille=9.5)
    barre_echelle(axB)
    axB.set_title(f"{nom_v} et sa région — {int(2 * marge * 111):d} km de côté",
                  pad=8, fontsize=11)
    axA.add_patch(plt.Rectangle((clon - marge, clat - marge), 2 * marge, 2 * marge,
                                facecolor="none", edgecolor=COULEURS["or_"], lw=1.3,
                                ls=(0, (4, 3)), zorder=7))
    barre_de_couleur(fig, imB, axB)
else:
    axB.axis("off")
    barre_de_couleur(fig, imA, axA)
axB.set_xlabel("longitude (°)")

titre_figure(fig, f"{NOM_PAYS} vu de nuit — et ce que nous allons apprendre à y lire",
             "Contour or : la frontière nationale. Cercles blancs : les villes repères.\n"
             "L'échelle de couleurs est logarithmique, parce qu'un centre-ville est mille fois plus lumineux qu'une campagne.",
             f"Source : {FICHIER_EXEMPLE.name}   ·   couche {couche_apercu}   ·   frontières Natural Earth"
             + ("   ·   MODE DÉMONSTRATION — valeurs simulées, structure conforme" if MODE_DEMO else ""))
plt.show()

print(f"Couche affichée  : {couche_apercu}")
print(f"Pixels non vides : {np.isfinite(radiance_apercu).sum():,} sur {radiance_apercu.size:,}"
      .replace(",", " "))
print("Aucun filtrage n'a encore été appliqué : c'est l'image brute, décodée, rien de plus.")
````

**Le point de départ.** Le capteur qui fournit ces données s’appelle **VIIRS-DNB**
(*Visible Infrared Imaging Radiometer Suite — Day/Night Band*). Il est embarqué sur les satellites
**Suomi-NPP** (depuis 2011) et **NOAA-20** (depuis 2017), qui passent au-dessus de chaque point du globe
vers **1 h 30 du matin, heure locale**.

Une mesure brute de nuit est cependant inutilisable telle quelle : elle dépend de la **lune**, des **nuages**,
de l’**angle de visée**, de l’**aérosol**, de la **neige**. Le projet *Black Marble* de la NASA applique une
chaîne de corrections — dont la correction **BRDF** qui neutralise l’effet de l’angle et de l’éclairement
lunaire — et publie quatre produits, qui diffèrent par le **degré d’agrégation dans le temps**.

````python
# --- Schéma : le même signal, quatre niveaux d'agrégation temporelle ---------
fig, ax = plt.subplots(figsize=(11.2, 4.3))
fig.subplots_adjust(top=0.72, bottom=0.17, left=0.11, right=0.985)

rng_fig = np.random.default_rng(3)
nuits = np.arange(365)
# Signal quotidien simulé : niveau de base + saison + nuages (creux aléatoires)
base = 100 + 9 * np.sin(2 * np.pi * (nuits - 60) / 365)
nuages = rng_fig.random(365) < 0.35
quotidien = base * (1 - 0.55 * nuages * rng_fig.random(365)) + rng_fig.normal(0, 4, 365)

ax.plot(nuits, quotidien, lw=0.8, color=COULEURS["sauge"], zorder=1,
        label="VNP46A2 — une nuit (bruité : nuages, lune)")
mois = pd.Series(quotidien).groupby(nuits // 30).mean()
centres = np.array(mois.index) * 30 + 15
ax.step(centres, mois.values, where="mid", lw=2.2, color=COULEURS["sarcelle"], zorder=2,
        label="VNP46A3 — composite mensuel")
ax.axhline(quotidien.mean(), lw=2.6, color=COULEURS["or_"], zorder=3,
           label="VNP46A4 — composite annuel")
ax.plot(nuits, base, lw=1.4, ls=(0, (4, 3)), color=COULEURS["vert_fonce"], zorder=2,
        label="Signal réel sous-jacent")

ax.set_xlim(0, 364)
ax.set_ylim(0, 135)
ax.set_xlabel("Jour de l'année")
ax.set_ylabel("Radiance (unités arbitraires)")
ax.set_xticks([0, 90, 181, 273, 364])
ax.set_xticklabels(["1 jan", "1 avr", "1 jui", "1 oct", "31 déc"])
ax.legend(loc="lower left", fontsize=8.6, ncol=2)

titre_figure(fig,
             "Le même lieu, vu par trois produits différents",
             "Plus on agrège dans le temps, plus le signal est stable — et moins il est daté.\n"
             "Le produit ne change pas la réalité mesurée : il change le compromis entre précision temporelle et robustesse.",
             "Illustration pédagogique à partir d'un signal simulé — ne constitue pas une mesure.")
plt.show()
````

**L’idée.** Un nom comme `VNP46A4.A2023001.h17v08.002.2024015113045.h5` ressemble à un code-barres.
Il l’est : chacun de ses six segments porte une information exploitable **sans ouvrir le fichier**. C’est ce
qui permet de dresser un inventaire de plusieurs milliers de fichiers en quelques secondes — exactement ce
qu’a fait l’étape 03.

Le point qui surprend le plus est le troisième segment : la date n’est pas au format calendaire, mais en
**jour julien** (*day-of-year*), un nombre de 001 à 366. `A2023001` se lit donc « année 2023, jour 1 »,
c’est-à-dire le 1ᵉʳ janvier 2023.

````python
# =============================================================================
#  ÉTAPE 05 — DÉCODER LE NOM DE FICHIER
# =============================================================================
nom_exemple = FICHIER_EXEMPLE.name
infos = decoder_nom_fichier(nom_exemple)
ouest, est, sud, nord = infos["emprise"]

# Lecture du sens du jour julien selon le produit
if infos["produit"] == "VNP46A4":
    sens_date = f"année {infos['annee']} entière (le jour 001 marque le début de la période)"
elif infos["produit"] == "VNP46A3":
    sens_date = f"mois de {infos['date'].strftime('%B %Y')} (le jour indique le 1ᵉʳ du mois)"
else:
    sens_date = f"nuit du {infos['date'].strftime('%d/%m/%Y')}"

lecture = pd.DataFrame([
    ("1", "Produit", infos["produit"],
     f"{infos['famille'].capitalize()} — VNP46 = suite Black Marble, A4 = annuel, A3 = mensuel, A2 = journalier"),
    ("2", "Année", f"A{infos['annee']}",
     f"Le « A » signale une date ; l'année est {infos['annee']}"),
    ("3", "Jour julien", f"{infos['jour']:03d}",
     f"{infos['date'].strftime('%d/%m/%Y')} — {sens_date}"),
    ("4", "Tuile", infos["tuile"],
     f"longitude [{ouest:+.0f}° , {est:+.0f}°]  ·  latitude [{sud:+.0f}° , {nord:+.0f}°]"),
    ("5", "Version", infos["version"],
     "002 = Collection 2. La version 001 est obsolète : ne mélangez jamais les deux."),
    ("6", "Horodatage", infos["horodatage"],
     ("Date et heure de production par la NASA — utile pour repérer un retraitement"
      if infos["horodatage"].isdigit() else
      f"Segment de production « {infos['horodatage']} »")),
], columns=["#", "Segment", "Valeur", "Interprétation"])

print(f"FICHIER ANALYSÉ : {nom_exemple}\n")
display(lecture.set_index("#"))

# --- Vérification : combien de fichiers l'inventaire décode-t-il sans erreur ? -
total_h5 = len(list(Path(DOSSIER_DONNEES).rglob("*.h5")))
print(f"\nSur {total_h5} fichier(s) .h5 présents, {len(INVENTAIRE)} ont été décodés "
      f"par cette seule lecture du nom ({100 * len(INVENTAIRE) / max(total_h5, 1):.0f} %).")
````

````python
# --- Schéma annoté du nom de fichier ----------------------------------------
segments = [(infos["produit"], "PRODUIT", COULEURS["vert"]),
            (f"A{infos['annee']}{infos['jour']:03d}", "DATE (année + jour julien)", COULEURS["or_"]),
            (infos["tuile"], "TUILE  (10° × 10°)", COULEURS["sarcelle"]),
            (infos["version"], "VERSION", COULEURS["terre"]),
            (infos["horodatage"], "PRODUCTION", COULEURS["ardoise"]),
            ("h5", "FORMAT", COULEURS["vert_fonce"])]

largeurs = np.array([max(len(v), 6) for v, _, _ in segments], dtype=float)
largeurs = largeurs / largeurs.sum()
bords = np.concatenate([[0], np.cumsum(largeurs)])

fig, ax = plt.subplots(figsize=(11.2, 2.9))
fig.subplots_adjust(top=0.66, bottom=0.04, left=0.02, right=0.98)
ax.axis("off")
ax.set_xlim(0, 1)
ax.set_ylim(0, 1)

for i, (valeur, etiquette, couleur) in enumerate(segments):
    x0, x1 = bords[i], bords[i + 1]
    ax.add_patch(plt.Rectangle((x0 + 0.004, 0.52), x1 - x0 - 0.008, 0.26,
                               facecolor=couleur, edgecolor="none", zorder=2))
    ax.text((x0 + x1) / 2, 0.65, valeur, ha="center", va="center", color="white",
            fontsize=11, fontweight="bold", family="monospace", zorder=3)
    ax.plot([(x0 + x1) / 2, (x0 + x1) / 2], [0.50, 0.40], lw=1, color=couleur, zorder=1)
    ax.text((x0 + x1) / 2, 0.355, etiquette, ha="center", va="top", color=COULEURS["encre"],
            fontsize=8.2, fontweight="bold", wrap=True)

ax.text(0.5, 0.90, ".".join(s[0] for s in segments), ha="center", va="center",
        family="monospace", fontsize=10.5, color=COULEURS["ardoise"])
titre_figure(fig, "Anatomie d'un nom de fichier VNP46",
             "Six segments séparés par des points — chacun exploitable sans ouvrir le fichier.")
plt.show()
````

### 06 · La grille de tuiles

<div style="border-left:6px solid #00A86A;background:#E8F5EF;border-radius:0 12px 12px 0;

 <div style="color:#D49A00;font-size:11px;font-weight:700;letter-spacing:2.5px;text-transform:uppercase;">
  ÉTAPE 06 · 20 &nbsp;&nbsp;<span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span  <div style="color:#231F20;font-size:14px;margin-top:4px;">Pourquoi votre pays tient en une à six cases</div>
</div>

**Le problème que résout la grille.** Une image mondiale à 500 m de résolution pèserait des
téraoctets. La NASA découpe donc le globe en **carrés de 10° × 10°** : 36 colonnes (`h00` à `h35`, d’ouest
en est) par 18 lignes (`v00` à `v17`, du nord au sud), soit 648 tuiles. Chaque tuile fait 2400 × 2400 pixels.

Le calcul du coin supérieur gauche d’une tuile tient en deux lignes, et c’est tout ce dont vous avez besoin
pour géoréférencer un fichier :

```
longitude_ouest = -180 + 10 × h
latitude_nord   =  +90 - 10 × v
```

La conséquence pratique : **un pays chevauche presque toujours plusieurs tuiles**, et il vous faudra les
assembler (*mosaïquer*) avant toute analyse nationale. La cellule ci-dessous calcule les tuiles nécessaires
pour quelques pays africains.

````python
# =============================================================================
#  ÉTAPE 06 — LA GRILLE DE TUILES
# =============================================================================
#  Les emprises viennent des frontières Natural Earth embarquées à l'étape 03 :
#  elles sont donc cohérentes avec le découpage qui sera fait à l'étape 14.
# -----------------------------------------------------------------------------
_reperes = ["CIV", "SEN", "NGA", "MAR", "KEN", "ZAF", "EGY", "COD", "TZA", "SDN"]
_a_montrer = [PAYS] + [c for c in _reperes if c != PAYS and c in EMPRISES_PAYS][:6]

lignes = []
for code in _a_montrer:
    o, e, s, n = EMPRISES_PAYS[code]
    tuiles = tuiles_couvrant(o, e, s, n)
    lignes.append({
        "code": code,
        "pays": NOMS_PAYS.get(code, code) + ("   ◄ votre pays" if code == PAYS else ""),
        "tuiles": " ".join(tuiles),
        "nombre": len(tuiles),
        "pixels bruts": f"{len(tuiles) * TUILE_PIXELS ** 2 / 1e6:.1f} M",
        "poids annuel ≈": f"{len(tuiles) * 25:.0f} Mo",
    })
display(pd.DataFrame(lignes).set_index("code"))

print(f"\nVotre pays — {NOM_PAYS} ({PAYS})")
print(f"  Emprise          : longitude [{EMPRISE_PAYS[0]:+.2f}° , {EMPRISE_PAYS[1]:+.2f}°]"
      f"   latitude [{EMPRISE_PAYS[2]:+.2f}° , {EMPRISE_PAYS[3]:+.2f}°]")
print(f"  Tuiles à couvrir : {', '.join(TUILES_PAYS)}")

o, e, s, n = emprise_tuile(infos["tuile"])
resolution_deg = TUILE_DEGRES / TUILE_PIXELS
print(f"\nTuile du fichier d'exemple : {infos['tuile']}")
print(f"  coin supérieur gauche  : longitude {o:+.4f}°  ·  latitude {n:+.4f}°")
print(f"  coin inférieur droit   : longitude {e:+.4f}°  ·  latitude {s:+.4f}°")
print(f"  taille d'un pixel      : {resolution_deg:.8f}° = {resolution_deg * 3600:.1f}\" d'arc")
print(f"                           ≈ {resolution_deg * 111320:.0f} m en longitude à l'équateur")
print(f"  pixels dans la tuile   : {TUILE_PIXELS} × {TUILE_PIXELS} = "
      f"{TUILE_PIXELS ** 2:,} pixels".replace(",", " "))
````

````python
# --- Carte : la grille Black Marble, et le pays dedans -----------------------
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13.2, 6.6),
                               gridspec_kw={"width_ratios": [1.12, 1]})
fig.subplots_adjust(top=0.775, bottom=0.155, left=0.055, right=0.985, wspace=0.18)

tuiles_presentes = set(INVENTAIRE["tuile"].unique())
tuiles_pays = set(TUILES_PAYS)
TITRE_TUILES = (f"{NOM_PAYS} et la tuile qui le couvre" if len(tuiles_pays) == 1
                else f"{NOM_PAYS} et les {len(tuiles_pays)} tuiles qui le couvrent")

# ---- Panneau 1 : l'Afrique, toutes frontières, et la grille -----------------
for code, anneaux in CONTOURS_PAYS.items():
    o, e, s, n = EMPRISES_PAYS[code]
    if e < -22 or o > 56 or n < -38 or s > 40:
        continue
    for anneau in anneaux:
        ax1.fill(anneau[:, 0], anneau[:, 1], facecolor=COULEURS["menthe"],
                 edgecolor=COULEURS["sauge"], lw=0.5, zorder=1)
for anneau in CONTOURS_PAYS[PAYS]:
    ax1.fill(anneau[:, 0], anneau[:, 1], facecolor=COULEURS["vert"], alpha=0.85,
             edgecolor=COULEURS["foret"], lw=1.3, zorder=3)

for h in range(int((-25 + 180) // 10), int((58 + 180) // 10) + 1):
    for v in range(int((90 - 42) // 10), int((90 + 40) // 10) + 1):
        tuile = f"h{h:02d}v{v:02d}"
        o, e, s, n = emprise_tuile(tuile)
        if e < -22 or o > 56 or n < -38 or s > 40:
            continue
        couvre = tuile in tuiles_pays
        ax1.add_patch(plt.Rectangle((o, s), 10, 10, facecolor="none",
                                    edgecolor=COULEURS["or_"] if couvre else "#C7D2CD",
                                    lw=2.2 if couvre else 0.7, zorder=5 if couvre else 2))
        if couvre:
            ax1.text(o + 0.4, n - 0.5, tuile, ha="left", va="top", fontsize=8,
                     fontweight="bold", color=COULEURS["ocre"], zorder=6)

ax1.set_xlim(-22, 56)
ax1.set_ylim(-38, 40)
ax1.set_aspect("equal")
ax1.set_xlabel("longitude (°)")
ax1.set_ylabel("latitude (°)")
ax1.set_title("La grille mondiale, restreinte à l'Afrique", pad=8, fontsize=11)
legende_sous_laxe(ax1, [(NOM_PAYS, COULEURS["vert"]),
                        ("tuiles à couvrir", COULEURS["or_"]),
                        ("grille 10° × 10°", "#C7D2CD")], colonnes=3)

# ---- Panneau 2 : le pays et ses tuiles -------------------------------------
o_p, e_p, s_p, n_p = EMPRISE_PAYS
marge = max(1.0, 0.16 * max(e_p - o_p, n_p - s_p))
# Les limites sont fixées AVANT de dessiner : l'étiquette de chaque tuile doit
# tomber dans la fenêtre visible, et non au coin de la tuile, souvent hors cadre.
ax2.set_xlim(o_p - marge, e_p + marge)
ax2.set_ylim(s_p - marge, n_p + marge)
for tuile in sorted(tuiles_pays):
    o, e, s, n = emprise_tuile(tuile)
    presente = tuile in tuiles_presentes
    ax2.add_patch(plt.Rectangle((o, s), 10, 10,
                                facecolor=COULEURS["or_"] if presente else "#F3F6F4",
                                edgecolor=COULEURS["ocre"], lw=1.6,
                                alpha=0.26 if presente else 1.0, zorder=1))
    x_etiquette = max(o, o_p - marge) + 0.03 * (e_p - o_p + 2 * marge)
    y_etiquette = max(s, s_p - marge) + 0.03 * (n_p - s_p + 2 * marge)
    ax2.text(x_etiquette, y_etiquette, tuile, ha="left", va="bottom", fontsize=8.5,
             fontweight="bold", color=COULEURS["ocre"], zorder=6,
             bbox=dict(boxstyle="round,pad=0.25", fc="white", ec=COULEURS["ocre"], lw=0.8))

for code, anneaux in CONTOURS_PAYS.items():
    o, e, s, n = EMPRISES_PAYS[code]
    if e < o_p - marge - 8 or o > e_p + marge + 8 or n < s_p - marge - 8 or s > n_p + marge + 8:
        continue
    for anneau in anneaux:
        ax2.fill(anneau[:, 0], anneau[:, 1],
                 facecolor=COULEURS["vert"] if code == PAYS else COULEURS["menthe"],
                 alpha=0.85 if code == PAYS else 0.6,
                 edgecolor=COULEURS["foret"] if code == PAYS else COULEURS["sauge"],
                 lw=1.4 if code == PAYS else 0.6, zorder=3 if code == PAYS else 2)

ax2.set_aspect("equal")
etiqueter_villes(ax2, villes_du_pays(nombre=7), couleur=COULEURS["encre"], taille=8.6)
ax2.set_xlabel("longitude (°)")
ax2.set_title(TITRE_TUILES, pad=8, fontsize=11)
legende_sous_laxe(ax2, [("tuile présente dans vos données", COULEURS["or_"]),
                        ("tuile manquante", "#F3F6F4")], colonnes=2)

titre_figure(fig, "Une tuile n'est pas un pays",
             "Le globe est découpé en 36 × 18 = 648 carrés de 10° × 10°, chacun de 2400 × 2400 pixels.\n"
             "Les frontières, elles, ignorent ce découpage : il faudra mosaïquer, puis découper.",
             "Grille : NASA Black Marble (VNP46, collection 2). Frontières : Natural Earth 1:50 M, domaine public.")
plt.show()
````

 <div style="color:#00A86A;font-weight:700;font-size:13.5px;">&#128200;&nbsp; Comment lire ces deux cartes</div>

 <div style="color:#231F20;font-size:14px;margin-top:4px;">Un fichier n’est pas une image : c’est un classeur</div>
</div>

**L’image mentale.** Ouvrez un fichier `.h5` en pensant à un **classeur à intercalaires**,
pas à une photo. Trois notions suffisent :

- un **groupe** est un intercalaire — il ne contient pas de données, seulement d’autres groupes ou des jeux
  de données ; c’est l’équivalent d’un dossier ;
- un **jeu de données** (*dataset*) est une feuille de chiffres — un tableau, ici presque toujours
  bidimensionnel, de type entier ;
- un **attribut** est une annotation collée sur un groupe ou sur une feuille : unité, facteur d’échelle,
  valeur de remplissage, description des codes.

Dans un fichier Black Marble, le chemin utile est toujours de la forme
`HDFEOS / GRIDS / <nom de grille> / Data Fields`. Le nom de grille, lui, **change selon le produit** :
`VIIRS_Grid_DNB_2d` pour les composites A3 et A4, `VNP_Grid_DNB` pour le journalier A2. C’est précisément
pour cela que la fonction `trouver_groupe_donnees()` cherche le groupe au lieu de le supposer.

````python
# =============================================================================
#  ÉTAPE 07 — L'ARBORESCENCE DU FICHIER
# =============================================================================
print(f"FICHIER : {FICHIER_EXEMPLE.name}")
print(f"TAILLE  : {FICHIER_EXEMPLE.stat().st_size / 1e6:.2f} Mo")
print("=" * 78)

compteurs = {"groupes": 0, "jeux": 0, "octets": 0}


def afficher_noeud(nom, objet):
    """Affiche un nœud de l'arborescence avec une indentation proportionnelle."""
    profondeur = nom.count("/")
    feuille = nom.split("/")[-1]
    marge = "   " * profondeur
    if isinstance(objet, h5py.Group):
        compteurs["groupes"] += 1
        n_attrs = len(objet.attrs)
        suffixe = f"   [{n_attrs} attribut(s)]" if n_attrs else ""
        print(f"{marge}├─ ▸ {feuille}/{suffixe}")
    else:
        compteurs["jeux"] += 1
        compteurs["octets"] += objet.size * objet.dtype.itemsize
        forme = "×".join(str(d) for d in objet.shape) if objet.shape else "scalaire"
        print(f"{marge}├─ ▪ {feuille:<42} {forme:>12}  {str(objet.dtype):<8} "
              f"{len(objet.attrs):>2} attr.")


with h5py.File(FICHIER_EXEMPLE, "r") as f:
    # 1. Les attributs globaux du fichier
    globaux = tous_les_attributs(f)
    if globaux:
        print("\nATTRIBUTS GLOBAUX DU FICHIER")
        for cle, valeur in list(globaux.items())[:14]:
            texte = str(valeur)
            print(f"   {cle:<26} = {texte[:46]}{'…' if len(texte) > 46 else ''}")

    # 2. L'arborescence complète
    print("\nARBORESCENCE")
    print("   ▸ groupe (intercalaire)   ▪ jeu de données (feuille de chiffres)\n")
    f.visititems(afficher_noeud)

    chemin_groupe = trouver_groupe_donnees(f)

print("\n" + "=" * 78)
print(f"  {compteurs['groupes']} groupe(s)  ·  {compteurs['jeux']} jeu(x) de données  ·  "
      f"{compteurs['octets'] / 1e6:.1f} Mo une fois décompressé en mémoire")
print(f"  Groupe utile détecté automatiquement :\n     {chemin_groupe}")
print("=" * 78)
````

### 08 · Le catalogue des couches

<div style="border-left:6px solid #00A86A;background:#E8F5EF;border-radius:0 12px 12px 0;

 <div style="color:#D49A00;font-size:11px;font-weight:700;letter-spacing:2.5px;text-transform:uppercase;">
  ÉTAPE 08 · 20 &nbsp;&nbsp;<span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span  <div style="color:#231F20;font-size:14px;margin-top:4px;">Une vingtaine de tableaux, cinq rôles</div>
</div>

**Ce qu’il faut comprendre.** Un fichier ne contient pas *une* image de radiance mais
**plusieurs versions** de cette image, plus des couches auxiliaires. Les distinguer est la clé pour choisir
la bonne. Les cinq familles :

<table style="width:100%;border-collapse:collapse;border:none;"><tr><td style="width:33.3333%;vertical-align:top;padding:5px;">
 <div style="background:#F4F7F5;border:1px solid #D5DED9;border-top:3px solid #00A86A;border-radius:10px;
  padding:13px 15px;height:100%;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
  <div style="color:#00704A;font-weight:700;font-size:14px;margin-bottom:5px;">Radiance</div>
  <div style="color:#231F20;font-size:13px;line-height:1.55;">La mesure elle-même, en nW·cm⁻²·sr⁻¹. Plusieurs variantes selon l’angle de visée (<code>AllAngle</code>, <code>NearNadir</code>, <code>OffNadir</code>) et selon la présence de neige (<code>Snow_Free</code>, <code>Snow_Covered</code>). <b>En Afrique, on prend systématiquement <code>AllAngle_Composite_Snow_Free</code>.</b></div>
 </div></td><td style="width:33.3333%;vertical-align:top;padding:5px;">
 <div style="background:#F4F7F5;border:1px solid #D5DED9;border-top:3px solid #00A86A;border-radius:10px;
  padding:13px 15px;height:100%;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
  <div style="color:#00704A;font-weight:700;font-size:14px;margin-bottom:5px;">Incertitude</div>
  <div style="color:#231F20;font-size:13px;line-height:1.55;"><code>…_Num</code> donne le <b>nombre de nuits</b> réellement utilisées pour construire le composite ; <code>…_Std</code> donne l’<b>écart-type</b> entre ces nuits. Un pixel calculé sur 2 nuits avec un écart-type énorme ne vaut pas un pixel calculé sur 25 nuits stables.</div>
 </div></td><td style="width:33.3333%;vertical-align:top;padding:5px;">
 <div style="background:#F4F7F5;border:1px solid #D5DED9;border-top:3px solid #00A86A;border-radius:10px;
  padding:13px 15px;height:100%;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
  <div style="color:#00704A;font-weight:700;font-size:14px;margin-bottom:5px;">Qualité</div>
  <div style="color:#231F20;font-size:13px;line-height:1.55;"><code>…_Quality</code> (composites) ou <code>Mandatory_Quality_Flag</code> (journalier) : un code par pixel qui dit s’il faut lui faire confiance. Détaillé à l’étape 11.</div>
 </div></td></tr><tr><td style="width:33.3333%;vertical-align:top;padding:5px;">
 <div style="background:#F4F7F5;border:1px solid #D5DED9;border-top:3px solid #00A86A;border-radius:10px;
  padding:13px 15px;height:100%;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
  <div style="color:#00704A;font-weight:700;font-size:14px;margin-bottom:5px;">Contexte</div>
  <div style="color:#231F20;font-size:13px;line-height:1.55;"><code>Land_Water_Mask</code> distingue terre, côte, eau intérieure et océan. <code>DNB_Platform</code> indique quel satellite a fourni la mesure — utile pour repérer une rupture de série.</div>
 </div></td><td style="width:33.3333%;vertical-align:top;padding:5px;">
 <div style="background:#F4F7F5;border:1px solid #D5DED9;border-top:3px solid #00A86A;border-radius:10px;
  padding:13px 15px;height:100%;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
  <div style="color:#00704A;font-weight:700;font-size:14px;margin-bottom:5px;">Coordonnées</div>
  <div style="color:#231F20;font-size:13px;line-height:1.55;"><code>lat</code> et <code>lon</code> : les coordonnées explicites des lignes et des colonnes, présentes dans les composites. Elles permettent de <b>vérifier</b> le géoréférencement calculé.</div>
 </div></td></tr></table>

````python
# =============================================================================
#  ÉTAPE 08 — LE CATALOGUE DES COUCHES
# =============================================================================

def classer_couche(nom: str) -> str:
    """Traduit le nom technique d'une couche en un rôle compréhensible."""
    if nom in ("lat", "lon", "latitude", "longitude"):
        return "5 · Coordonnées"
    if nom == "Land_Water_Mask":
        return "4 · Contexte — terre / eau"
    if nom == "DNB_Platform":
        return "4 · Contexte — satellite"
    if nom.endswith("_Quality") or "Quality_Flag" in nom:
        return "3 · Qualité"
    if "Cloud" in nom or nom == "Snow_Flag":
        return "3 · Qualité — nuage / neige"
    if nom.endswith("_Num"):
        return "2 · Incertitude — nb d'observations"
    if nom.endswith("_Std"):
        return "2 · Incertitude — écart-type"
    if "Lunar" in nom:
        return "4 · Contexte — éclairement lunaire"
    if "Retrieval" in nom:
        return "3 · Qualité — fraîcheur"
    if "Composite" in nom or "NTL" in nom:
        return "1 · RADIANCE"
    return "6 · Autre"


lignes = []
with h5py.File(FICHIER_EXEMPLE, "r") as f:
    groupe = f[trouver_groupe_donnees(f)]
    couche_principale = choisir_couche_radiance(groupe)
    for nom, objet in groupe.items():
        if not isinstance(objet, h5py.Dataset):
            continue
        lignes.append({
            "couche": nom,
            "rôle": classer_couche(nom),
            "forme": "×".join(str(d) for d in objet.shape) if objet.shape else "scalaire",
            "type": str(objet.dtype),
            "unité": lire_attribut(objet, "units", ""),
            "échelle": lire_attribut(objet, "scale_factor", ""),
            "décalage": lire_attribut(objet, "add_offset", lire_attribut(objet, "offset", "")),
            "_FillValue": lire_attribut(objet, "_FillValue", ""),
            "Mo": round(objet.size * objet.dtype.itemsize / 1e6, 2),
        })

CATALOGUE = pd.DataFrame(lignes).sort_values(["rôle", "couche"]).reset_index(drop=True)

print(f"{len(CATALOGUE)} couche(s) dans {FICHIER_EXEMPLE.name}")
print(f"Couche de radiance retenue automatiquement : {couche_principale}\n")
display(CATALOGUE)

print("\nPOIDS MÉMOIRE PAR RÔLE")
display(CATALOGUE.groupby("rôle").agg(couches=("couche", "size"), Mo=("Mo", "sum")))
````

<div style="background:#FDF4E0;border:1px solid #D49A0044;border-left:4px solid #D49A00;border-radius:8px;
 padding:13px 17px;margin:10px 0;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
 <div style="color:#D49A00;font-weight:700;font-size:13.5px;">&#9888;&nbsp; Le piège n°1 : croire que <code>_Num</code> et <code>_Std</code> sont accessoires</div>
 <div style="color:#231F20;font-size:13.5px;margin-top:5px;line-height:1.6;">Ils ne le sont pas. Dans une zone souvent nuageuse, un composite mensuel peut reposer sur <b>deux ou trois nuits seulement</b>. La valeur de radiance paraît normale, mais son incertitude est énorme. Prenez l’habitude de cartographier <code>…_Num</code> à côté de la radiance : c’est le meilleur moyen de repérer les régions où votre indicateur ne repose sur presque rien. C’est aussi ce qui vous permettra, le moment venu, de justifier une limitation dans votre note méthodologique.</div>
</div>

style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#F5C242;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span></div>
 <div style="color:#231F20;font-size:14px;margin-top:4px;">Trois attributs qui séparent le vrai du faux</div>
</div>

**Le problème.** Pour économiser de la place, la NASA ne stocke pas des nombres décimaux mais
des **entiers**. La radiance réelle s’obtient par une conversion, et les paramètres de cette conversion
sont écrits **dans le fichier lui-même**, sous forme d’attributs :

$$\text{radiance} = \text{valeur stockée} \times \texttt{scale\_factor} + \texttt{add\_offset}$$

à ceci près qu’une valeur particulière — le `_FillValue` — ne signifie pas « zéro » mais
**« aucune donnée ici »**. Elle doit être convertie en `NaN` **avant** toute opération.

Concrètement, pour `AllAngle_Composite_Snow_Free` : `scale_factor = 0,1`, `add_offset = 0`,
`_FillValue = 65535`. Une valeur stockée de `1247` vaut donc **124,7 nW·cm⁻²·sr⁻¹**, et une valeur
stockée de `65535` ne vaut **rien du tout**.

<div style="background:#FBEDEA;border:1px solid #B83B2E44;border-left:4px solid #B83B2E;border-radius:8px;
 padding:13px 17px;margin:10px 0;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
 <div style="color:#B83B2E;font-weight:700;font-size:13.5px;">&#9940;&nbsp; L’erreur qui invalide silencieusement une analyse entière</div>
 <div style="color:#231F20;font-size:13.5px;margin-top:5px;line-height:1.6;">Traiter le <code>_FillValue</code> comme un nombre. Avec <code>_FillValue = 65535</code> et <code>scale_factor = 0,1</code>, chaque pixel vide devient <b>6 553,5 nW·cm⁻²·sr⁻¹</b> — une valeur supérieure à celle du centre de Lagos. Une somme des lumières calculée ainsi peut être fausse d’un facteur dix, sans qu’aucun message d’erreur n’apparaisse. La carte produite reste jolie. <b>C’est la première chose à vérifier</b> quand un chiffre semble trop beau.</div>
</div>

````python
# =============================================================================
#  ÉTAPE 09 — DÉCODER LES NOMBRES : scale_factor, add_offset, _FillValue
# =============================================================================
with h5py.File(FICHIER_EXEMPLE, "r") as f:
    groupe = f[trouver_groupe_donnees(f)]
    jeu = groupe[couche_principale]

    attributs = tous_les_attributs(jeu)
    echelle = lire_attribut(jeu, "scale_factor", 1.0)
    decalage = lire_attribut(jeu, "add_offset", lire_attribut(jeu, "offset", 0.0))
    remplissage = lire_attribut(jeu, "_FillValue", None)

    # Un petit échantillon carré, pris là où il y a de la lumière
    cote = jeu.shape[0]
    apercu = jeu[::max(cote // 300, 1), ::max(cote // 300, 1)].astype("float64")
    masque_valide = apercu != remplissage if remplissage is not None else np.ones_like(apercu, bool)
    if masque_valide.any():
        i_max, j_max = np.unravel_index(np.argmax(np.where(masque_valide, apercu, -1)),
                                        apercu.shape)
        pas = max(cote // 300, 1)
        li = int(np.clip(i_max * pas - 2, 0, cote - 5))
        co = int(np.clip(j_max * pas - 2, 0, cote - 5))
    else:
        li = co = 0
    bloc_brut = jeu[li:li + 5, co:co + 5].astype("float64")

print(f"COUCHE : {couche_principale}\n")
print("ATTRIBUTS LUS DANS LE FICHIER (jamais codés en dur)")
for cle, valeur in attributs.items():
    texte = str(valeur)
    print(f"   {cle:<16} = {texte[:62]}{'…' if len(texte) > 62 else ''}")

print(f"\nL'ÉQUATION DE DÉCODAGE")
print(f"   radiance = valeur_stockée × {echelle} + {decalage}")
print(f"   puis     : toute valeur égale à {remplissage} devient NaN\n")

# --- Application sur un bloc 5 × 5 -------------------------------------------
bloc_decode = np.where(bloc_brut == remplissage, np.nan, bloc_brut * echelle + decalage)

print(f"BLOC 5 × 5 EXTRAIT À LA POSITION (ligne {li}, colonne {co})\n")
print("   valeurs stockées (entiers bruts)")
print(pd.DataFrame(bloc_brut.astype("int64")).to_string(
    header=[f"c{co + k}" for k in range(5)], index=[f"l{li + k}" for k in range(5)]))
print(f"\n   radiances décodées (nW·cm⁻²·sr⁻¹)")
print(pd.DataFrame(np.round(bloc_decode, 2)).to_string(
    header=[f"c{co + k}" for k in range(5)], index=[f"l{li + k}" for k in range(5)]))

# --- Démonstration du danger --------------------------------------------------
tableau_complet = None
with h5py.File(FICHIER_EXEMPLE, "r") as f:
    tableau_complet = f[trouver_groupe_donnees(f)][couche_principale][:].astype("float64")

faux = (tableau_complet * echelle + decalage)                    # sans traiter le _FillValue
juste = np.where(tableau_complet == remplissage, np.nan, faux)   # correctement traité
n_vides = int(np.sum(tableau_complet == remplissage))

print("\n" + "=" * 78)
print("  CE QUE COÛTE L'OUBLI DU _FillValue")
print("=" * 78)
print(f"  Pixels de remplissage  : {n_vides:,} sur {tableau_complet.size:,} "
      f"({100 * n_vides / tableau_complet.size:.1f} %)".replace(",", " "))
print(f"  Somme des lumières — traitement correct : {np.nansum(juste):>18,.0f}".replace(",", " "))
print(f"  Somme des lumières — _FillValue ignoré  : {np.nansum(faux):>18,.0f}".replace(",", " "))
if np.nansum(juste) > 0:
    print(f"  Facteur d'erreur                       : "
          f"× {np.nansum(faux) / np.nansum(juste):.1f}")
print("=" * 78)

# --- Le fil rouge commence ici ------------------------------------------------
noter_fil_rouge(9, "aucune précaution", np.nansum(faux), int(np.isfinite(faux).sum()),
                f"tuile {infos['tuile']}", "le _FillValue compté comme une mesure")
noter_fil_rouge(9, "_FillValue mis à NaN", np.nansum(juste), int(np.isfinite(juste).sum()),
                f"tuile {infos['tuile']}", "première valeur défendable")
````

````python
# --- Figure : de l'entier stocké à la radiance --------------------------------
fig, axes = plt.subplots(1, 3, figsize=(12.4, 4.3))
fig.subplots_adjust(top=0.70, bottom=0.13, left=0.04, right=0.98, wspace=0.22)

for ax, donnees, titre, cmap, fmt in [
        (axes[0], bloc_brut, "1 · Ce qui est stocké\n(entiers)", CMAP_VERT, "{:.0f}"),
        (axes[1], bloc_decode, "2 · Après décodage\n(× échelle + décalage)", CMAP_VERT, "{:.1f}"),
        (axes[2], np.isnan(bloc_decode).astype(float),
         "3 · Où sont les trous\n(_FillValue → NaN)",
         ListedColormap([COULEURS["menthe"], COULEURS["brique"]]), None)]:
    fini = donnees[np.isfinite(donnees)]
    vmin, vmax = (float(fini.min()), float(fini.max())) if fini.size else (0.0, 1.0)
    ax.imshow(np.nan_to_num(donnees, nan=vmin), cmap=cmap, vmin=vmin, vmax=max(vmax, vmin + 1e-9))
    for i in range(donnees.shape[0]):
        for j in range(donnees.shape[1]):
            v = donnees[i, j]
            if fmt is None:
                texte = "vide" if np.isnan(bloc_decode[i, j]) else "ok"
                couleur = "white" if np.isnan(bloc_decode[i, j]) else COULEURS["vert_fonce"]
            elif np.isnan(v):
                texte, couleur = "NaN", COULEURS["brique"]
            else:
                texte = fmt.format(v)
                couleur = "white" if (v - vmin) > 0.6 * (vmax - vmin + 1e-9) else COULEURS["encre"]
            ax.text(j, i, texte, ha="center", va="center", fontsize=8.2,
                    fontweight="bold", color=couleur)
    ax.set_title(titre, fontsize=10.5, pad=7)
    ax.set_xticks(range(5)); ax.set_xticklabels([f"c{co + k}" for k in range(5)], fontsize=7.5)
    ax.set_yticks(range(5)); ax.set_yticklabels([f"l{li + k}" for k in range(5)], fontsize=7.5)
    ax.tick_params(length=0)

titre_figure(fig, "La conversion, pixel par pixel",
             f"couche « {couche_principale} »   ·   "
             f"radiance = stocké × {echelle} + {decalage}   ·   _FillValue = {remplissage} → NaN",
             f"Source : {FICHIER_EXEMPLE.name}" + ("   ·   MODE DÉMONSTRATION" if MODE_DEMO else ""))
plt.show()
````

### 10 · Où se trouve chaque pixel ?

<div style="border-left:6px solid #00A86A;background:#E8F5EF;border-radius:0 12px 12px 0;
 padding:14px 22px;margin-top:22px;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
 <div style="color:#D49A00;font-size:11px;font-weight:700;letter-spacing:2.5px;text-transform:uppercase;">
  ÉTAPE 10 · 20 &nbsp;&nbsp;<span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span></div>
 <div style="color:#231F20;font-size:14px;margin-top:4px;">Passer de (ligne, colonne) à (longitude, latitude) sans bibliothèque SIG</div>
</div>

**Ce qu’il faut savoir.** Les produits Black Marble sont sur une **grille géographique
régulière** (EPSG:4326) : pas de projection compliquée, pas de rééchantillonnage. Le pixel (0, 0) est
le **coin nord-ouest** de la tuile, et la taille d’un pixel est constante en degrés :

```
résolution = 10° / 2400 = 0,00416667°  (soit 15 secondes d'arc)

longitude(colonne) = ouest + (colonne + 0,5) × résolution
latitude(ligne)    = nord  - (ligne    + 0,5) × résolution
```

Le `+ 0,5` place la coordonnée au **centre** du pixel, et non à son coin — une convention discrète mais
qui, oubliée, décale toute votre grille d’un demi-pixel, soit environ 230 mètres.

Les composites A3 et A4 embarquent en outre des jeux `lat` et `lon` explicites. La cellule ci-dessous
calcule les coordonnées, puis **les confronte** aux valeurs stockées : c’est une vérification que tout
notebook sérieux devrait faire une fois.

````python
# =============================================================================
#  ÉTAPE 10 — GÉORÉFÉRENCEMENT : DU PIXEL AUX COORDONNÉES
# =============================================================================
tuile_exemple = infos["tuile"]
ouest, est, sud, nord = emprise_tuile(tuile_exemple)
n_lignes, n_colonnes = tableau_complet.shape
resolution = TUILE_DEGRES / n_lignes


def coordonnees_pixel(ligne, colonne):
    """Centre du pixel (ligne, colonne) en degrés décimaux."""
    return (ouest + (colonne + 0.5) * resolution,
            nord - (ligne + 0.5) * resolution)


def pixel_depuis_coordonnees(longitude, latitude):
    """Opération inverse : quelle cellule contient ce point ?"""
    colonne = int((longitude - ouest) / resolution)
    ligne = int((nord - latitude) / resolution)
    if not (0 <= ligne < n_lignes and 0 <= colonne < n_colonnes):
        return None
    return ligne, colonne


# Axes complets (utiles pour l'affichage et l'export)
LONGITUDES = ouest + (np.arange(n_colonnes) + 0.5) * resolution
LATITUDES = nord - (np.arange(n_lignes) + 0.5) * resolution

# Emprise au format attendu par imshow : (gauche, droite, bas, haut)
EMPRISE_AFFICHAGE = (ouest, est, sud, nord)

# Transformation affine, au format GDAL / rasterio
TRANSFORMATION_AFFINE = (resolution, 0.0, ouest, 0.0, -resolution, nord)

print(f"TUILE {tuile_exemple}   ·   {n_lignes} × {n_colonnes} pixels\n")
print(f"  Coin nord-ouest   : longitude {ouest:+10.5f}°   latitude {nord:+9.5f}°")
print(f"  Coin sud-est      : longitude {est:+10.5f}°   latitude {sud:+9.5f}°")
print(f"  Résolution        : {resolution:.8f}°  =  {resolution * 3600:.2f}\" d'arc")
print(f"  Transformation    : (a={resolution:.8f}, b=0, c={ouest:.5f}, "
      f"d=0, e={-resolution:.8f}, f={nord:.5f})")

print("\nQUELQUES PIXELS DE CONTRÔLE")
controles = [(0, 0, "coin nord-ouest"),
             (0, n_colonnes - 1, "coin nord-est"),
             (n_lignes - 1, 0, "coin sud-ouest"),
             (n_lignes // 2, n_colonnes // 2, "centre de la tuile")]
table_controle = pd.DataFrame(
    [{"ligne": li_, "colonne": co_, "repère": nom_,
      "longitude": round(coordonnees_pixel(li_, co_)[0], 5),
      "latitude": round(coordonnees_pixel(li_, co_)[1], 5)}
     for li_, co_, nom_ in controles])
display(table_controle.set_index("repère"))

# --- Vérification contre les jeux lat / lon du fichier ------------------------
with h5py.File(FICHIER_EXEMPLE, "r") as f:
    groupe = f[trouver_groupe_donnees(f)]
    a_coordonnees = "lat" in groupe and "lon" in groupe
    if a_coordonnees:
        lat_fichier = groupe["lat"][:]
        lon_fichier = groupe["lon"][:]

if a_coordonnees:
    ecart_lat = float(np.max(np.abs(lat_fichier - LATITUDES)))
    ecart_lon = float(np.max(np.abs(lon_fichier - LONGITUDES)))
    print(f"\nVÉRIFICATION contre les jeux « lat » et « lon » du fichier")
    print(f"  écart maximal en latitude  : {ecart_lat:.3e}°  "
          f"({ecart_lat * 111320:.4f} m)")
    print(f"  écart maximal en longitude : {ecart_lon:.3e}°  "
          f"({ecart_lon * 111320:.4f} m)")
    verdict = "les deux méthodes concordent" if max(ecart_lat, ecart_lon) < 1e-6 \
              else "ÉCART SIGNIFICATIF — vérifiez la convention de centrage du pixel"
    print(f"  → {verdict}")
else:
    print("\n  Ce produit n'embarque pas de jeux « lat » / « lon » : "
          "le calcul ci-dessus est la seule source de géoréférencement.")

# --- Localiser une ville dans la grille ---------------------------------------
villes_tuile = villes_dans(EMPRISE_AFFICHAGE, nombre=10)
if villes_tuile:
    print("\nLOCALISATION DE QUELQUES VILLES DANS LA GRILLE")
    lignes_villes = []
    for nom_ville, vlon, vlat, _population in villes_tuile:
        position = pixel_depuis_coordonnees(vlon, vlat)
        if position is None:
            continue
        li_v, co_v = position
        valeur = tableau_complet[li_v, co_v]
        radiance_ville = (np.nan if valeur == remplissage
                          else valeur * echelle + decalage)
        lignes_villes.append({"ville": nom_ville, "longitude": vlon, "latitude": vlat,
                              "ligne": li_v, "colonne": co_v,
                              "radiance": round(radiance_ville, 1)})
    display(pd.DataFrame(lignes_villes).set_index("ville"))
````

````python
# --- Figure : le même fichier dans deux repères ------------------------------
radiance_complete = np.where(tableau_complet == remplissage, np.nan,
                             tableau_complet * echelle + decalage)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12.8, 5.9))
fig.subplots_adjust(top=0.775, bottom=0.125, left=0.065, right=0.945, wspace=0.26)

# ---- Panneau 1 : le repère du tableau (ligne, colonne) ---------------------
plafond = float(np.nanpercentile(radiance_complete[np.isfinite(radiance_complete)
                                                  & (radiance_complete > 0)], 99.7))
visible = np.where(np.isfinite(radiance_complete),
                   np.clip(radiance_complete, PLANCHER_RADIANCE, None), np.nan)
ax1.imshow(visible, cmap=CMAP_NUIT,
           norm=LogNorm(vmin=PLANCHER_RADIANCE, vmax=max(plafond, PLANCHER_RADIANCE * 20)),
           origin="upper", aspect="auto", interpolation="nearest")
ax1.set_facecolor(GRIS_SANS_DONNEE)
ax1.set_xlabel("colonne (indice du tableau)")
ax1.set_ylabel("ligne (indice du tableau)")
ax1.set_title("Repère informatique : (ligne, colonne)", pad=8, fontsize=11)
for x, y, texte, va, ha in [(0, 0, " (0, 0)  ligne 0 = le NORD", "top", "left"),
                            (0, n_lignes - 1, f" ({n_lignes - 1}, 0)  dernière ligne = le SUD",
                             "bottom", "left")]:
    ax1.annotate(texte, xy=(x, y), xytext=(x + 0.04 * n_colonnes,
                                           y + (0.06 if va == "top" else -0.06) * n_lignes),
                 fontsize=8.5, color="white", fontweight="bold", va=va, ha=ha,
                 bbox=dict(boxstyle="round,pad=0.32", fc=COULEURS["brique"], ec="none"),
                 arrowprops=dict(arrowstyle="->", color=COULEURS["brique"], lw=1.3))

# ---- Panneau 2 : le repère géographique ------------------------------------
im2 = carte_nocturne(ax2, radiance_complete, EMPRISE_AFFICHAGE)
ax2.set_xlim(EMPRISE_AFFICHAGE[0], EMPRISE_AFFICHAGE[1])
ax2.set_ylim(EMPRISE_AFFICHAGE[2], EMPRISE_AFFICHAGE[3])
tracer_contour_pays(ax2, couleur=COULEURS["or_"], lw=1.6)
etiqueter_villes(ax2, villes_dans(EMPRISE_AFFICHAGE, nombre=7), taille=8.4)
barre_echelle(ax2)
ax2.set_xlabel("longitude (°)")
ax2.set_ylabel("latitude (°)")
ax2.set_title("Repère géographique : (longitude, latitude)", pad=8, fontsize=11)
barre_de_couleur(fig, im2, ax2)

titre_figure(fig, "La même image, deux systèmes de repérage",
             "La ligne 0 du tableau est la plus au NORD : la latitude décroît quand l'indice de ligne augmente.\n"
             "C'est de l'oubli de ce signe moins que viennent presque toutes les cartes retournées.",
             f"Source : {FICHIER_EXEMPLE.name}   ·   couche {couche_principale}"
             + ("   ·   MODE DÉMONSTRATION — valeurs simulées" if MODE_DEMO else ""))
plt.show()
````

### 11 · Les drapeaux qualité

<div style="border-left:6px solid #00A86A;background:#E8F5EF;border-radius:0 12px 12px 0;
 padding:14px 22px;margin-top:22px;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
 <div style="color:#D49A00;font-size:11px;font-weight:700;letter-spacing:2.5px;text-transform:uppercase;">
  ÉTAPE 11 · 20 &nbsp;&nbsp;<span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span  <div style="color:#231F20;font-size:14px;margin-top:4px;">Savoir à quels pixels faire confiance</div>
</div>

**Le principe.** La NASA attache à chaque pixel un petit code entier qui résume la confiance
qu’on peut lui accorder. Sa signification **dépend du produit**, et la bonne pratique consiste à lire le
codage dans l’attribut `description` du fichier plutôt que de le supposer.

Pour les composites (A3, A4) — couche `…_Quality` :

| code | signification | faut-il le garder ? |
|:---:|---|---|
| **0** | bonne qualité | oui, toujours |
| **1** | qualité médiocre | selon l’usage |
| **2** | valeur comblée (*gap-filled*) | à éviter pour un indicateur publié |
| **255** | valeur de remplissage | jamais |

Pour le journalier (A2) — couche `Mandatory_Quality_Flag` :

| code | signification | commentaire |
|:---:|---|---|
| **0** | haute qualité, lumières **persistantes** | le cœur du signal : villes, infrastructures |
| **1** | haute qualité, lumières **éphémères** | feux, torchères, pêche, activité saisonnière |
| **2** | qualité médiocre, valeur aberrante corrigée | à manier avec précaution |
| **255** | aucune restitution | nuage épais, lune trop forte, données absentes |

````python
# =============================================================================
#  ÉTAPE 11 — LES DRAPEAUX QUALITÉ
# =============================================================================

# Codages de secours, si le fichier ne porte pas d'attribut « description »
CODAGES_DE_SECOURS = {
    "composite": {0: "Bonne qualité", 1: "Qualité médiocre",
                  2: "Valeur comblée (gap-filled)", 255: "Valeur de remplissage"},
    "journalier": {0: "Haute qualité — lumières persistantes",
                   1: "Haute qualité — lumières éphémères",
                   2: "Qualité médiocre — aberration corrigée",
                   255: "Aucune restitution"},
    "terre_eau": {0: "Océan peu profond", 1: "Terre", 2: "Trait de côte",
                  3: "Eau intérieure peu profonde", 4: "Eau temporaire",
                  5: "Eau intérieure profonde", 6: "Continental", 7: "Océan profond"},
}


def analyser_description(texte: str) -> dict:
    """Extrait un dictionnaire {code: libellé} d'un attribut « description ».

    Les descriptions NASA sont de la forme :
        "0 = Good quality 1 = Poor quality 2 = Gap filled 255 = Fill value"
    Le regard en avant (?=...) permet de couper juste avant le code suivant.
    """
    if not texte:
        return {}
    motif = r"(\d+)\s*=\s*([^\n\t0-9][^\n\t]*?)(?=\s+\d+\s*=|$)"
    return {int(code): libelle.strip() for code, libelle in re.findall(motif, str(texte))}


def trouver_couches_drapeaux(chemin) -> list:
    """Liste les couches de drapeaux présentes, dans un ordre de lecture utile."""
    trouvees = []
    with h5py.File(chemin, "r") as f:
        groupe = f[trouver_groupe_donnees(f)]
        for nom, objet in groupe.items():
            if not (isinstance(objet, h5py.Dataset) and objet.ndim == 2):
                continue
            if (nom.endswith("_Quality") or "Quality_Flag" in nom
                    or nom in ("Land_Water_Mask", "QF_Cloud_Mask", "Snow_Flag", "DNB_Platform")):
                trouvees.append(nom)
    priorite = {"Quality": 0, "Land_Water_Mask": 1, "QF_Cloud_Mask": 2}
    return sorted(trouvees, key=lambda n: (0 if "Quality" in n else priorite.get(n, 3), n))


couches_drapeaux = trouver_couches_drapeaux(FICHIER_EXEMPLE)
print(f"Couches de drapeaux détectées : {', '.join(couches_drapeaux)}\n")

# --- Analyse de la couche qualité principale ---------------------------------
nom_qualite = next((n for n in couches_drapeaux if "Quality" in n), None)

if nom_qualite is None:
    print("Aucune couche de qualité dans ce fichier.")
    drapeau_qualite, legende_qualite = None, {}
else:
    with h5py.File(FICHIER_EXEMPLE, "r") as f:
        objet = f[trouver_groupe_donnees(f)][nom_qualite]
        drapeau_qualite = objet[:]
        description = lire_attribut(objet, "description", "") or lire_attribut(objet, "long_name", "")

    legende_qualite = analyser_description(description)
    if not legende_qualite:
        famille = "journalier" if infos["produit"] == "VNP46A2" else "composite"
        legende_qualite = CODAGES_DE_SECOURS[famille]
        print("  (attribut « description » absent — codage de secours appliqué)")

    print(f"COUCHE : {nom_qualite}")
    print(f"description : {str(description)[:110]}\n")

    codes, effectifs = np.unique(drapeau_qualite, return_counts=True)
    repartition = pd.DataFrame({
        "code": codes,
        "signification": [legende_qualite.get(int(k), "code non documenté") for k in codes],
        "pixels": effectifs,
        "part_%": np.round(100 * effectifs / drapeau_qualite.size, 2),
        "retenu (QA_MAX=%d)" % QA_MAX: ["oui" if int(k) <= QA_MAX else "non" for k in codes],
    })
    display(repartition.set_index("code"))

    conserves = int(np.sum(drapeau_qualite <= QA_MAX))
    print(f"\nAvec QA_MAX = {QA_MAX} : {conserves:,} pixels conservés sur "
          f"{drapeau_qualite.size:,} ({100 * conserves / drapeau_qualite.size:.1f} %)"
          .replace(",", " "))

# --- Le fil rouge avance ------------------------------------------------------
if drapeau_qualite is not None:
    radiance_qualite = np.where(drapeau_qualite <= QA_MAX, radiance_complete, np.nan)
    noter_fil_rouge(11, f"+ filtre qualité ≤ {QA_MAX}", np.nansum(radiance_qualite),
                    int(np.isfinite(radiance_qualite).sum()), f"tuile {infos['tuile']}",
                    "les pixels douteux sortent du calcul")
````

````python
# --- Figure : les drapeaux qualité, et le prix du seuil ----------------------
if drapeau_qualite is not None:
    palette_q = {0: COULEURS["vert"], 1: COULEURS["or_"], 2: COULEURS["terre"],
                 3: COULEURS["sarcelle"], 255: COULEURS["sauge"]}

    fig, axes = plt.subplots(1, 3, figsize=(13.4, 5.5),
                             gridspec_kw={"width_ratios": [1, 1, 0.78]})
    fig.subplots_adjust(top=0.765, bottom=0.205, left=0.055, right=0.985, wspace=0.24)

    # 1. la carte des classes
    entrees = carte_categorielle(axes[0], drapeau_qualite, EMPRISE_AFFICHAGE,
                                 legende_qualite, palette_q,
                                 "Classe de qualité, pixel par pixel")
    tracer_contour_pays(axes[0], couleur=COULEURS["encre"], lw=1.2)
    barre_echelle(axes[0], couleur=COULEURS["encre"])
    axes[0].set_xlabel("longitude (°)")
    axes[0].set_ylabel("latitude (°)")
    legende_sous_laxe(axes[0], entrees, colonnes=2)

    # 2. ce que le seuil conserve
    conserve = (drapeau_qualite <= QA_MAX).astype("int16")
    entrees2 = carte_categorielle(
        axes[1], conserve, EMPRISE_AFFICHAGE,
        {0: "écarté par le seuil", 1: "conservé"},
        {0: COULEURS["brique"], 1: COULEURS["vert"]},
        f"Ce que conserve QA_MAX = {QA_MAX}")
    tracer_contour_pays(axes[1], couleur=COULEURS["encre"], lw=1.2)
    axes[1].set_xlabel("longitude (°)")
    legende_sous_laxe(axes[1], entrees2, colonnes=2)

    # 3. la courbe de l'arbitrage
    seuils = [0, 1, 2, 3]
    parts = [100 * np.mean(drapeau_qualite <= s) for s in seuils]
    barres = axes[2].bar([str(s) for s in seuils], parts, width=0.6, edgecolor="none",
                         color=[COULEURS["or_"] if s == QA_MAX else COULEURS["vert"]
                                for s in seuils])
    for barre, part in zip(barres, parts):
        axes[2].text(barre.get_x() + barre.get_width() / 2, part + 2.0, f"{part:.1f} %",
                     ha="center", fontsize=9, fontweight="bold", color=COULEURS["foret"])
    axes[2].set_ylim(0, 112)
    axes[2].set_xlabel("valeur de QA_MAX")
    axes[2].set_ylabel("pixels conservés (%)")
    axes[2].set_title("Le seuil est un arbitrage", pad=8, fontsize=10.5)
    axes[2].spines["left"].set_visible(False)
    axes[2].tick_params(left=False)
    axes[2].set_xticks(range(len(seuils)))
    axes[2].set_xticklabels([f"{s}\n◄ votre réglage" if s == QA_MAX else str(s)
                             for s in seuils], fontsize=9)
    for etiquette in axes[2].get_xticklabels():
        if "◄" in etiquette.get_text():
            etiquette.set_color(COULEURS["ocre"])
            etiquette.set_fontweight("bold")

    titre_figure(fig, "Les drapeaux qualité : à quels pixels faire confiance ?",
                 f"Couche « {nom_qualite} ». Plus le seuil est bas, plus l'analyse est solide — et moins la couverture est complète.\n"
                 "Cherchez des motifs sur la carte de gauche : des bandes trahissent les orbites, des taches trahissent les nuages.",
                 f"Source : {FICHIER_EXEMPLE.name}"
                 + ("   ·   MODE DÉMONSTRATION — distribution simulée" if MODE_DEMO else ""))
    plt.show()
````

<div style="background:#F4F7F5;border:1px solid #00A86A44;border-left:4px solid #00A86A;border-radius:8px;
 padding:13px 17px;margin:10px 0;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
 <div style="color:#00A86A;font-weight:700;font-size:13.5px;">&#128200;&nbsp; Comment lire ces trois panneaux</div>
 <div style="color:#231F20;font-size:13.5px;margin-top:5px;line-height:1.6;">À gauche, la répartition spatiale des classes de qualité : cherchez les <b>motifs</b>. Des bandes régulières trahissent la géométrie des orbites ; des taches concentrées trahissent des nuages persistants ; un bord entier dégradé trahit un problème de mosaïquage. Au centre, la carte binaire de ce que votre seuil conserve. À droite, la courbe de l’arbitrage : chaque cran de <code>QA_MAX</code> vous rend des pixels, au prix de leur fiabilité.</div>
</div>

<div style="background:#E8F5EF;border:1px solid #00704A44;border-left:4px solid #00704A;border-radius:8px;

 <div style="color:#00704A;font-weight:700;font-size:13.5px;">&#128273;&nbsp; Le réflexe à adopter, et à documenter</div>
 <div style="color:#231F20;font-size:13.5px;margin-top:5px;line-height:1.6;">Le choix de <code>QA_MAX</code> n’est pas technique, il est <b>méthodologique</b>. Il doit figurer dans vos métadonnées au même titre qu’un taux de non-réponse : « radiance moyenne, pixels de drapeau qualité ≤ 1, eau masquée, composite annuel VNP46A4 collection 2 ». Un lecteur doit pouvoir reproduire votre chiffre exactement. C’est aussi ce qui vous protégera le jour où quelqu’un obtiendra un résultat différent avec les mêmes fichiers.</div>
</div>

<div style="background:#F2F7FA;border:1px solid #0E7C8644;border-left:4px solid #0E7C86;border-radius:8px;padding:14px 18px;margin:14px 0;font-family:'Segoe UI',Calibri,sans-serif;"><div style="color:#0E7C86;font-weight:700;font-size:13.5px;">&#9999;&nbsp; À vous de jouer — choisir un seuil, et le défendre</div><div style="color:#231F20;font-size:13.5px;margin-top:6px;line-height:1.6;">Remontez à l’étape 02, passez <code>QA_MAX</code> de <code>1</code> à <code>0</code>, puis réexécutez les étapes 02 à 12.<br><b>(a)</b> Combien de pixels perdez-vous ? &nbsp; <b>(b)</b> La somme des lumières baisse-t-elle proportionnellement ? &nbsp; <b>(c)</b> Pour une série 2014-2024 destinée à mesurer une <i>tendance</i>, lequel des deux seuils défendriez-vous devant un comité de validation ?</div></div>

<details><summary><i>Afficher la correction</i></summary>

**(a)** Le troisième panneau de la figure de l’étape 11 donne la réponse directement : c’est l’écart entre la barre `0` et la barre `1`.

**(b)** Non, et c’est tout l’intérêt de la question. Les pixels de drapeau `1` sont des *lumières éphémères* — feux, torchères — souvent très brillantes. On perd donc **plus de lumière que de pixels** : la somme chute davantage que la couverture.

**(c)** `QA_MAX = 0` pour une tendance. Les lumières éphémères varient d’une année sur l’autre pour des raisons climatiques ou saisonnières, pas économiques : les inclure fabrique une volatilité qui n’a rien à voir avec le phénomène mesuré. L’essentiel reste de **documenter le choix**, et de montrer les deux séries quand elles divergent.

</details>

 <span style="color:#D49A00;font-size:10.5px;font-weight:700;letter-spacing:2px;text-transform:uppercase;">À retenir</span>

### 12 · Les masques : terre, eau et nuages

<div style="border-left:6px solid #00A86A;background:#E8F5EF;border-radius:0 12px 12px 0;
 padding:14px 22px;margin-top:22px;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
 <div style="color:#D49A00;font-size:11px;font-weight:700;letter-spacing:2.5px;text-transform:uppercase;">

**Deux masques, deux logiques très différentes.**

`Land_Water_Mask` est simple : un code de 0 à 7 par pixel, chaque code désignant un type de surface.
C’est ce qui permet d’exclure l’océan — indispensable, car la mer n’est pas noire : reflets lunaires,
navires et pêcheries y produisent un signal qui gonflerait artificiellement toute somme des lumières côtière.

`QF_Cloud_Mask` (produit journalier A2) est différent : **un seul entier de 16 bits y encode une dizaine
d’informations**, chacune occupant un ou plusieurs bits. Lire ce nombre comme un entier ordinaire n’a
aucun sens ; il faut en **extraire des tranches de bits**. La technique, appelée *masquage binaire*,
tient en une ligne :

```python
valeur_du_champ = (entier >> position_du_premier_bit) & ((1 << nombre_de_bits) - 1)
```

On décale l’entier vers la droite pour amener le champ voulu tout à droite, puis on efface tout ce qui
dépasse. Les principaux champs du `QF_Cloud_Mask` VIIRS :

  <div style="color:#00704A;font-weight:700;font-size:14px;margin-bottom:5px;">Bit 0 — Jour / nuit</div>
  <div style="color:#231F20;font-size:13px;line-height:1.55;">0 = nuit, 1 = jour. Un pixel acquis de jour n’a aucun sens pour les lumières nocturnes.</div>

  padding:13px 15px;height:100%;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
  <div style="color:#00704A;font-weight:700;font-size:14px;margin-bottom:5px;">Bits 1-3 — Fond terre/eau</div>
  <div style="color:#231F20;font-size:13px;line-height:1.55;">Une seconde description de la surface, plus fine que <code>Land_Water_Mask</code>.</div>
 </div></td><td style="width:33.3333%;vertical-align:top;padding:5px;">
 <div style="background:#F4F7F5;border:1px solid #D5DED9;border-top:3px solid #0E7C86;border-radius:10px;

  <div style="color:#231F20;font-size:13px;line-height:1.55;">De 00 (médiocre) à 11 (élevée). Si elle est médiocre, la détection de nuage qui suit vaut peu.</div>

  <div style="color:#231F20;font-size:13px;line-height:1.55;">00 = clair certain, 01 = probablement clair, 10 = probablement nuageux, 11 = nuageux certain.</div>
 </div></td><td style="width:33.3333%;vertical-align:top;padding:5px;">

  <div style="color:#00704A;font-weight:700;font-size:14px;margin-bottom:5px;">Bit 8 — Ombre</div>
  <div style="color:#231F20;font-size:13px;line-height:1.55;">Ombre de nuage détectée sur le pixel.</div>

  <div style="color:#231F20;font-size:13px;line-height:1.55;">Nuage fin de haute altitude — souvent invisible à l’œil sur l’image.</div>

````python
# =============================================================================
#  ÉTAPE 12 — LES MASQUES : TERRE / EAU, PUIS NUAGES BIT À BIT
# =============================================================================

# --- 1. Le masque terre / eau ------------------------------------------------
with h5py.File(FICHIER_EXEMPLE, "r") as f:
    groupe = f[trouver_groupe_donnees(f)]
    a_terre_eau = "Land_Water_Mask" in groupe
    if a_terre_eau:
        objet = groupe["Land_Water_Mask"]
        masque_terre_eau = objet[:]
        desc_terre_eau = lire_attribut(objet, "description", "")

if a_terre_eau:
    legende_terre_eau = analyser_description(desc_terre_eau) or CODAGES_DE_SECOURS["terre_eau"]
    codes, effectifs = np.unique(masque_terre_eau, return_counts=True)
    print("MASQUE TERRE / EAU  —  Land_Water_Mask\n")
    display(pd.DataFrame({
        "code": codes,
        "type de surface": [legende_terre_eau.get(int(k), "non documenté") for k in codes],
        "pixels": effectifs,
        "part_%": np.round(100 * effectifs / masque_terre_eau.size, 2),
    }).set_index("code"))

    # Codes considérés comme de l'eau à exclure (océan et eau profonde)
    CODES_EAU = [0, 7]
    part_eau = 100 * np.mean(np.isin(masque_terre_eau, CODES_EAU))
    print(f"\nAvec MASQUER_EAU = {MASQUER_EAU}, les codes {CODES_EAU} seraient écartés "
          f"({part_eau:.1f} % de la tuile).")
else:
    masque_terre_eau = None
    print("Ce produit ne contient pas de couche Land_Water_Mask.")


# --- 2. Le masque nuage, lu bit à bit ----------------------------------------
CHAMPS_NUAGE = {
    "Jour / nuit":                (0, 1, {0: "Nuit", 1: "Jour"}),
    "Fond terre / eau":           (1, 3, {0: "Terre et désert", 1: "Terre hors désert",
                                          2: "Eau intérieure", 3: "Eau de mer",
                                          5: "Zone côtière"}),
    "Qualité du masque nuage":    (4, 2, {0: "Médiocre", 1: "Faible",
                                          2: "Moyenne", 3: "Élevée"}),
    "Détection de nuage":         (6, 2, {0: "Clair (certain)", 1: "Probablement clair",
                                          2: "Probablement nuageux", 3: "Nuageux (certain)"}),
    "Ombre de nuage":             (8, 1, {0: "Non", 1: "Oui"}),
    "Cirrus":                     (9, 1, {0: "Non", 1: "Oui"}),
}


def extraire_bits(entier, premier_bit, nombre_de_bits):
    """Extrait un champ de bits d'un entier (ou d'un tableau d'entiers)."""
    return (np.asarray(entier) >> premier_bit) & ((1 << nombre_de_bits) - 1)


# On cherche un fichier journalier, seul porteur de QF_Cloud_Mask
journaliers = INVENTAIRE[INVENTAIRE["produit"] == "VNP46A2"]
if journaliers.empty:
    print("\nAucun fichier VNP46A2 dans l'inventaire : "
          "démonstration du décodage binaire sur un entier fabriqué.\n")
    exemple_entier = np.uint16(0b0000_0010_1000_0010)
    tableau_nuage = None
else:
    fichier_journalier = Path(journaliers.iloc[-1]["chemin"])
    with h5py.File(fichier_journalier, "r") as f:
        groupe_j = f[trouver_groupe_donnees(f)]
        tableau_nuage = groupe_j["QF_Cloud_Mask"][:] if "QF_Cloud_Mask" in groupe_j else None
    print(f"\nMASQUE NUAGE  —  QF_Cloud_Mask   ({fichier_journalier.name})\n")
    exemple_entier = (np.uint16(tableau_nuage.flat[0]) if tableau_nuage is not None
                      else np.uint16(0b0000_0010_1000_0010))

# Décodage d'un entier, champ par champ
print(f"Exemple : l'entier {int(exemple_entier)} vaut en binaire "
      f"{int(exemple_entier):016b}\n")
lignes_bits = []
for nom_champ, (premier, largeur, sens) in CHAMPS_NUAGE.items():
    valeur = int(extraire_bits(exemple_entier, premier, largeur))
    position = (f"bit {premier}" if largeur == 1
                else f"bits {premier}–{premier + largeur - 1}")
    lignes_bits.append({"champ": nom_champ, "position": position,
                        "bits": format(valeur, f"0{largeur}b"),
                        "valeur": valeur,
                        "signification": sens.get(valeur, "code non documenté")})
display(pd.DataFrame(lignes_bits).set_index("champ"))

# Statistiques sur toute la tuile
if tableau_nuage is not None:
    detection = extraire_bits(tableau_nuage, 6, 2)
    sens_detection = CHAMPS_NUAGE["Détection de nuage"][2]
    codes, effectifs = np.unique(detection, return_counts=True)
    print("\nRÉPARTITION DE LA DÉTECTION DE NUAGE SUR TOUTE LA TUILE")
    display(pd.DataFrame({
        "code": codes,
        "signification": [sens_detection.get(int(k), "?") for k in codes],
        "pixels": effectifs,
        "part_%": np.round(100 * effectifs / detection.size, 2),
    }).set_index("code"))
    part_nuageuse = 100 * np.mean(detection >= 2)
    print(f"\nNuit analysée : {part_nuageuse:.1f} % de la tuile est probablement ou "
          f"certainement nuageuse.")
    if part_nuageuse > 40:
        print("→ Au-delà de 40 %, une conclusion tirée de cette seule nuit serait imprudente.")

# --- Le fil rouge avance ------------------------------------------------------
radiance_apres_masques = radiance_complete.copy()
if masque_terre_eau is not None and MASQUER_EAU:
    radiance_apres_masques[np.isin(masque_terre_eau, CODES_EAU)] = np.nan
if drapeau_qualite is not None:
    radiance_apres_masques[drapeau_qualite > QA_MAX] = np.nan
noter_fil_rouge(12, "+ masque eau", np.nansum(radiance_apres_masques),
                int(np.isfinite(radiance_apres_masques).sum()), f"tuile {infos['tuile']}",
                "l'océan ne compte plus comme du territoire sombre")
````

````python
# --- Figure : les masques, et ce qu'il reste après leur application ----------
panneaux = []
if masque_terre_eau is not None:
    panneaux.append(("terre_eau", masque_terre_eau))
if "tableau_nuage" in dir() and tableau_nuage is not None:
    panneaux.append(("nuage", extraire_bits(tableau_nuage, 6, 2)))

if panneaux:
    fig, axes = plt.subplots(1, len(panneaux) + 1,
                             figsize=(4.55 * (len(panneaux) + 1), 5.5))
    fig.subplots_adjust(top=0.765, bottom=0.205, left=0.055, right=0.95, wspace=0.22)
    axes = np.atleast_1d(axes)

    for ax, (genre, donnees) in zip(axes, panneaux):
        if genre == "terre_eau":
            couleurs_codes = {0: "#2C7DA0", 1: "#BFE0CE", 2: COULEURS["or_"],
                              3: "#4EA8DE", 4: "#89C2D9", 5: "#1B3B6F",
                              6: COULEURS["vert"], 7: "#0A2342"}
            entrees = carte_categorielle(ax, donnees, EMPRISE_AFFICHAGE, legende_terre_eau,
                                         couleurs_codes, "Land_Water_Mask — type de surface")
        else:
            couleurs_codes = {0: COULEURS["menthe"], 1: "#BFE3D0",
                              2: COULEURS["or_"], 3: COULEURS["brique"]}
            entrees = carte_categorielle(ax, donnees, EMPRISE_AFFICHAGE,
                                         CHAMPS_NUAGE["Détection de nuage"][2],
                                         couleurs_codes,
                                         "QF_Cloud_Mask, bits 6-7 — détection de nuage")
        tracer_contour_pays(ax, couleur=COULEURS["encre"], lw=1.2)
        ax.set_xlabel("longitude (°)")
        legende_sous_laxe(ax, entrees, colonnes=2)
    axes[0].set_ylabel("latitude (°)")
    barre_echelle(axes[0], couleur=COULEURS["encre"])

    # Dernier panneau : l'effet cumulé des masques sur la radiance
    ax = axes[-1]
    radiance_masquee = radiance_complete.copy()
    if masque_terre_eau is not None and MASQUER_EAU:
        radiance_masquee[np.isin(masque_terre_eau, CODES_EAU)] = np.nan
    if drapeau_qualite is not None:
        radiance_masquee[drapeau_qualite > QA_MAX] = np.nan
    image = carte_nocturne(ax, radiance_masquee, EMPRISE_AFFICHAGE)
    ax.set_xlim(EMPRISE_AFFICHAGE[0], EMPRISE_AFFICHAGE[1])
    ax.set_ylim(EMPRISE_AFFICHAGE[2], EMPRISE_AFFICHAGE[3])
    tracer_contour_pays(ax, couleur=COULEURS["or_"], lw=1.6)
    barre_echelle(ax)
    restant = 100 * np.mean(np.isfinite(radiance_masquee))
    ax.set_title(f"Radiance après masques — {restant:.0f} % des pixels", pad=8, fontsize=10.5)
    ax.set_xlabel("longitude (°)")
    legende_sous_laxe(ax, [("pixels écartés (gris)", GRIS_SANS_DONNEE)], colonnes=1)
    barre_de_couleur(fig, image, ax)

    titre_figure(fig, "Les masques, et ce qu'il reste après leur application",
                 "En gris : les pixels écartés. Ce sont eux qui ne doivent jamais entrer dans une somme ou une moyenne.\n"
                 "Sur une tuile côtière, l'océan peut représenter 30 à 50 % des pixels — et il n'est pas noir.",
                 f"Source : {FICHIER_EXEMPLE.name}"
                 + ("   ·   MODE DÉMONSTRATION" if MODE_DEMO else ""))
    plt.show()
````

 <div style="color:#231F20;font-size:14px;margin-top:4px;">Les étapes 07 à 12 tiennent en une fonction</div>

**Ce que fait cette étape.** Tout ce que nous avons démonté séparément se remonte ici en une
seule fonction, `lire_ntl()`. Elle applique **dans le bon ordre** les six opérations, et cet ordre n’est
pas négociable :

1. ouvrir le fichier et **trouver** le groupe de données (ne jamais le supposer) ;
2. **choisir** la couche de radiance ;
3. lire les attributs et **décoder** : `valeur × scale_factor + add_offset` ;
4. remplacer le **`_FillValue` par `NaN`** — avant tout calcul ;
5. écarter les pixels dont le **drapeau qualité** dépasse le seuil ;
6. écarter les pixels d’**eau**, puis calculer le géoréférencement.

Inverser les étapes 3 et 4 — décoder avant de neutraliser le `_FillValue` — donne exactement le désastre
chiffré à l’étape 09. La fonction renvoie un dictionnaire contenant la radiance, l’emprise, la
transformation affine et un **journal de traitement** : le nombre de pixels perdus à chaque filtre.

<div style="background:#E8F5EF;border:1px solid #00704A44;border-left:4px solid #00704A;border-radius:8px;
 padding:13px 17px;margin:10px 0;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
 <div style="color:#00704A;font-weight:700;font-size:13.5px;">&#128273;&nbsp; Pourquoi un journal de traitement ?</div>
 <div style="color:#231F20;font-size:13.5px;margin-top:5px;line-height:1.6;">Parce qu’un indicateur sans traçabilité n’est pas publiable. Le champ <code>journal</code> retourné par <code>lire_ntl()</code> vous dit combien de pixels ont été écartés, et par quel filtre. Copié dans vos métadonnées, il répond d’avance à la question que posera le premier relecteur attentif : « sur quoi exactement ce chiffre a-t-il été calculé ? »</div>
</div>

````python
# =============================================================================
#  ÉTAPE 13 — LA FONCTION DE LECTURE PROPRE
# =============================================================================
CODES_EAU_A_EXCLURE = [0, 7]      # océan peu profond, océan profond


def _sous_echantillonner(tableau, facteur):
    """Moyenne des blocs facteur × facteur, en ignorant les NaN."""
    if facteur is None or facteur <= 1:
        return tableau
    h, w = tableau.shape
    h2, w2 = (h // facteur) * facteur, (w // facteur) * facteur
    decoupe = tableau[:h2, :w2].reshape(h2 // facteur, facteur, w2 // facteur, facteur)
    return np.nanmean(decoupe, axis=(1, 3))


def lire_ntl(chemin, couche=None, qa_max=None, masquer_eau=None,
             sous_echantillonnage=1, verbeux=False):
    """Lit un fichier VNP46 et renvoie une radiance propre et géoréférencée.

    Paramètres
    ----------
    chemin : str ou Path
        Le fichier .h5 à lire.
    couche : str, optionnel
        Nom de la couche de radiance. Par défaut, choisie automatiquement.
    qa_max : int, optionnel
        Drapeau qualité maximal accepté. Par défaut, la valeur de QA_MAX.
    masquer_eau : bool, optionnel
        Écarter les pixels d'océan. Par défaut, la valeur de MASQUER_EAU.
    sous_echantillonnage : int
        Facteur d'agrégation pour l'affichage (1 = pleine résolution).
    verbeux : bool
        Afficher le journal de traitement au fur et à mesure.

    Retour
    ------
    dict avec les clés : radiance, emprise, transformation, longitudes,
    latitudes, couche, metadonnees, journal.
    """
    chemin = Path(chemin)
    qa_max = QA_MAX if qa_max is None else qa_max
    masquer_eau = MASQUER_EAU if masquer_eau is None else masquer_eau

    infos_nom = decoder_nom_fichier(chemin.name)
    if infos_nom is None:
        raise ValueError(f"Nom de fichier non conforme à la convention VNP46 : {chemin.name}")

    journal = []

    with h5py.File(chemin, "r") as f:
        # --- 1. trouver le groupe, ne jamais le supposer ---------------------
        groupe = f[trouver_groupe_donnees(f)]

        # --- 2. choisir la couche de radiance --------------------------------
        nom_couche = couche or choisir_couche_radiance(groupe)
        jeu = groupe[nom_couche]

        # --- 3. lire les attributs DANS le fichier ---------------------------
        facteur = lire_attribut(jeu, "scale_factor", 1.0)
        offset = lire_attribut(jeu, "add_offset", lire_attribut(jeu, "offset", 0.0))
        vide = lire_attribut(jeu, "_FillValue", None)
        unite = lire_attribut(jeu, "units", "")

        brut = jeu[:].astype("float32")
        total = brut.size
        journal.append(("pixels du fichier", total, 100.0))

        # --- 4. neutraliser le _FillValue AVANT de décoder -------------------
        invalide = ~np.isfinite(brut)
        if vide is not None:
            invalide |= np.isclose(brut, float(vide))
        radiance = brut * facteur + offset
        radiance[invalide] = np.nan
        radiance[radiance < 0] = np.nan          # une radiance négative n'existe pas
        perdus = int(invalide.sum())
        journal.append(("écartés : _FillValue", -perdus, -100.0 * perdus / total))

        # --- 5. filtrer sur le drapeau qualité -------------------------------
        nom_drapeau = next((n for n in (f"{nom_couche}_Quality", "Mandatory_Quality_Flag")
                            if n in groupe), None)
        if nom_drapeau is not None and qa_max is not None:
            drapeau = groupe[nom_drapeau][:]
            a_ecarter = (drapeau > qa_max) & np.isfinite(radiance)
            radiance[a_ecarter] = np.nan
            journal.append((f"écartés : qualité > {qa_max}", -int(a_ecarter.sum()),
                            -100.0 * int(a_ecarter.sum()) / total))

        # --- 6. masquer l'eau -------------------------------------------------
        if masquer_eau and "Land_Water_Mask" in groupe:
            terre_eau = groupe["Land_Water_Mask"][:]
            a_ecarter = np.isin(terre_eau, CODES_EAU_A_EXCLURE) & np.isfinite(radiance)
            radiance[a_ecarter] = np.nan
            journal.append(("écartés : eau", -int(a_ecarter.sum()),
                            -100.0 * int(a_ecarter.sum()) / total))

        metadonnees = {"produit": infos_nom["produit"], "tuile": infos_nom["tuile"],
                       "date": infos_nom["date"].date(), "couche": nom_couche,
                       "scale_factor": facteur, "add_offset": offset,
                       "_FillValue": vide, "unite": unite,
                       "drapeau_qualite": nom_drapeau, "qa_max": qa_max,
                       "eau_masquee": bool(masquer_eau), "fichier": chemin.name}

    # --- 7. géoréférencement ---------------------------------------------------
    o, e, s, n = emprise_tuile(infos_nom["tuile"])
    if sous_echantillonnage and sous_echantillonnage > 1:
        radiance = _sous_echantillonner(radiance, sous_echantillonnage)
    lignes, colonnes = radiance.shape
    res_lon, res_lat = (e - o) / colonnes, (n - s) / lignes

    journal.append(("pixels exploitables", int(np.isfinite(radiance).sum()),
                    100.0 * float(np.isfinite(radiance).mean())))

    if verbeux:
        print(f"{chemin.name}")
        for etiquette, valeur, part in journal:
            print(f"   {etiquette:<28} {valeur:>12,}   {part:>7.2f} %".replace(",", " "))

    return {"radiance": radiance,
            "emprise": (o, e, s, n),
            "transformation": (res_lon, 0.0, o, 0.0, -res_lat, n),
            "longitudes": o + (np.arange(colonnes) + 0.5) * res_lon,
            "latitudes": n - (np.arange(lignes) + 0.5) * res_lat,
            "couche": metadonnees["couche"],
            "metadonnees": metadonnees,
            "journal": pd.DataFrame(journal, columns=["étape", "pixels", "part_%"])}


# --- Mise en oeuvre -----------------------------------------------------------
RESULTAT = lire_ntl(FICHIER_EXEMPLE, sous_echantillonnage=SOUS_ECHANTILLONNAGE)

print("JOURNAL DE TRAITEMENT")
display(RESULTAT["journal"].set_index("étape"))

print("\nMÉTADONNÉES DE SORTIE — à recopier dans votre documentation")
display(pd.Series(RESULTAT["metadonnees"], name="valeur").to_frame())

radiance = RESULTAT["radiance"]
valides = radiance[np.isfinite(radiance)]
print(f"\nRadiance : {valides.size:,} pixels exploitables".replace(",", " "))
print(f"   minimum {valides.min():.3f}   médiane {np.median(valides):.3f}   "
      f"maximum {valides.max():.1f}   (nW·cm⁻²·sr⁻¹)")
print(f"   somme des lumières (SoL) : {valides.sum():,.0f}".replace(",", " "))

# --- Le fil rouge : la fonction doit retrouver le chiffre de l'étape 12 -------
noter_fil_rouge(13, "via lire_ntl() (vérification)", valides.sum(), valides.size,
                f"tuile {RESULTAT['metadonnees']['tuile']}",
                "même résultat que l'étape 12 : la fonction est fidèle")
````

 padding:13px 17px;margin:10px 0;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
 <div style="color:#00A86A;font-weight:700;font-size:13.5px;">&#128200;&nbsp; Comment lire le journal</div>
 <div style="color:#231F20;font-size:13.5px;margin-top:5px;line-height:1.6;">Chaque ligne négative est une perte volontaire. Additionnez-les : si vous écartez plus de la moitié de la tuile, ce n’est pas forcément anormal (une tuile côtière est à moitié océanique), mais vous devez pouvoir l’expliquer. Une perte brutale sur la ligne « qualité » signale en général une période très nuageuse — c’est le moment de regarder la couche <code>…_Num</code>.</div>
</div>

 <span style="color:#D49A00;font-size:10.5px;font-weight:700;letter-spacing:2px;text-transform:uppercase;">À retenir</span>
 <span style="color:#231F20;font-size:13.5px;margin-left:12px;">Une fonction unique, un ordre d’opérations non négociable, et un journal pour le prouver.</span>
</div>

### 14 · Du carré au pays

<div style="border-left:6px solid #F5C242;background:#E8F5EF;border-radius:0 12px 12px 0;
 padding:14px 22px;margin-top:22px;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
 <div style="color:#D49A00;font-size:11px;font-weight:700;letter-spacing:2.5px;text-transform:uppercase;">
  ÉTAPE 14 · 20 &nbsp;&nbsp;<span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#F5C242;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span></div>
 <div style="color:#231F20;font-size:14px;margin-top:4px;">Mosaïquer les tuiles, puis découper sur une frontière</div>
</div>

**Le chaînon manquant.** Jusqu’ici nous avons travaillé sur *une tuile*. Or une tuile n’est pas un territoire : c’est un carré arbitraire de 10° × 10° qui ignore les frontières. Aucun indicateur publiable ne peut en sortir tel quel.

Deux opérations séparent la tuile de l’indicateur, et cette étape les fait toutes les deux :

1. **Mosaïquer** — recoller les tuiles qui couvrent le pays dans une grille commune. Comme elles partagent la même résolution et la même projection, le collage est un simple calcul d’indices : pas de rééchantillonnage, donc aucune déformation des valeurs.
2. **Découper** — écarter tout pixel situé hors de la frontière, avec `masque_pays()`.

L’ordre compte : on mosaïque d’abord, on découpe ensuite. L’inverse obligerait à gérer les pixels à cheval sur deux tuiles au bord de la frontière.

````python
# =============================================================================
#  ÉTAPE 14 — DU CARRÉ AU PAYS : MOSAÏQUER, PUIS DÉCOUPER
# =============================================================================

# --- 1. Quelles tuiles assembler ? -------------------------------------------
lot = INVENTAIRE[INVENTAIRE["produit"] == PRODUIT_PRINCIPAL]
if lot.empty:
    lot = INVENTAIRE
annee_cible = int(lot["annee"].max())
lot = lot[lot["annee"] == annee_cible]
tuiles_lot = [t for t in TUILES_PAYS if t in set(lot["tuile"])] or sorted(lot["tuile"].unique())

print(f"MOSAÏQUE  ·  produit {PRODUIT_PRINCIPAL}  ·  année {annee_cible}  ·  {NOM_PAYS}")
print(f"  tuiles du pays     : {', '.join(TUILES_PAYS)}")
print(f"  tuiles disponibles : {', '.join(tuiles_lot)}")
if set(TUILES_PAYS) - set(tuiles_lot):
    print(f"  ⚠ manquantes       : {', '.join(sorted(set(TUILES_PAYS) - set(tuiles_lot)))}")
    print("    (la mosaïque sera incomplète : le chiffre final sera sous-estimé)")
print()

# --- 2. Lire chaque tuile avec la MÊME fonction et les MÊMES réglages --------
morceaux = {}
for t in tuiles_lot:
    chemin_t = Path(lot[lot["tuile"] == t].iloc[-1]["chemin"])
    morceaux[t] = lire_ntl(chemin_t, sous_echantillonnage=SOUS_ECHANTILLONNAGE)
    print(f"  {t} : {morceaux[t]['radiance'].shape[0]} × "
          f"{morceaux[t]['radiance'].shape[1]} pixels")

# --- 3. Coller les morceaux dans une grille commune --------------------------
o_g = min(morceaux[t]["emprise"][0] for t in tuiles_lot)
e_g = max(morceaux[t]["emprise"][1] for t in tuiles_lot)
s_g = min(morceaux[t]["emprise"][2] for t in tuiles_lot)
n_g = max(morceaux[t]["emprise"][3] for t in tuiles_lot)

premier = morceaux[tuiles_lot[0]]
resolution_m = (premier["emprise"][1] - premier["emprise"][0]) / premier["radiance"].shape[1]

MOSAIQUE = np.full((int(round((n_g - s_g) / resolution_m)),
                    int(round((e_g - o_g) / resolution_m))), np.nan, dtype="float32")
for t, res_t in morceaux.items():
    o_t, _, _, n_t = res_t["emprise"]
    col0 = int(round((o_t - o_g) / resolution_m))
    lig0 = int(round((n_g - n_t) / resolution_m))
    h_t, w_t = res_t["radiance"].shape
    MOSAIQUE[lig0:lig0 + h_t, col0:col0 + w_t] = res_t["radiance"]

EMPRISE_MOSAIQUE = (o_g, e_g, s_g, n_g)
LON_M, LAT_M = grille_coordonnees(EMPRISE_MOSAIQUE, MOSAIQUE.shape)

print(f"\n  Mosaïque obtenue : {MOSAIQUE.shape[0]} × {MOSAIQUE.shape[1]} pixels")
print(f"  Emprise          : longitude [{o_g:+.0f}° , {e_g:+.0f}°]  ·  "
      f"latitude [{s_g:+.0f}° , {n_g:+.0f}°]")

# --- 4. Découper sur la frontière nationale ----------------------------------
DANS_LE_PAYS = masque_pays(LON_M, LAT_M)
RADIANCE_PAYS = np.where(DANS_LE_PAYS, MOSAIQUE, np.nan)

valides_mosaique = MOSAIQUE[np.isfinite(MOSAIQUE)]
valides_pays = RADIANCE_PAYS[np.isfinite(RADIANCE_PAYS)]

display(pd.DataFrame([
    {"périmètre": "mosaïque brute (les tuiles entières)",
     "pixels exploitables": f"{valides_mosaique.size:,}".replace(",", " "),
     "somme des lumières": f"{valides_mosaique.sum():,.0f}".replace(",", " "),
     "radiance médiane": round(float(np.median(valides_mosaique)), 3)},
    {"périmètre": f"territoire national — {NOM_PAYS}",
     "pixels exploitables": f"{valides_pays.size:,}".replace(",", " "),
     "somme des lumières": f"{valides_pays.sum():,.0f}".replace(",", " "),
     "radiance médiane": round(float(np.median(valides_pays)), 3) if valides_pays.size else np.nan},
]).set_index("périmètre"))

part_gardee = 100 * valides_pays.size / max(valides_mosaique.size, 1)
print(f"\nLe territoire de {NOM_PAYS} ne représente que {part_gardee:.1f} % des pixels")
print("exploitables de la mosaïque. Tout le reste appartient aux pays voisins ou à la")
print("mer : l'inclure dans un indicateur national n'aurait aucun sens.\n")

# --- 5. Le fil rouge avance ---------------------------------------------------
perimetre_tuiles = ("la tuile entière" if len(tuiles_lot) == 1
                    else f"{len(tuiles_lot)} tuiles entières")
noter_fil_rouge(14, "mosaïque des tuiles", valides_mosaique.sum(), valides_mosaique.size,
                perimetre_tuiles,
                "même chiffre qu'à l'étape 13 si une seule tuile suffit")
noter_fil_rouge(14, "découpage sur la frontière", valides_pays.sum(), valides_pays.size,
                f"territoire {PAYS}", "le chiffre devient un indicateur national")
````

````python
# --- Figure : trois échelles, de la mosaïque au quartier ---------------------
villes_p = villes_du_pays(nombre=9)
ville_phare = villes_p[0] if villes_p else None
o_p, e_p, s_p, n_p = EMPRISE_PAYS
marge_p = max(0.6, 0.08 * max(e_p - o_p, n_p - s_p))

fig, axes = plt.subplots(1, 3, figsize=(13.6, 5.9))
fig.subplots_adjust(top=0.765, bottom=0.135, left=0.05, right=0.945, wspace=0.26)

# (1) la mosaïque complète, frontière posée dessus
im1 = carte_nocturne(axes[0], MOSAIQUE, EMPRISE_MOSAIQUE)
axes[0].set_xlim(EMPRISE_MOSAIQUE[0], EMPRISE_MOSAIQUE[1])
axes[0].set_ylim(EMPRISE_MOSAIQUE[2], EMPRISE_MOSAIQUE[3])
tracer_contour_pays(axes[0], couleur=COULEURS["or_"], lw=1.7)
axes[0].add_patch(plt.Rectangle((o_p - marge_p, s_p - marge_p),
                                e_p - o_p + 2 * marge_p, n_p - s_p + 2 * marge_p,
                                facecolor="none", edgecolor="white", lw=1.0,
                                ls=(0, (4, 3)), alpha=0.85, zorder=7))
barre_echelle(axes[0])
axes[0].set_title(f"1 · La mosaïque — {' + '.join(tuiles_lot)}", pad=8, fontsize=10.5)
axes[0].set_ylabel("latitude (°)")

# (2) le pays seul, après découpage
im2 = carte_nocturne(axes[1], RADIANCE_PAYS, EMPRISE_MOSAIQUE)
axes[1].set_xlim(o_p - marge_p, e_p + marge_p)
axes[1].set_ylim(s_p - marge_p, n_p + marge_p)
tracer_contour_pays(axes[1], couleur=COULEURS["or_"], lw=1.7)
etiqueter_villes(axes[1], villes_p, taille=8.2)
barre_echelle(axes[1])
axes[1].set_title(f"2 · {NOM_PAYS} seul — après découpage", pad=8, fontsize=10.5)

# (3) la principale agglomération
if ville_phare is not None:
    nom_v, clon, clat, _pop = ville_phare
    marge_v = 0.55
    fenetre = ((np.abs(LON_M - clon) <= marge_v) & (np.abs(LAT_M - clat) <= marge_v)
               & np.isfinite(RADIANCE_PAYS))
    plafond_v = (float(np.nanpercentile(RADIANCE_PAYS[fenetre], 99.9))
                 if fenetre.any() else None)
    im3 = carte_nocturne(axes[2], RADIANCE_PAYS, EMPRISE_MOSAIQUE, vmax=plafond_v,
                         interpolation="bilinear")
    axes[2].set_xlim(clon - marge_v, clon + marge_v)
    axes[2].set_ylim(clat - marge_v, clat + marge_v)
    tracer_contour_pays(axes[2], couleur=COULEURS["or_"], lw=1.7)
    etiqueter_villes(axes[2], villes_dans(
        (clon - marge_v, clon + marge_v, clat - marge_v, clat + marge_v), nombre=5), taille=9.5)
    barre_echelle(axes[2])
    axes[2].set_title(f"3 · {nom_v} — {int(2 * marge_v * 111):d} km de côté",
                      pad=8, fontsize=10.5)
    axes[1].add_patch(plt.Rectangle((clon - marge_v, clat - marge_v), 2 * marge_v, 2 * marge_v,
                                    facecolor="none", edgecolor="white", lw=1.1,
                                    ls=(0, (4, 3)), zorder=8))
    barre_de_couleur(fig, im3, axes[2])
else:
    axes[2].axis("off")
    barre_de_couleur(fig, im2, axes[1])

for ax in axes[:2]:
    ax.set_xlabel("longitude (°)")
axes[2].set_xlabel("longitude (°)")

titre_figure(fig, f"Trois échelles, un seul jeu de données — {NOM_PAYS}",
             "La mosaïque assemble les tuiles ; le masque national écarte tout ce qui n'est pas le pays ;\n"
             "le zoom montre qu'à 500 m de résolution, une agglomération se lit quartier par quartier.",
             f"Produit {PRODUIT_PRINCIPAL}, année {annee_cible}. Frontières : Natural Earth 1:50 M, domaine public."
             + ("   ·   MODE DÉMONSTRATION — valeurs simulées" if MODE_DEMO else ""))
plt.show()
````

<div style="background:#F4F7F5;border:1px solid #00A86A44;border-left:4px solid #00A86A;border-radius:8px;
 padding:13px 17px;margin:10px 0;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
 <div style="color:#00A86A;font-weight:700;font-size:13.5px;">&#128200;&nbsp; Comment lire ces trois cartes</div>
 <div style="color:#231F20;font-size:13.5px;margin-top:5px;line-height:1.6;"><b>1 · la mosaïque</b> — les tuiles assemblées, bord à bord, sans coupure visible. <b>2 · le pays seul</b> — le même jeu de données, découpé sur la frontière : tout ce qui appartient aux voisins ou à la mer est passé en gris. <b>3 · la ville</b> — une centaine de kilomètres de côté, soit la fenêtre dans laquelle un urbaniste travaille.<br><br>Les deux premières cartes partagent la <b>même échelle</b> : la comparaison est donc honnête. La troisième a sa propre échelle, parce qu'un quartier n'a pas la dynamique d'un pays — et c'est précisément le genre de détail qu'il faut écrire dans une note de méthode.</div>
</div>

<div style="background:#E8F5EF;border:1px solid #00704A44;border-left:4px solid #00704A;border-radius:8px;

### 15 · La distribution des radiances

<div style="border-left:6px solid #00A86A;background:#E8F5EF;border-radius:0 12px 12px 0;
 padding:14px 22px;margin-top:22px;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
 <div style="color:#D49A00;font-size:11px;font-weight:700;letter-spacing:2.5px;text-transform:uppercase;">
  ÉTAPE 15 · 20 &nbsp;&nbsp;<span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#00A86A;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span></div>
 <div style="color:#231F20;font-size:14px;margin-top:4px;">La propriété statistique la plus importante, et la plus ignorée</div>
</div>

**Ce qu’il faut absolument comprendre.** La radiance nocturne n’est pas distribuée
normalement. Elle est **extrêmement asymétrique** : l’immense majorité des pixels sont très sombres, et
une poignée de pixels urbains concentre l’essentiel de la lumière totale. Il n’est pas rare que **1 % des
pixels porte plus de la moitié de la somme des lumières**.

Trois conséquences pratiques, qui décident de la qualité de votre analyse :

- **Cartographier en échelle linéaire ne montre rien** : la carte est noire, avec trois points blancs.
  Il faut passer en **échelle logarithmique** — c’est pourquoi toutes les cartes de ce notebook
  utilisent `log1p`.
- **La moyenne est un mauvais résumé.** Elle est dominée par quelques pixels. Pour une comparaison
  entre régions, la **somme** (SoL) et la **médiane** disent deux choses différentes, toutes deux utiles.
- **Un seul pixel aberrant peut déplacer un indicateur national.** Une torchère de gaz, un incendie,
  un navire : cartographiez toujours vos valeurs extrêmes avant de publier.

````python
# =============================================================================
#  ÉTAPE 14 — LA DISTRIBUTION DES RADIANCES
# =============================================================================
valides = radiance[np.isfinite(radiance)]
positifs = valides[valides > 0]

# --- Statistiques descriptives ------------------------------------------------
quantiles = [1, 5, 25, 50, 75, 90, 95, 99, 99.9]
resume = pd.DataFrame({
    "statistique": (["pixels exploitables", "moyenne", "écart-type", "minimum"]
                    + [f"centile {q}" for q in quantiles] + ["maximum", "somme (SoL)"]),
    "valeur": ([f"{valides.size:,}".replace(",", " "),
                f"{valides.mean():.3f}", f"{valides.std():.3f}", f"{valides.min():.3f}"]
               + [f"{np.percentile(valides, q):.3f}" for q in quantiles]
               + [f"{valides.max():.1f}", f"{valides.sum():,.0f}".replace(",", " ")]),
})
display(resume.set_index("statistique"))

# --- La concentration : quelle part de lumière pour quelle part de pixels ? ----
tries = np.sort(valides)[::-1]
cumul = np.cumsum(tries) / tries.sum()
concentration = pd.DataFrame({
    "part des pixels les plus lumineux": ["0,1 %", "1 %", "5 %", "10 %", "25 %", "50 %"],
    "part de la lumière totale": [
        f"{100 * cumul[max(int(p * tries.size / 100) - 1, 0)]:.1f} %"
        for p in (0.1, 1, 5, 10, 25, 50)],
})
print("\nCONCENTRATION DE LA LUMIÈRE")
display(concentration.set_index("part des pixels les plus lumineux"))

seuil_moitie = 100 * (np.searchsorted(cumul, 0.5) + 1) / tries.size
print(f"\nIl suffit des {seuil_moitie:.2f} % de pixels les plus lumineux pour "
      f"atteindre la moitié de la somme des lumières.")
````

````python
# --- Figure : quatre regards sur la même distribution ------------------------
fig = plt.figure(figsize=(12.8, 7.8))
grille = fig.add_gridspec(2, 2, left=0.075, right=0.945, top=0.80, bottom=0.085,
                          hspace=0.46, wspace=0.26)
axA = fig.add_subplot(grille[0, 0])
axB = fig.add_subplot(grille[0, 1])
axC = fig.add_subplot(grille[1, 0])
axD = fig.add_subplot(grille[1, 1])

# (a) en échelle linéaire : tout s'écrase à gauche
axA.hist(valides, bins=90, color=COULEURS["sauge"], edgecolor="none")
axA.set_yscale("log")
axA.set_title("(a) Échelle linéaire : tout est écrasé à gauche", pad=7, fontsize=10.5)
axA.set_xlabel("radiance (nW·cm⁻²·sr⁻¹)")
axA.set_ylabel("nombre de pixels (log)")
axA.text(0.97, 0.9, "la carte paraît noire,\net la moyenne ne veut rien dire",
         transform=axA.transAxes, ha="right", va="top", fontsize=8.2,
         color=COULEURS["brique"], style="italic")

# (b) en échelle logarithmique : la forme apparaît
if positifs.size:
    bornes = np.logspace(np.log10(max(positifs.min(), 1e-3)), np.log10(positifs.max()), 48)
    axB.hist(positifs, bins=bornes, color=COULEURS["vert_fonce"], edgecolor="none")
    axB.set_xscale("log")
    axB.set_yscale("log")
    bas, haut = axB.get_ylim()
    for rang, (centile, couleur, etiquette) in enumerate(
            [(50, COULEURS["or_"], "médiane"), (99, COULEURS["brique"], "centile 99")]):
        v = float(np.percentile(positifs, centile))
        axB.axvline(v, color=couleur, lw=1.8, ls="--", zorder=4)
        axB.text(v * 1.15, haut ** (0.92 - 0.14 * rang) * bas ** (0.08 + 0.14 * rang),
                 f"{etiquette}\n{v:.2f}", color=couleur, fontsize=8.4,
                 fontweight="bold", va="top", ha="left")
axB.set_title("(b) Échelle logarithmique : la forme apparaît", pad=7, fontsize=10.5)
axB.text(0.5, 0.04, "les barres isolées à gauche sont la quantification du fichier :\n"
                    "la radiance y est stockée par pas de 0,1 nW·cm⁻²·sr⁻¹",
         transform=axB.transAxes, ha="center", va="bottom", fontsize=7.6,
         color=COULEURS["ardoise"], style="italic")
axB.set_xlabel("radiance (échelle log)")
axB.set_ylabel("nombre de pixels (log)")

# (c) la courbe de concentration
part_pixels = 100 * np.arange(1, tries.size + 1) / tries.size
pas = max(tries.size // 4000, 1)
axC.plot(part_pixels[::pas], 100 * cumul[::pas], lw=2.4, color=COULEURS["sarcelle"], zorder=4)
axC.plot([0, 100], [0, 100], lw=1.1, ls=":", color=COULEURS["ardoise"], zorder=3)
axC.fill_between(part_pixels[::pas], 100 * cumul[::pas], part_pixels[::pas],
                 color=COULEURS["sarcelle"], alpha=0.14, zorder=2)
axC.axhline(50, color=COULEURS["or_"], lw=1.2, ls="--", zorder=3)
axC.scatter([seuil_moitie], [50], s=48, color=COULEURS["or_"], zorder=6)
axC.annotate(f"{seuil_moitie:.1f} % des pixels\nportent 50 % de la lumière",
             xy=(seuil_moitie, 50), xytext=(26, 26), fontsize=8.6,
             color=COULEURS["ocre"], fontweight="bold",
             arrowprops=dict(arrowstyle="->", color=COULEURS["ocre"], lw=1.3))
axC.set_xlim(0, 100)
axC.set_ylim(0, 102)
axC.set_xlabel("part des pixels, du plus lumineux au plus sombre (%)")
axC.set_ylabel("part cumulée de la lumière (%)")
axC.set_title("(c) Une poignée de pixels fait tout", pad=7, fontsize=10.5)
axC.text(72, 56, "répartition parfaitement\nhomogène (théorique)", fontsize=7.6,
         color=COULEURS["ardoise"], style="italic", ha="center")

# (d) la carte, avec la même convention que partout ailleurs
imD = carte_nocturne(axD, radiance, RESULTAT["emprise"])
axD.set_xlim(RESULTAT["emprise"][0], RESULTAT["emprise"][1])
axD.set_ylim(RESULTAT["emprise"][2], RESULTAT["emprise"][3])
tracer_contour_pays(axD, couleur=COULEURS["or_"], lw=1.5)
barre_echelle(axD)
axD.set_title("(d) La carte correspondante, en échelle log", pad=7, fontsize=10.5)
axD.set_xlabel("longitude (°)")
axD.set_ylabel("latitude (°)")
barre_de_couleur(fig, imD, axD, fraction=0.05)

titre_figure(fig, "La radiance nocturne est une grandeur très asymétrique",
             "Ignorer cette asymétrie, c'est produire des cartes noires, des moyennes trompeuses,\n"
             "et des indicateurs qu'un seul pixel aberrant peut faire basculer.",
             f"Source : {RESULTAT['metadonnees']['fichier']}   ·   couche {RESULTAT['couche']}"
             + ("   ·   MODE DÉMONSTRATION — valeurs simulées" if MODE_DEMO else ""))
plt.show()
````

### 16 · Comparer les couches entre elles

<div style="border-left:6px solid #00A86A;background:#E8F5EF;border-radius:0 12px 12px 0;
 padding:14px 22px;margin-top:22px;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
 <div style="color:#D49A00;font-size:11px;font-weight:700;letter-spacing:2.5px;text-transform:uppercase;">
  ÉTAPE 16 · 20 &nbsp;&nbsp;<span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#00A86A;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span></div>
 <div style="color:#231F20;font-size:14px;margin-top:4px;">Pourquoi le fichier contient plusieurs versions de la même image</div>
</div>

**Ce que révèle cette comparaison.** Un composite propose plusieurs vues de la même scène :
`AllAngle` (toutes les géométries de visée confondues), `NearNadir` (uniquement les passages proches de la
verticale), `OffNadir` (les passages obliques). Elles ne donnent pas exactement le même chiffre.

À cela s’ajoutent les deux couches d’incertitude, qui sont celles que l’on oublie le plus :
`…_Num` dit **sur combien de nuits** le composite a été calculé, et `…_Std` **à quel point ces nuits
divergeaient**. Un pixel à forte radiance et fort écart-type, calculé sur trois nuits, n’a pas la même
valeur probante qu’un pixel stable calculé sur vingt-cinq nuits — et pourtant les deux se ressemblent
sur une carte de radiance.

````python
# =============================================================================
#  ÉTAPE 15 — COMPARER LES COUCHES ENTRE ELLES
# =============================================================================
with h5py.File(FICHIER_EXEMPLE, "r") as f:
    groupe = f[trouver_groupe_donnees(f)]
    disponibles = {nom: obj for nom, obj in groupe.items()
                   if isinstance(obj, h5py.Dataset) and obj.ndim == 2}

    # (1) les variantes de radiance
    variantes = [n for n in disponibles
                 if ("Composite" in n or "NTL" in n)
                 and not n.endswith(("_Num", "_Std", "_Quality"))]

    lignes = []
    tableaux = {}
    for nom in variantes:
        objet = groupe[nom]
        fac = lire_attribut(objet, "scale_factor", 1.0)
        off = lire_attribut(objet, "add_offset", 0.0)
        vid = lire_attribut(objet, "_FillValue", None)
        valeurs = objet[:].astype("float32")
        valeurs = np.where(np.isclose(valeurs, float(vid)) if vid is not None else False,
                           np.nan, valeurs * fac + off)
        tableaux[nom] = valeurs
        ok = valeurs[np.isfinite(valeurs)]
        lignes.append({"couche": nom, "pixels_valides": ok.size,
                       "moyenne": round(float(ok.mean()), 3) if ok.size else np.nan,
                       "médiane": round(float(np.median(ok)), 3) if ok.size else np.nan,
                       "maximum": round(float(ok.max()), 1) if ok.size else np.nan,
                       "somme (SoL)": round(float(ok.sum()), 0) if ok.size else np.nan})

    # (2) les couches d'incertitude
    nom_num = next((n for n in disponibles if n.endswith("_Num")), None)
    nom_std = next((n for n in disponibles if n.endswith("_Std")), None)
    table_num = groupe[nom_num][:].astype("float32") if nom_num else None
    if nom_std is not None:
        objet_std = groupe[nom_std]
        f_std = lire_attribut(objet_std, "scale_factor", 1.0)
        v_std = lire_attribut(objet_std, "_FillValue", None)
        table_std = objet_std[:].astype("float32")
        table_std = np.where(np.isclose(table_std, float(v_std)) if v_std is not None else False,
                             np.nan, table_std * f_std)
    else:
        table_std = None

print("LES VARIANTES DE RADIANCE PRÉSENTES DANS LE FICHIER\n")
comparaison = pd.DataFrame(lignes).sort_values("somme (SoL)", ascending=False)
display(comparaison.set_index("couche"))

# Écart relatif entre la couche de référence et les autres
reference = RESULTAT["couche"]
if reference in tableaux and len(tableaux) > 1:
    base = np.nansum(tableaux[reference])
    print(f"\nÉCART DE SOMME DES LUMIÈRES PAR RAPPORT À « {reference} »")
    for nom, valeurs in tableaux.items():
        if nom == reference:
            continue
        ecart = 100 * (np.nansum(valeurs) - base) / base if base else np.nan
        print(f"   {nom:<44} {ecart:+7.2f} %")

if nom_num:
    fini = table_num[np.isfinite(table_num) & (table_num < 255)]
    print(f"\nNOMBRE D'OBSERVATIONS  ({nom_num})")
    print(f"   médiane {np.median(fini):.0f} nuits   ·   "
          f"minimum {fini.min():.0f}   ·   maximum {fini.max():.0f}")
    faibles = 100 * np.mean(fini <= 5)
    print(f"   {faibles:.1f} % des pixels reposent sur 5 nuits ou moins")
````

````python
# --- Figure : radiance, nombre d'observations, écart-type --------------------
panneaux = [("Radiance décodée", np.where(np.isfinite(radiance), np.log1p(radiance), np.nan),
             CMAP_NUIT, "log(1 + nW·cm⁻²·sr⁻¹)", True)]
if table_num is not None:
    n_affiche = np.where(table_num < 255, table_num, np.nan)
    panneaux.append((f"Nombre de nuits utilisées\n({nom_num})", n_affiche,
                     CMAP_VERT, "nuits", False))
if table_std is not None:
    panneaux.append((f"Écart-type entre ces nuits\n({nom_std})",
                     np.where(np.isfinite(table_std), np.log1p(table_std), np.nan),
                     "magma", "log(1 + écart-type)", False))

fig, axes = plt.subplots(1, len(panneaux), figsize=(4.55 * len(panneaux), 4.9))
fig.subplots_adjust(top=0.72, bottom=0.165, left=0.055, right=0.97, wspace=0.26)
axes = np.atleast_1d(axes)

for ax, (titre, donnees, cmap, etiquette, est_log) in zip(axes, panneaux):
    fini = donnees[np.isfinite(donnees)]
    vmin = float(np.nanpercentile(fini, 1)) if fini.size else 0.0
    vmax = float(np.nanpercentile(fini, 99.5)) if fini.size else 1.0
    im = ax.imshow(donnees, cmap=cmap, vmin=vmin, vmax=max(vmax, vmin + 1e-9),
                   extent=RESULTAT["emprise"], origin="upper", aspect="auto",
                   interpolation="nearest")
    ax.set_facecolor("#DCE4E0")
    ax.set_title(titre, pad=7, fontsize=10.2)
    ax.set_xlabel("longitude (°)")
    barre = fig.colorbar(im, ax=ax, fraction=0.045, pad=0.02)
    barre.set_label(etiquette, fontsize=7.6)
    barre.ax.tick_params(labelsize=7)
axes[0].set_ylabel("latitude (°)")

titre_figure(fig, "Une radiance seule ne dit pas si l'on peut s'y fier",
             "Les deux panneaux de droite sont ceux que l'on oublie : ils indiquent sur combien de nuits chaque pixel\n"
             "a été calculé, et à quel point ces nuits divergeaient. Ce sont eux qui justifient vos limitations.",
             f"Source : {RESULTAT['metadonnees']['fichier']}"
             + ("   ·   MODE DÉMONSTRATION" if MODE_DEMO else ""))
plt.show()
````

 <div style="color:#00704A;font-weight:700;font-size:13.5px;">&#128161;&nbsp; Quelle variante choisir ?</div>

**Le dernier point de structure.** Les trois produits ne diffèrent pas seulement par leur
période de référence. Ils n’ont pas le même chemin de groupe, pas les mêmes noms de couches, pas la même
couche de qualité, et pas le même codage de cette qualité. Un script écrit pour A4 qui reçoit un A2 ne
plante pas toujours : parfois il tourne, et produit des chiffres faux.

La cellule ci-dessous ouvre un fichier de chaque produit disponible et dresse le comparatif.

````python
# =============================================================================
#  ÉTAPE 16 — COMPARER LES PRODUITS ENTRE EUX
# =============================================================================
lignes = []
apercus = []

for produit in ["VNP46A4", "VNP46A3", "VNP46A2", "VNP46A1"]:
    sous = INVENTAIRE[INVENTAIRE["produit"] == produit]
    if sous.empty:
        continue
    chemin = Path(sous.iloc[-1]["chemin"])
    with h5py.File(chemin, "r") as f:
        chemin_groupe = trouver_groupe_donnees(f)
        groupe = f[chemin_groupe]
        jeux = {n: o for n, o in groupe.items() if isinstance(o, h5py.Dataset)}
        couche = choisir_couche_radiance(groupe)
        jeu = groupe[couche]
        nom_drapeau = next((n for n in (f"{couche}_Quality", "Mandatory_Quality_Flag")
                            if n in groupe), "—")
        lignes.append({
            "produit": produit,
            "famille": FAMILLE[produit],
            "groupe HDF5": chemin_groupe.replace("HDFEOS/GRIDS/", "…/").replace("/Data Fields", ""),
            "couches": len(jeux),
            "couche de radiance": couche,
            "type": str(jeu.dtype),
            "scale_factor": lire_attribut(jeu, "scale_factor", ""),
            "_FillValue": lire_attribut(jeu, "_FillValue", ""),
            "couche qualité": nom_drapeau,
            "forme": "×".join(str(d) for d in jeu.shape),
        })
    try:
        apercus.append((produit, lire_ntl(chemin, sous_echantillonnage=SOUS_ECHANTILLONNAGE)))
    except Exception as erreur:
        print(f"  {produit} : lecture impossible ({erreur})")

COMPARATIF = pd.DataFrame(lignes).set_index("produit")
print("COMPARATIF STRUCTUREL DES PRODUITS PRÉSENTS DANS VOS DONNÉES\n")
display(COMPARATIF.T)

if len(apercus) > 1:
    print("\nCE QUE CHAQUE PRODUIT DONNE SUR LA MÊME ZONE\n")
    resume = []
    for produit, res in apercus:
        v = res["radiance"][np.isfinite(res["radiance"])]
        resume.append({"produit": produit, "date": res["metadonnees"]["date"],
                       "tuile": res["metadonnees"]["tuile"],
                       "pixels exploitables": f"{v.size:,}".replace(",", " "),
                       "couverture_%": round(100 * v.size / res["radiance"].size, 1),
                       "médiane": round(float(np.median(v)), 3) if v.size else np.nan,
                       "SoL": f"{v.sum():,.0f}".replace(",", " ") if v.size else "—"})
    display(pd.DataFrame(resume).set_index("produit"))
````

````python
# --- Figure : la même zone vue par chaque produit ----------------------------
if apercus:
    fig, axes = plt.subplots(1, len(apercus), figsize=(4.5 * len(apercus), 4.9))
    fig.subplots_adjust(top=0.72, bottom=0.165, left=0.055, right=0.98, wspace=0.22)
    axes = np.atleast_1d(axes)

    plafond_commun = max(
        float(np.nanpercentile(np.log1p(r["radiance"][np.isfinite(r["radiance"])]), 99.5))
        for _, r in apercus if np.isfinite(r["radiance"]).any())

    for ax, (produit, res) in zip(axes, apercus):
        donnees = res["radiance"]
        image = np.where(np.isfinite(donnees), np.log1p(donnees), np.nan)
        ax.imshow(image, cmap=CMAP_NUIT, vmin=0, vmax=plafond_commun,
                  extent=res["emprise"], origin="upper", aspect="auto",
                  interpolation="nearest")
        ax.set_facecolor("#DCE4E0")
        couverture = 100 * float(np.isfinite(donnees).mean())
        ax.set_title(f"{produit} — {FAMILLE[produit]}\n{res['metadonnees']['date']}   "
                     f"·   {couverture:.0f} % de pixels exploitables",
                     pad=7, fontsize=10)
        ax.set_xlabel("longitude (°)")
    axes[0].set_ylabel("latitude (°)")

    titre_figure(fig, "Le même endroit, selon le produit choisi",
                 "Échelle de couleurs identique pour les trois. Le journalier montre plus de trous (nuages, lune) ;\n"
                 "l'annuel est le plus complet — au prix de toute information sur la date.",
                 "Toutes les cartes en log1p, mêmes bornes."
                 + ("   ·   MODE DÉMONSTRATION — valeurs simulées" if MODE_DEMO else ""))
    plt.show()
````

 <div style="color:#D49A00;font-size:11px;font-weight:700;letter-spacing:2.5px;text-transform:uppercase;">
  ÉTAPE 18 · 20 &nbsp;&nbsp;<span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span><span style="display:inline-block;width:7px;height:7px;border-radius:50%;margin-left:4px;background:#D5DED9;"></span></div>
 <div style="color:#231F20;font-size:14px;margin-top:4px;">Sortir du notebook vers un SIG, un tableur ou un dépôt</div>
</div>

**Pourquoi exporter.** Le notebook est un lieu d’exploration, pas un lieu de stockage. Le
produit de votre travail doit pouvoir être ouvert par un collègue sous QGIS, joint à un rapport, ou versé
dans un dépôt GitHub. Trois formats couvrent l’essentiel des besoins :

- **GeoTIFF** — le format universel des SIG. Il embarque le géoréférencement : QGIS et ArcGIS le placent
  au bon endroit sans aucun réglage. C’est le format à privilégier pour tout partage.
- **CSV** — un pixel par ligne, avec sa longitude, sa latitude et sa radiance. Lourd, mais compréhensible
  par tout le monde et joignable à des données administratives dans un tableur.
- **NPZ** — l’archive compressée de NumPy : rapide à relire en Python, idéale comme cache intermédiaire
  entre deux étapes d’un traitement.

Si `rasterio` n’est pas installé, la cellule produit tout de même un fichier `.tif` accompagné d’un
**fichier de calage** (`.tfw`) et d’un `.prj` : la plupart des SIG le géoréférenceront correctement.

````python
# =============================================================================
#  ÉTAPE 17 — EXPORTER
# =============================================================================
DOSSIER_SORTIES = Path.cwd() / "sorties_ntl"
DOSSIER_SORTIES.mkdir(exist_ok=True)

meta = RESULTAT["metadonnees"]
base = f"{meta['produit']}_{meta['tuile']}_{meta['date']}"
if MODE_DEMO:
    base = "DEMO_" + base
exportes = []

# --- 1. GeoTIFF ---------------------------------------------------------------
chemin_tif = DOSSIER_SORTIES / f"{base}_radiance.tif"
res_lon, _, ouest_e, _, res_lat_neg, nord_e = RESULTAT["transformation"]

if DISPO.get("rasterio"):
    import rasterio
    from rasterio.transform import Affine

    transformation = Affine(res_lon, 0.0, ouest_e, 0.0, res_lat_neg, nord_e)
    with rasterio.open(
            chemin_tif, "w", driver="GTiff",
            height=radiance.shape[0], width=radiance.shape[1], count=1,
            dtype="float32", crs="EPSG:4326", transform=transformation,
            nodata=np.nan, compress="deflate", tiled=True) as sortie:
        sortie.write(radiance.astype("float32"), 1)
        sortie.update_tags(**{k: str(v) for k, v in meta.items()})
    exportes.append(("GeoTIFF géoréférencé (rasterio)", chemin_tif))
else:
    # Repli sans rasterio : TIFF simple + fichier de calage + projection
    try:
        from PIL import Image
        Image.fromarray(radiance.astype("float32")).save(chemin_tif)
        (DOSSIER_SORTIES / f"{base}_radiance.tfw").write_text(
            f"{res_lon:.12f}\n0.0\n0.0\n{res_lat_neg:.12f}\n"
            f"{ouest_e + res_lon / 2:.12f}\n{nord_e + res_lat_neg / 2:.12f}\n")
        (DOSSIER_SORTIES / f"{base}_radiance.prj").write_text(
            'GEOGCS["WGS 84",DATUM["WGS_1984",SPHEROID["WGS 84",6378137,298.257223563]],'
            'PRIMEM["Greenwich",0],UNIT["degree",0.0174532925199433]]')
        exportes.append(("TIFF + calage .tfw + .prj (sans rasterio)", chemin_tif))
    except Exception as erreur:
        print(f"  export TIFF impossible : {erreur}")

# --- 2. CSV : un pixel par ligne ---------------------------------------------
pas = max(radiance.shape[0] // 220, 1)          # allégé pour rester lisible
sous_radiance = radiance[::pas, ::pas]
LON, LAT = np.meshgrid(RESULTAT["longitudes"][::pas], RESULTAT["latitudes"][::pas])
table_pixels = pd.DataFrame({
    "longitude": np.round(LON.ravel(), 6),
    "latitude": np.round(LAT.ravel(), 6),
    "radiance": np.round(sous_radiance.ravel(), 4),
}).dropna(subset=["radiance"])

chemin_csv = DOSSIER_SORTIES / f"{base}_pixels.csv"
table_pixels.to_csv(chemin_csv, index=False, encoding="utf-8")
exportes.append((f"CSV — {len(table_pixels):,} pixels (1 sur {pas})".replace(",", " "), chemin_csv))

# --- 3. Archive NumPy ---------------------------------------------------------
chemin_npz = DOSSIER_SORTIES / f"{base}_radiance.npz"
np.savez_compressed(chemin_npz, radiance=radiance.astype("float32"),
                    longitudes=RESULTAT["longitudes"], latitudes=RESULTAT["latitudes"],
                    emprise=np.array(RESULTAT["emprise"]))
exportes.append(("Archive NumPy compressée", chemin_npz))

# --- 4. Métadonnées lisibles --------------------------------------------------
chemin_json = DOSSIER_SORTIES / f"{base}_metadonnees.json"
documentation = {
    "source": "NASA Black Marble — suite VNP46, collection 2 (version 002)",
    "avertissement": ("VALEURS SIMULÉES — jeu de démonstration, ne pas utiliser pour produire "
                      "un chiffre" if MODE_DEMO else "Données NASA réelles"),
    "traitement": {k: str(v) for k, v in meta.items()},
    "emprise_WGS84": {"ouest": RESULTAT["emprise"][0], "est": RESULTAT["emprise"][1],
                      "sud": RESULTAT["emprise"][2], "nord": RESULTAT["emprise"][3]},
    "systeme_de_coordonnees": "EPSG:4326 (WGS 84)",
    "transformation_affine": list(RESULTAT["transformation"]),
    "dimensions": {"lignes": int(radiance.shape[0]), "colonnes": int(radiance.shape[1])},
    "sous_echantillonnage": SOUS_ECHANTILLONNAGE,
    "journal_de_traitement": RESULTAT["journal"].to_dict(orient="records"),
    "citation": ("Román, M. O. et al. (2018). NASA's Black Marble nighttime lights product "
                 "suite. Remote Sensing of Environment, 210, 113–143."),
}
chemin_json.write_text(json.dumps(documentation, indent=2, ensure_ascii=False,
                                  default=str), encoding="utf-8")
exportes.append(("Métadonnées JSON", chemin_json))

# --- Restitution --------------------------------------------------------------
print("FICHIERS ÉCRITS")
print("=" * 78)
for etiquette, chemin in exportes:
    print(f"  {etiquette}")
    print(f"     {chemin}   ({chemin.stat().st_size / 1e3:.0f} Ko)")
print("=" * 78)
print(f"\nDossier de sortie : {DOSSIER_SORTIES}")
if not DISPO.get("rasterio"):
    print("\n  (rasterio absent : le GeoTIFF a été produit avec un fichier de calage .tfw.")
    print("   Pour un GeoTIFF pleinement conforme :  pip install rasterio)")

print("\nAPERÇU DU CSV")
display(table_pixels.head(8))
````

 padding:13px 17px;margin:10px 0;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
 <div style="color:#D49A00;font-weight:700;font-size:13.5px;">&#9888;&nbsp; Ce qu’il faut toujours joindre à un export</div>
 <div style="color:#231F20;font-size:13.5px;margin-top:5px;line-height:1.6;">Le fichier <code>…_metadonnees.json</code> n’est pas décoratif. **Le moment de vérité.** Depuis l’étape 09, chaque correction a été enregistrée dans `FIL_ROUGE` avec la valeur de l’indicateur à cet instant. Les voici toutes ensemble.

Lisez ce tableau comme un **auditeur** le lirait : chacune de ces lignes est un chiffre qu’un analyste aurait pu publier de bonne foi, en pensant avoir fait le travail. Une seule est défendable. Les autres ne diffèrent pas par des arrondis — elles diffèrent par des facteurs.

````python
# =============================================================================
#  ÉTAPE 19 — LE FIL ROUGE : CE QUE CHAQUE CORRECTION A COÛTÉ
# =============================================================================
if not FIL_ROUGE:
    print("Aucun point de contrôle enregistré : exécutez les étapes 09 à 14 d'abord.")
else:
    print(f"LA MÊME QUESTION, POSÉE {len(FIL_ROUGE)} FOIS — {NOM_PAYS}\n")
    display(tableau_fil_rouge())

    valeurs = [e["somme des lumières"] for e in FIL_ROUGE]
    libelles = [f"{e['étape']:02d} · {e['traitement']}" for e in FIL_ROUGE]
    perimetres = [e["périmètre"] for e in FIL_ROUGE]

    palette_p = [COULEURS["brique"], COULEURS["vert"], COULEURS["sarcelle"], COULEURS["or_"],
                 COULEURS["terre"]]
    couleur_par_perimetre = {}
    for p in perimetres:
        couleur_par_perimetre.setdefault(
            p, palette_p[len(couleur_par_perimetre) % len(palette_p)])

    hauteur = 0.46 * len(valeurs) + 2.5
    fig, ax = plt.subplots(figsize=(12.0, hauteur))
    fig.subplots_adjust(top=1 - 1.18 / hauteur, bottom=1.02 / hauteur,
                        left=0.295, right=0.955)

    y = np.arange(len(valeurs))[::-1]
    ax.barh(y, valeurs, height=0.62, edgecolor="none",
            color=[couleur_par_perimetre[p] for p in perimetres])
    reference = valeurs[-1]
    for yi, v in zip(y, valeurs):
        facteur = v / reference if reference else np.nan
        suffixe = f"   (× {facteur:,.1f})".replace(",", " ") if facteur >= 1.05 else ""
        ax.text(v * 1.06, yi, f"{v:,.0f}".replace(",", " ") + suffixe, va="center",
                ha="left", fontsize=8.8, color=COULEURS["encre"], fontweight="bold")

    ax.set_yticks(y)
    ax.set_yticklabels(libelles, fontsize=9)
    ax.set_xscale("log")
    ax.set_xlim(max(min(valeurs) * 0.45, 1), max(valeurs) * 6)
    ax.set_xlabel("somme des lumières, échelle logarithmique (nW·cm⁻²·sr⁻¹)")
    ax.spines["left"].set_visible(False)
    ax.tick_params(left=False)
    ax.legend(handles=[plt.Rectangle((0, 0), 1, 1, fc=c, ec="none", label=p)
                       for p, c in couleur_par_perimetre.items()],
              loc="lower right", fontsize=8.2, frameon=True, framealpha=0.95,
              facecolor="white", edgecolor=COULEURS["sauge"], title="périmètre",
              title_fontsize=8.5)

    titre_figure(fig, f"Le même territoire, {len(set(np.round(valeurs, 3)))} chiffres différents",
                 "Chaque barre est une réponse défendable à la même question — et une seule est juste.\n"
                 "L'écart entre la première et la dernière n'est pas une marge d'erreur : c'est un ordre de grandeur.",
                 "Points de contrôle enregistrés aux étapes 09 à 14 du notebook."
                 + ("   ·   MODE DÉMONSTRATION — valeurs simulées" if MODE_DEMO else ""))
    plt.show()

    naif, final = valeurs[0], valeurs[-1]
    print("\n" + "=" * 78)
    print(f"  Première réponse, sans aucune précaution  : {naif:>18,.0f}".replace(",", " "))
    print(f"  Réponse finale, toutes corrections faites : {final:>18,.0f}".replace(",", " "))
    if final:
        print(f"  Rapport entre les deux                    : × {naif / final:,.1f}"
              .replace(",", " "))
    print("=" * 78)
    print("  Aucune de ces étapes n'est facultative, et aucune ne produit de message")
    print("  d'erreur. C'est précisément ce qui les rend dangereuses.")
````

 <div style="color:#D49A00;font-weight:700;font-size:13.5px;">&#9888;&nbsp; Ce que cette figure doit vous faire changer dans votre pratique</div>

  <div style="color:#231F20;font-size:13px;line-height:1.55;">Sans quoi un pays côtier et un pays enclavé ne sont plus comparables. <i>(étape 12)</i></div>
 </div></td><td style="width:33.3333%;vertical-align:top;padding:5px;">
 <div style="background:#F4F7F5;border:1px solid #D5DED9;border-top:3px solid #00A86A;border-radius:10px;
  padding:13px 15px;height:100%;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
  <div style="color:#00704A;font-weight:700;font-size:14px;margin-bottom:5px;">⑧ Le nombre d’observations a-t-il été regardé ?</div>
  <div style="color:#231F20;font-size:13px;line-height:1.55;">Une radiance calculée sur 2 nuits ne vaut pas une radiance calculée sur 25. <i>(étape 15)</i></div>
 </div></td><td style="width:33.3333%;vertical-align:top;padding:5px;">
 <div style="background:#F4F7F5;border:1px solid #D5DED9;border-top:3px solid #00A86A;border-radius:10px;
  padding:13px 15px;height:100%;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
  <div style="color:#00704A;font-weight:700;font-size:14px;margin-bottom:5px;">⑨ Le pixel le plus lumineux est-il plausible ?</div>
  <div style="color:#231F20;font-size:13px;line-height:1.55;">Une capitale, oui. Le désert ou la haute mer : cherchez la torchère, le feu ou le bogue. <i>(étape 14)</i></div>
 </div></td></tr><tr><td style="width:33.3333%;vertical-align:top;padding:5px;">
 <div style="background:#F4F7F5;border:1px solid #D5DED9;border-top:3px solid #00A86A;border-radius:10px;

  <div style="color:#00704A;font-weight:700;font-size:14px;margin-bottom:5px;">⑩ La même variante partout ?</div>
  <div style="color:#231F20;font-size:13px;line-height:1.55;"><code>AllAngle_Composite_Snow_Free</code> d’un bout à l’autre de la série. Changer de couche fabrique une fausse rupture. <i>(étape 15)</i></div>
 </div></td><td style="width:33.3333%;vertical-align:top;padding:5px;">
 <div style="background:#F4F7F5;border:1px solid #D5DED9;border-top:3px solid #00A86A;border-radius:10px;

  <div style="color:#00704A;font-weight:700;font-size:14px;margin-bottom:5px;">⑪ Le bon produit pour la bonne question ?</div>
  <div style="color:#231F20;font-size:13px;line-height:1.55;">Tendance → A4. Saison → A3. Événement daté → A2. Jamais l’inverse. <i>(étape 04)</i></div>
 </div></td><td style="width:33.3333%;vertical-align:top;padding:5px;">
 <div style="background:#F4F7F5;border:1px solid #D5DED9;border-top:3px solid #00A86A;border-radius:10px;

  <div style="color:#00704A;font-weight:700;font-size:14px;margin-bottom:5px;">⑫ Les limitations sont-elles écrites ?</div>
  <div style="color:#231F20;font-size:13px;line-height:1.55;">Halo urbain, saturation, torchères, nuages, rupture de capteur : ce paragraphe fait partie du livrable, pas de l’annexe.</div>
 </div></td></tr></table>

<div style="background:#FBEDEA;border:1px solid #B83B2E44;border-left:4px solid #B83B2E;border-radius:8px;
 padding:13px 17px;margin:10px 0;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
 <div style="color:#B83B2E;font-weight:700;font-size:13.5px;">&#9940;&nbsp; Les six artefacts à connaître avant de conclure quoi que ce soit</div>
 <div style="color:#231F20;font-size:13.5px;margin-top:5px;line-height:1.6;"><b>Halo urbain (<i>blooming</i>)</b> — la lumière déborde au-delà du bâti et gonfle les surfaces « éclairées ». &nbsp;<b>Saturation</b> — les cœurs urbains les plus intenses plafonnent, et une croissance y devient invisible. &nbsp;<b>Torchères de gaz</b> — un site pétrolier peut dominer la radiance d’une région entière sans rapport avec son activité économique. &nbsp;<b>Saisonnalité</b> — nuages et feux de brousse créent des cycles qui ne sont pas économiques. &nbsp;<b>Rupture de capteur</b> — l’arrivée de NOAA-20 aux côtés de Suomi-NPP peut introduire un décrochage dans une série longue. &nbsp;<b>Bruit en zone rurale peu éclairée</b> — près du seuil de détection, le rapport signal/bruit s’effondre : c’est précisément là que vivent les populations que l’indicateur prétend décrire.</div>
</div>
