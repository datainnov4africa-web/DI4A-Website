### 🎛️ Settings

**Express track:** only the first section matters. Keep `DOCUMENTS = ["example"]` for a first run, then replace it with your own files.

`DOCUMENTS` accepts a **list** mixing:
- `"example"` — three fictitious reports provided with the notebook;
- **files** — `"reports/labour_2025.pdf"`;
- **folders** — `"reports/"` (every PDF inside);
- **web addresses** — `"https://…/report.pdf"`.

`DOCUMENT_LANGUAGE = "auto"` detects the language of **each document** (and, in bilingual reports, of each passage): a collection can mix English, French, Spanish, Portuguese and other languages. Force one code (`"en"`, `"fr"`…) only if detection goes wrong. `DOCUMENTS = ["example-multilingual"]` adds a French note and a Spanish survey to the three English examples.

**Questions can be asked in any language.** Section **1b** of the settings decides how: `QUESTION_LANGUAGE` (detected by default), `ANSWER_LANGUAGE` (by default the language of the question), and `TRANSLATE_QUERIES` (with a model, the question is also searched in each language of the collection). Part **B10** explains the mechanism and measures it.

**Long institutional reports** (50 to 200 pages) are the normal case for a statistics office. Three settings matter for them:

| Setting | Effect |
|---|---|
| `USE_CACHE = True` | Extraction and embeddings are saved in `rag_cache`. The first run of an 80-page report costs a few seconds of table detection; every later run in the same session — or after a restart — reuses the cache. Delete the folder to force a fresh extraction. |
| `DROP_TABLE_OF_CONTENTS = True` | Removes "Chapter 3 …… 42" lines, lists of figures and stray page numbers, which otherwise become passages that answer nothing. |
| `MAX_PAGES_PER_DOCUMENT` | `0` reads everything. Set it to 30 or 40 to try a very long report quickly before indexing it in full. |

> ⚠️ **Web addresses do not always work.** Many institutional portals — `afdb.org` among them — refuse downloads that do not come from a browser and answer **HTTP 403**. The notebook tries browser-like headers, then `curl`, and if both fail it **skips that document and carries on with the others**, telling you which one failed.
>

````python
# @title 🦙 Install Ollama (if requested) { display-mode: "form" }
def _ollama_running():
    try:
        urllib.request.urlopen("http://localhost:11434/api/tags", timeout=2)
        return True
    except Exception:
        return False

def install_ollama(model):
    if _ollama_running():
        print("Ollama is already running.")
    elif platform.system() != "Linux":
        callout("Automatic installation is only available on Linux (Colab, Kaggle). On macOS or Windows, "
                "install Ollama from <code>ollama.com/download</code>, then run this cell again.", "warning")
        return False
    else:
        if not shutil.which("ollama"):
            print("1/3 · Installing Ollama (1–2 min)…")
            if shutil.which("apt-get"):
                subprocess.run("apt-get install -y -qq zstd pciutils > /dev/null 2>&1 || "
                               "(apt-get update -qq > /dev/null 2>&1 && apt-get install -y -qq zstd pciutils > /dev/null 2>&1)",
                               shell=True)
            subprocess.run("curl -fsSL https://ollama.com/install.sh | sh > ollama_install.log 2>&1", shell=True)
        if not shutil.which("ollama"):
            callout("Installation failed: see the file <code>ollama_install.log</code>.", "error")
            return False
        print("2/3 · Starting the service…")
        subprocess.Popen(["ollama", "serve"], stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL,
                         start_new_session=True)
        for _ in range(30):
            if _ollama_running():
                break
            time.sleep(1)
    print(f"3/3 · Downloading model {model} (a few minutes the first time)…")
    result = subprocess.run(["ollama", "pull", model], capture_output=True, text=True)
    if result.returncode != 0:
        callout(f"Could not download the model: <code>{_html.escape(result.stderr[-300:])}</code>", "error")
        return False
    callout(f"Ollama is ready with model <b>{model}</b> · unlimited runs."
            + ("" if ENV["gpu"] else " <br>⚠️ Without a GPU, each answer may take 10 to 60 seconds."), "success")
    return True

if INSTALL_OLLAMA:
    install_ollama(OLLAMA_MODEL)
else:
    print("Ollama installation not requested (INSTALL_OLLAMA = False).")
````

````python
# @title 🤖 Connect to the model { display-mode: "form" }
import urllib.error

def _secret(name):
    value = os.environ.get(name)
    if value:
        return value
    if ENV["name"] == "colab":
        try:
            from google.colab import userdata
            return userdata.get(name)
        except Exception:
            return None
    if ENV["name"] == "kaggle":
        try:
            from kaggle_secrets import UserSecretsClient
            return UserSecretsClient().get_secret(name)
        except Exception:
            return None
    return None

def _post(url, payload, headers, timeout=120):
    request = urllib.request.Request(url, data=json.dumps(payload).encode("utf-8"),
                                     headers={"Content-Type": "application/json", **headers})
    with urllib.request.urlopen(request, timeout=timeout) as response:
        return json.loads(response.read().decode("utf-8"))

class LLMClient:
    """One interface for Ollama, Groq, Anthropic or OpenAI.
    Temperature 0, quota-aware (pause between calls, retry after HTTP 429)."""

    MIN_PAUSE = {"groq": 2.2, "anthropic": 0.5, "openai": 0.5, "ollama": 0.0}

    def __init__(self, choice="auto"):
        self.found = {"ollama": _ollama_running(),
                      "groq": bool(_secret("GROQ_API_KEY")),
                      "anthropic": bool(_secret("ANTHROPIC_API_KEY")),
                      "openai": bool(_secret("OPENAI_API_KEY"))}
        order = ["ollama", "groq", "anthropic", "openai"] if choice == "auto" else [choice]
        self.name = next((p for p in order if self.found.get(p)), "none")
        self.model = {"ollama": OLLAMA_MODEL, "groq": GROQ_MODEL, "anthropic": ANTHROPIC_MODEL,
                      "openai": OPENAI_MODEL}.get(self.name, "—")
        self.calls, self.waited, self._last = 0, 0.0, 0.0

    @property
    def available(self):
        return self.name != "none"

    def _call(self, system, message, max_tokens):
        if self.name == "ollama":
            d = _post("http://localhost:11434/api/chat",
                      {"model": self.model, "stream": False, "options": {"temperature": 0},
                       "messages": [{"role": "system", "content": system},
                                    {"role": "user", "content": message}]}, {}, timeout=600)
            return d["message"]["content"]
        if self.name == "groq":
            payload = {"model": self.model, "temperature": 0, "max_completion_tokens": max_tokens + 600,
                       "messages": [{"role": "system", "content": system},
                                    {"role": "user", "content": message}]}
            if "gpt-oss" in self.model:
                payload["reasoning_effort"] = "low"          # short reasoning: fewer tokens used
            d = _post("https://api.groq.com/openai/v1/chat/completions", payload,
                      {"Authorization": f"Bearer {_secret('GROQ_API_KEY')}"})
            return d["choices"][0]["message"].get("content") or ""
        if self.name == "anthropic":
            d = _post("https://api.anthropic.com/v1/messages",
                      {"model": self.model, "max_tokens": max_tokens, "temperature": 0,
                       "system": system, "messages": [{"role": "user", "content": message}]},
                      {"x-api-key": _secret("ANTHROPIC_API_KEY"), "anthropic-version": "2023-06-01"})
            return "".join(b.get("text", "") for b in d["content"])
        if self.name == "openai":
            d = _post("https://api.openai.com/v1/chat/completions",
                      {"model": self.model, "temperature": 0, "max_completion_tokens": max_tokens,
                       "messages": [{"role": "system", "content": system},
                                    {"role": "user", "content": message}]},
                      {"Authorization": f"Bearer {_secret('OPENAI_API_KEY')}"})
            return d["choices"][0]["message"]["content"]
        raise RuntimeError("No model provider available.")

    def generate(self, system, message, max_tokens=400, attempts=5):
        for attempt in range(attempts):
            wait = self.MIN_PAUSE.get(self.name, 0) - (time.time() - self._last)
            if wait > 0:
                time.sleep(wait)
            try:
                self._last = time.time()
                text = self._call(system, message, max_tokens)
                self.calls += 1
                return re.sub(r"<think>.*?</think>", "", text, flags=re.S).strip()
            except urllib.error.HTTPError as error:
                if error.code == 429 and attempt < attempts - 1:
                    delay = min(float(error.headers.get("retry-after") or 2 ** (attempt + 2)), 90)
                    print(f"⏳ Rate limit reached ({self.name}): retrying in {delay:.0f} s…")
                    self.waited += delay
                    time.sleep(delay)
                    continue
                detail = error.read().decode("utf-8", "ignore")[:200]
                raise RuntimeError(f"HTTP {error.code}: {detail}") from error

# ── Where each key comes from, and what it looks like ──────────────────────────────────────
API_KEYS = {"groq":      ("GROQ_API_KEY",      "gsk_",    "console.groq.com/keys (free, no credit card)"),
            "anthropic": ("ANTHROPIC_API_KEY", "sk-ant-", "console.anthropic.com"),
            "openai":    ("OPENAI_API_KEY",    "sk-",     "platform.openai.com/api-keys")}

def ask_for_api_key(provider):
    """Ask for that provider's key, hidden as you type. Enter alone skips.
    The key stays in memory for this session: it is never written into the notebook."""
    variable, prefix, where = API_KEYS[provider]
    import getpass
    try:
        key = getpass.getpass(f"🔑 {provider.capitalize()} key ({prefix}…) — get one at {where} — "
                              f"press Enter alone to skip: ").strip()
    except Exception:
        return None                                  # headless run (nbconvert, papermill): skip
    if not key:
        callout("Skipped: the assistant will <b>quote</b> sentences instead of writing answers. "
                "You can paste a key later in the next cell.", "info")
        return None
    if not key.startswith(prefix):
        callout(f"A {provider.capitalize()} key starts with <code>{prefix}</code> — nothing was saved.", "error")
        return None
    os.environ[variable] = key
    return LLMClient(provider)

LLM = LLMClient("none" if PROVIDER == "none" else PROVIDER)

show_table(pd.DataFrame([
    {"Provider": "Ollama (local)", "Cost": "free · unlimited", "Detected": "✅" if LLM.found["ollama"] else "—",
     "How to enable": "INSTALL_OLLAMA = True, or Ollama installed locally"},
    {"Provider": "Groq", "Cost": "free · 1,000 req./day", "Detected": "✅" if LLM.found["groq"] else "—",
     "How to enable": "GROQ_API_KEY secret or variable"},
    {"Provider": "Anthropic", "Cost": "paid", "Detected": "✅" if LLM.found["anthropic"] else "—",
     "How to enable": "ANTHROPIC_API_KEY secret or variable"},
    {"Provider": "OpenAI", "Cost": "paid", "Detected": "✅" if LLM.found["openai"] else "—",
     "How to enable": "OPENAI_API_KEY secret or variable"},
]), title="Model providers detected")

# ── Asked for a provider but no key in sight: ask for it now, hidden ───────────────────────
if not LLM.available and PROVIDER != "none" and ASK_FOR_KEY:
    wanted = PROVIDER if PROVIDER in API_KEYS else "groq"
    callout(f"<b>{wanted.capitalize()}</b> was requested but no key was found in the secrets or the environment. "
            f"Paste it below — the typing is hidden, and the key is never written into the notebook.", "warning")
    LLM = ask_for_api_key(wanted) or LLM

if LLM.available:
    try:
        start = time.time()
        test = LLM.generate("You answer with a single word.", "Reply only: OK", max_tokens=20)
        callout(f"Provider selected: <b>{LLM.name}</b> · model <code>{LLM.model}</code> · "
                f"connection test passed in {time.time() - start:.1f} s (reply: “{_html.escape(test[:30])}”).", "success")
    except Exception as error:
        callout(f"Provider <b>{LLM.name}</b> was detected but the test failed: "
                f"<code>{_html.escape(str(error)[:220])}</code><br>The notebook switches to <b>extractive mode</b>. "
                "Check the key, the model name and Internet access.", "error")
        LLM.name, LLM.model = "none", "—"
else:
    callout("No model connected: the notebook runs in <b>extractive mode</b> — every step works, but answers are "
            "<b>quoted</b> from the documents instead of being written, and the assistant cannot count, compare "
            "or summarise. The next cell lets you paste a key at any time.", "warning")
````

````python
# @title ⚙️ T2 · Diagnosis and structured extraction { display-mode: "form" }
from dataclasses import dataclass, field

def diagnose(doc):
    rows = []
    for page in doc:
        chars = len(page.get_text().strip())
        images = len(page.get_images(full=True))
        status = "text" if chars >= 50 else ("scanned" if images else "empty")
        rows.append({"Page": page.number + 1, "Characters": chars, "Images": images, "Status": status})
    return pd.DataFrame(rows)

@dataclass
class Element:
    type: str           # "heading", "paragraph", "table" or "ocr"
    text: str
    page: int
    box: tuple          # rectangle (x0, y0, x1, y1) on the page
    level: int = 0      # heading level (1 = most important)
    section: str = ""   # nearest heading
    path: str = ""      # full heading breadcrumb, e.g. "2. Strategy › 2.4 Focus Areas"

FOOTNOTE_MARKER = re.compile(r"(?<=[a-z\)\u201d\"])\.(\d{1,3})(?=\s+[A-Z\u201c\"])")

def _nfkc(t):
    return unicodedata.normalize("NFKC", t)

def page_lines(page):
    lines = []
    for block in page.get_text("dict")["blocks"]:
        if block.get("type") != 0:
            continue
        for line in block["lines"]:
            spans = [s for s in line["spans"] if s["text"].strip()]
            if not spans:
                continue
            lines.append({"text": _nfkc("".join(s["text"] for s in line["spans"])).strip(),
                          "size": round(max(s["size"] for s in spans), 1),
                          "bold": any((s["flags"] & 16) or "Bold" in s.get("font", "") for s in spans),
                          "box": pymupdf.Rect(line["bbox"]), "block": block["number"]})
    return lines

def _signature(text):
    return re.sub(r"\d+", "#", text.lower()).strip()

def _tables(page):
    if not EXTRACT_TABLES:
        return []
    with contextlib.redirect_stdout(io.StringIO()):
        found = page.find_tables().tables
    result = []
    for t in found:
        rows = [[_nfkc((c or "").replace("\n", " ")).strip() for c in r] for r in t.extract()]
        headers = [_nfkc((c or "").replace("\n", " ")).strip() for c in t.header.names]
        if rows and rows[0] == headers:
            rows = rows[1:]
        if rows and len(headers) >= 2:
            result.append({"box": pymupdf.Rect(t.bbox), "headers": headers, "rows": rows})
    return result

def linearise(table, caption):
    sentences = []
    for row in table["rows"]:
        if not any(row):
            continue
        values = "; ".join(f"{table['headers'][j]}: {row[j]}" for j in range(1, len(row))
                           if j < len(table["headers"]) and row[j])
        sentences.append(f"{row[0]} — {values}.")
    return (f"{caption}\n" if caption else "") + "\n".join(sentences)

def ocr_page(page):
    ensure_package("pytesseract", "pytesseract")
    if not shutil.which("tesseract") and shutil.which("apt-get") and ENV["name"] in ("colab", "kaggle"):
        print("Installing Tesseract (about a minute)…")
        subprocess.run("apt-get install -y -qq tesseract-ocr tesseract-ocr-fra > /dev/null 2>&1 || "
                       "(apt-get update -qq > /dev/null 2>&1 && apt-get install -y -qq tesseract-ocr tesseract-ocr-fra > /dev/null 2>&1)",
                       shell=True)
    import pytesseract
    from PIL import Image
    installed = set(pytesseract.get_languages(config=""))
    missing = [l for l in OCR_LANGUAGE.split("+") if l not in installed]
    if missing and shutil.which("apt-get") and ENV["name"] in ("colab", "kaggle"):
        packs = " ".join(f"tesseract-ocr-{l}" for l in missing)
        print(f"Installing OCR languages: {', '.join(missing)}…")
        subprocess.run(f"apt-get install -y -qq {packs} > /dev/null 2>&1 || (apt-get update -qq > /dev/null 2>&1 "
                       f"&& apt-get install -y -qq {packs} > /dev/null 2>&1)", shell=True)
        installed = set(pytesseract.get_languages(config=""))
    language = "+".join(l for l in OCR_LANGUAGE.split("+") if l in installed) or "eng"
    if language != OCR_LANGUAGE:
        print(f"⚠️ OCR language “{OCR_LANGUAGE}” unavailable: using “{language}”.")
    pix = page.get_pixmap(dpi=300)
    image = Image.frombytes("RGB", (pix.width, pix.height), pix.samples)
    text = _nfkc(pytesseract.image_to_string(image, lang=language))
    return [re.sub(r"\s*\n\s*", " ", p).strip() for p in re.split(r"\n\s*\n", text) if len(p.strip()) > 3]

CAPTION = re.compile(r"^(Table|Figure|Chart|Graph|Tableau|Graphique)\b")
LEADER_LINE = re.compile(r"\.{4,}\s*\d{1,4}\s*$|\u2026{2,}")      # "Chapter 3 ......... 42"
PAGE_NUMBER_ONLY = re.compile(r"^(page\s*)?[ivxlcdm\d]{1,6}$", re.I)
NUMBERED_HEADING = re.compile(r"^\d{1,2}(\.\d{1,2}){0,3}[.)]?\s+\S")

def extract_structure(doc):
    last = doc.page_count if not MAX_PAGES_PER_DOCUMENT else min(doc.page_count, MAX_PAGES_PER_DOCUMENT)
    pages = {p.number + 1: page_lines(p) for p in doc if p.number < last}
    with_text = [n for n, l in pages.items() if l]
    # 1. repeated running headers and footers
    repeated = set()
    if REMOVE_HEADERS and len(with_text) >= 2:
        counts = Counter()
        for n in with_text:
            height = doc[n - 1].rect.height
            counts.update({_signature(l["text"]) for l in pages[n]
                           if l["box"].y1 < height * 0.1 or l["box"].y0 > height * 0.9})
        repeated = {s for s, c in counts.items() if c >= max(2, 0.5 * len(with_text))}
    # 2. body text size and heading sizes
    weights = Counter()
    for l in (l for n in with_text for l in pages[n]):
        weights[l["size"]] += len(l["text"])
    body = weights.most_common(1)[0][0] if weights else 10
    heading_sizes = sorted({l["size"] for n in with_text for l in pages[n]
                            if l["size"] >= body * 1.2 and len(l["text"]) <= 120}, reverse=True)
    level_of = {s: i + 1 for i, s in enumerate(heading_sizes)}
    def is_running_line(line, page):
        if _signature(line["text"]) in repeated:
            return True
        # single-page documents: nothing repeats, so use position and font size instead
        height = page.rect.height
        in_margin = line["box"].y1 < height * 0.08 or line["box"].y0 > height * 0.92
        return REMOVE_HEADERS and len(with_text) < 2 and in_margin and line["size"] < body * 0.85 and len(line["text"]) < 90
    elements, removed, dropped_toc = [], 0, 0
    for n in sorted(pages):
        page = doc[n - 1]
        if not pages[n]:
            if USE_OCR and page.get_images():
                for paragraph in ocr_page(page):
                    elements.append(Element("ocr", paragraph, n, tuple(page.rect)))
            continue
        tables = _tables(page)
        pieces, current = [], None                     # pieces: (y, element, block[, table])
        for l in pages[n]:
            if is_running_line(l, page) or PAGE_NUMBER_ONLY.match(l["text"]):
                removed += 1
                continue
            if DROP_TABLE_OF_CONTENTS and LEADER_LINE.search(l["text"]):
                dropped_toc += 1                      # table of contents, list of figures…
                continue
            if any(l["box"].intersects(t["box"]) for t in tables):
                continue
            numbered = (NUMBERED_HEADING.match(l["text"]) and len(l["text"]) <= 90
                        and l["size"] >= body * 0.98 and l["bold"])
            if l["size"] in level_of or numbered:
                if current: pieces.append(current); current = None
                if pieces and pieces[-1][1].type == "heading" and pieces[-1][2] == l["block"]:
                    heading = pieces[-1][1]
                    heading.text += " " + l["text"]
                    heading.box = tuple(pymupdf.Rect(heading.box) | l["box"])
                else:
                    pieces.append((l["box"].y0, Element("heading", l["text"], n, tuple(l["box"]),
                                                        level_of.get(l["size"], len(heading_sizes) + 1)), l["block"]))
                continue
            if current and current[2] == l["block"]:
                el = current[1]
                if el.text.endswith("-") and l["text"][:1].islower():
                    el.text = el.text[:-1] + l["text"]          # re-join hyphenated words
                else:
                    el.text += " " + l["text"]
                el.box = tuple(pymupdf.Rect(el.box) | l["box"])
            else:
                if current: pieces.append(current)
                current = (l["box"].y0, Element("paragraph", l["text"], n, tuple(l["box"])), l["block"])
        if current: pieces.append(current)
        for t in tables:
            pieces.append((t["box"].y0, Element("table", "", n, tuple(t["box"])), None, t))
        pieces.sort(key=lambda p: p[0])
        for p in pieces:
            el = p[1]
            if el.type == "table":
                previous = elements[-1] if elements else None
                table = p[3]
                if (previous is not None and previous.type == "table" and previous.page == n - 1
                        and len(previous.table["headers"]) == len(table["headers"])):
                    # continuation of the previous page's table: its "header" row is really a data row
                    previous.table["rows"] += [table["headers"]] + table["rows"]
                    previous.text = linearise(previous.table, previous.caption)
                    previous.extra.append((n, tuple(table["box"])))
                    continue
                caption = ""
                if previous and previous.type == "paragraph" and CAPTION.match(previous.text):
                    caption = previous.text
                    elements.pop()
                el.text = linearise(table, caption)
                el.table, el.caption, el.extra = table, caption, []
            elements.append(el)
    # 3. attach every element to its section and to its full heading path
    section, stack = "", {}
    for i, el in enumerate(elements):
        if el.type == "heading":
            document_title = i == 0 and el.level == 1 and len(heading_sizes) > 1
            section = "" if document_title else el.text
            if not document_title:
                stack = {lvl: t for lvl, t in stack.items() if lvl < el.level}
                stack[el.level] = el.text
        else:
            el.text = FOOTNOTE_MARKER.sub(". ", el.text)      # "…is high.5 The AI industry" → "…is high. The AI industry"
        el.section = section if el.type != "ocr" else f"Page {el.page} (OCR text)"
        el.path = " › ".join(stack[lvl] for lvl in sorted(stack)) or el.section
    return elements, {"headers removed": removed, "contents lines dropped": dropped_toc,
                      "body size": body, "heading sizes": heading_sizes, "pages read": len(pages)}
print("✅ T2 ready.")
````

````python
# @title ⚙️ T3 · Splitting into passages { display-mode: "form" }
@dataclass
class Passage:
    id: int
    doc_id: int
    doc: str             # short document title
    section: str
    text: str
    index_text: str      # text used for search (with a contextual header)
    boxes: list          # [(page, rectangle), …]
    kind: str = "text"

    @property
    def pages(self):
        return sorted({p for p, _ in self.boxes})

    path: str = ""       # heading breadcrumb of the passage
    parent_id: int = -1  # context passage this search window belongs to
    lang: str = ""       # language code of the passage ("en", "fr"…), set at indexing

    @property
    def pages_label(self):
        pages = self.pages
        return f"p. {pages[0]}" if len(pages) == 1 else f"pp. {pages[0]}–{pages[-1]}"

    @property
    def citation(self):
        return f"{self.doc} · {self.section or 'start of document'} · {self.pages_label}"

def _sentences(text):
    return [s.strip() for s in re.split(r"(?<=[.!?;؟])\s+(?=[A-ZÀ-Ý0-9«•–¿¡\"\u0400-\u042F\u0600-\u06FF-])", text) if s.strip()]

def split_passages(document, start=0, max_chars=1000, overlap=2, min_chars=280):
    passages = []
    def add(section, units, kind="text"):
        text = ("\n" if kind == "table" else " ").join(u[0] for u in units)
        boxes = list(dict.fromkeys((u[1], u[2]) for u in units))
        if len(text.strip()) < 40 or text.strip().lower() in seen:
            return                                   # scraps and duplicates add noise, never answers
        seen.add(text.strip().lower())
        breadcrumb = path or section
        # a short passage is glued to the previous one rather than left alone without context
        if (passages and len(text) < min_chars and passages[-1].kind == kind == "text"
                and passages[-1].section == section and len(passages[-1].text) + len(text) <= max_chars * 1.4):
            previous = passages[-1]
            previous.text += " " + text
            previous.index_text += " " + text
            previous.boxes += [b for b in boxes if b not in previous.boxes]
            return
        header = f"{document.title} — {breadcrumb}" if breadcrumb else document.title
        passages.append(Passage(start + len(passages), document.id, document.short, section, text,
                                f"{header}. {text}", boxes, kind, breadcrumb))
    seen = set()
    buffer, section, path = [], None, ""
    def flush():
        nonlocal buffer
        i = 0
        while i < len(buffer):
            batch, length, j = [], 0, i
            while j < len(buffer) and (not batch or length + len(buffer[j][0]) <= max_chars):
                batch.append(buffer[j]); length += len(buffer[j][0]) + 1; j += 1
            add(section, batch)
            if j >= len(buffer):
                break
            i = max(j - overlap, i + 1)
        buffer = []
    for el in document.elements:
        if el.type == "heading":
            continue
        if el.section != section:
            flush()
            section, path = el.section, el.path
        if el.type == "table":
            flush()
            lines = el.text.split("\n")
            batch = []
            for line in lines:
                if batch and sum(len(x) for x in batch) + len(line) > max_chars:
                    add(section, [(x, el.page, el.box) for x in ([lines[0]] if batch[0] != lines[0] else []) + batch], "table")
                    batch = []
                batch.append(line)
            if batch:
                add(section, [(x, el.page, el.box) for x in ([lines[0]] if batch[0] != lines[0] else []) + batch], "table")
            for p in passages:
                if p.kind == "table" and p.boxes and p.boxes[0] == (el.page, el.box):
                    p.boxes += [b for b in getattr(el, "extra", []) if b not in p.boxes]
            continue
        buffer += [(s, el.page, el.box) for s in _sentences(el.text)]
    flush()
    return passages
# ── The five levels of chunking, from the naive baseline to the one that measures best ──
CHUNKING_LEVELS = {
    "fixed":     ("1 · Fixed blocks",
                  "Cuts every MAX_CHARS characters, ignoring sentences, sections and tables."),
    "sentences": ("2 · Sentence windows",
                  "Whole sentences, N at a time, sliding. Nothing is cut mid-sentence, but a passage "
                  "no longer knows which section it comes from."),
    "sections":  ("3 · Structural passages",
                  "Follows the headings found by the extractor; each table is a passage of its own and "
                  "the heading path travels with the text."),
    "hybrid":    ("4 · Two granularities",
                  "Indexes the structural passages AND small sentence windows: the search becomes precise "
                  "while the reader still receives a whole passage."),
    "smart":     ("5 · Two granularities, tables and context",
                  "Adds each table row as its own unit and sends the neighbouring passages of the section "
                  "along with the answer."),
}

def _document_pieces(document):
    """Every text fragment of the document, in reading order, with its page and box."""
    return [(el.text, el.page, el.box) for el in document.elements if el.text.strip()]

def make_passages(document, start=0):
    """Levels 1 and 2 ignore the structure of the report; levels 3 to 5 use it."""
    if CHUNK_STRATEGY not in ("fixed", "sentences"):
        return split_passages(document, start, MAX_CHARS, OVERLAP_SENTENCES, MIN_CHARS)
    passages, pieces = [], _document_pieces(document)
    def add(text, boxes):
        text = " ".join(text.split())
        if len(text) < 40:
            return
        passages.append(Passage(start + len(passages), document.id, document.short, "", text,
                                f"{document.title}. {text}", list(dict.fromkeys(boxes)), "text", ""))
    if CHUNK_STRATEGY == "fixed":
        overlap, buffer, boxes = int(MAX_CHARS * 0.12), "", []
        for text, page, box in pieces:
            boxes.append((page, box))
            for word in text.split():
                if len(buffer) + len(word) + 1 > MAX_CHARS:     # cut in the middle of a sentence
                    add(buffer, boxes)
                    buffer, boxes = buffer[-overlap:], boxes[-2:]
                buffer += " " + word
        add(buffer, boxes)
    else:                                                        # "sentences"
        units = [(s, page, box) for text, page, box in pieces for s in (_sentences(text) or [text])]
        for i in range(0, max(1, len(units) - WINDOW_SENTENCES + 1), WINDOW_STRIDE):
            window = units[i:i + WINDOW_SENTENCES]
            add(" ".join(u[0] for u in window), [(u[1], u[2]) for u in window])
    return passages

def build_search_units(passages, strategy=None, size=None, stride=None):
    """Levels 4 and 5 index a second, finer granularity: small windows — and, at level 5, table
    rows — which point back through `parent_id` to the passage a reader should actually be given."""
    strategy = strategy or CHUNK_STRATEGY
    size, stride = size or WINDOW_SENTENCES, stride or WINDOW_STRIDE
    for p in passages:
        p.parent_id = p.id
    if strategy not in ("hybrid", "smart"):
        return list(passages)
    units, next_id = list(passages), max((p.id for p in passages), default=-1) + 1
    for p in passages:
        if p.kind == "table":
            if strategy != "smart":
                continue
            lines = p.text.split("\n")
            caption = lines[0] if lines and " — " not in lines[0] else ""
            for line in (lines[1:] if caption else lines):
                if len(line) < 15:
                    continue
                text = f"{caption} {line}".strip()                 # one row, with its caption
                units.append(Passage(next_id, p.doc_id, p.doc, p.section, text,
                                     f"{p.path or p.section}. {text}", p.boxes, "table", p.path, p.id))
                next_id += 1
            continue
        sentences = _sentences(p.text) or [p.text]
        for i in range(0, max(1, len(sentences) - size + 1), stride):
            text = " ".join(sentences[i:i + size])
            if len(text) < 60:
                continue
            units.append(Passage(next_id, p.doc_id, p.doc, p.section, text,
                                 f"{p.path or p.section}. {text}", p.boxes, p.kind, p.path, p.id))
            next_id += 1
    return units

print(f"✅ T3 ready · chunking level {CHUNKING_LEVELS[CHUNK_STRATEGY][0]}.")
````

````python
# @title ⚙️ T4 · Search — multilingual { display-mode: "form" }
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD
from sklearn.preprocessing import Normalizer
from functools import lru_cache
import difflib, hashlib

def strip_accents(text):
    return "".join(c for c in unicodedata.normalize("NFD", text) if unicodedata.category(c) != "Mn")

# ══ 1. Languages ════════════════════════════════════════════════════════════════════════════════
# A collection can mix languages: every document, every passage and every question carries its own
# language code, and each is analysed with the resources of ITS language.
LANGUAGE_NAMES = {"en": "English", "fr": "French", "es": "Spanish", "pt": "Portuguese", "de": "German",
                  "it": "Italian", "ar": "Arabic", "sw": "Swahili", "ru": "Russian"}
DEFAULT_LANG = DOCUMENT_LANGUAGE if DOCUMENT_LANGUAGE in LANGUAGE_NAMES else "en"
LANG = DEFAULT_LANG                        # kept for backward compatibility
DECIMAL_COMMA = {"fr", "es", "pt", "de", "it", "ru"}      # "7,8" is a decimal number in these languages

_STOPWORDS_RAW = {
"en": """a an the and or of in on at to for from by with is are was were be been being what which who
whom whose when where why how does do did can could will would shall should may might it its this that these those
there their they them his her he she we you i as than then so such not no nor about into over under also any all
each per much many get give tell me please document documents report reports say says said information show between
stand stands standing acronym abbreviation abbreviations initials mean means called number numbers proposed propose
proposes presented described mentioned according have has had""",
"fr": """le la les l un une des de du d et en au aux est sont ete ce cet cette ces qui que quoi quel quelle quels
quelles comment combien quand ou il elle se sa son ses leur leurs ne pas plus par pour sur dans avec y t n s qu etait
sera peut peuvent fait faire lors entre apres avant tout toute tous toutes ainsi dont tres rapport dit indique selon
informations donne donnent parle concernant sujet ils elles eux lui chez aussi meme on me nous vous je tu mon ma mes
a ont avoir etre etaient document documents nombre liste signifie veut dire""",
"es": """a al algo algunas algunos ante antes como con contra cual cuales cuando cuanto cuanta cuantos cuantas de del
desde donde dos el ella ellas ellos en entre era eran es esa esas ese eso esos esta estas este esto estos fue fueron
ha han hay la las le les lo los mas me mi muy nada ni no nos o otra otro para pero poco por porque que quien quienes
se ser si sin sobre su sus tambien tanto te tiene tienen todo todos tu un una uno unos y ya documento documentos
informe informes segun dice dicen numero lista significa""",
"pt": """a ao aos as com como da das de do dos e ela elas ele eles em entre era essa esse esta este eu foi foram ha
isso isto ja lhe mais mas me mesmo muito na nas nao nem no nos o os ou para pela pelas pelo pelos por qual quais
quando que quem se sem ser seu seus sua suas tambem te tem um uma umas uns documento documentos relatorio relatorios
segundo diz dizem quanto quantos quantas numero lista significa""",
"de": """der die das den dem des ein eine einer eines einem und oder ist sind war waren wird werden hat haben mit von
zu zum zur fur auf aus bei nach uber unter im in an als wie was wer wann wo warum nicht auch es sie er wir ich ihr
dass dieser diese dieses dokument bericht""",
"it": """il lo la i gli le un una uno di del della dei delle degli e ed o in su per con tra fra da che chi cosa come
quando dove perche non piu anche sono era erano essere ha hanno questo questa questi queste quale quali quanto quanti
documento rapporto""",
"sw": """na ya wa kwa ni katika za la cha vya hii huu hizi kuwa au pia lakini kama hiyo yake zake wake nini gani""",
"ar": """في من على إلى الى عن مع هذا هذه ذلك التي الذي ما ماذا كم هل كيف متى أين و أو ثم""",
}
STOPWORDS_BY_LANG = {l: {strip_accents(w.lower()) for w in ws.split()} for l, ws in _STOPWORDS_RAW.items()}
STOPWORDS = set().union(*STOPWORDS_BY_LANG.values())      # every language together (spotting function words only)

def stopwords(lang):
    return STOPWORDS_BY_LANG.get(lang, set())

# Very frequent function words, used to recognise a language (a question needs only one or two of them).
_MARKERS_RAW = {
    "en": "the and of to in is are was were for with that this which what how from by on be not have has it its who when",
    "fr": "le la les des du et est sont une un pour dans que qui sur avec au aux pas ce cette quel quelle quels quelles combien il elle nous par plus",
    "es": "el la los las del y es son una un para en que por con al se como cual cuales cuanto cuantos esta este estos entre su sus tiene tienen hay muy fue ha han desde",
    "pt": "o os a as do da dos das e em um uma para que com por nao ao se qual quais quanto quantos esta este no na nos nas pelo pela tem foi sao",
    "de": "der die das und ist sind ein eine mit fur von zu den dem nicht wie was auch im auf",
    "it": "il gli lo della delle dei e sono una per che con non come quale quali nel nella degli",
    "sw": "na ya wa kwa ni katika za la cha vya hii kuwa au pia hiyo",
}
LANGUAGE_MARKERS = {l: {strip_accents(w) for w in ws.split()} for l, ws in _MARKERS_RAW.items()}

def language_scores(text):
    """Number of marker words of each language found in the text."""
    t = text.lower()
    letters = [c for c in t if c.isalpha()]
    if letters:
        if sum("؀" <= c <= "ۿ" for c in letters) / len(letters) > 0.3:
            return {"ar": 99}
        if sum("Ѐ" <= c <= "ӿ" for c in letters) / len(letters) > 0.3:
            return {"ru": 99}
    counts = Counter(re.findall(r"[^\W\d_]+", strip_accents(t)))
    scores = {l: sum(counts[w] for w in markers) for l, markers in LANGUAGE_MARKERS.items()}
    scores["es"] += 2 * (t.count("¿") + t.count("¡")) + t.count("ñ")    # letters that give a language away
    scores["pt"] += t.count("ã") + t.count("õ")
    return scores

def detect_language(text, default=None, minimum=1):
    """Most likely language of a text, or `default` when the evidence is too thin or tied."""
    scores = language_scores(text or "")
    if not scores:
        return default
    ranked = sorted(scores.items(), key=lambda x: -x[1])
    best, hits = ranked[0]
    if hits < minimum:
        return default
    if len(ranked) > 1 and ranked[1][1] == hits:              # a tie: keep the default if it is one of them
        return default if default in (best, ranked[1][0]) else best
    return best

def question_language(question, collection=(), previous=None):
    """Language of a question: forced by QUESTION_LANGUAGE, detected, or inherited from the previous turn."""
    if QUESTION_LANGUAGE != "auto":
        return QUESTION_LANGUAGE
    return (detect_language(question, None, 1) or previous
            or (list(collection)[0] if collection else DEFAULT_LANG))

def passage_language(text, document_lang):
    """Bilingual reports: a long passage may be written in another language than its document.
    Deliberately conservative: links, lists of abbreviations and bibliographic references mix
    languages, so a passage changes language only on clear evidence (at least 12 marker words,
    and twice as many as for the language of its document)."""
    if not DETECT_PASSAGE_LANGUAGE or len(text) < 400:
        return document_lang
    clean = re.sub(r"\S*(?:https?://|www\.)\S*|\S+@\S+", " ", text)      # links say nothing about language
    scores = language_scores(clean)
    best = max(scores, key=scores.get)
    if best != document_lang and scores[best] >= 12 and scores[best] >= 2 * scores.get(document_lang, 0):
        return best
    return document_lang

def passage_lang(p):
    return getattr(p, "lang", "") or DEFAULT_LANG

# ══ 2. Normalisation, stems, numbers ════════════════════════════════════════════════════════════
SUFFIXES = {
    "en": sorted(["ational", "ations", "ation", "ments", "ment", "ness", "ings", "ing", "ities", "ity", "ies",
                  "ied", "ers", "er", "ed", "es", "ly", "al", "s", "e"], key=len, reverse=True),
    "fr": sorted(["ements", "ement", "ations", "ation", "itions", "ition", "ances", "ance", "ences", "ence",
                  "euses", "euse", "eurs", "eur", "ites", "ite", "iques", "ique", "ives", "ive", "ifs", "if",
                  "elles", "elle", "iels", "iel", "aux", "ants", "ant", "ents", "ent", "ees", "ee", "es", "er",
                  "age", "s", "e", "x"], key=len, reverse=True),
    # light fallbacks, used only when the snowballstemmer package is missing
    "es": sorted(["aciones", "acion", "ciones", "cion", "idades", "idad", "mente", "es", "s"], key=len, reverse=True),
    "pt": sorted(["acoes", "acao", "idades", "idade", "mente", "oes", "es", "s"], key=len, reverse=True),
}
# English and French keep the stemmer this notebook was tuned with; other languages use Snowball.
_SNOWBALL = {"es": "spanish", "pt": "portuguese", "de": "german", "it": "italian", "ar": "arabic", "ru": "russian"}
_STEMMERS = {}

def _snowball(lang):
    if lang not in _SNOWBALL:
        return None
    if lang not in _STEMMERS:
        try:
            import snowballstemmer
            _STEMMERS[lang] = snowballstemmer.stemmer(_SNOWBALL[lang])
        except Exception:
            _STEMMERS[lang] = None
    return _STEMMERS[lang]

@lru_cache(maxsize=300_000)
def stem(word, lang=None):
    lang = lang or DEFAULT_LANG
    if any(ch.isdigit() for ch in word):
        return word
    stemmer = _snowball(lang)
    if stemmer is not None:
        return stemmer.stemWord(word)
    for _ in range(2 if lang == "en" else 1):
        for suffix in SUFFIXES.get(lang, ()):
            if word.endswith(suffix) and len(word) - len(suffix) >= 4:
                word = word[: -len(suffix)]
                break
    return word

def normalise_numbers(text, lang):
    """One spelling for every number, whatever the language: 18,452,300 · 18 452 300 · 18.452.300 -> 18452300;
    7,8 (French, Spanish…) -> 7.8. A figure in an English question then matches a French table."""
    if lang in DECIMAL_COMMA:
        text = re.sub(r"(?<=\d)[\s  .](?=\d{3}\b)", "", text)
        return re.sub(r"(?<=\d),(?=\d)", ".", text)
    return re.sub(r"(?<=\d),(?=\d{3}\b)", "", text)

def analyse(text, lang=None):
    lang = lang or DEFAULT_LANG
    t = normalise_numbers(strip_accents(text.lower()), lang)
    stops = stopwords(lang)
    return [stem(w, lang) for w in re.findall(r"[^\W_]+(?:\.\d+)?", t) if w not in stops and len(w) > 1]

# ══ 3. Glossaries (within a language) and cross-lingual lexicon (between languages) ════════════
GLOSSARIES = {
    "en": {
        # users' words                   -> words used in publications
        "inhabitants": "population resident people", "residents": "population resident",
        "how many people": "population", "people live": "population",
        "jobless": "unemployment unemployed", "out of work": "unemployment unemployed",
        "young people": "youth 15 24", "youth": "young 15 24",
        "woman": "women", "man": "men", "female": "women", "male": "men",
        "city": "urban", "cities": "urban", "towns": "urban", "countryside": "rural", "villages": "rural",
        "proportion": "share rate", "percentage": "share rate",
        "microdata": "anonymised files", "individual data": "microdata anonymised",
        "get access": "available access", "obtain": "available access",
        "released": "release published", "come out": "released published", "published": "released",
        "when": "published released date", "publication": "published released",
        "did not respond": "non-response reweighting", "non-response": "reweighting",
        "first quarter": "q1", "second quarter": "q2", "third quarter": "q3", "fourth quarter": "q4",
        "change": "down up", "changed": "down up", "evolve": "down up", "trend": "down up",
        "informal": "social protection contract", "wage": "income", "salary": "income", "earnings": "income",
        "sample": "dwellings drawn areas", "definition": "glossary", "what is meant by": "glossary",
        "next census": "census scheduled", "household size": "average size household",
    },
    "fr": {
        "habitants": "population residente personnes", "gens": "population personnes",
        "vivent": "population residente", "combien de personnes": "population",
        "ville": "urbain", "villes": "urbain", "campagne": "rural",
        "sans emploi": "chomage chomeur", "chomeurs": "chomage", "travail": "emploi",
        "jeunes": "15 24 ans", "proportion": "taux part", "pourcentage": "taux part",
        "microdonnees": "fichiers donnees individuelles anonymises",
        "chercheur": "chercheurs", "obtenir": "acceder accessibles acces",
        "sort": "publie publies diffusion paraitra", "sortent": "publie publies diffusion paraitra",
        "sortira": "paraitra publie", "publication": "publies paraitra", "quand": "date publies paraitra",
        "pas repondu": "non reponse repondération", "non-reponse": "repondération",
        "premier trimestre": "t1", "deuxieme trimestre": "t2", "second trimestre": "t2",
        "troisieme trimestre": "t3", "quatrieme trimestre": "t4",
        "evolue": "hausse baisse", "evolution": "hausse baisse", "varie": "hausse baisse",
        "informel": "informelle protection sociale contrat", "salaire": "revenu",
        "echantillon": "logements tirage zones", "definition": "glossaire", "signifie": "glossaire",
        "prochain recensement": "recensement programme", "taille des menages": "taille moyenne menage",
    },
    "es": {"jovenes": "15 24", "cuando": "fecha publicado", "proporcion": "tasa porcentaje"},
    "pt": {"jovens": "15 24", "quando": "data publicado", "proporcao": "taxa percentagem"},
}
GLOSSARY = GLOSSARIES.setdefault(DEFAULT_LANG, {})   # 🧪 add your own: GLOSSARIES["fr"]["sans boulot"] = "chomage"

# Each line is ONE concept in several languages (alternatives separated by spaces). Without a language
# model, this is what lets an English question find a French passage with BM25, and lets the
# confidence score recognise it. 🧪 Extend it with the vocabulary of your own domain.
CROSS_LINGUAL_LEXICON = [
    {"en": "unemployment jobless", "fr": "chômage", "es": "desempleo paro", "pt": "desemprego"},
    {"en": "unemployed", "fr": "chômeur chômeurs", "es": "desempleado desempleados parados", "pt": "desempregado desempregados"},
    {"en": "employment job jobs", "fr": "emploi emplois", "es": "empleo empleos", "pt": "emprego empregos"},
    {"en": "employed", "fr": "occupés", "es": "ocupados", "pt": "ocupados"},
    {"en": "underemployment", "fr": "sous-emploi", "es": "subempleo", "pt": "subemprego"},
    {"en": "rate rates", "fr": "taux", "es": "tasa tasas", "pt": "taxa taxas"},
    {"en": "population inhabitants", "fr": "population habitants", "es": "población habitantes", "pt": "população habitantes"},
    {"en": "people persons", "fr": "personnes", "es": "personas", "pt": "pessoas"},
    {"en": "household households", "fr": "ménage ménages", "es": "hogar hogares", "pt": "agregado agregados domicílio domicílios"},
    {"en": "women woman female", "fr": "femmes femme", "es": "mujeres mujer", "pt": "mulheres mulher"},
    {"en": "men man male", "fr": "hommes homme", "es": "hombres hombre", "pt": "homens homem"},
    {"en": "youth young", "fr": "jeunes jeune jeunesse", "es": "jóvenes joven juventud", "pt": "jovens jovem juventude"},
    {"en": "age aged", "fr": "âge âgés", "es": "edad", "pt": "idade"},
    {"en": "region regions regional", "fr": "région régions régional", "es": "región regiones regional", "pt": "região regiões regional"},
    {"en": "urban city cities", "fr": "urbain urbaine urbains ville villes", "es": "urbano urbana urbanas urbanos ciudad", "pt": "urbano urbana urbanas urbanos cidade"},
    {"en": "rural countryside", "fr": "rural rurale ruraux campagne", "es": "rural rurales", "pt": "rural rurais"},
    {"en": "inflation", "fr": "inflation", "es": "inflación", "pt": "inflação"},
    {"en": "price prices", "fr": "prix", "es": "precio precios", "pt": "preço preços"},
    {"en": "consumer consumption", "fr": "consommation consommateurs", "es": "consumo consumidor consumidores", "pt": "consumo consumidor consumidores"},
    {"en": "quarter quarterly", "fr": "trimestre trimestriel trimestrielle", "es": "trimestre trimestral", "pt": "trimestre trimestral"},
    {"en": "year years annual", "fr": "année années annuel annuelle ans", "es": "año años anual", "pt": "ano anos anual"},
    {"en": "month monthly", "fr": "mois mensuel mensuelle", "es": "mes mensual", "pt": "mês mensal"},
    {"en": "average", "fr": "moyenne moyen", "es": "promedio medio media", "pt": "média médio"},
    {"en": "share proportion percentage", "fr": "part proportion pourcentage", "es": "proporción porcentaje parte", "pt": "proporção percentagem parte"},
    {"en": "income wage wages salary earnings", "fr": "revenu revenus salaire salaires", "es": "ingreso ingresos salario salarios", "pt": "rendimento rendimentos salário salários"},
    {"en": "informal", "fr": "informel informelle", "es": "informal", "pt": "informal"},
    {"en": "survey", "fr": "enquête", "es": "encuesta", "pt": "inquérito pesquisa"},
    {"en": "census", "fr": "recensement", "es": "censo", "pt": "censo recenseamento"},
    {"en": "growth", "fr": "croissance", "es": "crecimiento", "pt": "crescimento"},
    {"en": "activity", "fr": "activité", "es": "actividad", "pt": "atividade actividade"},
    {"en": "labour labor work", "fr": "travail", "es": "trabajo laboral", "pt": "trabalho laboral"},
    {"en": "sample", "fr": "échantillon", "es": "muestra", "pt": "amostra"},
    {"en": "response", "fr": "réponse", "es": "respuesta", "pt": "resposta"},
    {"en": "data microdata", "fr": "données microdonnées", "es": "datos microdatos", "pt": "dados microdados"},
    {"en": "access", "fr": "accès", "es": "acceso", "pt": "acesso"},
    {"en": "published publication release released", "fr": "publié publiée publication diffusion paraîtra",
     "es": "publicado publicación difusión", "pt": "publicado publicação divulgação"},
    {"en": "next", "fr": "prochain prochaine", "es": "próximo próxima", "pt": "próximo próxima"},
    {"en": "food", "fr": "alimentaire alimentaires alimentation", "es": "alimentos alimentario", "pt": "alimentos alimentar"},
    {"en": "energy", "fr": "énergie", "es": "energía", "pt": "energia"},
    {"en": "health", "fr": "santé", "es": "salud", "pt": "saúde"},
    {"en": "education school schooling enrolment", "fr": "éducation école scolarisation", "es": "educación escuela escolarización", "pt": "educação escola escolarização"},
    {"en": "poverty poor", "fr": "pauvreté pauvres", "es": "pobreza pobres", "pt": "pobreza pobres"},
    {"en": "agriculture agricultural", "fr": "agriculture agricole", "es": "agricultura agrícola", "pt": "agricultura agrícola"},
    {"en": "country countries national", "fr": "pays national nationale", "es": "país países nacional", "pt": "país países nacional"},
    {"en": "strategy", "fr": "stratégie", "es": "estrategia", "pt": "estratégia"},
    {"en": "artificial intelligence", "fr": "artificielle intelligence", "es": "artificial inteligencia", "pt": "artificial inteligência"},
    {"en": "innovation", "fr": "innovation", "es": "innovación", "pt": "inovação"},
    {"en": "size", "fr": "taille", "es": "tamaño", "pt": "tamanho dimensão"},
    {"en": "increase increased rise", "fr": "hausse augmentation augmente", "es": "aumento alza aumenta", "pt": "aumento alta aumenta"},
    {"en": "decrease decline fall", "fr": "baisse diminution recul", "es": "disminución descenso baja", "pt": "diminuição queda redução"},
    {"en": "north", "fr": "nord", "es": "norte", "pt": "norte"},
    {"en": "south", "fr": "sud", "es": "sur", "pt": "sul"},
    {"en": "coast coastal", "fr": "côte littoral", "es": "costa", "pt": "costa litoral"},
    {"en": "centre center central", "fr": "centre", "es": "centro", "pt": "centro"},
    {"en": "first", "fr": "premier première", "es": "primer primero primera", "pt": "primeiro primeira"},
    {"en": "second", "fr": "deuxième second", "es": "segundo segunda", "pt": "segundo segunda"},
    {"en": "third", "fr": "troisième", "es": "tercer tercero tercera", "pt": "terceiro terceira"},
    {"en": "fourth", "fr": "quatrième", "es": "cuarto cuarta", "pt": "quarto quarta"},
    {"en": "mobile phone telephone", "fr": "mobile téléphone portable", "es": "móvil teléfono celular", "pt": "móvel telemóvel telefone celular"},
]
_LEXICON_INDEX = {}        # (language, stem) -> {other language: stems of the same concept}

def build_lexicon_index():
    _LEXICON_INDEX.clear()
    for group in CROSS_LINGUAL_LEXICON:
        stems = {l: {s for w in words.split() for s in analyse(w, l)} for l, words in group.items()}
        for l, own in stems.items():
            for s in own:
                entry = _LEXICON_INDEX.setdefault((l, s), {})
                for other, theirs in stems.items():
                    if other != l:
                        entry.setdefault(other, set()).update(theirs)
build_lexicon_index()      # 🧪 run it again after extending CROSS_LINGUAL_LEXICON

def equivalents(term, lang, targets):
    entry = _LEXICON_INDEX.get((lang, term))
    return set().union(*(entry.get(t, set()) for t in targets)) if entry else set()

# ══ 4. Spelling tolerance (per language) ════════════════════════════════════════════════════════
# Users mistype: "coutries", "assessement". A mistyped word is corrected only against the vocabulary
# of the SAME language, so a French word is never "corrected" into an English one.
VOCABULARY_FORMS = {}      # language -> {stem: form as written in the documents}
TRIGRAM_INDEX = {}         # language -> {character trigram: stems}

def _trigrams(word):
    padded = f"  {word} "
    return {padded[i:i + 3] for i in range(len(padded) - 2)}

def build_vocabulary(passages):
    forms = {}
    for p in passages:
        lang = passage_lang(p)
        stops = stopwords(lang)
        for word in re.findall(r"[^\W\d_]{3,}", strip_accents(p.text.lower())):
            if word not in stops:
                forms.setdefault(lang, {}).setdefault(stem(word, lang), Counter())[word] += 1
    VOCABULARY_FORMS.clear(); TRIGRAM_INDEX.clear()
    for lang, roots in forms.items():
        VOCABULARY_FORMS[lang] = {root: c.most_common(1)[0][0] for root, c in roots.items()}
        index = TRIGRAM_INDEX.setdefault(lang, {})
        for root in VOCABULARY_FORMS[lang]:
            for gram in _trigrams(root):
                index.setdefault(gram, set()).add(root)

def closest_term(root, lang=None, minimum=0.86):
    """Nearest document term of the same language, or None. Deliberately strict: a genuinely foreign
    word ("maize", "hospital") must stay unknown, because that is what makes a question refusable."""
    lang = lang or DEFAULT_LANG
    forms, index = VOCABULARY_FORMS.get(lang, {}), TRIGRAM_INDEX.get(lang, {})
    if len(root) < 4 or root in forms:
        return None
    candidates = Counter()
    for gram in _trigrams(root):
        for other in index.get(gram, ()):
            candidates[other] += 1
    best, best_score = None, minimum
    for other, _ in candidates.most_common(40):
        if abs(len(other) - len(root)) > 2:
            continue
        score = difflib.SequenceMatcher(None, root, other).ratio()
        if score > best_score:
            best, best_score = other, score
    return best

def fix_spelling(question, lang=None):
    """Replace each mistyped word by the closest word the documents of that language really use."""
    lang = lang or DEFAULT_LANG
    forms, stops = VOCABULARY_FORMS.get(lang), stopwords(lang)
    if not forms:
        return question, []
    corrections = []
    def replace(match):
        word = match.group(0)
        plain = strip_accents(word.lower())
        if plain in stops or stem(plain, lang) in forms:
            return word
        near = closest_term(stem(plain, lang), lang)
        if not near:
            return word
        corrections.append((word, forms[near]))
        return forms[near]
    return re.sub(r"[^\W\d_]{4,}", replace, question), corrections

# ══ 5. Acronyms, expansion, concepts ═══════════════════════════════════════════════════════════
ACRONYMS = {}          # filled from the documents themselves: "DIL" -> "Data Innovation Lab"
_ACRONYM_IN_TEXT = re.compile(r"([A-Z][\w’'\-]*(?:\s+[a-z\w’'\-]+){1,5}(?:\s+[A-Z][\w’'\-]*)?)\s*\(([A-Z]{2,6})s?\)")
_ACRONYM_IN_LIST = re.compile(r"(?:^|\n|;|—|\|)\s*([A-Z]{2,6})\s*(?:[—:\-–]|\s{2,})\s*([A-Z][^.;\n|]{4,60})")

def build_acronyms(passages):
    """Institutional reports are full of acronyms: learn them from the documents (every language)."""
    found = {}
    for p in passages:
        for expansion, acronym in _ACRONYM_IN_TEXT.findall(p.text):
            found.setdefault(acronym, Counter())[expansion.strip()] += 1
        for acronym, expansion in _ACRONYM_IN_LIST.findall(p.text):
            found.setdefault(acronym, Counter())[expansion.strip()] += 1
    return {a: counts.most_common(1)[0][0] for a, counts in found.items()
            if len(counts.most_common(1)[0][0].split()) >= 2}

def expand_query(question, lang=None):
    lang = lang or DEFAULT_LANG
    q = " " + re.sub(r"[^\w-]+", " ", strip_accents(question.lower())) + " "
    added = [(key, terms) for key, terms in GLOSSARIES.get(lang, {}).items() if f" {key} " in q]
    for acronym, expansion in ACRONYMS.items():
        if f" {acronym.lower()} " in q:
            added.append((acronym, expansion))
    return question + " " + " ".join(t for _, t in added), added

_CONCEPT_CACHE = {}

def concepts(question, lang=None, targets=()):
    """Each term of the question (analysed in its language) with its synonyms: the glossary of that
    language, and — for each language in `targets` — the equivalents given by the lexicon."""
    lang = lang or DEFAULT_LANG
    other = tuple(sorted(t for t in targets if t != lang))
    key = (question, lang, other)
    if key in _CONCEPT_CACHE:
        return _CONCEPT_CACHE[key]
    _, added = expand_query(question, lang)
    synonyms = {}
    for k, terms in added:
        for word in analyse(k, lang) or [stem(strip_accents(k.split()[-1].lower()), lang)]:
            synonyms.setdefault(word, set()).update(analyse(terms, lang))
    items = []
    for t in dict.fromkeys(analyse(question, lang)):
        group = {t} | synonyms.get(t, set())
        if other:                                  # translations of the term itself, not of its synonyms
            group |= equivalents(t, lang, other)
        items.append((t, group))
    _CONCEPT_CACHE[key] = items
    return items

def query_terms(question, lang=None, targets=()):
    """Every term BM25 should look for: the question, its expansions and its lexicon equivalents."""
    lang = lang or DEFAULT_LANG
    terms = analyse(expand_query(question, lang)[0], lang)
    terms += [t for _, g in concepts(question, lang, targets) for t in sorted(g)]
    return list(dict.fromkeys(terms))

def query_view(passage, question, lang, queries):
    """The version of the question to compare with a passage: its translation into the passage's
    language when one exists, otherwise the original (then matched through the lexicon)."""
    pl = passage_lang(passage)
    if queries and queries.get(pl):
        return queries[pl], pl
    return question, lang

# ══ 6. Query translation (with a language model) ═══════════════════════════════════════════════
PROMPT_TRANSLATE = """You translate search questions for a document search engine. Translate the question into {language}.
Keep numbers, years, acronyms, codes and proper names exactly as written. Do not answer the question.
Return only the translation, on a single line."""
_TRANSLATIONS = {}

def translate_query(question, target):
    """Translation of the question into `target`, cached; None without a model or on failure."""
    key = (question, target)
    if key in _TRANSLATIONS:
        return _TRANSLATIONS[key]
    text = None
    if LLM.available and target in LANGUAGE_NAMES:
        try:
            out = LLM.generate(PROMPT_TRANSLATE.format(language=LANGUAGE_NAMES[target]), question, max_tokens=150)
            out = (out.strip().splitlines() or [""])[0].strip().strip('"«»“” ')
            text = out if 3 < len(out) < 400 else None
        except Exception:
            text = None
    _TRANSLATIONS[key] = text
    return text

# ══ 7. The two indexes ══════════════════════════════════════════════════════════════════════════
class BM25Index:
    """BM25 over a sparse term-document matrix; each unit is analysed with the rules of its language."""

    def __init__(self, passages, k1=1.5, b=0.75):
        from scipy.sparse import csr_matrix
        self.passages, self.k1, self.b = passages, k1, b
        self.docs = [analyse(p.index_text, passage_lang(p)) for p in passages]
        n = len(self.docs)
        self.vocabulary, rows, cols, values = {}, [], [], []
        for i, d in enumerate(self.docs):
            for term, f in Counter(d).items():
                j = self.vocabulary.setdefault(term, len(self.vocabulary))
                rows.append(i); cols.append(j); values.append(f)
        self.matrix = csr_matrix((values, (rows, cols)), shape=(n, max(1, len(self.vocabulary))), dtype=np.float32)
        lengths = np.array([len(d) for d in self.docs], dtype=np.float32)
        self.avg_len = max(1e-9, float(lengths.mean()) if n else 1.0)
        self.norm = (self.k1 * (1 - self.b + self.b * lengths / self.avg_len)).astype(np.float32)
        df = np.asarray((self.matrix > 0).sum(axis=0)).ravel()
        idf = np.log(1 + (n - df + 0.5) / (df + 0.5))
        self.idf_vector = idf.astype(np.float32)
        self.idf = {t: float(idf[j]) for t, j in self.vocabulary.items()}
        self.idf_max = max(self.idf.values()) if self.idf else 1.0

    def search(self, question, k=30, lang=None, targets=()):
        return self.search_terms(query_terms(question, lang, targets), k)

    def search_terms(self, terms, k=30):
        terms = [t for t in terms if t in self.vocabulary]
        if not terms:
            return []
        columns = [self.vocabulary[t] for t in dict.fromkeys(terms)]
        frequencies = self.matrix[:, columns].toarray()
        weights = self.idf_vector[columns] * (self.k1 + 1)
        scores = (weights * frequencies / (frequencies + self.norm[:, None])).sum(axis=1)
        order = [i for i in np.argsort(-scores)[:k] if scores[i] > 0]
        return [(self.passages[i], float(scores[i])) for i in order]

_EMBEDDING_CACHE = {}
CROSS_LINGUAL_MODELS = ("multilingual", "e5", "bge-m3", "labse", "laBSE")

def _embedding_prefixes():
    """E5 models expect "query: " and "passage: " in front of the texts."""
    return ("query: ", "passage: ") if "e5" in EMBEDDING_MODEL.lower() else ("", "")

class SemanticIndex:
    def __init__(self, passages):
        self.passages, self.mode, self.error = passages, None, None
        texts = [p.index_text for p in passages]
        if USE_EMBEDDINGS:
            try:
                if INSTALL_EMBEDDINGS:
                    ensure_package("sentence_transformers", "sentence-transformers")
                from sentence_transformers import SentenceTransformer
                if EMBEDDING_MODEL not in _EMBEDDING_CACHE:
                    _EMBEDDING_CACHE[EMBEDDING_MODEL] = SentenceTransformer(EMBEDDING_MODEL)
                self.model = _EMBEDDING_CACHE[EMBEDDING_MODEL]
                texts = [_embedding_prefixes()[1] + t for t in texts]
                key = hashlib.sha1((EMBEDDING_MODEL + "|" + "|".join(texts)).encode("utf-8")).hexdigest()[:16]
                cached = CACHE_DIR / f"embeddings_{key}.npy"
                if USE_CACHE and cached.exists():
                    self.matrix = np.load(cached)
                else:
                    self.matrix = self.model.encode(texts, normalize_embeddings=True, batch_size=32,
                                                    show_progress_bar=len(texts) > 200)
                    if USE_CACHE:
                        np.save(cached, self.matrix)
                self.mode, self.name = "embeddings", f"Embeddings · {EMBEDDING_MODEL.split('/')[-1]}"
            except Exception as error:
                self.error = type(error).__name__
        if self.mode is None:                      # offline fallback: latent semantic analysis
            self.vectorizer = TfidfVectorizer(analyzer="char_wb", ngram_range=(3, 5),
                                              strip_accents="unicode", sublinear_tf=True)
            m = self.vectorizer.fit_transform(texts)
            self.svd = TruncatedSVD(n_components=max(1, min(96, m.shape[0] - 1, m.shape[1] - 1)), random_state=0)
            self.normalizer = Normalizer()
            self.matrix = self.normalizer.fit_transform(self.svd.fit_transform(m))
            self.mode, self.name = "lsa", "Approximate semantic search · offline LSA"

    @property
    def cross_lingual(self):
        """Can this index match a question and a passage written in different languages?"""
        return self.mode == "embeddings" and any(m.lower() in EMBEDDING_MODEL.lower() for m in CROSS_LINGUAL_MODELS)

    def _encode(self, text):
        if self.mode == "embeddings":
            return self.model.encode([_embedding_prefixes()[0] + text], normalize_embeddings=True)
        return self.normalizer.transform(self.svd.transform(self.vectorizer.transform([text])))

    def search(self, question, k=30, lang=None):
        scores = (self.matrix @ self._encode(expand_query(question, lang)[0]).T).ravel()
        order = np.argsort(-scores)[:k]
        return [(self.passages[i], float(scores[i])) for i in order]

def rrf_fusion(rankings, k=60, top=12, weights=None):
    """Reciprocal rank fusion; `weights` lets the rankings of the original question count more than
    those of its translations."""
    scores, objects = {}, {}
    for n, ranking in enumerate(rankings):
        w = weights[n] if weights else 1.0
        for rank, (p, _) in enumerate(ranking, start=1):
            scores[p.id] = scores.get(p.id, 0.0) + w / (k + rank)
            objects[p.id] = p
    order = sorted(scores, key=scores.get, reverse=True)[:top]
    return [(objects[i], scores[i]) for i in order]

# ── What kind of answer the question is asking for (used by the reranker and the extractor) ──
QUESTION_TYPES = {
    "count": re.compile(r"\b(how many|how much|number of|combien|nombre de|cu[aá]nt[oa]s?|n[uú]mero de|"
                        r"quant[oa]s?|wie viele?)\b", re.I),
    "date": re.compile(r"\b(when|what date|which year|quand|en quelle ann[eé]e|cu[aá]ndo|en qu[eé] a[nñ]o|"
                       r"quando|em que ano|wann)\b", re.I),
    "list": re.compile(r"\b(list|name the|which \w+ (are|were|have|took)|what are the|quels|quelles|"
                       r"enum[eé]r\w*|cu[aá]les son|quais s[aã]o|lista)\b", re.I),
    "definition": re.compile(r"\b(what is an?|what are an?|define|definition of|stands? for|what does \w+ mean|"
                             r"signifie|qu'est-ce|qu[eé] es|o que [eé]|significa|siglas?|bedeutet)\b", re.I),
}   # order matters: the first pattern that matches wins
NUMBER_IN_TEXT = re.compile(r"\b\d[\d.,]*\b|\b(one|two|three|four|five|six|seven|eight|nine|ten|eleven|twelve|"
                            r"deux|trois|quatre|cinq|six|sept|huit|neuf|dix|onze|douze|"
                            r"tres|cuatro|cinco|seis|siete|ocho|nueve|diez|dois|quatro|sete|oito|nove|dez)\b", re.I)
CROSS_REFERENCE = re.compile(r"\b(figure|table|tableau|graph|chart|annex|annexe|section|page|p\.|cuadro|"
                             r"gr[aá]fico|quadro|anexo|tabla|tabela)\s*\d+", re.I)
ENUMERATION = re.compile(r"(?:,|;|\band\b|\bet\b|\by\b|\be\b)[^,;]{2,70}(?:,|;|\band\b|\bet\b|\by\b|\be\b)", re.I | re.S)
A_YEAR = re.compile(r"\b(19|20)\d{2}\b")

def question_type(question):
    for name, pattern in QUESTION_TYPES.items():
        if pattern.search(question):
            return name
    return "other"

def carries_a_number(sentence):
    """A real quantity, not a cross-reference such as "Figure 8" or "see Table 3"."""
    return bool(NUMBER_IN_TEXT.search(CROSS_REFERENCE.sub(" ", sentence)))

# ══ 8. The four levels of reranking ═════════════════════════════════════════════════════════════
RERANKING_LEVELS = {
    "none":          ("0 · None", "Keeps the order produced by the fusion of the two searches."),
    "coverage":      ("1 · Question coverage", "Share of the question's weighted terms present in the passage."),
    "smart":         ("2 · Coverage, density and answer type",
                      "Equal coverage is broken by the answer type (a “how many” question prefers a passage "
                      "carrying a quantity, a “list” question one carrying an enumeration), then by the language "
                      "of the question, then by term density."),
    "cross-encoder": ("3 · Neural cross-encoder",
                      "A model reads the question and the passage together and scores the pair; it reorders "
                      "passages of comparable coverage. The most accurate, and the slowest: it needs sentence-transformers."),
}
ENGLISH_CROSS_ENCODER = "cross-encoder/ms-marco-MiniLM-L-6-v2"
MULTILINGUAL_CROSS_ENCODER = "cross-encoder/mmarco-mMiniLMv2-L12-H384-v1"   # trained on mMARCO, 14 languages

class Reranker:
    def __init__(self, bm25, mode=None, languages=None):
        self.bm25 = bm25
        self.mode = mode or RERANKING
        self.languages = list(languages or [DEFAULT_LANG])
        self.cross, self.model_name = None, None
        if self.mode == "cross-encoder":
            model = CROSS_ENCODER_MODEL
            if model == "auto":     # the English model only for an English collection read in English
                english = set(self.languages) <= {"en"} and QUESTION_LANGUAGE in ("auto", "en")
                model = ENGLISH_CROSS_ENCODER if english and TRANSLATE_QUERIES and LLM.available else MULTILINGUAL_CROSS_ENCODER
            try:
                ensure_package("sentence_transformers", "sentence-transformers")
                from sentence_transformers import CrossEncoder
                self.cross, self.model_name = CrossEncoder(model), model
            except Exception as error:
                print(f"⚠️ Cross-encoder unavailable ({type(error).__name__}): reranking falls back to “smart”.")
                self.mode = "smart"

    @property
    def name(self):
        return RERANKING_LEVELS[self.mode][0]

    def answer_type_bonus(self, kind, passage):
        """Does this passage hold the *kind* of thing the question asks for?"""
        if kind == "count":
            return 1 if carries_a_number(passage.text) else 0
        if kind == "list":
            return 1 if ENUMERATION.search(passage.text) else 0
        if kind == "date":
            return 1 if A_YEAR.search(passage.text) else 0
        return 0

    def weight(self, term, group=None):
        """Weight of a question term: its rarity in the collection. A term unknown to the collection
        but whose synonym or translation is known takes that weight; otherwise it is a strong signal."""
        if term in self.bm25.idf:
            return self.bm25.idf[term]
        known = [self.bm25.idf[g] for g in (group or ()) if g in self.bm25.idf]
        return max(known) if known else self.bm25.idf_max * 1.5

    def coverage(self, question, passage, lang=None):
        lang = lang or DEFAULT_LANG
        present = set(analyse(passage.index_text, passage_lang(passage)))
        items = concepts(question, lang, (passage_lang(passage),))
        total = sum(self.weight(t, g) for t, g in items)
        return sum(self.weight(t, g) for t, g in items if g & present) / total if total else 0.0

    def rerank(self, question, candidates, density=None, kind="other", lang=None, queries=None):
        """The score returned is always the question coverage (measured with the version of the question
        in the passage's language), so that the confidence threshold keeps the same meaning in every
        language; only the ORDER depends on the level chosen."""
        lang = lang or DEFAULT_LANG
        queries = queries or {lang: question}
        views = [query_view(p, question, lang, queries) for p, _ in candidates]
        # the better of the two readings: the translation in the passage's language, or the original
        # question matched through the lexicon (a translated word absent from the whole collection
        # must not sink a passage that the original word already matches)
        scores = [max(self.coverage(question, p, lang), self.coverage(q, p, l) if q is not question else 0.0)
                  for (p, _), (q, l) in zip(candidates, views)]
        if self.mode == "none":
            order = range(len(candidates))
        elif self.mode == "coverage":
            order = sorted(range(len(candidates)), key=lambda i: (-scores[i], i))
        elif self.cross is not None:
            pairs = [(q, " ".join(p.text.split())[:900]) for (p, _), (q, _) in zip(candidates, views)]
            neural = self.cross.predict(pairs, show_progress_bar=False)
            # the neural model refines the order WITHIN a band of comparable coverage (steps of 0.2):
            # it can no longer put first a passage that shares no word with the question — which a
            # multilingual cross-encoder does easily when it compares pairs written in different languages
            same = lambda p: 1 if PREFER_QUESTION_LANGUAGE and passage_lang(p) == lang else 0
            order = sorted(range(len(candidates)),
                           key=lambda i: (-round(scores[i] / 0.2), -float(neural[i]), -same(candidates[i][0])))
        else:
            density = density or {}
            same = lambda p: 1 if PREFER_QUESTION_LANGUAGE and passage_lang(p) == lang else 0
            order = sorted(range(len(candidates)),
                           key=lambda i: (-round(scores[i] / 0.02),
                                          -self.answer_type_bonus(kind, candidates[i][0]),
                                          -same(candidates[i][0]),
                                          -density.get(candidates[i][0].id, 0.0), i))
        return [(candidates[i][0], scores[i]) for i in order]

def diversify(ranked, k, several_documents):
    """At most k-1 passages from the same document when several documents are searched."""
    cap = max(1, k - 1) if several_documents else k
    chosen, count = [], Counter()
    for p, s in ranked:
        if count[p.doc_id] < cap:
            chosen.append((p, s)); count[p.doc_id] += 1
        if len(chosen) == k:
            return chosen
    for p, s in ranked:
        if (p, s) not in chosen and len(chosen) < k:
            chosen.append((p, s))
    return chosen
print(f"✅ T4 ready · multilingual · default language {DEFAULT_LANG} · "
      f"{len(CROSS_LINGUAL_LEXICON)} cross-lingual concepts · reranking {RERANKING_LEVELS[RERANKING][0]}"
      + ("" if _snowball("es") else " · ⚠️ snowballstemmer missing: light stemmers for es/pt/de/it"))
````

````python
# @title ⚙️ T5 · Conversation memory { display-mode: "form" }
FOLLOW_UP_STARTS = ("and ", "and,", "what about", "how about", "same for", "same question", "same thing",
                    "also ", "for ", "et ", "et,", "pour ", "qu en est il", "qu'en est-il", "meme question",
                    "idem", "aussi ", "meme chose", "et si ",
                    "y ", "y,", "¿y ", "y para", "y en", "lo mismo", "tambien", "e ", "e,", "e para", "e no ", "e na ",
                    "e quanto", "o mesmo", "und ", "und fur", "ebenso")
INTERROGATIVE = re.compile(r"^(what|which|how|who|when|where|why|is |are |does |do |did |can |could |was |were |"
                           r"will |should |quel|quelle|quels|quelles|combien|comment|qui|quand|ou |pourquoi|"
                           r"est-ce|que |qu'|y a-t-il|¿|cual|cuales|cuanto|cuantos|como|donde|cuando|por que|que |"
                           r"qual|quais|quanto|quantos|onde|porque|o que|was |wie |wer |wann |welche)")
CATEGORIES = {
    "region": r"\b(coast|centre|center|north|south|littoral|nord|sud|costa|centro|norte|sur|sul)\b",
    "sex": r"\b(women|woman|men|man|female|male|femmes?|hommes?|mujer(?:es)?|hombres?|mulher(?:es)?|homens|homem|"
           r"frauen|manner)\b",
    "quarter": r"\b(q[1-4]|t[1-4]|(?:first|second|third|fourth) quarter|"
               r"(?:premier|deuxieme|second|troisieme|quatrieme) trimestre|"
               r"(?:primer|segundo|tercer|cuarto|primeiro|terceiro|quarto) trimestre)\b",
    "area": r"\b(urban|rural|cities|city|countryside|urbain|villes?|campagnes?|urbanos?|urbanas?|rurales|rurais)\b",
    "age": r"\b(young people|youth|adults|older people|jeunes|adultes|seniors|jovenes|jovens|adultos)\b",
    "year": r"\b((?:19|20)\d{2})\b",
    "indicator": r"\b((?:unemployment|activity|employment) rate|underemployment|informal employment|inflation|"
                 r"taux (?:de chomage|d'activite|d'emploi)|sous-emploi|emploi informel|"
                 r"tasa de (?:desempleo|actividad|empleo)|taxa de (?:desemprego|atividade|emprego)|inflacion|inflacao)\b",
}

def _aligned(text):
    """Lower case without accents, keeping the same length (same positions)."""
    return "".join((strip_accents(ch) or ch)[:1] for ch in text.lower())

def is_follow_up(question, history):
    if not history:
        return False
    q = _aligned(question).strip()
    if q.startswith(tuple(_aligned(s) for s in FOLLOW_UP_STARTS)):
        return True
    return len(analyse(question, detect_language(question, DEFAULT_LANG))) <= 2 and not INTERROGATIVE.match(q)

PROMPT_FOLLOW_UP = """You rewrite questions. Using the conversation history, rewrite the user's LAST question so
that it can be understood on its own, without the history. Do not answer the question. Return only the
rewritten question, in the same language as the user's question, on a single line."""

def rewrite_follow_up(question, history, use_model=True):
    """Returns (standalone question, rewritten?, method)."""
    if not is_follow_up(question, history):
        return question, False, None
    previous = history[-1]["standalone_question"]
    if use_model and LLM.available:
        try:
            turns = "\n".join(f"User: {h['standalone_question']}\nAssistant: {h['answer'][:200]}"
                              for h in history[-3:])
            text = LLM.generate(PROMPT_FOLLOW_UP, f"HISTORY\n{turns}\n\nLAST QUESTION\n{question}",
                                max_tokens=120).strip().strip('"«» ').splitlines()[0]
            if 5 < len(text) < 300:
                return text, True, "model"
        except Exception:
            pass
    result, q_al, replaced = previous, _aligned(question), 0
    for pattern in CATEGORIES.values():
        new = re.search(pattern, q_al)
        if not new:
            continue
        old = re.search(pattern, _aligned(result))
        if old:
            result = result[:old.start()] + question[new.start():new.end()] + result[old.end():]
            replaced += 1
    if not replaced:
        rest = re.sub(r"^\s*(and|what about|how about|same for|also|et|pour|qu'en est-il|aussi|¿?y|e|und|tambi[eé]n|tamb[eé]m)\b[\s,]*", "",
                      question, flags=re.I)
        result = previous.rstrip(" ?") + " — " + rest.rstrip(" ?") + "?"
    return result, True, "rules"
print("✅ T5 ready.")
````

````python
# @title ⚙️ T6 · The library: loading, indexing, answers, checks, feedback { display-mode: "form" }
REFUSALS = {
    "en": "Information not available in the documents.",
    "fr": "Information non disponible dans les documents.",
    "es": "Información no disponible en los documentos.",
    "pt": "Informação não disponível nos documentos.",
    "de": "Information in den Dokumenten nicht verfügbar.",
    "it": "Informazione non disponibile nei documenti.",
    "ar": "المعلومة غير متوفرة في الوثائق.",
    "sw": "Taarifa hii haipatikani katika nyaraka.",
}
REFUSAL = REFUSALS["en"]

def refusal_in(lang):
    return REFUSALS.get(lang, REFUSAL)

def strict_prompt(answer_lang):
    """The system prompt, with the refusal sentence and the answer language of this question."""
    language = LANGUAGE_NAMES.get(answer_lang, "the language of the question")
    return f"""You are the documentation assistant of a national statistics office.
You answer from excerpts of one or more PDF documents, which may be written in DIFFERENT languages.
Rules, in order of priority:
1. Answer ONLY from the numbered sources provided. If they do not contain the answer,
   write exactly: "{refusal_in(answer_lang)}" and nothing else.
2. Never use your general knowledge, even if you think you know the answer.
3. Copy figures exactly, with their unit, their period and the document they come from.
   A decimal comma ("7,8") and a decimal point ("7.8") write the same number.
4. If the question compares periods or documents, give each value with its source.
5. End every sentence with the number of its source in square brackets, for example [1].
6. Table sources are written "label — column: value; …".
7. Write the whole answer in {language}, whatever the language of the sources: translate faithfully,
   and keep acronyms, proper names and titles of programmes as written in the source.
8. Four sentences at most."""

STRICT_PROMPT = strict_prompt(DEFAULT_LANG)

COMPARISON = re.compile(r"\b(chang\w*|evolv\w*|evolu\w*|trend|between|entre|compar\w*|versus|vs|difference|"
                        r"increase\w*|decrease\w*|hausse|baisse|progress\w*|par rapport|cambi\w*|variac\w*|"
                        r"aument\w*|dismin\w*|diminu\w*|mudan\w*|frente a|em relacao)\b")
PRONOUN_START = re.compile(r"^(It|Its|They|Their|This|These|Il|Elle|Ils|Elles|Ce|Cette|Ces|Sa|Son|"
                           r"Él|Ella|Ellos|Ellas|Esta|Este|Estos|Su|Ele|Ela|Eles|Elas|Isso|Seu|Sua)\b")

@dataclass
class PDFDocument:
    id: int
    filename: str
    title: str
    data: bytes
    pages: int
    diagnosis: object
    elements: list
    info: dict
    lang: str = "en"    # detected at loading (or DOCUMENT_LANGUAGE)

    @property
    def short(self):
        t = re.sub(r"\s*\((fictitious|fictif|fictive|ficticia|ficticio)\)\s*$", "", self.title)
        return t if len(t) <= 42 else t[:40].rstrip() + "…"

    @property
    def scanned_pages(self):
        return self.diagnosis.loc[self.diagnosis["Status"] == "scanned", "Page"].tolist()

def is_refusal(text):
    t = strip_accents(text.lower())
    return any(m in t for m in ["information not available", "does not contain", "do not contain",
                                "no information", "i don't know", "i do not know", "cannot answer",
                                "does not mention", "not mentioned", "does not specify",
                                "information non disponible", "ne contient pas", "ne contiennent pas",
                                "pas d'information", "aucune information", "je ne sais pas",
                                "ne permet pas de repondre", "ne precise pas", "ne mentionne pas",
                                "informacion no disponible", "no contiene", "no contienen", "no hay informacion",
                                "no se menciona", "informacao nao disponivel", "nao contem", "nao ha informacao",
                                "nao menciona", "nicht verfugbar", "keine information"]
               + [strip_accents(r.lower()).rstrip(".") for r in REFUSALS.values()])

def _normalise_numbers(text):
    t = text.replace("\u00a0", " ").replace("\u202f", " ")
    t = re.sub(r"(?<=\d),(?=\d{3}\b)", "", t)       # 18,452,300 -> 18452300
    t = re.sub(r"(?<=\d) (?=\d{3}\b)", "", t)       # 18 452 300 -> 18452300
    return re.sub(r"(?<=\d),(?=\d)", ".", t)         # 7,8 -> 7.8

def check_figures(text, passages):
    """Every number in the answer must appear in at least one of the sources provided."""
    without_refs = re.sub(r"\[\d+\]", " ", _normalise_numbers(text))
    numbers = list(dict.fromkeys(re.findall(r"\d+(?:\.\d+)?", without_refs)))
    source = _normalise_numbers(" ".join(p.text for p, _ in passages))
    missing = [n for n in numbers if not re.search(rf"(?<![\d.]){re.escape(n)}(?!\d|\.\d)", source)]
    return {"numbers": len(numbers), "missing": missing}

def missing_terms(question, passage, lang=None):
    """Words of the question with no equivalent (synonym or translation) in the passage retained."""
    lang = lang or DEFAULT_LANG
    present = set(analyse(passage.index_text, passage_lang(passage)))
    absent = {t for t, g in concepts(question, lang, (passage_lang(passage),)) if not g & present}
    words = []
    for word in re.findall(r"[\w'-]+", question):
        if any(t in absent for t in analyse(word, lang)) and word not in words:
            words.append(word)
    return words

def relax_query(question, results, vocabulary, lang=None, targets=()):
    """Drop the words of the question that none of the retrieved passages contains.
    "the number of the dimensions in the framework proposed by AfDB" → "dimensions AfDB":
    question words and off-topic distractors disappear, the subject remains.

    Never relaxes when a word of the question is absent from the whole collection
    ("maize", "hospitals"): that word is the very signal of an out-of-scope question."""
    items = concepts(question, lang, targets)
    if not results or len(items) < 2:
        return ""
    if any(not (group & vocabulary) for _, group in items):
        return ""
    present = set()
    for p, _ in results:
        present |= set(analyse(p.index_text, passage_lang(p)))
    kept = [t for t, group in items if group & present]
    return " ".join(kept) if 1 < len(kept) < len(items) else ""

FEEDBACK = []
def record_feedback(turn, useful, comment="", channel="notebook"):
    row = {"timestamp": time.strftime("%Y-%m-%d %H:%M:%S"), "channel": channel, "useful": bool(useful),
           "question": turn.get("question", ""), "standalone_question": turn.get("standalone_question", ""),
           "answer": turn.get("answer", ""), "sources": " | ".join(turn.get("citations", [])),
           "comment": comment}
    FEEDBACK.append(row)
    path = OUTPUT_DIR / "user_feedback.csv"
    pd.DataFrame([row]).to_csv(path, mode="a", header=not path.exists(), index=False, encoding="utf-8-sig")
    return row

def _split_sentences(text):
    return [s.strip() for s in re.split(r"(?<=[.!?؟])\s+|\n", text) if len(s.strip()) > 15]

LIST_INTRO = re.compile(r"(:|following|as follows|are:|comprend|suivants?|suivantes?|siguientes?|seguintes?|folgende)\s*$", re.I)

def _best_sentence(items, passage, weight, text=None, kind="other"):
    """Best sentence of the passage, extended when it introduces a list or is very short.
    A "how many" question favours sentences that actually contain a number."""
    sentences = _split_sentences(text or passage.text)
    best, best_score = None, 0.0
    for n, sentence in enumerate(sentences):
        terms = set(analyse(sentence, passage_lang(passage)))
        if passage.kind == "table" and n > 0:
            terms |= set(analyse(sentences[0], passage_lang(passage)))   # a table row inherits its caption
        score = sum(weight(t, g) for t, g in items if g & terms)
        if kind == "count":
            score *= 1.6 if carries_a_number(sentence) else 0.6
        elif kind == "list":                                # "list the countries…" wants the list itself
            names = len(re.findall(r"(?<!^)(?<![.!?]\s)\b[A-Z][a-z]{2,}", sentence))
            score *= (1.35 if ENUMERATION.search(sentence) else 0.9) * (1 + 0.07 * min(names, 6))
        elif kind == "date" and re.search(r"\b(19|20)\d{2}\b|\b\d{1,2} (January|February|March|April|May|June|July|"
                                          r"August|September|October|November|December)\b", sentence):
            score *= 1.4
        if score > best_score:
            text = sentence
            if n > 0 and PRONOUN_START.match(text):
                text = sentences[n - 1] + " " + text
            if passage.kind == "table" and n > 0:
                text = f"{sentences[0]}: {text}"
            extra = n + 1
            while (extra < len(sentences) and len(text) < 420
                   and (LIST_INTRO.search(text) or len(text) < 130)):
                text += " " + sentences[extra]             # a list or a short sentence needs its follow-up
                extra += 1
            best, best_score = text, score
    return best, best_score

class Library:
    def __init__(self):
        self.documents, self.passages = [], []
        self.threshold, self.calibration, self.ready = 0.5, None, False
        self.language_counts = Counter()
        self.folder = ENV["root"] / "pdf"
        self.folder.mkdir(exist_ok=True)

    # --- loading -----------------------------------------------------------------------------
    BROWSER_HEADERS = {
        "User-Agent": ("Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) "
                       "Chrome/125.0.0.0 Safari/537.36"),
        "Accept": "application/pdf,text/html;q=0.9,*/*;q=0.8",
        "Accept-Language": "en-US,en;q=0.9,fr;q=0.8",
    }

    @staticmethod
    def direct_url(url):
        """Turn a sharing link into a direct download link (Google Drive, Dropbox, OneDrive)."""
        drive = re.search(r"(?:drive|docs)\.google\.com/(?:file/d/|open\?id=|uc\?id=|.*[?&]id=)([\w-]{20,})", url)
        if drive:
            return f"https://drive.usercontent.google.com/download?id={drive.group(1)}&export=download&confirm=t", f"gdrive_{drive.group(1)[:10]}.pdf"
        if "dropbox.com" in url:
            base = re.sub(r"[?&]dl=[01]", "", url)
            return base + ("&" if "?" in base else "?") + "dl=1", None
        if "1drv.ms" in url or "sharepoint.com" in url:
            return url + ("&" if "?" in url else "?") + "download=1", None
        return url, None

    def download(self, url, timeout=90):
        """Some institutional sites refuse anonymous scripts: send browser-like headers,
        then fall back to curl before giving up with a clear message.
        Sharing links (Google Drive, Dropbox, OneDrive) are converted to direct links."""
        url, forced_name = self.direct_url(url)
        name = forced_name or re.sub(r"[^\w.-]", "_", url.split("?")[0].rstrip("/").split("/")[-1]) or "document"
        target = self.folder / (name if name.lower().endswith(".pdf") else name + ".pdf")
        if target.exists() and target.stat().st_size > 1000:
            return target
        headers = dict(self.BROWSER_HEADERS)
        headers["Referer"] = "/".join(url.split("/")[:3]) + "/"
        try:
            request = urllib.request.Request(url, headers=headers)
            with urllib.request.urlopen(request, timeout=timeout) as response:
                content = response.read()
            if content[:4] == b"%PDF":
                target.write_bytes(content)
                return target
            first = "the address returned a web page, not a PDF (sharing link, login page or antivirus warning)"
            page = content[:4000].decode("utf-8", "ignore")
            token = re.search(r'name="(confirm|uuid)" value="([^"]+)"', page)
            if token:                                   # Google Drive antivirus warning for large files
                with urllib.request.urlopen(urllib.request.Request(
                        f"{url}&{token.group(1)}={token.group(2)}", headers=headers), timeout=timeout) as response:
                    content = response.read()
                if content[:4] == b"%PDF":
                    target.write_bytes(content)
                    return target
        except Exception as error:
            first = f"{type(error).__name__}: {str(error)[:80]}"
        if shutil.which("curl"):                    # curl negotiates TLS like a browser: often enough
            command = ["curl", "-sSL", "--max-time", str(timeout), "-A", headers["User-Agent"],
                       "-H", f"Accept: {headers['Accept']}", "-H", f"Referer: {headers['Referer']}",
                       "-o", str(target), url]
            if subprocess.run(command, capture_output=True).returncode == 0 and target.exists() \
                    and target.read_bytes()[:4] == b"%PDF":
                return target
            target.unlink(missing_ok=True)
        raise RuntimeError(f"download failed — {first}. Check that the link is public ('anyone with the link'), "
                           f"or download the PDF by hand and either set UPLOAD_PDF = True (Colab) "
                           f"or put its path in DOCUMENTS")

    def _resolve(self, sources):
        paths, failures = [], []
        for s in ([sources] if isinstance(sources, (str, Path)) else sources):
            s = str(s)
            try:
                if s == "example":
                    paths += create_example_documents(self.folder / "examples")
                elif s == "example-multilingual":
                    paths += create_example_documents(self.folder / "examples")
                    paths += create_multilingual_examples(self.folder / "examples")
                elif s.startswith(("http://", "https://")):
                    paths.append(self.download(s))      # a failure here must not stop the other documents
                elif Path(s).is_dir():
                    found = sorted(Path(s).rglob("*.pdf"))
                    paths += found
                    if not found:
                        failures.append((s, "no PDF in this folder"))
                elif Path(s).exists():
                    paths.append(Path(s))
                else:
                    failures.append((s, "source not found"))
            except Exception as error:
                failures.append((s.split("/")[-1][:60], str(error)[:200]))
        self._source_failures = failures
        return paths

    @staticmethod
    def _to_json(elements, info):
        def clean(t):
            return {"box": list(map(float, t["box"])), "headers": t["headers"], "rows": t["rows"]}
        return json.dumps({"info": info, "elements": [
            {"type": e.type, "text": e.text, "page": e.page, "box": list(map(float, e.box)),
             "level": e.level, "section": e.section,
             "table": clean(e.table) if getattr(e, "table", None) else None,
             "caption": getattr(e, "caption", ""), "extra": [[p, list(map(float, b))] for p, b in getattr(e, "extra", [])]}
            for e in elements]}, ensure_ascii=False)

    @staticmethod
    def _from_json(text):
        payload = json.loads(text)
        elements = []
        for item in payload["elements"]:
            e = Element(item["type"], item["text"], item["page"], tuple(item["box"]), item["level"], item["section"])
            if item["table"]:
                e.table = {"box": pymupdf.Rect(item["table"]["box"]), "headers": item["table"]["headers"],
                           "rows": item["table"]["rows"]}
                e.caption, e.extra = item["caption"], [(p, tuple(b)) for p, b in item["extra"]]
            elements.append(e)
        return elements, payload["info"]

    def _cache_key(self, data):
        settings = f"{REMOVE_HEADERS}|{EXTRACT_TABLES}|{USE_OCR}|{OCR_LANGUAGE}|{DROP_TABLE_OF_CONTENTS}|{MAX_PAGES_PER_DOCUMENT}|v2"
        return hashlib.sha1(data + settings.encode("utf-8")).hexdigest()[:16]

    def load(self, sources):
        added, errors = [], []
        known = {d.filename for d in self.documents}
        paths = self._resolve(sources)
        errors += getattr(self, "_source_failures", [])
        for number, path in enumerate(paths, 1):
            try:
                data = Path(path).read_bytes()
                if not data.startswith(b"%PDF"):
                    raise ValueError("not a PDF file")
                doc = pymupdf.open(stream=data, filetype="pdf")
                if doc.needs_pass:
                    raise ValueError("password-protected PDF")
                if Path(path).name in known:
                    continue
                start = time.time()
                cache_file = CACHE_DIR / f"extraction_{self._cache_key(data)}.json"
                if USE_CACHE and cache_file.exists():
                    elements, info = self._from_json(cache_file.read_text(encoding="utf-8"))
                    origin = "from cache"
                else:
                    elements, info = extract_structure(doc)
                    origin = "extracted"
                    if USE_CACHE:
                        cache_file.write_text(self._to_json(elements, info), encoding="utf-8")
                lang = (DOCUMENT_LANGUAGE if DOCUMENT_LANGUAGE in LANGUAGE_NAMES else
                        detect_language(" ".join(e.text for e in elements[:600])[:40000], DEFAULT_LANG, 3))
                print(f"  {number}/{len(paths)} · {Path(path).name[:48]} · {doc.page_count} pages · "
                      f"{origin} in {time.time() - start:.1f} s · language: {lang}")
                title = ((doc.metadata or {}).get("title") or "").strip()
                if not title:
                    title = elements[0].text if elements and elements[0].type == "heading" else Path(path).stem.replace("_", " ")
                d = PDFDocument(len(self.documents), Path(path).name, title, data, doc.page_count,
                                diagnose(doc), elements, info)
                d.path = str(path)
                d.lang = lang
                self.documents.append(d); added.append(d); known.add(d.filename)
            except Exception as error:
                errors.append((Path(path).name, str(error)[:120]))
        self.ready = False
        return added, errors

    # --- indexing ----------------------------------------------------------------------------
    def index(self):
        self.passages = []
        for d in self.documents:
            for p in make_passages(d, len(self.passages)):
                p.lang = passage_language(p.text, getattr(d, "lang", DEFAULT_LANG))
                self.passages.append(p)
        self.language_counts = Counter(passage_lang(p) for p in self.passages)
        if not self.passages:
            raise ValueError("No passages: check that the PDFs contain text (or turn on OCR).")
        start = time.time()
        self.units = build_search_units(self.passages)         # passages + small windows
        self.by_id = {p.id: p for p in self.passages}
        for u in self.units:                                   # a search window speaks its passage's language
            u.lang = passage_lang(self.by_id.get(u.parent_id, u))
        _CONCEPT_CACHE.clear()
        ACRONYMS.clear()
        ACRONYMS.update(build_acronyms(self.passages))
        build_vocabulary(self.passages)                        # for spelling tolerance
        self.bm25 = BM25Index(self.units)
        self.semantic = SemanticIndex(self.units)
        self.reranker = Reranker(self.bm25, languages=self.languages)
        print(f"  {len(self.passages)} passages / {len(self.units)} search units indexed in "
              f"{time.time() - start:.1f} s ({self.semantic.name}) · {len(ACRONYMS)} acronyms and "
              f"{sum(map(len, VOCABULARY_FORMS.values()))} words learned · languages "
              f"{', '.join(f'{l} ({n})' for l, n in self.language_counts.most_common())} · "
              f"calibrating the threshold…")
        start = time.time()
        self.calibrate()
        print(f"  threshold {self.threshold:.2f} calibrated in {time.time() - start:.1f} s")
        self.ready = True

    def calibrate(self):
        OFF_TOPIC = {
            "en": ["Which team won the last football World Cup?",
                   "What is the traditional recipe for couscous?",
                   "What will the weather be like tomorrow in the capital?",
                   "What is the price of maize per kilogram at the market?",
                   "How many measles vaccines were administered?",
                   "What is the national cocoa production?",
                   "Who wrote this year's best-selling novel?",
                   "How many hospitals are there in the North region?"],
            "fr": ["Quelle équipe a gagné la dernière coupe du monde de football ?",
                   "Quelle est la recette traditionnelle du couscous ?",
                   "Quel temps fera-t-il demain dans la capitale ?",
                   "Quel est le prix du maïs au kilogramme sur les marchés ?",
                   "Combien de vaccins ont été administrés contre la rougeole ?",
                   "Quelle est la production nationale de cacao ?",
                   "Qui est l'auteur du roman le plus vendu cette année ?",
                   "Combien d'hôpitaux compte la région du Nord ?"],
            "es": ["¿Qué equipo ganó el último Mundial de fútbol?",
                   "¿Cuál es la receta tradicional del cuscús?",
                   "¿Qué tiempo hará mañana en la capital?",
                   "¿Cuál es el precio del maíz por kilogramo en el mercado?",
                   "¿Cuántas vacunas contra el sarampión se administraron?",
                   "¿Cuál es la producción nacional de cacao?",
                   "¿Quién escribió la novela más vendida del año?",
                   "¿Cuántos hospitales hay en la región Norte?"],
            "pt": ["Que equipa ganhou o último Mundial de futebol?",
                   "Qual é a receita tradicional do cuscuz?",
                   "Como estará o tempo amanhã na capital?",
                   "Qual é o preço do milho por quilograma no mercado?",
                   "Quantas vacinas contra o sarampo foram administradas?",
                   "Qual é a produção nacional de cacau?",
                   "Quem escreveu o romance mais vendido do ano?",
                   "Quantos hospitais existem na região Norte?"],
        }
        # off-topic questions in every language of the collection: the threshold must hold in all of them
        off_topic = [q for l in ([l for l in self.languages if l in OFF_TOPIC] or ["en"]) for q in OFF_TOPIC[l]]
        import itertools
        per_document = [[(re.sub(r"^\s*[\dIVX]+[.)]?\s*", "", e.text), getattr(d, "lang", DEFAULT_LANG))
                         for e in d.elements if e.type == "heading" and e.level > 1] for d in self.documents]
        headings = [x for group in itertools.zip_longest(*per_document) for x in group if x]   # every document
        on_topic = [about_question(h, l) for h, l in dict.fromkeys(headings)][:15]
        rows = [{"Question": q, "Answer exists": True, "Confidence": round(self.confidence(q), 3)} for q in on_topic]
        rows += [{"Question": q, "Answer exists": False, "Confidence": round(self.confidence(q), 3)} for q in off_topic]
        self.calibration = pd.DataFrame(rows)
        ceiling = self.calibration.loc[~self.calibration["Answer exists"], "Confidence"].max()
        # on a long report, almost every word appears somewhere, so off-topic scores rise:
        # the threshold is capped to avoid refusing legitimate questions.
        self.calibrated_threshold = round(min(ceiling + 0.05, 0.7), 2)
        self.threshold = MANUAL_THRESHOLD if MANUAL_THRESHOLD is not None else self.calibrated_threshold

    # --- search and answers -------------------------------------------------------------------
    @property
    def languages(self):
        """Languages of the collection, the most frequent first: the language of every document, plus a
        passage language only if it weighs at least 5 % of the passages (a few stray passages must not
        trigger a translation of every question)."""
        total = max(1, sum(self.language_counts.values()))
        documents = {getattr(d, "lang", DEFAULT_LANG) for d in self.documents}
        kept = [l for l, n in self.language_counts.most_common() if l in documents or n / total >= 0.05]
        return kept or [DEFAULT_LANG]

    def queries_for(self, question, lang):
        """The question in each language of the collection: the original, plus a translation by the
        model for every other language (when a model is connected and TRANSLATE_QUERIES = True)."""
        queries = {lang: question}
        if TRANSLATE_QUERIES and LLM.available:
            for other in self.languages:
                if other != lang:
                    translation = translate_query(question, other)
                    if translation:
                        queries[other] = translation
        return queries

    def confidence(self, question):
        r = self.search(question, 1)
        return r[0][1] if r else 0.0

    def search(self, question, k=None, docs=None, kind=None, lang=None, queries=None):
        k = k or TOP_K
        lang = lang or question_language(question, self.languages)
        queries = dict(queries or {lang: question})
        queries.setdefault(lang, question)
        others = [l for l in self.languages if l != lang]
        allowed = set(range(len(self.documents))) if docs is None else set(docs)
        keep = lambda ranking: [(p, s) for p, s in ranking if p.doc_id in allowed]
        # BM25: the original question (plus its lexicon equivalents) over every unit, and each
        # translation over the units of its own language
        lexical = [keep(self.bm25.search(question, 40, lang, others))]
        for l, q in queries.items():
            if l != lang and q:
                lexical.append(keep([(u, s) for u, s in self.bm25.search(q, 40, l) if passage_lang(u) == l]))
        density = {}
        for ranking in lexical:
            for u, s in ranking:
                density[u.id] = max(density.get(u.id, 0.0), s)
        # multilingual embeddings already match the original question across languages: searching the
        # translations again would only add rankings leaning towards the twin document in another language
        semantic = [keep(self.semantic.search(question, 40, lang))]
        if not self.semantic.cross_lingual:
            for l, q in queries.items():
                if l != lang and q:
                    semantic.append(keep([(u, s) for u, s in self.semantic.search(q, 60, l) if passage_lang(u) == l][:40]))
        weights = [1.0] + [0.5] * (len(lexical) - 1) + [1.0] + [0.5] * (len(semantic) - 1)
        candidates = rrf_fusion(lexical + semantic, top=14 + 4 * (len(queries) - 1), weights=weights)
        ranked = self.reranker.rerank(question, candidates, density, kind or question_type(question), lang, queries)
        seen, mapped = set(), []
        for unit, score in ranked:                              # a window stands for its context passage
            parent = self.by_id.get(unit.parent_id, unit)
            if parent.id in seen:
                continue
            seen.add(parent.id)
            mapped.append((parent, score))
        return diversify(mapped, k, len(allowed) > 1)

    def answer(self, question, history=None, docs=None, k=None, threshold=None, guardrails=True, memory=True,
               lang=None):
        if not self.ready:
            self.index()
        start = time.time()
        _CONCEPT_CACHE.clear()                      # the glossary or the lexicon may have been edited
        threshold = self.threshold if threshold is None else threshold
        previous = history[-1].get("language") if history else None
        standalone, rewritten, method = (rewrite_follow_up(question, history or [], MODEL_FOLLOW_UP)
                                         if memory else (question, False, None))
        qlang = lang or question_language(question, self.languages, previous)
        answer_lang = qlang if ANSWER_LANGUAGE in ("question", "auto", None) else ANSWER_LANGUAGE
        standalone, corrections = fix_spelling(standalone, qlang)     # "coutries" -> "countries"
        kind = question_type(standalone)
        queries = self.queries_for(standalone, qlang)                 # the question in every language
        results = self.search(standalone, k, docs, kind, qlang, queries)
        conf = results[0][1] if results else 0.0
        relaxed = None
        if conf < threshold:
            # second pass: keep only the words the candidate passages really contain
            reduced = relax_query(standalone, results, set(self.bm25.vocabulary), qlang, self.languages)
            if reduced and reduced.lower() != standalone.lower():
                fallback = self.search(reduced, k, docs, kind, qlang)
                if fallback and fallback[0][1] > conf:
                    results, conf, relaxed = fallback, fallback[0][1], reduced
        alternative = None
        if docs is not None and set(docs) != set(range(len(self.documents))):
            best = self.search(standalone, 1, None, kind, qlang, queries)
            if best and best[0][0].doc_id not in set(docs) and best[0][1] >= max(conf + 0.15, threshold):
                alternative = best[0][0].doc
        missing = []
        if results:
            view_q, view_l = query_view(results[0][0], standalone, qlang, queries)
            missing = missing_terms(view_q, results[0][0], view_l)
        r = {"question": question, "standalone_question": standalone, "rewritten": rewritten,
             "rewrite_method": method, "confidence": conf, "passages": results, "reason": None, "relaxed": relaxed,
             "corrections": corrections, "chunking": CHUNK_STRATEGY, "reranking": self.reranker.mode,
             "language": qlang, "answer_language": answer_lang,
             "queries": {l: q for l, q in queries.items() if l != qlang},
             "source_languages": sorted({passage_lang(p) for p, _ in results}),
             "guardrails": guardrails, "time": time.strftime("%H:%M:%S"), "alternative": alternative,
             "missing_terms": missing, "check": None, "citations": [p.citation for p, _ in results], "kind": kind}
        if guardrails:
            years = set(re.findall(r"\b(?:19|20)\d{2}\b", standalone))
            content = " ".join(p.index_text for p, _ in results)
            absent_years = sorted(y for y in years if y not in content)
            if conf < threshold:
                r["reason"] = f"confidence {conf:.2f} below threshold {threshold:.2f}"
            elif absent_years:
                r["reason"] = f"year(s) {', '.join(absent_years)} not found in the passages"
            if r["reason"]:
                return {**r, "answer": refusal_in(answer_lang), "refused": True, "mode": "guardrail",
                        "duration_s": round(time.time() - start, 2)}
        text, mode = None, "extractive"
        if LLM.available:
            try:
                sources = "\n\n".join(
                    f"[{i}] ({p.citation} · language: {LANGUAGE_NAMES.get(passage_lang(p), passage_lang(p))}"
                    f"{', table' if p.kind == 'table' else ''})\n{self.context_of(p)}"
                    for i, (p, _) in enumerate(results, 1))
                text = LLM.generate(strict_prompt(answer_lang), f"SOURCES\n\n{sources}\n\nQUESTION\n{standalone}")
                mode = f"{LLM.name} · {LLM.model}"
            except Exception as error:
                r["reason"] = f"model unavailable ({str(error)[:80]}): extractive mode"
        if text is None:
            text = self._extract(standalone, results, kind, qlang, queries, answer_lang)
        refused = is_refusal(text)
        if not refused:
            r["check"] = check_figures(text, results)
        return {**r, "answer": text, "refused": refused, "mode": mode,
                "duration_s": round(time.time() - start, 2)}

    DEFINITION_QUESTION = re.compile(r"\b(stands? for|stand|mean|means|abbreviation|acronym|signifie|veut dire|"
                                     r"significa|siglas?|quer dizer|bedeutet)\b", re.I)

    def context_of(self, passage):
        """The passage plus its neighbours in the same section: a complete context for the model."""
        if not NEIGHBOUR_CONTEXT:
            return passage.text
        family = [p for p in self.passages if p.doc_id == passage.doc_id and p.section == passage.section]
        try:
            position = family.index(passage)
        except ValueError:
            return passage.text
        pieces, size, before, after = [passage.text], len(passage.text), position - 1, position + 1
        while size < CONTEXT_CHARS and (before >= 0 or after < len(family)):
            if before >= 0 and size + len(family[before].text) <= CONTEXT_CHARS:
                pieces.insert(0, family[before].text); size += len(family[before].text); before -= 1
            elif after < len(family) and size + len(family[after].text) <= CONTEXT_CHARS:
                pieces.append(family[after].text); size += len(family[after].text); after += 1
            else:
                break
        return " ".join(dict.fromkeys(pieces))

    def _extract(self, question, results, kind="other", lang=None, queries=None, answer_lang=None):
        lang = lang or DEFAULT_LANG
        queries = queries or {lang: question}
        answer_lang = answer_lang or lang
        # extractive mode cannot translate: a passage in another language is quoted as is, and labelled
        label = lambda p: ("" if passage_lang(p) == answer_lang
                           else f"({LANGUAGE_NAMES.get(passage_lang(p), passage_lang(p))} source) ")
        # "What does DIL stand for?" — answer from the acronyms learned in the documents
        if self.DEFINITION_QUESTION.search(question):
            for acronym in sorted(ACRONYMS, key=len, reverse=True):
                if re.search(rf"\b{re.escape(acronym)}\b", question):
                    expansion = ACRONYMS[acronym]
                    for j, (p, _) in enumerate(results):
                        sentence = next((s for s in _split_sentences(p.text)
                                         if expansion.lower() in s.lower() or f"({acronym})" in s), None)
                        if sentence:
                            return f"{acronym} — {expansion}. {sentence} [{j + 1}]"
                    return f"{acronym} — {expansion}."
        candidates = []
        for j, (p, _) in enumerate(results):
            q, l = query_view(p, question, lang, queries)       # the question in the passage's language
            items = concepts(q, l, (passage_lang(p),))
            sentence, score = _best_sentence(items, p, self.reranker.weight, kind=kind)   # quote the passage itself;
            # the expanded context (context_of) is only sent to the model, which needs completeness
            if sentence:
                candidates.append((score * (1 - 0.02 * j), j, p, sentence))
        if not candidates:
            return refusal_in(answer_lang)
        candidates.sort(key=lambda c: -c[0])
        score0, j0, p0, sentence0 = candidates[0]
        text = f"{label(p0)}{sentence0} [{j0 + 1}]"
        if COMPARISON.search(strip_accents(question.lower())):
            others = [c for c in candidates[1:] if c[2].doc_id != p0.doc_id and c[0] >= 0.5 * score0]
            if others:
                _, j1, p1, sentence1 = others[0]
                text += f" — {p1.doc}: {label(p1)}{sentence1} [{j1 + 1}]"
        return text

    # --- display ------------------------------------------------------------------------------
    def thumbnail(self, passage, **kw):
        doc = self.documents[passage.doc_id]
        first = passage.pages[0]
        return thumbnail_html(doc.data, first, [b for pg, b in passage.boxes if pg == first], **kw)

    def card_html(self, r, thumbnails=True):
        badges = pill("DECLINED", C["ochre"]) if r["refused"] else pill("ANSWER", C["dark_green"])
        badges += pill(f"confidence {r['confidence']:.2f}", C["teal"]) + pill(r["mode"], C["grey"])
        if r.get("language"):
            sources_lang = [l for l in r.get("source_languages", []) if l != r["language"]]
            badges += pill(f"🌐 {r['language']}" + (f" → sources in {', '.join(sources_lang)}" if sources_lang else ""),
                           C["teal"])
        if not r["guardrails"]:
            badges += pill("guardrails off", C["brick"])
        chk = r.get("check")
        if chk and chk["numbers"]:
            badges += (pill(f"figures checked {chk['numbers']}/{chk['numbers']}", C["dark_green"]) if not chk["missing"]
                       else pill(f"⚠ figure(s) not in sources: {', '.join(chk['missing'])}", C["brick"]))
        if r.get("relaxed"):
            badges += pill(f"relaxed search: {r['relaxed']}", C["ochre"])
        if r.get("corrections"):
            badges += pill("spelling: " + ", ".join(f"{a} → {b}" for a, b in r["corrections"]), C["teal"])
        notes = ""
        if r.get("queries"):
            notes += (f'<div style="font-size:12.5px;color:{C["teal"]};margin-top:6px;">🌐 Also searched as: '
                      + " · ".join(f'<b>{l}</b> <i>{_html.escape(q)}</i>' for l, q in r["queries"].items())
                      + '</div>')
        if r["rewritten"]:
            notes += (f'<div style="font-size:12.5px;color:{C["teal"]};margin-top:6px;">🧠 Understood as: '
                      f'<i>{_html.escape(r["standalone_question"])}</i> '
                      f'<span style="color:{C["slate"]};">({r["rewrite_method"]})</span></div>')
        if not r["refused"] and r.get("missing_terms"):
            notes += (f'<div style="font-size:12.5px;color:{C["ochre"]};margin-top:6px;">⚠️ Not found in the main '
                      f'source: <b>{_html.escape(", ".join(r["missing_terms"]))}</b> — check that the answer really '
                      f'addresses your question.</div>')
        if r.get("alternative"):
            notes += (f'<div style="font-size:12.5px;color:{C["teal"]};margin-top:6px;">💡 A more relevant answer is '
                      f'probably in “<b>{_html.escape(r["alternative"])}</b>”, excluded by the filter.</div>')
        reason = f'<div class="rg-note">Reason: {_html.escape(r["reason"])}</div>' if r.get("reason") else ""
        answer = re.sub(r"\[(\d+)\]", lambda m: f'<sup style="background:{C["mint"]};color:{C["dark_green"]};'
                        f'border-radius:4px;padding:0 4px;font-weight:700;">{m.group(1)}</sup>', _html.escape(r["answer"]))
        sources = []
        for i, (p, s) in enumerate(r["passages"], 1):
            thumb = self.thumbnail(p, dpi=55, width=105) if thumbnails and i <= 2 and not r["refused"] else ""
            sources.append(
                f'<div style="display:flex;gap:10px;align-items:flex-start;margin:6px 0;">{thumb}<div style="flex:1;">'
                f'<b>[{i}]</b> {pill(p.doc, C["dark_green"])}<b>{_html.escape(p.section or "start")}</b> · '
                f'p. {", ".join(map(str, p.pages))} · {p.kind} · {passage_lang(p)} · confidence {s:.2f}<br>'
                f'<span style="color:{C["slate"]};">“{_html.escape(" ".join(p.text.split())[:220])}…”</span></div></div>')
        return (f'<div class="rg rg-card" style="border-left:6px solid {C["ochre"] if r["refused"] else C["green"]};margin:0;">'
                f'<div>{badges}<span style="float:right;color:{C["slate"]};font-size:11px;">{r["time"]} · {r["duration_s"]} s</span></div>'
                f'{notes}<div style="font-size:14.5px;margin:10px 0;line-height:1.55;">{answer}</div>{reason}'
                f'<details {"open" if not r["refused"] else ""}><summary style="cursor:pointer;color:{C["dark_green"]};'
                f'font-weight:600;font-size:13px;">Sources ({len(r["passages"])})</summary>'
                f'<div class="rg-src">{"".join(sources) or "No source."}</div></details></div>')

    def show(self, r):
        display(HTML(f'{_CSS}<div class="rg" style="margin:10px 0;"><div style="font-weight:700;font-size:15px;'
                     f'margin-bottom:6px;">❓ {_html.escape(r["question"])}</div>{self.card_html(r)}</div>'))

    def summary(self):
        return pd.DataFrame([{"No.": d.id, "Document": d.title, "File": d.filename, "Pages": d.pages,
                              "Language": getattr(d, "lang", "—"),
                              "Scanned pages": ", ".join(map(str, d.scanned_pages)) or "—",
                              "Tables": sum(e.type == "table" for e in d.elements),
                              "Passages": sum(p.doc_id == d.id for p in self.passages) if self.passages else "—"}
                             for d in self.documents])

LIB = Library()
_DEMO_HISTORY = []

ABOUT_TEMPLATES = {"en": "What does the document say about: {}?", "fr": "Que dit le document sur : {} ?",
                   "es": "¿Qué dice el documento sobre: {}?", "pt": "O que diz o documento sobre: {}?",
                   "de": "Was sagt das Dokument über: {}?", "it": "Cosa dice il documento su: {}?"}

def about_question(heading, lang):
    """A question about a section, written in the language of that section."""
    return ABOUT_TEMPLATES.get(lang, ABOUT_TEMPLATES["en"]).format(heading)

def using_examples():
    return bool(DOCUMENTS) and set(map(str, DOCUMENTS)) <= {"example", "example-multilingual"}

def using_multilingual_examples():
    return using_examples() and "example-multilingual" in map(str, DOCUMENTS)

def suggested_questions(n=5):
    """Questions that fit the documents actually loaded (used by the demos and the interfaces)."""
    if using_multilingual_examples():
        return ["What is the unemployment rate for women in the Coast region?", "And for men?",
                "Quel est le taux de chômage des jeunes ?",
                "What was consumer price inflation in the first quarter of 2026?",
                "¿Qué proporción de hogares tiene acceso a internet?",
                "What is the price of maize?"][:n]
    if using_examples():
        return ["What is the unemployment rate for women in the Coast region?", "And for men?",
                "How did the unemployment rate change between the fourth quarter of 2025 and the first quarter of 2026?",
                "What is the average household size?", "What is the price of maize?"][:n]
    per_document = [[(re.sub(r"^\s*[\dIVX]+[.)]?\s*", "", e.text).strip(), getattr(d, "lang", DEFAULT_LANG))
                     for e in d.elements
                     if e.type == "heading" and e.level > 1 and e.page > 2 and 12 < len(e.text) < 70
                     and len(e.text.split()) >= 2 and e.text.lower() not in d.title.lower()] for d in LIB.documents]
    import itertools
    headings = [x for group in itertools.zip_longest(*per_document) for x in group if x]
    questions = [about_question(h, l) for h, l in dict.fromkeys(headings)][:max(1, n - 2)]
    if ACRONYMS:
        questions.append(f"What does {sorted(ACRONYMS, key=lambda a: -len(ACRONYMS[a]))[0]} stand for?")
    questions.append("What is the price of maize per kilogram?")       # out of scope, should be declined
    return questions[:n]

def ask(question, docs=None, memory=True, lang=None):
    """Ask the library a question, in any language, remembering the previous questions.
    `lang` forces the language of the question ("fr", "es"…); by default it is detected."""
    r = LIB.answer(question, history=_DEMO_HISTORY if memory else None, docs=docs, lang=lang)
    _DEMO_HISTORY.append(r)
    LIB.show(r)
    return r
print("✅ T6 ready · toolbox complete.")
````

````python
# @title ⚙️ T7 · The visible chain, the comparator and the benchmark { display-mode: "form" }

# ── 1. The chain, drawn ────────────────────────────────────────────────────────────────────
PIPELINE = [
    ("1", "Question", "typed by the user", "grey"),
    ("2", "Follow-up", "rewritten as a standalone question", "teal"),
    ("3", "Language & spelling", "language detected, question translated into each language of the collection, typos corrected", "teal"),
    ("4", "Normalisation", "lower case, accents, stems, question words removed", "grey"),
    ("5", "Expansion", "glossary + acronyms learned from the documents", "grey"),
    ("6", "BM25", "exact words, on every search unit", "green"),
    ("7", "Semantic", "meaning, on every search unit", "green"),
    ("8", "Fusion", "the two rankings merged (RRF)", "green"),
    ("9", "Reranking", "order decided by the level you chose", "gold"),
    ("10", "Windows → passages", "each unit gives back the passage to read", "gold"),
    ("11", "Guardrails", "threshold, year check, refusal", "ochre"),
    ("12", "Answer", "written by the model, or extracted", "dark_green"),
    ("13", "Checks", "figures, missing terms, filter", "dark_green"),
]

def show_pipeline():
    """The whole chain with the settings currently in force."""
    chosen = {"3": " · ".join(f"{l} {n}" for l, n in LIB.language_counts.most_common()) if getattr(LIB, "language_counts", None) else "",
              "9": RERANKING_LEVELS[LIB.reranker.mode][0] if getattr(LIB, "reranker", None) else RERANKING,
              "6": f"{len(getattr(LIB, 'units', []))} units",
              "7": getattr(getattr(LIB, "semantic", None), "name", "—"),
              "11": f"threshold {LIB.threshold:.2f}" if getattr(LIB, "threshold", None) else "",
              "12": f"{LLM.name} · {LLM.model}" if LLM.available else "extractive mode"}
    boxes = ""
    for number, title, subtitle, colour in PIPELINE:
        note = chosen.get(number, "")
        boxes += (
            f'<div style="flex:1 1 150px;min-width:150px;background:#fff;border:1px solid {C["sage"]};'
            f'border-top:3px solid {C[colour]};border-radius:10px;padding:8px 10px;">'
            f'<div style="font-size:10.5px;color:{C[colour]};font-weight:800;">STEP {number}</div>'
            f'<div style="font-size:13px;font-weight:700;color:{C["ink"]};">{title}</div>'
            f'<div style="font-size:11.5px;color:{C["slate"]};line-height:1.35;">{subtitle}</div>'
            + (f'<div style="font-size:11px;color:{C["dark_green"]};font-weight:700;margin-top:3px;">{_html.escape(str(note))}</div>' if note else "")
            + '</div>')
    display(HTML(
        f'{_CSS}<div class="rg"><div class="rg-title">🔗 The RAG chain, as configured right now</div>'
        f'<div style="display:flex;gap:8px;flex-wrap:wrap;">{boxes}</div>'
        f'<div class="rg-note">Chunking: <b>{CHUNKING_LEVELS[CHUNK_STRATEGY][0]}</b> · '
        f'Reranking: <b>{RERANKING_LEVELS[RERANKING][0]}</b> · Passages given to the answer: <b>{TOP_K}</b></div></div>'))

# ── 2. The chain, traced on a real question ────────────────────────────────────────────────
def explain(question, docs=None, k=None):
    """Run one question through every stage and show what each stage changed."""
    k = k or TOP_K
    steps = []
    qlang = question_language(question, LIB.languages)
    others = [l for l in LIB.languages if l != qlang]
    standalone, corrections = fix_spelling(question, qlang)
    kind = question_type(standalone)
    queries = LIB.queries_for(standalone, qlang)
    translated = "".join(f"<br>→ <b>{l}</b>: {_html.escape(q)}" for l, q in queries.items() if l != qlang)
    steps.append(("3 · Language", f"question in <b>{LANGUAGE_NAMES.get(qlang, qlang)}</b> · collection: "
                  + ", ".join(f"{l} ({n})" for l, n in LIB.language_counts.most_common())
                  + (translated or ("<br>no translation (no model): the lexicon links the languages" if others else ""))))
    if corrections:
        steps.append(("3 · Spelling", ", ".join(f"<b>{a}</b> → {b}" for a, b in corrections)))
    effect = {"count": "passages and sentences carrying a real quantity are favoured",
              "list": "passages and sentences carrying an enumeration are favoured",
              "date": "passages and sentences carrying a year are favoured",
              "definition": "the acronyms learned from the documents are checked first",
              "other": "no special treatment"}[kind]
    steps.append(("Question type", f"<b>{kind}</b> — {effect}"))
    terms = concepts(standalone, qlang, others)
    steps.append(("4 · Terms kept", ", ".join(f"{t} <span style='color:#5E6964'>({LIB.reranker.weight(t, g):.1f})</span>"
                                              for t, g in terms) or "—"))
    added = expand_query(standalone, qlang)[1]
    steps.append(("5 · Expansion", "<br>".join(f"<b>{a}</b> → {b}" for a, b in added) or "nothing to add"))
    lexical = LIB.bm25.search(standalone, 40, qlang, others)
    semantic = LIB.semantic.search(standalone, 40, qlang)
    line = lambda units: "<br>".join(f"{i}. <b>{u.section[:34] or '—'}</b> · {passage_lang(u)} · p. {u.pages[0]} "
                                     f"<span style='color:#5E6964'>({s:.2f})</span>"
                                     for i, (u, s) in enumerate(units[:4], 1)) or "nothing found"
    steps.append(("6 · BM25 (exact words)", line(lexical)))
    steps.append((f"7 · {LIB.semantic.name}", line(semantic)))
    fused = rrf_fusion([lexical, semantic], top=14)
    steps.append(("8 · Fusion", line(fused)))
    reranked = LIB.reranker.rerank(standalone, fused, {u.id: s for u, s in lexical}, kind, qlang, queries)
    steps.append((f"9 · Reranking · {LIB.reranker.name}",
                  "<br>".join(f"{i}. <b>{u.section[:34] or '—'}</b> · p. {u.pages[0]} "
                              f"<span style='color:#5E6964'>(coverage {s:.2f}"
                              + (", answer type ✓" if LIB.reranker.answer_type_bonus(kind, u) else "") + ")</span>"
                              for i, (u, s) in enumerate(reranked[:4], 1))))
    r = LIB.answer(question, docs=docs, k=k)
    steps.append(("10 · Passages returned", "<br>".join(f"{i}. {p.citation}" for i, (p, _) in enumerate(r["passages"], 1))))
    steps.append(("11 · Guardrails", f"confidence <b>{r['confidence']:.2f}</b> vs threshold <b>{LIB.threshold:.2f}</b> → "
                                     + ("<b style='color:#D49A00'>refusal</b>" if r["refused"] else "<b style='color:#00704A'>answer</b>")
                                     + (f" · {r['reason']}" if r.get("reason") else "")))
    rows = "".join(f'<tr><td style="white-space:nowrap;vertical-align:top;padding:6px 12px 6px 0;font-weight:700;'
                   f'color:{C["dark_green"]};font-size:12.5px;">{name}</td>'
                   f'<td style="padding:6px 0;font-size:12.5px;line-height:1.5;">{value}</td></tr>'
                   for name, value in steps)
    display(HTML(f'{_CSS}<div class="rg rg-card"><div class="rg-title">🔍 {_html.escape(question)}</div>'
                 f'<table style="width:100%;border-collapse:collapse;">{rows}</table></div>'))
    LIB.show(r)
    return r

# ── 3. The five chunking levels, on the same text ──────────────────────────────────────────
def compare_chunking(document=None, levels=("fixed", "sentences", "sections", "hybrid", "smart")):
    """Same document, five levels: how many pieces, how long, and what the first one looks like."""
    global CHUNK_STRATEGY
    document = document or LIB.documents[0]
    keep, rows = CHUNK_STRATEGY, []
    try:
        for level in levels:
            CHUNK_STRATEGY = level
            start = time.time()
            passages = make_passages(document)
            units = build_search_units(passages, level)
            lengths = [len(p.text) for p in passages] or [0]
            first = next((p for p in passages if len(p.text) > 200), passages[0] if passages else None)
            rows.append({"Level": CHUNKING_LEVELS[level][0],
                         "Passages": len(passages), "Search units": len(units),
                         "Average length": int(np.mean(lengths)),
                         "Knows its section": "✅" if any(p.section for p in passages) else "—",
                         "Seconds": round(time.time() - start, 2),
                         "First passage": _html.escape((first.text[:150] if first else "")) + "…"})
    finally:
        CHUNK_STRATEGY = keep
    show_table(pd.DataFrame(rows), title=f"✂️ The five levels on “{_html.escape(document.short)}”",
               note="Levels 1 and 2 ignore headings and tables, so a passage no longer knows where it comes from. "
                    "Levels 4 and 5 index a second, finer granularity — that is why they have more search units "
                    "than passages.")

# ── 4. The benchmark: which combination answers YOUR questions best ────────────────────────
def probe_questions(n=12, seed=0):
    """Questions whose answer is known, built from the documents when you have not written a test
    set. The words of ONE sentence must bring back a LATER sentence of the same topic: a chunking
    that cuts a topic in two, or that loses its section, fails the probe. It is only a proxy —
    write your own EVALUATION_QUESTIONS for a measurement you can trust."""
    import random
    rng = random.Random(seed)
    probes = []
    for minimum, gap in ((500, 3), (320, 2), (180, 2)):     # long reports first, small ones still work
        if len(probes) >= max(3, n // 2):
            break
        pool = [p for p in LIB.passages if p.kind == "text" and len(p.text) > minimum]
        if not pool:
            continue                                   # a short collection: try smaller passages
        probes = []
        for p in rng.sample(pool, len(pool)):
            sentences = [s for s in _split_sentences(p.text) if 40 < len(s) < 260]
            if len(sentences) < gap:
                continue
            question = " ".join(w for w in sentences[0].split() if len(w) > 3)[:170]
            target = sentences[-1].split()
            if len(target) < 6 or len(question.split()) < 5:
                continue
            probes.append((question, [" ".join(target[:6])]))
            if len(probes) == n:
                break
    return probes

def benchmark(chunking=("fixed", "sentences", "sections", "hybrid", "smart"),
              reranking=("smart",), questions=None, k=None, show_questions=False):
    """Re-index with each combination and measure it on questions whose answer is known.
    Answer found = the expected text is among the passages given to the model."""
    global CHUNK_STRATEGY, RERANKING
    k = k or TOP_K
    questions = questions or globals().get("EVALUATION_QUESTIONS") or probe_questions()
    if not questions:
        callout("No evaluation questions: load some documents first.", "warning")
        return
    if show_questions:
        show_table(pd.DataFrame([{"Question": q, "Expected in the context": " | ".join(e)} for q, e in questions]),
                   title=f"The {len(questions)} questions used by the benchmark")
    keep_chunk, keep_rank, keep_context, rows = CHUNK_STRATEGY, RERANKING, NEIGHBOUR_CONTEXT, []
    try:
        for level in chunking:
            for mode in reranking:
                CHUNK_STRATEGY, RERANKING = level, mode
                globals()["NEIGHBOUR_CONTEXT"] = keep_context and level == "smart"
                start = time.time()
                LIB.reranker = None
                LIB.index()
                build = time.time() - start
                found = first = 0
                start = time.time()
                for question, expected in questions:
                    results = LIB.search(question, k)
                    context = " ".join(LIB.context_of(p) for p, _ in results).lower()
                    found += any(e.lower() in context for e in expected)
                    if results:
                        first += any(e.lower() in LIB.context_of(results[0][0]).lower() for e in expected)
                rows.append({"Chunking": CHUNKING_LEVELS[level][0], "Reranking": RERANKING_LEVELS[mode][0],
                             "Passages": len(LIB.passages), "Units": len(LIB.units),
                             "Answer in the context": f"{found}/{len(questions)}",
                             "Right passage first": f"{first}/{len(questions)}",
                             "Indexing (s)": round(build, 1),
                             "Per question (ms)": int(1000 * (time.time() - start) / len(questions)),
                             "_score": (found, first)})
    finally:
        CHUNK_STRATEGY, RERANKING = keep_chunk, keep_rank
        globals()["NEIGHBOUR_CONTEXT"] = keep_context
        LIB.reranker = None
        LIB.index()
    table = pd.DataFrame(rows)
    best = table["_score"].max()
    table["Best"] = ["🏆" if s == best else "" for s in table["_score"]]
    show_table(table.drop(columns="_score"), title="🏁 Benchmark on your own documents",
               note="Measured on this collection only: re-run it on yours before trusting any ranking. "
                    "A combination that wins by one question is a tie, not a winner.")
    return table

print("✅ T7 ready · show_pipeline(), explain(\"…\"), compare_chunking(), benchmark().")
````

<div style="color:#5E6964;font-size:14px;line-height:1.45;">Headings, repeated headers, hyphenated words, tables and OCR.</div></div>

| Operation | Principle | Why |
|---|---|---|
| **Detect headings** | A line written clearly larger than the body text | Each passage knows its **section** |
| **Remove running headers and footers** | A line repeated at the top or bottom of at least half the pages | "Page 3" pollutes every passage |
| **Re-join hyphenated words** | "statis-⏎tics" → "statistics" | Search finds the whole word |
| **Linearise tables** | Each row becomes "Label — column: value; …" | A number on its own is meaningless |
| **OCR** (optional) | Text recognition on scanned pages | Makes scanned pages visible |

**NFKC** normalisation also replaces typographic ligatures ("ﬁ", "ﬀ") with ordinary letters: without it, "ﬁle" would never match "file".

The page below shows what the extractor recognised: **green** for headings, **grey** for paragraphs, **ochre** for tables, **red strikethrough** for removed headers and footers.

Chunking is the single decision that most affects the answers. Cut too small and a passage no longer contains enough to answer; cut too large and the search can no longer tell passages apart. The notebook offers **five levels**, from the naive baseline to the one that measures best. Change `CHUNK_STRATEGY` in the settings and run part A again.

| `CHUNK_STRATEGY` | What it does | What it costs you |
|---|---|---|
| `"fixed"` | **1 · Fixed blocks.** Cuts every `MAX_CHARS` characters. What almost every tutorial does. | Sentences, tables and headings are cut in the middle. A passage does not know which section it belongs to. |
| `"sentences"` | **2 · Sentence windows.** Groups N whole sentences, sliding. | Nothing is cut mid-sentence, but the passages are short and section-blind: the search finds fragments that cannot answer on their own. |
| `"sections"` | **3 · Structural passages.** Follows the headings found in part B2; each table is a passage of its own; the heading path travels with the text. | The unit of search is also the unit of reading, so one long section competes as a single block. |
| `"hybrid"` | **4 · Two granularities.** Indexes the structural passages **and** small sentence windows pointing back to them. | Twice the units to index (a few seconds more). |
| `"smart"` | **5 · Two granularities, tables and context.** Adds every table row as its own unit, and sends the neighbouring passages of the section along with the answer. | The most units and the slowest indexing — still only seconds on an 80-page report. |

### The five rules that levels 3 to 5 obey

| Rule | Setting | Why |
|---|---|---|
| Whole sentences, up to a size limit | `MAX_CHARS = 700` | A cut sentence can be retrieved but not quoted |
| One sentence repeated between passages | `OVERLAP_SENTENCES = 1` | The answer often sits astride a boundary |
| Short passages merged with the previous one | `MIN_CHARS = 280` | An isolated line has no context and answers nothing |
| Full heading path in the search text | — | "2. Strategy › 2.4 Focus Areas" separates sections with similar wording |
| Tables kept apart, rows carrying their caption | `EXTRACT_TABLES` | A number is meaningless without its row and column labels |

> 💡 **Measured, not assumed.** On the two institutional reports used to tune this notebook (66 and 81 pages, 12 questions with known answers):
>
> | Level | Answer in the context | Right passage first |
> |---|:-:|:-:|
> | 1 · Fixed blocks | 11/12 | 8/12 |
> | 2 · Sentence windows | 12/12 | **6/12** |
> | 3 · Structural passages | 11/12 | 10/12 |
> | 4 · Two granularities | **12/12** | **10/12** |
> | 5 · Two granularities, tables and context | **12/12** | **10/12** |
>
> The lesson that surprises most people: **indexing small chunks alone is the worst option** (level 2). A two-sentence window rarely contains all the words of a question, so the lexical search loses it. Small chunks help only when a larger passage stands behind them — that is levels 4 and 5.
>
> ⚠️ These numbers come from **two** reports and **twelve** questions. They are enough to separate very different strategies, not to prove that one level beats another by a single question. Run the benchmark below on your own documents.

### The thirteen steps, and where you can act

| # | Step | What it does | Setting |
|:-:|---|---|---|
| 1 | **Question** | as typed by the user | — |
| 2 | **Follow-up** | "And for men?" is rewritten as a standalone question (B7) | `MODEL_FOLLOW_UP` |
| 3 | **Language & spelling** | the language of the question is detected and, with a model, the question is translated into every language of the collection (B10); a word the documents never use is replaced by the closest one **of the same language**: *coutries* → *countries* | `QUESTION_LANGUAGE`, `TRANSLATE_QUERIES` |
| 4 | **Normalisation** | lower case, accents, word stems; question words ("number of", "list", "stand for") are dropped | `STOPWORDS` |
| 5 | **Expansion** | domain glossary, plus the acronyms learned from the documents | `GLOSSARY` |
| 6 | **BM25** | exact words, over every search unit | `CHUNK_STRATEGY` |
| 7 | **Semantic** | meaning rather than words | `USE_EMBEDDINGS` |
| 8 | **Fusion** | the two rankings merged (Reciprocal Rank Fusion) | — |
| 9 | **Reranking** | the order is decided here | `RERANKING` |
| 10 | **Windows → passages** | each small unit hands back the passage a human should read | `CHUNK_STRATEGY` |
| 11 | **Guardrails** | threshold, year check, refusal (B5) | `MANUAL_THRESHOLD` |
| 12 | **Answer** | written by the model, or extracted from the passage | Groq / Ollama |
| 13 | **Checks** | figures verified, terms not found, better answer excluded by a filter (B9) | — |

Step 7 is worth a word: **without `sentence-transformers` the notebook falls back to an offline approximation (LSA on character n-grams)**, which matches spellings rather than meaning. It never fails, but it cannot bridge "participating in the assessment" and "the Bank has conducted diagnostic missions in". ### The four levels of reranking

The fused ranking is rarely the right order. Reranking re-scores the shortlist; `RERANKING` chooses how.

| `RERANKING` | What it does | Cost |
|---|---|---|
| `"none"` | **0 · None.** Keeps the fusion order. | Free |
| `"coverage"` | **1 · Question coverage.** Share of the question's weighted terms present in the passage. A rare word weighs more than a frequent one; a word absent from the whole collection weighs most, which is what makes a refusal possible. | Free |
| `"smart"` | **2 · Coverage, density and answer type.** On a long report many passages cover the whole question, so equal scores are separated by **what the question asks for** — a "how many" question prefers a passage carrying a real quantity (not "see Figure 8"), a "list" question one carrying an enumeration — then by term density. | Free |
| `"cross-encoder"` | **3 · Neural cross-encoder.** A model reads the question and the passage *together* and scores the pair. The most accurate, and the slowest: it needs `sentence-transformers` and a few hundred milliseconds per question. | ~0.3–1 s/question |

>
> | Reranking | Answer in the context | Right passage first |
> |---|:-:|:-:|
> | 0 · None | 10/12 | 6/12 |
> | 1 · Question coverage | 12/12 | 8/12 |
> | 2 · Coverage, density and answer type | **12/12** | **10/12** |
>

Three guardrails decide whether the assistant may answer:

1. **Confidence threshold**: if the best passage is not relevant enough, the assistant declines **without calling the model**;
2. **Year check**: a year mentioned in the question must appear in the passages;
3. **Strict prompt**: answer only from the sources, cite them, and decline with a fixed sentence otherwise.

### How the threshold is calculated

**Confidence of a question** = weight of the question's words found in the best passage ÷ weight of all the question's words. Rare words weigh more; words found **nowhere** in the documents (such as "hospitals") weigh the most, because they signal an off-topic question.

**Calibration**: the notebook asks two series of reference questions — questions built from the **section headings** (the answer exists) and **off-topic** questions (no answer). The threshold is set just above the best off-topic score:

> **Threshold = best off-topic confidence + 0.05** (capped at 0.95)

> ⚠️ This threshold is a starting point: adjust it in the interfaces if the assistant declines too often, or not often enough. To fix it yourself: `MANUAL_THRESHOLD = 0.5`.

````python
cal = LIB.calibration
fig, ax = plt.subplots(figsize=(8.6, 2.8))
for rank, (exists, colour) in enumerate([(False, C["brick"]), (True, C["green"])]):
    v = cal.loc[cal["Answer exists"] == exists, "Confidence"]
    ax.scatter(v, np.full(len(v), rank) + np.linspace(-0.13, 0.13, len(v)), s=95, color=colour,
               alpha=0.85, edgecolor="white", linewidth=1.2, zorder=3)
ax.axvline(LIB.threshold, color=C["ochre"], linewidth=2.2, linestyle="--")
ax.axvspan(0, LIB.threshold, color=C["brick"], alpha=0.05); ax.axvspan(LIB.threshold, 1.02, color=C["green"], alpha=0.05)
ax.text(LIB.threshold, 1.55, f"  THRESHOLD = {LIB.threshold:.2f}", color=C["ochre"], fontweight="bold", va="center")
ax.set_yticks([0, 1], ["Off-topic", "Document headings"]); ax.set_xlim(0, 1.02); ax.set_ylim(-0.6, 1.8)
ax.set_xlabel("Confidence of the best passage"); ax.set_title("Threshold calibration on your documents")
plt.tight_layout(); fig.savefig(OUTPUT_DIR / "threshold_calibration.png", dpi=160); plt.show()
show_table(cal.sort_values("Confidence", ascending=False), title="Reference questions")

for q in ["What is the price of maize per kilogram?", "What will the unemployment rate be in 2030?"]:
    LIB.show(LIB.answer(q))
````

A follow-up question makes no sense on its own: "And for men?" contains no indicator. Before searching, the assistant **rewrites it as a standalone question**:

| Method | Principle | When |
|---|---|---|

A statistics office rarely works in one language: the continental strategy exists in English and French, the national survey is in Portuguese, the partner's report is in Spanish. A naive RAG fails in two ways: **it analyses every text with the rules of one language** (French stems applied to English, decimal commas misread), and **it only finds passages written in the language of the question** — an English question never retrieves the French note that holds the answer.

### Every text is read in its own language

| Step | What happens | Setting |
|---|---|---|
| **Document language** | Detected from its function words (*the, of* / *le, des* / *el, los* / *o, da*…) | `DOCUMENT_LANGUAGE = "auto"` |
| **Passage language** | A long passage of a bilingual report may differ from its document | `DETECT_PASSAGE_LANGUAGE` |
| **Question language** | Detected on every question; a short follow-up inherits the previous one | `QUESTION_LANGUAGE` |
| **Stop words and stems** | Each passage and each question is analysed with the resources of its language (Snowball stemmers for Spanish, Portuguese, German, Italian, Arabic) | — |
| **Numbers** | `7,8` (French) and `7.8` (English), `18 452 300` and `18,452,300` become the same token | — |
| **Spelling tolerance** | A typo is corrected only against words of the **same** language | — |

### Three bridges between languages

| Bridge | How it works | Needs |
|---|---|---|
| 🧠 **Multilingual embeddings** | The default model (`paraphrase-multilingual-MiniLM-L12-v2`) places a question and a passage with the same meaning close together, whatever their language | `sentence-transformers` (the offline LSA fallback is **not** cross-lingual) |
| 🔁 **Query translation** | The model translates the question into each language of the collection; each translation is searched in the passages of its language, and the rankings are fused | A model (Groq, Ollama…) and `TRANSLATE_QUERIES = True` — one short call per extra language |
| 📖 **Cross-lingual lexicon** | `CROSS_LINGUAL_LEXICON` links ~60 statistical concepts in English, French, Spanish and Portuguese (*unemployment · chômage · desempleo · desemprego*). It works **with no model and no Internet** | Nothing — extend it with your own domain |

The **confidence score** (B5) is measured with the version of the question written in the language of each passage, so a single threshold holds across languages; calibration uses off-topic questions **in every language of the collection**. The model then writes the answer in `ANSWER_LANGUAGE` — by default the language of the question — keeping figures, acronyms and programme titles as written in the source. The `cross-encoder` reranker switches automatically to a **multilingual** model (mMARCO).

> ⚠️ **Twin documents in two languages.** When the same report exists in English and French, both versions answer. `PREFER_QUESTION_LANGUAGE = True` favours the version in the language of the question at equal score; the document filter lets you keep only one.

> 💡 **Less-resourced languages** (Kinyarwanda, Wolof, Amharic…) are indexed without stems or stop words. Use a stronger multilingual embedding model (`intfloat/multilingual-e5-base`, `BAAI/bge-m3`) and a language model for translation, and add their Tesseract code to `OCR_LANGUAGE` for scanned pages.

````python
import ipywidgets as widgets

HISTORY = []
SUGGESTIONS = suggested_questions(6)

def _header():
    model_label = f"{LLM.name} · {LLM.model}" if LLM.available else "extractive mode"
    return (f'{_CSS}<div class="rg" style="background:linear-gradient(120deg,{C["forest"]} 0%,{C["green"]} 100%);'
            f'border-radius:14px;padding:14px 20px;color:#fff;"><div style="color:{C["gold"]};font-size:11px;'
            f'font-weight:700;letter-spacing:2.5px;">DOCUMENT ASSISTANT</div>'
            f'<div style="font-size:20px;font-weight:800;margin:2px 0;">💬 {len(LIB.documents)} document(s) · '
            f'{len(LIB.passages)} passages · 🌐 {", ".join(LIB.languages)}</div><div style="font-size:12.5px;color:#E6F6EE;">'
            f'ask in any language · model: '
            f'{_html.escape(model_label)} · search: {_html.escape(LIB.semantic.name)}</div></div>')

def _thread():
    if not HISTORY:
        return (f'{_CSS}<div class="rg rg-card" style="text-align:center;color:{C["slate"]};padding:26px;">'
                'Ask a question, or click a suggestion.</div>')
    bubbles = "".join(
        f'<div style="display:flex;justify-content:flex-end;margin:12px 0 6px 0;"><div style="background:{C["dark_green"]};'
        f'color:#fff;border-radius:14px 14px 2px 14px;padding:8px 14px;max-width:80%;font-size:14px;">'
        f'{_html.escape(r["question"])}</div></div><div style="display:flex;gap:8px;align-items:flex-start;">'
        f'<div style="font-size:22px;">🤖</div><div style="flex:1;">{LIB.card_html(r, ui_thumbnails.value)}</div></div>'
        for r in reversed(HISTORY))
    return (f'{_CSS}<div class="rg" style="max-height:720px;overflow-y:auto;padding:4px 6px;border:1px solid '
            f'{C["sage"]};border-radius:12px;background:{C["mist"]};">{bubbles}</div>')

ui_question = widgets.Text(placeholder="Ask a question about the documents, in any language…", layout=widgets.Layout(flex="1", width="auto"))
ui_send = widgets.Button(description="Send", button_style="success", icon="paper-plane", layout=widgets.Layout(width="120px"))
ui_status = widgets.HTML("")
ui_docs = [widgets.Checkbox(value=True, description=d.short, indent=False, layout=widgets.Layout(width="auto", margin="0 14px 0 0"))
           for d in LIB.documents]
ui_k = widgets.IntSlider(value=TOP_K, min=1, max=8, description="Passages", continuous_update=False)
ui_threshold = widgets.FloatSlider(value=float(LIB.threshold), min=0.0, max=1.0, step=0.01, description="Threshold",
                                   continuous_update=False, readout_format=".2f")
ui_guardrails = widgets.Checkbox(value=True, description="Guardrails on", indent=False)
ui_memory = widgets.Checkbox(value=True, description="Conversation memory", indent=False)
ui_thumbnails = widgets.Checkbox(value=True, description="Show pages", indent=False)
ui_clear = widgets.Button(description="New conversation", icon="refresh", layout=widgets.Layout(width="180px"))
ui_export = widgets.Button(description="Export", icon="download", layout=widgets.Layout(width="120px"))
ui_useful = widgets.Button(description="Useful", icon="thumbs-up", layout=widgets.Layout(width="100px"))
ui_review = widgets.Button(description="Needs review", icon="thumbs-down", layout=widgets.Layout(width="130px"))
ui_comment = widgets.Text(placeholder="Optional comment on the last answer…", layout=widgets.Layout(flex="1", width="auto"))
ui_thread = widgets.HTML(_thread())

def give_feedback(useful):
    if not HISTORY:
        ui_status.value = "Ask a question first."
        return
    record_feedback(HISTORY[-1], useful, ui_comment.value.strip(), "notebook")
    ui_comment.value = ""
    ui_status.value = (f'<span style="color:{C["dark_green"]};">Thank you! Feedback saved '
                       f'({len(FEEDBACK)} in total, file user_feedback.csv).</span>')

def send(_=None, question=None):
    q = (question or ui_question.value).strip()
    if not q:
        return
    docs = [d.id for d, c in zip(LIB.documents, ui_docs) if c.value]
    if not docs:
        ui_status.value = f'<span style="color:{C["brick"]};">Tick at least one document.</span>'
        return
    ui_question.value = ""
    ui_send.disabled = True
    ui_status.value = f'<span style="color:{C["teal"]};">⏳ Searching {len(docs)} document(s)…</span>'
    try:
        r = LIB.answer(q, history=HISTORY if ui_memory.value else None, docs=docs, k=ui_k.value,
                       threshold=ui_threshold.value, guardrails=ui_guardrails.value, memory=ui_memory.value)
        HISTORY.append(r)
        ui_status.value = (f'<span style="color:{C["slate"]};font-size:12px;">{len(HISTORY)} turn(s) · '
                           f'last answer in {r["duration_s"]} s' + (f' · {LLM.calls} model call(s)' if LLM.available else "") + "</span>")
    except Exception as error:
        ui_status.value = f'<span style="color:{C["brick"]};">⛔ {_html.escape(str(error)[:200])}</span>'
    finally:
        ui_send.disabled = False
        ui_thread.value = _thread()

def new_conversation(_=None):
    HISTORY.clear()
    ui_thread.value = _thread()
    ui_status.value = ""

def _serialise(r):
    return {"question": r["question"], "standalone_question": r["standalone_question"], "answer": r["answer"],
            "refused": r["refused"], "confidence": round(r["confidence"], 3), "reason": r["reason"], "mode": r["mode"],
            "guardrails": r["guardrails"], "check": r["check"], "time": r["time"],
            "sources": [{"document": p.doc, "section": p.section, "pages": p.pages, "kind": p.kind,
                         "confidence": round(s, 3), "text": p.text} for p, s in r["passages"]]}

def export(_=None):
    if not HISTORY:
        ui_status.value = "Nothing to export yet."
        return
    stamp = time.strftime("%Y%m%d_%H%M%S")
    (OUTPUT_DIR / f"conversation_{stamp}.json").write_text(
        json.dumps({"documents": [d.title for d in LIB.documents], "turns": [_serialise(r) for r in HISTORY]},
                   ensure_ascii=False, indent=2), encoding="utf-8")
    html_path = OUTPUT_DIR / f"conversation_{stamp}.html"
    html_path.write_text(f'<html><head><meta charset="utf-8"><title>Conversation</title></head>'
                         f'<body style="max-width:900px;margin:auto;">{_header()}{_thread()}</body></html>', encoding="utf-8")
    ui_status.value = f"✅ Conversation exported: <code>{html_path.name}</code> (and its JSON version)"
    if ENV["name"] == "colab":
        from google.colab import files
        files.download(str(html_path))

buttons = []
for s in SUGGESTIONS:
    b = widgets.Button(description=(s[:44] + "…") if len(s) > 45 else s, tooltip=s,
                       layout=widgets.Layout(width="auto", margin="2px 4px 2px 0"))
    b.on_click(lambda _, s=s: send(question=s))
    buttons.append(b)
ui_send.on_click(send)
ui_useful.on_click(lambda _: give_feedback(True))
ui_review.on_click(lambda _: give_feedback(False))
ui_clear.on_click(new_conversation)
ui_export.on_click(export)
ui_thumbnails.observe(lambda _: setattr(ui_thread, "value", _thread()), names="value")
try:
    import warnings
    with warnings.catch_warnings():
        warnings.simplefilter("ignore")
        ui_question.on_submit(send)
except Exception:
    pass

settings_panel = widgets.Accordion(children=[widgets.VBox([
    ui_k, ui_threshold, widgets.HBox([ui_guardrails, ui_memory, ui_thumbnails])])])
settings_panel.set_title(0, "⚙️ Settings")
settings_panel.selected_index = None

INTERFACE = widgets.VBox([
    widgets.HTML(_header()),
    widgets.HTML('<div style="font-size:12.5px;margin:10px 0 2px 0;"><b>Documents searched</b></div>'),
    widgets.HBox(ui_docs, layout=widgets.Layout(flex_flow="row wrap")),
    widgets.HTML('<div style="font-size:12.5px;margin:8px 0 2px 0;"><b>Suggestions</b></div>'),
    widgets.HBox(buttons, layout=widgets.Layout(flex_flow="row wrap")),
    widgets.HBox([ui_question, ui_send], layout=widgets.Layout(margin="8px 0 2px 0")),
    ui_status, settings_panel,
    widgets.HBox([widgets.HTML('<span style="font-size:12.5px;margin-right:6px;"><b>Last answer:</b></span>'),
                  ui_useful, ui_review, ui_comment]),
    widgets.HBox([ui_clear, ui_export]), ui_thread,
], layout=widgets.Layout(width="100%", max_width="980px"))
display(INTERFACE)
````

**Gradio** is a free, open-source Python library that turns Python functions into an interactive web page — no HTML or JavaScript required. The app brings every feature of the assistant to a web page:

- **drag and drop** new PDFs, indexed on the fly;
- **filter** the documents searched;
- **conversation with memory**, with the rewrite displayed;
- **sources** with highlighted page thumbnails and automatic checks;
- **👍 / 👎** under each answer;
- **settings**: number of passages, threshold, guardrails, memory.

On **Colab** and **Kaggle**, Gradio creates a **temporary public link** (`https://….gradio.live`), valid while the session is running and for a limited time. **Locally**, the app opens at `http://127.0.0.1:7860`. The app runs **inside your notebook session**: if the session stops, the page stops working.

> ⚠️ **Confidentiality.** Anyone with the public link can query your documents and upload files. **Set `WEB_PASSWORD`** before sharing a link, and do not use a public link for unpublished documents.

> ⚠️ **Shared app.** All users of the same link share the same library: a PDF uploaded by one becomes searchable by all. Each user keeps their own conversation.

````python
def _ensure_gradio():
    try:
        import gradio as gr
        if int(gr.__version__.split(".")[0]) >= 5:
            return gr
        print(f"Gradio {gr.__version__} is too old: upgrading…")
    except ImportError:
        print("Installing Gradio (about a minute)…")
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-U", "gradio>=5"], check=False)
    importlib.invalidate_caches()
    import gradio as gr
    if int(gr.__version__.split(".")[0]) < 5:
        raise RuntimeError("Restart the session (Runtime → Restart session), then run the notebook again.")
    return gr

def _doc_labels():
    return [f"{d.id} · {d.short}" for d in LIB.documents]

def _library_summary():
    return (f"**{len(LIB.documents)} document(s)** · {len(LIB.passages)} passages · 🌐 {', '.join(LIB.languages)} · "
            f"threshold {LIB.threshold:.2f} · "
            + (f"model **{LLM.name}**" if LLM.available else "extractive mode"))

def build_app():
    gr = _ensure_gradio()
    import inspect
    theme = gr.themes.Soft(primary_hue="emerald", secondary_hue="amber", neutral_hue="slate")
    css = ".gradio-container {max-width: 1250px !important} footer {display: none !important}"
    blocks_args = {"title": "RAG document assistant"}
    blocks_params = inspect.signature(gr.Blocks.__init__).parameters
    if "theme" in blocks_params:
        blocks_args.update(theme=theme, css=css)
    chat_params = inspect.signature(gr.Chatbot.__init__).parameters
    chat_args = {"height": 470, "label": "Conversation"}
    if "type" in chat_params:
        chat_args["type"] = "messages"
    if "feedback_options" in chat_params:
        chat_args["feedback_options"] = ["Like", "Dislike"]

    def web_answer(question, messages, memory_state, docs, k, threshold, guardrails, memory):
        messages = list(messages or [])
        if not question or not question.strip():
            return "", messages, memory_state, gr.update()
        ids = [int(x.split(" · ")[0]) for x in (docs or [])]
        if not ids:
            messages.append({"role": "assistant", "content": "Tick at least one document."})
            return "", messages, memory_state, gr.update()
        hist = [{"standalone_question": h["standalone_question"], "answer": h["answer"],
                 "language": h.get("language")} for h in (memory_state or [])]
        r = LIB.answer(question.strip(), history=hist if memory else None, docs=ids, k=int(k),
                       threshold=float(threshold), guardrails=bool(guardrails), memory=bool(memory))
        text = r["answer"]
        if r.get("queries"):
            text = "_🌐 Also searched as: " + " · ".join(f"[{l}] {q}" for l, q in r["queries"].items()) + "_\n\n" + text
        if r["rewritten"]:
            text = f"_🧠 Understood as: {r['standalone_question']}_\n\n" + text
        if r["reason"]:
            text += f"\n\n_Reason: {r['reason']}_"
        if not r["refused"]:
            text += "\n\n" + "\n".join(f"**[{i}]** {p.citation}" for i, (p, _) in enumerate(r["passages"], 1))
        messages += [{"role": "user", "content": question.strip()}, {"role": "assistant", "content": text}]
        memory_state = (memory_state or []) + [{"question": r["question"], "standalone_question": r["standalone_question"],
                                                "answer": r["answer"], "citations": r["citations"],
                                                "language": r["language"]}]
        return "", messages, memory_state, f"{_CSS}{LIB.card_html(r)}"

    def rate(data: gr.LikeData, memory_state):
        index = data.index[0] if isinstance(data.index, (list, tuple)) else data.index
        turns = memory_state or []
        position = int(index) // 2
        if not turns or position >= len(turns):
            return "Feedback not saved: turn not found."
        useful = data.liked if isinstance(data.liked, bool) else str(data.liked).lower() in ("true", "like")
        record_feedback(turns[position], useful, channel="web")
        return f"Thank you! Feedback saved ({len(FEEDBACK)} in total)."

    def add_pdfs(files, docs):
        if not files:
            return gr.update(), _library_summary(), ""
        paths = [f if isinstance(f, str) else getattr(f, "name", str(f)) for f in files]
        added, errors = LIB.load(paths)
        LIB.index()
        choices = _doc_labels()
        ticked = list(docs or []) + [f"{d.id} · {d.short}" for d in added]
        message = f"✅ {len(added)} document(s) added." + "".join(f"\n⚠️ {n}: {e}" for n, e in errors)
        return gr.update(choices=choices, value=[c for c in choices if c in ticked]), _library_summary(), message

    with gr.Blocks(**blocks_args) as app:
        gr.HTML('<div style="background:linear-gradient(120deg,#00553A 0%,#00A86A 100%);border-radius:14px;'
                'padding:16px 22px;color:#fff;font-family:\'Segoe UI\',Calibri,sans-serif;">'
                '<div style="color:#F5C242;font-size:11px;font-weight:700;letter-spacing:2.5px;">DOCUMENT ASSISTANT · RAG</div>'
                '<div style="font-size:24px;font-weight:800;">💬 Ask questions about your PDF reports</div>'
                '<div style="font-size:13px;color:#E6F6EE;">Sourced answers, highlighted pages, and a refusal when '
                'the information is not in the documents.</div></div>')
        memory_state = gr.State([])
        with gr.Row():
            with gr.Column(scale=1, min_width=300):
                summary = gr.Markdown(_library_summary())
                docs = gr.CheckboxGroup(choices=_doc_labels(), value=_doc_labels(), label="Documents searched")
                upload = gr.File(label="Add PDFs", file_count="multiple", file_types=[".pdf"])
                add_button = gr.Button("📥 Add and index")
                add_message = gr.Markdown("")
                with gr.Accordion("⚙️ Settings", open=False):
                    k = gr.Slider(1, 8, value=TOP_K, step=1, label="Passages provided")
                    threshold = gr.Slider(0.0, 1.0, value=float(LIB.threshold), step=0.01, label="Confidence threshold")
                    guardrails = gr.Checkbox(value=True, label="Guardrails on")
                    memory = gr.Checkbox(value=True, label="Conversation memory")
            with gr.Column(scale=2):
                chat = gr.Chatbot(**chat_args)
                with gr.Row():
                    question = gr.Textbox(placeholder="Ask a question about the documents, in any language…", show_label=False,
                                          scale=5, autofocus=True)
                    send_button = gr.Button("Send", variant="primary", scale=1)
                gr.Examples(examples=[[s] for s in SUGGESTIONS], inputs=[question], label="Suggestions")
                with gr.Row():
                    new_chat = gr.Button("🔄 New conversation", size="sm")
                    feedback_message = gr.Markdown("_Use 👍 / 👎 under each answer to give your feedback._")
                with gr.Accordion("📄 Sources of the last answer", open=True):
                    sources = gr.HTML("<i>Sources will appear here.</i>")
        inputs = [question, chat, memory_state, docs, k, threshold, guardrails, memory]
        outputs = [question, chat, memory_state, sources]
        send_button.click(web_answer, inputs, outputs)
        question.submit(web_answer, inputs, outputs)
        new_chat.click(lambda: ([], [], "<i>Sources will appear here.</i>"), None, [chat, memory_state, sources])
        add_button.click(add_pdfs, [upload, docs], [docs, summary, add_message])
        if hasattr(chat, "like"):
            chat.like(rate, [memory_state], [feedback_message])
    app._launch_args = {} if "theme" in blocks_params else {"theme": theme, "css": css}
    return app

def launch_app(app):
    import inspect
    auth = (WEB_USERNAME, WEB_PASSWORD) if WEB_PASSWORD else None
    options = dict(share=SHARE_LINK, auth=auth, prevent_thread_lock=True, quiet=True, show_error=True,
                   inbrowser=(ENV["name"] == "local"), inline=(ENV["name"] == "colab"), **app._launch_args)
    params = inspect.signature(app.launch).parameters
    options = {k: v for k, v in options.items() if k in params}
    try:
        app.launch(**options)
    except Exception as error:
        if options.get("share"):
            print(f"⚠️ Public link unavailable ({type(error).__name__}): launching locally.")
            options["share"] = False
            app.launch(**options)
        else:
            raise
    public_url = getattr(app, "share_url", None)
    local_url = getattr(app, "local_url", None) or "http://127.0.0.1:7860"
    lines = [f'<div style="font-size:14px;margin:4px 0;">🔗 <b>Public link:</b> <a href="{public_url}" target="_blank">{public_url}</a></div>'
             if public_url else "",
             f'<div style="font-size:14px;margin:4px 0;">💻 <b>Local address:</b> <code>{local_url}</code></div>',
             f'<div style="font-size:13px;margin:4px 0;">🔒 Access: ' + (f'username <code>{WEB_USERNAME}</code> and password set'
             if auth else f'<span style="color:{C["brick"]};"><b>no password</b></span>') + '</div>']
    display(HTML(f'{_CSS}<div class="rg rg-card" style="border-left:6px solid {C["green"]};">'
                 f'<div class="rg-title">🌐 Web app running</div>{"".join(lines)}'
                 f'<div class="rg-note">To stop it: run <code>APP.close()</code>. It also stops when the session ends.</div></div>'))
    if public_url and not auth:
        callout("The link is public and has <b>no password</b>: set <code>WEB_PASSWORD</code> before sharing it.", "warning")

if LAUNCH_WEB_APP:
    APP = build_app()
    launch_app(APP)
else:
    callout("Web app not launched (<code>LAUNCH_WEB_APP = False</code>).", "info")
````

### Limits of this assistant

- **Complex layouts** (multiple columns, borderless tables, footnotes) are imperfectly extracted. Check step B2 on your documents.
- **Charts are not read**, only their titles and captions.
- **Extractive mode** quotes sentences or table rows: it can neither **calculate** nor **rank** ("which region has the highest unemployment?"). Such questions need a language model.
- **Rule-based memory** recognises changes of region, sex, quarter, year, area, age and indicator. Beyond that, a model is needed.
- **Figure checking is literal**: a figure calculated by the model (a difference, an average) is flagged as missing, even if it is correct.
- **Language resources.** Stop words and stems exist for English, French, Spanish, Portuguese, German, Italian, Arabic and (stop words only) Swahili; the glossary and the cross-lingual lexicon cover English, French, Spanish and Portuguese. Another language (Kinyarwanda, Wolof, Amharic…) is still read, but without stems or stop words: rely on a multilingual embedding model (`multilingual-e5`, `bge-m3`) and on a language model for translation.
- **Language detection** is based on frequent function words. It is reliable on documents and on ordinary questions, less so on a two-word question ("Coast region?"): such a question inherits the language of the previous turn, and `ask("…", lang="fr")` forces it.
- **Extractive mode cannot translate**: a French passage is quoted in French, labelled *(French source)*. Answers in the language of the question need a model.
- **Without `sentence-transformers`** the semantic search is an offline approximation that matches spellings, not meaning. It cannot bridge a question written in your users' words and a report written in the institution's. This is the single biggest limitation of the default setup: `INSTALL_EMBEDDINGS = True` removes it.
- **Spelling tolerance has a deliberate blind spot**: a word the documents never use is corrected only if a very close one exists. "maize" stays unknown — which is exactly what lets the assistant refuse an out-of-scope question.
- **On very long reports**, most passages contain every word of a question, so the confidence score saturates near 1.00: it then measures coverage, not precision. Ranking relies on term density and on the fused search, and the threshold is capped at 0.70.
- **Final verification remains human**: the assistant gives the page, a statistician checks before any release.

### Suggested improvements

| # | Improvement | Principle | Effort | Gain |
|:-:|---|---|:-:|:-:|
| 1 | **Neural embeddings** | `INSTALL_EMBEDDINGS = True`: the semantic search stops matching spellings and starts matching meaning | Low | ⭐⭐⭐ |
| 2 | **Advanced extraction** | Specialised tools (Docling, Marker, PyMuPDF4LLM) for columns and complex tables | Medium | ⭐⭐⭐ |
| 3 | **Cross-encoder reranking** | `RERANKING = "cross-encoder"`: a model reads question and passage together (B4) | Low | ⭐⭐⭐ |
| 3b | **Faithfulness judge** | A second model call checks that every sentence is supported by its source (beyond figures) | Medium | ⭐⭐⭐ |
| 4 | **Metadata** | Filter by year, topic or publication type, not only by file | Medium | ⭐⭐ |
| 5 | **Reading charts** | Describe figures with a multimodal model | Medium | ⭐⭐ |
| 6 | **Persistent library** | Extraction and embeddings are already cached; going further means storing the index in a vector database | Medium | ⭐⭐ |
| 7 | **Feedback dashboard** | Track the share of useful answers and declined questions, week after week | Low | ⭐⭐ |
| 8 | **Durable deployment** | Host the app (internal server, Hugging Face Spaces) instead of a temporary link | High | ⭐⭐⭐ |
| 9 | **Access rights** | Restrict some documents to some users | High | Essential in production |

### ✅ Checkpoint

<details><summary><b>1. Why display “Understood as…”?</b></summary>
Because memory can be wrong. Users see the question actually searched and can rephrase if it does not match their intent.
</details>

<details><summary><b>2. Why limit the number of passages from the same document?</b></summary>
So that a comparison question receives excerpts from several documents, rather than three passages from the same report.
</details>

<details><summary><b>3. Why a password on the web app?</b></summary>
A public link is accessible to anyone who has it: without a password, anyone can query your documents or upload new ones.
</details>

<details><summary><b>4. A wrong answer with the right passage: what should you fix?</b></summary>
The <b>generation</b>: the prompt or the model. With an off-topic passage: the <b>retrieval</b> (extraction, glossary, threshold).
</details>

<details><summary><b>5. What does “figure not in sources” mean?</b></summary>
A number in the answer appears in none of the passages provided: either the model invented or rounded it, or it calculated it. In both cases, check it against the cited sources.
</details>

### 🛠️ Troubleshooting

| Symptom | Likely cause | Fix |
|---|---|---|
| Interface C1 does not show | Widgets not supported | Use `ask("…")` or the web app |
| "Public link unavailable" | Internet off or filtered network | Kaggle: *Settings → Internet: On*; otherwise use the local address |
| "Restart the session" after installing Gradio | An older version was already loaded | *Runtime → Restart session*, then *Run all* |
| An uploaded PDF does not appear | Protected or scanned file | See the message below the button; turn on OCR if needed |
| `HTTP Error 403: Forbidden` on a web address | The site refuses automated downloads (common on institutional portals, including afdb.org from Colab) | The other documents still load. Download the PDF in your browser, then `UPLOAD_PDF = True` on Colab, or give the file path in `DOCUMENTS` |
| "Understood as" is wrong | Rule-based memory is limited | Ask a complete question, or enable a model (Groq, Ollama) |
| "Figure not in sources" on a correct answer | Figure calculated or reformatted by the model | Check the calculation against the cited sources |
| Too many refusals | Threshold too high | Lower it in the interface settings; extend the glossary |
| A question about your own field returns something adjacent | Your users' words are not the report's words | Add the pair to `GLOSSARY`, or turn on neural embeddings. `explain("your question")` shows exactly where the chain went astray |
| A typo silently breaks a question | Rare: the word is corrected and shown on the card ("spelling: coutries → countries") | If the correction is wrong, the word is probably legitimate but absent from the documents |
| Which chunking should I use? | — | Run `benchmark()` (B3) on your own documents instead of guessing |
| Indexing an 80-page report is slow the first time | Table detection runs page by page | Normal: 5 to 10 seconds per report. Later runs reuse `rag_cache` |
| Extraction settings changed but nothing moved | A cached extraction was reused | The cache key includes the settings; if in doubt, delete the `rag_cache` folder |
| A question about a section returns the table of contents | `DROP_TABLE_OF_CONTENTS = False` | Set it back to `True` and re-index |
| A document is given the wrong language | Very short document, or mostly tables | Check the *Language* column in B10; force it with `DOCUMENT_LANGUAGE = "fr"` if the whole collection is in one language, or set `LIB.documents[i].lang` then `LIB.index()` |
| A question in another language than the documents is declined | No model and no multilingual embeddings: only the lexicon links the languages | Connect a model (Groq, Ollama) so the question is translated, or add the missing words to `CROSS_LINGUAL_LEXICON` and run `build_lexicon_index()` |
| The same report in two languages returns both versions | Normal: each version is a document | Untick one in the interfaces' document filter, or keep `PREFER_QUESTION_LANGUAGE = True` (default) |
| Scanned pages in Arabic, Swahili… come out empty | OCR language missing | Add its Tesseract code to `OCR_LANGUAGE` (`"ara"`, `"swa"`…) |
| "Rate limit reached" | Groq per-minute limit | The notebook waits and retries; for heavy use, switch to Ollama |
| Ollama very slow | No GPU | Turn on a T4 GPU *before* running, or use `qwen2.5:1.5b` |
