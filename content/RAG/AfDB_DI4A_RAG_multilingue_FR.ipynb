<div style="color:#E6F6EE;font-size:15px;font-style:italic;">Séance pratique — interroger les documents de votre institution avec un seul notebook.</div></div>

L'apprentissage automatique permet de **classer** du texte (par exemple des transactions dans les catégories de la balance des paiements). Ce notebook montre comment l'**interroger** : un assistant qui lit les publications PDF de votre institution, répond avec le **document, la section et la page**, et **refuse** de répondre lorsque l'information n'y figure pas. La partie F montre comment votre équipe peut en faire un prototype.

### Quatre objectifs, rattachés à ce notebook

| # | Objectif | Où dans le notebook | Durée |
|:-:|---|---|:-:|
| 1 | **Créer un assistant pour les documents statistiques** | 🧪 Exercice 1 → A1 · A2 | ≈ 7 min |
| 2 | **S'exercer à la recherche documentaire et aux réponses fondées sur les sources** | 🧪 Exercice 2 → A3 · C1 | ≈ 6 min |
| 3 | **Comparer les méthodes de recherche** | 🧪 Exercice 3 → B3 · B4 | ≈ 5 min |
| 4 | **Diagnostiquer les erreurs et gérer les questions hors champ** | 🧪 Exercice 4 → B5 · B8 · B9 | ≈ 5 min |
| — | Bilan et suite pour votre propre projet | Partie F | ≈ 2 min |
| ➕ | **Collections bilingues** (français · anglais · portugais) | B10 | parcours approfondi |

Les quatre 🧪 **exercices** sont les encadrés dorés répartis dans le notebook. Une séance de 30 minutes suit le **parcours express** ; le **parcours approfondi** (90 min) sert à approfondir et à préparer votre propre projet.

### ✅ Avant de commencer — trois minutes qui font gagner du temps

1. **Ouvrez le notebook dans Google Colab** (connecté à un compte Google) et lancez **Exécution → Tout exécuter** tout de suite : l'installation prend quelques minutes.
2. **Choisissez le modèle.** Sur un créneau de 30 minutes, une clé **Groq** gratuite est la voie la plus rapide : `PROVIDER = "groq"`, puis collez la clé à l'étape 0b (2 minutes, sans carte bancaire). **Ollama** (le réglage par défaut) exécute le modèle dans la session — illimité et confidentiel, mais il faut un GPU T4 et un téléchargement d'environ 2 Go : idéal pour les séances plus longues.
3. **Ayez sous la main un PDF de votre institution** — note méthodologique de l'IPC, publication des comptes nationaux trimestriels, bulletin de la balance des paiements, déclaration de politique monétaire, rapport annuel. Téléchargez-le d'abord dans votre navigateur : `UPLOAD_PDF = True` permet ensuite de l'importer.

> ⚠️ **Confidentialité** — avec un modèle en ligne (Groq, Anthropic, OpenAI), n'utilisez que des documents **publiés**. Les documents non publiés ou sous embargo restent avec **Ollama**, où aucun texte ne quitte la session — et jamais sur un lien Gradio public sans `WEB_PASSWORD`.

Deux réglages déterminent la qualité de tout le reste, et la partie B les mesure tous les deux sur vos propres documents : **`CHUNK_STRATEGY`** (cinq niveaux de découpage, des blocs fixes jusqu'à deux granularités avec tableaux et contexte) et **`RERANKING`** (quatre niveaux de reclassement, d'aucun jusqu'à un cross-encoder neuronal). La chaîne complète est dessinée après l'indexation, et `explain("votre question")` suit n'importe quelle question à travers ses treize étapes.

Chaque réponse cite **le document, la section et la page**, affiche une **vignette de la page avec le passage surligné**, effectue des **vérifications automatiques** (chiffres, termes manquants), ou **refuse de répondre** lorsque l'information ne figure pas dans les documents.

````python
# @title ⚙️ Installer les bibliothèques { display-mode: "form" }
import importlib, subprocess, sys

def ensure_package(module, package):
    try:
        importlib.import_module(module)
    except ImportError:
        print(f"Installation de {package}…")
        subprocess.run([sys.executable, "-m", "pip", "install", "-q", package], check=False)

for module, package in [("numpy", "numpy"), ("pandas", "pandas"), ("sklearn", "scikit-learn"),
                        ("matplotlib", "matplotlib"), ("pymupdf", "pymupdf"), ("ipywidgets", "ipywidgets"),
                        ("snowballstemmer", "snowballstemmer")]:
    ensure_package(module, package)
print("✅ Bibliothèques prêtes.")
````

````python
# @title ⚙️ Fonctions d'affichage { display-mode: "form" }
import html as _html
import re, time, json, math, unicodedata
from collections import Counter
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import HTML, display

C = dict(green="#00A86A", dark_green="#00704A", forest="#00553A", gold="#F5C242", ochre="#D49A00",
         teal="#0E7C86", terra="#C4621D", brick="#B83B2E", ink="#231F20", slate="#5E6964",
         mist="#F4F7F5", mint="#E8F5EF", sage="#D5DED9", grey="#A9B5B0")

_CSS = f"""<style>
.rg {{font-family:'Segoe UI',Calibri,system-ui,sans-serif;color:{C['ink']};}}
.rg-title {{font-weight:700;font-size:15px;margin:6px 0 6px 0;color:{C['ink']};}}
.rg-note {{font-size:12.5px;color:{C['slate']};font-style:italic;margin:6px 0 2px 0;}}
table.rg-t {{border-collapse:collapse;font-size:13px;width:100%;background:#fff;}}
table.rg-t th {{background:{C['forest']};color:#fff;padding:7px 10px;text-align:left;font-weight:600;}}
table.rg-t td {{padding:6px 10px;border-bottom:1px solid {C['sage']};color:{C['ink']};background:#fff;vertical-align:top;}}
table.rg-t tr:nth-child(even) td {{background:{C['mist']};}}
.rg-card {{background:#fff;border:1px solid {C['sage']};border-radius:12px;padding:14px 18px;margin:10px 0;color:{C['ink']};}}
.rg-pill {{display:inline-block;border-radius:12px;padding:2px 10px;font-size:11.5px;font-weight:700;margin:0 6px 3px 0;}}
.rg-src {{font-size:12.5px;color:{C['slate']};border-top:1px dashed {C['sage']};padding-top:6px;margin-top:6px;}}
</style>"""

def pill(text, color):
    return (f'<span class="rg-pill" style="background:{color}22;color:{color};'
            f'border:1px solid {color};">{_html.escape(str(text))}</span>')

def show_table(df, title=None, note=None):
    body = df.to_html(index=False, escape=False, classes="rg-t", border=0)
    t = f'<div class="rg-title">{title}</div>' if title else ""
    n = f'<div class="rg-note">{note}</div>' if note else ""
    display(HTML(f'{_CSS}<div class="rg">{t}<div style="overflow-x:auto">{body}</div>{n}</div>'))

def callout(text, kind="info"):
    styles = {"info": (C["teal"], "💡"), "success": (C["green"], "✅"),
              "warning": (C["ochre"], "⚠️"), "error": (C["brick"], "⛔")}
    color, icon = styles[kind]
    display(HTML(f'{_CSS}<div class="rg" style="background:{color}14;border:1px solid {color};'
                 f'border-radius:10px;padding:10px 16px;margin:8px 0;">{icon}&nbsp; {text}</div>'))

plt.rcParams.update({
    "figure.dpi": 110, "figure.facecolor": "white", "axes.facecolor": "white",
    "axes.spines.top": False, "axes.spines.right": False, "axes.edgecolor": C["sage"],
    "axes.labelcolor": C["slate"], "xtick.color": C["slate"], "ytick.color": C["slate"],
    "axes.titleweight": "bold", "axes.titlesize": 12.5, "axes.titlecolor": C["ink"],
    "font.size": 10, "legend.frameon": False,
})
callout("Fonctions d'affichage prêtes.", "success")
````

### 🎛️ Réglages

**Parcours express :** seule la première section compte. Gardez `DOCUMENTS = ["exemple"]` pour un premier essai, puis remplacez-le par vos propres fichiers.

`DOCUMENTS` accepte une **liste** qui mélange :
- `"exemple"` — trois rapports fictifs en français fournis avec le notebook ;
- des **fichiers** — `"rapports/emploi_2025.pdf"` ;
- des **dossiers** — `"rapports/"` (tous les PDF qu'il contient) ;
- des **adresses web** — `"https://…/rapport.pdf"`.

`DOCUMENT_LANGUAGE = "auto"` détecte la langue de **chaque paragraphe, tableau et titre**, puis de chaque document dans son ensemble : une collection peut mélanger le français, l'anglais, l'espagnol, le portugais et d'autres langues, **un même rapport aussi** (un bulletin bilingue, un rapport en français avec une annexe en anglais, un tableau aux libellés dans deux langues). Un passage ne mélange jamais deux langues, et un passage et sa traduction ne comptent qu'une fois dans une réponse. N'imposez un code (`"fr"`, `"en"`…) que si la détection se trompe. `DOCUMENTS = ["exemple-multilingue"]` ajoute aux trois exemples en français une note en anglais, une enquête en espagnol et une **enquête bilingue français/anglais**.

**Les questions peuvent être posées dans n'importe quelle langue.** La section **1b** des réglages décide comment : `QUESTION_LANGUAGE` (détectée par défaut), `ANSWER_LANGUAGE` (par défaut, la langue de la question) et `TRANSLATE_QUERIES` (avec un modèle, la question est aussi recherchée dans chaque langue de la collection). La partie **B10** explique le mécanisme et le mesure.

**Les longs rapports institutionnels** (50 à 200 pages) sont le cas normal pour un institut de statistique. Trois réglages comptent pour eux :

| Réglage | Effet |
|---|---|
| `USE_CACHE = True` | L'extraction et les embeddings sont enregistrés dans `rag_cache`. La première exécution sur un rapport de 80 pages coûte quelques secondes de détection des tableaux ; toutes les exécutions suivantes de la même session — ou après un redémarrage — réutilisent le cache. Supprimez le dossier pour forcer une nouvelle extraction. |
| `DROP_TABLE_OF_CONTENTS = True` | Supprime les lignes « Chapitre 3 …… 42 », les listes des figures et les numéros de page isolés, qui deviendraient sinon des passages ne répondant à rien. |
| `MAX_PAGES_PER_DOCUMENT` | `0` lit tout. Réglez-le sur 30 ou 40 pour essayer rapidement un très long rapport avant de l'indexer en entier. |

> ⚠️ **Les adresses web ne fonctionnent pas toujours.** De nombreux portails institutionnels — dont `afdb.org` — refusent les téléchargements qui ne proviennent pas d'un navigateur et répondent **HTTP 403**. Le notebook essaie des en-têtes de navigateur, puis `curl`, et si les deux échouent, il **ignore ce document et continue avec les autres**, en vous indiquant lequel a échoué.
>
> **Les liens de partage fonctionnent aussi.** Un lien Google Drive, Dropbox ou OneDrive est converti automatiquement en téléchargement direct, y compris l'avertissement antivirus de Drive pour les gros fichiers. Le fichier doit être partagé avec **« tous les utilisateurs disposant du lien »** ; un lien privé renvoie une page web au lieu d'un PDF, et le notebook vous le signale.

Un modèle n'est **pas obligatoire** : sans modèle, le notebook fonctionne en **mode extractif** (il cite la phrase la plus pertinente des documents). Mais pour voir comment un RAG se comporte vraiment — et surtout comment il **refuse de répondre** — il vous faut un modèle. Voici les options, de la mieux adaptée à un usage intensif à la moins adaptée :

| Option | Coût | Volume | Installation | Pour qui ? |
|---|---|---|---|---|
| 🥇 **Ollama** (modèle ouvert exécuté sur la machine) | Gratuit | **Illimité** | 3 à 5 min (installation + ~2 Go à télécharger) | Formations, pratique intensive, documents sensibles |
| 🥈 **Groq** (API) | Gratuit, sans carte bancaire | **1 000 requêtes/jour par modèle**, 30/min, 8 000 tokens/min | 1 min (clé sur `console.groq.com/keys`) | Démarrage rapide, réponses très rapides |
| Anthropic, OpenAI (API) | Payant | Selon le budget | 1 min | Comparaison avec des modèles commerciaux |

**Ordre de grandeur.** Une exécution complète de ce notebook demande environ **30 appels au modèle** : avec Groq, une trentaine d'exécutions par jour et par modèle ; avec Ollama, aucune limite.

> ⚠️ **Quotas.** Les limites de Groq ci-dessus sont celles de sa documentation officielle en septembre 2026. Elles évoluent : consultez `console.groq.com/settings/limits` pour connaître les valeurs de votre compte. Lorsqu'une limite est atteinte, le notebook **attend automatiquement** le délai demandé par le service, puis réessaie.

> 💡 **Ollama et confidentialité.** Avec Ollama, aucun texte ne quitte jamais la machine. C'est l'option à privilégier dès que la collection comprend des documents non publiés.

> 💡 **Ollama sur Colab ou Kaggle.** Activez un **GPU T4 avant d'exécuter le notebook**. Ollama détecte le GPU et y charge le modèle automatiquement ; `qwen2.5:3b` utilise environ 3 Go des 16 Go du T4.

⚠️ **Ne collez jamais une clé API directement dans une cellule.** Utilisez les secrets de Colab ou de Kaggle, ou une variable d'environnement en local.

````python
# @title 🦙 Installer Ollama (si demandé) { display-mode: "form" }
def _ollama_running():
    try:
        urllib.request.urlopen("http://localhost:11434/api/tags", timeout=2)
        return True
    except Exception:
        return False

def install_ollama(model):
    if _ollama_running():
        print("Ollama est déjà en cours d'exécution.")
    elif platform.system() != "Linux":
        callout("L'installation automatique n'est disponible que sous Linux (Colab, Kaggle). Sous macOS ou Windows, "
                "installez Ollama depuis <code>ollama.com/download</code>, puis exécutez de nouveau cette cellule.", "warning")
        return False
    else:
        if not shutil.which("ollama"):
            print("1/3 · Installation d'Ollama (1 à 2 min)…")
            if shutil.which("apt-get"):
                subprocess.run("apt-get install -y -qq zstd pciutils > /dev/null 2>&1 || "
                               "(apt-get update -qq > /dev/null 2>&1 && apt-get install -y -qq zstd pciutils > /dev/null 2>&1)",
                               shell=True)
            subprocess.run("curl -fsSL https://ollama.com/install.sh | sh > ollama_install.log 2>&1", shell=True)
        if not shutil.which("ollama"):
            callout("L'installation a échoué : consultez le fichier <code>ollama_install.log</code>.", "error")
            return False
        print("2/3 · Démarrage du service…")
        subprocess.Popen(["ollama", "serve"], stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL,
                         start_new_session=True)
        for _ in range(30):
            if _ollama_running():
                break
            time.sleep(1)
    print(f"3/3 · Téléchargement du modèle {model} (quelques minutes la première fois)…")
    result = subprocess.run(["ollama", "pull", model], capture_output=True, text=True)
    if result.returncode != 0:
        callout(f"Impossible de télécharger le modèle : <code>{_html.escape(result.stderr[-300:])}</code>", "error")
        return False
    callout(f"Ollama est prêt avec le modèle <b>{model}</b> · exécutions illimitées."
            + ("" if ENV["gpu"] else " <br>⚠️ Sans GPU, chaque réponse peut prendre de 10 à 60 secondes."), "success")
    return True

if INSTALL_OLLAMA:
    install_ollama(OLLAMA_MODEL)
else:
    print("Installation d'Ollama non demandée (INSTALL_OLLAMA = False).")
````

````python
# @title 🤖 Se connecter au modèle { display-mode: "form" }
import urllib.error

def _secret(name):
    value = os.environ.get(name)
    if value:
        return value
    if ENV["name"] == "colab":
        try:
            from google.colab import userdata
            return userdata.get(name)
        except Exception:
            return None
    if ENV["name"] == "kaggle":
        try:
            from kaggle_secrets import UserSecretsClient
            return UserSecretsClient().get_secret(name)
        except Exception:
            return None
    return None

def _post(url, payload, headers, timeout=120):
    request = urllib.request.Request(url, data=json.dumps(payload).encode("utf-8"),
                                     headers={"Content-Type": "application/json", **headers})
    with urllib.request.urlopen(request, timeout=timeout) as response:
        return json.loads(response.read().decode("utf-8"))

class LLMClient:
    """Une seule interface pour Ollama, Groq, Anthropic ou OpenAI.
    Température 0, respect des quotas (pause entre les appels, nouvel essai après HTTP 429)."""

    MIN_PAUSE = {"groq": 2.2, "anthropic": 0.5, "openai": 0.5, "ollama": 0.0}

    def __init__(self, choice="auto"):
        self.found = {"ollama": _ollama_running(),
                      "groq": bool(_secret("GROQ_API_KEY")),
                      "anthropic": bool(_secret("ANTHROPIC_API_KEY")),
                      "openai": bool(_secret("OPENAI_API_KEY"))}
        order = ["ollama", "groq", "anthropic", "openai"] if choice == "auto" else [choice]
        self.name = next((p for p in order if self.found.get(p)), "none")
        self.model = {"ollama": OLLAMA_MODEL, "groq": GROQ_MODEL, "anthropic": ANTHROPIC_MODEL,
                      "openai": OPENAI_MODEL}.get(self.name, "—")
        self.calls, self.waited, self._last = 0, 0.0, 0.0

    @property
    def available(self):
        return self.name != "none"

    def _call(self, system, message, max_tokens):
        if self.name == "ollama":
            d = _post("http://localhost:11434/api/chat",
                      {"model": self.model, "stream": False, "options": {"temperature": 0},
                       "messages": [{"role": "system", "content": system},
                                    {"role": "user", "content": message}]}, {}, timeout=600)
            return d["message"]["content"]
        if self.name == "groq":
            payload = {"model": self.model, "temperature": 0, "max_completion_tokens": max_tokens + 600,
                       "messages": [{"role": "system", "content": system},
                                    {"role": "user", "content": message}]}
            if "gpt-oss" in self.model:
                payload["reasoning_effort"] = "low"          # raisonnement court : moins de tokens consommés
            d = _post("https://api.groq.com/openai/v1/chat/completions", payload,
                      {"Authorization": f"Bearer {_secret('GROQ_API_KEY')}"})
            return d["choices"][0]["message"].get("content") or ""
        if self.name == "anthropic":
            d = _post("https://api.anthropic.com/v1/messages",
                      {"model": self.model, "max_tokens": max_tokens, "temperature": 0,
                       "system": system, "messages": [{"role": "user", "content": message}]},
                      {"x-api-key": _secret("ANTHROPIC_API_KEY"), "anthropic-version": "2023-06-01"})
            return "".join(b.get("text", "") for b in d["content"])
        if self.name == "openai":
            d = _post("https://api.openai.com/v1/chat/completions",
                      {"model": self.model, "temperature": 0, "max_completion_tokens": max_tokens,
                       "messages": [{"role": "system", "content": system},
                                    {"role": "user", "content": message}]},
                      {"Authorization": f"Bearer {_secret('OPENAI_API_KEY')}"})
            return d["choices"][0]["message"]["content"]
        raise RuntimeError("Aucun fournisseur de modèle disponible.")

    def generate(self, system, message, max_tokens=400, attempts=5):
        for attempt in range(attempts):
            wait = self.MIN_PAUSE.get(self.name, 0) - (time.time() - self._last)
            if wait > 0:
                time.sleep(wait)
            try:
                self._last = time.time()
                text = self._call(system, message, max_tokens)
                self.calls += 1
                return re.sub(r"<think>.*?</think>", "", text, flags=re.S).strip()
            except urllib.error.HTTPError as error:
                if error.code == 429 and attempt < attempts - 1:
                    delay = min(float(error.headers.get("retry-after") or 2 ** (attempt + 2)), 90)
                    print(f"⏳ Limite de débit atteinte ({self.name}) : nouvel essai dans {delay:.0f} s…")
                    self.waited += delay
                    time.sleep(delay)
                    continue
                detail = error.read().decode("utf-8", "ignore")[:200]
                raise RuntimeError(f"HTTP {error.code}: {detail}") from error

# ── D'où vient chaque clé, et à quoi elle ressemble ───────────────────────────────────────
API_KEYS = {"groq":      ("GROQ_API_KEY",      "gsk_",    "console.groq.com/keys (gratuit, sans carte bancaire)"),
            "anthropic": ("ANTHROPIC_API_KEY", "sk-ant-", "console.anthropic.com"),
            "openai":    ("OPENAI_API_KEY",    "sk-",     "platform.openai.com/api-keys")}

def ask_for_api_key(provider):
    """Demande la clé de ce fournisseur, masquée pendant la saisie. Entrée seule pour passer.
    La clé reste en mémoire pour cette session : elle n'est jamais écrite dans le notebook."""
    variable, prefix, where = API_KEYS[provider]
    import getpass
    try:
        key = getpass.getpass(f"🔑 Clé {provider.capitalize()} ({prefix}…) — à obtenir sur {where} — "
                              f"appuyez sur Entrée seule pour passer : ").strip()
    except Exception:
        return None                                  # exécution sans interface (nbconvert, papermill) : on passe
    if not key:
        callout("Étape passée : l'assistant <b>citera</b> des phrases au lieu de rédiger des réponses. "
                "Vous pourrez coller une clé plus tard dans la cellule suivante.", "info")
        return None
    if not key.startswith(prefix):
        callout(f"Une clé {provider.capitalize()} commence par <code>{prefix}</code> — rien n'a été enregistré.", "error")
        return None
    os.environ[variable] = key
    return LLMClient(provider)

LLM = LLMClient("none" if PROVIDER == "none" else PROVIDER)

show_table(pd.DataFrame([
    {"Fournisseur": "Ollama (local)", "Coût": "gratuit · illimité", "Détecté": "✅" if LLM.found["ollama"] else "—",
     "Comment l'activer": "INSTALL_OLLAMA = True, ou Ollama installé en local"},
    {"Fournisseur": "Groq", "Coût": "gratuit · 1 000 req./jour", "Détecté": "✅" if LLM.found["groq"] else "—",
     "Comment l'activer": "secret ou variable GROQ_API_KEY"},
    {"Fournisseur": "Anthropic", "Coût": "payant", "Détecté": "✅" if LLM.found["anthropic"] else "—",
     "Comment l'activer": "secret ou variable ANTHROPIC_API_KEY"},
    {"Fournisseur": "OpenAI", "Coût": "payant", "Détecté": "✅" if LLM.found["openai"] else "—",
     "Comment l'activer": "secret ou variable OPENAI_API_KEY"},
]), title="Fournisseurs de modèles détectés")

# ── Fournisseur demandé mais aucune clé en vue : on la demande maintenant, masquée ─────────
if not LLM.available and PROVIDER != "none" and ASK_FOR_KEY:
    wanted = PROVIDER if PROVIDER in API_KEYS else "groq"
    callout(f"<b>{wanted.capitalize()}</b> a été demandé, mais aucune clé n'a été trouvée dans les secrets ni dans l'environnement. "
            f"Collez-la ci-dessous — la saisie est masquée, et la clé n'est jamais écrite dans le notebook.", "warning")
    LLM = ask_for_api_key(wanted) or LLM

if LLM.available:
    try:
        start = time.time()
        test = LLM.generate("Tu réponds par un seul mot.", "Réponds uniquement : OK", max_tokens=20)
        callout(f"Fournisseur retenu : <b>{LLM.name}</b> · modèle <code>{LLM.model}</code> · "
                f"test de connexion réussi en {time.time() - start:.1f} s (réponse : « {_html.escape(test[:30])} »).", "success")
    except Exception as error:
        callout(f"Le fournisseur <b>{LLM.name}</b> a été détecté, mais le test a échoué : "
                f"<code>{_html.escape(str(error)[:220])}</code><br>Le notebook passe en <b>mode extractif</b>. "
                "Vérifiez la clé, le nom du modèle et l'accès à Internet.", "error")
        LLM.name, LLM.model = "none", "—"
else:
    callout("Aucun modèle connecté : le notebook fonctionne en <b>mode extractif</b> — toutes les étapes fonctionnent, mais les réponses sont "
            "<b>citées</b> à partir des documents au lieu d'être rédigées, et l'assistant ne peut ni compter, ni comparer, "
            "ni résumer. La cellule suivante vous permet de coller une clé à tout moment.", "warning")
````

### 🔑 Étape 0b — Vos clés (Groq, Hugging Face)

Le panneau ci-dessous est **toujours affiché**, quoi qu'il ait été détecté plus haut. Il accepte deux clés, toutes deux facultatives :

| Clé | À quoi elle sert | Nécessaire ? |
|---|---|---|
| **Groq** (`gsk_…`) | Le modèle qui **rédige** les réponses à partir des sources | Seulement si vous voulez des réponses rédigées plutôt que des phrases citées, et qu'aucun Ollama ne tourne |
| **Hugging Face** (`hf_…`) | Télécharger le modèle de recherche sémantique | **Pas nécessaire** pour le modèle par défaut, qui est public. Utile pour un modèle privé ou à accès restreint, ou lorsque les téléchargements anonymes sont limités sur Colab |

### Obtenir une clé Groq gratuite (2 minutes)

Sans modèle, l'assistant cite des phrases ; avec un modèle, il **rédige** la réponse à partir des sources. Groq est le moyen le plus rapide d'y parvenir gratuitement :

1. Ouvrez **`console.groq.com/keys`** et connectez-vous (compte Google ou GitHub, sans carte bancaire).
2. Cliquez sur **Create API Key**, donnez-lui un nom et copiez la clé (`gsk_…`) — elle ne s'affiche qu'une seule fois.
3. Enregistrez-la, par ordre de préférence :
   - **Colab** : 🔑 *Secrets* dans la barre latérale gauche → nom `GROQ_API_KEY`, collez la valeur, activez *Accès au notebook* (*Notebook access*), puis exécutez de nouveau les cellules ci-dessus.
   - **Kaggle** : *Add-ons → Secrets* → même nom, associez-la au notebook.
   - **En local** : `export GROQ_API_KEY=gsk_…` avant de lancer Jupyter.
   - **Ou collez-la dans le champ ci-dessous**, pour cette session uniquement.

Un jeton Hugging Face se crée de la même façon, sur **`huggingface.co/settings/tokens`** (un jeton *Read* suffit), et peut être enregistré sous le nom `HF_TOKEN`.

> ⚠️ Les champs ci-dessous gardent les clés **en mémoire pour cette session uniquement**. Elles ne sont jamais écrites dans le notebook et disparaissent à la fin de la session. Ne tapez jamais une clé dans une cellule de code : les notebooks sont partagés et archivés.

> 💡 **Pas de widgets dans votre éditeur ?** Utilisez plutôt les fonctions : `set_groq_key("gsk_…")` et `set_hf_token("hf_…")`.

````python
# 🔑 Connectez vos clés pour cette session (Groq pour rédiger les réponses, Hugging Face pour le modèle de recherche).
import ipywidgets as widgets

def set_groq_key(key):
    """Utilise une clé Groq pour cette session (enregistrée nulle part)."""
    global LLM
    key = (key or "").strip()
    if not key.startswith("gsk_"):
        return "⛔ Une clé Groq commence par gsk_."
    os.environ["GROQ_API_KEY"] = key
    candidate = LLMClient("groq")
    try:
        candidate.generate("Tu réponds par un seul mot.", "Réponds uniquement : OK", max_tokens=20)
    except Exception as error:
        os.environ.pop("GROQ_API_KEY", None)
        return f"⛔ Clé refusée : {str(error)[:160]}"
    LLM = candidate
    return f"✅ Connecté à Groq · modèle {LLM.model}. Les réponses sont désormais rédigées par le modèle — reposez simplement votre question."

def set_hf_token(token):
    """Utilise un jeton Hugging Face pour cette session (modèles privés ou à accès restreint, limites de téléchargement)."""
    token = (token or "").strip()
    if not token.startswith("hf_"):
        return "⛔ Un jeton Hugging Face commence par hf_."
    os.environ["HF_TOKEN"] = os.environ["HUGGING_FACE_HUB_TOKEN"] = token
    try:
        from huggingface_hub import whoami
        return f"✅ Hugging Face connecté en tant que {whoami(token=token).get('name', 'utilisateur')}."
    except ImportError:
        return "✅ Jeton Hugging Face enregistré pour cette session."
    except Exception as error:
        os.environ.pop("HF_TOKEN", None); os.environ.pop("HUGGING_FACE_HUB_TOKEN", None)
        return f"⛔ Jeton refusé : {str(error)[:160]}"

def keys_panel():
    groq_field = widgets.Password(placeholder="gsk_…", description="Clé Groq", layout=widgets.Layout(width="420px"))
    hf_field = widgets.Password(placeholder="hf_… (facultatif)", description="Jeton HF", layout=widgets.Layout(width="420px"))
    groq_button = widgets.Button(description="Connecter Groq", button_style="success", icon="key")
    hf_button = widgets.Button(description="Connecter HF", icon="key")
    message = widgets.HTML()

    def say(text):
        colour = C["dark_green"] if text.startswith("✅") else C["brick"]
        message.value = f'<span style="color:{colour};">{_html.escape(text)}</span>'

    def connect_groq(_=None):
        value, groq_field.value = groq_field.value, ""
        say(set_groq_key(value))

    def connect_hf(_=None):
        value, hf_field.value = hf_field.value, ""
        say(set_hf_token(value))

    import warnings
    groq_button.on_click(connect_groq); hf_button.on_click(connect_hf)
    with warnings.catch_warnings():
        warnings.simplefilter("ignore")
        groq_field.on_submit(connect_groq); hf_field.on_submit(connect_hf)
    display(widgets.VBox([widgets.HBox([groq_field, groq_button]),
                          widgets.HBox([hf_field, hf_button]), message]))

if LLM.available:
    callout(f"Un modèle est déjà connecté : <b>{LLM.name}</b> · <code>{LLM.model}</code>. "
            "Vous pouvez tout de même coller une autre clé ci-dessous pour en changer.", "success")
else:
    display(HTML(f'{_CSS}<div class="rg" style="background:{C["gold"]}22;border:2px solid {C["gold"]};'
                 f'border-radius:12px;padding:14px 20px;"><div style="font-size:17px;font-weight:800;">'
                 f'🔑 Une clé Groq gratuite est nécessaire pour que les réponses soient <u>rédigées</u></div>'
                 f'<div style="font-size:13.5px;color:{C["slate"]};margin-top:4px;">Créez-en une en deux minutes sur '
                 f'<b>console.groq.com/keys</b> (sans carte bancaire). Sans elle, l\'assistant fonctionne quand même, mais il '
                 f'<b>cite</b> des phrases au lieu de rédiger des réponses — et il ne peut ni compter, ni comparer, ni résumer.'
                 f'</div></div>'))
keys_panel()
````

````python
# @title ⚙️ T1 · Rapports d'exemple et vignettes { display-mode: "form" }
import io, base64, contextlib
import pymupdf

_REPORT_HTML = """
<h1>Rapport annuel sur le marché du travail 2025</h1>
<p class="warn">DONNÉES FICTIVES — document de formation. Aucun chiffre de ce rapport n'est réel.</p>
<h2>1. Avant-propos</h2>
<p>Ce rapport présente les principaux résultats de l'Enquête permanente sur l'emploi (EPE) réalisée en 2025 par l'Institut national de la statistique. Elle couvre la population âgée de 15 ans et plus vivant dans les ménages ordinaires. Les estimations trimestrielles sont corrigées des variations saisonnières.</p>
<h2>2. Principaux résultats</h2>
<ul>
<li>Le taux de chômage annuel moyen s'établit à 7,9 % en 2025, contre 8,4 % en 2024.</li>
<li>Le taux d'activité atteint 61,2 % en moyenne annuelle.</li>
<li>La part de l'emploi informel dans l'emploi total est estimée à 71,5 %.</li>
<li>Le chômage des jeunes (personnes âgées de 15 à 24 ans) reste élevé, à 17,6 %.</li>
</ul>
<h2>3. Population en âge de travailler</h2>
<p>En 2025, la population en âge de travailler compte 11,2 millions de personnes. Les femmes en représentent 50,8 %. La population active est estimée à 6,85 millions de personnes, dont 6,31 millions ont un emploi et 0,54 million sont au chômage.</p>
<p>La population en âge de travailler progresse de 2,7 % par an, plus vite que l'emploi, ce qui maintient une pression continue sur le marché du travail.</p>
<h2>4. Activité et emploi</h2>
<p>Le tableau 1 présente l'évolution trimestrielle des principaux indicateurs. Le taux d'emploi progresse de 0,9 point entre le premier et le quatrième trimestre.</p>
<p><b>Tableau 1 — Indicateurs trimestriels du marché du travail en 2025 (%)</b></p>
<table>
<tr><th>Indicateur</th><th>T1</th><th>T2</th><th>T3</th><th>T4</th></tr>
<tr><td>Taux d'activité</td><td>60,9</td><td>61,5</td><td>61,1</td><td>61,3</td></tr>
<tr><td>Taux d'emploi</td><td>55,8</td><td>56,7</td><td>56,4</td><td>56,7</td></tr>
<tr><td>Taux de chômage</td><td>8,3</td><td>7,8</td><td>7,7</td><td>7,6</td></tr>
<tr><td>Sous-emploi lié à la durée du travail</td><td>12,9</td><td>12,4</td><td>12,2</td><td>12,0</td></tr>
</table>
<p>Source : Institut national de la statistique, EPE 2025 (données fictives).</p>
<h2>5. Chômage</h2>
<p>Le chômage touche davantage les femmes (9,3 %) que les hommes (6,7 %). Il est plus élevé en milieu urbain (10,4 %) qu'en milieu rural (5,2 %). Le tableau 2 détaille le taux de chômage par région.</p>
<p><b>Tableau 2 — Taux de chômage par région en 2025 (%)</b></p>
<table>
<tr><th>Région</th><th>Hommes</th><th>Femmes</th><th>Total</th></tr>
<tr><td>Côte</td><td>8,1</td><td>11,2</td><td>9,6</td></tr>
<tr><td>Centre</td><td>6,4</td><td>8,9</td><td>7,6</td></tr>
<tr><td>Nord</td><td>5,9</td><td>8,0</td><td>6,9</td></tr>
<tr><td>Sud</td><td>6,2</td><td>8,7</td><td>7,4</td></tr>
</table>
<p>La recherche d'emploi dure en moyenne 14 mois. Un chômeur sur trois cherche un emploi depuis plus de deux ans.</p>
<h2>6. Emploi informel</h2>
<p>L'emploi informel regroupe les emplois sans protection sociale ni contrat écrit. Sa part atteint 71,5 % de l'emploi total, et 88,3 % dans l'agriculture. Le revenu mensuel moyen d'un travailleur informel est inférieur de 42 % à celui d'un salarié du secteur formel.</p>
<h2>7. Méthodologie</h2>
<p>L'enquête repose sur un échantillon aléatoire à deux degrés. Six cents zones de dénombrement sont tirées avec une probabilité proportionnelle à leur taille, puis vingt logements sont tirés dans chaque zone, soit 12 000 logements par trimestre.</p>
<p>Le taux de réponse est de 91 %. La non-réponse totale est corrigée par repondération au sein de groupes homogènes définis par la région et le milieu de résidence. Les concepts suivent les recommandations de la 19e Conférence internationale des statisticiens du travail.</p>
<h2>8. Diffusion et confidentialité</h2>
<p>Les résultats trimestriels sont publiés 45 jours après la fin du trimestre de référence. Les fichiers de microdonnées anonymisées sont accessibles aux chercheurs dans le centre d'accès sécurisé aux données de l'Institut, après signature d'un engagement de confidentialité.</p>
<h2>9. Glossaire</h2>
<p><b>Chômeur</b> : personne âgée de 15 ans ou plus, sans emploi, disponible pour travailler et ayant activement recherché un emploi au cours des quatre dernières semaines.</p>
<p><b>Taux d'activité</b> : rapport entre la population active et la population en âge de travailler.</p>
<p><b>Sous-emploi lié à la durée du travail</b> : situation des personnes en emploi qui souhaitent travailler davantage d'heures et sont disponibles pour le faire.</p>
"""

_BULLETIN_HTML = """
<h1>Bulletin trimestriel de l'emploi — premier trimestre 2026</h1>
<p class="warn">DONNÉES FICTIVES — document de formation. Aucun chiffre de ce bulletin n'est réel.</p>
<h2>Chômage</h2>
<p>Au premier trimestre 2026, le taux de chômage s'établit à 7,4 % de la population active, en baisse de 0,2 point par rapport au quatrième trimestre 2025. Il atteint 16,9 % chez les jeunes de 15 à 24 ans, 8,8 % chez les femmes et 6,2 % chez les hommes.</p>
<h2>Activité et emploi</h2>
<p>Le taux d'activité est de 61,4 % et le taux d'emploi atteint 56,9 %. Le sous-emploi lié à la durée du travail touche 11,8 % des personnes en emploi.</p>
<h2>Révisions et diffusion</h2>
<p>Les estimations du quatrième trimestre 2025 publiées dans le rapport annuel sont confirmées. Les séries sont corrigées des variations saisonnières. Le prochain bulletin paraîtra à la mi-août 2026.</p>
"""

_CENSUS_HTML = """
<h1>Recensement général de la population et de l'habitat 2024 — Principaux résultats</h1>
<p class="warn">DONNÉES FICTIVES — document de formation. Aucun chiffre de ce document n'est réel.</p>
<h2>Effectifs</h2>
<p>Au 15 mars 2024, date de référence du recensement, la population résidente du pays s'établit à 18 452 300 personnes, contre 14 208 900 au recensement de 2013. Le taux de croissance annuel moyen entre les deux recensements est estimé à 2,4 %.</p>
<h2>Répartition régionale</h2>
<p><b>Tableau 1 — Population résidente par région en 2024</b></p>
<table>
<tr><th>Région</th><th>Population</th><th>Part (%)</th><th>Part urbaine (%)</th></tr>
<tr><td>Côte</td><td>5 112 000</td><td>27,7</td><td>78,4</td></tr>
<tr><td>Centre</td><td>4 870 500</td><td>26,4</td><td>44,1</td></tr>
<tr><td>Nord</td><td>4 305 800</td><td>23,3</td><td>29,6</td></tr>
<tr><td>Sud</td><td>4 164 000</td><td>22,6</td><td>31,2</td></tr>
</table>
<p>La part de la population vivant en milieu urbain atteint 47,3 % au niveau national.</p>
<h2>Ménages</h2>
<p>Le recensement dénombre 3 618 100 ménages ordinaires. La taille moyenne des ménages est de 5,1 personnes, contre 5,4 en 2013. Un ménage sur quatre est dirigé par une femme.</p>
"""

_REPORT_CSS = """
body {font-family: sans-serif; font-size: 10.5pt; line-height: 1.45;}
h1 {font-size: 20pt; color: #00553A; margin-bottom: 4pt;}
h2 {font-size: 14pt; color: #00704A; margin-top: 14pt; margin-bottom: 4pt;}
p.warn {color: #B83B2E; font-weight: bold;}
table {border-collapse: collapse; margin: 6pt 0;}
th {background-color: #E8F5EF; border: 0.5pt solid #5E6964; padding: 3pt 6pt;}
td {border: 0.5pt solid #5E6964; padding: 3pt 6pt;}
"""

def _render_pdf(html, path, title, running_header, annex=None):
    story = pymupdf.Story(html=html, user_css=_REPORT_CSS)
    buffer = io.BytesIO()
    writer = pymupdf.DocumentWriter(buffer)
    page_size = pymupdf.paper_rect("a5")
    more = True
    while more:
        device = writer.begin_page(page_size)
        more, _ = story.place(page_size + (42, 60, -42, -54))
        story.draw(device)
        writer.end_page()
    writer.close()
    doc = pymupdf.open("pdf", buffer.getvalue())
    for n, page in enumerate(doc, 1):                     # en-tête et pied de page répétés
        page.insert_text((42, 34), running_header, fontsize=7.5, color=(0.37, 0.41, 0.39))
        page.draw_line((42, 40), (page.rect.width - 42, 40), color=(0, 0.66, 0.42), width=0.8)
        page.insert_text((page.rect.width - 70, page.rect.height - 28), f"Page {n}", fontsize=7.5,
                         color=(0.37, 0.41, 0.39))
    if annex:                                             # annexe numérisée : une image, sans couche texte
        draft = pymupdf.open()
        sheet = draft.new_page(width=page_size.width, height=page_size.height)
        sheet.insert_textbox(sheet.rect + (42, 60, -42, -60), annex, fontsize=11)
        scan = doc.new_page(width=page_size.width, height=page_size.height)
        scan.insert_image(scan.rect, stream=sheet.get_pixmap(dpi=150, colorspace=pymupdf.csGRAY).tobytes("png"))
    doc.set_metadata({"title": title, "author": "Institut national de la statistique (fictif)"})
    doc.subset_fonts()
    doc.save(str(path), garbage=3, deflate=True)
    return Path(path)

def create_example_documents(folder):
    folder = Path(folder)
    folder.mkdir(parents=True, exist_ok=True)
    return [
        _render_pdf(_REPORT_HTML, folder / "rapport_marche_travail_2025_fictif.pdf",
                    "Rapport annuel sur le marché du travail 2025 (fictif)",
                    "Institut national de la statistique · Rapport annuel 2025 (fictif)",
                    "Annexe - Note interne (document numérisé)\n\nLe prochain recensement général de la "
                    "population est prévu pour novembre 2027. Le budget prévisionnel de l'opération "
                    "s'élève à 38 millions d'unités monétaires."),
        _render_pdf(_BULLETIN_HTML, folder / "bulletin_emploi_T1_2026_fictif.pdf",
                    "Bulletin trimestriel de l'emploi T1 2026 (fictif)",
                    "Institut national de la statistique · Bulletin de l'emploi (fictif)"),
        _render_pdf(_CENSUS_HTML, folder / "recensement_2024_fictif.pdf",
                    "Recensement de la population 2024 (fictif)",
                    "Institut national de la statistique · Recensement 2024 (fictif)"),
    ]

_PRICES_EN_HTML = """
<h1>Quarterly Consumer Price Note — first quarter 2026</h1>
<p class="warn">FICTITIOUS DATA — training document. None of the figures in this note are real.</p>
<h2>1. Inflation</h2>
<p>In the first quarter of 2026, consumer price inflation stands at 3.2% year on year, compared with 3.9% in the fourth quarter of 2025. The slowdown is mainly explained by the fall in energy prices.</p>
<p><b>Table 1 — Price change by consumption item, first quarter 2026 (%)</b></p>
<table>
<tr><th>Item</th><th>Weight</th><th>Year-on-year change</th></tr>
<tr><td>Food products</td><td>38.5</td><td>5.1</td></tr>
<tr><td>Energy</td><td>9.2</td><td>-2.4</td></tr>
<tr><td>Transport</td><td>11.7</td><td>1.8</td></tr>
<tr><td>Health</td><td>4.3</td><td>2.6</td></tr>
<tr><td>Education</td><td>3.1</td><td>4.0</td></tr>
</table>
<p>Source: National Statistics Institute, consumer price index (fictitious data).</p>
<h2>2. Food prices</h2>
<p>Food prices rise by 5.1% over one year. The increase is stronger in rural areas (6.3%) than in urban areas (4.4%).</p>
<h2>3. Methodology</h2>
<p>The index is computed every month from 14,200 price quotes collected in 42 markets. The index base year is 2020. The next quarterly note will be released at the end of July 2026.</p>
"""

_HOUSEHOLD_ES_HTML = """
<h1>Encuesta Nacional de Hogares 2025 — Principales resultados</h1>
<p class="warn">DATOS FICTICIOS — documento de formación. Ninguna cifra de este documento es real.</p>
<h2>1. Acceso a internet</h2>
<p>En 2025, la proporción de hogares con acceso a internet alcanza el 58,4 %, frente al 49,7 % en 2023. El acceso es mucho mayor en las zonas urbanas (74,2 %) que en las zonas rurales (39,8 %).</p>
<h2>2. Teléfono móvil</h2>
<p>El 91,3 % de las personas de 15 años o más posee un teléfono móvil. Entre las mujeres, la proporción es del 88,6 %.</p>
<h2>3. Educación</h2>
<p>La tasa neta de escolarización en primaria es del 86,1 %. El 12,4 % de los jóvenes de 15 a 24 años no estudia ni trabaja.</p>
<h2>4. Metodología</h2>
<p>La encuesta se basa en una muestra de 9 600 hogares seleccionados en dos etapas. La tasa de respuesta es del 93 %.</p>
"""

_BILINGUAL_HTML = """
<h1>Household Living Conditions Survey 2025 / Enquête sur les conditions de vie des ménages 2025</h1>
<p class="warn">FICTITIOUS DATA — training document. DONNÉES FICTIVES — document de formation.</p>
<h2>Key figures / Chiffres clés</h2>
<p><b>Table 1 / Tableau 1 — Poverty rate by region / Taux de pauvreté par région (%)</b></p>
<table>
<tr><th>Region / Région</th><th>2021</th><th>2025</th></tr>
<tr><td>Coast / Côte</td><td>29.8</td><td>26.1</td></tr>
<tr><td>Centre / Centre</td><td>40.2</td><td>36.9</td></tr>
<tr><td>North / Nord</td><td>52.6</td><td>49.3</td></tr>
<tr><td>South / Sud</td><td>45.1</td><td>41.8</td></tr>
</table>
<h2>1. Poverty</h2>
<p>In 2025, the national poverty rate stands at 38.2% of the population, compared with 41.5% in 2021. Poverty remains much higher in rural areas (51.7%) than in urban areas (22.4%).</p>
<h2>1. Pauvreté</h2>
<p>En 2025, le taux de pauvreté national s'établit à 38,2 % de la population, contre 41,5 % en 2021. La pauvreté reste beaucoup plus élevée en milieu rural (51,7 %) qu'en milieu urbain (22,4 %).</p>
<h2>2. Access to drinking water</h2>
<p>In 2025, 67.3% of households have access to drinking water, compared with 61.0% in 2021. On average, a household that fetches its water spends 24 minutes on each trip.</p>
<h2>3. Accès à l'électricité</h2>
<p>Le taux d'accès des ménages à l'électricité atteint 54,6 % en 2025, contre 47,9 % en 2021. En milieu rural, il n'est que de 31,2 %. La principale source d'éclairage des ménages non raccordés est la lampe solaire.</p>
<h2>4. Methodology / Méthodologie</h2>
<p>The survey covers 8,400 households selected in two stages, and the response rate is 94%. L'enquête couvre 8 400 ménages sélectionnés en deux degrés, et le taux de réponse est de 94 %.</p>
"""

def create_multilingual_examples(folder):
    """Une note en anglais, une enquête en espagnol et une enquête bilingue, en complément des trois rapports en français."""
    folder = Path(folder)
    folder.mkdir(parents=True, exist_ok=True)
    return [
        _render_pdf(_PRICES_EN_HTML, folder / "price_note_Q1_2026_fictitious.pdf",
                    "Quarterly Price Note Q1 2026 (fictitious)",
                    "National Statistics Institute · Price Note (fictitious)"),
        _render_pdf(_HOUSEHOLD_ES_HTML, folder / "encuesta_hogares_2025_ficticia.pdf",
                    "Encuesta Nacional de Hogares 2025 (ficticia)",
                    "Instituto Nacional de Estadística · Encuesta de Hogares (ficticia)"),
        _render_pdf(_BILINGUAL_HTML, folder / "living_conditions_2025_bilingual_fictitious.pdf",
                    "Conditions de vie des ménages 2025 / Living Conditions Survey 2025 (fictif)",
                    "Institut national de la statistique · National Statistics Institute (fictif)"),
    ]

def render_page(data, number, boxes=(), dpi=60):
    """PNG (base64) de la page `number` (numérotée à partir de 1), avec les zones surlignées."""
    doc = pymupdf.open("pdf", data)
    page = doc[number - 1]
    for box in boxes:
        page.draw_rect(pymupdf.Rect(box), color=(0.96, 0.76, 0.26), fill=(0.96, 0.76, 0.26),
                       fill_opacity=0.33, width=1.2)
    return base64.b64encode(page.get_pixmap(dpi=dpi).tobytes("png")).decode()

def thumbnail_html(data, number, boxes=(), dpi=60, width=150, caption=None):
    img = render_page(data, number, boxes, dpi)
    text = caption if caption is not None else f"page {number}"
    return (f'<div style="display:inline-block;margin:4px 8px 4px 0;vertical-align:top;text-align:center;">'
            f'<img src="data:image/png;base64,{img}" style="width:{width}px;border:1px solid #D5DED9;'
            f'border-radius:6px;box-shadow:0 1px 4px rgba(0,0,0,.12);">'
            f'<div style="font-size:11px;color:#5E6964;max-width:{width}px;">{text}</div></div>')
print("✅ T1 prêt.")
````

````python
# @title ⚙️ T2 · Diagnostic et extraction structurée { display-mode: "form" }
from dataclasses import dataclass, field

def diagnose(doc):
    rows = []
    for page in doc:
        chars = len(page.get_text().strip())
        images = len(page.get_images(full=True))
        status = "text" if chars >= 50 else ("scanned" if images else "empty")
        rows.append({"Page": page.number + 1, "Characters": chars, "Images": images, "Status": status})
    return pd.DataFrame(rows)

@dataclass
class Element:
    type: str           # "heading" (titre), "paragraph", "table" ou "ocr"
    text: str
    page: int
    box: tuple          # rectangle (x0, y0, x1, y1) sur la page
    level: int = 0      # niveau de titre (1 = le plus important)
    section: str = ""   # titre le plus proche
    path: str = ""      # chemin complet des titres, p. ex. « 2. Stratégie › 2.4 Domaines prioritaires »
    lang: str = ""      # langue de cet élément ("en", "fr"…), fixée au chargement

FOOTNOTE_MARKER = re.compile(r"(?<=[a-z\)\u201d\"])\.(\d{1,3})(?=\s+[A-Z\u201c\"])")

def _nfkc(t):
    return unicodedata.normalize("NFKC", t)

def page_lines(page):
    lines = []
    for block in page.get_text("dict")["blocks"]:
        if block.get("type") != 0:
            continue
        for line in block["lines"]:
            spans = [s for s in line["spans"] if s["text"].strip()]
            if not spans:
                continue
            lines.append({"text": _nfkc("".join(s["text"] for s in line["spans"])).strip(),
                          "size": round(max(s["size"] for s in spans), 1),
                          "bold": any((s["flags"] & 16) or "Bold" in s.get("font", "") for s in spans),
                          "box": pymupdf.Rect(line["bbox"]), "block": block["number"]})
    return lines

def _signature(text):
    return re.sub(r"\d+", "#", text.lower()).strip()

def _tables(page):
    if not EXTRACT_TABLES:
        return []
    with contextlib.redirect_stdout(io.StringIO()):
        found = page.find_tables().tables
    result = []
    for t in found:
        rows = [[_nfkc((c or "").replace("\n", " ")).strip() for c in r] for r in t.extract()]
        headers = [_nfkc((c or "").replace("\n", " ")).strip() for c in t.header.names]
        if rows and rows[0] == headers:
            rows = rows[1:]
        if rows and len(headers) >= 2:
            result.append({"box": pymupdf.Rect(t.bbox), "headers": headers, "rows": rows})
    return result

def linearise(table, caption):
    sentences = []
    for row in table["rows"]:
        if not any(row):
            continue
        values = "; ".join(f"{table['headers'][j]}: {row[j]}" for j in range(1, len(row))
                           if j < len(table["headers"]) and row[j])
        sentences.append(f"{row[0]} — {values}.")
    return (f"{caption}\n" if caption else "") + "\n".join(sentences)

def ocr_page(page):
    ensure_package("pytesseract", "pytesseract")
    if not shutil.which("tesseract") and shutil.which("apt-get") and ENV["name"] in ("colab", "kaggle"):
        print("Installation de Tesseract (environ une minute)…")
        subprocess.run("apt-get install -y -qq tesseract-ocr tesseract-ocr-fra > /dev/null 2>&1 || "
                       "(apt-get update -qq > /dev/null 2>&1 && apt-get install -y -qq tesseract-ocr tesseract-ocr-fra > /dev/null 2>&1)",
                       shell=True)
    import pytesseract
    from PIL import Image
    installed = set(pytesseract.get_languages(config=""))
    missing = [l for l in OCR_LANGUAGE.split("+") if l not in installed]
    if missing and shutil.which("apt-get") and ENV["name"] in ("colab", "kaggle"):
        packs = " ".join(f"tesseract-ocr-{l}" for l in missing)
        print(f"Installation des langues OCR : {', '.join(missing)}…")
        subprocess.run(f"apt-get install -y -qq {packs} > /dev/null 2>&1 || (apt-get update -qq > /dev/null 2>&1 "
                       f"&& apt-get install -y -qq {packs} > /dev/null 2>&1)", shell=True)
        installed = set(pytesseract.get_languages(config=""))
    language = "+".join(l for l in OCR_LANGUAGE.split("+") if l in installed) or "eng"
    if language != OCR_LANGUAGE:
        print(f"⚠️ Langue OCR « {OCR_LANGUAGE} » indisponible : utilisation de « {language} ».")
    pix = page.get_pixmap(dpi=300)
    image = Image.frombytes("RGB", (pix.width, pix.height), pix.samples)
    text = _nfkc(pytesseract.image_to_string(image, lang=language))
    return [re.sub(r"\s*\n\s*", " ", p).strip() for p in re.split(r"\n\s*\n", text) if len(p.strip()) > 3]

CAPTION = re.compile(r"^(Table|Figure|Chart|Graph|Tableau|Graphique)\b")
LEADER_LINE = re.compile(r"\.{4,}\s*\d{1,4}\s*$|\u2026{2,}")      # « Chapitre 3 ......... 42 »
PAGE_NUMBER_ONLY = re.compile(r"^(page\s*)?[ivxlcdm\d]{1,6}$", re.I)
NUMBERED_HEADING = re.compile(r"^\d{1,2}(\.\d{1,2}){0,3}[.)]?\s+\S")

def extract_structure(doc):
    last = doc.page_count if not MAX_PAGES_PER_DOCUMENT else min(doc.page_count, MAX_PAGES_PER_DOCUMENT)
    pages = {p.number + 1: page_lines(p) for p in doc if p.number < last}
    with_text = [n for n, l in pages.items() if l]
    # 1. en-têtes et pieds de page répétés
    repeated = set()
    if REMOVE_HEADERS and len(with_text) >= 2:
        counts = Counter()
        for n in with_text:
            height = doc[n - 1].rect.height
            counts.update({_signature(l["text"]) for l in pages[n]
                           if l["box"].y1 < height * 0.1 or l["box"].y0 > height * 0.9})
        repeated = {s for s, c in counts.items() if c >= max(2, 0.5 * len(with_text))}
    # 2. taille du corps de texte et tailles des titres
    weights = Counter()
    for l in (l for n in with_text for l in pages[n]):
        weights[l["size"]] += len(l["text"])
    body = weights.most_common(1)[0][0] if weights else 10
    heading_sizes = sorted({l["size"] for n in with_text for l in pages[n]
                            if l["size"] >= body * 1.2 and len(l["text"]) <= 120}, reverse=True)
    level_of = {s: i + 1 for i, s in enumerate(heading_sizes)}
    def is_running_line(line, page):
        if _signature(line["text"]) in repeated:
            return True
        # documents d'une seule page : rien ne se répète, on utilise donc la position et la taille de police
        height = page.rect.height
        in_margin = line["box"].y1 < height * 0.08 or line["box"].y0 > height * 0.92
        return REMOVE_HEADERS and len(with_text) < 2 and in_margin and line["size"] < body * 0.85 and len(line["text"]) < 90
    elements, removed, dropped_toc = [], 0, 0
    for n in sorted(pages):
        page = doc[n - 1]
        if not pages[n]:
            if USE_OCR and page.get_images():
                for paragraph in ocr_page(page):
                    elements.append(Element("ocr", paragraph, n, tuple(page.rect)))
            continue
        tables = _tables(page)
        pieces, current = [], None                     # pieces : (y, élément, bloc[, tableau])
        for l in pages[n]:
            if is_running_line(l, page) or PAGE_NUMBER_ONLY.match(l["text"]):
                removed += 1
                continue
            if DROP_TABLE_OF_CONTENTS and LEADER_LINE.search(l["text"]):
                dropped_toc += 1                      # table des matières, liste des figures…
                continue
            if any(l["box"].intersects(t["box"]) for t in tables):
                continue
            numbered = (NUMBERED_HEADING.match(l["text"]) and len(l["text"]) <= 90
                        and l["size"] >= body * 0.98 and l["bold"])
            if l["size"] in level_of or numbered:
                if current: pieces.append(current); current = None
                if pieces and pieces[-1][1].type == "heading" and pieces[-1][2] == l["block"]:
                    heading = pieces[-1][1]
                    heading.text += " " + l["text"]
                    heading.box = tuple(pymupdf.Rect(heading.box) | l["box"])
                else:
                    pieces.append((l["box"].y0, Element("heading", l["text"], n, tuple(l["box"]),
                                                        level_of.get(l["size"], len(heading_sizes) + 1)), l["block"]))
                continue
            if current and current[2] == l["block"]:
                el = current[1]
                if el.text.endswith("-") and l["text"][:1].islower():
                    el.text = el.text[:-1] + l["text"]          # recolle les mots coupés par un trait d'union
                else:
                    el.text += " " + l["text"]
                el.box = tuple(pymupdf.Rect(el.box) | l["box"])
            else:
                if current: pieces.append(current)
                current = (l["box"].y0, Element("paragraph", l["text"], n, tuple(l["box"])), l["block"])
        if current: pieces.append(current)
        for t in tables:
            pieces.append((t["box"].y0, Element("table", "", n, tuple(t["box"])), None, t))
        pieces.sort(key=lambda p: p[0])
        for p in pieces:
            el = p[1]
            if el.type == "table":
                previous = elements[-1] if elements else None
                table = p[3]
                if (previous is not None and previous.type == "table" and previous.page == n - 1
                        and len(previous.table["headers"]) == len(table["headers"])):
                    # suite du tableau de la page précédente : sa ligne « d'en-tête » est en fait une ligne de données
                    previous.table["rows"] += [table["headers"]] + table["rows"]
                    previous.text = linearise(previous.table, previous.caption)
                    previous.extra.append((n, tuple(table["box"])))
                    continue
                caption = ""
                if previous and previous.type == "paragraph" and CAPTION.match(previous.text):
                    caption = previous.text
                    elements.pop()
                el.text = linearise(table, caption)
                el.table, el.caption, el.extra = table, caption, []
            elements.append(el)
    # 3. rattache chaque élément à sa section et à son chemin complet de titres
    section, stack = "", {}
    for i, el in enumerate(elements):
        if el.type == "heading":
            document_title = i == 0 and el.level == 1 and len(heading_sizes) > 1
            section = "" if document_title else el.text
            if not document_title:
                stack = {lvl: t for lvl, t in stack.items() if lvl < el.level}
                stack[el.level] = el.text
        else:
            el.text = FOOTNOTE_MARKER.sub(". ", el.text)      # « …est élevé.5 Le secteur » → « …est élevé. Le secteur »
        el.section = section if el.type != "ocr" else f"Page {el.page} (texte OCR)"
        el.path = " › ".join(stack[lvl] for lvl in sorted(stack)) or el.section
    return elements, {"headers removed": removed, "contents lines dropped": dropped_toc,
                      "body size": body, "heading sizes": heading_sizes, "pages read": len(pages)}
print("✅ T2 prêt.")
````

````python
# @title ⚙️ T3 · Découpage en passages { display-mode: "form" }
@dataclass
class Passage:
    id: int
    doc_id: int
    doc: str             # titre court du document
    section: str
    text: str
    index_text: str      # texte utilisé pour la recherche (avec un en-tête de contexte)
    boxes: list          # [(page, rectangle), …]
    kind: str = "text"

    @property
    def pages(self):
        return sorted({p for p, _ in self.boxes})

    path: str = ""       # chemin des titres du passage
    parent_id: int = -1  # passage de contexte auquel appartient cette fenêtre de recherche
    lang: str = ""       # langue principale du passage ("en", "fr"…)
    langs: tuple = ()    # toutes les langues présentes (un tableau bilingue, un paragraphe en deux langues)

    @property
    def pages_label(self):
        pages = self.pages
        return f"p. {pages[0]}" if len(pages) == 1 else f"pp. {pages[0]}–{pages[-1]}"

    @property
    def citation(self):
        return f"{self.doc} · {self.section or 'début du document'} · {self.pages_label}"

def _sentences(text):
    return [s.strip() for s in re.split(r"(?<=[.!?;؟])\s+(?=[A-ZÀ-Ý0-9«•–¿¡\"\u0400-\u042F\u0600-\u06FF-])", text) if s.strip()]

def split_passages(document, start=0, max_chars=1000, overlap=2, min_chars=280):
    passages = []
    def add(section, units, kind="text"):
        text = ("\n" if kind == "table" else " ").join(u[0] for u in units)
        boxes = list(dict.fromkeys((u[1], u[2]) for u in units))
        if len(text.strip()) < 40 or text.strip().lower() in seen:
            return                                   # fragments et doublons ajoutent du bruit, jamais de réponses
        seen.add(text.strip().lower())
        breadcrumb = path or section
        # un passage court est recollé au précédent plutôt que laissé seul sans contexte
        if (passages and len(text) < min_chars and passages[-1].kind == kind == "text"
                and passages[-1].section == section and passages[-1].lang == (language or "")
                and len(passages[-1].text) + len(text) <= max_chars * 1.4):
            previous = passages[-1]
            previous.text += " " + text
            previous.index_text += " " + text
            previous.boxes += [b for b in boxes if b not in previous.boxes]
            return
        header = f"{document.title} — {breadcrumb}" if breadcrumb else document.title
        passages.append(Passage(start + len(passages), document.id, document.short, section, text,
                                f"{header}. {text}", boxes, kind, breadcrumb, -1, language or ""))
    seen = set()
    buffer, section, path, language = [], None, "", None
    def flush():
        nonlocal buffer
        i = 0
        while i < len(buffer):
            batch, length, j = [], 0, i
            while j < len(buffer) and (not batch or length + len(buffer[j][0]) <= max_chars):
                batch.append(buffer[j]); length += len(buffer[j][0]) + 1; j += 1
            add(section, batch)
            if j >= len(buffer):
                break
            i = max(j - overlap, i + 1)
        buffer = []
    for el in document.elements:
        if el.type == "heading":
            continue
        el_lang = getattr(el, "lang", "") or getattr(document, "lang", "")
        if el.section != section or el_lang != language:   # nouvelle section, ou changement de langue
            flush()
            section, path, language = el.section, el.path, el_lang
        if el.type == "table":
            flush()
            lines = el.text.split("\n")
            batch = []
            for line in lines:
                if batch and sum(len(x) for x in batch) + len(line) > max_chars:
                    add(section, [(x, el.page, el.box) for x in ([lines[0]] if batch[0] != lines[0] else []) + batch], "table")
                    batch = []
                batch.append(line)
            if batch:
                add(section, [(x, el.page, el.box) for x in ([lines[0]] if batch[0] != lines[0] else []) + batch], "table")
            for p in passages:
                if p.kind == "table" and p.boxes and p.boxes[0] == (el.page, el.box):
                    p.boxes += [b for b in getattr(el, "extra", []) if b not in p.boxes]
            continue
        buffer += [(s, el.page, el.box) for s in _sentences(el.text)]
    flush()
    return passages
# ── Les cinq niveaux de découpage, de la base naïve à celui qui obtient les meilleurs scores ──
CHUNKING_LEVELS = {
    "fixed":     ("1 · Blocs fixes",
                  "Coupe tous les MAX_CHARS caractères, sans tenir compte des phrases, des sections ni des tableaux."),
    "sentences": ("2 · Fenêtres de phrases",
                  "Des phrases entières, N à la fois, en glissant. Rien n'est coupé au milieu d'une phrase, mais "
                  "un passage ne sait plus de quelle section il provient."),
    "sections":  ("3 · Passages structurels",
                  "Suit les titres repérés par l'extracteur ; chaque tableau forme un passage à part et "
                  "le chemin des titres accompagne le texte."),
    "hybrid":    ("4 · Deux granularités",
                  "Indexe les passages structurels ET de petites fenêtres de phrases : la recherche devient précise "
                  "tandis que le lecteur reçoit toujours un passage entier."),
    "smart":     ("5 · Deux granularités, tableaux et contexte",
                  "Ajoute chaque ligne de tableau comme unité à part entière et transmet avec la réponse les "
                  "passages voisins de la section."),
}

def _document_pieces(document):
    """Tous les fragments de texte du document, dans l'ordre de lecture, avec leur page et leur zone."""
    return [(el.text, el.page, el.box) for el in document.elements if el.text.strip()]

def make_passages(document, start=0):
    """Les niveaux 1 et 2 ignorent la structure du rapport ; les niveaux 3 à 5 l'exploitent."""
    if CHUNK_STRATEGY not in ("fixed", "sentences"):
        return split_passages(document, start, MAX_CHARS, OVERLAP_SENTENCES, MIN_CHARS)
    passages, pieces = [], _document_pieces(document)
    def add(text, boxes):
        text = " ".join(text.split())
        if len(text) < 40:
            return
        passages.append(Passage(start + len(passages), document.id, document.short, "", text,
                                f"{document.title}. {text}", list(dict.fromkeys(boxes)), "text", ""))
    if CHUNK_STRATEGY == "fixed":
        overlap, buffer, boxes = int(MAX_CHARS * 0.12), "", []
        for text, page, box in pieces:
            boxes.append((page, box))
            for word in text.split():
                if len(buffer) + len(word) + 1 > MAX_CHARS:     # coupe au milieu d'une phrase
                    add(buffer, boxes)
                    buffer, boxes = buffer[-overlap:], boxes[-2:]
                buffer += " " + word
        add(buffer, boxes)
    else:                                                        # "sentences"
        units = [(s, page, box) for text, page, box in pieces for s in (_sentences(text) or [text])]
        for i in range(0, max(1, len(units) - WINDOW_SENTENCES + 1), WINDOW_STRIDE):
            window = units[i:i + WINDOW_SENTENCES]
            add(" ".join(u[0] for u in window), [(u[1], u[2]) for u in window])
    return passages

def build_search_units(passages, strategy=None, size=None, stride=None):
    """Les niveaux 4 et 5 indexent une seconde granularité, plus fine : de petites fenêtres — et, au
    niveau 5, les lignes de tableau — qui renvoient via `parent_id` au passage à fournir réellement au lecteur."""
    strategy = strategy or CHUNK_STRATEGY
    size, stride = size or WINDOW_SENTENCES, stride or WINDOW_STRIDE
    for p in passages:
        p.parent_id = p.id
    if strategy not in ("hybrid", "smart"):
        return list(passages)
    units, next_id = list(passages), max((p.id for p in passages), default=-1) + 1
    for p in passages:
        if p.kind == "table":
            if strategy != "smart":
                continue
            lines = p.text.split("\n")
            caption = lines[0] if lines and " — " not in lines[0] else ""
            for line in (lines[1:] if caption else lines):
                if len(line) < 15:
                    continue
                text = f"{caption} {line}".strip()                 # une ligne, avec sa légende
                units.append(Passage(next_id, p.doc_id, p.doc, p.section, text,
                                     f"{p.path or p.section}. {text}", p.boxes, "table", p.path, p.id))
                next_id += 1
            continue
        sentences = _sentences(p.text) or [p.text]
        for i in range(0, max(1, len(sentences) - size + 1), stride):
            text = " ".join(sentences[i:i + size])
            if len(text) < 60:
                continue
            units.append(Passage(next_id, p.doc_id, p.doc, p.section, text,
                                 f"{p.path or p.section}. {text}", p.boxes, p.kind, p.path, p.id))
            next_id += 1
    return units

print(f"✅ T3 prêt · niveau de découpage {CHUNKING_LEVELS[CHUNK_STRATEGY][0]}.")
````

````python
# @title ⚙️ T4 · Recherche — multilingue { display-mode: "form" }
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD
from sklearn.preprocessing import Normalizer
from functools import lru_cache
import difflib, hashlib

def strip_accents(text):
    return "".join(c for c in unicodedata.normalize("NFD", text) if unicodedata.category(c) != "Mn")

# ══ 1. Langues ════════════════════════════════════════════════════════════════════════════════════
# Une collection peut mêler plusieurs langues : chaque document, chaque passage et chaque question porte
# son propre code de langue, et chacun est analysé avec les ressources de SA langue.
LANGUAGE_NAMES = {"en": "English", "fr": "French", "es": "Spanish", "pt": "Portuguese", "de": "German",
                  "it": "Italian", "ar": "Arabic", "sw": "Swahili", "ru": "Russian"}
DEFAULT_LANG = DOCUMENT_LANGUAGE if DOCUMENT_LANGUAGE in LANGUAGE_NAMES else "fr"
LANG = DEFAULT_LANG                        # conservé pour la compatibilité ascendante
DECIMAL_COMMA = {"fr", "es", "pt", "de", "it", "ru"}      # « 7,8 » est un nombre décimal dans ces langues

_STOPWORDS_RAW = {
"en": """a an the and or of in on at to for from by with is are was were be been being what which who
whom whose when where why how does do did can could will would shall should may might it its this that these those
there their they them his her he she we you i as than then so such not no nor about into over under also any all
each per much many get give tell me please document documents report reports say says said information show between
stand stands standing acronym abbreviation abbreviations initials mean means called number numbers proposed propose
proposes presented described mentioned according have has had""",
"fr": """le la les l un une des de du d et en au aux est sont ete ce cet cette ces qui que quoi quel quelle quels
quelles comment combien quand ou il elle se sa son ses leur leurs ne pas plus par pour sur dans avec y t n s qu etait
sera peut peuvent fait faire lors entre apres avant tout toute tous toutes ainsi dont tres rapport dit indique selon
informations donne donnent parle concernant sujet ils elles eux lui chez aussi meme on me nous vous je tu mon ma mes
a ont avoir etre etaient document documents nombre liste signifie veut dire""",
"es": """a al algo algunas algunos ante antes como con contra cual cuales cuando cuanto cuanta cuantos cuantas de del
desde donde dos el ella ellas ellos en entre era eran es esa esas ese eso esos esta estas este esto estos fue fueron
ha han hay la las le les lo los mas me mi muy nada ni no nos o otra otro para pero poco por porque que quien quienes
se ser si sin sobre su sus tambien tanto te tiene tienen todo todos tu un una uno unos y ya documento documentos
informe informes segun dice dicen numero lista significa""",
"pt": """a ao aos as com como da das de do dos e ela elas ele eles em entre era essa esse esta este eu foi foram ha
isso isto ja lhe mais mas me mesmo muito na nas nao nem no nos o os ou para pela pelas pelo pelos por qual quais
quando que quem se sem ser seu seus sua suas tambem te tem um uma umas uns documento documentos relatorio relatorios
segundo diz dizem quanto quantos quantas numero lista significa""",
"de": """der die das den dem des ein eine einer eines einem und oder ist sind war waren wird werden hat haben mit von
zu zum zur fur auf aus bei nach uber unter im in an als wie was wer wann wo warum nicht auch es sie er wir ich ihr
dass dieser diese dieses dokument bericht""",
"it": """il lo la i gli le un una uno di del della dei delle degli e ed o in su per con tra fra da che chi cosa come
quando dove perche non piu anche sono era erano essere ha hanno questo questa questi queste quale quali quanto quanti
documento rapporto""",
"sw": """na ya wa kwa ni katika za la cha vya hii huu hizi kuwa au pia lakini kama hiyo yake zake wake nini gani""",
"ar": """في من على إلى الى عن مع هذا هذه ذلك التي الذي ما ماذا كم هل كيف متى أين و أو ثم""",
}
STOPWORDS_BY_LANG = {l: {strip_accents(w.lower()) for w in ws.split()} for l, ws in _STOPWORDS_RAW.items()}
STOPWORDS = set().union(*STOPWORDS_BY_LANG.values())      # toutes les langues ensemble (pour repérer les mots outils seulement)

def stopwords(lang):
    return STOPWORDS_BY_LANG.get(lang, set())

# Mots outils très fréquents, servant à reconnaître une langue (une question n'en a besoin que d'un ou deux).
_MARKERS_RAW = {
    "en": "the and of to in is are was were for with that this which what how from by on be not have has it its who when",
    "fr": "le la les des du et est sont une un pour dans que qui sur avec au aux pas ce cette quel quelle quels quelles combien il elle nous par plus en l d qu n c j ont dont leur leurs ces ne se",
    "es": "el la los las del y es son una un para en que por con al se como cual cuales cuanto cuantos esta este estos entre su sus tiene tienen hay muy fue ha han desde",
    "pt": "o os a as do da dos das e em um uma para que com por nao ao se qual quais quanto quantos esta este no na nos nas pelo pela tem foi sao",
    "de": "der die das und ist sind ein eine mit fur von zu den dem nicht wie was auch im auf",
    "it": "il gli lo della delle dei e sono una per che con non come quale quali nel nella degli",
    "sw": "na ya wa kwa ni katika za la cha vya hii kuwa au pia hiyo",
}
LANGUAGE_MARKERS = {l: {strip_accents(w) for w in ws.split()} for l, ws in _MARKERS_RAW.items()}

def language_scores(text):
    """Nombre de mots marqueurs de chaque langue trouvés dans le texte."""
    t = text.lower()
    letters = [c for c in t if c.isalpha()]
    if letters:
        if sum("؀" <= c <= "ۿ" for c in letters) / len(letters) > 0.3:
            return {"ar": 99}
        if sum("Ѐ" <= c <= "ӿ" for c in letters) / len(letters) > 0.3:
            return {"ru": 99}
    counts = Counter(re.findall(r"[^\W\d_]+", strip_accents(t)))
    scores = {l: sum(counts[w] for w in markers) for l, markers in LANGUAGE_MARKERS.items()}
    scores["es"] += 2 * (t.count("¿") + t.count("¡")) + t.count("ñ")    # lettres qui trahissent une langue
    scores["pt"] += t.count("ã") + t.count("õ")
    scores["fr"] += sum(t.count(ch) for ch in "èùûîïœë") + t.count(" à ")        # lettres que l'espagnol et le portugais n'utilisent jamais
    return scores

def detect_language(text, default=None, minimum=1):
    """Langue la plus probable d'un texte, ou `default` quand les indices sont trop minces ou à égalité."""
    scores = language_scores(text or "")
    if not scores:
        return default
    ranked = sorted(scores.items(), key=lambda x: -x[1])
    best, hits = ranked[0]
    if hits < minimum:
        return default
    if len(ranked) > 1 and ranked[1][1] == hits:              # égalité : garder la valeur par défaut si elle en fait partie
        return default if default in (best, ranked[1][0]) else best
    return best

def question_language(question, collection=(), previous=None):
    """Langue d'une question : imposée par QUESTION_LANGUAGE, détectée, ou héritée du tour précédent."""
    if QUESTION_LANGUAGE != "auto":
        return QUESTION_LANGUAGE
    return (detect_language(question, None, 1) or previous
            or (list(collection)[0] if collection else DEFAULT_LANG))

def passage_language(text, document_lang):
    """Langue d'un passage construit sans la structure du document (niveaux de découpage 1 et 2)."""
    if not DETECT_PASSAGE_LANGUAGE:
        return document_lang
    return detect_language(text, document_lang, 3 if len(text) < 250 else 4)

def passage_lang(p):
    return getattr(p, "lang", "") or DEFAULT_LANG

def passage_langs(p):
    """Toutes les langues d'un passage, la principale en premier."""
    return getattr(p, "langs", ()) or (passage_lang(p),)

# ── Plusieurs langues dans UN SEUL document ───────────────────────────────────────────────────────
def element_language(text, minimum=2):
    """Langue d'un paragraphe, d'un titre ou d'un tableau, ou None quand les indices sont trop minces
    (un titre de deux mots, un tableau de chiffres, un titre bilingue « Poverty / Pauvreté »)."""
    ranked = sorted(language_scores(text or "").items(), key=lambda x: -x[1])
    if not ranked or ranked[0][1] < minimum:
        return None
    if len(ranked) > 1 and ranked[0][1] - ranked[1][1] < max(1, 0.25 * ranked[0][1]):
        return None
    return ranked[0][0]

def assign_languages(elements, forced=None, default=None):
    """Attribue à chaque élément d'un document sa propre langue et renvoie le mélange du document
    (caractères par langue). Un élément trop court pour être reconnu prend la langue de ses
    voisins dans la même section — le paragraphe qui le précède, sinon celui qui le suit — et, en dernier
    recours, la langue principale du document."""
    if forced:
        for e in elements:
            e.lang = forced
        return Counter({forced: sum(len(e.text) for e in elements) or 1})
    found = [element_language(e.text, 1 if e.type == "heading" else 2) for e in elements]
    weights = Counter()
    for e, l in zip(elements, found):
        if l:
            weights[l] += len(e.text)
    majority = weights.most_common(1)[0][0] if weights else (default or DEFAULT_LANG)
    for i, e in enumerate(elements):
        lang = found[i]
        if not lang:
            before = [j for j in range(i - 1, max(-1, i - 7), -1) if found[j] and elements[j].section == e.section]
            after = [j for j in range(i + 1, min(len(elements), i + 7)) if found[j] and elements[j].section == e.section]
            order = (after + before) if e.type == "heading" else (before + after)   # un titre annonce ce qui suit
            lang = found[order[0]] if order else majority
        e.lang = lang
    mix = Counter()
    for e in elements:                  # les tableaux (souvent libellés en deux langues) ne décident pas du mélange
        if e.type != "table":
            mix[e.lang] += len(e.text)
    return mix or Counter(e.lang for e in elements)

def document_languages(document, share=0.1):
    """Langues qui représentent au moins `share` d'un document, la principale en premier."""
    mix = getattr(document, "lang_mix", None) or Counter({getattr(document, "lang", DEFAULT_LANG): 1})
    total = sum(mix.values()) or 1
    return [l for l, n in mix.most_common() if n / total >= share] or [getattr(document, "lang", DEFAULT_LANG)]

def mix_label(mix):
    total = sum(mix.values()) or 1
    return " · ".join(f"{l} {100 * n / total:.0f} %" for l, n in mix.most_common() if n / total >= 0.01)

def passage_languages(p, document_langs=()):
    """La langue principale d'un passage plus toute autre langue réellement présente : un tableau dont
    les libellés sont écrits en deux langues, un paragraphe qui se répète dans une seconde langue."""
    primary = passage_lang(p)
    langs = [primary]
    if p.kind == "table":
        langs += [l for l in document_langs if l != primary]
    elif DETECT_PASSAGE_LANGUAGE:
        # phrase par phrase : un paragraphe qui change de langue en cours de route est analysé dans les deux
        langs += [l for l in (element_language(s, 2) for s in _sentences(p.text) if len(s) > 40) if l]
    return tuple(dict.fromkeys(langs))

# ══ 2. Normalisation, racines, nombres ════════════════════════════════════════════════════════════
SUFFIXES = {
    "en": sorted(["ational", "ations", "ation", "ments", "ment", "ness", "ings", "ing", "ities", "ity", "ies",
                  "ied", "ers", "er", "ed", "es", "ly", "al", "s", "e"], key=len, reverse=True),
    "fr": sorted(["ements", "ement", "ations", "ation", "itions", "ition", "ances", "ance", "ences", "ence",
                  "euses", "euse", "eurs", "eur", "ites", "ite", "iques", "ique", "ives", "ive", "ifs", "if",
                  "elles", "elle", "iels", "iel", "aux", "ants", "ant", "ents", "ent", "ees", "ee", "es", "er",
                  "age", "s", "e", "x"], key=len, reverse=True),
    # solutions de repli légères, utilisées seulement si le paquet snowballstemmer est absent
    "es": sorted(["aciones", "acion", "ciones", "cion", "idades", "idad", "mente", "es", "s"], key=len, reverse=True),
    "pt": sorted(["acoes", "acao", "idades", "idade", "mente", "oes", "es", "s"], key=len, reverse=True),
}
# L'anglais et le français gardent le raciniseur avec lequel ce notebook a été réglé ; les autres langues utilisent Snowball.
_SNOWBALL = {"es": "spanish", "pt": "portuguese", "de": "german", "it": "italian", "ar": "arabic", "ru": "russian"}
_STEMMERS = {}

def _snowball(lang):
    if lang not in _SNOWBALL:
        return None
    if lang not in _STEMMERS:
        try:
            import snowballstemmer
            _STEMMERS[lang] = snowballstemmer.stemmer(_SNOWBALL[lang])
        except Exception:
            _STEMMERS[lang] = None
    return _STEMMERS[lang]

@lru_cache(maxsize=300_000)
def stem(word, lang=None):
    lang = lang or DEFAULT_LANG
    if any(ch.isdigit() for ch in word):
        return word
    stemmer = _snowball(lang)
    if stemmer is not None:
        return stemmer.stemWord(word)
    for _ in range(2 if lang == "en" else 1):
        for suffix in SUFFIXES.get(lang, ()):
            if word.endswith(suffix) and len(word) - len(suffix) >= 4:
                word = word[: -len(suffix)]
                break
    return word

def normalise_numbers(text, lang):
    """Une seule graphie pour chaque nombre, quelle que soit la langue : 18,452,300 · 18 452 300 · 18.452.300 -> 18452300 ;
    7,8 (français, espagnol…) -> 7.8. Un chiffre d'une question en anglais correspond alors à un tableau en français."""
    if lang in DECIMAL_COMMA:
        text = re.sub(r"(?<=\d)[\s  .](?=\d{3}\b)", "", text)
        return re.sub(r"(?<=\d),(?=\d)", ".", text)
    return re.sub(r"(?<=\d),(?=\d{3}\b)", "", text)

def analyse(text, lang=None):
    lang = lang or DEFAULT_LANG
    t = normalise_numbers(strip_accents(text.lower()), lang)
    stops = stopwords(lang)
    return [stem(w, lang) for w in re.findall(r"[^\W_]+(?:\.\d+)?", t) if w not in stops and len(w) > 1]

def analyse_multi(text, langs):
    """Un texte écrit en plusieurs langues : les mots outils de toutes ces langues sont retirés, et
    chaque mot est réduit à sa racine avec les règles de chaque langue (la langue principale en premier)."""
    langs = list(dict.fromkeys(langs)) or [DEFAULT_LANG]
    if len(langs) == 1:
        return analyse(text, langs[0])
    stops = set().union(*(stopwords(l) for l in langs))
    tokens, known = [], set()
    for lang in langs:
        t = normalise_numbers(strip_accents(text.lower()), lang)
        for w in re.findall(r"[^\W_]+(?:\.\d+)?", t):
            if w in stops or len(w) < 2:
                continue
            s = stem(w, lang)
            if lang == langs[0] or s not in known:
                tokens.append(s)
                known.add(s)
    return tokens

_TERMS_CACHE = {}

def passage_terms(p):
    """Ensemble des termes d'un passage, analysé dans chaque langue qu'il contient (mis en cache)."""
    key = (p.id, p.index_text, passage_langs(p))
    if key not in _TERMS_CACHE:
        _TERMS_CACHE[key] = set(analyse_multi(p.index_text, passage_langs(p)))
    return _TERMS_CACHE[key]

def figures_of(p):
    """Les nombres d'un passage dans une seule graphie : 38,2 (français) et 38.2 (anglais) sont le même chiffre."""
    return set(re.findall(r"\d+(?:\.\d+)?", normalise_numbers(p.text.replace("\u00a0", " "), passage_lang(p))))

# ══ 3. Glossaires (dans une langue) et lexique interlangue (entre les langues) ════════════════════
GLOSSARIES = {
    "en": {
        # mots des utilisateurs          -> mots employés dans les publications
        "inhabitants": "population resident people", "residents": "population resident",
        "how many people": "population", "people live": "population",
        "jobless": "unemployment unemployed", "out of work": "unemployment unemployed",
        "young people": "youth 15 24", "youth": "young 15 24",
        "woman": "women", "man": "men", "female": "women", "male": "men",
        "city": "urban", "cities": "urban", "towns": "urban", "countryside": "rural", "villages": "rural",
        "proportion": "share rate", "percentage": "share rate",
        "microdata": "anonymised files", "individual data": "microdata anonymised",
        "get access": "available access", "obtain": "available access",
        "released": "release published", "come out": "released published", "published": "released",
        "when": "published released date", "publication": "published released",
        "did not respond": "non-response reweighting", "non-response": "reweighting",
        "first quarter": "q1", "second quarter": "q2", "third quarter": "q3", "fourth quarter": "q4",
        "change": "down up", "changed": "down up", "evolve": "down up", "trend": "down up",
        "informal": "social protection contract", "wage": "income", "salary": "income", "earnings": "income",
        "sample": "dwellings drawn areas", "definition": "glossary", "what is meant by": "glossary",
        "next census": "census scheduled", "household size": "average size household",
    },
    "fr": {
        "habitants": "population residente personnes", "gens": "population personnes",
        "vivent": "population residente", "combien de personnes": "population",
        "ville": "urbain", "villes": "urbain", "campagne": "rural",
        "sans emploi": "chomage chomeur", "chomeurs": "chomage", "travail": "emploi",
        "jeunes": "15 24 ans", "proportion": "taux part", "pourcentage": "taux part",
        "microdonnees": "fichiers donnees individuelles anonymises",
        "chercheur": "chercheurs", "obtenir": "acceder accessibles acces",
        "sort": "publie publies diffusion paraitra", "sortent": "publie publies diffusion paraitra",
        "sortira": "paraitra publie", "publication": "publies paraitra", "quand": "date publies paraitra",
        "pas repondu": "non reponse repondération", "non-reponse": "repondération",
        "premier trimestre": "t1", "deuxieme trimestre": "t2", "second trimestre": "t2",
        "troisieme trimestre": "t3", "quatrieme trimestre": "t4",
        "evolue": "hausse baisse", "evolution": "hausse baisse", "varie": "hausse baisse",
        "informel": "informelle protection sociale contrat", "salaire": "revenu",
        "echantillon": "logements tirage zones", "definition": "glossaire", "signifie": "glossaire",
        "prochain recensement": "recensement programme", "taille des menages": "taille moyenne menage",
    },
    "es": {"jovenes": "15 24", "cuando": "fecha publicado", "proporcion": "tasa porcentaje"},
    "pt": {"jovens": "15 24", "quando": "data publicado", "proporcao": "taxa percentagem"},
}
GLOSSARY = GLOSSARIES.setdefault(DEFAULT_LANG, {})   # 🧪 ajoutez les vôtres : GLOSSARIES["fr"]["sans boulot"] = "chomage"

# Chaque ligne est UN concept en plusieurs langues (variantes séparées par des espaces). Sans modèle de
# langage, c'est ce qui permet à une question en anglais de trouver un passage en français avec BM25, et au
# score de confiance de le reconnaître. 🧪 Complétez-le avec le vocabulaire de votre domaine.
CROSS_LINGUAL_LEXICON = [
    {"en": "unemployment jobless", "fr": "chômage", "es": "desempleo paro", "pt": "desemprego"},
    {"en": "unemployed", "fr": "chômeur chômeurs", "es": "desempleado desempleados parados", "pt": "desempregado desempregados"},
    {"en": "employment job jobs", "fr": "emploi emplois", "es": "empleo empleos", "pt": "emprego empregos"},
    {"en": "employed", "fr": "occupés", "es": "ocupados", "pt": "ocupados"},
    {"en": "underemployment", "fr": "sous-emploi", "es": "subempleo", "pt": "subemprego"},
    {"en": "rate rates", "fr": "taux", "es": "tasa tasas", "pt": "taxa taxas"},
    {"en": "population inhabitants", "fr": "population habitants", "es": "población habitantes", "pt": "população habitantes"},
    {"en": "people persons", "fr": "personnes", "es": "personas", "pt": "pessoas"},
    {"en": "household households", "fr": "ménage ménages", "es": "hogar hogares", "pt": "agregado agregados domicílio domicílios"},
    {"en": "women woman female", "fr": "femmes femme", "es": "mujeres mujer", "pt": "mulheres mulher"},
    {"en": "men man male", "fr": "hommes homme", "es": "hombres hombre", "pt": "homens homem"},
    {"en": "youth young", "fr": "jeunes jeune jeunesse", "es": "jóvenes joven juventud", "pt": "jovens jovem juventude"},
    {"en": "age aged", "fr": "âge âgés", "es": "edad", "pt": "idade"},
    {"en": "region regions regional", "fr": "région régions régional", "es": "región regiones regional", "pt": "região regiões regional"},
    {"en": "urban city cities", "fr": "urbain urbaine urbains ville villes", "es": "urbano urbana urbanas urbanos ciudad", "pt": "urbano urbana urbanas urbanos cidade"},
    {"en": "rural countryside", "fr": "rural rurale ruraux campagne", "es": "rural rurales", "pt": "rural rurais"},
    {"en": "inflation", "fr": "inflation", "es": "inflación", "pt": "inflação"},
    {"en": "price prices", "fr": "prix", "es": "precio precios", "pt": "preço preços"},
    {"en": "consumer consumption", "fr": "consommation consommateurs", "es": "consumo consumidor consumidores", "pt": "consumo consumidor consumidores"},
    {"en": "quarter quarterly", "fr": "trimestre trimestriel trimestrielle", "es": "trimestre trimestral", "pt": "trimestre trimestral"},
    {"en": "year years annual", "fr": "année années annuel annuelle ans", "es": "año años anual", "pt": "ano anos anual"},
    {"en": "month monthly", "fr": "mois mensuel mensuelle", "es": "mes mensual", "pt": "mês mensal"},
    {"en": "average", "fr": "moyenne moyen", "es": "promedio medio media", "pt": "média médio"},
    {"en": "share proportion percentage", "fr": "part proportion pourcentage", "es": "proporción porcentaje parte", "pt": "proporção percentagem parte"},
    {"en": "income wage wages salary earnings", "fr": "revenu revenus salaire salaires", "es": "ingreso ingresos salario salarios", "pt": "rendimento rendimentos salário salários"},
    {"en": "informal", "fr": "informel informelle", "es": "informal", "pt": "informal"},
    {"en": "survey", "fr": "enquête", "es": "encuesta", "pt": "inquérito pesquisa"},
    {"en": "census", "fr": "recensement", "es": "censo", "pt": "censo recenseamento"},
    {"en": "growth", "fr": "croissance", "es": "crecimiento", "pt": "crescimento"},
    {"en": "activity", "fr": "activité", "es": "actividad", "pt": "atividade actividade"},
    {"en": "labour labor work", "fr": "travail", "es": "trabajo laboral", "pt": "trabalho laboral"},
    {"en": "sample", "fr": "échantillon", "es": "muestra", "pt": "amostra"},
    {"en": "response", "fr": "réponse", "es": "respuesta", "pt": "resposta"},
    {"en": "data microdata", "fr": "données microdonnées", "es": "datos microdatos", "pt": "dados microdados"},
    {"en": "access", "fr": "accès", "es": "acceso", "pt": "acesso"},
    {"en": "published publication release released", "fr": "publié publiée publication diffusion paraîtra",
     "es": "publicado publicación difusión", "pt": "publicado publicação divulgação"},
    {"en": "next", "fr": "prochain prochaine", "es": "próximo próxima", "pt": "próximo próxima"},
    {"en": "food", "fr": "alimentaire alimentaires alimentation", "es": "alimentos alimentario", "pt": "alimentos alimentar"},
    {"en": "energy", "fr": "énergie", "es": "energía", "pt": "energia"},
    {"en": "health", "fr": "santé", "es": "salud", "pt": "saúde"},
    {"en": "education school schooling enrolment", "fr": "éducation école scolarisation", "es": "educación escuela escolarización", "pt": "educação escola escolarização"},
    {"en": "poverty poor", "fr": "pauvreté pauvres", "es": "pobreza pobres", "pt": "pobreza pobres"},
    {"en": "agriculture agricultural", "fr": "agriculture agricole", "es": "agricultura agrícola", "pt": "agricultura agrícola"},
    {"en": "country countries national", "fr": "pays national nationale", "es": "país países nacional", "pt": "país países nacional"},
    {"en": "strategy", "fr": "stratégie", "es": "estrategia", "pt": "estratégia"},
    {"en": "artificial intelligence", "fr": "artificielle intelligence", "es": "artificial inteligencia", "pt": "artificial inteligência"},
    {"en": "innovation", "fr": "innovation", "es": "innovación", "pt": "inovação"},
    {"en": "size", "fr": "taille", "es": "tamaño", "pt": "tamanho dimensão"},
    {"en": "increase increased rise", "fr": "hausse augmentation augmente", "es": "aumento alza aumenta", "pt": "aumento alta aumenta"},
    {"en": "decrease decline fall", "fr": "baisse diminution recul", "es": "disminución descenso baja", "pt": "diminuição queda redução"},
    {"en": "north", "fr": "nord", "es": "norte", "pt": "norte"},
    {"en": "south", "fr": "sud", "es": "sur", "pt": "sul"},
    {"en": "coast coastal", "fr": "côte littoral", "es": "costa", "pt": "costa litoral"},
    {"en": "centre center central", "fr": "centre", "es": "centro", "pt": "centro"},
    {"en": "first", "fr": "premier première", "es": "primer primero primera", "pt": "primeiro primeira"},
    {"en": "second", "fr": "deuxième second", "es": "segundo segunda", "pt": "segundo segunda"},
    {"en": "third", "fr": "troisième", "es": "tercer tercero tercera", "pt": "terceiro terceira"},
    {"en": "fourth", "fr": "quatrième", "es": "cuarto cuarta", "pt": "quarto quarta"},
    {"en": "mobile phone telephone", "fr": "mobile téléphone portable", "es": "móvil teléfono celular", "pt": "móvel telemóvel telefone celular"},
    {"en": "water", "fr": "eau", "es": "agua", "pt": "água"},
    {"en": "drinking", "fr": "potable", "es": "potable", "pt": "potável"},
    {"en": "electricity electric", "fr": "électricité électrique", "es": "electricidad eléctrica", "pt": "eletricidade electricidade elétrica"},
    {"en": "lighting light", "fr": "éclairage", "es": "iluminación alumbrado", "pt": "iluminação"},
]
_LEXICON_INDEX = {}        # (langue, racine) -> {autre langue : racines du même concept}

def build_lexicon_index():
    _LEXICON_INDEX.clear()
    for group in CROSS_LINGUAL_LEXICON:
        stems = {l: {s for w in words.split() for s in analyse(w, l)} for l, words in group.items()}
        for l, own in stems.items():
            for s in own:
                entry = _LEXICON_INDEX.setdefault((l, s), {})
                for other, theirs in stems.items():
                    if other != l:
                        entry.setdefault(other, set()).update(theirs)
build_lexicon_index()      # 🧪 relancez-la après avoir complété CROSS_LINGUAL_LEXICON

def equivalents(term, lang, targets):
    entry = _LEXICON_INDEX.get((lang, term))
    return set().union(*(entry.get(t, set()) for t in targets)) if entry else set()

# ══ 4. Tolérance aux fautes de frappe (par langue) ════════════════════════════════════════════════
# Les utilisateurs font des fautes de frappe : « coutries », « assessement ». Un mot mal tapé n'est corrigé
# qu'avec le vocabulaire de la MÊME langue : un mot français n'est donc jamais « corrigé » en mot anglais.
VOCABULARY_FORMS = {}      # langue -> {racine : forme telle qu'écrite dans les documents}
TRIGRAM_INDEX = {}         # langue -> {trigramme de caractères : racines}

def _trigrams(word):
    padded = f"  {word} "
    return {padded[i:i + 3] for i in range(len(padded) - 2)}

def build_vocabulary(passages):
    forms = {}
    for p in passages:
        mixed = len(passage_langs(p)) > 1
        # un passage en deux langues : chaque phrase alimente le vocabulaire de SA langue
        for piece in (_sentences(p.text) if mixed else [p.text]):
            lang = (element_language(piece, 1) if mixed else None) or passage_lang(p)
            if lang not in passage_langs(p):
                lang = passage_lang(p)
            stops = stopwords(lang)
            for word in re.findall(r"[^\W\d_]{3,}", strip_accents(piece.lower())):
                if word not in stops:
                    forms.setdefault(lang, {}).setdefault(stem(word, lang), Counter())[word] += 1
    VOCABULARY_FORMS.clear(); TRIGRAM_INDEX.clear()
    for lang, roots in forms.items():
        VOCABULARY_FORMS[lang] = {root: c.most_common(1)[0][0] for root, c in roots.items()}
        index = TRIGRAM_INDEX.setdefault(lang, {})
        for root in VOCABULARY_FORMS[lang]:
            for gram in _trigrams(root):
                index.setdefault(gram, set()).add(root)

def closest_term(root, lang=None, minimum=0.86):
    """Terme des documents le plus proche dans la même langue, ou None. Volontairement strict : un mot
    réellement étranger (« maize », « hospital ») doit rester inconnu, car c'est ce qui rend une question refusable."""
    lang = lang or DEFAULT_LANG
    forms, index = VOCABULARY_FORMS.get(lang, {}), TRIGRAM_INDEX.get(lang, {})
    if len(root) < 4 or root in forms:
        return None
    candidates = Counter()
    for gram in _trigrams(root):
        for other in index.get(gram, ()):
            candidates[other] += 1
    best, best_score = None, minimum
    for other, _ in candidates.most_common(40):
        if abs(len(other) - len(root)) > 2:
            continue
        score = difflib.SequenceMatcher(None, root, other).ratio()
        if score > best_score:
            best, best_score = other, score
    return best

def fix_spelling(question, lang=None):
    """Remplace chaque mot mal tapé par le mot le plus proche réellement employé dans les documents de cette langue."""
    lang = lang or DEFAULT_LANG
    forms, stops = VOCABULARY_FORMS.get(lang), stopwords(lang)
    if not forms:
        return question, []
    corrections = []
    def replace(match):
        word = match.group(0)
        plain = strip_accents(word.lower())
        if plain in stops or stem(plain, lang) in forms:
            return word
        near = closest_term(stem(plain, lang), lang)
        if not near:
            return word
        corrections.append((word, forms[near]))
        return forms[near]
    return re.sub(r"[^\W\d_]{4,}", replace, question), corrections

# ══ 5. Sigles, expansion, concepts ════════════════════════════════════════════════════════════════
ACRONYMS = {}          # rempli à partir des documents eux-mêmes : "DIL" -> "Data Innovation Lab"
_ACRONYM_IN_TEXT = re.compile(r"([A-Z][\w’'\-]*(?:\s+[a-z\w’'\-]+){1,5}(?:\s+[A-Z][\w’'\-]*)?)\s*\(([A-Z]{2,6})s?\)")
_ACRONYM_IN_LIST = re.compile(r"(?:^|\n|;|—|\|)\s*([A-Z]{2,6})\s*(?:[—:\-–]|\s{2,})\s*([A-Z][^.;\n|]{4,60})")

def build_acronyms(passages):
    """Les rapports institutionnels regorgent de sigles : on les apprend dans les documents (toutes langues)."""
    found = {}
    for p in passages:
        for expansion, acronym in _ACRONYM_IN_TEXT.findall(p.text):
            found.setdefault(acronym, Counter())[expansion.strip()] += 1
        for acronym, expansion in _ACRONYM_IN_LIST.findall(p.text):
            found.setdefault(acronym, Counter())[expansion.strip()] += 1
    return {a: counts.most_common(1)[0][0] for a, counts in found.items()
            if len(counts.most_common(1)[0][0].split()) >= 2}

def expand_query(question, lang=None):
    lang = lang or DEFAULT_LANG
    q = " " + re.sub(r"[^\w-]+", " ", strip_accents(question.lower())) + " "
    added = [(key, terms) for key, terms in GLOSSARIES.get(lang, {}).items() if f" {key} " in q]
    for acronym, expansion in ACRONYMS.items():
        if f" {acronym.lower()} " in q:
            added.append((acronym, expansion))
    return question + " " + " ".join(t for _, t in added), added

_CONCEPT_CACHE = {}

def concepts(question, lang=None, targets=()):
    """Chaque terme de la question (analysé dans sa langue) avec ses synonymes : le glossaire de cette
    langue et — pour chaque langue de `targets` — les équivalents fournis par le lexique."""
    lang = lang or DEFAULT_LANG
    other = tuple(sorted(t for t in targets if t != lang))
    key = (question, lang, other)
    if key in _CONCEPT_CACHE:
        return _CONCEPT_CACHE[key]
    _, added = expand_query(question, lang)
    synonyms = {}
    for k, terms in added:
        for word in analyse(k, lang) or [stem(strip_accents(k.split()[-1].lower()), lang)]:
            synonyms.setdefault(word, set()).update(analyse(terms, lang))
    items = []
    for t in dict.fromkeys(analyse(question, lang)):
        group = {t} | synonyms.get(t, set())
        if other:                                  # traductions du terme lui-même, pas de ses synonymes
            group |= equivalents(t, lang, other)
        items.append((t, group))
    _CONCEPT_CACHE[key] = items
    return items

def query_terms(question, lang=None, targets=()):
    """Tous les termes que BM25 doit chercher : la question, ses expansions et ses équivalents du lexique."""
    lang = lang or DEFAULT_LANG
    terms = analyse(expand_query(question, lang)[0], lang)
    terms += [t for _, g in concepts(question, lang, targets) for t in sorted(g)]
    return list(dict.fromkeys(terms))

def query_view(passage, question, lang, queries):
    """La version de la question à comparer à un passage : sa traduction dans la langue du passage
    quand elle existe, sinon l'originale (alors rapprochée via le lexique)."""
    pl = passage_lang(passage)
    if queries and queries.get(pl):
        return queries[pl], pl
    return question, lang

# ══ 6. Traduction de la question (avec un modèle de langage) ══════════════════════════════════════
PROMPT_TRANSLATE = """You translate search questions for a document search engine. Translate the question into {language}.
Keep numbers, years, acronyms, codes and proper names exactly as written. Do not answer the question.
Return only the translation, on a single line."""
_TRANSLATIONS = {}

def translate_query(question, target):
    """Traduction de la question vers `target`, mise en cache ; None sans modèle ou en cas d'échec."""
    key = (question, target)
    if key in _TRANSLATIONS:
        return _TRANSLATIONS[key]
    text = None
    if LLM.available and target in LANGUAGE_NAMES:
        try:
            out = LLM.generate(PROMPT_TRANSLATE.format(language=LANGUAGE_NAMES[target]), question, max_tokens=150)
            out = (out.strip().splitlines() or [""])[0].strip().strip('"«»“” ')
            text = out if 3 < len(out) < 400 else None
        except Exception:
            text = None
    _TRANSLATIONS[key] = text
    return text

# ══ 7. Les deux index ═════════════════════════════════════════════════════════════════════════════
class BM25Index:
    """BM25 sur une matrice creuse termes-documents ; chaque unité est analysée avec les règles de sa langue."""

    def __init__(self, passages, k1=1.5, b=0.75):
        from scipy.sparse import csr_matrix
        self.passages, self.k1, self.b = passages, k1, b
        self.docs = [analyse_multi(p.index_text, passage_langs(p)) for p in passages]
        n = len(self.docs)
        self.vocabulary, rows, cols, values = {}, [], [], []
        for i, d in enumerate(self.docs):
            for term, f in Counter(d).items():
                j = self.vocabulary.setdefault(term, len(self.vocabulary))
                rows.append(i); cols.append(j); values.append(f)
        self.matrix = csr_matrix((values, (rows, cols)), shape=(n, max(1, len(self.vocabulary))), dtype=np.float32)
        lengths = np.array([len(d) for d in self.docs], dtype=np.float32)
        self.avg_len = max(1e-9, float(lengths.mean()) if n else 1.0)
        self.norm = (self.k1 * (1 - self.b + self.b * lengths / self.avg_len)).astype(np.float32)
        df = np.asarray((self.matrix > 0).sum(axis=0)).ravel()
        idf = np.log(1 + (n - df + 0.5) / (df + 0.5))
        self.idf_vector = idf.astype(np.float32)
        self.idf = {t: float(idf[j]) for t, j in self.vocabulary.items()}
        self.idf_max = max(self.idf.values()) if self.idf else 1.0

    def search(self, question, k=30, lang=None, targets=()):
        return self.search_terms(query_terms(question, lang, targets), k)

    def search_terms(self, terms, k=30):
        terms = [t for t in terms if t in self.vocabulary]
        if not terms:
            return []
        columns = [self.vocabulary[t] for t in dict.fromkeys(terms)]
        frequencies = self.matrix[:, columns].toarray()
        weights = self.idf_vector[columns] * (self.k1 + 1)
        scores = (weights * frequencies / (frequencies + self.norm[:, None])).sum(axis=1)
        order = [i for i in np.argsort(-scores)[:k] if scores[i] > 0]
        return [(self.passages[i], float(scores[i])) for i in order]

_EMBEDDING_CACHE = {}
CROSS_LINGUAL_MODELS = ("multilingual", "e5", "bge-m3", "labse", "laBSE")

def _embedding_prefixes():
    """Les modèles E5 attendent "query: " et "passage: " devant les textes."""
    return ("query: ", "passage: ") if "e5" in EMBEDDING_MODEL.lower() else ("", "")

class SemanticIndex:
    def __init__(self, passages):
        self.passages, self.mode, self.error = passages, None, None
        self.rows = {p.id: i for i, p in enumerate(passages)}      # vecteur de chaque unité, pour repérer les traductions
        texts = [p.index_text for p in passages]
        if USE_EMBEDDINGS:
            try:
                if INSTALL_EMBEDDINGS:
                    ensure_package("sentence_transformers", "sentence-transformers")
                from sentence_transformers import SentenceTransformer
                if EMBEDDING_MODEL not in _EMBEDDING_CACHE:
                    _EMBEDDING_CACHE[EMBEDDING_MODEL] = SentenceTransformer(EMBEDDING_MODEL)
                self.model = _EMBEDDING_CACHE[EMBEDDING_MODEL]
                texts = [_embedding_prefixes()[1] + t for t in texts]
                key = hashlib.sha1((EMBEDDING_MODEL + "|" + "|".join(texts)).encode("utf-8")).hexdigest()[:16]
                cached = CACHE_DIR / f"embeddings_{key}.npy"
                if USE_CACHE and cached.exists():
                    self.matrix = np.load(cached)
                else:
                    self.matrix = self.model.encode(texts, normalize_embeddings=True, batch_size=32,
                                                    show_progress_bar=len(texts) > 200)
                    if USE_CACHE:
                        np.save(cached, self.matrix)
                self.mode, self.name = "embeddings", f"Embeddings · {EMBEDDING_MODEL.split('/')[-1]}"
            except Exception as error:
                self.error = type(error).__name__
        if self.mode is None:                      # repli hors ligne : analyse sémantique latente
            self.vectorizer = TfidfVectorizer(analyzer="char_wb", ngram_range=(3, 5),
                                              strip_accents="unicode", sublinear_tf=True)
            m = self.vectorizer.fit_transform(texts)
            self.svd = TruncatedSVD(n_components=max(1, min(96, m.shape[0] - 1, m.shape[1] - 1)), random_state=0)
            self.normalizer = Normalizer()
            self.matrix = self.normalizer.fit_transform(self.svd.fit_transform(m))
            self.mode, self.name = "lsa", "Recherche sémantique approchée · LSA hors ligne"

    @property
    def cross_lingual(self):
        """Cet index peut-il rapprocher une question et un passage écrits dans des langues différentes ?"""
        return self.mode == "embeddings" and any(m.lower() in EMBEDDING_MODEL.lower() for m in CROSS_LINGUAL_MODELS)

    def _encode(self, text):
        if self.mode == "embeddings":
            return self.model.encode([_embedding_prefixes()[0] + text], normalize_embeddings=True)
        return self.normalizer.transform(self.svd.transform(self.vectorizer.transform([text])))

    def search(self, question, k=30, lang=None):
        scores = (self.matrix @ self._encode(expand_query(question, lang)[0]).T).ravel()
        order = np.argsort(-scores)[:k]
        return [(self.passages[i], float(scores[i])) for i in order]

def rrf_fusion(rankings, k=60, top=12):
    scores, objects = {}, {}
    for ranking in rankings:
        for rank, (p, _) in enumerate(ranking, start=1):
            scores[p.id] = scores.get(p.id, 0.0) + 1.0 / (k + rank)
            objects[p.id] = p
    order = sorted(scores, key=scores.get, reverse=True)[:top]
    return [(objects[i], scores[i]) for i in order]

# ── Quel type de réponse la question attend (utilisé par le reclassement et l'extracteur) ─────────
QUESTION_TYPES = {
    "count": re.compile(r"\b(how many|how much|number of|combien|nombre de|cu[aá]nt[oa]s?|n[uú]mero de|"
                        r"quant[oa]s?|wie viele?)\b", re.I),
    "date": re.compile(r"\b(when|what date|which year|quand|en quelle ann[eé]e|cu[aá]ndo|en qu[eé] a[nñ]o|"
                       r"quando|em que ano|wann)\b", re.I),
    "list": re.compile(r"\b(list|name the|which \w+ (are|were|have|took)|what are the|quels|quelles|"
                       r"enum[eé]r\w*|cu[aá]les son|quais s[aã]o|lista)\b", re.I),
    "definition": re.compile(r"\b(what is an?|what are an?|define|definition of|stands? for|what does \w+ mean|"
                             r"signifie|qu'est-ce|qu[eé] es|o que [eé]|significa|siglas?|bedeutet)\b", re.I),
}   # l'ordre compte : le premier motif qui correspond l'emporte
NUMBER_IN_TEXT = re.compile(r"\b\d[\d.,]*\b|\b(one|two|three|four|five|six|seven|eight|nine|ten|eleven|twelve|"
                            r"deux|trois|quatre|cinq|six|sept|huit|neuf|dix|onze|douze|"
                            r"tres|cuatro|cinco|seis|siete|ocho|nueve|diez|dois|quatro|sete|oito|nove|dez)\b", re.I)
CROSS_REFERENCE = re.compile(r"\b(figure|table|tableau|graph|chart|annex|annexe|section|page|p\.|cuadro|"
                             r"gr[aá]fico|quadro|anexo|tabla|tabela)\s*\d+", re.I)
ENUMERATION = re.compile(r"(?:,|;|\band\b|\bet\b|\by\b|\be\b)[^,;]{2,70}(?:,|;|\band\b|\bet\b|\by\b|\be\b)", re.I | re.S)
A_YEAR = re.compile(r"\b(19|20)\d{2}\b")

def question_type(question):
    for name, pattern in QUESTION_TYPES.items():
        if pattern.search(question):
            return name
    return "other"

def carries_a_number(sentence):
    """Une vraie quantité, pas un renvoi comme « Figure 8 » ou « voir Tableau 3 »."""
    return bool(NUMBER_IN_TEXT.search(CROSS_REFERENCE.sub(" ", sentence)))

# ══ 8. Les quatre niveaux de reclassement ═════════════════════════════════════════════════════════
RERANKING_LEVELS = {
    "none":          ("0 · Aucun", "Conserve l'ordre produit par la fusion des deux recherches."),
    "coverage":      ("1 · Couverture de la question", "Part des termes pondérés de la question présents dans le passage."),
    "smart":         ("2 · Couverture, densité et type de réponse",
                      "À couverture égale, on départage par le type de réponse (une question « combien » préfère un passage "
                      "contenant une quantité, une question « liste » un passage contenant une énumération), puis par la langue "
                      "de la question, puis par la densité des termes."),
    "cross-encoder": ("3 · Cross-encoder neuronal",
                      "Un modèle lit la question et le passage ensemble et note la paire. Le plus "
                      "précis, et le plus lent : il nécessite sentence-transformers."),
}
ENGLISH_CROSS_ENCODER = "cross-encoder/ms-marco-MiniLM-L-6-v2"
MULTILINGUAL_CROSS_ENCODER = "cross-encoder/mmarco-mMiniLMv2-L12-H384-v1"   # entraîné sur mMARCO, 14 langues

class Reranker:
    def __init__(self, bm25, mode=None, languages=None):
        self.bm25 = bm25
        self.mode = mode or RERANKING
        self.languages = list(languages or [DEFAULT_LANG])
        self.cross, self.model_name = None, None
        if self.mode == "cross-encoder":
            model = CROSS_ENCODER_MODEL
            if model == "auto":     # le modèle anglais seulement pour une collection en anglais interrogée en anglais
                english = set(self.languages) <= {"en"} and QUESTION_LANGUAGE in ("auto", "en")
                model = ENGLISH_CROSS_ENCODER if english and TRANSLATE_QUERIES and LLM.available else MULTILINGUAL_CROSS_ENCODER
            try:
                ensure_package("sentence_transformers", "sentence-transformers")
                from sentence_transformers import CrossEncoder
                self.cross, self.model_name = CrossEncoder(model), model
            except Exception as error:
                print(f"⚠️ Cross-encoder indisponible ({type(error).__name__}) : le reclassement repasse en « smart ».")
                self.mode = "smart"

    @property
    def name(self):
        return RERANKING_LEVELS[self.mode][0]

    def answer_type_bonus(self, kind, passage):
        """Ce passage contient-il le *type* d'élément que demande la question ?"""
        if kind == "count":
            return 1 if carries_a_number(passage.text) else 0
        if kind == "list":
            return 1 if ENUMERATION.search(passage.text) else 0
        if kind == "date":
            return 1 if A_YEAR.search(passage.text) else 0
        return 0

    def weight(self, term, group=None):
        """Poids d'un terme de la question : sa rareté dans la collection. Un terme inconnu de la collection
        mais dont un synonyme ou une traduction est connu prend ce poids ; sinon, c'est un signal fort."""
        if term in self.bm25.idf:
            return self.bm25.idf[term]
        known = [self.bm25.idf[g] for g in (group or ()) if g in self.bm25.idf]
        return max(known) if known else self.bm25.idf_max * 1.5

    def coverage(self, question, passage, lang=None):
        lang = lang or DEFAULT_LANG
        present = passage_terms(passage)
        items = concepts(question, lang, passage_langs(passage))
        total = sum(self.weight(t, g) for t, g in items)
        return sum(self.weight(t, g) for t, g in items if g & present) / total if total else 0.0

    def rerank(self, question, candidates, density=None, kind="other", lang=None, queries=None):
        """Le score renvoyé est toujours la couverture de la question (mesurée avec la version de la question
        dans la langue du passage), afin que le seuil de confiance garde le même sens dans toutes les
        langues ; seul l'ORDRE dépend du niveau choisi."""
        lang = lang or DEFAULT_LANG
        queries = queries or {lang: question}
        views = [query_view(p, question, lang, queries) for p, _ in candidates]
        # la meilleure des deux lectures : la traduction dans la langue du passage, ou la question
        # originale rapprochée via le lexique (un mot traduit absent de toute la collection
        # ne doit pas couler un passage que le mot original trouve déjà)
        scores = [max(self.coverage(question, p, lang), self.coverage(q, p, l) if q is not question else 0.0)
                  for (p, _), (q, l) in zip(candidates, views)]
        if self.mode == "none":
            order = range(len(candidates))
        elif self.mode == "coverage":
            order = sorted(range(len(candidates)), key=lambda i: (-scores[i], i))
        elif self.cross is not None:
            pairs = [(q, " ".join(p.text.split())[:900]) for (p, _), (q, _) in zip(candidates, views)]
            neural = self.cross.predict(pairs, show_progress_bar=False)
            order = sorted(range(len(candidates)), key=lambda i: -float(neural[i]))
        else:
            density = density or {}
            same = lambda p: 1 if PREFER_QUESTION_LANGUAGE and lang in passage_langs(p) else 0
            order = sorted(range(len(candidates)),
                           key=lambda i: (-round(scores[i] / 0.02),
                                          -self.answer_type_bonus(kind, candidates[i][0]),
                                          -same(candidates[i][0]),
                                          -density.get(candidates[i][0].id, 0.0), i))
        return [(candidates[i][0], scores[i]) for i in order]

def diversify(ranked, k, several_documents):
    """Au plus k-1 passages d'un même document quand plusieurs documents sont interrogés."""
    cap = max(1, k - 1) if several_documents else k
    chosen, count = [], Counter()
    for p, s in ranked:
        if count[p.doc_id] < cap:
            chosen.append((p, s)); count[p.doc_id] += 1
        if len(chosen) == k:
            return chosen
    for p, s in ranked:
        if (p, s) not in chosen and len(chosen) < k:
            chosen.append((p, s))
    return chosen
print(f"✅ T4 prêt · multilingue · langue par défaut {DEFAULT_LANG} · "
      f"{len(CROSS_LINGUAL_LEXICON)} concepts interlangues · reclassement {RERANKING_LEVELS[RERANKING][0]}"
      + ("" if _snowball("es") else " · ⚠️ snowballstemmer absent : raciniseurs légers pour es/pt/de/it"))
````

````python
# @title ⚙️ T5 · Mémoire de la conversation { display-mode: "form" }
FOLLOW_UP_STARTS = ("and ", "and,", "what about", "how about", "same for", "same question", "same thing",
                    "also ", "for ", "et ", "et,", "pour ", "qu en est il", "qu'en est-il", "meme question",
                    "idem", "aussi ", "meme chose", "et si ",
                    "y ", "y,", "¿y ", "y para", "y en", "lo mismo", "tambien", "e ", "e,", "e para", "e no ", "e na ",
                    "e quanto", "o mesmo", "und ", "und fur", "ebenso")
INTERROGATIVE = re.compile(r"^(what|which|how|who|when|where|why|is |are |does |do |did |can |could |was |were |"
                           r"will |should |quel|quelle|quels|quelles|combien|comment|qui|quand|ou |pourquoi|"
                           r"est-ce|que |qu'|y a-t-il|¿|cual|cuales|cuanto|cuantos|como|donde|cuando|por que|que |"
                           r"qual|quais|quanto|quantos|onde|porque|o que|was |wie |wer |wann |welche)")
CATEGORIES = {
    "region": r"\b(coast|c[oô]te|centre|center|north|south|littoral|nord|sud|costa|centro|norte|sur|sul)\b",
    "sex": r"\b(women|woman|men|man|female|male|femmes?|hommes?|mujer(?:es)?|hombres?|mulher(?:es)?|homens|homem|"
           r"frauen|manner)\b",
    "quarter": r"\b(q[1-4]|t[1-4]|(?:first|second|third|fourth) quarter|"
               r"(?:premier|deuxieme|second|troisieme|quatrieme) trimestre|"
               r"(?:primer|segundo|tercer|cuarto|primeiro|terceiro|quarto) trimestre)\b",
    "area": r"\b(urban|rural|cities|city|countryside|urbain|villes?|campagnes?|urbanos?|urbanas?|rurales|rurais)\b",
    "age": r"\b(young people|youth|adults|older people|jeunes|adultes|seniors|jovenes|jovens|adultos)\b",
    "year": r"\b((?:19|20)\d{2})\b",
    "indicator": r"\b((?:unemployment|activity|employment) rate|underemployment|informal employment|inflation|"
                 r"taux (?:de chomage|d'activite|d'emploi)|sous-emploi|emploi informel|"
                 r"tasa de (?:desempleo|actividad|empleo)|taxa de (?:desemprego|atividade|emprego)|inflacion|inflacao)\b",
}

def _aligned(text):
    """Minuscules sans accents, en conservant la même longueur (mêmes positions)."""
    return "".join((strip_accents(ch) or ch)[:1] for ch in text.lower())

def is_follow_up(question, history):
    if not history:
        return False
    q = _aligned(question).strip()
    if q.startswith(tuple(_aligned(s) for s in FOLLOW_UP_STARTS)):
        return True
    return len(analyse(question, detect_language(question, DEFAULT_LANG))) <= 2 and not INTERROGATIVE.match(q)

PROMPT_FOLLOW_UP = """Vous reformulez des questions. À l'aide de l'historique de la conversation, réécrivez la DERNIÈRE
question de l'utilisateur pour qu'elle se comprenne seule, sans l'historique. Ne répondez pas à la question.
Renvoyez uniquement la question réécrite, dans la même langue que la question de l'utilisateur, sur une seule ligne."""

def rewrite_follow_up(question, history, use_model=True):
    """Renvoie (question autonome, réécrite ?, méthode)."""
    if not is_follow_up(question, history):
        return question, False, None
    previous = history[-1]["standalone_question"]
    if use_model and LLM.available:
        try:
            turns = "\n".join(f"Utilisateur : {h['standalone_question']}\nAssistant : {h['answer'][:200]}"
                              for h in history[-3:])
            text = LLM.generate(PROMPT_FOLLOW_UP, f"HISTORIQUE\n{turns}\n\nDERNIÈRE QUESTION\n{question}",
                                max_tokens=120).strip().strip('"«» ').splitlines()[0]
            if 5 < len(text) < 300:
                return text, True, "model"
        except Exception:
            pass
    result, q_al, replaced = previous, _aligned(question), 0
    for pattern in CATEGORIES.values():
        new = re.search(pattern, q_al)
        if not new:
            continue
        old = re.search(pattern, _aligned(result))
        if old:
            result = result[:old.start()] + question[new.start():new.end()] + result[old.end():]
            replaced += 1
    if not replaced:
        rest = re.sub(r"^\s*(and|what about|how about|same for|also|et|pour|qu'en est-il|aussi|¿?y|e|und|tambi[eé]n|tamb[eé]m)\b[\s,]*", "",
                      question, flags=re.I)
        result = previous.rstrip(" ?") + " — " + rest.rstrip(" ?") + "?"
    return result, True, "rules"
print("✅ T5 prêt.")
````

````python
# @title ⚙️ T6 · La bibliothèque : chargement, indexation, réponses, vérifications, retours { display-mode: "form" }
REFUSALS = {
    "en": "Information not available in the documents.",
    "fr": "Information non disponible dans les documents.",
    "es": "Información no disponible en los documentos.",
    "pt": "Informação não disponível nos documentos.",
    "de": "Information in den Dokumenten nicht verfügbar.",
    "it": "Informazione non disponibile nei documenti.",
    "ar": "المعلومة غير متوفرة في الوثائق.",
    "sw": "Taarifa hii haipatikani katika nyaraka.",
}
REFUSAL = REFUSALS["fr"]      # refus par défaut : le français

# noms des langues en français, pour l'affichage et la consigne du modèle uniquement
NOMS_LANGUES = {"en": "anglais", "fr": "français", "es": "espagnol", "pt": "portugais", "de": "allemand",
                "it": "italien", "ar": "arabe", "sw": "swahili", "ru": "russe"}

def nom_langue(lang):
    return NOMS_LANGUES.get(lang, LANGUAGE_NAMES.get(lang, lang))

def refusal_in(lang):
    return REFUSALS.get(lang, REFUSAL)

def strict_prompt(answer_lang):
    """La consigne système, avec la phrase de refus et la langue de réponse de cette question."""
    language = f"en {NOMS_LANGUES[answer_lang]}" if answer_lang in NOMS_LANGUES else "dans la langue de la question"
    return f"""Vous êtes l'assistant documentaire d'un institut national de statistique.
Vous répondez à partir d'extraits d'un ou plusieurs documents PDF, qui peuvent être rédigés dans des langues DIFFÉRENTES.
Règles, par ordre de priorité :
1. Répondez UNIQUEMENT à partir des sources numérotées fournies. Si elles ne contiennent pas la réponse,
   écrivez exactement : "{refusal_in(answer_lang)}" et rien d'autre.
2. N'utilisez jamais vos connaissances générales, même si vous pensez connaître la réponse.
3. Recopiez les chiffres exactement, avec leur unité, leur période et le document dont ils proviennent.
   Une virgule décimale ("7,8") et un point décimal ("7.8") écrivent le même nombre.
4. Si la question compare des périodes ou des documents, donnez chaque valeur avec sa source.
5. Terminez chaque phrase par le numéro de sa source entre crochets, par exemple [1].
6. Les sources tirées de tableaux s'écrivent "libellé — colonne : valeur ; …".
   Une source peut elle-même mêler deux langues (un tableau bilingue, un paragraphe répété dans une seconde langue) :
   lisez-la dans les deux, et n'utilisez qu'une seule fois une information répétée.
7. Rédigez toute la réponse {language}, quelle que soit la langue des sources : traduisez fidèlement,
   et conservez les sigles, les noms propres et les intitulés de programmes tels qu'ils figurent dans la source.
8. Quatre phrases au maximum."""

STRICT_PROMPT = strict_prompt(DEFAULT_LANG)

COMPARISON = re.compile(r"\b(chang\w*|evolv\w*|evolu\w*|trend|between|entre|compar\w*|versus|vs|difference|"
                        r"increase\w*|decrease\w*|hausse|baisse|progress\w*|par rapport|cambi\w*|variac\w*|"
                        r"aument\w*|dismin\w*|diminu\w*|mudan\w*|frente a|em relacao)\b")
PRONOUN_START = re.compile(r"^(It|Its|They|Their|This|These|Il|Elle|Ils|Elles|Ce|Cette|Ces|Sa|Son|"
                           r"Él|Ella|Ellos|Ellas|Esta|Este|Estos|Su|Ele|Ela|Eles|Elas|Isso|Seu|Sua)\b")

@dataclass
class PDFDocument:
    id: int
    filename: str
    title: str
    data: bytes
    pages: int
    diagnosis: object
    elements: list
    info: dict
    lang: str = "fr"    # langue principale, détectée au chargement (ou DOCUMENT_LANGUAGE)
    lang_mix: object = None   # caractères par langue : un rapport peut en contenir plusieurs

    @property
    def short(self):
        t = re.sub(r"\s*\((fictitious|fictif|fictive|ficticia|ficticio)\)\s*$", "", self.title)
        return t if len(t) <= 42 else t[:40].rstrip() + "…"

    @property
    def scanned_pages(self):
        return self.diagnosis.loc[self.diagnosis["Status"] == "scanned", "Page"].tolist()

def is_refusal(text):
    t = strip_accents(text.lower())
    return any(m in t for m in ["information not available", "does not contain", "do not contain",
                                "no information", "i don't know", "i do not know", "cannot answer",
                                "does not mention", "not mentioned", "does not specify",
                                "information non disponible", "ne contient pas", "ne contiennent pas",
                                "pas d'information", "aucune information", "je ne sais pas",
                                "ne permet pas de repondre", "ne precise pas", "ne mentionne pas",
                                "informacion no disponible", "no contiene", "no contienen", "no hay informacion",
                                "no se menciona", "informacao nao disponivel", "nao contem", "nao ha informacao",
                                "nao menciona", "nicht verfugbar", "keine information"]
               + [strip_accents(r.lower()).rstrip(".") for r in REFUSALS.values()])

def _normalise_numbers(text):
    t = text.replace("\u00a0", " ").replace("\u202f", " ")
    t = re.sub(r"(?<=\d),(?=\d{3}\b)", "", t)       # 18,452,300 -> 18452300
    t = re.sub(r"(?<=\d) (?=\d{3}\b)", "", t)       # 18 452 300 -> 18452300
    return re.sub(r"(?<=\d),(?=\d)", ".", t)         # 7,8 -> 7.8

def check_figures(text, passages):
    """Chaque nombre de la réponse doit figurer dans au moins une des sources fournies."""
    without_refs = re.sub(r"\[\d+\]", " ", _normalise_numbers(text))
    numbers = list(dict.fromkeys(re.findall(r"\d+(?:\.\d+)?", without_refs)))
    source = _normalise_numbers(" ".join(p.text for p, _ in passages))
    missing = [n for n in numbers if not re.search(rf"(?<![\d.]){re.escape(n)}(?!\d|\.\d)", source)]
    return {"numbers": len(numbers), "missing": missing}

def missing_terms(question, passage, lang=None):
    """Mots de la question sans équivalent (synonyme ou traduction) dans le passage retenu."""
    lang = lang or DEFAULT_LANG
    present = passage_terms(passage)
    absent = {t for t, g in concepts(question, lang, passage_langs(passage)) if not g & present}
    words = []
    for word in re.findall(r"[\w'-]+", question):
        if any(t in absent for t in analyse(word, lang)) and word not in words:
            words.append(word)
    return words

def relax_query(question, results, vocabulary, lang=None, targets=()):
    """Retire les mots de la question qu'aucun des passages retrouvés ne contient.
    « le nombre de dimensions du cadre proposé par l'AfDB » → « dimensions AfDB » :
    les mots interrogatifs et les distracteurs hors sujet disparaissent, le sujet reste.

    N'assouplit jamais quand un mot de la question est absent de toute la collection
    (« maïs », « hôpitaux ») : ce mot est justement le signal d'une question hors du champ."""
    items = concepts(question, lang, targets)
    if not results or len(items) < 2:
        return ""
    if any(not (group & vocabulary) for _, group in items):
        return ""
    present = set()
    for p, _ in results:
        present |= passage_terms(p)
    kept = [t for t, group in items if group & present]
    return " ".join(kept) if 1 < len(kept) < len(items) else ""

FEEDBACK = []
def record_feedback(turn, useful, comment="", channel="notebook"):
    row = {"timestamp": time.strftime("%Y-%m-%d %H:%M:%S"), "channel": channel, "useful": bool(useful),
           "question": turn.get("question", ""), "standalone_question": turn.get("standalone_question", ""),
           "answer": turn.get("answer", ""), "sources": " | ".join(turn.get("citations", [])),
           "comment": comment}
    FEEDBACK.append(row)
    path = OUTPUT_DIR / "user_feedback.csv"
    pd.DataFrame([row]).to_csv(path, mode="a", header=not path.exists(), index=False, encoding="utf-8-sig")
    return row

def _split_sentences(text):
    return [s.strip() for s in re.split(r"(?<=[.!?؟])\s+|\n", text) if len(s.strip()) > 15]

LIST_INTRO = re.compile(r"(:|following|as follows|are:|comprend|suivants?|suivantes?|siguientes?|seguintes?|folgende)\s*$", re.I)

def _best_sentence(items, passage, weight, text=None, kind="other", prefer_lang=None):
    """Meilleure phrase du passage, prolongée quand elle introduit une liste ou est très courte.
    Une question « combien » privilégie les phrases qui contiennent vraiment un nombre."""
    sentences = _split_sentences(text or passage.text)
    best, best_score = None, 0.0
    for n, sentence in enumerate(sentences):
        terms = set(analyse_multi(sentence, passage_langs(passage)))
        if passage.kind == "table" and n > 0:
            terms |= set(analyse_multi(sentences[0], passage_langs(passage)))   # une ligne de tableau hérite de sa légende
        score = sum(weight(t, g) for t, g in items if g & terms)
        if kind == "count":
            score *= 1.6 if carries_a_number(sentence) else 0.6
        elif kind == "list":                                # « listez les pays… » attend la liste elle-même
            names = len(re.findall(r"(?<!^)(?<![.!?]\s)\b[A-Z][a-z]{2,}", sentence))
            score *= (1.35 if ENUMERATION.search(sentence) else 0.9) * (1 + 0.07 * min(names, 6))
        elif kind == "date" and re.search(r"\b(19|20)\d{2}\b|\b\d{1,2} (January|February|March|April|May|June|July|"
                                          r"August|September|October|November|December)\b", sentence):
            score *= 1.4
        if prefer_lang and len(passage_langs(passage)) > 1 and element_language(sentence, 1) == prefer_lang:
            score *= 1.2                                       # passage bilingue : citer la langue du lecteur
        if score > best_score:
            text = sentence
            if n > 0 and PRONOUN_START.match(text):
                text = sentences[n - 1] + " " + text
            if passage.kind == "table" and n > 0:
                text = f"{sentences[0]}: {text}"
            extra = n + 1
            while (extra < len(sentences) and len(text) < 420
                   and (LIST_INTRO.search(text) or len(text) < 130)):
                text += " " + sentences[extra]             # une liste ou une phrase courte a besoin de sa suite
                extra += 1
            best, best_score = text, score
    return best, best_score

class Library:
    def __init__(self):
        self.documents, self.passages = [], []
        self.threshold, self.calibration, self.ready = 0.5, None, False
        self.language_counts = Counter()
        self.folder = ENV["root"] / "pdf"
        self.folder.mkdir(exist_ok=True)

    # --- chargement -----------------------------------------------------------------------------
    BROWSER_HEADERS = {
        "User-Agent": ("Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) "
                       "Chrome/125.0.0.0 Safari/537.36"),
        "Accept": "application/pdf,text/html;q=0.9,*/*;q=0.8",
        "Accept-Language": "en-US,en;q=0.9,fr;q=0.8",
    }

    @staticmethod
    def direct_url(url):
        """Transforme un lien de partage en lien de téléchargement direct (Google Drive, Dropbox, OneDrive)."""
        drive = re.search(r"(?:drive|docs)\.google\.com/(?:file/d/|open\?id=|uc\?id=|.*[?&]id=)([\w-]{20,})", url)
        if drive:
            return f"https://drive.usercontent.google.com/download?id={drive.group(1)}&export=download&confirm=t", f"gdrive_{drive.group(1)[:10]}.pdf"
        if "dropbox.com" in url:
            base = re.sub(r"[?&]dl=[01]", "", url)
            return base + ("&" if "?" in base else "?") + "dl=1", None
        if "1drv.ms" in url or "sharepoint.com" in url:
            return url + ("&" if "?" in url else "?") + "download=1", None
        return url, None

    def download(self, url, timeout=90):
        """Certains sites institutionnels refusent les scripts anonymes : on envoie des en-têtes de navigateur,
        puis on se rabat sur curl avant d'abandonner avec un message clair.
        Les liens de partage (Google Drive, Dropbox, OneDrive) sont convertis en liens directs."""
        url, forced_name = self.direct_url(url)
        name = forced_name or re.sub(r"[^\w.-]", "_", url.split("?")[0].rstrip("/").split("/")[-1]) or "document"
        target = self.folder / (name if name.lower().endswith(".pdf") else name + ".pdf")
        if target.exists() and target.stat().st_size > 1000:
            return target
        headers = dict(self.BROWSER_HEADERS)
        headers["Referer"] = "/".join(url.split("/")[:3]) + "/"
        try:
            request = urllib.request.Request(url, headers=headers)
            with urllib.request.urlopen(request, timeout=timeout) as response:
                content = response.read()
            if content[:4] == b"%PDF":
                target.write_bytes(content)
                return target
            first = "l'adresse a renvoyé une page web, pas un PDF (lien de partage, page de connexion ou avertissement antivirus)"
            page = content[:4000].decode("utf-8", "ignore")
            token = re.search(r'name="(confirm|uuid)" value="([^"]+)"', page)
            if token:                                   # avertissement antivirus de Google Drive pour les gros fichiers
                with urllib.request.urlopen(urllib.request.Request(
                        f"{url}&{token.group(1)}={token.group(2)}", headers=headers), timeout=timeout) as response:
                    content = response.read()
                if content[:4] == b"%PDF":
                    target.write_bytes(content)
                    return target
        except Exception as error:
            first = f"{type(error).__name__}: {str(error)[:80]}"
        if shutil.which("curl"):                    # curl négocie TLS comme un navigateur : cela suffit souvent
            command = ["curl", "-sSL", "--max-time", str(timeout), "-A", headers["User-Agent"],
                       "-H", f"Accept: {headers['Accept']}", "-H", f"Referer: {headers['Referer']}",
                       "-o", str(target), url]
            if subprocess.run(command, capture_output=True).returncode == 0 and target.exists() \
                    and target.read_bytes()[:4] == b"%PDF":
                return target
            target.unlink(missing_ok=True)
        raise RuntimeError(f"échec du téléchargement — {first}. Vérifiez que le lien est public (« tous les utilisateurs disposant du lien »), "
                           f"ou téléchargez le PDF à la main, puis réglez UPLOAD_PDF = True (Colab) "
                           f"ou indiquez son chemin dans DOCUMENTS")

    def _resolve(self, sources):
        paths, failures = [], []
        for s in ([sources] if isinstance(sources, (str, Path)) else sources):
            s = str(s)
            try:
                if s in ("exemple", "example"):              # « example » : ancien mot-clé, toujours accepté
                    paths += create_example_documents(self.folder / "examples")
                elif s in ("exemple-multilingue", "example-multilingual"):
                    paths += create_example_documents(self.folder / "examples")
                    paths += create_multilingual_examples(self.folder / "examples")
                elif s.startswith(("http://", "https://")):
                    paths.append(self.download(s))      # un échec ici ne doit pas arrêter les autres documents
                elif Path(s).is_dir():
                    found = sorted(Path(s).rglob("*.pdf"))
                    paths += found
                    if not found:
                        failures.append((s, "aucun PDF dans ce dossier"))
                elif Path(s).exists():
                    paths.append(Path(s))
                else:
                    failures.append((s, "source introuvable"))
            except Exception as error:
                failures.append((s.split("/")[-1][:60], str(error)[:200]))
        self._source_failures = failures
        return paths

    @staticmethod
    def _to_json(elements, info):
        def clean(t):
            return {"box": list(map(float, t["box"])), "headers": t["headers"], "rows": t["rows"]}
        return json.dumps({"info": info, "elements": [
            {"type": e.type, "text": e.text, "page": e.page, "box": list(map(float, e.box)),
             "level": e.level, "section": e.section,
             "table": clean(e.table) if getattr(e, "table", None) else None,
             "caption": getattr(e, "caption", ""), "extra": [[p, list(map(float, b))] for p, b in getattr(e, "extra", [])]}
            for e in elements]}, ensure_ascii=False)

    @staticmethod
    def _from_json(text):
        payload = json.loads(text)
        elements = []
        for item in payload["elements"]:
            e = Element(item["type"], item["text"], item["page"], tuple(item["box"]), item["level"], item["section"])
            if item["table"]:
                e.table = {"box": pymupdf.Rect(item["table"]["box"]), "headers": item["table"]["headers"],
                           "rows": item["table"]["rows"]}
                e.caption, e.extra = item["caption"], [(p, tuple(b)) for p, b in item["extra"]]
            elements.append(e)
        return elements, payload["info"]

    def _cache_key(self, data):
        settings = f"{REMOVE_HEADERS}|{EXTRACT_TABLES}|{USE_OCR}|{OCR_LANGUAGE}|{DROP_TABLE_OF_CONTENTS}|{MAX_PAGES_PER_DOCUMENT}|v2"
        return hashlib.sha1(data + settings.encode("utf-8")).hexdigest()[:16]

    def load(self, sources):
        added, errors = [], []
        known = {d.filename for d in self.documents}
        paths = self._resolve(sources)
        errors += getattr(self, "_source_failures", [])
        for number, path in enumerate(paths, 1):
            try:
                data = Path(path).read_bytes()
                if not data.startswith(b"%PDF"):
                    raise ValueError("ce n'est pas un fichier PDF")
                doc = pymupdf.open(stream=data, filetype="pdf")
                if doc.needs_pass:
                    raise ValueError("PDF protégé par un mot de passe")
                if Path(path).name in known:
                    continue
                start = time.time()
                cache_file = CACHE_DIR / f"extraction_{self._cache_key(data)}.json"
                if USE_CACHE and cache_file.exists():
                    elements, info = self._from_json(cache_file.read_text(encoding="utf-8"))
                    origin = "depuis le cache"
                else:
                    elements, info = extract_structure(doc)
                    origin = "extrait"
                    if USE_CACHE:
                        cache_file.write_text(self._to_json(elements, info), encoding="utf-8")
                if DOCUMENT_LANGUAGE in LANGUAGE_NAMES:            # forcée : une seule langue partout
                    lang_mix = assign_languages(elements, DOCUMENT_LANGUAGE)
                elif DETECT_PASSAGE_LANGUAGE:                       # paragraphe par paragraphe
                    lang_mix = assign_languages(elements, None, DEFAULT_LANG)
                else:                                               # une seule langue pour tout le document
                    lang_mix = assign_languages(elements, detect_language(
                        " ".join(e.text for e in elements[:600])[:40000], DEFAULT_LANG, 3))
                lang = lang_mix.most_common(1)[0][0] if lang_mix else DEFAULT_LANG
                print(f"  {number}/{len(paths)} · {Path(path).name[:48]} · {doc.page_count} pages · "
                      f"{origin} en {time.time() - start:.1f} s · langue : {mix_label(lang_mix) or lang}")
                title = ((doc.metadata or {}).get("title") or "").strip()
                if not title:
                    title = elements[0].text if elements and elements[0].type == "heading" else Path(path).stem.replace("_", " ")
                d = PDFDocument(len(self.documents), Path(path).name, title, data, doc.page_count,
                                diagnose(doc), elements, info)
                d.path = str(path)
                d.lang, d.lang_mix = lang, lang_mix
                self.documents.append(d); added.append(d); known.add(d.filename)
            except Exception as error:
                errors.append((Path(path).name, str(error)[:120]))
        self.ready = False
        return added, errors

    # --- indexation ----------------------------------------------------------------------------
    def index(self):
        self.passages = []
        for d in self.documents:
            languages_of_document = document_languages(d)
            for p in make_passages(d, len(self.passages)):
                # les niveaux 3 à 5 héritent de la langue de leurs paragraphes ; les niveaux 1 et 2 sont détectés ici
                p.lang = p.lang or passage_language(p.text, getattr(d, "lang", DEFAULT_LANG))
                p.langs = passage_languages(p, languages_of_document)
                self.passages.append(p)
        self.language_counts = Counter(passage_lang(p) for p in self.passages)
        if not self.passages:
            raise ValueError("Aucun passage : vérifiez que les PDF contiennent du texte (ou activez l'OCR).")
        start = time.time()
        self.units = build_search_units(self.passages)         # passages + petites fenêtres
        self.by_id = {p.id: p for p in self.passages}
        for u in self.units:                                   # une fenêtre de recherche parle les langues de son passage
            parent = self.by_id.get(u.parent_id, u)
            u.lang, u.langs = passage_lang(parent), passage_langs(parent)
        _CONCEPT_CACHE.clear()
        _TERMS_CACHE.clear()
        self.last_twins = {}
        ACRONYMS.clear()
        ACRONYMS.update(build_acronyms(self.passages))
        build_vocabulary(self.passages)                        # pour la tolérance orthographique
        self.bm25 = BM25Index(self.units)
        self.semantic = SemanticIndex(self.units)
        self.reranker = Reranker(self.bm25, languages=self.languages)
        print(f"  {len(self.passages)} passages / {len(self.units)} unités de recherche indexés en "
              f"{time.time() - start:.1f} s ({self.semantic.name}) · {len(ACRONYMS)} sigles et "
              f"{sum(map(len, VOCABULARY_FORMS.values()))} mots appris · langues "
              f"{', '.join(f'{l} ({n})' for l, n in self.language_counts.most_common())} · "
              f"calibrage du seuil…")
        start = time.time()
        self.calibrate()
        print(f"  seuil {self.threshold:.2f} calibré en {time.time() - start:.1f} s")
        self.ready = True

    def calibrate(self):
        OFF_TOPIC = {
            "en": ["Which team won the last football World Cup?",
                   "What is the traditional recipe for couscous?",
                   "What will the weather be like tomorrow in the capital?",
                   "What is the price of maize per kilogram at the market?",
                   "How many measles vaccines were administered?",
                   "What is the national cocoa production?",
                   "Who wrote this year's best-selling novel?",
                   "How many hospitals are there in the North region?"],
            "fr": ["Quelle équipe a gagné la dernière coupe du monde de football ?",
                   "Quelle est la recette traditionnelle du couscous ?",
                   "Quel temps fera-t-il demain dans la capitale ?",
                   "Quel est le prix du maïs au kilogramme sur les marchés ?",
                   "Combien de vaccins ont été administrés contre la rougeole ?",
                   "Quelle est la production nationale de cacao ?",
                   "Qui est l'auteur du roman le plus vendu cette année ?",
                   "Combien d'hôpitaux compte la région du Nord ?"],
            "es": ["¿Qué equipo ganó el último Mundial de fútbol?",
                   "¿Cuál es la receta tradicional del cuscús?",
                   "¿Qué tiempo hará mañana en la capital?",
                   "¿Cuál es el precio del maíz por kilogramo en el mercado?",
                   "¿Cuántas vacunas contra el sarampión se administraron?",
                   "¿Cuál es la producción nacional de cacao?",
                   "¿Quién escribió la novela más vendida del año?",
                   "¿Cuántos hospitales hay en la región Norte?"],
            "pt": ["Que equipa ganhou o último Mundial de futebol?",
                   "Qual é a receita tradicional do cuscuz?",
                   "Como estará o tempo amanhã na capital?",
                   "Qual é o preço do milho por quilograma no mercado?",
                   "Quantas vacinas contra o sarampo foram administradas?",
                   "Qual é a produção nacional de cacau?",
                   "Quem escreveu o romance mais vendido do ano?",
                   "Quantos hospitais existem na região Norte?"],
        }
        # questions hors sujet dans chaque langue de la collection : le seuil doit tenir dans toutes
        off_topic = [q for l in ([l for l in self.languages if l in OFF_TOPIC] or ["fr"]) for q in OFF_TOPIC[l]]
        import itertools
        per_document = [[(re.sub(r"^\s*[\dIVX]+[.)]?\s*", "", e.text), getattr(e, "lang", "") or getattr(d, "lang", DEFAULT_LANG))
                         for e in d.elements if e.type == "heading" and e.level > 1] for d in self.documents]
        headings = [x for group in itertools.zip_longest(*per_document) for x in group if x]   # chaque document
        on_topic = [about_question(h, l) for h, l in dict.fromkeys(headings)][:15]
        rows = [{"Question": q, "Answer exists": True, "Confidence": round(self.confidence(q), 3)} for q in on_topic]
        rows += [{"Question": q, "Answer exists": False, "Confidence": round(self.confidence(q), 3)} for q in off_topic]
        self.calibration = pd.DataFrame(rows)
        ceiling = self.calibration.loc[~self.calibration["Answer exists"], "Confidence"].max()
        # dans un long rapport, presque chaque mot apparaît quelque part, donc les scores hors sujet montent :
        # le seuil est plafonné pour ne pas refuser des questions légitimes.
        self.calibrated_threshold = round(min(ceiling + 0.05, 0.7), 2)
        self.threshold = MANUAL_THRESHOLD if MANUAL_THRESHOLD is not None else self.calibrated_threshold

    # --- recherche et réponses -------------------------------------------------------------------
    @property
    def languages(self):
        """Langues de la collection, la plus fréquente en premier."""
        return [l for l, _ in self.language_counts.most_common()] or [DEFAULT_LANG]

    def queries_for(self, question, lang):
        """La question dans chaque langue de la collection : l'originale, plus une traduction par le
        modèle pour chaque autre langue (quand un modèle est connecté et que TRANSLATE_QUERIES = True)."""
        queries = {lang: question}
        if TRANSLATE_QUERIES and LLM.available:
            for other in self.languages:
                if other != lang:
                    translation = translate_query(question, other)
                    if translation:
                        queries[other] = translation
        return queries

    def confidence(self, question):
        r = self.search(question, 1)
        return r[0][1] if r else 0.0

    def search(self, question, k=None, docs=None, kind=None, lang=None, queries=None):
        k = k or TOP_K
        lang = lang or question_language(question, self.languages)
        queries = dict(queries or {lang: question})
        queries.setdefault(lang, question)
        others = [l for l in self.languages if l != lang]
        allowed = set(range(len(self.documents))) if docs is None else set(docs)
        keep = lambda ranking: [(p, s) for p, s in ranking if p.doc_id in allowed]
        # BM25 : la question originale (plus ses équivalents du lexique) sur toutes les unités, et chaque
        # traduction sur les unités de sa propre langue
        lexical = [keep(self.bm25.search(question, 40, lang, others))]
        for l, q in queries.items():
            if l != lang and q:
                lexical.append(keep([(u, s) for u, s in self.bm25.search(q, 40, l) if l in passage_langs(u)]))
        density = {}
        for ranking in lexical:
            for u, s in ranking:
                density[u.id] = max(density.get(u.id, 0.0), s)
        semantic = [keep(self.semantic.search(q, 40, l)) for l, q in queries.items() if q]
        candidates = rrf_fusion(lexical + semantic, top=14 + 4 * (len(queries) - 1))
        ranked = self.reranker.rerank(question, candidates, density, kind or question_type(question), lang, queries)
        seen, mapped = set(), []
        for unit, score in ranked:                              # une fenêtre représente son passage de contexte
            parent = self.by_id.get(unit.parent_id, unit)
            if parent.id in seen:
                continue
            seen.add(parent.id)
            mapped.append((parent, score))
        preferred = lang if ANSWER_LANGUAGE in ("question", "auto", None) else ANSWER_LANGUAGE
        mapped, self.last_twins = self.merge_translations(mapped, preferred)
        return diversify(mapped, k, len(allowed) > 1)

    # --- le même contenu dans deux langues ------------------------------------------------------
    def are_translations(self, a, b):
        """Deux passages en langues différentes qui disent la même chose : les deux moitiés d'un
        rapport bilingue, ou les éditions anglaise et française d'une même stratégie. Mêmes chiffres
        (38.2 et 38,2) ou — avec des embeddings multilingues — presque le même sens."""
        if passage_lang(a) == passage_lang(b) or a.kind != b.kind:
            return False
        fa, fb = figures_of(a), figures_of(b)
        shared = len(fa & fb) / max(1, len(fa | fb))
        if len(fa) >= 3 and len(fb) >= 3 and shared >= 0.8:
            return True
        if getattr(self.semantic, "cross_lingual", False):
            ia, ib = self.semantic.rows.get(a.id), self.semantic.rows.get(b.id)
            if ia is not None and ib is not None:
                similarity = float(self.semantic.matrix[ia] @ self.semantic.matrix[ib])
                return similarity >= TWIN_SIMILARITY and (not (fa or fb) or shared >= 0.5)
        return False

    def merge_translations(self, ranked, lang):
        """Garde un seul passage par contenu : la version dans la langue du lecteur quand elle existe, au
        meilleur rang des deux. Libère les places du TOP_K pour d'autres informations plutôt que des traductions."""
        if not MERGE_TRANSLATIONS:
            return ranked, {}
        kept, twins = [], {}
        for p, s in ranked:
            for i, (q, sq) in enumerate(kept):
                if self.are_translations(p, q):
                    if passage_lang(p) == lang and passage_lang(q) != lang:
                        kept[i] = (p, max(s, sq))
                        twins[p.id] = twins.pop(q.id, []) + [q]
                    else:
                        twins.setdefault(q.id, []).append(p)
                    break
            else:
                kept.append((p, s))
        return kept, twins

    def answer(self, question, history=None, docs=None, k=None, threshold=None, guardrails=True, memory=True,
               lang=None):
        if not self.ready:
            self.index()
        start = time.time()
        _CONCEPT_CACHE.clear()                      # le glossaire ou le lexique ont pu être modifiés
        threshold = self.threshold if threshold is None else threshold
        previous = history[-1].get("language") if history else None
        standalone, rewritten, method = (rewrite_follow_up(question, history or [], MODEL_FOLLOW_UP)
                                         if memory else (question, False, None))
        qlang = lang or question_language(question, self.languages, previous)
        answer_lang = qlang if ANSWER_LANGUAGE in ("question", "auto", None) else ANSWER_LANGUAGE
        standalone, corrections = fix_spelling(standalone, qlang)     # « chomage » -> « chômage »
        kind = question_type(standalone)
        queries = self.queries_for(standalone, qlang)                 # la question dans chaque langue
        results = self.search(standalone, k, docs, kind, qlang, queries)
        twins = self.last_twins
        conf = results[0][1] if results else 0.0
        relaxed = None
        if conf < threshold:
            # deuxième tentative : ne garder que les mots que les passages candidats contiennent vraiment
            reduced = relax_query(standalone, results, set(self.bm25.vocabulary), qlang, self.languages)
            if reduced and reduced.lower() != standalone.lower():
                fallback = self.search(reduced, k, docs, kind, qlang)
                if fallback and fallback[0][1] > conf:
                    results, conf, relaxed, twins = fallback, fallback[0][1], reduced, self.last_twins
        alternative = None
        if docs is not None and set(docs) != set(range(len(self.documents))):
            best = self.search(standalone, 1, None, kind, qlang, queries)
            if best and best[0][0].doc_id not in set(docs) and best[0][1] >= max(conf + 0.15, threshold):
                alternative = best[0][0].doc
        missing = []
        if results:
            view_q, view_l = query_view(results[0][0], standalone, qlang, queries)
            missing = missing_terms(view_q, results[0][0], view_l)
        r = {"question": question, "standalone_question": standalone, "rewritten": rewritten,
             "rewrite_method": method, "confidence": conf, "passages": results, "reason": None, "relaxed": relaxed,
             "corrections": corrections, "chunking": CHUNK_STRATEGY, "reranking": self.reranker.mode,
             "language": qlang, "answer_language": answer_lang,
             "queries": {l: q for l, q in queries.items() if l != qlang},
             "source_languages": sorted({passage_lang(p) for p, _ in results}),
             "translations": {p.id: [f"{passage_lang(t)} · {t.doc} · {t.pages_label}" for t in twins.get(p.id, [])]
                              for p, _ in results if twins.get(p.id)},
             "guardrails": guardrails, "time": time.strftime("%H:%M:%S"), "alternative": alternative,
             "missing_terms": missing, "check": None, "citations": [p.citation for p, _ in results], "kind": kind}
        if guardrails:
            years = set(re.findall(r"\b(?:19|20)\d{2}\b", standalone))
            content = " ".join(p.index_text for p, _ in results)
            absent_years = sorted(y for y in years if y not in content)
            if conf < threshold:
                r["reason"] = f"confiance {conf:.2f} inférieure au seuil {threshold:.2f}"
            elif absent_years:
                r["reason"] = f"année(s) {', '.join(absent_years)} absente(s) des passages"
            if r["reason"]:
                return {**r, "answer": refusal_in(answer_lang), "refused": True, "mode": "guardrail",
                        "duration_s": round(time.time() - start, 2)}
        text, mode = None, "extractive"
        if LLM.available:
            try:
                sources = "\n\n".join(
                    f"[{i}] ({p.citation} · langue : {' + '.join(nom_langue(l) for l in passage_langs(p))}"
                    f"{', tableau' if p.kind == 'table' else ''})\n{self.context_of(p)}"
                    for i, (p, _) in enumerate(results, 1))
                text = LLM.generate(strict_prompt(answer_lang), f"SOURCES\n\n{sources}\n\nQUESTION\n{standalone}")
                mode = f"{LLM.name} · {LLM.model}"
            except Exception as error:
                r["reason"] = f"modèle indisponible ({str(error)[:80]}) : mode extractif"
        if text is None:
            text = self._extract(standalone, results, kind, qlang, queries, answer_lang)
        refused = is_refusal(text)
        if not refused:
            r["check"] = check_figures(text, results)
        return {**r, "answer": text, "refused": refused, "mode": mode,
                "duration_s": round(time.time() - start, 2)}

    DEFINITION_QUESTION = re.compile(r"\b(stands? for|stand|mean|means|abbreviation|acronym|signifie|veut dire|"
                                     r"significa|siglas?|quer dizer|bedeutet)\b", re.I)

    def context_of(self, passage):
        """Le passage et ses voisins de la même section : un contexte complet pour le modèle."""
        if not NEIGHBOUR_CONTEXT:
            return passage.text
        family = [p for p in self.passages if p.doc_id == passage.doc_id and p.section == passage.section
                  and passage_lang(p) == passage_lang(passage)]    # une traduction voisine n'apporte rien
        try:
            position = family.index(passage)
        except ValueError:
            return passage.text
        pieces, size, before, after = [passage.text], len(passage.text), position - 1, position + 1
        while size < CONTEXT_CHARS and (before >= 0 or after < len(family)):
            if before >= 0 and size + len(family[before].text) <= CONTEXT_CHARS:
                pieces.insert(0, family[before].text); size += len(family[before].text); before -= 1
            elif after < len(family) and size + len(family[after].text) <= CONTEXT_CHARS:
                pieces.append(family[after].text); size += len(family[after].text); after += 1
            else:
                break
        return " ".join(dict.fromkeys(pieces))

    def _extract(self, question, results, kind="other", lang=None, queries=None, answer_lang=None):
        lang = lang or DEFAULT_LANG
        queries = queries or {lang: question}
        answer_lang = answer_lang or lang
        # le mode extractif ne traduit pas : un passage dans une autre langue est cité tel quel, avec une étiquette
        label = lambda p: ("" if passage_lang(p) == answer_lang
                           else f"(source en {nom_langue(passage_lang(p))}) ")
        # « Que signifie DIL ? » — réponse tirée des sigles appris dans les documents
        if self.DEFINITION_QUESTION.search(question):
            for acronym in sorted(ACRONYMS, key=len, reverse=True):
                if re.search(rf"\b{re.escape(acronym)}\b", question):
                    expansion = ACRONYMS[acronym]
                    for j, (p, _) in enumerate(results):
                        sentence = next((s for s in _split_sentences(p.text)
                                         if expansion.lower() in s.lower() or f"({acronym})" in s), None)
                        if sentence:
                            return f"{acronym} — {expansion}. {sentence} [{j + 1}]"
                    return f"{acronym} — {expansion}."
        candidates = []
        for j, (p, coverage) in enumerate(results):
            q, l = query_view(p, question, lang, queries)       # la question dans la langue du passage
            items = concepts(q, l, passage_langs(p))
            sentence, score = _best_sentence(items, p, self.reranker.weight, kind=kind, prefer_lang=answer_lang)   # citer le passage lui-même ;
            # le contexte élargi (context_of) n'est envoyé qu'au modèle, qui a besoin d'exhaustivité
            if sentence:
                candidates.append((score * (1 - 0.02 * j) * (0.6 + 0.4 * min(1.0, coverage)), j, p, sentence))
        if not candidates:
            return refusal_in(answer_lang)
        candidates.sort(key=lambda c: -c[0])
        score0, j0, p0, sentence0 = candidates[0]
        text = f"{label(p0)}{sentence0} [{j0 + 1}]"
        if COMPARISON.search(strip_accents(question.lower())):
            others = [c for c in candidates[1:] if c[2].doc_id != p0.doc_id and c[0] >= 0.5 * score0]
            if others:
                _, j1, p1, sentence1 = others[0]
                text += f" — {p1.doc}: {label(p1)}{sentence1} [{j1 + 1}]"
        return text

    # --- affichage ------------------------------------------------------------------------------
    def thumbnail(self, passage, **kw):
        doc = self.documents[passage.doc_id]
        first = passage.pages[0]
        return thumbnail_html(doc.data, first, [b for pg, b in passage.boxes if pg == first], **kw)

    def card_html(self, r, thumbnails=True):
        badges = pill("REFUS", C["ochre"]) if r["refused"] else pill("RÉPONSE", C["dark_green"])
        badges += pill(f"confiance {r['confidence']:.2f}", C["teal"]) + pill(MODE_LABELS.get(r["mode"], r["mode"]), C["grey"])
        if r.get("language"):
            sources_lang = [l for l in r.get("source_languages", []) if l != r["language"]]
            badges += pill(f"🌐 {r['language']}" + (f" → sources en {', '.join(map(nom_langue, sources_lang))}" if sources_lang else ""),
                           C["teal"])
        if not r["guardrails"]:
            badges += pill("garde-fous désactivés", C["brick"])
        chk = r.get("check")
        if chk and chk["numbers"]:
            badges += (pill(f"chiffres vérifiés {chk['numbers']}/{chk['numbers']}", C["dark_green"]) if not chk["missing"]
                       else pill(f"⚠ chiffre(s) absent(s) des sources : {', '.join(chk['missing'])}", C["brick"]))
        if r.get("relaxed"):
            badges += pill(f"recherche assouplie : {r['relaxed']}", C["ochre"])
        if r.get("corrections"):
            badges += pill("orthographe : " + ", ".join(f"{a} → {b}" for a, b in r["corrections"]), C["teal"])
        notes = ""
        if r.get("queries"):
            notes += (f'<div style="font-size:12.5px;color:{C["teal"]};margin-top:6px;">🌐 Recherché aussi sous la forme : '
                      + " · ".join(f'<b>{l}</b> <i>{_html.escape(q)}</i>' for l, q in r["queries"].items())
                      + '</div>')
        if r.get("translations"):
            notes += (f'<div style="font-size:12.5px;color:{C["teal"]};margin-top:6px;">🔁 Même contenu trouvé aussi dans '
                      f'une autre langue, fusionné : ' + " · ".join(_html.escape(t) for ts in r["translations"].values() for t in ts)
                      + '</div>')
        if r["rewritten"]:
            notes += (f'<div style="font-size:12.5px;color:{C["teal"]};margin-top:6px;">🧠 Compris comme : '
                      f'<i>{_html.escape(r["standalone_question"])}</i> '
                      f'<span style="color:{C["slate"]};">({ {"model": "modèle", "rules": "règles"}.get(r["rewrite_method"], r["rewrite_method"]) })</span></div>')
        if not r["refused"] and r.get("missing_terms"):
            notes += (f'<div style="font-size:12.5px;color:{C["ochre"]};margin-top:6px;">⚠️ Introuvable dans la source '
                      f'principale : <b>{_html.escape(", ".join(r["missing_terms"]))}</b> — vérifiez que la réponse '
                      f'traite vraiment votre question.</div>')
        if r.get("alternative"):
            notes += (f'<div style="font-size:12.5px;color:{C["teal"]};margin-top:6px;">💡 Une réponse plus pertinente se trouve '
                      f'probablement dans « <b>{_html.escape(r["alternative"])}</b> », exclu par le filtre.</div>')
        reason = f'<div class="rg-note">Motif : {_html.escape(r["reason"])}</div>' if r.get("reason") else ""
        answer = re.sub(r"\[(\d+)\]", lambda m: f'<sup style="background:{C["mint"]};color:{C["dark_green"]};'
                        f'border-radius:4px;padding:0 4px;font-weight:700;">{m.group(1)}</sup>', _html.escape(r["answer"]))
        sources = []
        for i, (p, s) in enumerate(r["passages"], 1):
            thumb = self.thumbnail(p, dpi=55, width=105) if thumbnails and i <= 2 and not r["refused"] else ""
            sources.append(
                f'<div style="display:flex;gap:10px;align-items:flex-start;margin:6px 0;">{thumb}<div style="flex:1;">'
                f'<b>[{i}]</b> {pill(p.doc, C["dark_green"])}<b>{_html.escape(p.section or "début")}</b> · '
                f'p. {", ".join(map(str, p.pages))} · {KIND_LABELS.get(p.kind, p.kind)} · {"+".join(passage_langs(p))} · confiance {s:.2f}<br>'
                f'<span style="color:{C["slate"]};">“{_html.escape(" ".join(p.text.split())[:220])}…”</span></div></div>')
        return (f'<div class="rg rg-card" style="border-left:6px solid {C["ochre"] if r["refused"] else C["green"]};margin:0;">'
                f'<div>{badges}<span style="float:right;color:{C["slate"]};font-size:11px;">{r["time"]} · {r["duration_s"]} s</span></div>'
                f'{notes}<div style="font-size:14.5px;margin:10px 0;line-height:1.55;">{answer}</div>{reason}'
                f'<details {"open" if not r["refused"] else ""}><summary style="cursor:pointer;color:{C["dark_green"]};'
                f'font-weight:600;font-size:13px;">Sources ({len(r["passages"])})</summary>'
                f'<div class="rg-src">{"".join(sources) or "Aucune source."}</div></details></div>')

    def show(self, r):
        display(HTML(f'{_CSS}<div class="rg" style="margin:10px 0;"><div style="font-weight:700;font-size:15px;'
                     f'margin-bottom:6px;">❓ {_html.escape(r["question"])}</div>{self.card_html(r)}</div>'))

    def summary(self):
        return pd.DataFrame([{"N°": d.id, "Document": d.title, "Fichier": d.filename, "Pages": d.pages,
                              "Langue": mix_label(d.lang_mix) if getattr(d, "lang_mix", None) else getattr(d, "lang", "—"),
                              "Pages scannées": ", ".join(map(str, d.scanned_pages)) or "—",
                              "Tableaux": sum(e.type == "table" for e in d.elements),
                              "Passages": sum(p.doc_id == d.id for p in self.passages) if self.passages else "—"}
                             for d in self.documents])

KIND_LABELS = {"text": "texte", "table": "tableau"}     # affichage seulement
MODE_LABELS = {"extractive": "extractif", "guardrail": "garde-fou"}

LIB = Library()
_DEMO_HISTORY = []

ABOUT_TEMPLATES = {"en": "What does the document say about: {}?", "fr": "Que dit le document sur : {} ?",
                   "es": "¿Qué dice el documento sobre: {}?", "pt": "O que diz o documento sobre: {}?",
                   "de": "Was sagt das Dokument über: {}?", "it": "Cosa dice il documento su: {}?"}

def about_question(heading, lang):
    """Une question sur une section, rédigée dans la langue de cette section."""
    return ABOUT_TEMPLATES.get(lang, ABOUT_TEMPLATES["fr"]).format(heading)

EXAMPLE_KEYWORDS = {"exemple", "example"}                          # « example » : alias anglais
MULTILINGUAL_KEYWORDS = {"exemple-multilingue", "example-multilingual"}

def using_examples():
    return bool(DOCUMENTS) and set(map(str, DOCUMENTS)) <= EXAMPLE_KEYWORDS | MULTILINGUAL_KEYWORDS

def using_multilingual_examples():
    return using_examples() and bool(MULTILINGUAL_KEYWORDS & set(map(str, DOCUMENTS)))

def suggested_questions(n=5):
    """Questions adaptées aux documents réellement chargés (utilisées par les démos et les interfaces)."""
    if using_multilingual_examples():
        return ["Quel est le taux de chômage des femmes dans la région Côte ?", "Et pour les hommes ?",
                "Quel est le taux de chômage des jeunes de 15 à 24 ans ?",
                "Quelle a été l'inflation des prix à la consommation au premier trimestre 2026 ?",
                "Quel est le taux d'accès des ménages à l'électricité ?",
                "Quelle proportion des ménages a accès à internet ?",
                "Quel est le prix du maïs ?"][:n]
    if using_examples():
        return ["Quel est le taux de chômage des femmes dans la région Côte ?", "Et pour les hommes ?",
                "Comment le taux de chômage a-t-il évolué entre le quatrième trimestre 2025 et le premier trimestre 2026 ?",
                "Quelle est la taille moyenne des ménages ?", "Quel est le prix du maïs ?"][:n]
    per_document = [[(re.sub(r"^\s*[\dIVX]+[.)]?\s*", "", e.text).strip(), getattr(e, "lang", "") or getattr(d, "lang", DEFAULT_LANG))
                     for e in d.elements
                     if e.type == "heading" and e.level > 1 and e.page > 2 and 12 < len(e.text) < 70
                     and len(e.text.split()) >= 2 and e.text.lower() not in d.title.lower()] for d in LIB.documents]
    import itertools
    headings = [x for group in itertools.zip_longest(*per_document) for x in group if x]
    questions = [about_question(h, l) for h, l in dict.fromkeys(headings)][:max(1, n - 2)]
    if ACRONYMS:
        questions.append(f"Que signifie {sorted(ACRONYMS, key=lambda a: -len(ACRONYMS[a]))[0]} ?")
    questions.append("Quel est le prix du maïs au kilogramme ?")       # hors du champ : doit être refusée
    return questions[:n]

def ask(question, docs=None, memory=True, lang=None):
    """Pose une question à la bibliothèque, dans n'importe quelle langue, en se souvenant des questions précédentes.
    `lang` force la langue de la question ("fr", "es"…) ; par défaut, elle est détectée."""
    r = LIB.answer(question, history=_DEMO_HISTORY if memory else None, docs=docs, lang=lang)
    _DEMO_HISTORY.append(r)
    LIB.show(r)
    return r
print("✅ T6 prêt · boîte à outils complète.")
````

````python
# @title ⚙️ T7 · La chaîne visible, le comparateur et le banc d'essai { display-mode: "form" }

# ── 1. La chaîne, dessinée ─────────────────────────────────────────────────────────────────
PIPELINE = [
    ("1", "Question", "saisie par l'utilisateur", "grey"),
    ("2", "Relance", "réécrite en question autonome", "teal"),
    ("3", "Langue et orthographe", "langue détectée, question traduite dans chaque langue de la collection (et de chaque rapport), fautes de frappe corrigées", "teal"),
    ("4", "Normalisation", "minuscules, accents, racines, mots interrogatifs retirés", "grey"),
    ("5", "Expansion", "glossaire + acronymes appris dans les documents", "grey"),
    ("6", "BM25", "mots exacts, sur chaque unité de recherche", "green"),
    ("7", "Sémantique", "sens, sur chaque unité de recherche", "green"),
    ("8", "Fusion", "les deux classements fusionnés (RRF)", "green"),
    ("9", "Reclassement", "ordre fixé par le niveau que vous avez choisi", "gold"),
    ("10", "Fenêtres → passages", "chaque unité renvoie au passage à lire ; un passage et sa traduction ne comptent qu'une fois", "gold"),
    ("11", "Garde-fous", "seuil, contrôle de l'année, refus", "ochre"),
    ("12", "Réponse", "rédigée par le modèle, ou extraite", "dark_green"),
    ("13", "Vérifications", "chiffres, termes manquants, filtre", "dark_green"),
]

def show_pipeline():
    """Toute la chaîne, avec les paramètres actuellement en vigueur."""
    chosen = {"3": " · ".join(f"{l} {n}" for l, n in LIB.language_counts.most_common()) if getattr(LIB, "language_counts", None) else "",
              "9": RERANKING_LEVELS[LIB.reranker.mode][0] if getattr(LIB, "reranker", None) else RERANKING,
              "6": f"{len(getattr(LIB, 'units', []))} unités",
              "7": getattr(getattr(LIB, "semantic", None), "name", "—"),
              "11": f"seuil {LIB.threshold:.2f}" if getattr(LIB, "threshold", None) else "",
              "12": f"{LLM.name} · {LLM.model}" if LLM.available else "mode extractif"}
    boxes = ""
    for number, title, subtitle, colour in PIPELINE:
        note = chosen.get(number, "")
        boxes += (
            f'<div style="flex:1 1 150px;min-width:150px;background:#fff;border:1px solid {C["sage"]};'
            f'border-top:3px solid {C[colour]};border-radius:10px;padding:8px 10px;">'
            f'<div style="font-size:10.5px;color:{C[colour]};font-weight:800;">ÉTAPE {number}</div>'
            f'<div style="font-size:13px;font-weight:700;color:{C["ink"]};">{title}</div>'
            f'<div style="font-size:11.5px;color:{C["slate"]};line-height:1.35;">{subtitle}</div>'
            + (f'<div style="font-size:11px;color:{C["dark_green"]};font-weight:700;margin-top:3px;">{_html.escape(str(note))}</div>' if note else "")
            + '</div>')
    display(HTML(
        f'{_CSS}<div class="rg"><div class="rg-title">🔗 La chaîne RAG, telle qu\'elle est configurée en ce moment</div>'
        f'<div style="display:flex;gap:8px;flex-wrap:wrap;">{boxes}</div>'
        f'<div class="rg-note">Découpage : <b>{CHUNKING_LEVELS[CHUNK_STRATEGY][0]}</b> · '
        f'Reclassement : <b>{RERANKING_LEVELS[RERANKING][0]}</b> · Passages transmis pour la réponse : <b>{TOP_K}</b></div></div>'))

# ── 2. La chaîne, suivie sur une vraie question ────────────────────────────────────────────
def explain(question, docs=None, k=None):
    """Fait passer une question par chaque étape et montre ce que chaque étape a changé."""
    k = k or TOP_K
    steps = []
    qlang = question_language(question, LIB.languages)
    others = [l for l in LIB.languages if l != qlang]
    standalone, corrections = fix_spelling(question, qlang)
    kind = question_type(standalone)
    queries = LIB.queries_for(standalone, qlang)
    translated = "".join(f"<br>→ <b>{l}</b> : {_html.escape(q)}" for l, q in queries.items() if l != qlang)
    steps.append(("3 · Langue", f"question en <b>{nom_langue(qlang)}</b> · collection : "
                  + ", ".join(f"{l} ({n})" for l, n in LIB.language_counts.most_common())
                  + (translated or ("<br>pas de traduction (pas de modèle) : le lexique relie les langues" if others else ""))))
    if corrections:
        steps.append(("3 · Orthographe", ", ".join(f"<b>{a}</b> → {b}" for a, b in corrections)))
    effect = {"count": "les passages et phrases portant une vraie quantité sont favorisés",
              "list": "les passages et phrases portant une énumération sont favorisés",
              "date": "les passages et phrases portant une année sont favorisés",
              "definition": "les acronymes appris dans les documents sont consultés en premier",
              "other": "aucun traitement particulier"}[kind]
    steps.append(("Type de question", f"<b>{kind}</b> — {effect}"))
    terms = concepts(standalone, qlang, others)
    steps.append(("4 · Termes retenus", ", ".join(f"{t} <span style='color:#5E6964'>({LIB.reranker.weight(t, g):.1f})</span>"
                                              for t, g in terms) or "—"))
    added = expand_query(standalone, qlang)[1]
    steps.append(("5 · Expansion", "<br>".join(f"<b>{a}</b> → {b}" for a, b in added) or "rien à ajouter"))
    lexical = LIB.bm25.search(standalone, 40, qlang, others)
    semantic = LIB.semantic.search(standalone, 40, qlang)
    line = lambda units: "<br>".join(f"{i}. <b>{u.section[:34] or '—'}</b> · {passage_lang(u)} · p. {u.pages[0]} "
                                     f"<span style='color:#5E6964'>({s:.2f})</span>"
                                     for i, (u, s) in enumerate(units[:4], 1)) or "rien trouvé"
    steps.append(("6 · BM25 (mots exacts)", line(lexical)))
    steps.append((f"7 · {LIB.semantic.name}", line(semantic)))
    fused = rrf_fusion([lexical, semantic], top=14)
    steps.append(("8 · Fusion", line(fused)))
    reranked = LIB.reranker.rerank(standalone, fused, {u.id: s for u, s in lexical}, kind, qlang, queries)
    steps.append((f"9 · Reclassement · {LIB.reranker.name}",
                  "<br>".join(f"{i}. <b>{u.section[:34] or '—'}</b> · p. {u.pages[0]} "
                              f"<span style='color:#5E6964'>(couverture {s:.2f}"
                              + (", type de réponse ✓" if LIB.reranker.answer_type_bonus(kind, u) else "") + ")</span>"
                              for i, (u, s) in enumerate(reranked[:4], 1))))
    r = LIB.answer(question, docs=docs, k=k)
    steps.append(("10 · Passages renvoyés", "<br>".join(f"{i}. {p.citation} · {'+'.join(passage_langs(p))}"
                                                        for i, (p, _) in enumerate(r["passages"], 1))))
    if r.get("translations"):
        steps.append(("10 · Traductions fusionnées", "<br>".join(_html.escape(t) for ts in r["translations"].values() for t in ts)))
    steps.append(("11 · Garde-fous", f"confiance <b>{r['confidence']:.2f}</b> contre seuil <b>{LIB.threshold:.2f}</b> → "
                                     + ("<b style='color:#D49A00'>refus</b>" if r["refused"] else "<b style='color:#00704A'>réponse</b>")
                                     + (f" · {r['reason']}" if r.get("reason") else "")))
    rows = "".join(f'<tr><td style="white-space:nowrap;vertical-align:top;padding:6px 12px 6px 0;font-weight:700;'
                   f'color:{C["dark_green"]};font-size:12.5px;">{name}</td>'
                   f'<td style="padding:6px 0;font-size:12.5px;line-height:1.5;">{value}</td></tr>'
                   for name, value in steps)
    display(HTML(f'{_CSS}<div class="rg rg-card"><div class="rg-title">🔍 {_html.escape(question)}</div>'
                 f'<table style="width:100%;border-collapse:collapse;">{rows}</table></div>'))
    LIB.show(r)
    return r

# ── 3. Les cinq niveaux de découpage, sur le même texte ────────────────────────────────────
def compare_chunking(document=None, levels=("fixed", "sentences", "sections", "hybrid", "smart")):
    """Même document, cinq niveaux : combien de morceaux, de quelle longueur, et à quoi ressemble le premier."""
    global CHUNK_STRATEGY
    document = document or LIB.documents[0]
    keep, rows = CHUNK_STRATEGY, []
    try:
        for level in levels:
            CHUNK_STRATEGY = level
            start = time.time()
            passages = make_passages(document)
            units = build_search_units(passages, level)
            lengths = [len(p.text) for p in passages] or [0]
            first = next((p for p in passages if len(p.text) > 200), passages[0] if passages else None)
            rows.append({"Niveau": CHUNKING_LEVELS[level][0],
                         "Passages": len(passages), "Unités de recherche": len(units),
                         "Longueur moyenne": int(np.mean(lengths)),
                         "Connaît sa section": "✅" if any(p.section for p in passages) else "—",
                         "Secondes": round(time.time() - start, 2),
                         "Premier passage": _html.escape((first.text[:150] if first else "")) + "…"})
    finally:
        CHUNK_STRATEGY = keep
    show_table(pd.DataFrame(rows), title=f"✂️ Les cinq niveaux sur « {_html.escape(document.short)} »",
               note="Les niveaux 1 et 2 ignorent les titres et les tableaux : un passage ne sait donc plus d'où il vient. "
                    "Les niveaux 4 et 5 indexent une seconde granularité, plus fine — c'est pourquoi ils comptent plus "
                    "d'unités de recherche que de passages.")

# ── 4. Le banc d'essai : quelle combinaison répond le mieux à VOS questions ────────────────
def probe_questions(n=12, seed=0):
    """Questions dont la réponse est connue, construites à partir des documents quand vous n'avez pas
    écrit de jeu de tests. Les mots d'UNE phrase doivent ramener une phrase ULTÉRIEURE du même sujet :
    un découpage qui coupe un sujet en deux, ou qui perd sa section, échoue à la sonde. Ce n'est qu'un
    indicateur approché — écrivez vos propres EVALUATION_QUESTIONS pour une mesure fiable."""
    import random
    rng = random.Random(seed)
    probes = []
    for minimum, gap in ((500, 3), (320, 2), (180, 2)):     # longs rapports d'abord, les petits fonctionnent quand même
        if len(probes) >= max(3, n // 2):
            break
        pool = [p for p in LIB.passages if p.kind == "text" and len(p.text) > minimum]
        if not pool:
            continue                                   # collection courte : essayer des passages plus petits
        probes = []
        for p in rng.sample(pool, len(pool)):
            sentences = [s for s in _split_sentences(p.text) if 40 < len(s) < 260]
            if len(sentences) < gap:
                continue
            question = " ".join(w for w in sentences[0].split() if len(w) > 3)[:170]
            target = sentences[-1].split()
            if len(target) < 6 or len(question.split()) < 5:
                continue
            probes.append((question, [" ".join(target[:6])]))
            if len(probes) == n:
                break
    return probes

def benchmark(chunking=("fixed", "sentences", "sections", "hybrid", "smart"),
              reranking=("smart",), questions=None, k=None, show_questions=False):
    """Réindexe avec chaque combinaison et la mesure sur des questions dont la réponse est connue.
    Réponse trouvée = le texte attendu figure parmi les passages transmis au modèle."""
    global CHUNK_STRATEGY, RERANKING
    k = k or TOP_K
    questions = questions or globals().get("EVALUATION_QUESTIONS") or probe_questions()
    if not questions:
        callout("Aucune question d'évaluation : chargez d'abord des documents.", "warning")
        return
    if show_questions:
        show_table(pd.DataFrame([{"Question": q, "Attendu dans le contexte": " | ".join(e)} for q, e in questions]),
                   title=f"Les {len(questions)} questions utilisées par le banc d'essai")
    keep_chunk, keep_rank, keep_context, rows = CHUNK_STRATEGY, RERANKING, NEIGHBOUR_CONTEXT, []
    try:
        for level in chunking:
            for mode in reranking:
                CHUNK_STRATEGY, RERANKING = level, mode
                globals()["NEIGHBOUR_CONTEXT"] = keep_context and level == "smart"
                start = time.time()
                LIB.reranker = None
                LIB.index()
                build = time.time() - start
                found = first = 0
                start = time.time()
                for question, expected in questions:
                    results = LIB.search(question, k)
                    context = " ".join(LIB.context_of(p) for p, _ in results).lower()
                    found += any(e.lower() in context for e in expected)
                    if results:
                        first += any(e.lower() in LIB.context_of(results[0][0]).lower() for e in expected)
                rows.append({"Découpage": CHUNKING_LEVELS[level][0], "Reclassement": RERANKING_LEVELS[mode][0],
                             "Passages": len(LIB.passages), "Unités": len(LIB.units),
                             "Réponse dans le contexte": f"{found}/{len(questions)}",
                             "Bon passage en premier": f"{first}/{len(questions)}",
                             "Indexation (s)": round(build, 1),
                             "Par question (ms)": int(1000 * (time.time() - start) / len(questions)),
                             "_score": (found, first)})
    finally:
        CHUNK_STRATEGY, RERANKING = keep_chunk, keep_rank
        globals()["NEIGHBOUR_CONTEXT"] = keep_context
        LIB.reranker = None
        LIB.index()
    table = pd.DataFrame(rows)
    best = table["_score"].max()
    table["Meilleur"] = ["🏆" if s == best else "" for s in table["_score"]]
    show_table(table.drop(columns="_score"), title="🏁 Banc d'essai sur vos propres documents",
               note="Mesuré sur cette collection uniquement : relancez-le sur la vôtre avant de vous fier à un classement. "
                    "Une combinaison qui l'emporte d'une seule question est à égalité, pas gagnante.")
    return table

print("✅ T7 prêt · show_pipeline(), explain(\"…\"), compare_chunking(), benchmark().")
````

````python
if ENV["name"] == "colab" and UPLOAD_PDF:
    from google.colab import files
    print("Sélectionnez un ou plusieurs PDF…")
    uploaded = LIB.folder / "uploaded"
    uploaded.mkdir(exist_ok=True)
    for name, content in files.upload().items():
        if name.lower().endswith(".pdf"):
            (uploaded / name).write_bytes(content)
    DOCUMENTS = list(DOCUMENTS) + [str(uploaded)]

if ENV["name"] == "kaggle" and Path("/kaggle/input").exists():
    found = sorted(Path("/kaggle/input").rglob("*.pdf"))
    if found:
        show_table(pd.DataFrame([{"PDF disponible": p.name, "À copier dans DOCUMENTS": f'"{p}"'} for p in found[:25]]),
                   title="PDF joints à ce notebook Kaggle")

start = time.time()
added, errors = LIB.load(DOCUMENTS)
cards = "".join(
    f'<div style="border:1px solid {C["sage"]};border-radius:12px;padding:10px;background:#fff;width:190px;">'
    f'{thumbnail_html(d.data, 1, dpi=40, width=168, caption="")}'
    f'<div style="font-weight:700;font-size:13px;margin-top:4px;">{_html.escape(d.short)}</div>'
    f'<div style="font-size:11.5px;color:{C["slate"]};">{d.pages} pages · '
    f'{sum(e.type == "table" for e in d.elements)} tableau(x)'
    + (f' · <span style="color:{C["brick"]};">{len(d.scanned_pages)} page(s) numérisée(s)</span>' if d.scanned_pages else "")
    + '</div></div>' for d in LIB.documents)
display(HTML(f'{_CSS}<div class="rg"><div class="rg-title">📚 {len(LIB.documents)} document(s) chargé(s) '
             f'en {time.time() - start:.1f} s</div><div style="display:flex;gap:10px;flex-wrap:wrap;">{cards}</div></div>'))
for name, why in errors:
    callout(f"<b>{_html.escape(name)}</b> ignoré : {_html.escape(why)}", "error")
if errors and not LIB.documents:
    callout("Aucun document n'a pu être chargé. Si un site web a refusé le téléchargement (HTTP 403), récupérez le PDF "
            "dans votre navigateur, puis mettez <code>UPLOAD_PDF = True</code> (Colab) ou indiquez le chemin du fichier "
            "dans <code>DOCUMENTS</code>.",
            "warning")
if any(d.scanned_pages for d in LIB.documents) and not USE_OCR:
    callout("Certaines pages sont <b>numérisées</b> et restent invisibles pour l'assistant. "
            "Mettez <code>USE_OCR = True</code> pour les lire.", "warning")
````

<div style="border-left:6px solid #00A86A;background:#F4F7F5;padding:14px 22px;border-radius:0 12px 12px 0;margin:8px 0 6px 0;font-family:'Segoe UI',Calibri,system-ui,sans-serif;">

````python
start = time.time()
LIB.index()
model_label = f"{LLM.name} · {LLM.model}" if LLM.available else "mode extractif (sans modèle)"
display(HTML(
    f'{_CSS}<div class="rg" style="display:flex;gap:10px;flex-wrap:wrap;">'
    + "".join(f'<div style="flex:1;min-width:150px;background:#fff;border:1px solid {C["sage"]};border-radius:12px;padding:12px 16px;">'
              f'<div style="font-size:12px;color:{C["slate"]};font-weight:600;">{t}</div>'
              f'<div style="font-size:26px;font-weight:800;color:{C["dark_green"]};line-height:1.25;">{v}</div>'
              f'<div style="font-size:11.5px;color:{C["slate"]};">{s}</div></div>'
              for t, v, s in [("Documents", len(LIB.documents), f"{sum(d.pages for d in LIB.documents)} pages au total"),
                              ("Passages", len(LIB.passages), f"{len(LIB.units)} unités de recherche · {time.time() - start:.1f} s"),
                              ("Seuil de confiance", f"{LIB.threshold:.2f}", "calibré automatiquement"),
                              ("Découpage", CHUNKING_LEVELS[CHUNK_STRATEGY][0].split(" · ")[0],
                               CHUNKING_LEVELS[CHUNK_STRATEGY][0].split(" · ")[1]),
                              ("Reclassement", LIB.reranker.name.split(" · ")[0], LIB.reranker.name.split(" · ")[1]),
                              ("Rédaction", "modèle" if LLM.available else "extraits", model_label)])
    + '</div>'))
show_pipeline()
if LIB.semantic.mode != "embeddings":
    callout(f"Recherche sémantique : <b>{LIB.semantic.name}</b> — elle rapproche des graphies, pas du sens. "
            "Un modèle neuronal fait le lien entre les mots de vos utilisateurs et ceux du rapport : mettez "
            "<code>INSTALL_EMBEDDINGS = True</code> (un seul téléchargement d'environ 470 Mo, ensuite mis en cache).", "warning")
````

> 💡 **Ce que signifient ces cartes.** **Documents** et **Passages** : ce qui a été lu et comment il a été découpé. **Seuil de confiance** : en dessous de ce score, l'assistant refuse de répondre (B5). **Découpage** et **Reclassement** : les deux leviers qui déterminent la qualité des réponses — la partie B les compare et vous permet de les modifier. **Rédaction** : *modèle* si un modèle de langage rédige les réponses, *extraits* en mode extractif.

> Le schéma ci-dessous représente **toute la chaîne**, telle qu'elle est configurée en ce moment. La partie B4 y suit le parcours d'une vraie question.

<div style="border-left:6px solid #00A86A;background:#F4F7F5;padding:14px 22px;border-radius:0 12px 12px 0;margin:8px 0 6px 0;font-family:'Segoe UI',Calibri,system-ui,sans-serif;">

Une page **numérisée** (scannée) ne contient qu'une image : sans OCR, l'assistant l'ignore **sans le dire**. Il répondra « information non disponible » alors que l'information figure dans le document. Ce diagnostic prend une seconde et évite ce piège.

````python
colours = {"text": C["green"], "scanned": C["brick"], "empty": C["grey"]}
labels = {"text": "texte", "scanned": "numérisée", "empty": "vide"}
n = len(LIB.documents)
fig, axes = plt.subplots(1, n, figsize=(max(7.5, 3.3 * n), 3.0), squeeze=False)
for ax, d in zip(axes[0], LIB.documents):
    diag = d.diagnosis
    ax.bar(diag["Page"], diag["Characters"].clip(lower=15), color=[colours[s] for s in diag["Status"]])
    for p in d.scanned_pages:
        ax.text(p, diag["Characters"].max() * 0.08, "numérisée", ha="center", rotation=90, color=C["brick"],
                fontweight="bold", fontsize=8.5)
    ax.set_title(d.short[:34], fontsize=10.5)
    pages = diag["Page"]
    ax.set_xticks(pages if len(pages) <= 20 else np.linspace(1, len(pages), 8, dtype=int))
    ax.set_xlabel("Page")
axes[0][0].set_ylabel("Caractères extraits")
from matplotlib.patches import Patch
fig.legend(handles=[Patch(color=v, label=labels[k]) for k, v in colours.items()], loc="lower center", ncol=3,
           bbox_to_anchor=(0.5, -0.08))
fig.suptitle("Texte extractible, page par page", fontweight="bold", color=C["ink"])
plt.tight_layout()
fig.savefig(OUTPUT_DIR / "page_diagnosis.png", dpi=160, bbox_inches="tight")
plt.show()
show_table(LIB.summary(), title="📚 Documents de la bibliothèque")
````

<div style="color:#0E7C86;font-size:11.5px;font-weight:700;letter-spacing:2.5px;text-transform:uppercase;">Parcours approfondi</div>
<div style="font-size:23px;font-weight:800;color:#231F20;margin:3px 0 4px 0;">B2 · Extraire la structure</div>
<div style="color:#5E6964;font-size:14px;line-height:1.45;">Titres, en-têtes répétés, mots coupés, tableaux et OCR.</div></div>

| Opération | Principe | Pourquoi |
|---|---|---|
| **Détecter les titres** | Une ligne écrite nettement plus grand que le corps du texte | Chaque passage connaît sa **section** |
| **Supprimer les en-têtes et pieds de page** | Une ligne répétée en haut ou en bas d'au moins la moitié des pages | « Page 3 » pollue chaque passage |
| **Recoller les mots coupés** | « statis-⏎tiques » → « statistiques » | La recherche trouve le mot entier |
| **Linéariser les tableaux** | Chaque ligne devient « Libellé — colonne : valeur ; … » | Un nombre isolé n'a pas de sens |
| **OCR** (facultatif) | Reconnaissance du texte sur les pages numérisées | Rend visibles les pages numérisées |

La normalisation **NFKC** remplace aussi les ligatures typographiques (« ﬁ », « ﬀ ») par des lettres ordinaires : sans elle, « ﬁchier » ne correspondrait jamais à « fichier ».

La page ci-dessous montre ce que l'extracteur a reconnu : en **vert** les titres, en **gris** les paragraphes, en **ocre** les tableaux, en **rouge barré** les en-têtes et pieds de page supprimés.

````python
def annotated_page(d, number, dpi=75):
    doc = pymupdf.open("pdf", d.data)
    page = doc[number - 1]
    styles = {"heading": (0.0, 0.66, 0.42), "paragraph": (0.37, 0.41, 0.39),
              "table": (0.83, 0.60, 0.0), "ocr": (0.05, 0.49, 0.53)}
    for e in d.elements:
        if e.page == number:
            page.draw_rect(pymupdf.Rect(e.box), color=styles[e.type], width=1.4)
    height = page.rect.height
    for l in page_lines(page):
        if (l["box"].y1 < height * 0.1 or l["box"].y0 > height * 0.9) and not any(
                e.page == number and pymupdf.Rect(e.box).intersects(l["box"]) for e in d.elements):
            page.draw_rect(l["box"], color=(0.72, 0.23, 0.18), width=1.2)
            page.draw_line(l["box"].bl, l["box"].tr, color=(0.72, 0.23, 0.18), width=0.8)
    return base64.b64encode(page.get_pixmap(dpi=dpi).tobytes("png")).decode()

d_demo = next((d for d in LIB.documents if any(e.type == "table" for e in d.elements)), LIB.documents[0])
tab = next((e for e in d_demo.elements if e.type == "table"), None)
demo_page = tab.page if tab else 1
block = ""
if tab:
    df_t = pd.DataFrame(tab.table["rows"], columns=tab.table["headers"][:len(tab.table["rows"][0])])
    block = (f'<div class="rg-title" style="margin-top:10px;">Le tableau tel qu\'il est lu…</div>'
             f'{df_t.to_html(index=False, classes="rg-t", border=0)}'
             f'<div class="rg-title" style="margin-top:10px;">… et tel qu\'il est indexé</div>'
             f'<pre style="background:{C["mist"]};padding:10px;border-radius:8px;white-space:pre-wrap;font-size:12px;'
             f'color:{C["ink"]};">{_html.escape(tab.text)}</pre>')
headings = [e for e in d_demo.elements if e.type == "heading"]
outline = "".join(f'<div style="margin-left:{(e.level - 1) * 16}px;">{"📘" if e.level == 1 else "▸"} {_html.escape(e.text[:70])} '
                  f'<span style="color:{C["slate"]};font-size:11px;">p. {e.page}</span></div>'
                  for e in headings[:25])
if len(headings) > 25:
    outline += f'<div style="color:{C["slate"]};">… et {len(headings) - 25} autres titres</div>'
display(HTML(
    f'{_CSS}<div class="rg rg-card"><div class="rg-title">🔍 {_html.escape(d_demo.short)} · page {demo_page} vue par l\'extracteur</div>'
    f'<div style="display:flex;gap:20px;flex-wrap:wrap;align-items:flex-start;">'
    f'<img src="data:image/png;base64,{annotated_page(d_demo, demo_page)}" style="width:320px;border:1px solid {C["sage"]};border-radius:6px;">'
    f'<div style="flex:1;min-width:280px;font-size:13px;"><div class="rg-title">🧭 Plan détecté</div>'
    f'<div style="line-height:1.6;">{outline}</div>{block}</div></div></div>'))
show_table(pd.DataFrame([{"Document": d.short, "Titres": sum(e.type == "heading" for e in d.elements),
                          "Paragraphes": sum(e.type == "paragraph" for e in d.elements),
                          "Tableaux": sum(e.type == "table" for e in d.elements),
                          "Paragraphes OCR": sum(e.type == "ocr" for e in d.elements),
                          "En-têtes supprimés": d.info["headers removed"],
                          "Lignes de sommaire écartées": d.info.get("contents lines dropped", 0)} for d in LIB.documents]),
           title="Ce que l'extraction a produit, document par document")
````

> 💡 **Ce qu'il faut observer.** BM25 l'emporte sur les codes, sigles et noms exacts d'indicateurs (*« COICOP 01.1 »*, *« IHPC »*) ; la recherche sémantique l'emporte quand les mots des utilisateurs diffèrent de ceux du rapport (*« hausse des prix »* contre *« inflation »*). L'hybride garde les deux.


<div style="border-left:6px solid #0E7C86;background:#F4F7F5;padding:14px 22px;border-radius:0 12px 12px 0;margin:8px 0 6px 0;font-family:'Segoe UI',Calibri,system-ui,sans-serif;">
<div style="color:#0E7C86;font-size:11.5px;font-weight:700;letter-spacing:2.5px;text-transform:uppercase;">Parcours approfondi</div>
<div style="font-size:23px;font-weight:800;color:#231F20;margin:3px 0 4px 0;">B3 · Découper en passages</div>
<div style="color:#5E6964;font-size:14px;line-height:1.45;">Des passages qui conservent leur document, leur section, leurs pages et leur position.</div></div>

Le découpage (*chunking*) est la décision qui pèse le plus sur les réponses. Découpez trop finement et un passage ne contient plus assez pour répondre ; découpez trop largement et la recherche ne parvient plus à distinguer les passages. Le notebook propose **cinq niveaux**, de la référence naïve à celui qui obtient les meilleures mesures. Modifiez `CHUNK_STRATEGY` dans les paramètres et relancez la partie A.

| `CHUNK_STRATEGY` | Ce qu'il fait | Ce qu'il vous coûte |
|---|---|---|
| `"fixed"` | **1 · Blocs fixes.** Coupe tous les `MAX_CHARS` caractères. C'est ce que font presque tous les tutoriels. | Les phrases, les tableaux et les titres sont coupés au milieu. Un passage ne sait pas à quelle section il appartient. |
| `"sentences"` | **2 · Fenêtres de phrases.** Regroupe N phrases entières, en glissant. | Rien n'est coupé au milieu d'une phrase, mais les passages sont courts et ignorent leur section : la recherche trouve des fragments qui ne peuvent pas répondre seuls. |
| `"sections"` | **3 · Passages structurels.** Suit les titres trouvés en partie B2 ; chaque tableau forme un passage à part ; le chemin des titres accompagne le texte. | L'unité de recherche est aussi l'unité de lecture : une longue section est donc mise en concurrence comme un seul bloc. |
| `"hybrid"` | **4 · Deux granularités.** Indexe les passages structurels **et** de petites fenêtres de phrases qui renvoient vers eux. | Deux fois plus d'unités à indexer (quelques secondes de plus). |
| `"smart"` | **5 · Deux granularités, tableaux et contexte.** Ajoute chaque ligne de tableau comme unité à part entière, et transmet avec la réponse les passages voisins de la section. | Le plus grand nombre d'unités et l'indexation la plus lente — toujours quelques secondes seulement pour un rapport de 80 pages. |

### Les cinq règles que respectent les niveaux 3 à 5

| Règle | Paramètre | Pourquoi |
|---|---|---|
| Des phrases entières, jusqu'à une taille limite | `MAX_CHARS = 700` | Une phrase coupée peut être retrouvée mais pas citée |
| Une phrase répétée entre deux passages | `OVERLAP_SENTENCES = 1` | La réponse se trouve souvent à cheval sur une frontière |
| Les passages courts fusionnés avec le précédent | `MIN_CHARS = 280` | Une ligne isolée n'a pas de contexte et ne répond à rien |
| Chemin complet des titres dans le texte de recherche | — | « 2. Stratégie › 2.4 Domaines prioritaires » distingue des sections aux formulations proches |
| Tableaux tenus à part, chaque ligne portant sa légende | `EXTRACT_TABLES` | Un nombre n'a pas de sens sans les libellés de sa ligne et de sa colonne |

> 💡 **Mesuré, pas supposé.** Sur les deux rapports institutionnels utilisés pour régler ce notebook (66 et 81 pages, 12 questions dont la réponse est connue) :
>
> | Niveau | Réponse dans le contexte | Bon passage en premier |
> |---|:-:|:-:|
> | 1 · Blocs fixes | 11/12 | 8/12 |
> | 2 · Fenêtres de phrases | 12/12 | **6/12** |
> | 3 · Passages structurels | 11/12 | 10/12 |
> | 4 · Deux granularités | **12/12** | **10/12** |
> | 5 · Deux granularités, tableaux et contexte | **12/12** | **10/12** |
>
> La leçon qui surprend le plus : **indexer uniquement de petits morceaux est la pire option** (niveau 2). Une fenêtre de deux phrases contient rarement tous les mots d'une question, si bien que la recherche lexicale la manque. Les petits morceaux n'aident que lorsqu'un passage plus large se tient derrière eux — ce sont les niveaux 4 et 5.
>
> ⚠️ Ces chiffres proviennent de **deux** rapports et de **douze** questions. Ils suffisent à séparer des stratégies très différentes, pas à prouver qu'un niveau en bat un autre d'une seule question. Lancez le banc d'essai ci-dessous sur vos propres documents.

````python
sample = LIB.passages[:40]
show_table(pd.DataFrame([{"N°": p.id, "Document": p.doc[:30], "Section": (p.section or "—")[:36],
                          "Pages": ", ".join(map(str, p.pages)), "Type": p.kind, "Caractères": len(p.text),
                          "Début": _html.escape(p.text[:80].replace(chr(10), " ")) + "…"} for p in sample]),
           title=f"✂️ {len(LIB.passages)} passages indexés",
           note=("Seuls les 40 premiers sont affichés ; la liste complète est exportée dans passages.csv en partie D."
                 if len(LIB.passages) > 40 else None))
lengths = [len(p.text) for p in LIB.passages]
show_table(pd.DataFrame([{"Passages de contexte": len(LIB.passages), "Unités de recherche": len(LIB.units),
                          "Longueur moyenne d'un passage": int(np.mean(lengths)),
                          "Passages de tableau": sum(p.kind == "table" for p in LIB.passages),
                          "Stratégie": CHUNK_STRATEGY, "Acronymes appris": len(ACRONYMS)}]),
           title="L'index en un coup d'œil")
````

#### Choisir sur des preuves, pas par conviction

`benchmark()` réindexe votre collection avec chaque combinaison et la mesure. Il utilise, par ordre de préférence :

1. **vos propres questions**, si vous avez rempli `EVALUATION_QUESTIONS` dans les paramètres — de loin le plus significatif :
   ```python
   EVALUATION_QUESTIONS = [
       ("Combien de dimensions compte le modèle de maturité ?", ["12 dimensions"]),
       ("Quels pays ont participé au diagnostic ?", ["Tunisie", "Sénégal"]),
   ]
   ```
   Chaque entrée est une question et le texte qui doit figurer dans les passages transmis au modèle.
2. sinon, des **questions-sondes construites à partir des documents eux-mêmes** : les mots d'une phrase doivent ramener une phrase *ultérieure* du même sujet. Un découpage qui coupe un sujet en deux échoue. C'est un indicateur approché, pas un verdict.

L'indexation a lieu une fois par combinaison : cinq niveaux sur un rapport de 80 pages prennent environ une demi-minute — et bien plus avec des embeddings neuronaux.

> ⚠️ **Un petit corpus ne permet pas de départager les niveaux.** Sur les rapports d'exemple de 5 pages (« Rapport annuel sur le marché du travail 2025 », « Bulletin trimestriel de l'emploi T1 2026 », « Recensement de la population 2024 »), tous les niveaux obtiennent le même score et plusieurs se partagent le 🏆. Ce n'est pas une égalité à généraliser : les écarts apparaissent sur les rapports de 60 à 200 pages avec lesquels un institut de statistique travaille réellement, comme le montre le tableau de mesures plus haut. Évaluez les documents que vous utiliserez vraiment.

<div style="color:#5E6964;font-size:14px;line-height:1.45;">Treize étapes entre la question et la réponse — et les quatre niveaux de reclassement.</div></div>

### Les treize étapes, et où vous pouvez agir

| # | Étape | Ce qu'elle fait | Réglage |
|:-:|---|---|---|
| 1 | **Question** | telle que saisie par l'utilisateur | — |
| 2 | **Relance** | « Et pour les hommes ? » est réécrite en question autonome (B7) | `MODEL_FOLLOW_UP` |
| 3 | **Langue et orthographe** | la langue de la question est détectée et, avec un modèle, la question est traduite dans chaque langue de la collection (B10) ; un mot que les documents n'emploient jamais est remplacé par le plus proche **de la même langue** : *chaumage* → *chômage* | `QUESTION_LANGUAGE`, `TRANSLATE_QUERIES` |
| 4 | **Normalisation** | minuscules, accents, racines des mots ; les mots interrogatifs (« nombre de », « liste », « signifie ») sont écartés | `STOPWORDS` |
| 5 | **Expansion** | glossaire du domaine, plus les sigles appris dans les documents | `GLOSSARY` |
| 6 | **BM25** | mots exacts, sur chaque unité de recherche | `CHUNK_STRATEGY` |
| 7 | **Sémantique** | le sens plutôt que les mots | `USE_EMBEDDINGS` |
| 8 | **Fusion** | les deux classements fusionnés (Reciprocal Rank Fusion) | — |
| 9 | **Reclassement** | c'est ici que l'ordre est décidé | `RERANKING` |
| 10 | **Fenêtres → passages** | chaque petite unité renvoie le passage qu'un humain devrait lire | `CHUNK_STRATEGY` |
| 11 | **Garde-fous** | seuil, contrôle de l'année, refus (B5) | `MANUAL_THRESHOLD` |
| 12 | **Réponse** | rédigée par le modèle, ou extraite du passage | Groq / Ollama |
| 13 | **Contrôles** | chiffres vérifiés, termes non trouvés, meilleure réponse exclue par un filtre (B9) | — |

L'étape 7 mérite un mot : **sans `sentence-transformers`, le notebook se rabat sur une approximation hors ligne (LSA sur des n-grammes de caractères)**, qui rapproche des graphies plutôt que des sens. Elle n'échoue jamais, mais elle ne peut pas relier « participating in the assessment » et « the Bank has conducted diagnostic missions in ». Si vos questions emploient les mots de vos utilisateurs plutôt que ceux du rapport, installez le modèle neuronal : `INSTALL_EMBEDDINGS = True`.

### Les quatre niveaux de reclassement

Le classement fusionné est rarement le bon ordre. Le reclassement (reranking) note à nouveau la liste restreinte ; `RERANKING` choisit comment.

| `RERANKING` | Ce qu'il fait | Coût |
|---|---|---|
| `"none"` | **0 · Aucun.** Conserve l'ordre de la fusion. | Gratuit |
| `"coverage"` | **1 · Couverture de la question.** Part des termes pondérés de la question présents dans le passage. Un mot rare pèse plus qu'un mot fréquent ; un mot absent de toute la collection pèse le plus, ce qui rend un refus possible. | Gratuit |
| `"smart"` | **2 · Couverture, densité et type de réponse.** Sur un long rapport, de nombreux passages couvrent toute la question : les ex æquo sont départagés par **ce que demande la question** (une question « combien » préfère un passage portant une vraie quantité, pas « voir Figure 8 » ; une question « liste », un passage portant une énumération), puis par la densité des termes. | Gratuit |
| `"cross-encoder"` | **3 · Cross-encoder neuronal.** Un modèle lit la question et le passage *ensemble* et note la paire. Le plus précis, et le plus lent : il nécessite `sentence-transformers` et quelques centaines de millisecondes par question. | ~0,3–1 s/question |

> 💡 **Mesuré sur les deux mêmes rapports**, découpage fixé au niveau 5 :
>
> | Reclassement | Réponse dans le contexte | Bon passage en premier |
> |---|:-:|:-:|
> | 0 · Aucun | 10/12 | 6/12 |
> | 1 · Couverture de la question | 12/12 | 8/12 |
> | 2 · Couverture, densité et type de réponse | **12/12** | **10/12** |
>
> Le cross-encoder est absent de ce tableau à dessein : il n'a pas pu être téléchargé dans l'environnement où le notebook a été mis au point. Lancez vous-même `benchmark(reranking=("smart", "cross-encoder"))` : sur les bancs d'essai publics, un cross-encoder l'emporte généralement, au prix de la latence.

> 💡 **Le glossaire vous appartient.** Enrichissez `GLOSSARY` (cellule T4) avec le vocabulaire de vos utilisateurs, par exemple « framework » → « maturity model diagnostic » sur le guide de l'AfDB. C'est souvent l'amélioration la plus rentable de toutes, et c'est la bonne réponse lorsqu'une question échoue sans cesse parce que vos utilisateurs et vos rapports n'emploient pas les mêmes mots.


````python
q = ("Quel était le taux de chômage au quatrième trimestre 2025 ?" if using_examples()
     else f"Que dit le document sur : {next((e.text for d in LIB.documents for e in d.elements if e.type == 'heading' and e.level > 1), 'le thème principal')} ?")
show_table(pd.DataFrame([{"Question": q, "Termes analysés": ", ".join(analyse(q)),
                          "Ajouts du glossaire": "<br>".join(f"<b>{k}</b> → {t}" for k, t in expand_query(q)[1]) or "—"}]),
           title="Ce que le moteur de recherche « voit » dans une question")

def _ranking(docs):
    return [f"{p.doc[:28]}… › {p.section[:22]} ({s:.2f})" for p, s in LIB.search(q, 4, docs)]
columns = {"Tous les documents": _ranking(None)}
for d in LIB.documents[:2]:
    columns[f"Seulement : {d.short[:30]}"] = _ranking([d.id])
length = max(len(v) for v in columns.values())
show_table(pd.DataFrame({"Rang": range(1, length + 1),
                         **{k: v + ["—"] * (length - len(v)) for k, v in columns.items()}}),
           title=f"Effet du filtre par document · « {q} »",
           note=f"Recherche sémantique : {LIB.semantic.name}. Entre parenthèses : la confiance.")
````

Trois garde-fous décident si l'assistant peut répondre :

1. **Seuil de confiance** : si le meilleur passage n'est pas assez pertinent, l'assistant refuse **sans appeler le modèle** ;
2. **Contrôle de l'année** : une année mentionnée dans la question doit figurer dans les passages ;
3. **Prompt strict** : répondre uniquement à partir des sources, les citer, et refuser avec une phrase fixe sinon.

### Comment le seuil est calculé

**Confiance d'une question** = poids des mots de la question trouvés dans le meilleur passage ÷ poids de tous les mots de la question. Les mots rares pèsent plus ; les mots trouvés **nulle part** dans les documents (comme « hôpitaux ») pèsent le plus, car ils signalent une question hors sujet.

**Calibrage** : le notebook pose deux séries de questions de référence : des questions construites à partir des **titres de section** (la réponse existe) et des questions **hors sujet** (pas de réponse). Le seuil est placé juste au-dessus du meilleur score hors sujet :

> **Seuil = meilleure confiance hors sujet + 0,05** (plafonné à 0,95)

> ⚠️ Ce seuil est un point de départ : ajustez-le dans les interfaces si l'assistant refuse trop souvent, ou pas assez. Pour le fixer vous-même : `MANUAL_THRESHOLD = 0.5`.


````python
cal = LIB.calibration
fig, ax = plt.subplots(figsize=(8.6, 2.8))
for rank, (exists, colour) in enumerate([(False, C["brick"]), (True, C["green"])]):
    v = cal.loc[cal["Answer exists"] == exists, "Confidence"]
    ax.scatter(v, np.full(len(v), rank) + np.linspace(-0.13, 0.13, len(v)), s=95, color=colour,
               alpha=0.85, edgecolor="white", linewidth=1.2, zorder=3)
ax.axvline(LIB.threshold, color=C["ochre"], linewidth=2.2, linestyle="--")
ax.axvspan(0, LIB.threshold, color=C["brick"], alpha=0.05); ax.axvspan(LIB.threshold, 1.02, color=C["green"], alpha=0.05)
ax.text(LIB.threshold, 1.55, f"  SEUIL = {LIB.threshold:.2f}", color=C["ochre"], fontweight="bold", va="center")
ax.set_yticks([0, 1], ["Hors sujet", "Titres des documents"]); ax.set_xlim(0, 1.02); ax.set_ylim(-0.6, 1.8)
ax.set_xlabel("Confiance du meilleur passage"); ax.set_title("Calibrage du seuil sur vos documents")
plt.tight_layout(); fig.savefig(OUTPUT_DIR / "threshold_calibration.png", dpi=160); plt.show()
show_table(cal.sort_values("Confidence", ascending=False)
              .rename(columns={"Answer exists": "La réponse existe", "Confidence": "Confiance"}),
           title="Questions de référence")

for q in ["Quel est le prix du maïs au kilogramme ?", "Quel sera le taux de chômage en 2030 ?"]:
    LIB.show(LIB.answer(q))
````

<div style="color:#5E6964;font-size:14px;line-height:1.45;">Comprendre « et pour les hommes ? » grâce à la question précédente.</div></div>


Une question de relance n'a pas de sens seule : « Et pour les hommes ? » ne contient aucun indicateur. Avant de chercher, l'assistant **la réécrit en question autonome** :

| Méthode | Principe | Quand |
|---|---|---|
| **Par le modèle** | Le modèle reçoit les derniers échanges et réécrit la question | Un modèle est disponible et `MODEL_FOLLOW_UP = True` |
| **Par des règles** | La question précédente est reprise et la valeur qui change (région, sexe, trimestre, année, milieu, âge, indicateur) est **remplacée** | Sans modèle, ou en secours |

La réécriture est **toujours affichée** (« 🧠 Compris comme : … ») pour que l'utilisateur puisse la vérifier.

> ⚠️ **La mémoire peut se tromper.** Si l'utilisateur change de sujet avec une question très courte, celle-ci peut être rattachée à tort à la précédente. D'où l'affichage systématique, et la possibilité de désactiver la mémoire dans les interfaces.


````python
if using_examples():
    conversation = ["Quel est le taux de chômage des femmes dans la région Côte ?",
                    "Et pour les hommes ?",
                    "Qu'en est-il de la région Nord ?",
                    "Quel est le taux de chômage au deuxième trimestre ?",
                    "Et au troisième trimestre ?"]
else:
    starters = suggested_questions(4)
    conversation = [starters[0], "Et plus en détail ?", starters[1] if len(starters) > 1 else "Et ensuite ?"]
METHOD_LABELS = {"model": "modèle", "rules": "règles"}
history, rows = [], []
for q in conversation:
    r = LIB.answer(q, history=history)
    history.append(r)
    rows.append({"Question posée": q,
                 "Question recherchée": ("🧠 " if r["rewritten"] else "") + _html.escape(r["standalone_question"]),
                 "Méthode": METHOD_LABELS.get(r["rewrite_method"], r["rewrite_method"]) or "—",
                 "Réponse": _html.escape(r["answer"][:110])})
show_table(pd.DataFrame(rows), title="Une conversation avec mémoire")
````

<div style="color:#5E6964;font-size:14px;line-height:1.45;">Trois contrôles automatiques affichés avec chaque réponse.</div></div>


Même avec des garde-fous, une réponse peut n'être fondée qu'**en partie**. Trois contrôles sont donc affichés sur chaque carte de réponse :

| Contrôle | Principe | Ce qu'il détecte |
|---|---|---|
| **Chiffres vérifiés** | Chaque nombre de la réponse doit figurer dans une source fournie | Un modèle qui arrondit, confond ou **invente** un chiffre |
| **Termes non trouvés** | Les mots de la question absents du passage principal sont listés | Une réponse portant sur **un autre indicateur** que celui demandé |
| **Meilleure réponse ailleurs** | Si le filtre exclut un document nettement plus pertinent, c'est signalé | Un filtre par document trop restrictif |

> 💡 **Point clé.** Ces contrôles ne remplacent pas la relecture : ils **orientent** l'attention du lecteur vers les réponses à vérifier en priorité.


````python
# 1. Contrôle des chiffres : une réponse correcte et une réponse « inventée », avec les mêmes sources.
r = LIB.answer("Quel est le taux de chômage au deuxième trimestre ?")
show_table(pd.DataFrame([
    {"Réponse testée": "Le taux de chômage s'établit à 7,8 % au deuxième trimestre [1].",
     "Contrôle": str(check_figures("Le taux de chômage s'établit à 7,8 % au deuxième trimestre [1].", r["passages"]))},
    {"Réponse testée": "Le taux de chômage s'établit à 7,3 % au deuxième trimestre [1].",
     "Contrôle": str(check_figures("Le taux de chômage s'établit à 7,3 % au deuxième trimestre [1].", r["passages"]))},
]), title="Contrôle des chiffres", note="7,3 n'apparaît dans aucune source : il est signalé.")

# 2. Termes non trouvés et 3. meilleure réponse ailleurs : une question de population, filtrée sur le rapport annuel.
report = next((d for d in LIB.documents if any(k in d.filename.lower() for k in ("rapport", "report"))), LIB.documents[0])
LIB.show(LIB.answer("Quelle est la population de la région Nord ?", docs=[report.id]))
````

<div style="color:#5E6964;font-size:14px;line-height:1.45;">Des documents en plusieurs langues, des questions dans n'importe quelle langue, des réponses dans la vôtre.</div></div>


Un office statistique travaille rarement dans une seule langue : la stratégie continentale existe en anglais et en français, l'enquête nationale est en portugais, le rapport du partenaire est en espagnol. Un RAG naïf échoue de deux façons : **il analyse chaque texte avec les règles d'une seule langue** (racines françaises appliquées à l'anglais, virgules décimales mal lues), et **il ne trouve que les passages écrits dans la langue de la question** : une question en français ne retrouve jamais la note en anglais qui contient la réponse.

### Chaque texte est lu dans sa propre langue

| Étape | Ce qui se passe | Réglage |
|---|---|---|
| **Langue du document** | Détectée à partir de ses mots-outils (*le, des* / *the, of* / *el, los* / *o, da*…) | `DOCUMENT_LANGUAGE = "auto"` |
| **Langue du paragraphe** | Chaque paragraphe, titre et tableau d'un rapport reçoit sa propre langue ; un titre trop court pour en juger prend la langue du paragraphe qu'il annonce | `DETECT_PASSAGE_LANGUAGE` |
| **Limites des passages** | Un passage est coupé quand la langue change, même à l'intérieur d'une section : aucun passage ne mélange la moitié anglaise et la moitié française d'un rapport | — |
| **Langue de la question** | Détectée pour chaque question ; une relance courte hérite de la précédente | `QUESTION_LANGUAGE` |
| **Mots vides et racines** | Chaque passage et chaque question est analysé avec les ressources de sa langue (racinisateurs Snowball pour l'espagnol, le portugais, l'allemand, l'italien, l'arabe) | — |
| **Nombres** | `7,8` (français) et `7.8` (anglais), `18 452 300` et `18,452,300` deviennent le même jeton | — |
| **Tolérance orthographique** | Une faute de frappe n'est corrigée qu'avec des mots de la **même** langue | — |

### Trois ponts entre les langues

| Pont | Fonctionnement | Prérequis |
|---|---|---|
| 🧠 **Embeddings multilingues** | Le modèle par défaut (`paraphrase-multilingual-MiniLM-L12-v2`) place près l'une de l'autre une question et un passage de même sens, quelle que soit leur langue | `sentence-transformers` (l'approximation LSA hors ligne n'est **pas** interlingue) |
| 🔁 **Traduction de la question** | Le modèle traduit la question dans chaque langue de la collection ; chaque traduction est cherchée dans les passages de sa langue, et les classements sont fusionnés | Un modèle (Groq, Ollama…) et `TRANSLATE_QUERIES = True` : un court appel par langue supplémentaire |
| 📖 **Lexique interlingue** | `CROSS_LINGUAL_LEXICON` relie une soixantaine de concepts statistiques en anglais, français, espagnol et portugais (*unemployment · chômage · desempleo · desemprego*). Il fonctionne **sans modèle et sans Internet** | Rien : enrichissez-le avec votre propre domaine |

### Plusieurs langues dans un même rapport

Bulletins bilingues, annexe en français dans un rapport en anglais, tableaux intitulés « Region / Région » : un seul PDF peut contenir plusieurs langues. Quatre mécanismes permettent à la chaîne de fonctionner sur de tels rapports :

| Mécanisme | Ce qu'il fait | Pourquoi c'est important |
|---|---|---|
| **Langue par paragraphe** | le document n'est plus « anglais » ou « français » : il est, par exemple, *en 55 % · fr 45 %* | racines, mots vides et virgules décimales sont traités avec les bonnes règles |
| **Passages monolingues** | le découpage coupe à chaque changement de langue | une question en français n'est plus comparée à un passage à moitié en anglais |
| **Passages mixtes** | un tableau bilingue ou un paragraphe qui se répète dans une seconde langue est analysé dans les **deux** langues (`langs`) | l'une ou l'autre version de la question le trouve ; le mode extractif cite la phrase dans la langue du lecteur |
| **Traductions fusionnées** | un passage et sa traduction (deux moitiés d'un rapport bilingue, ou deux éditions d'une même stratégie) sont reconnus (mêmes chiffres `38.2` / `38,2`, ou embeddings multilingues au-dessus de `TWIN_SIMILARITY`) et ne comptent **qu'une fois**, dans la langue du lecteur | les `TOP_K` passages apportent des informations différentes au lieu de la même en double ; la carte montre 🔁 ce qui a été fusionné |

Le **score de confiance** (B5) est mesuré avec la version de la question écrite dans la langue de chaque passage, si bien qu'un seul seuil vaut pour toutes les langues ; le calibrage utilise des questions hors sujet **dans chaque langue de la collection**. Le modèle rédige ensuite la réponse dans `ANSWER_LANGUAGE` (par défaut, la langue de la question) en conservant les chiffres, sigles et intitulés de programmes tels qu'écrits dans la source. Le reclasseur `cross-encoder` passe automatiquement à un modèle **multilingue** (mMARCO).

> ⚠️ **Documents jumeaux en deux langues.** Quand le même rapport existe en anglais et en français, les deux versions répondent. `PREFER_QUESTION_LANGUAGE = True` favorise, à score égal, la version dans la langue de la question ; le filtre par document permet de n'en garder qu'une.

> 💡 **Langues moins dotées** (kinyarwanda, wolof, amharique…) : elles sont indexées sans racines ni mots vides. Utilisez un modèle d'embeddings multilingue plus puissant (`intfloat/multilingual-e5-base`, `BAAI/bge-m3`) et un modèle de langage pour la traduction, et ajoutez leur code Tesseract à `OCR_LANGUAGE` pour les pages scannées.


````python
# 1. Les langues que contient la collection, et les ponts actifs
show_table(pd.DataFrame([{
    "Document": d.short, "Langue principale": f"{nom_langue(d.lang)} ({d.lang})",
    "Langues présentes": mix_label(d.lang_mix) if d.lang_mix else d.lang,
    "Passages": sum(p.doc_id == d.id for p in LIB.passages),
    "Passages dans une autre langue": sum(p.doc_id == d.id and p.lang != d.lang for p in LIB.passages),
    "Passages mixtes": sum(p.doc_id == d.id and len(passage_langs(p)) > 1 for p in LIB.passages)}
    for d in LIB.documents]),
    title=f"🌍 {len(LIB.languages)} langue(s) dans la collection : "
          + ", ".join(f"{nom_langue(l)} ({n} passages)" for l, n in LIB.language_counts.most_common()))

translation_on = TRANSLATE_QUERIES and LLM.available
show_table(pd.DataFrame([
    {"Pont entre les langues": "🧠 Embeddings multilingues", "Actif": "✅" if LIB.semantic.cross_lingual else "—",
     "Détail": LIB.semantic.name},
    {"Pont entre les langues": "🔁 Traduction de la question", "Actif": "✅" if translation_on else "—",
     "Détail": f"{LLM.name} · {LLM.model}" if translation_on else "nécessite un modèle et TRANSLATE_QUERIES = True"},
    {"Pont entre les langues": "📖 Lexique interlingue", "Actif": "✅",
     "Détail": f"{len(CROSS_LINGUAL_LEXICON)} concepts · en, fr, es, pt"},
    {"Pont entre les langues": "🎯 Cross-encoder multilingue", "Actif": "✅" if LIB.reranker.cross is not None else "—",
     "Détail": LIB.reranker.model_name or f"reclassement : {LIB.reranker.name}"},
]), title="Ce qui permet à une question de trouver un passage écrit dans une autre langue")
if len(LIB.languages) > 1 and not (LIB.semantic.cross_lingual or translation_on):
    callout("Seul le <b>lexique</b> relie les langues pour l'instant : les questions employant des mots qu'il ne contient pas "
            "peuvent manquer des passages écrits dans une autre langue. Connectez un modèle (étape 0b) ou installez les "
            "embeddings multilingues (<code>USE_EMBEDDINGS = INSTALL_EMBEDDINGS = True</code>).", "warning")

# 2. Le même besoin, exprimé dans plusieurs langues
if using_multilingual_examples():
    PROBES = ["Quel est le taux de chômage des jeunes ?", "What is the youth unemployment rate?",
              "¿Cuál es la tasa de desempleo de los jóvenes?", "Qual é a taxa de desemprego dos jovens?",
              "Quelle était l'inflation des prix à la consommation au premier trimestre 2026 ?",   # réponse dans la note en anglais
              "Quelle proportion des ménages a accès à internet ?",                                 # réponse dans l'enquête en espagnol
              "What is the household access rate to electricity?",           # réponse dans la moitié française d'un rapport
              "Quelle proportion des ménages a accès à l'eau potable ?"]      # réponse dans la moitié anglaise
else:
    base = suggested_questions(1)[0]
    base_lang = question_language(base, LIB.languages)
    PROBES = [base] + [t for t in (translate_query(base, l) for l in ("fr", "en", "es", "pt") if l != base_lang) if t]
    if len(PROBES) == 1:
        callout("Connectez un modèle pour traduire automatiquement la question de démonstration, ou écrivez vos propres "
                "questions dans d'autres langues dans <code>PROBES</code>.", "info")
rows = []
for q in PROBES:
    r = LIB.answer(q)
    top = r["passages"][0][0] if r["passages"] else None
    rows.append({"Question": _html.escape(q), "Détectée": r["language"],
                 "Aussi cherchée comme": "<br>".join(f"<b>{l}</b> {_html.escape(t)}" for l, t in r["queries"].items()) or "—",
                 "Source principale": f"{top.doc} ({top.lang})" if top else "—",
                 "Confiance": f"{r['confidence']:.2f}",
                 "Réponse": ("⛔ " if r["refused"] else "") + _html.escape(r["answer"][:140])})
show_table(pd.DataFrame(rows), title="🌐 Un besoin, plusieurs langues",
           note="La source principale ne dépend pas de la langue de la question : c'est ce qu'un RAG multilingue doit obtenir. "
                "En mode extractif, le passage est cité dans sa propre langue ; avec un modèle, la réponse est rédigée dans la langue de la question.")
````

#### À l'intérieur d'un rapport multilingue

La cellule ci-dessous ouvre chaque rapport qui contient **plus d'une langue** : la langue attribuée à chaque paragraphe, les passages construits à partir d'eux, et la même question posée dans deux langues ; le passage renvoyé est la version dans la langue de la question, et sa traduction est fusionnée (🔁 sur la carte).


````python
mixed_documents = [d for d in LIB.documents if len(document_languages(d)) > 1]
if not mixed_documents:
    callout("Aucun rapport de cette collection ne mélange plusieurs langues. Essayez <code>DOCUMENTS = [\"exemple-multilingue\"]</code>, "
            "qui comprend une enquête bilingue français/anglais.", "info")
for d in mixed_documents[:3]:
    colours = {l: c for l, c in zip(dict.fromkeys(e.lang for e in d.elements),
                                    [C["dark_green"], C["teal"], C["terra"], C["ochre"], C["brick"]] * 3)}
    rows = [{"Page": e.page, "Élément": e.type, "Section": _html.escape((e.section or "—")[:38]),
             "Langue": pill(e.lang, colours[e.lang]), "Texte": _html.escape(" ".join(e.text.split())[:110]) + "…"}
            for e in d.elements[:40]]
    show_table(pd.DataFrame(rows), title=f"🧬 {_html.escape(d.short)} · {mix_label(d.lang_mix)}",
               note="Chaque paragraphe porte sa propre langue ; les titres trop courts pour être reconnus prennent la langue "
                    "du paragraphe qui les suit.")
    show_table(pd.DataFrame([{"N°": p.id, "Section": _html.escape((p.section or "—")[:38]), "Type": p.kind,
                              "Langues": " + ".join(passage_langs(p)),
                              "Début": _html.escape(" ".join(p.text.split())[:110]) + "…"}
                             for p in LIB.passages if p.doc_id == d.id][:30]),
               title="Les passages construits à partir de lui : une seule langue chacun, ou explicitement mixtes")

if using_multilingual_examples():
    for q in ["Quel est le taux de pauvreté national ?", "What is the national poverty rate?"]:
        LIB.show(LIB.answer(q))
````

> 💡 **Les retours des utilisateurs** sont enregistrés dans `user_feedback.csv`, avec la question, la réponse et ses sources. Les réponses marquées « À revoir » sont la meilleure matière pour **enrichir le glossaire et le jeu de tests**.


<div style="border-left:6px solid #00704A;background:#F4F7F5;padding:14px 22px;border-radius:0 12px 12px 0;margin:8px 0 6px 0;font-family:'Segoe UI',Calibri,system-ui,sans-serif;">
<div style="color:#00704A;font-size:11.5px;font-weight:700;letter-spacing:2.5px;text-transform:uppercase;">Interfaces</div>
<div style="font-size:23px;font-weight:800;color:#231F20;margin:3px 0 4px 0;">C1 · L'interface dans le notebook</div>
<div style="color:#5E6964;font-size:14px;line-height:1.45;">Questions suggérées, filtre de documents, mémoire, paramètres et retours.</div></div>


````python
import ipywidgets as widgets

HISTORY = []
SUGGESTIONS = suggested_questions(6)

def _header():
    model_label = f"{LLM.name} · {LLM.model}" if LLM.available else "mode extractif"
    return (f'{_CSS}<div class="rg" style="background:linear-gradient(120deg,{C["forest"]} 0%,{C["green"]} 100%);'
            f'border-radius:14px;padding:14px 20px;color:#fff;"><div style="color:{C["gold"]};font-size:11px;'
            f'font-weight:700;letter-spacing:2.5px;">ASSISTANT DOCUMENTAIRE</div>'
            f'<div style="font-size:20px;font-weight:800;margin:2px 0;">💬 {len(LIB.documents)} document(s) · '
            f'{len(LIB.passages)} passages · 🌐 {", ".join(LIB.languages)}</div><div style="font-size:12.5px;color:#E6F6EE;">'
            f'posez vos questions dans n’importe quelle langue · modèle : '
            f'{_html.escape(model_label)} · recherche : {_html.escape(LIB.semantic.name)}</div></div>')

def _thread():
    if not HISTORY:
        return (f'{_CSS}<div class="rg rg-card" style="text-align:center;color:{C["slate"]};padding:26px;">'
                'Posez une question, ou cliquez sur une suggestion.</div>')
    bubbles = "".join(
        f'<div style="display:flex;justify-content:flex-end;margin:12px 0 6px 0;"><div style="background:{C["dark_green"]};'
        f'color:#fff;border-radius:14px 14px 2px 14px;padding:8px 14px;max-width:80%;font-size:14px;">'
        f'{_html.escape(r["question"])}</div></div><div style="display:flex;gap:8px;align-items:flex-start;">'
        f'<div style="font-size:22px;">🤖</div><div style="flex:1;">{LIB.card_html(r, ui_thumbnails.value)}</div></div>'
        for r in reversed(HISTORY))
    return (f'{_CSS}<div class="rg" style="max-height:720px;overflow-y:auto;padding:4px 6px;border:1px solid '
            f'{C["sage"]};border-radius:12px;background:{C["mist"]};">{bubbles}</div>')

ui_question = widgets.Text(placeholder="Posez une question sur les documents, dans n’importe quelle langue…", layout=widgets.Layout(flex="1", width="auto"))
ui_send = widgets.Button(description="Envoyer", button_style="success", icon="paper-plane", layout=widgets.Layout(width="120px"))
ui_status = widgets.HTML("")
ui_docs = [widgets.Checkbox(value=True, description=d.short, indent=False, layout=widgets.Layout(width="auto", margin="0 14px 0 0"))
           for d in LIB.documents]
ui_k = widgets.IntSlider(value=TOP_K, min=1, max=8, description="Passages", continuous_update=False)
ui_threshold = widgets.FloatSlider(value=float(LIB.threshold), min=0.0, max=1.0, step=0.01, description="Seuil",
                                   continuous_update=False, readout_format=".2f")
ui_guardrails = widgets.Checkbox(value=True, description="Garde-fous actifs", indent=False)
ui_memory = widgets.Checkbox(value=True, description="Mémoire de conversation", indent=False)
ui_thumbnails = widgets.Checkbox(value=True, description="Afficher les pages", indent=False)
ui_clear = widgets.Button(description="Nouvelle conversation", icon="refresh", layout=widgets.Layout(width="190px"))
ui_export = widgets.Button(description="Exporter", icon="download", layout=widgets.Layout(width="120px"))
ui_useful = widgets.Button(description="Utile", icon="thumbs-up", layout=widgets.Layout(width="100px"))
ui_review = widgets.Button(description="À revoir", icon="thumbs-down", layout=widgets.Layout(width="130px"))
ui_comment = widgets.Text(placeholder="Commentaire facultatif sur la dernière réponse…", layout=widgets.Layout(flex="1", width="auto"))
ui_thread = widgets.HTML(_thread())

def give_feedback(useful):
    if not HISTORY:
        ui_status.value = "Posez d’abord une question."
        return
    record_feedback(HISTORY[-1], useful, ui_comment.value.strip(), "notebook")
    ui_comment.value = ""
    ui_status.value = (f'<span style="color:{C["dark_green"]};">Merci ! Retour enregistré '
                       f'({len(FEEDBACK)} au total, fichier user_feedback.csv).</span>')

def send(_=None, question=None):
    q = (question or ui_question.value).strip()
    if not q:
        return
    docs = [d.id for d, c in zip(LIB.documents, ui_docs) if c.value]
    if not docs:
        ui_status.value = f'<span style="color:{C["brick"]};">Cochez au moins un document.</span>'
        return
    ui_question.value = ""
    ui_send.disabled = True
    ui_status.value = f'<span style="color:{C["teal"]};">⏳ Recherche dans {len(docs)} document(s)…</span>'
    try:
        r = LIB.answer(q, history=HISTORY if ui_memory.value else None, docs=docs, k=ui_k.value,
                       threshold=ui_threshold.value, guardrails=ui_guardrails.value, memory=ui_memory.value)
        HISTORY.append(r)
        ui_status.value = (f'<span style="color:{C["slate"]};font-size:12px;">{len(HISTORY)} échange(s) · '
                           f'dernière réponse en {r["duration_s"]} s' + (f' · {LLM.calls} appel(s) au modèle' if LLM.available else "") + "</span>")
    except Exception as error:
        ui_status.value = f'<span style="color:{C["brick"]};">⛔ {_html.escape(str(error)[:200])}</span>'
    finally:
        ui_send.disabled = False
        ui_thread.value = _thread()

def new_conversation(_=None):
    HISTORY.clear()
    ui_thread.value = _thread()
    ui_status.value = ""

def _serialise(r):
    return {"question": r["question"], "standalone_question": r["standalone_question"], "answer": r["answer"],
            "refused": r["refused"], "confidence": round(r["confidence"], 3), "reason": r["reason"], "mode": r["mode"],
            "guardrails": r["guardrails"], "check": r["check"], "time": r["time"],
            "sources": [{"document": p.doc, "section": p.section, "pages": p.pages, "kind": p.kind,
                         "confidence": round(s, 3), "text": p.text} for p, s in r["passages"]]}

def export(_=None):
    if not HISTORY:
        ui_status.value = "Rien à exporter pour l’instant."
        return
    stamp = time.strftime("%Y%m%d_%H%M%S")
    (OUTPUT_DIR / f"conversation_{stamp}.json").write_text(
        json.dumps({"documents": [d.title for d in LIB.documents], "turns": [_serialise(r) for r in HISTORY]},
                   ensure_ascii=False, indent=2), encoding="utf-8")
    html_path = OUTPUT_DIR / f"conversation_{stamp}.html"
    html_path.write_text(f'<html lang="fr"><head><meta charset="utf-8"><title>Conversation</title></head>'
                         f'<body style="max-width:900px;margin:auto;">{_header()}{_thread()}</body></html>', encoding="utf-8")
    ui_status.value = f"✅ Conversation exportée : <code>{html_path.name}</code> (et sa version JSON)"
    if ENV["name"] == "colab":
        from google.colab import files
        files.download(str(html_path))

buttons = []
for s in SUGGESTIONS:
    b = widgets.Button(description=(s[:44] + "…") if len(s) > 45 else s, tooltip=s,
                       layout=widgets.Layout(width="auto", margin="2px 4px 2px 0"))
    b.on_click(lambda _, s=s: send(question=s))
    buttons.append(b)
ui_send.on_click(send)
ui_useful.on_click(lambda _: give_feedback(True))
ui_review.on_click(lambda _: give_feedback(False))
ui_clear.on_click(new_conversation)
ui_export.on_click(export)
ui_thumbnails.observe(lambda _: setattr(ui_thread, "value", _thread()), names="value")
try:
    import warnings
    with warnings.catch_warnings():
        warnings.simplefilter("ignore")
        ui_question.on_submit(send)
except Exception:
    pass

settings_panel = widgets.Accordion(children=[widgets.VBox([
    ui_k, ui_threshold, widgets.HBox([ui_guardrails, ui_memory, ui_thumbnails])])])
settings_panel.set_title(0, "⚙️ Paramètres")
settings_panel.selected_index = None

INTERFACE = widgets.VBox([
    widgets.HTML(_header()),
    widgets.HTML('<div style="font-size:12.5px;margin:10px 0 2px 0;"><b>Documents interrogés</b></div>'),
    widgets.HBox(ui_docs, layout=widgets.Layout(flex_flow="row wrap")),
    widgets.HTML('<div style="font-size:12.5px;margin:8px 0 2px 0;"><b>Suggestions</b></div>'),
    widgets.HBox(buttons, layout=widgets.Layout(flex_flow="row wrap")),
    widgets.HBox([ui_question, ui_send], layout=widgets.Layout(margin="8px 0 2px 0")),
    ui_status, settings_panel,
    widgets.HBox([widgets.HTML('<span style="font-size:12.5px;margin-right:6px;"><b>Dernière réponse :</b></span>'),
                  ui_useful, ui_review, ui_comment]),
    widgets.HBox([ui_clear, ui_export]), ui_thread,
], layout=widgets.Layout(width="100%", max_width="980px"))
display(INTERFACE)
````

**Gradio** est une bibliothèque Python libre et gratuite qui transforme des fonctions Python en page web interactive, sans HTML ni JavaScript. L'application apporte toutes les fonctionnalités de l'assistant dans une page web :

- **glisser-déposer** de nouveaux PDF, indexés à la volée ;
- **filtre** des documents interrogés ;
- **conversation avec mémoire**, avec affichage de la reformulation ;
- **sources** avec vignettes des pages surlignées et vérifications automatiques ;
- **👍 / 👎** sous chaque réponse ;
- **paramètres** : nombre de passages, seuil, garde-fous, mémoire.

Sur **Colab** et **Kaggle**, Gradio crée un **lien public temporaire** (`https://….gradio.live`), valable tant que la session tourne et pour une durée limitée. **En local**, l'application s'ouvre à l'adresse `http://127.0.0.1:7860`. L'application tourne **dans votre session de notebook** : si la session s'arrête, la page cesse de fonctionner.

> ⚠️ **Confidentialité.** Toute personne disposant du lien public peut interroger vos documents et déposer des fichiers. **Définissez `WEB_PASSWORD`** avant de partager un lien, et n'utilisez pas de lien public pour des documents non publiés.

> ⚠️ **Application partagée.** Tous les utilisateurs d'un même lien partagent la même bibliothèque : un PDF déposé par l'un devient interrogeable par tous. Chacun garde sa propre conversation.


````python
def _ensure_gradio():
    try:
        import gradio as gr
        if int(gr.__version__.split(".")[0]) >= 5:
            return gr
        print(f"Gradio {gr.__version__} est trop ancien : mise à jour…")
    except ImportError:
        print("Installation de Gradio (environ une minute)…")
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-U", "gradio>=5"], check=False)
    importlib.invalidate_caches()
    import gradio as gr
    if int(gr.__version__.split(".")[0]) < 5:
        raise RuntimeError("Redémarrez la session (Exécution → Redémarrer la session), puis relancez le notebook.")
    return gr

def _doc_labels():
    return [f"{d.id} · {d.short}" for d in LIB.documents]

def _library_summary():
    return (f"**{len(LIB.documents)} document(s)** · {len(LIB.passages)} passages · 🌐 {', '.join(LIB.languages)} · "
            f"seuil {f'{LIB.threshold:.2f}'.replace('.', ',')} · "
            + (f"modèle **{LLM.name}**" if LLM.available else "mode extractif"))

def build_app():
    gr = _ensure_gradio()
    import inspect
    theme = gr.themes.Soft(primary_hue="emerald", secondary_hue="amber", neutral_hue="slate")
    css = ".gradio-container {max-width: 1250px !important} footer {display: none !important}"
    blocks_args = {"title": "Assistant documentaire RAG"}
    blocks_params = inspect.signature(gr.Blocks.__init__).parameters
    if "theme" in blocks_params:
        blocks_args.update(theme=theme, css=css)
    chat_params = inspect.signature(gr.Chatbot.__init__).parameters
    chat_args = {"height": 470, "label": "Conversation"}
    if "type" in chat_params:
        chat_args["type"] = "messages"
    if "feedback_options" in chat_params:
        chat_args["feedback_options"] = ["Like", "Dislike"]

    def web_answer(question, messages, memory_state, docs, k, threshold, guardrails, memory):
        messages = list(messages or [])
        if not question or not question.strip():
            return "", messages, memory_state, gr.update()
        ids = [int(x.split(" · ")[0]) for x in (docs or [])]
        if not ids:
            messages.append({"role": "assistant", "content": "Cochez au moins un document."})
            return "", messages, memory_state, gr.update()
        hist = [{"standalone_question": h["standalone_question"], "answer": h["answer"],
                 "language": h.get("language")} for h in (memory_state or [])]
        r = LIB.answer(question.strip(), history=hist if memory else None, docs=ids, k=int(k),
                       threshold=float(threshold), guardrails=bool(guardrails), memory=bool(memory))
        text = r["answer"]
        if r.get("queries"):
            text = "_🌐 Également recherché sous la forme : " + " · ".join(f"[{l}] {q}" for l, q in r["queries"].items()) + "_\n\n" + text
        if r["rewritten"]:
            text = f"_🧠 Compris comme : {r['standalone_question']}_\n\n" + text
        if r["reason"]:
            text += f"\n\n_Motif : {r['reason']}_"
        if not r["refused"]:
            text += "\n\n" + "\n".join(f"**[{i}]** {p.citation}" for i, (p, _) in enumerate(r["passages"], 1))
        messages += [{"role": "user", "content": question.strip()}, {"role": "assistant", "content": text}]
        memory_state = (memory_state or []) + [{"question": r["question"], "standalone_question": r["standalone_question"],
                                                "answer": r["answer"], "citations": r["citations"],
                                                "language": r["language"]}]
        return "", messages, memory_state, f"{_CSS}{LIB.card_html(r)}"

    def rate(data: gr.LikeData, memory_state):
        index = data.index[0] if isinstance(data.index, (list, tuple)) else data.index
        turns = memory_state or []
        position = int(index) // 2
        if not turns or position >= len(turns):
            return "Retour non enregistré : échange introuvable."
        useful = data.liked if isinstance(data.liked, bool) else str(data.liked).lower() in ("true", "like")
        record_feedback(turns[position], useful, channel="web")
        return f"Merci ! Retour enregistré ({len(FEEDBACK)} au total)."

    def add_pdfs(files, docs):
        if not files:
            return gr.update(), _library_summary(), ""
        paths = [f if isinstance(f, str) else getattr(f, "name", str(f)) for f in files]
        added, errors = LIB.load(paths)
        LIB.index()
        choices = _doc_labels()
        ticked = list(docs or []) + [f"{d.id} · {d.short}" for d in added]
        message = f"✅ {len(added)} document(s) ajouté(s)." + "".join(f"\n⚠️ {n}: {e}" for n, e in errors)
        return gr.update(choices=choices, value=[c for c in choices if c in ticked]), _library_summary(), message

    with gr.Blocks(**blocks_args) as app:
        gr.HTML('<div style="background:linear-gradient(120deg,#00553A 0%,#00A86A 100%);border-radius:14px;'
                'padding:16px 22px;color:#fff;font-family:\'Segoe UI\',Calibri,sans-serif;">'
                '<div style="color:#F5C242;font-size:11px;font-weight:700;letter-spacing:2.5px;">ASSISTANT DOCUMENTAIRE · RAG</div>'
                '<div style="font-size:24px;font-weight:800;">💬 Interrogez vos rapports PDF</div>'
                '<div style="font-size:13px;color:#E6F6EE;">Des réponses sourcées, des pages surlignées, et un refus quand '
                'l’information ne figure pas dans les documents.</div></div>')
        memory_state = gr.State([])
        with gr.Row():
            with gr.Column(scale=1, min_width=300):
                summary = gr.Markdown(_library_summary())
                docs = gr.CheckboxGroup(choices=_doc_labels(), value=_doc_labels(), label="Documents interrogés")
                upload = gr.File(label="Ajouter des PDF", file_count="multiple", file_types=[".pdf"])
                add_button = gr.Button("📥 Ajouter et indexer")
                add_message = gr.Markdown("")
                with gr.Accordion("⚙️ Paramètres", open=False):
                    k = gr.Slider(1, 8, value=TOP_K, step=1, label="Passages fournis")
                    threshold = gr.Slider(0.0, 1.0, value=float(LIB.threshold), step=0.01, label="Seuil de confiance")
                    guardrails = gr.Checkbox(value=True, label="Garde-fous actifs")
                    memory = gr.Checkbox(value=True, label="Mémoire de conversation")
            with gr.Column(scale=2):
                chat = gr.Chatbot(**chat_args)
                with gr.Row():
                    question = gr.Textbox(placeholder="Posez une question sur les documents, dans n’importe quelle langue…", show_label=False,
                                          scale=5, autofocus=True)
                    send_button = gr.Button("Envoyer", variant="primary", scale=1)
                gr.Examples(examples=[[s] for s in SUGGESTIONS], inputs=[question], label="Suggestions")
                with gr.Row():
                    new_chat = gr.Button("🔄 Nouvelle conversation", size="sm")
                    feedback_message = gr.Markdown("_Utilisez 👍 / 👎 sous chaque réponse pour donner votre avis._")
                with gr.Accordion("📄 Sources de la dernière réponse", open=True):
                    sources = gr.HTML("<i>Les sources apparaîtront ici.</i>")
        inputs = [question, chat, memory_state, docs, k, threshold, guardrails, memory]
        outputs = [question, chat, memory_state, sources]
        send_button.click(web_answer, inputs, outputs)
        question.submit(web_answer, inputs, outputs)
        new_chat.click(lambda: ([], [], "<i>Les sources apparaîtront ici.</i>"), None, [chat, memory_state, sources])
        add_button.click(add_pdfs, [upload, docs], [docs, summary, add_message])
        if hasattr(chat, "like"):
            chat.like(rate, [memory_state], [feedback_message])
    app._launch_args = {} if "theme" in blocks_params else {"theme": theme, "css": css}
    return app

def launch_app(app):
    import inspect
    auth = (WEB_USERNAME, WEB_PASSWORD) if WEB_PASSWORD else None
    options = dict(share=SHARE_LINK, auth=auth, prevent_thread_lock=True, quiet=True, show_error=True,
                   inbrowser=(ENV["name"] == "local"), inline=(ENV["name"] == "colab"), **app._launch_args)
    params = inspect.signature(app.launch).parameters
    options = {k: v for k, v in options.items() if k in params}
    try:
        app.launch(**options)
    except Exception as error:
        if options.get("share"):
            print(f"⚠️ Lien public indisponible ({type(error).__name__}) : lancement en local.")
            options["share"] = False
            app.launch(**options)
        else:
            raise
    public_url = getattr(app, "share_url", None)
    local_url = getattr(app, "local_url", None) or "http://127.0.0.1:7860"
    lines = [f'<div style="font-size:14px;margin:4px 0;">🔗 <b>Lien public :</b> <a href="{public_url}" target="_blank">{public_url}</a></div>'
             if public_url else "",
             f'<div style="font-size:14px;margin:4px 0;">💻 <b>Adresse locale :</b> <code>{local_url}</code></div>',
             f'<div style="font-size:13px;margin:4px 0;">🔒 Accès : ' + (f'identifiant <code>{WEB_USERNAME}</code> et mot de passe définis'
             if auth else f'<span style="color:{C["brick"]};"><b>aucun mot de passe</b></span>') + '</div>']
    display(HTML(f'{_CSS}<div class="rg rg-card" style="border-left:6px solid {C["green"]};">'
                 f'<div class="rg-title">🌐 Application web en cours d’exécution</div>{"".join(lines)}'
                 f'<div class="rg-note">Pour l’arrêter : exécutez <code>APP.close()</code>. Elle s’arrête aussi à la fin de la session.</div></div>'))
    if public_url and not auth:
        callout("Le lien est public et n’a <b>aucun mot de passe</b> : définissez <code>WEB_PASSWORD</code> avant de le partager.", "warning")

if LAUNCH_WEB_APP:
    APP = build_app()
    launch_app(APP)
else:
    callout("Application web non lancée (<code>LAUNCH_WEB_APP = False</code>).", "info")
````

````python
config = {
    "documents": [{"title": d.title, "file": d.filename, "pages": d.pages, "language": d.lang, "scanned_pages": d.scanned_pages,
                   "elements": dict(Counter(e.type for e in d.elements))} for d in LIB.documents],
    "document_language": DOCUMENT_LANGUAGE, "languages": dict(LIB.language_counts),
    "question_language": QUESTION_LANGUAGE, "answer_language": ANSWER_LANGUAGE,
    "translate_queries": TRANSLATE_QUERIES, "cross_lingual_concepts": len(CROSS_LINGUAL_LEXICON), "passages": len(LIB.passages), "ocr": USE_OCR, "max_chars": MAX_CHARS,
    "chunk_strategy": CHUNK_STRATEGY, "chunk_level": CHUNKING_LEVELS[CHUNK_STRATEGY][0],
    "reranking": LIB.reranker.mode, "reranking_level": LIB.reranker.name, "search_units": len(LIB.units),
    "vocabulary": len(VOCABULARY_FORMS),
    "acronyms_learned": len(ACRONYMS), "cache": bool(USE_CACHE), "contents_lines_dropped":
        sum(d.info.get("contents lines dropped", 0) for d in LIB.documents),
    "top_k": TOP_K, "semantic_search": LIB.semantic.name, "threshold": LIB.threshold,
    "calibrated_threshold": LIB.calibrated_threshold, "provider": LLM.name, "model": LLM.model,
    "model_calls": LLM.calls, "environment": ENV["name"], "date": time.strftime("%Y-%m-%d %H:%M"),
}
(OUTPUT_DIR / "configuration.json").write_text(json.dumps(config, ensure_ascii=False, indent=2), encoding="utf-8")
pd.DataFrame([{"id": p.id, "document": p.doc, "section": p.section, "pages": ",".join(map(str, p.pages)),
               "kind": p.kind, "language": p.lang, "text": p.text} for p in LIB.passages]).to_csv(
    OUTPUT_DIR / "passages.csv", index=False, encoding="utf-8-sig")
LIB.calibration.to_csv(OUTPUT_DIR / "calibration.csv", index=False, encoding="utf-8-sig")
if TEST_ROWS:
    TEST.drop(columns="_ok").to_csv(OUTPUT_DIR / "test_set.csv", index=False, encoding="utf-8-sig")
if HISTORY:
    export()
if FEEDBACK:
    fb = pd.DataFrame(FEEDBACK)
    show_table(fb[["timestamp", "channel", "useful", "question", "comment"]].assign(
                   useful=lambda d: d["useful"].map({True: "👍", False: "👎"})),
               title=f"🗳️ {len(fb)} retour(s) · {fb['useful'].mean() * 100:.0f} % jugés utiles",
               note="Les réponses marquées « À revoir » sont à ajouter au jeu de tests, avec la bonne réponse.")

archive = shutil.make_archive(str(ENV["root"] / "rag_deliverables"), "zip", OUTPUT_DIR)
files_list = sorted(p.name for p in OUTPUT_DIR.iterdir())
show_table(pd.DataFrame({"Fichier": ["rag_deliverables.zip"] + files_list}),
           title=f"📦 {len(files_list)} livrable(s)",
           note="configuration.json : paramètres · passages.csv : texte indexé · calibration.csv et test_set.csv : "
                "éléments de validation · conversation_* : conversations exportées · user_feedback.csv : retours 👍 / 👎.")
if ENV["name"] == "colab":
    if COPY_TO_DRIVE:
        from google.colab import drive
        drive.mount("/content/drive")
        target = Path("/content/drive/MyDrive/rag_deliverables")
        target.mkdir(parents=True, exist_ok=True)
        shutil.copy(archive, target / f"rag_deliverables_{time.strftime('%Y%m%d_%H%M')}.zip")
        callout(f"Archive copiée sur Google Drive, dans <code>{target}</code>.", "success")
    if DOWNLOAD_OUTPUTS:
        from google.colab import files
        files.download(archive)
        callout("Téléchargement de l’archive lancé : les fichiers de la session Colab sont supprimés à sa fermeture.", "warning")
elif ENV["name"] == "kaggle":
    callout("Les livrables sont dans <code>/kaggle/working</code> : cliquez sur <b>Save Version</b> pour les conserver.", "info")
else:
    callout(f"Archive disponible sur votre disque : <code>{archive}</code>.", "success")
````

### Limites de cet assistant

- **Les mises en page complexes** (plusieurs colonnes, tableaux sans bordures, notes de bas de page) sont extraites imparfaitement. Vérifiez l'étape B2 sur vos documents.
- **Les graphiques ne sont pas lus**, seulement leurs titres et légendes.
- **Le mode extractif** cite des phrases ou des lignes de tableau : il ne sait ni **calculer** ni **classer** (« quelle région a le taux de chômage le plus élevé ? »). Ces questions nécessitent un modèle de langage.
- **La mémoire à base de règles** reconnaît les changements de région, de sexe, de trimestre, d'année, de milieu, d'âge et d'indicateur. Au-delà, un modèle est nécessaire.
- **La vérification des chiffres est littérale** : un chiffre calculé par le modèle (une différence, une moyenne) est signalé comme absent, même s'il est juste.
- **Ressources linguistiques.** Les mots vides et les racines existent pour l'anglais, le français, l'espagnol, le portugais, l'allemand, l'italien, l'arabe et (mots vides uniquement) le swahili ; le glossaire et le lexique interlangue couvrent l'anglais, le français, l'espagnol et le portugais. Une autre langue (kinyarwanda, wolof, amharique…) est tout de même lue, mais sans racines ni mots vides : appuyez-vous sur un modèle d'embeddings multilingue (`multilingual-e5`, `bge-m3`) et sur un modèle de langage pour la traduction.
- **La détection de la langue** repose sur les mots grammaticaux fréquents. Elle est fiable sur les documents et sur les questions ordinaires, moins sur une question de deux mots (« Région Côte ? ») : une telle question hérite de la langue de l'échange précédent, et `ask("…", lang="fr")` permet de la forcer.
- **Une phrase mêlant deux langues** (« Taux de chômage / Unemployment rate: 7.9 % ») est analysée dans les deux, mais pas scindée. Les colonnes parallèles (anglais à gauche, français à droite sur la même page) sont lues dans l'ordre de lecture par PyMuPDF et se séparent en général proprement ; sinon, vérifiez B10.
- **Les traductions sont reconnues** par leurs chiffres ou, avec des embeddings multilingues, par leur sens. Deux passages sans chiffres et sans embeddings ne sont pas fusionnés : les deux peuvent apparaître. Baissez ou augmentez `TWIN_SIMILARITY` si trop peu ou trop de passages sont fusionnés.
- **Le mode extractif ne sait pas traduire** : un passage anglais est cité en anglais, avec la mention *(source en anglais)*. Pour des réponses dans la langue de la question, il faut un modèle.
- **Sans `sentence-transformers`**, la recherche sémantique est une approximation hors ligne qui rapproche des graphies, pas des sens. Elle ne peut pas faire le lien entre une question formulée avec les mots de vos utilisateurs et un rapport rédigé avec ceux de l'institution. C'est la principale limite de la configuration par défaut : `INSTALL_EMBEDDINGS = True` la lève.
- **La tolérance aux fautes d'orthographe a un angle mort voulu** : un mot que les documents n'emploient jamais n'est corrigé que s'il en existe un très proche. « manioc » reste inconnu — et c'est précisément ce qui permet à l'assistant de refuser une question hors sujet.
- **Sur les rapports très longs**, la plupart des passages contiennent tous les mots d'une question : le score de confiance sature alors près de 1,00 et mesure la couverture, non la précision. Le classement s'appuie sur la densité des termes et sur la recherche fusionnée, et le seuil est plafonné à 0,70.
- **La vérification finale reste humaine** : l'assistant donne la page, un statisticien vérifie avant toute diffusion.

### Améliorations possibles

| # | Amélioration | Principe | Effort | Gain |
|:-:|---|---|:-:|:-:|
| 1 | **Embeddings neuronaux** | `INSTALL_EMBEDDINGS = True` : la recherche sémantique cesse de rapprocher des graphies et commence à rapprocher des sens | Faible | ⭐⭐⭐ |
| 2 | **Extraction avancée** | Outils spécialisés (Docling, Marker, PyMuPDF4LLM) pour les colonnes et les tableaux complexes | Moyen | ⭐⭐⭐ |
| 3 | **Reclassement par cross-encoder** | `RERANKING = "cross-encoder"` : un modèle lit ensemble la question et le passage (B4) | Faible | ⭐⭐⭐ |
| 3b | **Juge de fidélité** | Un second appel au modèle vérifie que chaque phrase est étayée par sa source (au-delà des chiffres) | Moyen | ⭐⭐⭐ |
| 4 | **Métadonnées** | Filtrer par année, thème ou type de publication, pas seulement par fichier | Moyen | ⭐⭐ |
| 5 | **Lecture des graphiques** | Décrire les figures avec un modèle multimodal | Moyen | ⭐⭐ |
| 6 | **Bibliothèque persistante** | L'extraction et les embeddings sont déjà mis en cache ; aller plus loin, c'est stocker l'index dans une base de données vectorielle | Moyen | ⭐⭐ |
| 7 | **Tableau de bord des retours** | Suivre la part de réponses utiles et de questions refusées, semaine après semaine | Faible | ⭐⭐ |
| 8 | **Déploiement durable** | Héberger l'application (serveur interne, Hugging Face Spaces) au lieu d'un lien temporaire | Élevé | ⭐⭐⭐ |
| 9 | **Droits d'accès** | Réserver certains documents à certains utilisateurs | Élevé | Indispensable en production |

### ✅ Point de contrôle

<details><summary><b>1. Pourquoi afficher « Compris comme… » ?</b></summary>
Parce que la mémoire peut se tromper. L'utilisateur voit la question réellement recherchée et peut reformuler si elle ne correspond pas à son intention.
</details>

<details><summary><b>2. Pourquoi limiter le nombre de passages issus d'un même document ?</b></summary>
Pour qu'une question de comparaison reçoive des extraits de plusieurs documents, plutôt que trois passages du même rapport.
</details>

<details><summary><b>3. Pourquoi un mot de passe sur l'application web ?</b></summary>
Un lien public est accessible à quiconque le possède : sans mot de passe, n'importe qui peut interroger vos documents ou en déposer de nouveaux.
</details>

<details><summary><b>4. Une mauvaise réponse avec le bon passage : que faut-il corriger ?</b></summary>
La <b>génération</b> : le prompt ou le modèle. Avec un passage hors sujet : la <b>recherche</b> (extraction, glossaire, seuil).
</details>

<details><summary><b>5. Que signifie « chiffre absent des sources » ?</b></summary>
Un nombre de la réponse ne figure dans aucun des passages fournis : soit le modèle l'a inventé ou arrondi, soit il l'a calculé. Dans les deux cas, vérifiez-le dans les sources citées.
</details>

### 🛠️ Dépannage

| Symptôme | Cause probable | Solution |
|---|---|---|
| L'interface C1 ne s'affiche pas | Widgets non pris en charge | Utilisez `ask("…")` ou l'application web |
| « Lien public indisponible » | Pas d'Internet ou réseau filtré | Kaggle : *Settings → Internet: On* ; sinon, utilisez l'adresse locale |
| « Redémarrez la session » après l'installation de Gradio | Une version plus ancienne était déjà chargée | *Exécution → Redémarrer la session*, puis *Tout exécuter* |
| Un PDF déposé n'apparaît pas | Fichier protégé ou numérisé | Lisez le message sous le bouton ; activez l'OCR si nécessaire |
| `HTTP Error 403: Forbidden` sur une adresse web | Le site refuse les téléchargements automatisés (fréquent sur les portails institutionnels, y compris afdb.org depuis Colab) | Les autres documents se chargent quand même. Téléchargez le PDF dans votre navigateur, puis `UPLOAD_PDF = True` sur Colab, ou indiquez le chemin du fichier dans `DOCUMENTS` |
| « Compris comme » est erroné | La mémoire à base de règles est limitée | Posez une question complète, ou activez un modèle (Groq, Ollama) |
| « Chiffre absent des sources » sur une réponse juste | Chiffre calculé ou reformaté par le modèle | Vérifiez le calcul dans les sources citées |
| Trop de refus | Seuil trop élevé | Baissez-le dans les paramètres de l'interface ; enrichissez le glossaire |
| Une question sur votre propre domaine renvoie un sujet voisin | Les mots de vos utilisateurs ne sont pas ceux du rapport | Ajoutez la paire au `GLOSSARY`, ou activez les embeddings neuronaux. `explain("votre question")` montre exactement où la chaîne a dévié |
| Une faute de frappe fait échouer une question sans bruit | Rare : le mot est corrigé et affiché sur la carte (« orthographe : chomage → chômage ») | Si la correction est fausse, le mot est probablement légitime mais absent des documents |
| Quel découpage utiliser ? | — | Lancez `benchmark()` (B3) sur vos propres documents plutôt que de deviner |
| L'indexation d'un rapport de 80 pages est lente la première fois | La détection des tableaux s'effectue page par page | Normal : 5 à 10 secondes par rapport. Les exécutions suivantes réutilisent `rag_cache` |
| Les paramètres d'extraction ont changé mais rien n'a bougé | Une extraction en cache a été réutilisée | La clé du cache inclut les paramètres ; en cas de doute, supprimez le dossier `rag_cache` |
| Une question sur une section renvoie la table des matières | `DROP_TABLE_OF_CONTENTS = False` | Remettez-le à `True` et réindexez |
| Un document se voit attribuer la mauvaise langue | Document très court, ou surtout composé de tableaux | Vérifiez la colonne *Langue* en B10 ; forcez-la avec `DOCUMENT_LANGUAGE = "fr"` si toute la collection est dans une seule langue, ou modifiez `LIB.documents[i].lang` puis `LIB.index()` |
| Une question posée dans une autre langue que celle des documents est refusée | Ni modèle ni embeddings multilingues : seul le lexique relie les langues | Connectez un modèle (Groq, Ollama) pour que la question soit traduite, ou ajoutez les mots manquants à `CROSS_LINGUAL_LEXICON` et exécutez `build_lexicon_index()` |
| Le même rapport en deux langues renvoie les deux versions | Normal : chaque version est un document | Décochez-en une dans le filtre de documents des interfaces, ou gardez `PREFER_QUESTION_LANGUAGE = True` (par défaut) |
| Des pages numérisées en arabe, en swahili… ressortent vides | Langue OCR manquante | Ajoutez son code Tesseract à `OCR_LANGUAGE` (`"ara"`, `"swa"`…) |
| « Rate limit reached » | Limite par minute de Groq | Le notebook attend puis réessaie ; pour un usage intensif, passez à Ollama |
| Ollama très lent | Pas de GPU | Activez un GPU T4 *avant* l'exécution, ou utilisez `qwen2.5:1.5b` |

<b style="color:#F5C242;">À retenir.</b> Un assistant documentaire utile <b>lit</b> bien ses PDF, <b>cite</b> le document et la page, <b>comprend</b> les questions de relance — et <b>refuse</b> lorsqu'il ne sait pas.

### Des idées de projets construits sur cet assistant

| Projet | Documents | Utilisateurs | Premier test |
|---|---|---|---|
| **Assistance méthodologique** pour les équipes de l'IPC ou des comptes nationaux | Manuels, notes méthodologiques, procédures internes | Compilateurs, nouveaux agents | 20 questions posées par de nouvelles recrues |
| **Questions-réponses sur les publications** pour les usagers et la presse | Publications statistiques, calendrier de diffusion, politique de révision | Service de communication, journalistes | Les questions reçues par le service d'assistance au dernier trimestre |
| **Appui à la classification de la BdP** — complète la classification des transactions par apprentissage automatique | Chapitres du MBP6, guide d'établissement, nomenclatures nationales | Compilateurs de la BdP | Transactions ambiguës classées avec une règle citée |
| **Assistant pour les déclarations de politique monétaire** | Déclarations, procès-verbaux, rapports sur l'inflation | Économistes, chercheurs | « Qu'a dit le comité sur l'inflation alimentaire ? » |
| **Assistant bilingue** (français · anglais) | Les mêmes rapports dans les deux langues | Équipes régionales | Même question dans les deux langues, même réponse |

### Ce qu'une équipe projet doit consigner

- **Problème et utilisateurs** — qui pose quelles questions aujourd'hui, et combien de temps prend la réponse.
- **Documents** — liste, langue, nombre de pages, scannés ou non, niveau de confidentialité → **choix du modèle** (Ollama pour les documents non publiés).
- **Évaluation** — un jeu de test de 10 à 20 questions à réponse connue, **dont des questions à refuser** ; le résultat du banc d'essai ; la part de réponses correctes et sourcées.
- **Livrables** — les fichiers exportés en partie D (configuration, passages, résultats des tests, retours des utilisateurs).
- **Étape suivante** — qui l'hébergerait, qui tient les documents à jour, comment les droits d'accès sont gérés.
