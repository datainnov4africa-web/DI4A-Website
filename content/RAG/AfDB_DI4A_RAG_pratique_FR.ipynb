### 🎛️ Réglages

**Parcours express :** seule la première section compte. Conservez `DOCUMENTS = ["exemple"]` pour un premier essai, puis remplacez-la par vos propres fichiers.

`DOCUMENTS` accepte une **liste** qui mélange :
- `"exemple"` — trois rapports fictifs fournis avec le notebook ;
- des **fichiers** — `"reports/labour_2025.pdf"` ;
- des **dossiers** — `"reports/"` (tous les PDF qu’ils contiennent) ;
- des **adresses web** — `"https://…/report.pdf"`.

`LANGUE_DOCUMENTS` indique au moteur de recherche comment lire le texte : `"en"` pour des documents en anglais, `"fr"` pour des documents en français.

**Les longs rapports institutionnels** (50 à 200 pages) sont le cas normal dans un institut de statistique. Trois réglages comptent pour eux :

| Réglage | Effet |
|---|---|
| `UTILISER_CACHE = True` | L’extraction et les embeddings sont enregistrés dans `cache_rag`. Le premier traitement d’un rapport de 80 pages coûte quelques secondes de détection des tableaux ; tous les traitements suivants dans la même session — ou après un redémarrage — réutilisent le cache. Supprimez le dossier pour forcer une nouvelle extraction. |
| `SUPPRIMER_SOMMAIRE = True` | Supprime les lignes du type « Chapitre 3 …… 42 », les listes de figures et les numéros de page isolés, qui deviendraient sinon des passages ne répondant à rien. |
| `MAX_PAGES_PAR_DOCUMENT` | `0` lit tout. Mettez 30 ou 40 pour essayer rapidement un rapport très long avant de l’indexer en entier. |

> ⚠️ **Les adresses web ne fonctionnent pas toujours.** De nombreux portails institutionnels — `afdb.org` parmi eux — refusent les téléchargements qui ne proviennent pas d’un navigateur et répondent **HTTP 403**. Le notebook essaie des en-têtes imitant un navigateur, puis `curl`, et si les deux échouent il **ignore ce document et poursuit avec les autres**, en vous indiquant lequel a échoué.
>

````python
# @title 🦙 Installation d'Ollama (si demandée) { display-mode: "form" }
def _ollama_running():
    try:
        urllib.request.urlopen("http://localhost:11434/api/tags", timeout=2)
        return True
    except Exception:
        return False

def installer_ollama(model):
    if _ollama_running():
        print("Ollama tourne déjà.")
    elif platform.system() != "Linux":
        encadre("L'installation automatique n'est disponible que sur Linux (Colab, Kaggle). Sur macOS ou Windows, "
                "installez Ollama depuis <code>ollama.com/download</code>, puis réexécutez cette cellule.", "warning")
        return False
    else:
        if not shutil.which("ollama"):
            print("1/3 · Installation d'Ollama (1 à 2 min)…")
            if shutil.which("apt-get"):
                subprocess.run("apt-get install -y -qq zstd pciutils > /dev/null 2>&1 || "
                               "(apt-get update -qq > /dev/null 2>&1 && apt-get install -y -qq zstd pciutils > /dev/null 2>&1)",
                               shell=True)
            subprocess.run("curl -fsSL https://ollama.com/install.sh | sh > ollama_install.log 2>&1", shell=True)
        if not shutil.which("ollama"):
            encadre("L'installation a échoué : voir le fichier <code>ollama_install.log</code>.", "error")
            return False
        print("2/3 · Démarrage du service…")
        subprocess.Popen(["ollama", "serve"], stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL,
                         start_new_session=True)
        for _ in range(30):
            if _ollama_running():
                break
            time.sleep(1)
    print(f"3/3 · Téléchargement du modèle {model} (quelques minutes la première fois)…")
    result = subprocess.run(["ollama", "pull", model], capture_output=True, text=True)
    if result.returncode != 0:
        encadre(f"Téléchargement du modèle impossible : <code>{_html.escape(result.stderr[-300:])}</code>", "error")
        return False
    encadre(f"Ollama est prêt avec le modèle <b>{model}</b> · essais illimités."
            + ("" if ENV["gpu"] else " <br>⚠️ Sans GPU, chaque réponse peut demander de 10 à 60 secondes."), "success")
    return True

if INSTALLER_OLLAMA:
    installer_ollama(MODELE_OLLAMA)
else:
    print("Installation d'Ollama non demandée (INSTALLER_OLLAMA = False).")
````

````python
# @title 🤖 Connexion au modèle { display-mode: "form" }
import urllib.error

def _secret(name):
    value = os.environ.get(name)
    if value:
        return value
    if ENV["name"] == "colab":
        try:
            from google.colab import userdata
            return userdata.get(name)
        except Exception:
            return None
    if ENV["name"] == "kaggle":
        try:
            from kaggle_secrets import UserSecretsClient
            return UserSecretsClient().get_secret(name)
        except Exception:
            return None
    return None

# Cloudflare (devant api.groq.com) refuse la signature par défaut « Python-urllib/3.x »
# avec « HTTP 403: error code: 1010 ». Un User-Agent explicite règle le problème.
API_USER_AGENT = "AfDB-DI4A-RAG-notebook/1.1"

def _poster(url, payload, headers, timeout=120):
    request = urllib.request.Request(url, data=json.dumps(payload).encode("utf-8"),
                                     headers={"Content-Type": "application/json", "Accept": "application/json",
                                              "User-Agent": API_USER_AGENT, **headers})
    with urllib.request.urlopen(request, timeout=timeout) as response:
        return json.loads(response.read().decode("utf-8"))

class ClientLLM:
    """Une seule interface pour Ollama, Groq, Anthropic ou OpenAI.
    Température 0, respect des quotas : pause entre les appels, nouvel essai après un HTTP 429."""

    MIN_PAUSE = {"groq": 2.2, "anthropic": 0.5, "openai": 0.5, "ollama": 0.0}

    def __init__(self, choice="auto"):
        self.found = {"ollama": _ollama_running(),
                      "groq": bool(_secret("GROQ_API_KEY")),
                      "anthropic": bool(_secret("ANTHROPIC_API_KEY")),
                      "openai": bool(_secret("OPENAI_API_KEY"))}
        order = ["ollama", "groq", "anthropic", "openai"] if choice == "auto" else [choice]
        self.name = next((p for p in order if self.found.get(p)), "none")
        self.model = {"ollama": MODELE_OLLAMA, "groq": MODELE_GROQ, "anthropic": MODELE_ANTHROPIC,
                      "openai": MODELE_OPENAI}.get(self.name, "—")
        self.calls, self.waited, self._last = 0, 0.0, 0.0

    @property
    def available(self):
        return self.name != "none"

    def _call(self, system, message, max_tokens):
        if self.name == "ollama":
            d = _poster("http://localhost:11434/api/chat",
                      {"model": self.model, "stream": False, "options": {"temperature": 0},
                       "messages": [{"role": "system", "content": system},
                                    {"role": "user", "content": message}]}, {}, timeout=600)
            return d["message"]["content"]
        if self.name == "groq":
            payload = {"model": self.model, "temperature": 0, "max_completion_tokens": max_tokens + 600,
                       "messages": [{"role": "system", "content": system},
                                    {"role": "user", "content": message}]}
            if "gpt-oss" in self.model:
                payload["reasoning_effort"] = "low"          # raisonnement court : moins de tokens consommés
            d = _poster("https://api.groq.com/openai/v1/chat/completions", payload,
                      {"Authorization": f"Bearer {_secret('GROQ_API_KEY')}"})
            return d["choices"][0]["message"].get("content") or ""
        if self.name == "anthropic":
            d = _poster("https://api.anthropic.com/v1/messages",
                      {"model": self.model, "max_tokens": max_tokens, "temperature": 0,
                       "system": system, "messages": [{"role": "user", "content": message}]},
                      {"x-api-key": _secret("ANTHROPIC_API_KEY"), "anthropic-version": "2023-06-01"})
            return "".join(b.get("text", "") for b in d["content"])
        if self.name == "openai":
            d = _poster("https://api.openai.com/v1/chat/completions",
                      {"model": self.model, "temperature": 0, "max_completion_tokens": max_tokens,
                       "messages": [{"role": "system", "content": system},
                                    {"role": "user", "content": message}]},
                      {"Authorization": f"Bearer {_secret('OPENAI_API_KEY')}"})
            return d["choices"][0]["message"]["content"]
        raise RuntimeError("Aucun fournisseur de modèle disponible.")

    def generate(self, system, message, max_tokens=400, attempts=5):
        for attempt in range(attempts):
            wait = self.MIN_PAUSE.get(self.name, 0) - (time.time() - self._last)
            if wait > 0:
                time.sleep(wait)
            try:
                self._last = time.time()
                text = self._call(system, message, max_tokens)
                self.calls += 1
                return re.sub(r"<think>.*?</think>", "", text, flags=re.S).strip()
            except urllib.error.HTTPError as error:
                if error.code == 429 and attempt < attempts - 1:
                    delay = min(float(error.headers.get("retry-after") or 2 ** (attempt + 2)), 90)
                    print(f"⏳ Quota momentanément atteint ({self.name}) : nouvel essai dans {delay:.0f} s…")
                    self.waited += delay
                    time.sleep(delay)
                    continue
                detail = error.read().decode("utf-8", "ignore")[:200]
                if error.code == 403 and "1010" in detail:
                    detail += " — Cloudflare a bloqué la requête (signature du client). Vérifiez le réseau ou utilisez Ollama."
                raise RuntimeError(f"HTTP {error.code}: {detail}") from error

# ── D'où vient chaque clé, et à quoi elle ressemble ────────────────────────────────────────
CLES_API = {"groq":      ("GROQ_API_KEY",      "gsk_",    "console.groq.com/keys (gratuit, sans carte bancaire)"),
            "anthropic": ("ANTHROPIC_API_KEY", "sk-ant-", "console.anthropic.com"),
            "openai":    ("OPENAI_API_KEY",    "sk-",     "platform.openai.com/api-keys")}

def demander_cle_api(provider):
    """Demande la clé du fournisseur, saisie masquée. Entrée seule passe l'étape.
    La clé reste en mémoire pour cette session : elle n'est jamais écrite dans le notebook."""
    variable, prefix, where = CLES_API[provider]
    import getpass
    try:
        key = getpass.getpass(f"🔑 Clé {provider.capitalize()} ({prefix}…) — à créer sur {where} — "
                              f"Entrée seule pour passer : ").strip()
    except Exception:
        return None                                  # exécution automatisée (nbconvert) : on passe
    if not key:
        encadre("Étape passée : l'assistant <b>recopiera</b> des phrases au lieu de rédiger les réponses. "
                "Vous pourrez coller une clé plus tard, dans la cellule suivante.", "info")
        return None
    if not key.startswith(prefix):
        encadre(f"Une clé {provider.capitalize()} commence par <code>{prefix}</code> — rien n'a été enregistré.", "error")
        return None
    os.environ[variable] = key
    return ClientLLM(provider)

SANS_MODELE = FOURNISSEUR in ("none", "aucun")
LLM = ClientLLM("none" if SANS_MODELE else FOURNISSEUR)

tableau(pd.DataFrame([
    {"Fournisseur": "Ollama (local)", "Coût": "gratuit · illimité", "Détecté": "✅" if LLM.found["ollama"] else "—",
     "Comment l'activer": "INSTALLER_OLLAMA = True, ou Ollama installé en local"},
    {"Fournisseur": "Groq", "Coût": "gratuit · 1 000 req./jour", "Détecté": "✅" if LLM.found["groq"] else "—",
     "Comment l'activer": "secret ou variable GROQ_API_KEY"},
    {"Fournisseur": "Anthropic", "Coût": "payant", "Détecté": "✅" if LLM.found["anthropic"] else "—",
     "Comment l'activer": "secret ou variable ANTHROPIC_API_KEY"},
    {"Fournisseur": "OpenAI", "Coût": "payant", "Détecté": "✅" if LLM.found["openai"] else "—",
     "Comment l'activer": "secret ou variable OPENAI_API_KEY"},
]), title="Fournisseurs de modèle détectés")

# ── Un fournisseur est demandé mais aucune clé n'est visible : on la demande, saisie masquée ────
if not LLM.available and not SANS_MODELE and DEMANDER_CLE:
    wanted = FOURNISSEUR if FOURNISSEUR in CLES_API else "groq"
    encadre(f"<b>{wanted.capitalize()}</b> est demandé, mais aucune clé n'a été trouvée dans les secrets ni dans "
            f"l'environnement. Collez-la ci-dessous : la saisie est masquée et la clé n'est jamais écrite dans "
            f"le notebook.", "warning")
    LLM = demander_cle_api(wanted) or LLM

if LLM.available:
    try:
        start = time.time()
        test = LLM.generate("Tu réponds par un seul mot.", "Réponds seulement : OK", max_tokens=20)
        encadre(f"Fournisseur retenu : <b>{LLM.name}</b> · modèle <code>{LLM.model}</code> · "
                f"test de connexion réussi en {time.time() - start:.1f} s (réponse : « {_html.escape(test[:30])} »).", "success")
    except Exception as error:
        encadre(f"Le fournisseur <b>{LLM.name}</b> a été détecté mais le test a échoué : "
                f"<code>{_html.escape(str(error)[:220])}</code><br>Le notebook bascule en <b>mode extractif</b>. "
                "Vérifiez la clé, le nom du modèle et l'accès à Internet.", "error")
        LLM.name, LLM.model = "none", "—"
        if _ollama_running():
            secours = ClientLLM("ollama")
            try:
                secours.generate("Tu réponds par un seul mot.", "Réponds seulement : OK", max_tokens=20)
                LLM = secours
                encadre(f"Ollama tourne ici : bascule sur <b>ollama</b> · <code>{LLM.model}</code>.", "success")
            except Exception:
                pass
else:
    encadre("Aucun modèle connecté : le notebook fonctionne en <b>mode extractif</b> — toutes les étapes "
            "fonctionnent, mais les réponses sont <b>recopiées</b> des documents au lieu d'être rédigées, et "
            "l'assistant ne sait ni compter, ni comparer, ni synthétiser. La cellule suivante permet de coller "
            "une clé à tout moment.", "warning")
````

   - **Ou collez-la dans le champ ci-dessous** pour cette session uniquement.

Un jeton Hugging Face se crée de la même façon, sur **`huggingface.co/settings/tokens`** (un jeton *Read* suffit), et peut être conservé sous le nom `HF_TOKEN`.

````python
# @title ⚙️ T2 · Diagnostic et extraction structurée { display-mode: "form" }
from dataclasses import dataclass, field

def diagnostiquer(doc):
    rows = []
    for page in doc:
        chars = len(page.get_text().strip())
        images = len(page.get_images(full=True))
        status = "texte" if chars >= 50 else ("scannée" if images else "vide")
        rows.append({"Page": page.number + 1, "Caractères": chars, "Images": images, "État": status})
    return pd.DataFrame(rows)

@dataclass
class Element:
    type: str           # "heading" (titre), "paragraph", "table" ou "ocr"
    text: str
    page: int
    box: tuple          # rectangle (x0, y0, x1, y1) sur la page
    level: int = 0      # niveau de titre (1 = le plus important)
    section: str = ""   # titre le plus proche
    path: str = ""      # fil d'Ariane complet des titres, par exemple « 2. Stratégie › 2.4 Domaines »

APPEL_DE_NOTE = re.compile(r"(?<=[a-z\)\u201d\"])\.(\d{1,3})(?=\s+[A-Z\u201c\"])")

def _nfkc(t):
    return unicodedata.normalize("NFKC", t)

def lignes_de_page(page):
    lines = []
    for block in page.get_text("dict")["blocks"]:
        if block.get("type") != 0:
            continue
        for line in block["lines"]:
            spans = [s for s in line["spans"] if s["text"].strip()]
            if not spans:
                continue
            lines.append({"text": _nfkc("".join(s["text"] for s in line["spans"])).strip(),
                          "size": round(max(s["size"] for s in spans), 1),
                          "bold": any((s["flags"] & 16) or "Bold" in s.get("font", "") for s in spans),
                          "box": pymupdf.Rect(line["bbox"]), "block": block["number"]})
    return lines

def _signature(text):
    return re.sub(r"\d+", "#", text.lower()).strip()

def _tableaux(page):
    if not EXTRAIRE_TABLEAUX:
        return []
    with contextlib.redirect_stdout(io.StringIO()):
        found = page.find_tables().tables
    result = []
    for t in found:
        rows = [[_nfkc((c or "").replace("\n", " ")).strip() for c in r] for r in t.extract()]
        headers = [_nfkc((c or "").replace("\n", " ")).strip() for c in t.header.names]
        if rows and rows[0] == headers:
            rows = rows[1:]
        if rows and len(headers) >= 2:
            result.append({"box": pymupdf.Rect(t.bbox), "headers": headers, "rows": rows})
    return result

def lineariser(table, caption):
    sentences = []
    for row in table["rows"]:
        if not any(row):
            continue
        values = "; ".join(f"{table['headers'][j]}: {row[j]}" for j in range(1, len(row))
                           if j < len(table["headers"]) and row[j])
        sentences.append(f"{row[0]} — {values}.")
    return (f"{caption}\n" if caption else "") + "\n".join(sentences)

def ocr_de_page(page):
    installer("pytesseract", "pytesseract")
    if not shutil.which("tesseract") and shutil.which("apt-get") and ENV["name"] in ("colab", "kaggle"):
        print("Installation de Tesseract (environ une minute)…")
        subprocess.run("apt-get install -y -qq tesseract-ocr tesseract-ocr-fra > /dev/null 2>&1 || "
                       "(apt-get update -qq > /dev/null 2>&1 && apt-get install -y -qq tesseract-ocr tesseract-ocr-fra > /dev/null 2>&1)",
                       shell=True)
    import pytesseract
    from PIL import Image
    installed = set(pytesseract.get_languages(config=""))
    language = "+".join(l for l in LANGUE_OCR.split("+") if l in installed) or "eng"
    if language != LANGUE_OCR:
        print(f"⚠️ Langue OCR « {LANGUE_OCR} » indisponible : utilisation de « {language} ».")
    pix = page.get_pixmap(dpi=300)
    image = Image.frombytes("RGB", (pix.width, pix.height), pix.samples)
    text = _nfkc(pytesseract.image_to_string(image, lang=language))
    return [re.sub(r"\s*\n\s*", " ", p).strip() for p in re.split(r"\n\s*\n", text) if len(p.strip()) > 3]

LEGENDE = re.compile(r"^(Table|Figure|Chart|Graph|Tableau|Graphique)\b")
LIGNE_SOMMAIRE = re.compile(r"\.{4,}\s*\d{1,4}\s*$|\u2026{2,}")      # "Chapter 3 ......... 42"
NUMERO_DE_PAGE_SEUL = re.compile(r"^(page\s*)?[ivxlcdm\d]{1,6}$", re.I)
TITRE_NUMEROTE = re.compile(r"^\d{1,2}(\.\d{1,2}){0,3}[.)]?\s+\S")

def extraire_structure(doc):
    last = doc.page_count if not MAX_PAGES_PAR_DOCUMENT else min(doc.page_count, MAX_PAGES_PAR_DOCUMENT)
    pages = {p.number + 1: lignes_de_page(p) for p in doc if p.number < last}
    with_text = [n for n, l in pages.items() if l]
    # 1. en-têtes et pieds de page répétés
    repeated = set()
    if SUPPRIMER_ENTETES and len(with_text) >= 2:
        counts = Counter()
        for n in with_text:
            height = doc[n - 1].rect.height
            counts.update({_signature(l["text"]) for l in pages[n]
                           if l["box"].y1 < height * 0.1 or l["box"].y0 > height * 0.9})
        repeated = {s for s, c in counts.items() if c >= max(2, 0.5 * len(with_text))}
    # 2. taille du corps de texte et tailles des titres
    weights = Counter()
    for l in (l for n in with_text for l in pages[n]):
        weights[l["size"]] += len(l["text"])
    body = weights.most_common(1)[0][0] if weights else 10
    heading_sizes = sorted({l["size"] for n in with_text for l in pages[n]
                            if l["size"] >= body * 1.2 and len(l["text"]) <= 120}, reverse=True)
    level_of = {s: i + 1 for i, s in enumerate(heading_sizes)}
    def is_running_line(line, page):
        if _signature(line["text"]) in repeated:
            return True
        # document d'une seule page : rien ne se répète, on se fie à la position et à la taille
        height = page.rect.height
        in_margin = line["box"].y1 < height * 0.08 or line["box"].y0 > height * 0.92
        return SUPPRIMER_ENTETES and len(with_text) < 2 and in_margin and line["size"] < body * 0.85 and len(line["text"]) < 90
    elements, removed, dropped_toc = [], 0, 0
    for n in sorted(pages):
        page = doc[n - 1]
        if not pages[n]:
            if UTILISER_OCR and page.get_images():
                for paragraph in ocr_de_page(page):
                    elements.append(Element("ocr", paragraph, n, tuple(page.rect)))
            continue
        tables = _tableaux(page)
        pieces, current = [], None                     # pieces: (y, element, block[, table])
        for l in pages[n]:
            if is_running_line(l, page) or NUMERO_DE_PAGE_SEUL.match(l["text"]):
                removed += 1
                continue
            if SUPPRIMER_SOMMAIRE and LIGNE_SOMMAIRE.search(l["text"]):
                dropped_toc += 1                      # table of contents, list of figures…
                continue
            if any(l["box"].intersects(t["box"]) for t in tables):
                continue
            numbered = (TITRE_NUMEROTE.match(l["text"]) and len(l["text"]) <= 90
                        and l["size"] >= body * 0.98 and l["bold"])
            if l["size"] in level_of or numbered:
                if current: pieces.append(current); current = None
                if pieces and pieces[-1][1].type == "heading" and pieces[-1][2] == l["block"]:
                    heading = pieces[-1][1]
                    heading.text += " " + l["text"]
                    heading.box = tuple(pymupdf.Rect(heading.box) | l["box"])
                else:
                    pieces.append((l["box"].y0, Element("heading", l["text"], n, tuple(l["box"]),
                                                        level_of.get(l["size"], len(heading_sizes) + 1)), l["block"]))
                continue
            if current and current[2] == l["block"]:
                el = current[1]
                if el.text.endswith("-") and l["text"][:1].islower():
                    el.text = el.text[:-1] + l["text"]          # recolle les mots coupés en fin de ligne
                else:
                    el.text += " " + l["text"]
                el.box = tuple(pymupdf.Rect(el.box) | l["box"])
            else:
                if current: pieces.append(current)
                current = (l["box"].y0, Element("paragraph", l["text"], n, tuple(l["box"])), l["block"])
        if current: pieces.append(current)
        for t in tables:
            pieces.append((t["box"].y0, Element("table", "", n, tuple(t["box"])), None, t))
        pieces.sort(key=lambda p: p[0])
        for p in pieces:
            el = p[1]
            if el.type == "table":
                previous = elements[-1] if elements else None
                table = p[3]
                if (previous is not None and previous.type == "table" and previous.page == n - 1
                        and len(previous.table["headers"]) == len(table["headers"])):
                    # suite du tableau de la page précédente : sa ligne d'« en-tête » est une ligne de données
                    previous.table["rows"] += [table["headers"]] + table["rows"]
                    previous.text = lineariser(previous.table, previous.caption)
                    previous.extra.append((n, tuple(table["box"])))
                    continue
                caption = ""
                if previous and previous.type == "paragraph" and LEGENDE.match(previous.text):
                    caption = previous.text
                    elements.pop()
                el.text = lineariser(table, caption)
                el.table, el.caption, el.extra = table, caption, []
            elements.append(el)
    # 3. rattache chaque élément à sa section et à son fil d'Ariane complet
    section, stack = "", {}
    for i, el in enumerate(elements):
        if el.type == "heading":
            document_title = i == 0 and el.level == 1 and len(heading_sizes) > 1
            section = "" if document_title else el.text
            if not document_title:
                stack = {lvl: t for lvl, t in stack.items() if lvl < el.level}
                stack[el.level] = el.text
        else:
            el.text = APPEL_DE_NOTE.sub(". ", el.text)      # « …élevé.5 Le secteur » → « …élevé. Le secteur »
        el.section = section if el.type != "ocr" else f"Page {el.page} (texte OCR)"
        el.path = " › ".join(stack[lvl] for lvl in sorted(stack)) or el.section
    return elements, {"en-têtes retirés": removed, "lignes de sommaire écartées": dropped_toc,
                      "taille du corps": body, "tailles des titres": heading_sizes, "pages lues": len(pages)}
print("✅ T2 prêt.")
````

````python
# @title ⚙️ T3 · Découpage en passages { display-mode: "form" }
@dataclass
class Passage:
    id: int
    doc_id: int
    doc: str             # titre court du document
    section: str
    text: str
    index_text: str      # texte utilisé pour la recherche (avec un en-tête de contexte)
    boxes: list          # [(page, rectangle), …]
    kind: str = "text"

    @property
    def pages(self):
        return sorted({p for p, _ in self.boxes})

    path: str = ""       # fil d'Ariane des titres du passage
    parent_id: int = -1  # passage de contexte auquel cette fenêtre de recherche appartient

    @property
    def pages_label(self):
        pages = self.pages
        return f"p. {pages[0]}" if len(pages) == 1 else f"pp. {pages[0]}–{pages[-1]}"

    @property
    def citation(self):
        return f"{self.doc} · {self.section or 'début du document'} · {self.pages_label}"

def _phrases(text):
    return [s.strip() for s in re.split(r"(?<=[.!?;])\s+(?=[A-ZÀ-Ý0-9«•–-])", text) if s.strip()]

def decouper_passages(document, start=0, max_chars=1000, overlap=2, min_chars=280):
    passages = []
    def add(section, units, kind="text"):
        text = ("\n" if kind == "table" else " ").join(u[0] for u in units)
        boxes = list(dict.fromkeys((u[1], u[2]) for u in units))
        if len(text.strip()) < 40 or text.strip().lower() in seen:
            return                                   # bribes et doublons : du bruit, jamais une réponse
        seen.add(text.strip().lower())
        breadcrumb = path or section
        # un passage trop court est collé au précédent plutôt que laissé seul, sans contexte
        if (passages and len(text) < min_chars and passages[-1].kind == kind == "text"
                and passages[-1].section == section and len(passages[-1].text) + len(text) <= max_chars * 1.4):
            previous = passages[-1]
            previous.text += " " + text
            previous.index_text += " " + text
            previous.boxes += [b for b in boxes if b not in previous.boxes]
            return
        header = f"{document.title} — {breadcrumb}" if breadcrumb else document.title
        passages.append(Passage(start + len(passages), document.id, document.short, section, text,
                                f"{header}. {text}", boxes, kind, breadcrumb))
    seen = set()
    buffer, section, path = [], None, ""
    def flush():
        nonlocal buffer
        i = 0
        while i < len(buffer):
            batch, length, j = [], 0, i
            while j < len(buffer) and (not batch or length + len(buffer[j][0]) <= max_chars):
                batch.append(buffer[j]); length += len(buffer[j][0]) + 1; j += 1
            add(section, batch)
            if j >= len(buffer):
                break
            i = max(j - overlap, i + 1)
        buffer = []
    for el in document.elements:
        if el.type == "heading":
            continue
        if el.section != section:
            flush()
            section, path = el.section, el.path
        if el.type == "table":
            flush()
            lines = el.text.split("\n")
            batch = []
            for line in lines:
                if batch and sum(len(x) for x in batch) + len(line) > max_chars:
                    add(section, [(x, el.page, el.box) for x in ([lines[0]] if batch[0] != lines[0] else []) + batch], "table")
                    batch = []
                batch.append(line)
            if batch:
                add(section, [(x, el.page, el.box) for x in ([lines[0]] if batch[0] != lines[0] else []) + batch], "table")
            for p in passages:
                if p.kind == "table" and p.boxes and p.boxes[0] == (el.page, el.box):
                    p.boxes += [b for b in getattr(el, "extra", []) if b not in p.boxes]
            continue
        buffer += [(s, el.page, el.box) for s in _phrases(el.text)]
    flush()
    return passages
# ── Les cinq niveaux de découpage, de la référence naïve à celui qui mesure le mieux ──
NIVEAUX_DECOUPAGE = {
    "fixed":     ("1 · Blocs fixes",
                  "Découpe tous les TAILLE_MAX caractères, sans égard aux phrases, aux sections ni aux tableaux."),
    "sentences": ("2 · Fenêtres de phrases",
                  "Des phrases entières, N à la fois, en glissant. Rien n'est coupé en plein milieu, mais un "
                  "passage ne sait plus de quelle section il vient."),
    "sections":  ("3 · Passages structurels",
                  "Suit les titres repérés par l'extracteur ; chaque tableau forme un passage à part et le fil "
                  "d'Ariane des titres accompagne le texte."),
    "hybrid":    ("4 · Deux granularités",
                  "Indexe les passages structurels ET de petites fenêtres de phrases : la recherche gagne en "
                  "précision, le lecteur reçoit toujours un passage entier."),
    "smart":     ("5 · Deux granularités, tableaux et contexte",
                  "Ajoute chaque ligne de tableau comme unité à part entière et joint les passages voisins de "
                  "la section à la réponse."),
}

def _morceaux_document(document):
    """Chaque fragment de texte du document, dans l'ordre de lecture, avec sa page et sa zone."""
    return [(el.text, el.page, el.box) for el in document.elements if el.text.strip()]

def creer_passages(document, start=0):
    """Les niveaux 1 et 2 ignorent la structure du rapport ; les niveaux 3 à 5 s'en servent."""
    if STRATEGIE_DECOUPAGE not in ("fixed", "sentences"):
        return decouper_passages(document, start, TAILLE_MAX, CHEVAUCHEMENT_PHRASES, TAILLE_MIN)
    passages, pieces = [], _morceaux_document(document)
    def add(text, boxes):
        text = " ".join(text.split())
        if len(text) < 40:
            return
        passages.append(Passage(start + len(passages), document.id, document.short, "", text,
                                f"{document.title}. {text}", list(dict.fromkeys(boxes)), "text", ""))
    if STRATEGIE_DECOUPAGE == "fixed":
        overlap, buffer, boxes = int(TAILLE_MAX * 0.12), "", []
        for text, page, box in pieces:
            boxes.append((page, box))
            for word in text.split():
                if len(buffer) + len(word) + 1 > TAILLE_MAX:     # coupe en plein milieu d'une phrase
                    add(buffer, boxes)
                    buffer, boxes = buffer[-overlap:], boxes[-2:]
                buffer += " " + word
        add(buffer, boxes)
    else:                                                        # "sentences"
        units = [(s, page, box) for text, page, box in pieces for s in (_phrases(text) or [text])]
        for i in range(0, max(1, len(units) - FENETRE_PHRASES + 1), PAS_FENETRE):
            window = units[i:i + FENETRE_PHRASES]
            add(" ".join(u[0] for u in window), [(u[1], u[2]) for u in window])
    return passages

def creer_unites_recherche(passages, strategy=None, size=None, stride=None):
    """Les niveaux 4 et 5 indexent une seconde granularité, plus fine : de petites fenêtres — et, au
    niveau 5, des lignes de tableau — qui renvoient par `parent_id` au passage à donner à lire."""
    strategy = strategy or STRATEGIE_DECOUPAGE
    size, stride = size or FENETRE_PHRASES, stride or PAS_FENETRE
    for p in passages:
        p.parent_id = p.id
    if strategy not in ("hybrid", "smart"):
        return list(passages)
    units, next_id = list(passages), max((p.id for p in passages), default=-1) + 1
    for p in passages:
        if p.kind == "table":
            if strategy != "smart":
                continue
            lines = p.text.split("\n")
            caption = lines[0] if lines and " — " not in lines[0] else ""
            for line in (lines[1:] if caption else lines):
                if len(line) < 15:
                    continue
                text = f"{caption} {line}".strip()                 # une ligne, avec sa légende
                units.append(Passage(next_id, p.doc_id, p.doc, p.section, text,
                                     f"{p.path or p.section}. {text}", p.boxes, "table", p.path, p.id))
                next_id += 1
            continue
        sentences = _phrases(p.text) or [p.text]
        for i in range(0, max(1, len(sentences) - size + 1), stride):
            text = " ".join(sentences[i:i + size])
            if len(text) < 60:
                continue
            units.append(Passage(next_id, p.doc_id, p.doc, p.section, text,
                                 f"{p.path or p.section}. {text}", p.boxes, p.kind, p.path, p.id))
            next_id += 1
    return units

print(f"✅ T3 prêt · niveau de découpage {NIVEAUX_DECOUPAGE[STRATEGIE_DECOUPAGE][0]}.")
````

````python
# @title ⚙️ T4 · Recherche { display-mode: "form" }
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD
from sklearn.preprocessing import Normalizer

LANGUE = LANGUE_DOCUMENTS if LANGUE_DOCUMENTS in ("en", "fr") else "en"

# --- Ressources linguistiques (les mots vides des deux langues sont toujours ignorés) ----------
MOTS_VIDES = set("""a an the and or of in on at to for from by with is are was were be been being what which who
whom whose when where why how does do did can could will would shall should may might it its this that these those
there their they them his her he she we you i as than then so such not no nor about into over under also any all
each per much many get give tell me please document documents report reports say says said information show between stand stands standing acronym abbreviation abbreviations initials mean means called number numbers proposed propose proposes presented described mentioned according
le la les l un une des de du d et en au aux est sont ete ce cet cette ces qui que quoi quel quelle quels quelles
comment combien quand ou il elle se sa son ses leur leurs ne pas plus par pour sur dans avec y t n s qu etait
sera peut peuvent fait faire lors entre apres avant tout toute tous toutes ainsi dont tres rapport dit indique selon
informations donne donnent parle concernant sujet ils elles eux lui chez aussi meme""".split())

SUFFIXES = {
    "en": sorted(["ational", "ations", "ation", "ments", "ment", "ness", "ings", "ing", "ities", "ity", "ies",
                  "ied", "ers", "er", "ed", "es", "ly", "al", "s", "e"], key=len, reverse=True),
    "fr": sorted(["ements", "ement", "ations", "ation", "itions", "ition", "ances", "ance", "ences", "ence",
                  "euses", "euse", "eurs", "eur", "ites", "ite", "iques", "ique", "ives", "ive", "ifs", "if",
                  "elles", "elle", "iels", "iel", "aux", "ants", "ant", "ents", "ent", "ees", "ee", "es", "er",
                  "age", "s", "e", "x"], key=len, reverse=True),
}

GLOSSAIRES = {
    "en": {
        # mots des usagers               -> mots employés dans les publications
        "inhabitants": "population resident people", "residents": "population resident",
        "how many people": "population", "people live": "population",
        "jobless": "unemployment unemployed", "out of work": "unemployment unemployed",
        "young people": "youth 15 24", "youth": "young 15 24",
        "woman": "women", "man": "men", "female": "women", "male": "men",
        "city": "urban", "cities": "urban", "towns": "urban", "countryside": "rural", "villages": "rural",
        "proportion": "share rate", "percentage": "share rate",
        "microdata": "anonymised files", "individual data": "microdata anonymised",
        "get access": "available access", "obtain": "available access",
        "released": "release published", "come out": "released published", "published": "released",
        "when": "published released date", "publication": "published released",
        "did not respond": "non-response reweighting", "non-response": "reweighting",
        "first quarter": "q1", "second quarter": "q2", "third quarter": "q3", "fourth quarter": "q4",
        "change": "down up", "changed": "down up", "evolve": "down up", "trend": "down up",
        "informal": "social protection contract", "wage": "income", "salary": "income", "earnings": "income",
        "sample": "dwellings drawn areas", "definition": "glossary", "what is meant by": "glossary",
        "next census": "census scheduled", "household size": "average size household",
    },
    "fr": {
        "habitants": "population residente personnes", "gens": "population personnes",
        "vivent": "population residente", "combien de personnes": "population",
        "ville": "urbain", "villes": "urbain", "campagne": "rural",
        "sans emploi": "chomage chomeur", "chomeurs": "chomage", "travail": "emploi",
        "jeunes": "15 24 ans", "proportion": "taux part", "pourcentage": "taux part",
        "microdonnees": "fichiers donnees individuelles anonymises",
        "chercheur": "chercheurs", "obtenir": "acceder accessibles acces",
        "sort": "publie publies diffusion paraitra", "sortent": "publie publies diffusion paraitra",
        "sortira": "paraitra publie", "publication": "publies paraitra", "quand": "date publies paraitra",
        "pas repondu": "non reponse reponderation", "non-reponse": "reponderation",
        "premier trimestre": "t1", "deuxieme trimestre": "t2", "second trimestre": "t2",
        "troisieme trimestre": "t3", "quatrieme trimestre": "t4",
        "evolue": "hausse baisse", "evolution": "hausse baisse", "varie": "hausse baisse",
        "informel": "informelle protection sociale contrat", "salaire": "revenu",
        "echantillon": "logements tirage zones", "definition": "glossaire", "signifie": "glossaire",
        "prochain recensement": "recensement programme", "taille des menages": "taille moyenne menage",
    },
}
GLOSSAIRE = GLOSSAIRES[LANGUE]          # 🧪 ajoutez les vôtres : GLOSSAIRE["cadre"] = "modele maturite diagnostic"

def sans_accents(text):
    return "".join(c for c in unicodedata.normalize("NFD", text) if unicodedata.category(c) != "Mn")

def raciner(word):
    if any(ch.isdigit() for ch in word):
        return word
    for _ in range(2 if LANGUE == "en" else 1):
        for suffix in SUFFIXES[LANGUE]:
            if word.endswith(suffix) and len(word) - len(suffix) >= 4:
                word = word[: -len(suffix)]
                break
    return word

def analyser(text):
    t = sans_accents(text.lower())
    if LANGUE == "en":
        t = re.sub(r"(?<=\d),(?=\d{3}\b)", "", t)                    # 18,452,300 -> 18452300
        tokens = re.findall(r"[a-z0-9]+(?:\.\d+)?", t)
    else:
        t = re.sub(r"(?<=\d)[\s\u00a0\u202f](?=\d{3}\b)", "", t)      # 18 452 300 -> 18452300
        tokens = re.findall(r"[a-z0-9]+(?:,\d+)?", t)
    return [raciner(w) for w in tokens if w not in MOTS_VIDES and len(w) > 1]

# --- Tolérance orthographique -------------------------------------------------------------------
# Les usagers font des fautes de frappe : « chômge », « developement », « statisitque ».
# Un mot que les documents ne contiennent jamais est sinon traité comme « inconnu, donc décisif »,
# ce qui coule toute la question. Ces fonctions le rapprochent du mot le plus proche réellement employé.
import difflib

FORMES_VOCABULAIRE = {}      # racine -> la forme telle qu'elle est écrite dans les documents
INDEX_TRIGRAMMES = {}        # trigramme de caractères -> racines qui le contiennent (recherche rapide)

def _trigrammes(word):
    padded = f"  {word} "
    return {padded[i:i + 3] for i in range(len(padded) - 2)}

def construire_vocabulaire(passages):
    """Collecte les formes de surface des documents et les indexe par trigramme de caractères."""
    forms = {}
    for p in passages:
        for word in re.findall(r"[^\W\d_]{3,}", sans_accents(p.text.lower())):
            if word not in MOTS_VIDES:
                forms.setdefault(raciner(word), Counter())[word] += 1
    FORMES_VOCABULAIRE.clear()
    FORMES_VOCABULAIRE.update({root: c.most_common(1)[0][0] for root, c in forms.items()})
    INDEX_TRIGRAMMES.clear()
    for root in FORMES_VOCABULAIRE:
        for gram in _trigrammes(root):
            INDEX_TRIGRAMMES.setdefault(gram, set()).add(root)

def terme_le_plus_proche(root, minimum=0.86):
    """Terme du document le plus proche d'un mot que la collection ne contient pas, ou None.
    Le seuil est volontairement strict : un mot réellement étranger (« maïs », « hôpital ») doit
    rester inconnu, car c'est lui qui rend une question hors sujet refusable."""
    if len(root) < 4 or root in FORMES_VOCABULAIRE:
        return None
    candidates = Counter()
    for gram in _trigrammes(root):
        for other in INDEX_TRIGRAMMES.get(gram, ()):
            candidates[other] += 1
    best, best_score = None, minimum
    for other, _ in candidates.most_common(40):
        if abs(len(other) - len(root)) > 2:
            continue
        score = difflib.SequenceMatcher(None, root, other).ratio()
        if score > best_score:
            best, best_score = other, score
    return best

def corriger_orthographe(question):
    """Remplace chaque mot mal orthographié par le mot le plus proche réellement employé.
    Renvoie la question corrigée et la liste des corrections, toujours affichée."""
    corrections = []
    def replace(match):
        word = match.group(0)
        plain = sans_accents(word.lower())
        if plain in MOTS_VIDES or raciner(plain) in FORMES_VOCABULAIRE:
            return word
        near = terme_le_plus_proche(raciner(plain))
        if not near:
            return word
        corrections.append((word, FORMES_VOCABULAIRE[near]))
        return FORMES_VOCABULAIRE[near]
    return (re.sub(r"[^\W\d_]{4,}", replace, question) if FORMES_VOCABULAIRE else question), corrections

SIGLES = {}          # appris des documents eux-mêmes : « INS » -> « Institut national de la statistique »
_SIGLE_DANS_TEXTE = re.compile(r"([A-Z][\w’'\-]*(?:\s+[a-z\w’'\-]+){1,5}(?:\s+[A-Z][\w’'\-]*)?)\s*\(([A-Z]{2,6})s?\)")
_SIGLE_DANS_LISTE = re.compile(r"(?:^|\n|;|—|\|)\s*([A-Z]{2,6})\s*(?:[—:\-–]|\s{2,})\s*([A-Z][^.;\n|]{4,60})")

def construire_sigles(passages):
    """Les rapports institutionnels sont pleins de sigles : on les apprend des documents."""
    found = {}
    for p in passages:
        for expansion, acronym in _SIGLE_DANS_TEXTE.findall(p.text):
            found.setdefault(acronym, Counter())[expansion.strip()] += 1
        for acronym, expansion in _SIGLE_DANS_LISTE.findall(p.text):
            found.setdefault(acronym, Counter())[expansion.strip()] += 1
    return {a: counts.most_common(1)[0][0] for a, counts in found.items()
            if counts.most_common(1)[0][1] >= 1 and len(counts.most_common(1)[0][0].split()) >= 2}

def enrichir_question(question):
    q = " " + re.sub(r"[^a-z0-9-]+", " ", sans_accents(question.lower())) + " "
    added = [(key, terms) for key, terms in GLOSSAIRE.items() if f" {key} " in q]
    for acronym, expansion in SIGLES.items():
        if f" {acronym.lower()} " in q:
            added.append((acronym, expansion))
    return question + " " + " ".join(t for _, t in added), added

def concepts(question):
    """Chaque terme de la question, avec les synonymes que le glossaire lui attache."""
    _, added = enrichir_question(question)
    synonyms = {}
    for key, terms in added:
        for word in analyser(key) or [raciner(sans_accents(key.split()[-1]))]:
            synonyms.setdefault(word, set()).update(analyser(terms))
    return [(t, {t} | synonyms.get(t, set())) for t in dict.fromkeys(analyser(question))]

class IndexBM25:
    """BM25 sur une matrice creuse termes × documents : reste rapide sur des milliers de passages."""

    def __init__(self, passages, k1=1.5, b=0.75):
        from scipy.sparse import csr_matrix
        self.passages, self.k1, self.b = passages, k1, b
        self.docs = [analyser(p.index_text) for p in passages]
        n = len(self.docs)
        self.vocabulary, rows, cols, values = {}, [], [], []
        for i, d in enumerate(self.docs):
            for term, f in Counter(d).items():
                j = self.vocabulary.setdefault(term, len(self.vocabulary))
                rows.append(i); cols.append(j); values.append(f)
        self.matrix = csr_matrix((values, (rows, cols)), shape=(n, max(1, len(self.vocabulary))), dtype=np.float32)
        lengths = np.array([len(d) for d in self.docs], dtype=np.float32)
        self.avg_len = max(1e-9, float(lengths.mean()) if n else 1.0)
        self.norm = (self.k1 * (1 - self.b + self.b * lengths / self.avg_len)).astype(np.float32)
        df = np.asarray((self.matrix > 0).sum(axis=0)).ravel()
        idf = np.log(1 + (n - df + 0.5) / (df + 0.5))
        self.idf_vector = idf.astype(np.float32)
        self.idf = {t: float(idf[j]) for t, j in self.vocabulary.items()}
        self.idf_max = max(self.idf.values()) if self.idf else 1.0

    def rechercher(self, question, k=30):
        terms = [t for t in analyser(enrichir_question(question)[0]) if t in self.vocabulary]
        if not terms:
            return []                                  # aucun mot commun : aucun résultat                                  # aucun mot commun : aucun résultat
        columns = [self.vocabulary[t] for t in dict.fromkeys(terms)]
        frequencies = self.matrix[:, columns].toarray()
        weights = self.idf_vector[columns] * (self.k1 + 1)
        scores = (weights * frequencies / (frequencies + self.norm[:, None])).sum(axis=1)
        order = [i for i in np.argsort(-scores)[:k] if scores[i] > 0]
        return [(self.passages[i], float(scores[i])) for i in order]

import hashlib
_CACHE_EMBEDDINGS = {}
class IndexSemantique:
    def __init__(self, passages):
        self.passages, self.mode, self.error = passages, None, None
        texts = [p.index_text for p in passages]
        if UTILISER_EMBEDDINGS:
            try:
                if INSTALLER_EMBEDDINGS:
                    installer("sentence_transformers", "sentence-transformers")
                from sentence_transformers import SentenceTransformer
                if MODELE_EMBEDDINGS not in _CACHE_EMBEDDINGS:
                    _CACHE_EMBEDDINGS[MODELE_EMBEDDINGS] = SentenceTransformer(MODELE_EMBEDDINGS)
                self.model = _CACHE_EMBEDDINGS[MODELE_EMBEDDINGS]
                key = hashlib.sha1((MODELE_EMBEDDINGS + "|" + "|".join(texts)).encode("utf-8")).hexdigest()[:16]
                cached = DOSSIER_CACHE / f"embeddings_{key}.npy"
                if UTILISER_CACHE and cached.exists():
                    self.matrix = np.load(cached)
                else:
                    self.matrix = self.model.encode(texts, normalize_embeddings=True, batch_size=32,
                                                    show_progress_bar=len(texts) > 200)
                    if UTILISER_CACHE:
                        np.save(cached, self.matrix)
                self.mode, self.name = "embeddings", f"Embeddings · {MODELE_EMBEDDINGS.split('/')[-1]}"
            except Exception as error:
                self.error = type(error).__name__
        if self.mode is None:                      # secours hors ligne : analyse sémantique latente
            self.vectorizer = TfidfVectorizer(analyzer="char_wb", ngram_range=(3, 5),
                                              strip_accents="unicode", sublinear_tf=True)
            m = self.vectorizer.fit_transform(texts)
            self.svd = TruncatedSVD(n_components=max(1, min(96, m.shape[0] - 1, m.shape[1] - 1)), random_state=0)
            self.normalizer = Normalizer()
            self.matrix = self.normalizer.fit_transform(self.svd.fit_transform(m))
            self.mode, self.name = "lsa", "Sémantique approchée · LSA hors ligne"

    def _encoder(self, text):
        if self.mode == "embeddings":
            return self.model.encode([text], normalize_embeddings=True)
        return self.normalizer.transform(self.svd.transform(self.vectorizer.transform([text])))

    def rechercher(self, question, k=30):
        scores = (self.matrix @ self._encoder(enrichir_question(question)[0]).T).ravel()
        order = np.argsort(-scores)[:k]
        return [(self.passages[i], float(scores[i])) for i in order]

def fusion_rrf(rankings, k=60, top=12):
    scores, objects = {}, {}
    for ranking in rankings:
        for rank, (p, _) in enumerate(ranking, start=1):
            scores[p.id] = scores.get(p.id, 0.0) + 1.0 / (k + rank)
            objects[p.id] = p
    order = sorted(scores, key=scores.get, reverse=True)[:top]
    return [(objects[i], scores[i]) for i in order]

# ── Quel type de réponse la question attend (utilisé par le reclasseur et l'extracteur) ──
TYPES_QUESTION = {
    "count": re.compile(r"\b(how many|how much|number of|combien|nombre de)\b", re.I),
    "date": re.compile(r"\b(when|what date|which year|quand|en quelle annee)\b", re.I),
    "list": re.compile(r"\b(list|name the|which \w+ (are|were|have|took)|what are the|quels|quelles)\b", re.I),
    "definition": re.compile(r"\b(what is an?|what are an?|define|definition of|stands? for|what does \w+ mean|"
                             r"signifie|qu'est-ce)\b", re.I),
}   # l'ordre compte : le premier motif qui correspond l'emporte
NOMBRE_DANS_TEXTE = re.compile(r"\b\d[\d.,]*\b|\b(one|two|three|four|five|six|seven|eight|nine|ten|eleven|twelve|"
                            r"deux|trois|quatre|cinq|six|sept|huit|neuf|dix|onze|douze)\b", re.I)
RENVOI = re.compile(r"\b(figure|table|tableau|graph|chart|annex|annexe|section|page|p\.)\s*\d+", re.I)
ENUMERATION = re.compile(r"(?:,|;|\band\b|\bet\b)[^,;]{2,70}(?:,|;|\band\b|\bet\b)", re.I | re.S)
UNE_ANNEE = re.compile(r"\b(19|20)\d{2}\b")

def type_de_question(question):
    for name, pattern in TYPES_QUESTION.items():
        if pattern.search(question):
            return name
    return "other"

def porte_un_nombre(sentence):
    """Une vraie quantité, pas un renvoi du type « Figure 8 » ou « voir tableau 3 »."""
    return bool(NOMBRE_DANS_TEXTE.search(RENVOI.sub(" ", sentence)))

# ── Les quatre niveaux de reclassement ──
NIVEAUX_RECLASSEMENT = {
    "none":          ("0 · Aucun", "Conserve l'ordre issu de la fusion des deux recherches."),
    "coverage":      ("1 · Couverture de la question",
                      "Part des termes pondérés de la question présents dans le passage."),
    "smart":         ("2 · Couverture, densité et type de réponse",
                      "À couverture égale, le type de réponse départage (une question « combien » préfère un "
                      "passage portant une quantité, une question « liste » un passage portant une énumération), "
                      "puis la densité des termes."),
    "cross-encoder": ("3 · Cross-encoder neuronal",
                      "Un modèle lit la question et le passage ensemble et note la paire. Le plus précis, et le "
                      "plus lent : il demande sentence-transformers."),
}

class Reclasseur:
    def __init__(self, bm25, mode=None):
        self.bm25 = bm25
        self.mode = mode or RECLASSEMENT
        self.cross = None
        if self.mode == "cross-encoder":
            try:
                installer("sentence_transformers", "sentence-transformers")
                from sentence_transformers import CrossEncoder
                self.cross = CrossEncoder(MODELE_CROSS_ENCODER)
            except Exception as error:
                print(f"⚠️ Cross-encoder indisponible ({type(error).__name__}) : reclassement « smart » par défaut.")
                self.mode = "smart"

    @property
    def name(self):
        return NIVEAUX_RECLASSEMENT[self.mode][0]

    def bonus_type_reponse(self, kind, passage):
        """Ce passage porte-t-il le *type* de chose que la question demande ?"""
        if kind == "count":
            return 1 if porte_un_nombre(passage.text) else 0
        if kind == "list":
            return 1 if ENUMERATION.search(passage.text) else 0
        if kind == "date":
            return 1 if UNE_ANNEE.search(passage.text) else 0
        return 0

    def poids(self, term):
        return self.bm25.idf.get(term, self.bm25.idf_max * 1.5)   # terme inconnu = signal fort

    def couverture(self, question, passage):
        present = set(analyser(passage.index_text))
        items = concepts(question)
        total = sum(self.poids(t) for t, _ in items)
        return sum(self.poids(t) for t, g in items if g & present) / total if total else 0.0

    def reclasser(self, question, candidates, density=None, kind="other"):
        """Le score renvoyé est toujours la couverture de la question, pour que le seuil de confiance
        garde le même sens ; seul l'ORDRE dépend du niveau choisi."""
        scores = [self.couverture(question, p) for p, _ in candidates]
        if self.mode == "none":
            order = range(len(candidates))
        elif self.mode == "coverage":
            order = sorted(range(len(candidates)), key=lambda i: (-scores[i], i))
        elif self.cross is not None:
            pairs = [(question, " ".join(p.text.split())[:900]) for p, _ in candidates]
            neural = self.cross.predict(pairs, show_progress_bar=False)
            order = sorted(range(len(candidates)), key=lambda i: -float(neural[i]))
        else:
            # sur un long rapport, beaucoup de passages couvrent toute la question : à 0,02 près,
            # le type de réponse départage, puis la densité des termes, puis le rang de la fusion
            density = density or {}
            order = sorted(range(len(candidates)),
                           key=lambda i: (-round(scores[i] / 0.02),
                                          -self.bonus_type_reponse(kind, candidates[i][0]),
                                          -density.get(candidates[i][0].id, 0.0), i))
        return [(candidates[i][0], scores[i]) for i in order]

def diversifier(ranked, k, several_documents):
    """Au plus k-1 passages d'un même document quand plusieurs documents sont interrogés."""
    cap = max(1, k - 1) if several_documents else k
    chosen, count = [], Counter()
    for p, s in ranked:
        if count[p.doc_id] < cap:
            chosen.append((p, s)); count[p.doc_id] += 1
        if len(chosen) == k:
            return chosen
    for p, s in ranked:
        if (p, s) not in chosen and len(chosen) < k:
            chosen.append((p, s))
    return chosen
print(f"✅ T4 prêt · langue {LANGUE} · reclassement {NIVEAUX_RECLASSEMENT[RECLASSEMENT][0]}.")
````

````python
# @title ⚙️ T5 · Mémoire de conversation { display-mode: "form" }
DEBUTS_DE_SUIVI = ("et ", "et,", "pour ", "qu en est il", "qu'en est-il", "meme question", "idem",
                    "aussi ", "meme chose", "et si ", "and ", "and,", "what about", "how about",
                    "same for", "same question", "same thing", "also ", "for ")
INTERROGATIF = re.compile(r"^(what|which|how|who|when|where|why|is |are |does |do |did |can |could |was |were |"
                           r"will |should |quel|quelle|quels|quelles|combien|comment|qui|quand|ou |pourquoi|"
                           r"est-ce|que |qu'|y a-t-il)")
CATEGORIES = {
    "region": r"\b(coast|centre|center|north|south|littoral|nord|sud)\b",
    "sex": r"\b(women|woman|men|man|female|male|femmes?|hommes?)\b",
    "quarter": r"\b(q[1-4]|t[1-4]|(?:first|second|third|fourth) quarter|"
               r"(?:premier|deuxieme|second|troisieme|quatrieme) trimestre)\b",
    "area": r"\b(urban|rural|cities|city|countryside|urbain|villes?|campagnes?)\b",
    "age": r"\b(young people|youth|adults|older people|jeunes|adultes|seniors)\b",
    "year": r"\b((?:19|20)\d{2})\b",
    "indicator": r"\b((?:unemployment|activity|employment) rate|underemployment|informal employment|"
                 r"taux (?:de chomage|d'activite|d'emploi)|sous-emploi|emploi informel)\b",
}

def _aligned(text):
    """Minuscules sans accents, en conservant la longueur (mêmes positions)."""
    return "".join((sans_accents(ch) or ch)[:1] for ch in text.lower())

def est_suivi(question, history):
    if not history:
        return False
    q = _aligned(question).strip()
    if q.startswith(tuple(_aligned(s) for s in DEBUTS_DE_SUIVI)):
        return True
    return len(analyser(question)) <= 2 and not INTERROGATIF.match(q)

PROMPT_SUIVI = """Tu réécris des questions. À partir de l'historique de la conversation, réécris la DERNIÈRE
question de l'utilisateur pour qu'elle se comprenne seule, sans l'historique. Ne réponds pas à la question.
Renvoie uniquement la question réécrite, dans la langue de l'utilisateur, sur une seule ligne."""

def reformuler_suivi(question, history, use_model=True):
    """Renvoie (question autonome, réécrite ?, méthode)."""
    if not est_suivi(question, history):
        return question, False, None
    previous = history[-1]["question_autonome"]
    if use_model and LLM.available:
        try:
            turns = "\n".join(f"Utilisateur : {h['question_autonome']}\nAssistant : {h['reponse'][:200]}"
                              for h in history[-3:])
            text = LLM.generate(PROMPT_SUIVI, f"HISTORIQUE\n{turns}\n\nDERNIÈRE QUESTION\n{question}",
                                max_tokens=120).strip().strip('"«» ').splitlines()[0]
            if 5 < len(text) < 300:
                return text, True, "modèle"
        except Exception:
            pass
    result, q_al, replaced = previous, _aligned(question), 0
    for pattern in CATEGORIES.values():
        new = re.search(pattern, q_al)
        if not new:
            continue
        old = re.search(pattern, _aligned(result))
        if old:
            result = result[:old.start()] + question[new.start():new.end()] + result[old.end():]
            replaced += 1
    if not replaced:
        rest = re.sub(r"^\s*(and|what about|how about|same for|also|et|pour|qu'en est-il|aussi)\b[\s,]*", "",
                      question, flags=re.I)
        result = previous.rstrip(" ?") + " — " + rest.rstrip(" ?") + "?"
    return result, True, "règles"
print("✅ T5 prêt.")
````

````python
# @title ⚙️ T6 · La bibliothèque : chargement, indexation, réponses, contrôles, avis { display-mode: "form" }
REFUS = "Information non disponible dans les documents."

PROMPT_STRICT = f"""Tu es l'assistant documentaire d'un institut national de la statistique.
Tu réponds à partir d'extraits d'un ou de plusieurs documents PDF.
Règles, par ordre de priorité :
1. Réponds UNIQUEMENT à partir des sources numérotées fournies. Si elles ne contiennent pas la
   réponse, écris exactement : « {REFUS} » et rien d'autre.
2. N'utilise jamais tes connaissances générales, même si tu crois connaître la réponse.
3. Recopie les chiffres exactement, avec leur unité, leur période et le document dont ils viennent.
4. Si la question compare des périodes ou des documents, donne chaque valeur avec sa source.
5. Termine chaque phrase par le numéro de sa source entre crochets, par exemple [1].
6. Les sources de type tableau s'écrivent « libellé — colonne : valeur ; … ».
7. Réponds dans la langue de la question, en quatre phrases au maximum."""

COMPARAISON = re.compile(r"\b(chang\w*|evolv\w*|evolu\w*|trend|between|entre|compar\w*|versus|vs|difference|"
                        r"increase\w*|decrease\w*|hausse|baisse|progress\w*|par rapport)\b")
DEBUT_PRONOM = re.compile(r"^(It|Its|They|Their|This|These|Il|Elle|Ils|Elles|Ce|Cette|Ces|Sa|Son)\b")

@dataclass
class DocumentPDF:
    id: int
    filename: str
    title: str
    data: bytes
    pages: int
    diagnosis: object
    elements: list
    info: dict

    @property
    def short(self):
        t = re.sub(r"\s*\((fictitious|fictif|fictive)\)\s*$", "", self.title)
        return t if len(t) <= 42 else t[:40].rstrip() + "…"

    @property
    def scanned_pages(self):
        return self.diagnosis.loc[self.diagnosis["État"] == "scannée", "Page"].tolist()

def est_refus(text):
    t = sans_accents(text.lower())
    return any(m in t for m in ["information non disponible", "ne contient pas", "ne contiennent pas",
                                "pas d'information", "aucune information", "je ne sais pas",
                                "ne permet pas de repondre", "ne precise pas", "ne mentionne pas",
                                "information not available", "does not contain", "do not contain",
                                "no information", "i don't know", "cannot answer", "does not mention"])

def _normaliser_nombres(text):
    t = text.replace("\u00a0", " ").replace("\u202f", " ")
    t = re.sub(r"(?<=\d),(?=\d{3}\b)", "", t)       # 18,452,300 -> 18452300
    t = re.sub(r"(?<=\d) (?=\d{3}\b)", "", t)       # 18 452 300 -> 18452300
    return re.sub(r"(?<=\d),(?=\d)", ".", t)         # 7,8 -> 7.8

def verifier_chiffres(text, passages):
    """Chaque nombre de la réponse doit figurer dans au moins une des sources fournies."""
    without_refs = re.sub(r"\[\d+\]", " ", _normaliser_nombres(text))
    numbers = list(dict.fromkeys(re.findall(r"\d+(?:\.\d+)?", without_refs)))
    source = _normaliser_nombres(" ".join(p.text for p, _ in passages))
    missing = [n for n in numbers if not re.search(rf"(?<![\d.]){re.escape(n)}(?!\d|\.\d)", source)]
    return {"nombres": len(numbers), "absents": missing}

def termes_absents(question, passage):
    """Mots de la question sans équivalent dans le passage retenu."""
    present = set(analyser(passage.index_text))
    absent = {t for t, g in concepts(question) if not g & present}
    words = []
    for word in re.findall(r"[\w'-]+", question):
        if any(t in absent for t in analyser(word)) and word not in words:
            words.append(word)
    return words

def alleger_question(question, results, vocabulary):
    """Écarte les mots de la question qu'aucun passage retrouvé ne contient.
    « le nombre de dimensions du cadre proposé par l'AfDB » → « dimensions AfDB » : les mots
    interrogatifs et les distracteurs hors sujet disparaissent, le sujet demeure.

    N'allège jamais quand un mot de la question est absent de toute la collection
    (« maïs », « hôpitaux ») : ce mot est précisément le signal d'une question hors sujet."""
    items = concepts(question)
    if not results or len(items) < 2:
        return ""
    if any(not (group & vocabulary) for _, group in items):
        return ""
    present = set()
    for p, _ in results:
        present |= set(analyser(p.index_text))
    kept = [t for t, group in items if group & present]
    return " ".join(kept) if 1 < len(kept) < len(items) else ""

AVIS = []
def enregistrer_avis(turn, useful, comment="", channel="notebook"):
    row = {"horodatage": time.strftime("%Y-%m-%d %H:%M:%S"), "canal": channel, "utile": bool(useful),
           "question": turn.get("question", ""), "question_autonome": turn.get("question_autonome", ""),
           "reponse": turn.get("reponse", ""), "sources": " | ".join(turn.get("citations", [])),
           "commentaire": comment}
    AVIS.append(row)
    path = DOSSIER_SORTIE / "avis_utilisateurs.csv"
    pd.DataFrame([row]).to_csv(path, mode="a", header=not path.exists(), index=False, encoding="utf-8-sig")
    return row

def _decouper_phrases(text):
    return [s.strip() for s in re.split(r"(?<=[.!?])\s+|\n", text) if len(s.strip()) > 15]

INTRO_LISTE = re.compile(r"(:|following|as follows|are:|comprend|suivants?|suivantes?)\s*$", re.I)

def _meilleure_phrase(items, passage, weight, text=None, kind="other"):
    """Meilleure phrase du passage, prolongée quand elle annonce une liste ou reste très courte.
    Une question « combien » privilégie les phrases qui portent réellement un nombre."""
    sentences = _decouper_phrases(text or passage.text)
    best, best_score = None, 0.0
    for n, sentence in enumerate(sentences):
        terms = set(analyser(sentence))
        if passage.kind == "table" and n > 0:
            terms |= set(analyser(sentences[0]))            # une ligne de tableau hérite de sa légende
        score = sum(weight(t) for t, g in items if g & terms)
        if kind == "count":
            score *= 1.6 if porte_un_nombre(sentence) else 0.6
        elif kind == "list":                                # « liste les pays… » attend la liste elle-même
            names = len(re.findall(r"(?<!^)(?<![.!?]\s)\b[A-Z][a-z]{2,}", sentence))
            score *= (1.35 if ENUMERATION.search(sentence) else 0.9) * (1 + 0.07 * min(names, 6))
        elif kind == "date" and re.search(r"\b(19|20)\d{2}\b|\b\d{1,2} (January|February|March|April|May|June|July|"
                                          r"August|September|October|November|December)\b", sentence):
            score *= 1.4
        if score > best_score:
            text = sentence
            if n > 0 and DEBUT_PRONOM.match(text):
                text = sentences[n - 1] + " " + text
            if passage.kind == "table" and n > 0:
                text = f"{sentences[0]}: {text}"
            extra = n + 1
            while (extra < len(sentences) and len(text) < 420
                   and (INTRO_LISTE.search(text) or len(text) < 130)):
                text += " " + sentences[extra]             # une liste ou une phrase courte appelle la suite
                extra += 1
            best, best_score = text, score
    return best, best_score

class Bibliotheque:
    def __init__(self):
        self.documents, self.passages = [], []
        self.threshold, self.calibration, self.ready = 0.5, None, False
        self.folder = ENV["root"] / "pdf"
        self.folder.mkdir(exist_ok=True)

    # --- loading -----------------------------------------------------------------------------
    BROWSER_HEADERS = {
        "User-Agent": ("Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) "
                       "Chrome/125.0.0.0 Safari/537.36"),
        "Accept": "application/pdf,text/html;q=0.9,*/*;q=0.8",
        "Accept-Language": "en-US,en;q=0.9,fr;q=0.8",
    }

    @staticmethod
    def lien_direct(url):
        """Transforme un lien de partage en lien de téléchargement direct (Google Drive, Dropbox, OneDrive)."""
        drive = re.search(r"(?:drive|docs)\.google\.com/(?:file/d/|open\?id=|uc\?id=|.*[?&]id=)([\w-]{20,})", url)
        if drive:
            return f"https://drive.usercontent.google.com/download?id={drive.group(1)}&export=download&confirm=t", f"gdrive_{drive.group(1)[:10]}.pdf"
        if "dropbox.com" in url:
            base = re.sub(r"[?&]dl=[01]", "", url)
            return base + ("&" if "?" in base else "?") + "dl=1", None
        if "1drv.ms" in url or "sharepoint.com" in url:
            return url + ("&" if "?" in url else "?") + "download=1", None
        return url, None

    def telecharger(self, url, timeout=90):
        """Certains sites institutionnels refusent les scripts anonymes : on envoie des en-têtes de
        navigateur, puis on tente curl avant d'abandonner avec un message clair.
        Les liens de partage (Google Drive, Dropbox, OneDrive) sont convertis en liens directs."""
        url, forced_name = self.lien_direct(url)
        name = forced_name or re.sub(r"[^\w.-]", "_", url.split("?")[0].rstrip("/").split("/")[-1]) or "document"
        target = self.folder / (name if name.lower().endswith(".pdf") else name + ".pdf")
        if target.exists() and target.stat().st_size > 1000:
            return target
        headers = dict(self.BROWSER_HEADERS)
        headers["Referer"] = "/".join(url.split("/")[:3]) + "/"
        try:
            request = urllib.request.Request(url, headers=headers)
            with urllib.request.urlopen(request, timeout=timeout) as response:
                content = response.read()
            if content[:4] == b"%PDF":
                target.write_bytes(content)
                return target
            first = "l'adresse a renvoyé une page web, pas un PDF (lien de partage, page de connexion ou avertissement antivirus)"
            page = content[:4000].decode("utf-8", "ignore")
            token = re.search(r'name="(confirm|uuid)" value="([^"]+)"', page)
            if token:                                   # avertissement antivirus de Google Drive (gros fichier)
                with urllib.request.urlopen(urllib.request.Request(
                        f"{url}&{token.group(1)}={token.group(2)}", headers=headers), timeout=timeout) as response:
                    content = response.read()
                if content[:4] == b"%PDF":
                    target.write_bytes(content)
                    return target
        except Exception as error:
            first = f"{type(error).__name__}: {str(error)[:80]}"
        if shutil.which("curl"):                    # curl negotiates TLS like a browser: often enough
            command = ["curl", "-sSL", "--max-time", str(timeout), "-A", headers["User-Agent"],
                       "-H", f"Accept: {headers['Accept']}", "-H", f"Referer: {headers['Referer']}",
                       "-o", str(target), url]
            if subprocess.run(command, capture_output=True).returncode == 0 and target.exists() \
                    and target.read_bytes()[:4] == b"%PDF":
                return target
            target.unlink(missing_ok=True)
        raise RuntimeError(f"téléchargement impossible — {first}. Vérifiez que le lien est public (« tous ceux "
                           f"qui ont le lien »), ou téléchargez le PDF à la main puis réglez IMPORTER_PDF = True "
                           f"(Colab) ou indiquez son chemin dans DOCUMENTS")

    def _resoudre(self, sources):
        paths, failures = [], []
        for s in ([sources] if isinstance(sources, (str, Path)) else sources):
            s = str(s)
            try:
                if s == "exemple":
                    paths += creer_documents_exemple(self.folder / "examples")
                elif s.startswith(("http://", "https://")):
                    paths.append(self.telecharger(s))      # un échec ici ne doit pas bloquer les autres documents
                elif Path(s).is_dir():
                    found = sorted(Path(s).rglob("*.pdf"))
                    paths += found
                    if not found:
                        failures.append((s, "aucun PDF dans ce dossier"))
                elif Path(s).exists():
                    paths.append(Path(s))
                else:
                    failures.append((s, "source introuvable"))
            except Exception as error:
                failures.append((s.split("/")[-1][:60], str(error)[:200]))
        self._source_failures = failures
        return paths

    @staticmethod
    def _vers_json(elements, info):
        def clean(t):
            return {"box": list(map(float, t["box"])), "headers": t["headers"], "rows": t["rows"]}
        return json.dumps({"info": info, "elements": [
            {"type": e.type, "text": e.text, "page": e.page, "box": list(map(float, e.box)),
             "level": e.level, "section": e.section,
             "table": clean(e.table) if getattr(e, "table", None) else None,
             "caption": getattr(e, "caption", ""), "extra": [[p, list(map(float, b))] for p, b in getattr(e, "extra", [])]}
            for e in elements]}, ensure_ascii=False)

    @staticmethod
    def _depuis_json(text):
        payload = json.loads(text)
        elements = []
        for item in payload["elements"]:
            e = Element(item["type"], item["text"], item["page"], tuple(item["box"]), item["level"], item["section"])
            if item["table"]:
                e.table = {"box": pymupdf.Rect(item["table"]["box"]), "headers": item["table"]["headers"],
                           "rows": item["table"]["rows"]}
                e.caption, e.extra = item["caption"], [(p, tuple(b)) for p, b in item["extra"]]
            elements.append(e)
        return elements, payload["info"]

    def _cle_cache(self, data):
        settings = f"{SUPPRIMER_ENTETES}|{EXTRAIRE_TABLEAUX}|{UTILISER_OCR}|{LANGUE_OCR}|{SUPPRIMER_SOMMAIRE}|{MAX_PAGES_PAR_DOCUMENT}|v2"
        return hashlib.sha1(data + settings.encode("utf-8")).hexdigest()[:16]

    def charger(self, sources):
        added, errors = [], []
        known = {d.filename for d in self.documents}
        paths = self._resoudre(sources)
        errors += getattr(self, "_source_failures", [])
        for number, path in enumerate(paths, 1):
            try:
                data = Path(path).read_bytes()
                if not data.startswith(b"%PDF"):
                    raise ValueError("ce n'est pas un fichier PDF")
                doc = pymupdf.open(stream=data, filetype="pdf")
                if doc.needs_pass:
                    raise ValueError("PDF protégé par mot de passe")
                if Path(path).name in known:
                    continue
                start = time.time()
                cache_file = DOSSIER_CACHE / f"extraction_{self._cle_cache(data)}.json"
                if UTILISER_CACHE and cache_file.exists():
                    elements, info = self._depuis_json(cache_file.read_text(encoding="utf-8"))
                    origin = "depuis le cache"
                else:
                    elements, info = extraire_structure(doc)
                    origin = "extrait"
                    if UTILISER_CACHE:
                        cache_file.write_text(self._vers_json(elements, info), encoding="utf-8")
                print(f"  {number}/{len(paths)} · {Path(path).name[:48]} · {doc.page_count} pages · "
                      f"{origin} en {time.time() - start:.1f} s")
                title = ((doc.metadata or {}).get("title") or "").strip()
                if not title:
                    title = elements[0].text if elements and elements[0].type == "heading" else Path(path).stem.replace("_", " ")
                d = DocumentPDF(len(self.documents), Path(path).name, title, data, doc.page_count,
                                diagnostiquer(doc), elements, info)
                d.path = str(path)
                self.documents.append(d); added.append(d); known.add(d.filename)
            except Exception as error:
                errors.append((Path(path).name, str(error)[:120]))
        self.ready = False
        return added, errors

    # --- indexing ----------------------------------------------------------------------------
    def indexer(self):
        self.passages = []
        for d in self.documents:
            self.passages += creer_passages(d, len(self.passages))
        if not self.passages:
            raise ValueError("Aucun passage : vérifiez que les PDF contiennent du texte (ou activez l'OCR).")
        start = time.time()
        self.units = creer_unites_recherche(self.passages)         # passages + petites fenêtres
        self.by_id = {p.id: p for p in self.passages}
        SIGLES.clear()
        SIGLES.update(construire_sigles(self.passages))
        construire_vocabulaire(self.passages)                        # pour la tolérance orthographique
        self.bm25 = IndexBM25(self.units)
        self.semantic = IndexSemantique(self.units)
        self.reranker = Reclasseur(self.bm25)
        print(f"  {len(self.passages)} passages / {len(self.units)} unités de recherche indexés en "
              f"{time.time() - start:.1f} s ({self.semantic.name}) · {len(SIGLES)} sigles et "
              f"{len(FORMES_VOCABULAIRE)} mots appris · calibration du seuil…")
        start = time.time()
        self.calibrer()
        print(f"  seuil {self.threshold:.2f} calibré en {time.time() - start:.1f} s")
        self.ready = True

    def calibrer(self):
        off_topic = {
            "fr": ["Quelle équipe a gagné la dernière coupe du monde de football ?",
                   "Quelle est la recette traditionnelle du couscous ?",
                   "Quel temps fera-t-il demain dans la capitale ?",
                   "Quel est le prix du maïs au kilogramme sur les marchés ?",
                   "Combien de vaccins ont été administrés contre la rougeole ?",
                   "Quelle est la production nationale de cacao ?",
                   "Qui est l'auteur du roman le plus vendu cette année ?",
                   "Combien d'hôpitaux compte la région du Nord ?"],
            "en": ["Which team won the last football World Cup?",
                   "What is the traditional recipe for couscous?",
                   "What will the weather be like tomorrow in the capital?",
                   "What is the price of maize per kilogram at the market?",
                   "How many measles vaccines were administered?",
                   "What is the national cocoa production?",
                   "Who wrote this year's best-selling novel?",
                   "How many hospitals are there in the North region?"],
        }[LANGUE]
        headings = [re.sub(r"^\s*[\dIVX]+[.)]?\s*", "", e.text) for d in self.documents
                    for e in d.elements if e.type == "heading" and e.level > 1]
        on_topic = [f"Que dit le document sur : {h} ?" for h in dict.fromkeys(headings)][:15]
        rows = [{"Question": q, "Réponse existe": True, "Confiance": round(self.confiance(q), 3)} for q in on_topic]
        rows += [{"Question": q, "Réponse existe": False, "Confiance": round(self.confiance(q), 3)} for q in off_topic]
        self.calibration = pd.DataFrame(rows)
        ceiling = self.calibration.loc[~self.calibration["Réponse existe"], "Confiance"].max()
        # sur un long rapport, presque chaque mot apparaît quelque part : les scores hors sujet
        # montent, on plafonne donc le seuil pour ne pas refuser des questions légitimes.
        self.calibrated_threshold = round(min(ceiling + 0.05, 0.7), 2)
        self.threshold = SEUIL_MANUEL if SEUIL_MANUEL is not None else self.calibrated_threshold

    # --- recherche et réponses ----------------------------------------------------------------
    def confiance(self, question):
        r = self.rechercher(question, 1)
        return r[0][1] if r else 0.0

    def rechercher(self, question, k=None, docs=None, kind=None):
        k = k or K_RESULTATS
        allowed = set(range(len(self.documents))) if docs is None else set(docs)
        keep = lambda ranking: [(p, s) for p, s in ranking if p.doc_id in allowed]
        lexical = keep(self.bm25.rechercher(question, 40))
        density = {u.id: s for u, s in lexical}
        candidates = fusion_rrf([lexical, keep(self.semantic.rechercher(question, 40))], top=14)
        ranked = self.reranker.reclasser(question, candidates, density, kind or type_de_question(question))
        seen, mapped = set(), []
        for unit, score in ranked:                              # une fenêtre vaut pour son passage de contexte
            parent = self.by_id.get(unit.parent_id, unit)
            if parent.id in seen:
                continue
            seen.add(parent.id)
            mapped.append((parent, score))
        return diversifier(mapped, k, len(allowed) > 1)

    def repondre(self, question, history=None, docs=None, k=None, threshold=None, guardrails=True, memory=True):
        if not self.ready:
            self.indexer()
        start = time.time()
        threshold = self.threshold if threshold is None else threshold
        standalone, rewritten, method = (reformuler_suivi(question, history or [], MEMOIRE_PAR_MODELE)
                                         if memory else (question, False, None))
        standalone, corrections = corriger_orthographe(standalone)     # « chômge » -> « chômage »
        kind = type_de_question(standalone)
        results = self.rechercher(standalone, k, docs, kind)
        conf = results[0][1] if results else 0.0
        relaxed = None
        if conf < threshold:
            # second passage : on ne garde que les mots que les passages candidats contiennent
            reduced = alleger_question(standalone, results, set(self.bm25.vocabulary))
            if reduced and reduced.lower() != standalone.lower():
                fallback = self.rechercher(reduced, k, docs, kind)
                if fallback and fallback[0][1] > conf:
                    results, conf, relaxed = fallback, fallback[0][1], reduced
        alternative = None
        if docs is not None and set(docs) != set(range(len(self.documents))):
            best = self.rechercher(standalone, 1, None)
            if best and best[0][0].doc_id not in set(docs) and best[0][1] >= max(conf + 0.15, threshold):
                alternative = best[0][0].doc
        r = {"question": question, "question_autonome": standalone, "reformulee": rewritten,
             "methode_reformulation": method, "confiance": conf, "passages": results, "motif": None, "allegee": relaxed,
             "corrections": corrections, "decoupage": STRATEGIE_DECOUPAGE, "reclassement": self.reranker.mode,
             "type_question": type_de_question(standalone),
             "garde_fous": guardrails, "heure": time.strftime("%H:%M:%S"), "alternative": alternative,
             "termes_absents": termes_absents(standalone, results[0][0]) if results else [],
             "verification": None, "citations": [p.citation for p, _ in results], "type_question": kind}
        if guardrails:
            years = set(re.findall(r"\b(?:19|20)\d{2}\b", standalone))
            content = " ".join(p.index_text for p, _ in results)
            absent_years = sorted(y for y in years if y not in content)
            if conf < threshold:
                r["motif"] = f"confiance {conf:.2f} inférieure au seuil {threshold:.2f}"
            elif absent_years:
                r["motif"] = f"année(s) {', '.join(absent_years)} absente(s) des passages"
            if r["motif"]:
                return {**r, "reponse": REFUS, "refus": True, "mode": "guardrail",
                        "duree_s": round(time.time() - start, 2)}
        text, mode = None, "extractive"
        if LLM.available:
            try:
                sources = "\n\n".join(f"[{i}] ({p.citation}{', table' if p.kind == 'table' else ''})\n{self.contexte_de(p)}"
                                      for i, (p, _) in enumerate(results, 1))
                text = LLM.generate(PROMPT_STRICT, f"SOURCES\n\n{sources}\n\nQUESTION\n{standalone}")
                mode = f"{LLM.name} · {LLM.model}"
            except Exception as error:
                r["motif"] = f"modèle indisponible ({str(error)[:80]}) : mode extractif"
        if text is None:
            text = self._extraire(standalone, results, kind)
        refused = est_refus(text)
        if not refused:
            r["verification"] = verifier_chiffres(text, results)
        return {**r, "reponse": text, "refus": refused, "mode": mode,
                "duree_s": round(time.time() - start, 2)}

    QUESTION_DEFINITION = re.compile(r"\b(stands? for|stand|mean|means|abbreviation|acronym|signifie|veut dire)\b", re.I)

    def contexte_de(self, passage):
        """Le passage et ses voisins de la même section : un contexte complet pour le modèle."""
        if not CONTEXTE_VOISIN:
            return passage.text
        family = [p for p in self.passages if p.doc_id == passage.doc_id and p.section == passage.section]
        try:
            position = family.index(passage)
        except ValueError:
            return passage.text
        pieces, size, before, after = [passage.text], len(passage.text), position - 1, position + 1
        while size < TAILLE_CONTEXTE and (before >= 0 or after < len(family)):
            if before >= 0 and size + len(family[before].text) <= TAILLE_CONTEXTE:
                pieces.insert(0, family[before].text); size += len(family[before].text); before -= 1
            elif after < len(family) and size + len(family[after].text) <= TAILLE_CONTEXTE:
                pieces.append(family[after].text); size += len(family[after].text); after += 1
            else:
                break
        return " ".join(dict.fromkeys(pieces))

    def _extraire(self, question, results, kind="other"):
        # « Que signifie INS ? » — réponse tirée des sigles appris dans les documents
        if self.QUESTION_DEFINITION.search(question):
            for acronym in sorted(SIGLES, key=len, reverse=True):
                if re.search(rf"\b{re.escape(acronym)}\b", question):
                    expansion = SIGLES[acronym]
                    for j, (p, _) in enumerate(results):
                        sentence = next((s for s in _decouper_phrases(p.text)
                                         if expansion.lower() in s.lower() or f"({acronym})" in s), None)
                        if sentence:
                            return f"{acronym} — {expansion}. {sentence} [{j + 1}]"
                    return f"{acronym} — {expansion}."
        items = concepts(question)
        candidates = []
        for j, (p, _) in enumerate(results):
            sentence, score = _meilleure_phrase(items, p, self.reranker.poids, kind=kind)   # on cite le passage ;
            # le contexte élargi (contexte_de) n'est envoyé qu'au modèle, qui a besoin de complétude
            if sentence:
                candidates.append((score * (1 - 0.02 * j), j, p, sentence))
        if not candidates:
            return REFUS
        candidates.sort(key=lambda c: -c[0])
        score0, j0, p0, sentence0 = candidates[0]
        text = f"{sentence0} [{j0 + 1}]"
        if COMPARAISON.search(sans_accents(question.lower())):
            others = [c for c in candidates[1:] if c[2].doc_id != p0.doc_id and c[0] >= 0.5 * score0]
            if others:
                _, j1, p1, sentence1 = others[0]
                text += f" — {p1.doc}: {sentence1} [{j1 + 1}]"
        return text

    # --- display ------------------------------------------------------------------------------
    def vignette(self, passage, **kw):
        doc = self.documents[passage.doc_id]
        first = passage.pages[0]
        return vignette_html(doc.data, first, [b for pg, b in passage.boxes if pg == first], **kw)

    def carte_html(self, r, thumbnails=True):
        badges = pastille("REFUS", C["ochre"]) if r["refus"] else pastille("RÉPONSE", C["dark_green"])
        badges += pastille(f"confiance {r['confiance']:.2f}", C["teal"]) + pastille(r["mode"], C["grey"])
        if not r["garde_fous"]:
            badges += pastille("garde-fous désactivés", C["brick"])
        chk = r.get("verification")
        if chk and chk["nombres"]:
            badges += (pastille(f"chiffres vérifiés {chk['nombres']}/{chk['nombres']}", C["dark_green"]) if not chk["absents"]
                       else pastille(f"⚠ chiffre(s) absent(s) des sources : {', '.join(chk['absents'])}", C["brick"]))
        if r.get("allegee"):
            badges += pastille(f"recherche allégée : {r['allegee']}", C["ochre"])
        if r.get("corrections"):
            badges += pastille("orthographe : " + ", ".join(f"{a} → {b}" for a, b in r["corrections"]), C["teal"])
        notes = ""
        if r["reformulee"]:
            notes += (f'<div style="font-size:12.5px;color:{C["teal"]};margin-top:6px;">🧠 Compris comme : '
                      f'<i>{_html.escape(r["question_autonome"])}</i> '
                      f'<span style="color:{C["slate"]};">({r["methode_reformulation"]})</span></div>')
        if not r["refus"] and r.get("termes_absents"):
            notes += (f'<div style="font-size:12.5px;color:{C["ochre"]};margin-top:6px;">⚠️ Introuvable dans la source '
                      f'principale : <b>{_html.escape(", ".join(r["termes_absents"]))}</b> — vérifiez que la réponse '
                      f'porte bien sur votre question.</div>')
        if r.get("alternative"):
            notes += (f'<div style="font-size:12.5px;color:{C["teal"]};margin-top:6px;">💡 Une réponse plus pertinente '
                      f'figure probablement dans « <b>{_html.escape(r["alternative"])}</b> », exclu par le filtre.</div>')
        reason = f'<div class="rg-note">Motif : {_html.escape(r["motif"])}</div>' if r.get("motif") else ""
        answer = re.sub(r"\[(\d+)\]", lambda m: f'<sup style="background:{C["mint"]};color:{C["dark_green"]};'
                        f'border-radius:4px;padding:0 4px;font-weight:700;">{m.group(1)}</sup>', _html.escape(r["reponse"]))
        sources = []
        for i, (p, s) in enumerate(r["passages"], 1):
            thumb = self.vignette(p, dpi=55, width=105) if thumbnails and i <= 2 and not r["refus"] else ""
            sources.append(
                f'<div style="display:flex;gap:10px;align-items:flex-start;margin:6px 0;">{thumb}<div style="flex:1;">'
                f'<b>[{i}]</b> {pastille(p.doc, C["dark_green"])}<b>{_html.escape(p.section or "début")}</b> · '
                f'p. {", ".join(map(str, p.pages))} · {p.kind} · confiance {s:.2f}<br>'
                f'<span style="color:{C["slate"]};">“{_html.escape(" ".join(p.text.split())[:220])}…”</span></div></div>')
        return (f'<div class="rg rg-card" style="border-left:6px solid {C["ochre"] if r["refus"] else C["green"]};margin:0;">'
                f'<div>{badges}<span style="float:right;color:{C["slate"]};font-size:11px;">{r["heure"]} · {r["duree_s"]} s</span></div>'
                f'{notes}<div style="font-size:14.5px;margin:10px 0;line-height:1.55;">{answer}</div>{reason}'
                f'<details {"open" if not r["refus"] else ""}><summary style="cursor:pointer;color:{C["dark_green"]};'
                f'font-weight:600;font-size:13px;">Sources ({len(r["passages"])})</summary>'
                f'<div class="rg-src">{"".join(sources) or "Aucune source."}</div></details></div>')

    def afficher(self, r):
        display(HTML(f'{_CSS}<div class="rg" style="margin:10px 0;"><div style="font-weight:700;font-size:15px;'
                     f'margin-bottom:6px;">❓ {_html.escape(r["question"])}</div>{self.carte_html(r)}</div>'))

    def resume(self):
        return pd.DataFrame([{"N°": d.id, "Document": d.title, "Fichier": d.filename, "Pages": d.pages,
                              "Pages scannées": ", ".join(map(str, d.scanned_pages)) or "—",
                              "Tableaux": sum(e.type == "table" for e in d.elements),
                              "Passages": sum(p.doc_id == d.id for p in self.passages) if self.passages else "—"}
                             for d in self.documents])

BIB = Bibliotheque()
_HISTORIQUE_DEMO = []

def questions_suggerees(n=5):
    """Questions adaptées aux documents réellement chargés (utilisées par les démonstrations et les interfaces)."""
    if DOCUMENTS == ["exemple"]:
        return ["Quel est le taux de chômage des femmes dans la région du Littoral ?", "Et pour les hommes ?",
                "Comment le taux de chômage a-t-il évolué entre le quatrième trimestre 2025 et le premier trimestre 2026 ?",
                "Quelle est la taille moyenne d'un ménage ?", "Quel est le prix du maïs ?"][:n]
    headings = [re.sub(r"^\s*[\dIVX]+[.)]?\s*", "", e.text).strip()
                for d in BIB.documents for e in d.elements
                if e.type == "heading" and e.level > 1 and e.page > 2 and 12 < len(e.text) < 70
                and len(e.text.split()) >= 2 and e.text.lower() not in d.title.lower()]
    questions = [f"Que dit le document sur : {h} ?" for h in dict.fromkeys(headings)][:max(1, n - 2)]
    if SIGLES:
        questions.append(f"Que signifie {sorted(SIGLES, key=lambda a: -len(SIGLES[a]))[0]} ?")
    questions.append("Quel est le prix du maïs au kilogramme ?")       # hors sujet, doit être refusée
    return questions[:n]

def demander(question, docs=None, memory=True):
    """Pose une question à la bibliothèque, en se souvenant des questions précédentes."""
    r = BIB.repondre(question, history=_HISTORIQUE_DEMO if memory else None, docs=docs)
    _HISTORIQUE_DEMO.append(r)
    BIB.afficher(r)
    return r
print("✅ T6 prêt · boîte à outils complète.")
````

````python
# @title ⚙️ T7 · La chaîne visible, le comparateur et le banc d'essai { display-mode: "form" }

# ── 1. La chaîne, dessinée ─────────────────────────────────────────────────────────────────
CHAINE = [
    ("1", "Question", "saisie par l'utilisateur", "grey"),
    ("2", "Relance", "réécrite en question autonome", "teal"),
    ("3", "Orthographe", "les mots absents des documents sont corrigés", "teal"),
    ("4", "Normalisation", "minuscules, accents, racines, mots interrogatifs retirés", "grey"),
    ("5", "Enrichissement", "glossaire + sigles appris dans les documents", "grey"),
    ("6", "BM25", "mots exacts, sur chaque unité de recherche", "green"),
    ("7", "Sémantique", "sens, sur chaque unité de recherche", "green"),
    ("8", "Fusion", "les deux classements fusionnés (RRF)", "green"),
    ("9", "Reclassement", "ordre décidé par le niveau que vous avez choisi", "gold"),
    ("10", "Fenêtres → passages", "chaque unité redonne le passage à lire", "gold"),
    ("11", "Garde-fous", "seuil, contrôle des années, refus", "ochre"),
    ("12", "Réponse", "rédigée par le modèle, ou extraite", "dark_green"),
    ("13", "Vérifications", "chiffres, termes manquants, filtre", "dark_green"),
]

def afficher_chaine():
    """Toute la chaîne avec les réglages actuellement en vigueur."""
    chosen = {"9": NIVEAUX_RECLASSEMENT[BIB.reranker.mode][0] if getattr(BIB, "reranker", None) else RECLASSEMENT,
              "6": f"{len(getattr(BIB, 'units', []))} unités",
              "7": getattr(getattr(BIB, "semantic", None), "name", "—"),
              "11": f"seuil {BIB.threshold:.2f}" if getattr(BIB, "threshold", None) else "",
              "12": f"{LLM.name} · {LLM.model}" if LLM.available else "mode extractif"}
    boxes = ""
    for number, title, subtitle, colour in CHAINE:
        note = chosen.get(number, "")
        boxes += (
            f'<div style="flex:1 1 150px;min-width:150px;background:#fff;border:1px solid {C["sage"]};'
            f'border-top:3px solid {C[colour]};border-radius:10px;padding:8px 10px;">'
            f'<div style="font-size:10.5px;color:{C[colour]};font-weight:800;">ÉTAPE {number}</div>'
            f'<div style="font-size:13px;font-weight:700;color:{C["ink"]};">{title}</div>'
            f'<div style="font-size:11.5px;color:{C["slate"]};line-height:1.35;">{subtitle}</div>'
            + (f'<div style="font-size:11px;color:{C["dark_green"]};font-weight:700;margin-top:3px;">{_html.escape(str(note))}</div>' if note else "")
            + '</div>')
    display(HTML(
        f'{_CSS}<div class="rg"><div class="rg-title">🔗 La chaîne RAG, telle que configurée en ce moment</div>'
        f'<div style="display:flex;gap:8px;flex-wrap:wrap;">{boxes}</div>'
        f'<div class="rg-note">Découpage : <b>{NIVEAUX_DECOUPAGE[STRATEGIE_DECOUPAGE][0]}</b> · '
        f'Reclassement : <b>{NIVEAUX_RECLASSEMENT[RECLASSEMENT][0]}</b> · Passages remis à la réponse : <b>{K_RESULTATS}</b></div></div>'))

# ── 2. La chaîne, suivie sur une vraie question ────────────────────────────────────────────
def expliquer(question, docs=None, k=None):
    """Fait passer une question par toutes les étapes et montre ce que chacune a changé."""
    k = k or K_RESULTATS
    steps = []
    standalone, corrections = corriger_orthographe(question)
    kind = type_de_question(standalone)
    if corrections:
        steps.append(("3 · Orthographe", ", ".join(f"<b>{a}</b> → {b}" for a, b in corrections)))
    effect = {"count": "les passages et phrases portant une vraie quantité sont favorisés",
              "list": "les passages et phrases portant une énumération sont favorisés",
              "date": "les passages et phrases portant une année sont favorisés",
              "definition": "les sigles appris dans les documents sont consultés en premier",
              "other": "aucun traitement particulier"}[kind]
    steps.append(("Type de question", f"<b>{kind}</b> — {effect}"))
    terms = concepts(standalone)
    steps.append(("4 · Termes retenus", ", ".join(f"{t} <span style='color:#5E6964'>({BIB.reranker.poids(t):.1f})</span>"
                                              for t, _ in terms) or "—"))
    added = enrichir_question(standalone)[1]
    steps.append(("5 · Enrichissement", "<br>".join(f"<b>{a}</b> → {b}" for a, b in added) or "rien à ajouter"))
    lexical = BIB.bm25.rechercher(standalone, 40)
    semantic = BIB.semantic.rechercher(standalone, 40)
    line = lambda units: "<br>".join(f"{i}. <b>{u.section[:34] or '—'}</b> · p. {u.pages[0]} "
                                     f"<span style='color:#5E6964'>({s:.2f})</span>"
                                     for i, (u, s) in enumerate(units[:4], 1)) or "rien trouvé"
    steps.append(("6 · BM25 (mots exacts)", line(lexical)))
    steps.append((f"7 · {BIB.semantic.name}", line(semantic)))
    fused = fusion_rrf([lexical, semantic], top=14)
    steps.append(("8 · Fusion", line(fused)))
    reranked = BIB.reranker.reclasser(standalone, fused, {u.id: s for u, s in lexical}, kind)
    steps.append((f"9 · Reclassement · {BIB.reranker.name}",
                  "<br>".join(f"{i}. <b>{u.section[:34] or '—'}</b> · p. {u.pages[0]} "
                              f"<span style='color:#5E6964'>(couverture {s:.2f}"
                              + (", type de réponse ✓" if BIB.reranker.bonus_type_reponse(kind, u) else "") + ")</span>"
                              for i, (u, s) in enumerate(reranked[:4], 1))))
    r = BIB.repondre(question, docs=docs, k=k)
    steps.append(("10 · Passages retournés", "<br>".join(f"{i}. {p.citation}" for i, (p, _) in enumerate(r["passages"], 1))))
    steps.append(("11 · Garde-fous", f"confiance <b>{r['confiance']:.2f}</b> face au seuil <b>{BIB.threshold:.2f}</b> → "
                                     + ("<b style='color:#D49A00'>refus</b>" if r["refus"] else "<b style='color:#00704A'>réponse</b>")
                                     + (f" · {r['motif']}" if r.get("motif") else "")))
    rows = "".join(f'<tr><td style="white-space:nowrap;vertical-align:top;padding:6px 12px 6px 0;font-weight:700;'
                   f'color:{C["dark_green"]};font-size:12.5px;">{name}</td>'
                   f'<td style="padding:6px 0;font-size:12.5px;line-height:1.5;">{value}</td></tr>'
                   for name, value in steps)
    display(HTML(f'{_CSS}<div class="rg rg-card"><div class="rg-title">🔍 {_html.escape(question)}</div>'
                 f'<table style="width:100%;border-collapse:collapse;">{rows}</table></div>'))
    BIB.afficher(r)
    return r

# ── 3. Les cinq niveaux de découpage, sur le même texte ────────────────────────────────────
def comparer_decoupage(document=None, levels=("fixed", "sentences", "sections", "hybrid", "smart")):
    """Même document, cinq niveaux : combien de morceaux, de quelle longueur, et à quoi ressemble le premier."""
    global STRATEGIE_DECOUPAGE
    document = document or BIB.documents[0]
    keep, rows = STRATEGIE_DECOUPAGE, []
    try:
        for level in levels:
            STRATEGIE_DECOUPAGE = level
            start = time.time()
            passages = creer_passages(document)
            units = creer_unites_recherche(passages, level)
            lengths = [len(p.text) for p in passages] or [0]
            first = next((p for p in passages if len(p.text) > 200), passages[0] if passages else None)
            rows.append({"Niveau": NIVEAUX_DECOUPAGE[level][0],
                         "Passages": len(passages), "Unités de recherche": len(units),
                         "Longueur moyenne": int(np.mean(lengths)),
                         "Connaît sa section": "✅" if any(p.section for p in passages) else "—",
                         "Secondes": round(time.time() - start, 2),
                         "Premier passage": _html.escape((first.text[:150] if first else "")) + "…"})
    finally:
        STRATEGIE_DECOUPAGE = keep
    tableau(pd.DataFrame(rows), title=f"✂️ Les cinq niveaux sur « {_html.escape(document.short)} »",
               note="Les niveaux 1 et 2 ignorent les titres et les tableaux : un passage ne sait plus d'où il vient. "
                    "Les niveaux 4 et 5 indexent une seconde granularité, plus fine — c'est pour cela qu'ils ont plus "
                    "d'unités de recherche que de passages.")

# ── 4. Le banc d'essai : quelle combinaison répond le mieux à VOS questions ─────────────────
def questions_sondes(n=12, seed=0):
    """Questions dont la réponse est connue, fabriquées à partir des documents quand vous n'avez pas
    écrit de jeu de test. Les mots d'UNE phrase doivent ramener une phrase PLUS LOIN dans le même
    sujet : un découpage qui coupe un sujet en deux, ou qui perd sa section, échoue à la sonde. Ce
    n'est qu'un indicateur indirect — écrivez vos propres QUESTIONS_EVALUATION pour une mesure fiable."""
    import random
    rng = random.Random(seed)
    probes = []
    for minimum, gap in ((500, 3), (320, 2), (180, 2)):     # les rapports longs d'abord, les courts marchent quand même
        if len(probes) >= max(3, n // 2):
            break
        pool = [p for p in BIB.passages if p.kind == "text" and len(p.text) > minimum]
        if not pool:
            continue                                   # collection courte : on essaie des passages plus petits
        probes = []
        for p in rng.sample(pool, len(pool)):
            sentences = [s for s in _decouper_phrases(p.text) if 40 < len(s) < 260]
            if len(sentences) < gap:
                continue
            question = " ".join(w for w in sentences[0].split() if len(w) > 3)[:170]
            target = sentences[-1].split()
            if len(target) < 6 or len(question.split()) < 5:
                continue
            probes.append((question, [" ".join(target[:6])]))
            if len(probes) == n:
                break
    return probes

def banc_essai(chunking=("fixed", "sentences", "sections", "hybrid", "smart"),
              reranking=("smart",), questions=None, k=None, show_questions=False):
    """Réindexe avec chaque combinaison et la mesure sur des questions dont la réponse est connue.
    Réponse trouvée = le texte attendu figure parmi les passages remis au modèle."""
    global STRATEGIE_DECOUPAGE, RECLASSEMENT
    k = k or K_RESULTATS
    questions = questions or globals().get("QUESTIONS_EVALUATION") or questions_sondes()
    if not questions:
        encadre("Aucune question d'évaluation : chargez d'abord des documents.", "warning")
        return
    if show_questions:
        tableau(pd.DataFrame([{"Question": q, "Attendu dans le contexte": " | ".join(e)} for q, e in questions]),
                   title=f"Les {len(questions)} questions utilisées par le banc d'essai")
    keep_chunk, keep_rank, keep_context, rows = STRATEGIE_DECOUPAGE, RECLASSEMENT, CONTEXTE_VOISIN, []
    try:
        for level in chunking:
            for mode in reranking:
                STRATEGIE_DECOUPAGE, RECLASSEMENT = level, mode
                globals()["CONTEXTE_VOISIN"] = keep_context and level == "smart"
                start = time.time()
                BIB.reranker = None
                BIB.indexer()
                build = time.time() - start
                found = first = 0
                start = time.time()
                for question, expected in questions:
                    results = BIB.rechercher(question, k)
                    context = " ".join(BIB.contexte_de(p) for p, _ in results).lower()
                    found += any(e.lower() in context for e in expected)
                    if results:
                        first += any(e.lower() in BIB.contexte_de(results[0][0]).lower() for e in expected)
                rows.append({"Découpage": NIVEAUX_DECOUPAGE[level][0], "Reclassement": NIVEAUX_RECLASSEMENT[mode][0],
                             "Passages": len(BIB.passages), "Unités": len(BIB.units),
                             "Réponse dans le contexte": f"{found}/{len(questions)}",
                             "Bon passage en tête": f"{first}/{len(questions)}",
                             "Indexation (s)": round(build, 1),
                             "Par question (ms)": int(1000 * (time.time() - start) / len(questions)),
                             "_score": (found, first)})
    finally:
        STRATEGIE_DECOUPAGE, RECLASSEMENT = keep_chunk, keep_rank
        globals()["CONTEXTE_VOISIN"] = keep_context
        BIB.reranker = None
        BIB.indexer()
    table = pd.DataFrame(rows)
    best = table["_score"].max()
    table["Meilleur"] = ["🏆" if s == best else "" for s in table["_score"]]
    tableau(table.drop(columns="_score"), title="🏁 Banc d'essai sur vos propres documents",
               note="Mesuré sur cette collection uniquement : relancez-le sur la vôtre avant de vous fier à un classement. "
                    "Une combinaison qui gagne d'une seule question fait match nul, elle ne gagne pas.")
    return table

print("✅ T7 prêt · afficher_chaine(), expliquer(\"…\"), comparer_decoupage(), banc_essai().")
````

<div style="color:#5E6964;font-size:14px;line-height:1.45;">Titres, en-têtes répétés, mots coupés, tableaux et OCR.</div></div>

| Opération | Principe | Pourquoi |
|---|---|---|
| **Détecter les titres** | Une ligne écrite nettement plus grande que le corps du texte | Chaque passage connaît sa **section** |
| **Supprimer les en-têtes et pieds de page** | Une ligne répétée en haut ou en bas d'au moins la moitié des pages | « Page 3 » pollue chaque passage |
| **Recoller les mots coupés** | « statis-⏎tiques » → « statistiques » | La recherche retrouve le mot entier |
| **Linéariser les tableaux** | Chaque ligne devient « Libellé — colonne : valeur ; … » | Un nombre isolé ne veut rien dire |
| **OCR** (facultatif) | Reconnaissance de texte sur les pages scannées | Rend visibles les pages scannées |

La normalisation **NFKC** remplace aussi les ligatures typographiques (« ﬁ », « ﬀ ») par des lettres ordinaires : sans elle, « ﬁche » ne correspondrait jamais à « fiche ».

La page ci-dessous montre ce que l'extracteur a reconnu : **vert** pour les titres, **gris** pour les paragraphes, **ocre** pour les tableaux, **rouge barré** pour les en-têtes et pieds de page supprimés.

````python
def page_annotee(d, number, dpi=75):
    doc = pymupdf.open("pdf", d.data)
    page = doc[number - 1]
    styles = {"heading": (0.0, 0.66, 0.42), "paragraph": (0.37, 0.41, 0.39),
              "table": (0.83, 0.60, 0.0), "ocr": (0.05, 0.49, 0.53)}
    for e in d.elements:
        if e.page == number:
            page.draw_rect(pymupdf.Rect(e.box), color=styles[e.type], width=1.4)
    height = page.rect.height
    for l in lignes_de_page(page):
        if (l["box"].y1 < height * 0.1 or l["box"].y0 > height * 0.9) and not any(
                e.page == number and pymupdf.Rect(e.box).intersects(l["box"]) for e in d.elements):
            page.draw_rect(l["box"], color=(0.72, 0.23, 0.18), width=1.2)
            page.draw_line(l["box"].bl, l["box"].tr, color=(0.72, 0.23, 0.18), width=0.8)
    return base64.b64encode(page.get_pixmap(dpi=dpi).tobytes("png")).decode()

d_demo = next((d for d in BIB.documents if any(e.type == "table" for e in d.elements)), BIB.documents[0])
tab = next((e for e in d_demo.elements if e.type == "table"), None)
demo_page = tab.page if tab else 1
block = ""
if tab:
    df_t = pd.DataFrame(tab.table["rows"], columns=tab.table["headers"][:len(tab.table["rows"][0])])
    block = (f'<div class="rg-title" style="margin-top:10px;">Le tableau tel qu\'il est lu…</div>'
             f'{df_t.to_html(index=False, classes="rg-t", border=0)}'
             f'<div class="rg-title" style="margin-top:10px;">… et tel qu\'il est indexé</div>'
             f'<pre style="background:{C["mist"]};padding:10px;border-radius:8px;white-space:pre-wrap;font-size:12px;'
             f'color:{C["ink"]};">{_html.escape(tab.text)}</pre>')
headings = [e for e in d_demo.elements if e.type == "heading"]
outline = "".join(f'<div style="margin-left:{(e.level - 1) * 16}px;">{"📘" if e.level == 1 else "▸"} {_html.escape(e.text[:70])} '
                  f'<span style="color:{C["slate"]};font-size:11px;">p. {e.page}</span></div>'
                  for e in headings[:25])
if len(headings) > 25:
    outline += f'<div style="color:{C["slate"]};">… et {len(headings) - 25} autres titres</div>'
display(HTML(
    f'{_CSS}<div class="rg rg-card"><div class="rg-title">🔍 {_html.escape(d_demo.short)} · page {demo_page} vue par l\'extracteur</div>'
    f'<div style="display:flex;gap:20px;flex-wrap:wrap;align-items:flex-start;">'
    f'<img src="data:image/png;base64,{page_annotee(d_demo, demo_page)}" style="width:320px;border:1px solid {C["sage"]};border-radius:6px;">'
    f'<div style="flex:1;min-width:280px;font-size:13px;"><div class="rg-title">🧭 Plan détecté</div>'
    f'<div style="line-height:1.6;">{outline}</div>{block}</div></div></div>'))
tableau(pd.DataFrame([{"Document": d.short, "Titres": sum(e.type == "heading" for e in d.elements),
                          "Paragraphes": sum(e.type == "paragraph" for e in d.elements),
                          "Tableaux": sum(e.type == "table" for e in d.elements),
                          "Paragraphes OCR": sum(e.type == "ocr" for e in d.elements),
                          "En-têtes retirés": d.info["en-têtes retirés"],
                          "Lignes de sommaire écartées": d.info.get("lignes de sommaire écartées", 0)} for d in BIB.documents]),
           title="Ce que l'extraction a produit, document par document")
````

Le découpage est la décision qui influence le plus les réponses. Découpez trop petit et un passage ne contient plus de quoi répondre ; découpez trop grand et la recherche ne distingue plus les passages les uns des autres. Le notebook propose **cinq niveaux**, de la référence naïve à celui qui mesure le mieux. Changez `STRATEGIE_DECOUPAGE` dans les réglages et réexécutez la partie A.

| `STRATEGIE_DECOUPAGE` | Ce qu'il fait | Ce qu'il vous coûte |
|---|---|---|
| `"fixed"` | **1 · Blocs fixes.** Découpe tous les `TAILLE_MAX` caractères. Ce que fait la quasi-totalité des tutoriels. | Les phrases, les tableaux et les titres sont coupés en plein milieu. Un passage ignore à quelle section il appartient. |
| `"sentences"` | **2 · Fenêtres de phrases.** Regroupe N phrases entières, en glissant. | Rien n'est coupé au milieu d'une phrase, mais les passages sont courts et aveugles aux sections : la recherche trouve des fragments incapables de répondre à eux seuls. |
| `"sections"` | **3 · Passages structurels.** Suit les titres trouvés en partie B2 ; chaque tableau forme un passage à part entière ; le fil d'Ariane des titres voyage avec le texte. | L'unité de recherche est aussi l'unité de lecture, si bien qu'une longue section concourt comme un seul bloc. |
| `"hybrid"` | **4 · Deux granularités.** Indexe les passages structurels **et** de petites fenêtres de phrases qui y renvoient. | Deux fois plus d'unités à indexer (quelques secondes de plus). |
| `"smart"` | **5 · Deux granularités, tableaux et contexte.** Ajoute chaque ligne de tableau comme unité à part entière, et joint à la réponse les passages voisins de la section. | Le plus grand nombre d'unités et l'indexation la plus lente — quelques secondes tout de même sur un rapport de 80 pages. |

### Les cinq règles que respectent les niveaux 3 à 5

| Règle | Réglage | Pourquoi |
|---|---|---|
| Des phrases entières, jusqu'à une taille limite | `TAILLE_MAX = 700` | Une phrase coupée peut être retrouvée, mais pas citée |
| Une phrase répétée entre les passages | `CHEVAUCHEMENT_PHRASES = 1` | La réponse est souvent à cheval sur une frontière |
| Les passages courts fusionnés avec le précédent | `TAILLE_MIN = 280` | Une ligne isolée n'a aucun contexte et ne répond à rien |
| Fil d'Ariane des titres complet dans le texte de recherche | — | « 2. Stratégie › 2.4 Axes prioritaires » sépare des sections aux formulations voisines |
| Tableaux tenus à part, lignes portant leur légende | `EXTRAIRE_TABLEAUX` | Un nombre ne veut rien dire sans les libellés de sa ligne et de sa colonne |

### Les treize étapes, et là où vous pouvez agir

| # | Étape | Ce qu'elle fait | Réglage |
|:-:|---|---|---|
| 1 | **Question** | telle que saisie par l'utilisateur | — |
| 2 | **Question de suivi** | « Et pour les hommes ? » est réécrite en question autonome (B7) | `MEMOIRE_PAR_MODELE` |
| 3 | **Orthographe** | un mot que les documents n'emploient jamais est remplacé par le plus proche qu'ils emploient : *régon* → *région* | — |
| 4 | **Normalisation** | minuscules, accents, racines des mots ; les mots interrogatifs (« nombre de », « liste », « signifie ») sont écartés | `MOTS_VIDES` |
| 5 | **Expansion** | glossaire métier, plus les sigles appris des documents | `GLOSSAIRE` |
| 6 | **BM25** | mots exacts, sur chaque unité de recherche | `STRATEGIE_DECOUPAGE` |
| 7 | **Sémantique** | le sens plutôt que les mots | `UTILISER_EMBEDDINGS` |
| 8 | **Fusion** | les deux classements fusionnés (Reciprocal Rank Fusion) | — |
| 9 | **Reclassement** | l'ordre se décide ici | `RECLASSEMENT` |
| 10 | **Fenêtres → passages** | chaque petite unité restitue le passage qu'un humain devrait lire | `STRATEGIE_DECOUPAGE` |
| 11 | **Garde-fous** | seuil, contrôle de l'année, refus (B5) | `SEUIL_MANUEL` |
| 12 | **Réponse** | rédigée par le modèle, ou extraite du passage | Groq / Ollama |
| 13 | **Contrôles** | chiffres vérifiés, termes introuvables, meilleure réponse écartée par un filtre (B9) | — |

L'étape 7 mérite un mot : **sans `sentence-transformers`, le notebook se rabat sur une approximation hors ligne (LSA sur n-grammes de caractères)**, qui rapproche des orthographes plutôt que des sens. Elle ne tombe jamais en panne, mais elle ne peut pas relier « participer à l'évaluation » et « la Banque a mené des missions de diagnostic en ». Si vos questions emploient les mots de vos utilisateurs plutôt que ceux du rapport, installez le modèle neuronal — `INSTALLER_EMBEDDINGS = True`.

### Les quatre niveaux de reclassement

Le classement fusionné est rarement le bon ordre. Le reclassement réévalue la liste courte ; `RECLASSEMENT` choisit comment.

| `RECLASSEMENT` | Ce qu'il fait | Coût |
|---|---|---|
| `"none"` | **0 · Aucun.** Conserve l'ordre de la fusion. | Gratuit |
| `"coverage"` | **1 · Couverture de la question.** Part des termes pondérés de la question présents dans le passage. Un mot rare pèse plus qu'un mot fréquent ; un mot absent de toute la collection pèse le plus, et c'est ce qui rend un refus possible. | Gratuit |
| `"smart"` | **2 · Couverture, densité et type de réponse.** Sur un long rapport, beaucoup de passages couvrent toute la question : les scores égaux sont alors départagés par **ce que demande la question** — une question « combien » préfère un passage portant une quantité réelle (et non « voir la figure 8 »), une question « liste » un passage portant une énumération — puis par la densité des termes. | Gratuit |
| `"cross-encoder"` | **3 · Cross-encoder neuronal.** Un modèle lit la question et le passage *ensemble* et note la paire. Le plus précis, et le plus lent : il exige `sentence-transformers` et quelques centaines de millisecondes par question. | ~0,3–1 s/question |

>
> | Reclassement | Réponse dans le contexte | Bon passage en tête |

> | 0 · Aucun | 10/12 | 6/12 |
> | 1 · Couverture de la question | 12/12 | 8/12 |
> | 2 · Couverture, densité et type de réponse | **12/12** | **10/12** |
>

Lancez vous-même `banc_essai(reranking=("smart", "cross-encoder"))` — sur les bancs d'essai publics, un cross-encoder l'emporte généralement, au prix de la latence.

> 💡 **Le glossaire vous appartient.** Enrichissez `GLOSSAIRE` (cellule T4) du vocabulaire de vos utilisateurs — « cadre » → « diagnostic du modèle de maturité » sur le guide de l'AfDB, par exemple. C'est souvent l'amélioration la plus rentable de toutes, et c'est la bonne réponse quand une question échoue systématiquement parce que vos utilisateurs et vos rapports emploient des mots différents.

2. **Contrôle de l'année** : une année mentionnée dans la question doit figurer dans les passages ;
3. **Consigne stricte** : répondre uniquement à partir des sources, les citer, et refuser avec une phrase fixe sinon.

### Comment le seuil est calculé

**Confiance d'une question** = poids des mots de la question retrouvés dans le meilleur passage ÷ poids de tous les mots de la question. Les mots rares pèsent plus ; les mots retrouvés **nulle part** dans les documents (comme « hôpitaux ») pèsent le plus, car ils signalent une question hors sujet.

**Calibration** : le notebook pose deux séries de questions de référence — des questions construites à partir des **titres de section** (la réponse existe) et des questions **hors sujet** (aucune réponse). Le seuil est fixé juste au-dessus du meilleur score hors sujet :

> **Seuil = meilleure confiance hors sujet + 0,05** (plafonné à 0,95)

> ⚠️ Ce seuil est un point de départ : ajustez-le dans les interfaces si l'assistant refuse trop souvent, ou pas assez. Pour le fixer vous-même : `SEUIL_MANUEL = 0.5`.

````python
cal = BIB.calibration
fig, ax = plt.subplots(figsize=(8.6, 2.8))
for rank, (exists, colour) in enumerate([(False, C["brick"]), (True, C["green"])]):
    v = cal.loc[cal["Réponse existe"] == exists, "Confiance"]
    ax.scatter(v, np.full(len(v), rank) + np.linspace(-0.13, 0.13, len(v)), s=95, color=colour,
               alpha=0.85, edgecolor="white", linewidth=1.2, zorder=3)
ax.axvline(BIB.threshold, color=C["ochre"], linewidth=2.2, linestyle="--")
ax.axvspan(0, BIB.threshold, color=C["brick"], alpha=0.05); ax.axvspan(BIB.threshold, 1.02, color=C["green"], alpha=0.05)
ax.text(BIB.threshold, 1.55, f"  SEUIL = {BIB.threshold:.2f}", color=C["ochre"], fontweight="bold", va="center")
ax.set_yticks([0, 1], ["Hors sujet", "Titres du document"]); ax.set_xlim(0, 1.02); ax.set_ylim(-0.6, 1.8)
ax.set_xlabel("Confiance du meilleur passage"); ax.set_title("Calibration du seuil sur vos documents")
plt.tight_layout(); fig.savefig(DOSSIER_SORTIE / "calibration_seuil.png", dpi=160); plt.show()
tableau(cal.sort_values("Confiance", ascending=False), title="Questions de référence")

for q in ["Quel est le prix du maïs au kilogramme ?", "Quel sera le taux de chômage en 2030 ?"]:
    BIB.afficher(BIB.repondre(q))
````

Une question de suivi n'a pas de sens à elle seule : « Et pour les hommes ? » ne contient aucun indicateur. Avant de rechercher, l'assistant **la réécrit en question autonome** :

| Méthode | Principe | Quand |
|---|---|---|
| **Par le modèle** | Le modèle reçoit les derniers échanges et réécrit la question | Un modèle est disponible et `MEMOIRE_PAR_MODELE = True` |
| **Par règles** | La question précédente est reprise et la valeur qui change (région, sexe, trimestre, année, milieu, âge, indicateur) est **remplacée** | Sans modèle, ou en solution de repli |

La réécriture est **toujours affichée** (« 🧠 Compris comme : … ») pour que l'utilisateur puisse la vérifier.

> ⚠️ **La mémoire peut se tromper.** Si l'utilisateur change de sujet avec une question très courte, celle-ci peut être rattachée à tort à la précédente. D'où l'affichage systématique, et la possibilité de désactiver la mémoire dans les interfaces.

````python
import ipywidgets as widgets

HISTORIQUE = []
SUGGESTIONS = questions_suggerees(5)

def _entete():
    model_label = f"{LLM.name} · {LLM.model}" if LLM.available else "mode extractif"
    return (f'{_CSS}<div class="rg" style="background:linear-gradient(120deg,{C["forest"]} 0%,{C["green"]} 100%);'
            f'border-radius:14px;padding:14px 20px;color:#fff;"><div style="color:{C["gold"]};font-size:11px;'
            f'font-weight:700;letter-spacing:2.5px;">ASSISTANT DOCUMENTAIRE</div>'
            f'<div style="font-size:20px;font-weight:800;margin:2px 0;">💬 {len(BIB.documents)} document(s) · '
            f'{len(BIB.passages)} passages</div><div style="font-size:12.5px;color:#E6F6EE;">modèle : '
            f'{_html.escape(model_label)} · recherche : {_html.escape(BIB.semantic.name)}</div></div>')

def _fil():
    if not HISTORIQUE:
        return (f'{_CSS}<div class="rg rg-card" style="text-align:center;color:{C["slate"]};padding:26px;">'
                'Posez une question, ou cliquez sur une suggestion.</div>')
    bubbles = "".join(
        f'<div style="display:flex;justify-content:flex-end;margin:12px 0 6px 0;"><div style="background:{C["dark_green"]};'
        f'color:#fff;border-radius:14px 14px 2px 14px;padding:8px 14px;max-width:80%;font-size:14px;">'
        f'{_html.escape(r["question"])}</div></div><div style="display:flex;gap:8px;align-items:flex-start;">'
        f'<div style="font-size:22px;">🤖</div><div style="flex:1;">{BIB.carte_html(r, ui_vignettes.value)}</div></div>'
        for r in reversed(HISTORIQUE))
    return (f'{_CSS}<div class="rg" style="max-height:720px;overflow-y:auto;padding:4px 6px;border:1px solid '
            f'{C["sage"]};border-radius:12px;background:{C["mist"]};">{bubbles}</div>')

ui_question = widgets.Text(placeholder="Posez une question sur les documents…", layout=widgets.Layout(flex="1", width="auto"))
ui_envoyer = widgets.Button(description="Envoyer", button_style="success", icon="paper-plane", layout=widgets.Layout(width="120px"))
ui_etat = widgets.HTML("")
ui_docs = [widgets.Checkbox(value=True, description=d.short, indent=False, layout=widgets.Layout(width="auto", margin="0 14px 0 0"))
           for d in BIB.documents]
ui_k = widgets.IntSlider(value=K_RESULTATS, min=1, max=8, description="Passages", continuous_update=False)
ui_seuil = widgets.FloatSlider(value=float(BIB.threshold), min=0.0, max=1.0, step=0.01, description="Seuil",
                                   continuous_update=False, readout_format=".2f")
ui_garde_fous = widgets.Checkbox(value=True, description="Garde-fous actifs", indent=False)
ui_memoire = widgets.Checkbox(value=True, description="Mémoire de conversation", indent=False)
ui_vignettes = widgets.Checkbox(value=True, description="Afficher les pages", indent=False)
ui_effacer = widgets.Button(description="Nouvelle conversation", icon="refresh", layout=widgets.Layout(width="200px"))
ui_exporter = widgets.Button(description="Exporter", icon="download", layout=widgets.Layout(width="120px"))
ui_utile = widgets.Button(description="Utile", icon="thumbs-up", layout=widgets.Layout(width="100px"))
ui_a_revoir = widgets.Button(description="À revoir", icon="thumbs-down", layout=widgets.Layout(width="130px"))
ui_commentaire = widgets.Text(placeholder="Commentaire facultatif sur la dernière réponse…", layout=widgets.Layout(flex="1", width="auto"))
ui_fil = widgets.HTML(_fil())

def donner_avis(useful):
    if not HISTORIQUE:
        ui_etat.value = "Posez d'abord une question."
        return
    enregistrer_avis(HISTORIQUE[-1], useful, ui_commentaire.value.strip(), "notebook")
    ui_commentaire.value = ""
    ui_etat.value = (f'<span style="color:{C["dark_green"]};">Merci ! Avis enregistré '
                       f'({len(AVIS)} au total, fichier avis_utilisateurs.csv).</span>')

def envoyer(_=None, question=None):
    q = (question or ui_question.value).strip()
    if not q:
        return
    docs = [d.id for d, c in zip(BIB.documents, ui_docs) if c.value]
    if not docs:
        ui_etat.value = f'<span style="color:{C["brick"]};">Cochez au moins un document.</span>'
        return
    ui_question.value = ""
    ui_envoyer.disabled = True
    ui_etat.value = f'<span style="color:{C["teal"]};">⏳ Recherche dans {len(docs)} document(s)…</span>'
    try:
        r = BIB.repondre(q, history=HISTORIQUE if ui_memoire.value else None, docs=docs, k=ui_k.value,
                       threshold=ui_seuil.value, guardrails=ui_garde_fous.value, memory=ui_memoire.value)
        HISTORIQUE.append(r)
        ui_etat.value = (f'<span style="color:{C["slate"]};font-size:12px;">{len(HISTORIQUE)} échange(s) · '
                           f'dernière réponse en {r["duree_s"]} s' + (f' · {LLM.calls} appel(s) au modèle' if LLM.available else "") + "</span>")
    except Exception as error:
        ui_etat.value = f'<span style="color:{C["brick"]};">⛔ {_html.escape(str(error)[:200])}</span>'
    finally:
        ui_envoyer.disabled = False
        ui_fil.value = _fil()

def nouvelle_conversation(_=None):
    HISTORIQUE.clear()
    ui_fil.value = _fil()
    ui_etat.value = ""

def _serialiser(r):
    return {"question": r["question"], "question_autonome": r["question_autonome"], "reponse": r["reponse"],
            "refus": r["refus"], "confiance": round(r["confiance"], 3), "motif": r["motif"], "mode": r["mode"],
            "garde_fous": r["garde_fous"], "verification": r["verification"], "heure": r["heure"],
            "sources": [{"document": p.doc, "section": p.section, "pages": p.pages, "nature": p.kind,
                         "confiance": round(s, 3), "text": p.text} for p, s in r["passages"]]}

def exporter(_=None):
    if not HISTORIQUE:
        ui_etat.value = "Rien à exporter pour l'instant."
        return
    stamp = time.strftime("%Y%m%d_%H%M%S")
    (DOSSIER_SORTIE / f"conversation_{stamp}.json").write_text(
        json.dumps({"documents": [d.title for d in BIB.documents], "turns": [_serialiser(r) for r in HISTORIQUE]},
                   ensure_ascii=False, indent=2), encoding="utf-8")
    html_path = DOSSIER_SORTIE / f"conversation_{stamp}.html"
    html_path.write_text(f'<html><head><meta charset="utf-8"><title>Conversation</title></head>'
                         f'<body style="max-width:900px;margin:auto;">{_entete()}{_fil()}</body></html>', encoding="utf-8")
    ui_etat.value = f"✅ Conversation exportée : <code>{html_path.name}</code> (et sa version JSON)"
    if ENV["name"] == "colab":
        from google.colab import files
        files.download(str(html_path))

buttons = []
for s in SUGGESTIONS:
    b = widgets.Button(description=(s[:44] + "…") if len(s) > 45 else s, tooltip=s,
                       layout=widgets.Layout(width="auto", margin="2px 4px 2px 0"))
    b.on_click(lambda _, s=s: envoyer(question=s))
    buttons.append(b)
ui_envoyer.on_click(envoyer)
ui_utile.on_click(lambda _: donner_avis(True))
ui_a_revoir.on_click(lambda _: donner_avis(False))
ui_effacer.on_click(nouvelle_conversation)
ui_exporter.on_click(exporter)
ui_vignettes.observe(lambda _: setattr(ui_fil, "value", _fil()), names="value")
try:
    import warnings
    with warnings.catch_warnings():
        warnings.simplefilter("ignore")
        ui_question.on_submit(envoyer)
except Exception:
    pass

panneau_reglages = widgets.Accordion(children=[widgets.VBox([
    ui_k, ui_seuil, widgets.HBox([ui_garde_fous, ui_memoire, ui_vignettes])])])
panneau_reglages.set_title(0, "⚙️ Réglages")
panneau_reglages.selected_index = None

INTERFACE = widgets.VBox([
    widgets.HTML(_entete()),
    widgets.HTML('<div style="font-size:12.5px;margin:10px 0 2px 0;"><b>Documents interrogés</b></div>'),
    widgets.HBox(ui_docs, layout=widgets.Layout(flex_flow="row wrap")),
    widgets.HTML('<div style="font-size:12.5px;margin:8px 0 2px 0;"><b>Suggestions</b></div>'),
    widgets.HBox(buttons, layout=widgets.Layout(flex_flow="row wrap")),
    widgets.HBox([ui_question, ui_envoyer], layout=widgets.Layout(margin="8px 0 2px 0")),
    ui_etat, panneau_reglages,
    widgets.HBox([widgets.HTML('<span style="font-size:12.5px;margin-right:6px;"><b>Dernière réponse :</b></span>'),
                  ui_utile, ui_a_revoir, ui_commentaire]),
    widgets.HBox([ui_effacer, ui_exporter]), ui_fil,
], layout=widgets.Layout(width="100%", max_width="980px"))
display(INTERFACE)
````

**Gradio** est une bibliothèque Python libre et gratuite qui transforme des fonctions Python en page web interactive — sans HTML ni JavaScript. L'application porte toutes les fonctionnalités de l'assistant sur une page web :

- **glisser-déposer** de nouveaux PDF, indexés à la volée ;
- **filtrer** les documents interrogés ;
- **conversation avec mémoire**, avec la réécriture affichée ;
- **sources** avec vignettes de page surlignées et contrôles automatiques ;
- **👍 / 👎** sous chaque réponse ;
- **réglages** : nombre de passages, seuil, garde-fous, mémoire.

Sur **Colab** et **Kaggle**, Gradio crée un **lien public temporaire** (`https://….gradio.live`), valable tant que la session tourne et pour une durée limitée. **En local**, l'application s'ouvre à l'adresse `http://127.0.0.1:7860`. L'application tourne **dans votre session de notebook** : si la session s'arrête, la page cesse de fonctionner.

> ⚠️ **Confidentialité.** Toute personne disposant du lien public peut interroger vos documents et téléverser des fichiers. **Définissez `MOT_DE_PASSE_WEB`** avant de partager un lien, et n'utilisez pas de lien public pour des documents non publiés.

> ⚠️ **Application partagée.** Tous les utilisateurs d'un même lien partagent la même bibliothèque : un PDF téléversé par l'un devient interrogeable par tous. Chaque utilisateur conserve sa propre conversation.

````python
def _installer_gradio():
    try:
        import gradio as gr
        if int(gr.__version__.split(".")[0]) >= 5:
            return gr
        print(f"Gradio {gr.__version__} est trop ancien : mise à jour…")
    except ImportError:
        print("Installation de Gradio (environ une minute)…")
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-U", "gradio>=5"], check=False)
    importlib.invalidate_caches()
    import gradio as gr
    if int(gr.__version__.split(".")[0]) < 5:
        raise RuntimeError("Redémarrez la session (Exécution → Redémarrer la session), puis relancez le notebook.")
    return gr

def _libelles_docs():
    return [f"{d.id} · {d.short}" for d in BIB.documents]

def _resume_bibliotheque():
    return (f"**{len(BIB.documents)} document(s)** · {len(BIB.passages)} passages · seuil {BIB.threshold:.2f} · "
            + (f"modèle **{LLM.name}**" if LLM.available else "mode extractif"))

def creer_application():
    gr = _installer_gradio()
    import inspect
    theme = gr.themes.Soft(primary_hue="emerald", secondary_hue="amber", neutral_hue="slate")
    css = ".gradio-container {max-width: 1250px !important} footer {display: none !important}"
    blocks_args = {"title": "Assistant documentaire RAG"}
    blocks_params = inspect.signature(gr.Blocks.__init__).parameters
    if "theme" in blocks_params:
        blocks_args.update(theme=theme, css=css)
    chat_params = inspect.signature(gr.Chatbot.__init__).parameters
    chat_args = {"height": 470, "label": "Conversation"}
    if "type" in chat_params:
        chat_args["type"] = "messages"
    if "feedback_options" in chat_params:
        chat_args["feedback_options"] = ["Utile", "À revoir"]

    def reponse_web(question, messages, memory_state, docs, k, threshold, guardrails, memory):
        messages = list(messages or [])
        if not question or not question.strip():
            return "", messages, memory_state, gr.update()
        ids = [int(x.split(" · ")[0]) for x in (docs or [])]
        if not ids:
            messages.append({"role": "assistant", "content": "Cochez au moins un document."})
            return "", messages, memory_state, gr.update()
        hist = [{"question_autonome": h["question_autonome"], "reponse": h["reponse"]} for h in (memory_state or [])]
        r = BIB.repondre(question.strip(), history=hist if memory else None, docs=ids, k=int(k),
                       threshold=float(threshold), guardrails=bool(guardrails), memory=bool(memory))
        text = r["reponse"]
        if r["reformulee"]:
            text = f"_🧠 Comprise comme : {r['question_autonome']}_\n\n" + text
        if r["motif"]:
            text += f"\n\n_Motif : {r['motif']}_"
        if not r["refus"]:
            text += "\n\n" + "\n".join(f"**[{i}]** {p.citation}" for i, (p, _) in enumerate(r["passages"], 1))
        messages += [{"role": "user", "content": question.strip()}, {"role": "assistant", "content": text}]
        memory_state = (memory_state or []) + [{"question": r["question"], "question_autonome": r["question_autonome"],
                                                "reponse": r["reponse"], "citations": r["citations"]}]
        return "", messages, memory_state, f"{_CSS}{BIB.carte_html(r)}"

    def noter(data: gr.LikeData, memory_state):
        index = data.index[0] if isinstance(data.index, (list, tuple)) else data.index
        turns = memory_state or []
        position = int(index) // 2
        if not turns or position >= len(turns):
            return "Avis non enregistré : échange introuvable."
        useful = data.liked if isinstance(data.liked, bool) else str(data.liked).lower() in ("true", "like", "utile")
        enregistrer_avis(turns[position], useful, channel="web")
        return f"Merci ! Avis enregistré ({len(AVIS)} au total)."

    def ajouter_pdfs(files, docs):
        if not files:
            return gr.update(), _resume_bibliotheque(), ""
        paths = [f if isinstance(f, str) else getattr(f, "name", str(f)) for f in files]
        added, errors = BIB.charger(paths)
        BIB.indexer()
        choices = _libelles_docs()
        ticked = list(docs or []) + [f"{d.id} · {d.short}" for d in added]
        message = f"✅ {len(added)} document(s) ajouté(s)." + "".join(f"\n⚠️ {n} : {e}" for n, e in errors)
        return gr.update(choices=choices, value=[c for c in choices if c in ticked]), _resume_bibliotheque(), message

    with gr.Blocks(**blocks_args) as app:
        gr.HTML('<div style="background:linear-gradient(120deg,#00553A 0%,#00A86A 100%);border-radius:14px;'
                'padding:16px 22px;color:#fff;font-family:\'Segoe UI\',Calibri,sans-serif;">'
                '<div style="color:#F5C242;font-size:11px;font-weight:700;letter-spacing:2.5px;">ASSISTANT DOCUMENTAIRE · RAG</div>'
                '<div style="font-size:24px;font-weight:800;">💬 Posez vos questions sur vos rapports PDF</div>'
                '<div style="font-size:13px;color:#E6F6EE;">Des réponses sourcées, les pages surlignées, et un refus '
                'quand l\'information n\'est pas dans les documents.</div></div>')
        memory_state = gr.State([])
        with gr.Row():
            with gr.Column(scale=1, min_width=300):
                summary = gr.Markdown(_resume_bibliotheque())
                docs = gr.CheckboxGroup(choices=_libelles_docs(), value=_libelles_docs(), label="Documents interrogés")
                upload = gr.File(label="Ajouter des PDF", file_count="multiple", file_types=[".pdf"])
                add_button = gr.Button("📥 Ajouter et indexer")
                add_message = gr.Markdown("")
                with gr.Accordion("⚙️ Réglages", open=False):
                    k = gr.Slider(1, 8, value=K_RESULTATS, step=1, label="Passages fournis")
                    threshold = gr.Slider(0.0, 1.0, value=float(BIB.threshold), step=0.01, label="Seuil de confiance")
                    guardrails = gr.Checkbox(value=True, label="Garde-fous actifs")
                    memory = gr.Checkbox(value=True, label="Mémoire de conversation")
            with gr.Column(scale=2):
                chat = gr.Chatbot(**chat_args)
                with gr.Row():
                    question = gr.Textbox(placeholder="Posez une question sur les documents…", show_label=False,
                                          scale=5, autofocus=True)
                    send_button = gr.Button("Envoyer", variant="primary", scale=1)
                gr.Examples(examples=[[s] for s in SUGGESTIONS], inputs=[question], label="Suggestions")
                with gr.Row():
                    new_chat = gr.Button("🔄 Nouvelle conversation", size="sm")
                    feedback_message = gr.Markdown("_Utilisez 👍 / 👎 sous chaque réponse pour donner votre avis._")
                with gr.Accordion("📄 Sources de la dernière réponse", open=True):
                    sources = gr.HTML("<i>Les sources apparaîtront ici.</i>")
        inputs = [question, chat, memory_state, docs, k, threshold, guardrails, memory]
        outputs = [question, chat, memory_state, sources]
        send_button.click(reponse_web, inputs, outputs)
        question.submit(reponse_web, inputs, outputs)
        new_chat.click(lambda: ([], [], "<i>Les sources apparaîtront ici.</i>"), None, [chat, memory_state, sources])
        add_button.click(ajouter_pdfs, [upload, docs], [docs, summary, add_message])
        if hasattr(chat, "like"):
            chat.like(noter, [memory_state], [feedback_message])
    app._launch_args = {} if "theme" in blocks_params else {"theme": theme, "css": css}
    return app

def lancer_application(app):
    import inspect
    auth = (UTILISATEUR_WEB, MOT_DE_PASSE_WEB) if MOT_DE_PASSE_WEB else None
    options = dict(share=PARTAGER_LIEN, auth=auth, prevent_thread_lock=True, quiet=True, show_error=True,
                   inbrowser=(ENV["name"] == "local"), inline=(ENV["name"] == "colab"), **app._launch_args)
    params = inspect.signature(app.launch).parameters
    options = {k: v for k, v in options.items() if k in params}
    try:
        app.launch(**options)
    except Exception as error:
        if options.get("share"):
            print(f"⚠️ Lien public indisponible ({type(error).__name__}) : lancement en local.")
            options["share"] = False
            app.launch(**options)
        else:
            raise
    public_url = getattr(app, "share_url", None)
    local_url = getattr(app, "local_url", None) or "http://127.0.0.1:7860"
    lines = [f'<div style="font-size:14px;margin:4px 0;">🔗 <b>Lien public :</b> <a href="{public_url}" target="_blank">{public_url}</a></div>'
             if public_url else "",
             f'<div style="font-size:14px;margin:4px 0;">💻 <b>Adresse locale :</b> <code>{local_url}</code></div>',
             f'<div style="font-size:13px;margin:4px 0;">🔒 Accès : ' + (f'identifiant <code>{UTILISATEUR_WEB}</code> et mot de passe défini'
             if auth else f'<span style="color:{C["brick"]};"><b>aucun mot de passe</b></span>') + '</div>']
    display(HTML(f'{_CSS}<div class="rg rg-card" style="border-left:6px solid {C["green"]};">'
                 f'<div class="rg-title">🌐 Application web en service</div>{"".join(lines)}'
                 f'<div class="rg-note">Pour l\'arrêter : exécutez <code>APP.close()</code>. Elle s\'arrête aussi à la fin de la session.</div></div>'))
    if public_url and not auth:
        encadre("Le lien est public et <b>sans mot de passe</b> : définissez <code>MOT_DE_PASSE_WEB</code> avant de le partager.", "warning")

if LANCER_INTERFACE_WEB:
    APP = creer_application()
    lancer_application(APP)
else:
    encadre("Application web non lancée (<code>LANCER_INTERFACE_WEB = False</code>).", "info")
````

### Limites de cet assistant

- **Les mises en page complexes** (colonnes multiples, tableaux sans bordures, notes de bas de page) sont imparfaitement extraites. Vérifiez l'étape B2 sur vos documents.
- **Les graphiques ne sont pas lus**, seulement leurs titres et légendes.
- **Le mode extractif** recopie des phrases ou des lignes de tableau : il ne sait ni **calculer** ni **classer** (« quelle région a le taux de chômage le plus élevé ? »). De telles questions demandent un modèle de langue.
- **La mémoire par règles** reconnaît les changements de région, de sexe, de trimestre, d'année, de milieu, d'âge et d'indicateur. Au-delà, un modèle est nécessaire.
- **La vérification des chiffres est littérale** : un chiffre calculé par le modèle (une différence, une moyenne) est signalé comme absent, même s'il est correct.
- **Les ressources linguistiques** (mots vides, radicaux, glossaire) sont fournies pour l'anglais et le français ; les autres langues demandent leurs propres ressources.
- **Sans `sentence-transformers`**, la recherche sémantique est une approximation hors ligne qui rapproche des orthographes, pas des sens. Elle ne peut pas faire le lien entre une question écrite avec les mots de vos utilisateurs et un rapport écrit avec ceux de l'institution. C'est la limite la plus forte de la configuration par défaut : `INSTALLER_EMBEDDINGS = True` la lève.
- **La tolérance orthographique a un angle mort délibéré** : un mot que les documents n'emploient jamais n'est corrigé que s'il en existe un très proche. « maïs » reste inconnu — ce qui est précisément ce qui permet à l'assistant de refuser une question hors périmètre.
- **Sur les rapports très longs**, la plupart des passages contiennent tous les mots d'une question, si bien que le score de confiance sature près de 1,00 : il mesure alors la couverture, et non la précision. Le classement repose sur la densité des termes et sur la recherche fusionnée, et le seuil est plafonné à 0,70.
- **La vérification finale reste humaine** : l'assistant donne la page, un statisticien vérifie avant toute diffusion.

### ✅ Point de contrôle

<details><summary><b>1. Pourquoi afficher « Compris comme… » ?</b></summary>
Parce que la mémoire peut se tromper. Les utilisateurs voient la question réellement recherchée et peuvent la reformuler si elle ne correspond pas à leur intention.
</details>

<details><summary><b>2. Pourquoi limiter le nombre de passages issus d'un même document ?</b></summary>
Pour qu'une question de comparaison reçoive des extraits de plusieurs documents, plutôt que trois passages du même rapport.
</details>

<details><summary><b>3. Pourquoi un mot de passe sur l'application web ?</b></summary>
Un lien public est accessible à quiconque le possède : sans mot de passe, n'importe qui peut interroger vos documents ou en téléverser de nouveaux.
</details>

<details><summary><b>4. Une réponse fausse avec le bon passage : que faut-il corriger ?</b></summary>
La <b>génération</b> : la consigne ou le modèle. Avec un passage hors sujet : la <b>recherche</b> (extraction, glossaire, seuil).
</details>

<details><summary><b>5. Que signifie « chiffre absent des sources » ?</b></summary>
Un nombre de la réponse ne figure dans aucun des passages fournis : soit le modèle l'a inventé ou arrondi, soit il l'a calculé. Dans les deux cas, vérifiez-le à partir des sources citées.
</details>
