# Online price index — analysis

**Who it is for.** There is nothing to install and not a single line of code to write.
Each cell is run with the ▶ button on its left, from top to bottom.

**What this notebook does.** It starts from the price records produced by notebooks 1
and 2, **pools** them, checks them, classifies the products in the international COICOP
classification, **compares the classification methods against one another**, computes a
price index and produces a report to download.

**What it does not do.** It never visits a retail website. Collection takes place
elsewhere, on the institute's workstation. This notebook works on files
only.

---

**Everything is set in the next cell.** The default values lead to the demonstration
dataset: you can run the whole notebook without changing anything, and
see what a complete result looks like.

## 4 · Classifying the products

Each product is assigned to a **COICOP** class, the classification used by
statistical institutes. Two methods:

| | |
|---|---|
| **`regles`** | keywords and regular expressions. Free, instantaneous, identical on every run. |
| **`mixte`** | the language model when it is confident, the rules otherwise. |

Cell 5 will tell you which one is best **on your data**. In the
meantime, `regles` is a good starting point: it requires no key.

```python
#@title Classification { display-mode: "form" }
S = config.Settings(mode="live", country=(MANIFESTE or {}).get("pays", "CI"),
                    llm_provider=("none" if MODELE == "aucun" else MODELE),
                    model_name=("" if NOM_DU_MODELE == "automatique"
                                else NOM_DU_MODELE),
                    llm_max_items=MAX_PRODUITS,
                    contact_email="analyse@local",
                    root=str(Path(DOSSIER_SORTIE) if DOSSIER_SORTIE.strip()
                             else (Path("/content/sorties") if SUR_COLAB
                                   else RACINE / "sorties")))

# The key never appears in this notebook. Three routes, and none of them writes
# anything to disk:
#
#   secrets        the Colab vault (🔑 icon). The key is read by the kernel,
#                  never by the cell, and does not travel with the shared
#                  notebook. This is the recommended route.
#   saisie_masquee `getpass`: you type it, it is not displayed and is
#                  not kept in the cell output.
#   variable       an environment variable or a `.env` file next to the
#                  project — handy locally, not to be committed.
#
# A text field in the form would have been simpler, and that is
# precisely why there is none: its value would be saved with the
# notebook, and would travel with it.
CLIENT = None
if METHODE != "regles" and MODELE in ("groq", "gemini"):
    from src.utils import get_secret, mask

    variable = config.SECRET_NAMES[MODELE]
    cle = get_secret(variable, ask=(SAISIE_CLE == "saisie_masquee"))
    if cle:
        os.environ[variable] = cle
        encadre("Key <code>{}</code> found: {}. It is neither displayed nor "
                "written to disk.".format(variable, mask(cle)), *SUCCES)
    else:
        ou = {"secrets": "the Colab vault (🔑 icon on the left, then "
                         "“Add new secret” and grant the notebook access)",
              "saisie_masquee": "the masked input — you left it empty",
              "variable": "the environment variable {} or a .env file"
                          .format(variable)}[SAISIE_CLE]
        encadre("No <code>{}</code> key in {}. Classification will be done "
                "by the rules — which remains entirely usable."
                .format(variable, ou), *ATTENTION)

if METHODE != "regles":
    CLIENT = pipeline.make_llm_client(S)
    display(viz.table(CLIENT.status(), caption="Providers tried, in this order"))
    # The list in the form is that of the models known at the time of writing.
    # Providers add and withdraw models: so we query yours to find out
    # what it offers today.
    for fournisseur in CLIENT.providers:
        if getattr(fournisseur, "name", "") != MODELE:
            continue
        try:
            offerts = fournisseur.list_models()
        except Exception:
            offerts = []
        if offerts:
            display(viz.table(pd.DataFrame({"available model": sorted(offerts)[:25]}),
                              caption="{} offers {} model(s) — enter one "
                                      "in NOM_DU_MODELE".format(MODELE, len(offerts))))
    effectif = getattr(CLIENT, "active_provider", None)
    if effectif in (None, "rules", "règles"):
        encadre("No model reachable — key missing or service unavailable. "
                "Classification will be done by the rules. The "
                "“détail” column above says what is missing.", *ATTENTION)
    else:
        encadre("Model used: <b>{}</b>.".format(effectif), *SUCCES)
else:
    encadre("<b>Rules</b> method: no model called, no cost, "
            "identical result on every run.")

try:
    ENRICHI, OBS = pipeline.enrich_panel(OBS_BRUT, CLIENT, S, progress=True)
except Exception as erreur:
    encadre("Classification failed: {}. Check that the file really comes "
            "from the collection notebook.".format(erreur), *ALERTE)
    raise SystemExit("Classification impossible.")

part = OBS["coicop"].notna().mean() if "coicop" in OBS else 0
encadre("{:.0%} of price records are assigned to a COICOP class, across {} distinct "
        "products.".format(part, OBS["sku"].nunique()), *SUCCES)
if "coicop_source" in ENRICHI:
    display(viz.table(ENRICHI["coicop_source"].value_counts().rename("produits")
                      .reset_index(), caption="Where each classification comes from"))
display(viz.table(OBS.groupby("division", dropna=False)["sku"].nunique()
                  .rename("produits").reset_index().head(15),
                  caption="Products by COICOP division"))
```

## 5 · Test bench — which method to choose

A classification method is not chosen in advance: it is **measured**.
This cell runs each method on a dataset whose correct answer is
known, and reports precision, recall, F1, coverage, time and cost.

**Two figures never to be separated.** *Coverage* is the share of products
to which a method dares to assign a class; *accuracy* is the share of
correct answers. A method that classifies 10% of products without ever being
wrong is not perfect — it is unusable.

<div style='font-family:Segoe UI,Roboto,Helvetica,Arial,sans-serif;background:#FBF5E6;border-left:6px solid #C8A13A;border-radius:6px;padding:12px 16px;margin:8px 0;font-size:14px;color:#2B2B2B'><b>BANC_ESSAI = “exemple” does not compare methods.</b><br>The synthetic dataset draws its labels from the same templates the rules were written against: they score <b>100%</b> on it, and that figure proves nothing. It only checks that the measurement chain works.<br><br>To decide, you need <code>BANC_ESSAI = "mes_donnees"</code> and a sample of your own price records annotated by hand. The next cell prepares it.</div>

```python
#@title Compare the methods { display-mode: "form" }
COMPARAISON = None
if BANC_ESSAI == "aucun":
    encadre("Test bench skipped. You can enable it in cell 1.")
else:
    if BANC_ESSAI == "exemple":
        chemin_ref = RACINE / "data" / "exemple" / "benchmark_exemple_realiste.csv"
        encadre("<b>Synthetic</b> dataset: a check of the measurement chain, not "
                "a comparison of methods. The rules score close to 100% on it "
                "because they share its vocabulary.", *ATTENTION)
    else:
        chemin_ref = Path(CHEMIN_ANNOTE)

    REF = evaluation.charger_reference(chemin_ref)
    if not REF["ok"]:
        for p in REF["problèmes"]:
            encadre(p, *ALERTE, titre="Reference dataset unusable")
        encadre("Run the next cell to produce a sample to "
                "annotate, then come back here.", *ATTENTION)
    else:
        for a in REF.get("avertissements", []):
            encadre(a, *ATTENTION)
        methodes = ("regles",) if CLIENT is None else ("regles", "llm", "mixte")
        if CLIENT is None:
            encadre("No model available: only the rules are measured. "
                    "To compare, choose a MODELE and provide a key.",
                    *ATTENTION)
        COMPARAISON = evaluation.comparer(
            REF["produits"], REF["vérité"], methodes=methodes, client=CLIENT,
            reglages=S, max_items=MAX_PRODUITS, progress=True)

        T = COMPARAISON["tableau"]
        colonnes = [c for c in ("méthode", "couverture", "justesse_globale",
                                "justesse_sur_classés", "précision_macro",
                                "rappel_macro", "F1_macro", "F1_pondéré",
                                "secondes", "produits_par_seconde",
                                "appels_modèle", "jetons", "coût_usd",
                                "coût_pour_1000") if c in T.columns]
        display(viz.table(T[colonnes], caption="{} annotated products, {} classes"
                          .format(REF["produits_annotés"], REF["classes"])))

        for phrase in evaluation.recommander(T):
            encadre(phrase)

        for methode, detail in COMPARAISON["détails"].items():
            conf = detail["confusions"]
            if len(conf):
                display(viz.table(conf.head(8), caption="{} — most frequent "
                                  "confusions".format(methode)))
```

## 6 · The weights

An all-items index is a **weighted average**: each division counts
according to its share of household expenditure. These shares come from the national CPI.

Enter them in cell 1, in the form `01=45.2, 02=1.8, 05=6.4`. The
weights need not sum to 100 — only their proportions matter.

```python
#@title Apply the weights { display-mode: "form" }
divisions = sorted(d for d in OBS["division"].dropna().unique())
POIDS = {}
for morceau in PONDERATIONS.split(","):
    if "=" in morceau:
        cle, _, valeur = morceau.partition("=")
        try:
            POIDS[cle.strip()] = float(valeur.strip().replace(",", "."))
        except ValueError:
            encadre("“{}” is not a number: this division is ignored."
                    .format(morceau.strip()), *ATTENTION)
if not POIDS:
    POIDS = {d: 1.0 for d in divisions}
    encadre("No weights provided: <b>equal shares</b> across the {} "
            "divisions covered. The index is then a trend indicator, "
            "not a measure comparable to the national CPI.".format(len(divisions)),
            *ATTENTION)

total = sum(POIDS.values()) or 1
display(viz.table(pd.DataFrame({
    "division": list(POIDS),
    "label": [config.COICOP_DIVISIONS.get(d, "—") for d in POIDS],
    "weight": [POIDS[d] for d in POIDS],
    "share": ["{:.1%}".format(POIDS[d] / total) for d in POIDS],
    "data available": ["yes" if d in divisions else "NO" for d in POIDS],
}), caption="Weights applied"))

absentes = [d for d in POIDS if d not in divisions]
if absentes:
    encadre("Divisions {} have a weight but no data: they will carry "
            "no weight. The share of the basket actually covered is therefore lower "
            "than that of the national CPI.".format(", ".join(absentes)),
            *ATTENTION)
```

## 7 · The index

The computation follows the CPI Manual (IMF, ILO, Eurostat). The
**Jevons** is the recommended formula; the **Carli** is computed to show its
upward drift when prices oscillate.

```python
#@title Compute the index { display-mode: "form" }
RES = index.compute_all(OBS, POIDS, reference=PERIODE_DE_REFERENCE.strip())
if not RES.get("ready", True):
    encadre("Only one month of price records ({}). An index compares two periods: "
            "a second collection is needed, next month. Everything else in this "
            "notebook remains valid."
            .format(", ".join(RES.get("periods", []))), *ATTENTION)
    raise SystemExit("Not computable yet — this is not an error.")

ENS = RES["indices"].query("série == 'Ensemble'").sort_values("period")
dernier = ENS.iloc[-1]
encadre("All-items index: <b>{:.1f}</b> in {} (base 100 = {}). "
        "Over one month: <b>{}</b>. Over one year: <b>{}</b>.".format(
            dernier["index"], dernier["period"], RES["reference"],
            "{:+.2f} %".format(dernier["mensuelle_%"]) if pd.notna(dernier["mensuelle_%"]) else "—",
            "{:+.1f} %".format(dernier["annuelle_%"]) if pd.notna(dernier["annuelle_%"]) else "—"),
        *SUCCES, titre="Result")
for phrase in messages.interpret_index(RES):
    encadre(phrase)
display(viz.table(RES["indices"].pivot(index="period", columns="série", values="index")
                  .reset_index(), caption="Indices by division (base 100)"))
```

## What this index measures, and what it does not

It measures the change in **prices displayed online**, at constant quality —
that is, on the same items from one month to the next (matched model). It is a
**complementary indicator**, not a substitute for the official CPI.

| Limitation | Why it matters |
|---|---|
| Online shoppers are urban and connected | open-air markets and the informal sector are not covered |
| Goods only | housing, transport, health and education are absent |
| Displayed prices, not prices paid | no quantities sold; promotions carry weight |
| The catalogue turns over quickly | new products only enter the following month |
| Classification is imperfect | cell 5 tells you by how much, **on your data** |

**Before publishing a figure**, check four things: that the weights
are those of your institute and not equal shares; that the file carries a
manifest, otherwise its provenance cannot be cited; that it is not the
demonstration dataset; and that the test bench was run on **your** data.

### If something does not work

| Message | What to do |
|---|---|
| “These files cannot be used” | the reason is written just below, in plain language |
| “Only one month of price records” | normal on the first run: a second collection is needed |
| “No model reachable” | no consequence, the rules take over |
| “Reference dataset unusable” | produce the sample to annotate, cell 5a |
| A cell stays grey for a long time | installation or classification is at work; wait |

For any question about the collection itself, notebook 1 is the place to
look — it runs on the institute's workstation.
